# Rappresentazione dell'informazione numerica

---

## 0. Il problema della rappresentazione

### 0.1 Dalla stringa di bit all'informazione

Un elaboratore memorizza sequenze finite di bit. Un bit (binary digit) è un’entità che può assumere uno tra due valori, convenzionalmente indicati con $0$ e $1$. La stessa stringa di $32$ bit
$$
111000001000000100100000000000000
$$
può rappresentare oggetti completamente diversi:

- un’istruzione assembly, ad esempio `add r2, r1, r0`;
- un numero naturale in base $2$, cioè $4034994176$;
- un numero intero in complemento a due, cioè $-259973120$;
- un numero razionale in virgola mobile, cioè $-7.44354946411796 \times 10^{19}$;
- una stringa di caratteri ASCII, ad esempio `<SOH><NUL>aü<NUL>`;
- molte altre interpretazioni.

Dunque **una stringa di bit non ha significato intrinseco**: il significato dipende dal tipo di dato e dalla codifica adottata. Questo è il punto di partenza concettuale di tutta la rappresentazione dell’informazione.

Un sistema di numerazione posizionale in matematica ammette segni, cifre e virgole. In un elaboratore sono disponibili **solo bit**. Non esistono il segno “$-$”, la virgola e le cifre decimali $0,\dots,9$ come simboli separati: tutto deve essere codificato con sequenze di bit.

Inoltre la rappresentazione è **finita**: con $k$ bit si possono rappresentare al più $2^k$ oggetti distinti. Il numero di oggetti rappresentabili cresce esponenzialmente con $k$:

| $k$ | $2^k$ |
|---|---|
| $4$ | $16$ |
| $10$ | $1024$ |
| $20$ | $2^{20}$ |
| $30$ | $2^{30}\approx 10^9$ |
| $k$ | $2^k$ |

Per $k$ grande si usano prefissi binari: $1$ Ki $=2^{10}$, $1$ Mi $=2^{20}$, $1$ Gi $=2^{30}$, $1$ Ti $=2^{40}$, e così via.

La conseguenza fondamentale è che **non tutti i numeri sono rappresentabili**. Ogni rappresentazione finita introduce un insieme finito di valori ammessi e un insieme di valori non rappresentabili.

---

## 1. Sistemi di numerazione posizionali

### Definizione 1.1 — Sistema posizionale in base $b$

Sia $b\in\mathbb{N}$, $b\ge 2$. Un numero reale $x$ è rappresentato in base $b$ dalla scrittura
$$
x=(\dots a_2a_1a_0\,.\,a_{-1}a_{-2}\dots)_b
$$
dove le cifre $a_i$ appartengono all’alfabeto $\{0,1,\dots,b-1\}$, e il valore numerico è
$$
x=\sum_{i=-\infty}^{+\infty} a_i b^i.
$$
Le cifre con indice maggiore sono le più **significative**, quelle con indice minore le meno significative. Il valore di una cifra dipende dalla sua **posizione**.

> **Osservazione 1.2.**  
> Il sistema posizionale è una rappresentazione **ridondante** se si ammettono zeri non significativi a sinistra: $(0011)_2=(11)_2$. In un calcolatore la lunghezza è fissa, quindi gli zeri a sinistra diventano parte della rappresentazione.

### 1.3 Esempi di basi

**Base 10.**
$$
2095.42_{10}
=
2\cdot 10^3+0\cdot 10^2+9\cdot 10+5+\frac{4}{10}+\frac{2}{10^2}.
$$

**Base 2.**
$$
11011.101_2
=
1\cdot 2^4+1\cdot 2^3+0\cdot 2^2+1\cdot 2+1+\frac{1}{2}+\frac{0}{2^2}+\frac{1}{2^3}
=
27.625_{10}.
$$

**Base 8.**
$$
675.1_8
=
6\cdot 8^2+7\cdot 8+5+\frac{1}{8}
=
445.125_{10}.
$$

**Base 16.**
$$
8B9.1_{16}
=
8\cdot 16^2+11\cdot 16+9+\frac{1}{16}
=
2233.0625_{10}.
$$

Per la base $16$ si usano le cifre $0,\dots,9,A,B,C,D,E,F$, con $A=10,\dots,F=15$.

## 2. Cambio di base

### 2.1 Da base $b$ a base $10$

Si applica direttamente la definizione:
$$
(\dots a_2a_1a_0\,.\,a_{-1}a_{-2}\dots)_b
=
\sum_{i=-\infty}^{+\infty} a_i b^i.
$$
Le operazioni si svolgono in base $10$.

### 2.2 Da base $10$ a base $b$: parte intera

Sia $n\in\mathbb{N}$. Si eseguono divisioni successive per $b$ e si raccolgono i resti. Formalmente, se
$$
n=q_0 b+r_0,\qquad 0\le r_0<b,
$$
$$
q_0=q_1 b+r_1,\qquad 0\le r_1<b,
$$
e così via fino a $q_m=0$, allora
$$
n=(q_{m-1}\dots q_1 q_0)_b
=
(r_m r_{m-1}\dots r_1 r_0)_b.
$$
L’ultimo resto è la cifra più significativa.

**Esempio.** $2009$ in base $16$:
$$
2009=125\cdot 16+9,
\qquad
125=7\cdot 16+13,
\qquad
7=0\cdot 16+7.
$$
Quindi
$$
2009_{10}=7D9_{16}.
$$

### 2.3 Da base $10$ a base $b$: parte frazionaria

Sia $f\in[0,1)$. Si moltiplica ripetutamente per $b$ e si raccolgono le parti intere:
$$
f b=c_{-1}+f_1,\qquad 0\le f_1<1,
$$
$$
f_1 b=c_{-2}+f_2,\qquad 0\le f_2<1,
$$
e così via. Allora
$$
f=(0.c_{-1}c_{-2}\dots)_b.
$$
La prima cifra è la più significativa. Il procedimento può **non arrestarsi**: ad esempio $0.3_{10}$ in base $2$ è periodico.

**Esempio.** $0.6875$ in base $2$:
$$
0.6875\cdot 2=1.375\Rightarrow c_{-1}=1,
$$
$$
0.375\cdot 2=0.75\Rightarrow c_{-2}=0,
$$
$$
0.75\cdot 2=1.5\Rightarrow c_{-3}=1,
$$
$$
0.5\cdot 2=1.0\Rightarrow c_{-4}=1.
$$
Quindi
$$
0.6875_{10}=0.1011_2.
$$

**Esempio.** $0.3_{10}$ in base $2$:
$$
0.3\cdot 2=0.6\Rightarrow 0,
$$
$$
0.6\cdot 2=1.2\Rightarrow 1,
$$
$$
0.2\cdot 2=0.4\Rightarrow 0,
$$
$$
0.4\cdot 2=0.8\Rightarrow 0,
$$
$$
0.8\cdot 2=1.6\Rightarrow 1,
$$
$$
0.6\cdot 2=1.2\Rightarrow 1,
$$
e il ciclo si ripete. Quindi
$$
0.3_{10}=0.0\overline{1001}_2.
$$

### 2.4 Conversioni rapide tra basi $2$, $8$, $16$

Poiché $8=2^3$ e $16=2^4$, ogni cifra ottale corrisponde a $3$ bit e ogni cifra esadecimale a $4$ bit.

**Da ottale a binario:** si espande ogni cifra in una terna di bit.
$$
475_8
=
\underbrace{100}_{4}\underbrace{111}_{7}\underbrace{101}_{5}
=
100111101_2.
$$

**Da binario a ottale:** si raggruppa in terne da destra e si sostituisce ogni terna con la cifra ottale corrispondente.

**Da esadecimale a binario:** si espande ogni cifra in una quaterna di bit.
$$
8B9_{16}
=
\underbrace{1000}_{8}\underbrace{1011}_{B}\underbrace{1001}_{9}
=
100010111001_2.
$$

**Da binario a esadecimale:** si raggruppa in quaterne da destra.

## 3. Rappresentazione finita nei calcolatori

### 3.1 Bit e lunghezza fissa

Un elaboratore lavora con un numero **finito** $k$ di bit. La rappresentazione di un numero naturale $n$ con $k$ bit è la stringa
$$
R_{N,k}(n)=(x_{k-1},\dots,x_0)
$$
costituita dalle $k$ cifre binarie meno significative della rappresentazione binaria di $n$. Se la rappresentazione binaria di $n$ ha meno di $k$ bit, si aggiungono zeri a sinistra.

**Vincolo fondamentale.** Sono rappresentabili solo i numeri
$$
n\in\{0,1,\dots,2^k-1\}.
$$
Un numero $n\ge 2^k$ non è rappresentabile con $k$ bit.

### 3.2 Decodifica

La stringa $x=(x_{k-1},\dots,x_0)$ rappresenta il numero naturale
$$
R_{N,k}^{-1}(x)=\sum_{i=0}^{k-1}x_i 2^i.
$$

### 3.3 Struttura ciclica dei naturali

La rappresentazione dei numeri naturali con $k$ bit ha una **struttura ciclica**: aggiungendo $1$ al massimo numero rappresentabile $2^k-1$ si ottiene $100\dots 00$, che richiede $k+1$ bit e quindi “esce” dall’insieme rappresentabile. Il bit che separa l’ultimo numero rappresentabile dal primo è detto **bit di guardia** (guardian line).

Per $k=32$, i numeri rappresentabili vanno da $0$ a $2^{32}-1\approx 4.29\times 10^9$.

## 4. Numeri interi: le rappresentazioni

Con $k$ bit si vogliono rappresentare numeri interi, inclusi i negativi. Non esiste una sola rappresentazione: se ne studiano quattro, ciascuna con proprietà diverse.

### 4.1 Ampiezza e segno

Il bit più significativo è usato come **bit di segno**:
$$
x_{k-1}=
\begin{cases}
1,&\text{se } q<0,\\
0,&\text{altrimenti}.
\end{cases}
$$
I restanti $k-1$ bit rappresentano $|q|$ in base $2$.

**Encoding:**
$$
R_{AS,k}(q)=(x_{k-1},x_{k-2},\dots,x_0),
\qquad
(x_{k-2},\dots,x_0)=R_{N,k-1}(|q|).
$$

**Decoding:**
$$
R_{AS,k}^{-1}(x)=(1-2x_{k-1})\sum_{i=0}^{k-2}x_i 2^i.
$$

**Numeri rappresentabili:**
$$
q\in\{-2^{k-1}+1,\dots,0,\dots,2^{k-1}-1\}.
$$

> **Osservazione.** Ampiezza e segno ha **due rappresentazioni dello zero** ($+0$ e $-0$) e una discontinuità tra $-0$ e $+0$. Riduce inoltre il numero di interi positivi rappresentabili, perché il bit di segno è “sprecato”.

### 4.2 Eccesso $P$

Si sceglie una costante $P$ (tipicamente $P=2^{k-1}$ oppure $P=2^{k-1}-1$) e si rappresenta $q$ come il numero naturale $q+P$:
$$
R_{E,P,k}(q)=R_{N,k}(q+P).
$$

**Decoding:**
$$
R_{E,P,k}^{-1}(x)=-P+\sum_{i=0}^{k-1}x_i 2^i.
$$

**Numeri rappresentabili:**
$$
q\in\{-P,\dots,0,\dots,P-1\}.
$$

**Esempio ($k=4$, $P=8$).**
$$
0000_2\leftrightarrow -8,
\quad
0001_2\leftrightarrow -7,
\quad
\dots,
\quad
1000_2\leftrightarrow 0,
\quad
\dots,
\quad
1111_2\leftrightarrow 7.
$$

L’eccesso $P$ ha una **unica rappresentazione dello zero** e una struttura asimmetrica. È usata nei formati in virgola mobile (IEEE 754, Microsoft Binary Format).

### 4.3 Complemento a $1$

I numeri positivi coincidono con la rappresentazione naturale. Per $q<0$ si prende il **complemento bit a bit** di $R_{N,k}(|q|)$:
$$
R_{c1,k}(q)=\mathrm{NOT}(R_{N,k}(|q|)),
\qquad
\mathrm{NOT}(x)_i=1-x_i.
$$

**Decoding:**
$$
R_{c1,k}^{-1}(x)
=
-x_{k-1}(2^{k-1}-1)+\sum_{i=0}^{k-1}x_i 2^i.
$$

**Numeri rappresentabili:**
$$
q\in\{-2^{k-1}+1,\dots,0,\dots,2^{k-1}-1\}.
$$

> **Osservazione.** Il complemento a $1$ ha **due rappresentazioni dello zero**: $00\dots 0$ e $11\dots 1$. È stato usato in CDC, PDP e UNIVAC.

### 4.4 Complemento a $2$

I numeri positivi coincidono con la rappresentazione naturale. Per $q<0$ si somma $1$ al complemento a $1$:
$$
R_{c2,k}(q)=\mathrm{NOT}(R_{N,k}(|q|))+1.
$$

**Decoding:**
$$
R_{c2,k}^{-1}(x)
=
-x_{k-1}2^{k-1}+\sum_{i=0}^{k-1}x_i 2^i.
$$

**Numeri rappresentabili:**
$$
q\in\{-2^{k-1},\dots,0,\dots,2^{k-1}-1\}.
$$

Il complemento a $2$ è l’unica rappresentazione, tra quelle viste, che soddisfa **tutte** le seguenti proprietà:

1. ha una sola rappresentazione dello zero;
2. il bit più significativo indica il segno;
3. ha una struttura ciclica: aggiungendo $1$ al massimo si ottiene il minimo;
4. le operazioni aritmetiche sui numeri negativi usano le stesse regole dei positivi.

Per questo è la rappresentazione standard di x86, ARM, MIPS, SPARC, Power Architecture, Itanium.

---

## 5. Complemento a $2$: struttura approfondita

### 5.1 Interpretazione come peso dei bit

La formula di decoding
$$
x=-x_{k-1}2^{k-1}+\sum_{i=0}^{k-2}x_i 2^i
$$
mostra che nel complemento a $2$ il bit più significativo ha **peso negativo** $-2^{k-1}$, mentre tutti gli altri hanno peso positivo $2^i$. Questa è la differenza concettuale rispetto alla rappresentazione naturale, in cui tutti i bit hanno peso positivo.

**Esempio ($k=8$).** Per $x=10000000_2$,
$$
x=-1\cdot 2^7+0=-128.
$$
Per $x=11111111_2$,
$$
x=-128+\sum_{i=0}^{6}2^i=-128+127=-1.
$$
Per $x=01111111_2$,
$$
x=0+127=127.
$$

### 5.2 Struttura ciclica

La rappresentazione in complemento a $2$ ha una struttura **ciclica**: i numeri sono disposti su una circonferenza in cui il successore del massimo $2^{k-1}-1$ è il minimo $-2^{k-1}$. Aggiungere $1$ corrisponde a muoversi in senso antiorario; sottrarre $1$ a muoversi in senso orario.

Per $k=32$,
$$
\begin{aligned}
2^{31}-1&=0111\dots 11,\\
2^{31}&=1000\dots 00=-2^{31},\\
2^{31}+1&=1000\dots 01=-2^{31}+1,\\
&\vdots\\
2^{32}-2&=1111\dots 10=-2,\\
2^{32}-1&=1111\dots 11=-1.
\end{aligned}
$$
L’“overflow” è esattamente il superamento del bit di guardia.

### 5.3 Aritmetica

Il complemento a $2$ consente di eseguire somme e sottrazioni con le stesse regole dei numeri naturali. Ad esempio, con $4$ bit:
$$
0011_2+0100_2=0111_2,
\qquad
3+4=7,
$$
$$
0011_2+1100_2=1111_2,
\qquad
3+(-4)=-1.
$$

### 5.4 Overflow

Nell’eseguire un’operazione, il risultato potrebbe non essere rappresentabile con $k$ bit. Si parla allora di **overflow**. Le regole per riconoscerlo sono:

- se si sommano due numeri positivi e si ottiene un numero negativo, c’è overflow;
- se si sommano due numeri negativi e si ottiene un numero positivo, c’è overflow;
- se si sommano un numero positivo e uno negativo, non c’è mai overflow.

Equivalentemente, l’overflow si verifica quando il **bit di segno** e il **bit di riporto in uscita** dalla posizione più significativa sono diversi.

**Esempio ($4$ bit).**
$$
0101_2+0010_2=0111_2,
\qquad
5+2=7\ \text{OK},
$$
$$
1011_2+1110_2=1\,1001_2,
\qquad
-5+(-2)=-7\ \text{OK},
$$
$$
0101_2+0101_2=1010_2,
\qquad
5+5=10\ \text{overflow},
$$
perché il risultato $1010_2$ rappresenta $-6$, non $+10$.

### 5.5 Estensione del numero di bit

Se si passa da $N$ a $P$ bit, con $P>N$, la rappresentazione deve essere **coerente**: il valore numerico deve restare lo stesso.

**Ampiezza e segno.** Si sposta il bit di segno nel nuovo bit più significativo e si mettono a $0$ i bit intermedi:
$$
1011_2\ (\text{con }4\text{ bit},\, -3)
\longrightarrow
10000011_2\ (\text{con }8\text{ bit},\, -3).
$$

**Complemento a $2$.** Si applica la **regola dell’estensione del segno**: i nuovi bit vengono impostati uguali al bit di segno originale. Se il numero è positivo, si aggiungono zeri; se è negativo, si aggiungono uni:
$$
1101_2\ (\text{con }4\text{ bit},\, -3)
\longrightarrow
11111101_2\ (\text{con }8\text{ bit},\, -3),
$$
$$
0110_2\ (\text{con }4\text{ bit},\, +6)
\longrightarrow
00000110_2\ (\text{con }8\text{ bit},\, +6).
$$

**Giustificazione formale.** Sia $x=(x_{N-1},\dots,x_0)$ con $x_{N-1}=1$ (numero negativo). Il valore è
$$
v=-2^{N-1}+\sum_{i=0}^{N-2}x_i 2^i.
$$
Se si estende a $P$ bit replicando il segno, la nuova stringa ha i bit $x_{P-1}=\dots=x_{N-1}=1$ e gli stessi $x_{N-2},\dots,x_0$. Il nuovo valore è
$$
v'
=
-2^{P-1}+\sum_{i=N-1}^{P-1}2^i+\sum_{i=0}^{N-2}x_i 2^i.
$$
Ma
$$
\sum_{i=N-1}^{P-1}2^i=2^{P-1}-2^{N-1},
$$
quindi
$$
v'
=
-2^{P-1}+(2^{P-1}-2^{N-1})+\sum_{i=0}^{N-2}x_i 2^i
=
-2^{N-1}+\sum_{i=0}^{N-2}x_i 2^i
=
v.
$$
Dunque il valore è preservato. $\square$
