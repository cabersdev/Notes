### Esercizio 0.1

In $\mathbb{R}^2$ determiniamo massimi e minimi assoluti per $f(x,y) = x^3 - y^3$ su 

$$
A = \{(x,y) \in \mathbb{R}^3 : x \geq -1, y \leq -x + 1, y \geq x - 1\}
$$

### Soluzione 

$A$ è chiuso poichè intersezione di $3$ insiemi di sotto/sopra livello con il $\geq$ o $\leq$ di funzioni continue definite in $\mathbb{R}^3$

$A$ è limitato infatti: 

Dalle ultime due si ha 

$$
x-1 \leq y \leq 1-x
$$

Perché esista un $y$ che soddisfa entrambe, deve valere $x-1 \leq 1-x$ quindi 

$$
2x \leq 2 \implies x \leq 1
$$

Insieme alla condizione iniziale $x \geq -1$, otteniamo 

$$
-1 \leq x \leq 1
$$

Ora limitiamo $y$. Poiché $x \geq -1$ si ha 

$$
y \geq x-1 \geq -1-1 = -2
$$

e poichè $x \geq -1$

$$
y \leq 1 - x \leq 1 - (-1) = 2
$$

Dunque 

$$
-2 \leq y \leq 2
$$

$f \in C^k(\mathbb{R}^3) \implies f \in C^k(A)$, per il teorema di bolzano weierstrass $f$ ammette quindi massimo e minimo assoluto su $A$

Decomponiamo la frontiera in 

$$
A = A^{\circ} \cup \partial A
$$

$$
A^{\circ} = \{(x,y) \in \mathbb{R}^2 : x > -1, y < - x + 1, y > x-1 \}
$$

$$
\nabla_f(x,y) = 0 \iff 
\begin{pmatrix}
\frac{\partial f}{\partial x} = 0 \\
\frac{\partial f}{\partial y} = 0
\end{pmatrix} \iff
\begin{cases}
3x^2 = 0 \\
-3y^2 = 0
\end{cases} \iff 
\begin{cases}
x = 0 \\
y = 0
\end{cases}
$$

$(0,0) \in A^{\circ}$

Decomponiamo la frontiera: 

$$
\begin{aligned}
\partial A = & \{x = -1, y > x - 1, y < -x + 1\} \\
    \cup     & \{x > -1, y = x - 1, y < -x + 1\} \\
    \cup     & \{x > -1, y > x - 1, y = -x + 1\} \\
    \cup     & \{x = -1, y = x - 1, y < -x + 1\} \\
    \cup     & \{x = -1, y > x - 1, y = -x + 1\} \\
    \cup     & \{x > -1, y = x - 1, y = -x + 1\} \\
    \cup     & \{x = -1, y = x - 1, y = -x + 1\}
\end{aligned}
$$

Si ha che 

$$
\{x = -1, y > x - 1, y < -x + 1\}
$$

per $f(x,y) = x^3 - y^3$ su $(-1, y)$ non ammette massimo e minimo in quanto $f$ è una funzione monotona su un aperto di $\mathbb{R}^2$

$$
\{x > -1, y = x - 1, y < -x + 1\}
$$

$f(x, x-1) = x^3 - (x-1)^3 = 3x^2 - 3x+1$ con $x > -1, x < 1$ c'è un punto di minimo per $f$: $(\frac{1}{2}, -\frac{1}{2})$

$$
\{x > -1, y > x - 1, y = -x + 1\}
$$

$f(x, -x+1) = x^3 -(-x+1)^3 = 2x^3 - 3x^2 + 3x-1$ con $-1 < x < 1$, $f$ non ammette massimo e minimo su $\{x > -1, y > x - 1, y = -x + 1\}$ in quanto funzione monotona su un un aperto

$$
\{x = -1, y = x - 1, y < -x + 1\} = (-1, -2)
$$

$$
\{x = -1, y > x - 1, y = -x + 1\} = (-1, 2)
$$

$$
\{x > -1, y = x - 1, y = -x + 1\} = (1, 0) 
$$

$$
\{x = -1, y = x - 1, y = -x + 1\} = \emptyset
$$

Mettendo assieme i risultati ottenuti 

$$
f(0,0) = 0, \; f(\frac{1}{2}, -\frac{1}{2}) = \frac{1}{4}, \; f(-1, 2) = -9, \; f(-1, 2) = 7, \; f(1, 0) = 1
$$

Si ha quindi che il punto $(-1, 2)$ è di minimo assoluto e il punto $(-1, -2)$ è di massimo assoluto per $f$ in $A$

### Esercizio 0.2 

In $\mathbb{R}^2$ determinare massimi e minimi assoluti di $f(x,y) = x-y$ su $A = \{(x,y) \in \mathbb{R}^2 : x^2 + y^2 \leq 1\}$

### Soluzione

$A$ è chiuso poiché è un insieme di sottolivvelo con il $\leq$ di una funzione continua in $\mathbb{R}^2$

$A$ è limitato poichè 

$$
x^2 + y^2 \iff x \leq \sqrt{1-y^2} \implies 1-y^2 \geq 0 \implies -1 \leq y \leq 1
$$

Analogamente:

$$
x^2 + y^2 \iff y \leq \sqrt{1-x^2} \implies 1-x^2 \geq 0 \implies -1 \leq x \leq 1
$$

$f \in C^k(\mathbb{R}^2)$ poiché composizione di funzioni elementari, si ha quindi che $f \in C^k(A)$

Per il teorema di weierstrass $f$ ammette quindi massimi e minimi assoluti su $A$

Decomponiamo $A$

$$
A = A^{\circ} \cup \partial A = \{(x,y) \in \mathbb{R}^2 : x^2 + y^2 < 1\} \cup \{(x,y) \in \mathbb{R}^2 : x^2 + y^2 = 1\}
$$

$$
\nabla_f(x,y) = 0 \iff 
\begin{cases}
\frac{\partial f}{\partial x} = 0 \\ 
\frac{\partial f}{\partial y} = 0
\end{cases} \iff 
\begin{cases}
1 = 0 \\
-1 = 0
\end{cases}
$$

$f$ non ammette punti critici in $A^{\circ}$

Scriviamo $f$ in coordinate polari

$$
x = \rho\cos\theta, \; y = \rho\sin\theta, \; \rho = 1
$$

(Reppresentazione parametrica della circonferenza rispetto al parametro $\theta \in \mathbb{R}$)

$$
f(x,y) = f(\cos\theta, \sin\theta) = \cos\theta - \sin\theta := h(\theta)
$$

$h(\theta)$ è $2\pi$-periodica consideriamo la 

$$
h'(\theta) = -\sin\theta - \cos\theta
$$

$$
h'(\theta) = 0 \iff \sin\theta + \cos\theta = 0 \iff \sin\theta = -\cos\theta \iff \theta = \frac{3}{4}\pi \cup \theta = \frac{7}{4}\pi 
$$

$$
h'(\theta) > 0 \iff \sin\theta + \cos\theta < 0 \iff \sin\theta < -\cos\theta \iff \frac{3}{4}\pi < \theta < \frac{7}{4}\pi
$$

$h(\theta)$ ha minimo in $\frac{3}{4}\pi$ da cui $(-\frac{1}{\sqrt{2}}, \frac{1}{\sqrt{2}})$ è minimo assoluto per $f$ in $A$

$h(\theta)$ ha massimo in $\frac{7}{4}\pi$ da cui $(\frac{1}{\sqrt{2}},-\frac{1}{\sqrt{2}})$ è massimo assoluto per $f$ in $A$

### Moltiplicatori di lagrange

$f(x,y) = x-y \in C^1(\mathbb{R}^2)$, $g(x,y) = x^2 + y^2 - 1 \in C^1(\mathbb{R}^2)$

$$
\nabla_g(x,y) = 0 \iff (2x, 2y) = (0,0) \notin A
$$

$\forall (x,y) \in A \implies \nabla_g(x,y) \neq (0,0)$ il vincolo è regolare

$$
\nabla_f(x,y) = \lambda \nabla_g(x,y)
$$ 

Da cui 

$$
\begin{cases}
1 = \lambda 2x \\
-1 = \lambda 2y \\ 
x^2 + y^2 = 1
\end{cases} \iff 
\begin{cases}
\frac{1}{\lambda} = 2x \\
\frac{1}{\lambda} = -2y \\
x^2 + y^2 = 1
\end{cases}
$$

Da cui 

$$
\begin{cases}
2x = -2y \\
x^2 + y^2 = 1 
\end{cases} \iff 
\begin{cases}
y = -\frac{1}{\sqrt{2}} \\
x = \frac{1}{\sqrt{2}}
\end{cases} \cup 
\begin{cases}
y = \frac{1}{\sqrt{2}} \\
x = -\frac{1}{\sqrt{2}} 
\end{cases}
$$

Da cui si ha 

$$
f(\frac{1}{\sqrt{2}}, -\frac{1}{\sqrt{2}}) = \sqrt{2}, \quad f(-\frac{1}{\sqrt{2}}, \frac{1}{\sqrt{2}}) = -\sqrt{2}
$$

$(\frac{1}{\sqrt{2}}, -\frac{1}{\sqrt{2}})$ è minimo assoluto per $f$ in $A$, $(-\frac{1}{\sqrt{2}}, \frac{1}{\sqrt{2}})$ è massimo assoluto per $f$ in $A$

### Esercizio 0.3

Sia $f(x,y,z) = x-y+z$ determiniamo massimi e minimi assoluti su 

$$
A = \{(x,y,z) \in \mathbb{R}^3 : x^2 + \frac{y^2}{4}+ \frac{z^2}{9} = 1\}
$$

$f \in C^k(\mathbb{R}^3) \implies f \in C^k(A)$ essendo composizione di funzioni elementari 

$A$ è chiuso essendo una curva di livello di una funzione $g: \mathbb{R}^3 \to \mathbb{R}$ continua

$$
\frac{z^2}{9} \leq x^2 + \frac{y^2}{4}+ \frac{z^2}{9} = 1 \implies -3 \leq z \leq 3
$$

$$
x^2 \leq x^2 + \frac{y^2}{4}+ \frac{z^2}{9} = 1 \implies -1 \leq x \leq 1
$$

$$
\frac{y^2}{4} \leq x^2 + \frac{y^2}{4}+ \frac{z^2}{9} = 1 \implies -2 \leq y \leq 2
$$

$A$ è limitato, e quindi compatto 

Per il teorema di Weierstrass esistono massimi e minimi assoluti per $f$ in $A$

$$
g(x,y,z) = x^2 + \frac{y^2}{4} + \frac{z^2}{9} \in C^1(\mathbb{R}^3)
$$

$$
\nabla_g(x,y,z) = (2x, \frac{y}{2}, \frac{2}{9}z)
$$

$$
\nabla_g(x,y,z) = 0 \iff 
\begin{cases}
x = 0 \\ y = 0 \\ z = 0
\end{cases}
$$

$(0,0,0) \notin A \implies \forall (x,y,z) \in A$ il vincolo è regolare

$$
\nabla_f(x,y,z) = \lambda \nabla_g(x,y,z) \iff
$$

$$
\begin{cases}
    1 = \lambda 2x\\
    -1 = \lambda \frac{1}{2}y  \\
    1 = \lambda \frac{2}{9}z \\
    x^2 + \frac{y^2}{4}+ \frac{z^2}{9} = 1
\end{cases} \iff 
\begin{cases}
    1 = \lambda 2x \\
    1 = -\lambda \frac{1}{2}y \\
    1 = \lambda \frac{2}{9}y \\
    x^2 + \frac{y^2}{4}+ \frac{z^2}{9} = 1
\end{cases} \iff 
\begin{cases}
    1 = \lambda 2x \\
    \lambda 2x = -\lambda \frac{1}{2}y \\
    \lambda 2x = \lambda \frac{2}{9}y \\
    x^2 + \frac{y^2}{4}+ \frac{z^2}{9} = 1
\end{cases} \iff 
\begin{cases}
    1 = \lambda 2x \\
    \lambda(2x + \frac{1}{2}y) = 0\\
    2\lambda(x - \frac{1}{9}z) = 0 \\
    x^2 + \frac{y^2}{4}+ \frac{z^2}{9} = 1
\end{cases}
$$

Dalla seconda equazione si ha $\lambda = 0, 2x+\frac{1}{2} = y$

$\lambda = 0$ tuttavia rende il sistema impossibile da cui 

$$
\begin{cases}
    1 = 2\lambda x \\
    2x + \frac{1}{2} y = 0 \\
    x -\frac{z}{9} = 0 \\
    x^2 + \frac{y^2}{4}+ \frac{z^2}{9} = 1
\end{cases} \iff 
\begin{cases} 
    1 = 2\lambda x \\
    y = -4x \\
    z = 9x \\ 
    x^2 + 4x^2 + 9x^2 = 1
\end{cases} \iff 
\begin{cases}
    1 = 2\lambda x \\
    y = -4x \\
    z = 9x \\ 
    x^2 = \frac{1}{14}
\end{cases}
$$

$$
f(\frac{1}{14}, -\frac{4}{14}, \frac{9}{14}) = \sqrt{14}, \quad f(-\frac{1}{14}, \frac{4}{14}, -\frac{9}{14}) = -\sqrt{14}
$$

Il punto $(\frac{1}{14}, -\frac{4}{14}, \frac{9}{14})$ è quindi massimo assoluto per $f$ in $A$, il punto $(-\frac{1}{14}, \frac{4}{14}, -\frac{9}{14})$ è quindi minimo assoluto per $f$ in $A$
