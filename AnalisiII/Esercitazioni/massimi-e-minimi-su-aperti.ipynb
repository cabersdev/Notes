# Esercitazioni su massimi e minimi su insiemi aperti

---

### Esercizio 0.1

$$
f(x,y,z) = \frac{x^3}{3} - x - y^2 + z^3 - 3z
$$

Determinare massimi e minimi relativi e/o assoluti

### Soluzione

Dominio di $f = \{(x,y,z) \in \mathbb{R}^3\}$

In particolare $f \in C^2(\mathbb{R})$, valgono quindi le condizioni sufficienti per l'esistenza di massimi e minimi in insiemi aperti

Determiniamo le derivate parziali di $f$

$$
\frac{\partial}{\partial x} \frac{x^3}{3} - x - y^2 + z^3 - 3z = x^2 - 1
$$

$$
\frac{\partial}{\partial y} \frac{x^3}{3} - x - y^2 + z^3 - 3z = -2y
$$

$$
\frac{\partial}{\partial z} \frac{x^3}{3} - x - y^2 + z^3 - 3z = 3z^2 - 3
$$

Da cui 

$$
\nabla_f(x,y,z) = 
\begin{pmatrix}
    x^2 - 1 \\
    -2y \\
    3z^2 - 3
\end{pmatrix}
$$

Poniamo 
$$
\nabla_f(x,y,z) = 0 \iff \begin{cases}
    x^2 - 1 = 0 \\
    -2y = 0\\
    3z^2 - 3 = 0
\end{cases} \iff \begin{cases}
    x = \pm 1 \\
    y = 0 \\
    z = \pm 1
\end{cases}
$$

Vi sono quindi $4$ punti critici $\{(1,0,1), (1,0, -1), (-1, 0, 1), (-1, 0, -1)\}$

$$
H_f(x,y,z) = 
\begin{pmatrix}
    2x & 0 & 0 \\
    0 & -2 & 0 \\
    0 & 0 & 6z 
\end{pmatrix}
$$

Studiamo la matrice hessiana nei vari punti critici 

$$
H_f(1,0,1) = 
\begin{pmatrix}
    2 & 0 & 0 \\
    0 & -2 & 0 \\
    0 & 0 & 6 
\end{pmatrix}
$$

$\text{Spec}(H_f(1,0,1)) = \{2, -2, 6\}$ il che implica che $H_f(1,0,1)$ è indefinita poiché presenta autovalori con segno diverso, si ha quindi che il punto $(1,0,1)$ è un punto di sella

$$
H_f(1,0,-1) = 
\begin{pmatrix}
    2 & 0 & 0 \\
    0 & -2 & 0 \\
    0 & 0 & -6
\end{pmatrix}
$$

$\text{Spec}(H_f(1,0,-1)) = \{2, -2, -6\}$ il che implica che $H_f(1,0,-1)$ è indefinita poiché presenta autovalori con segno opposto, si ha quindi che il punto $(1,0,-1)$ è un punto di sella


$$
H_f(-1,0,1) = 
\begin{pmatrix}
    -2 & 0 & 0 \\
    0 & -2 & 0 \\
    0 & 0 & 6 
\end{pmatrix}
$$

$\text{Spec}(H_f(-1,0,1)) = \{-2, -2, 6\}$ il che implica che $H_f(-1,0,1)$ è indefinita poiché presenta autovalori con segno opposto, si ha quindi che il punto $(-1,0,1)$ è un punto di sella

$$
H_f(-1,0,-1) = 
\begin{pmatrix}
    -2 & 0 & 0 \\
    0 & -2 & 0 \\
    0 & 0 & -6 
\end{pmatrix}
$$

$\text{Spec}(H_f(-1,0,-1)) = \{-2,-2,6\}$ tutti gli autovalori sono strettamenti negativi il che implica che $H_f(-1, 0, -1)$ è definita negativa e di conseguenza il punto $(-1,0,-1)$ è di massimo relativo

$f$ non presenta minimi relativi, non ammette quindi minimi assoluti

Essendo che $f$ ammette un unico massimo relativo, allora se $f$ ammette un massimo assoluto il massimo assoluto coincide con il massimo relativo 

$(-1, 0, -1)$ è massimo assoluto se $\forall x \in \text{Dom}f, \; f(x) \leq f(-1, 0, -1)$

Devo quindi dimostrare che 

$$
\frac{x^3}{3} -x - y^2 +z^3 - 3z \leq \frac{8}{3}
$$

Considerando il punto $(0,0, 10) = 1000 - 30 > \frac{8}{3}$ possiamo concludere che $f$ non ammette massimo assoluto

### Esercizio 0.2

$$
f(x,y,z) = 2x^2 + y^2 + z^2 + 2xz + 2yz
$$

Determinare massimi e minimi relativi e/o assoluti

### Soluzione

Dominio di $f = \{(x,y,z) \in \mathbb{R}^3\}$

In particolare $f \in C^2(\mathbb{R})$, valgono quindi le condizioni sufficienti per l'esistenza di massimi e minimi in insiemi aperti

Calcoliamo le derivate parziali

$$
\frac{\partial}{\partial x} 2x^2 + y^2 + z^2 + 2xz + 2yz = 4x + 2z
$$

$$
\frac{\partial}{\partial x} 2x^2 + y^2 + z^2 + 2xz + 2yz = 2y + 2z
$$

$$
\frac{\partial}{\partial x} 2x^2 + y^2 + z^2 + 2xz + 2yz = 2z + 2x + 2y

$$

Da cui 

$$
\nabla_f(x,y,z) = 
\begin{pmatrix}
    4x + 2z \\
    2y + 2z \\
    2z + 2x + 2y
\end{pmatrix}
$$

Poniamo il gradiente di $f$ pari a $0$

$$
\nabla_f(x,y,z) = 
\begin{pmatrix}
    4x + 2z \\
    2y + 2z \\
    2z + 2x + 2y
\end{pmatrix} = 0 \iff 
\begin{cases}
    4x + 2z = 0 \\
    2y + 2z = 0\\
    2z + 2x + 2y = 0
\end{cases} \iff 
\begin{cases}
    x = 0 \\
    y = 0 \\
    z = 0
\end{cases}
$$

Il punto $(0,0,0)$ è quindi critico per $f$

Costruiamo l'hessiana di $f$ in un generico punto del dominio

$$
H_f(x,y,z) = 
\begin{pmatrix}
    4 & 0 & 2 \\
    0 & 2 & 2 \\
    2 & 2 & 2
\end{pmatrix}
$$

Poiché l'hessiana è costante $H_f(x,y,z) = H_f(0,0,0)$, cerchiamo gli autovalori della matrice hessiana

$$
p_\lambda = \det(H_f(0,0,0) - \lambda I_3) = \det \begin{pmatrix}
    4 - \lambda& 0 & 2 \\
    0 & 2 - \lambda& 2 \\
    2 & 2 & 2 - \lambda
\end{pmatrix} = -\lambda^3 + 8\lambda^2 - 12\lambda - 8 
$$

$$
p_\lambda = 0 \iff -\lambda^3 + 8\lambda^2 - 12\lambda - 8 = 0 
$$

Il punto $(0,0,0)$ è di sella in quanto gli autovalori non possono essere tutti negativi.
