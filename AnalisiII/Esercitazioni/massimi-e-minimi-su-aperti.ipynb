### Esercizio 0.1

$$
f(x,y,z) = \frac{x^3}{3} - x - y^2 + z^3 - 3z
$$

Determinare massimi e minimi relativi e/o assoluti

### Soluzione

Dominio di $f = \{(x,y,z) \in \mathbb{R}^3\}$

In particolare $f \in C^2(\mathbb{R})$, valgono quindi le condizioni sufficienti per l'esistenza di massimi e minimi in insiemi aperti

Determiniamo le derivate parziali di $f$

$$
\frac{\partial}{\partial x} \frac{x^3}{3} - x - y^2 + z^3 - 3z = x^2 - 1
$$

$$
\frac{\partial}{\partial y} \frac{x^3}{3} - x - y^2 + z^3 - 3z = -2y
$$

$$
\frac{\partial}{\partial z} \frac{x^3}{3} - x - y^2 + z^3 - 3z = 3z^2 - 3
$$

Da cui 

$$
\nabla_f(x,y,z) = 
\begin{pmatrix}
    x^2 - 1 \\
    -2y \\
    3z^2 - 3
\end{pmatrix}
$$

Poniamo 
$$
\nabla_f(x,y,z) = 0 \iff \begin{cases}
    x^2 - 1 = 0 \\
    -2y = 0\\
    3z^2 - 3 = 0
\end{cases} \iff \begin{cases}
    x = \pm 1 \\
    y = 0 \\
    z = \pm 1
\end{cases}
$$

Vi sono quindi $4$ punti critici $\{(1,0,1), (1,0, -1), (-1, 0, 1), (-1, 0, -1)\}$

$$
H_f(x,y,z) = 
\begin{pmatrix}
    2x & 0 & 0 \\
    0 & -2 & 0 \\
    0 & 0 & 6z 
\end{pmatrix}
$$

Studiamo la matrice hessiana nei vari punti critici 

$$
H_f(1,0,1) = 
\begin{pmatrix}
    2 & 0 & 0 \\
    0 & -2 & 0 \\
    0 & 0 & 6 
\end{pmatrix}
$$

$\text{Spec}(H_f(1,0,1)) = \{2, -2, 6\}$ il che implica che $H_f(1,0,1)$ è indefinita poiché presenta autovalori con segno diverso, si ha quindi che il punto $(1,0,1)$ è un punto di sella

$$
H_f(1,0,-1) = 
\begin{pmatrix}
    2 & 0 & 0 \\
    0 & -2 & 0 \\
    0 & 0 & -6
\end{pmatrix}
$$

$\text{Spec}(H_f(1,0,-1)) = \{2, -2, -6\}$ il che implica che $H_f(1,0,-1)$ è indefinita poiché presenta autovalori con segno opposto, si ha quindi che il punto $(1,0,-1)$ è un punto di sella


$$
H_f(-1,0,1) = 
\begin{pmatrix}
    -2 & 0 & 0 \\
    0 & -2 & 0 \\
    0 & 0 & 6 
\end{pmatrix}
$$

$\text{Spec}(H_f(-1,0,1)) = \{-2, -2, 6\}$ il che implica che $H_f(-1,0,1)$ è indefinita poiché presenta autovalori con segno opposto, si ha quindi che il punto $(-1,0,1)$ è un punto di sella

$$
H_f(-1,0,-1) = 
\begin{pmatrix}
    -2 & 0 & 0 \\
    0 & -2 & 0 \\
    0 & 0 & -6 
\end{pmatrix}
$$

$\text{Spec}(H_f(-1,0,-1)) = \{-2,-2,6\}$ tutti gli autovalori sono strettamenti negativi il che implica che $H_f(-1, 0, -1)$ è definita negativa e di conseguenza il punto $(-1,0,-1)$ è di massimo relativo

$f$ non presenta minimi relativi, non ammette quindi minimi assoluti

Essendo che $f$ ammette un unico massimo relativo, allora se $f$ ammette un massimo assoluto il massimo assoluto coincide con il massimo relativo 

$(-1, 0, -1)$ è massimo assoluto se $\forall x \in \text{Dom}f, \; f(x) \leq f(-1, 0, -1)$

Devo quindi dimostrare che 

$$
\frac{x^3}{3} -x - y^2 +z^3 - 3z \leq \frac{8}{3}
$$

Considerando il punto $(0,0, 10) = 1000 - 30 > \frac{8}{3}$ possiamo concludere che $f$ non ammette massimo assoluto

### Esercizio 0.2

$$
f(x,y,z) = 2x^2 + y^2 + z^2 + 2xz + 2yz
$$

Determinare massimi e minimi relativi e/o assoluti

### Soluzione

Dominio di $f = \{(x,y,z) \in \mathbb{R}^3\}$

In particolare $f \in C^2(\mathbb{R})$, valgono quindi le condizioni sufficienti per l'esistenza di massimi e minimi in insiemi aperti

Calcoliamo le derivate parziali

$$
\frac{\partial}{\partial x} 2x^2 + y^2 + z^2 + 2xz + 2yz = 4x + 2z
$$

$$
\frac{\partial}{\partial x} 2x^2 + y^2 + z^2 + 2xz + 2yz = 2y + 2z
$$

$$
\frac{\partial}{\partial x} 2x^2 + y^2 + z^2 + 2xz + 2yz = 2z + 2x + 2y

$$

Da cui 

$$
\nabla_f(x,y,z) = 
\begin{pmatrix}
    4x + 2z \\
    2y + 2z \\
    2z + 2x + 2y
\end{pmatrix}
$$

Poniamo il gradiente di $f$ pari a $0$

$$
\nabla_f(x,y,z) = 
\begin{pmatrix}
    4x + 2z \\
    2y + 2z \\
    2z + 2x + 2y
\end{pmatrix} = 0 \iff 
\begin{cases}
    4x + 2z = 0 \\
    2y + 2z = 0\\
    2z + 2x + 2y = 0
\end{cases} \iff 
\begin{cases}
    x = 0 \\
    y = 0 \\
    z = 0
\end{cases}
$$

Il punto $(0,0,0)$ è quindi critico per $f$

Costruiamo l'hessiana di $f$ in un generico punto del dominio

$$
H_f(x,y,z) = 
\begin{pmatrix}
    4 & 0 & 2 \\
    0 & 2 & 2 \\
    2 & 2 & 2
\end{pmatrix}
$$

Poiché l'hessiana è costante $H_f(x,y,z) = H_f(0,0,0)$, cerchiamo gli autovalori della matrice hessiana

$$
p_\lambda = \det(H_f(0,0,0) - \lambda I_3) = \det \begin{pmatrix}
    4 - \lambda& 0 & 2 \\
    0 & 2 - \lambda& 2 \\
    2 & 2 & 2 - \lambda
\end{pmatrix} = -\lambda^3 + 8\lambda^2 - 12\lambda - 8 
$$

$$
p_\lambda = 0 \iff -\lambda^3 + 8\lambda^2 - 12\lambda - 8 = 0 
$$

Il punto $(0,0,0)$ è di sella in quanto gli autovalori non possono essere tutti negativi.

In alternativa (o se non riesco a concludere diversavemente) posso fare uno studio di funzione e determinare il segno dei vari zeri

$-\lambda^3 + 8\lambda^2 - 12\lambda - 8 = f(\lambda)$

### Esercizio 1

Determinare gli eventuali massimi e minimi relativi e/o assoluti della funzione

$$
    f(x,y,z) = y^2z^2 + 2y^2x^2 − y^4x^4 − 1
$$

### Soluzione 

$f \in C^k(\mathbb{R})$ poiché composizione di funzioni elementari, $\mathbb{R}^3$ è aperto, non posso quindi applicare il teorema di Weierstrass e quindi concludere in linea di principio l'esistenza di massimi e minimi assoluti

$$
\begin{cases}
    \frac{\partial f}{\partial x} = 0 \\
    \frac{\partial f}{\partial y} = 0 \\
    \frac{\partial f}{\partial z} = 0
\end{cases} \iff 
\begin{cases}
    4y^2x - 4x^3y^4 = 0 \\
    2yz^2 + 4x^2y - 4x^4y^3 = 0 \\
    2y^2z = 0
\end{cases}
$$

Da cui si ottengono i seguenti punti critici 

$$
C = \left\{
\begin{pmatrix}
    a \\ 0 \\ b
\end{pmatrix} : a,b \in \mathbb{R}
\right\} \cup 
\left\{
\begin{pmatrix}
    \frac{1}{c} \\ c \\ 0
\end{pmatrix} : c \in \mathbb{R}\backslash\{0\}
\right\} \cup
\left\{
\begin{pmatrix}
    -\frac{1}{c} \\ c \\ 0
\end{pmatrix} : c \in \mathbb{R}\backslash\{0\}
\right\} \cup 
\left\{
\begin{pmatrix}
    0 \\ d \\ 0
\end{pmatrix} : d \in \mathbb{R}\backslash\{0\}
\right\}
$$

Determiniamo la matrice hessiana dal teorema di swartz sappiamo che $H_f(x,y,z)$ è simmetrica: 

$$
H_f(x,y,z) = 
\begin{pmatrix}
    4y^2 - 12x^2y^4 & 8xy - 16x^3y^3 & 0 \\
    8xy - 16x^3y^3 & 2z^2 + 4x^2 -12x^4y^2 & 4yz\\
    0 & 4yz & 2y^2 
\end{pmatrix}
$$

Consideriamo i punti della forma $\left\{
\begin{pmatrix}
    a \\ 0 \\ b
\end{pmatrix} : a,b \in \mathbb{R}
\right\}$ si ha che: 

$$
H_f(a, 0, b) = 
\begin{pmatrix}
    0 & 0 & 0 \\
    0 & 2b^2 + 4a^2 & 0 \\
    0 & 0 & 0 
\end{pmatrix}
$$

Si ha quindi che 

$$
\operatorname{Spec}(H_f(a, 0, b)) = 
\left\{ 
    0, 2b^2 + 4a^2, 0 : a,b \in \mathbb{R}
\right\}
$$

Consideriamo il caso $a,b \in \mathbb{R}\backslash\{0\}$

Si ha che 

$$
\lambda \geq 0 \; \forall \lambda \in \operatorname{Spec}(H_f(a, 0, b))
$$

L'hessiana è quindi semidefinita, i punti della forma $(a, 0, b), \; a,b \in \mathbb{R}\backslash\{0\}$ sono quindi o minimi relativi o selle, poichè $f \in C^k(\mathbb{R})$ sfruttiamo la continuità per determinare la natura dei punti critici

$$
f(x,y, z) + 1 = y^2z^2 + 2y^2x^2 − y^4x^4 = y^2(z^2 + 2x^2 - y^2x^4)
$$

Consideriamo i punti della forma $(x,y,z) = (a, 0, b)$


$$
z^2 + 2x^2 - y^2x^4\bigg|_{(a,0,b)} = b^2 + 2a^2
$$

Se $(a,b)\neq (0,0)$ si ha che

$$
b^2 + 2a^2 > 0 
$$

La funzione 

$$
(x,y,z) \to z^2 + 2x^2 - y^2x^4
$$

E' continua. dunque, per continuità, esiste un intorno $U$ di $(a,0, b)$ tale che in tutto $U$ si abbia 

$$
z^2 + 2x^2 - y^2x^4 > 0
$$

Poiché $y^2 \geq 0$, segue che in $U$ 

$$
f(x,y, z) + 1 = y^2(z^2 + 2x^2 - y^2x^4) \geq 0 \implies f(x,y,z) \geq -1 = f(a,0,b)
$$

i punti della forma 

$$
\left\{
\begin{pmatrix}
a \\ 0 \\ b
\end{pmatrix} : a,b \in \mathbb{R}\backslash\{0\}
\right\}
$$
Sono quindi punti di minimo relativi

Consideriamo ora il caso $a = b = 0$, quindi il punto $(0,0,0)$

$$
H_f(0,0,0) = 0
$$

L'hessiana è contemporaneamente semidefinita positiva e negativa, il criterio è quindi inconcludente

Sfruttiamo analogamente a prima la continuità di $f$ per comprendere la natura del punto critico

$$
f(x,y,z) + 1 = y^2z^2 + y^2x^2 (2-y^2x^2)
$$

In un intorno dell'origine, ad esempio se $|x| \leq 1$ e $|y| \leq 1$, si ha 

$$
y^2x^2 \leq 1 < 2
$$

Quindi 

$$
2 - y^2x^2 > 0
$$

Pertanto 

$$
f(x,y,z) + 1 = y^2z^2 + y^2x^2(2-y^2x^2) \geq 0 \implies f(x,y,z) \geq -1 = f(0,0,0)
$$

Quindi anche l'origine è un minimo relativo

Consideriamo ora i punti della forma $\left\{
\begin{pmatrix}
    \frac{1}{c} \\ c \\ 0
\end{pmatrix} : c \in \mathbb{R}\backslash\{0\}
\right\}$ si ha che 

$$
H_f(\frac{1}{c}, c, 0) = 
\begin{pmatrix}
    -8c^2 & -8 & 0 \\
    -8 & -8\frac{1}{c^2} & 0 \\ 
    0 & 0 & 2c^2 
\end{pmatrix}
$$

Determiniamo il polinomio caratteristico

$$
p_\lambda = \det(H_f(\frac{1}{c}, c, 0)  - \lambda I_3) = (2c^2 - \lambda) \cdot \det\begin{pmatrix}
-8c^2 - \lambda & - 8 \\
-8 & -8\frac{1}{c^2} - \lambda
\end{pmatrix} = (2c^2 - \lambda) \cdot [8c^2\lambda + 8\frac{1}{c^2}\lambda +
\lambda^2] 
$$ 

$$
= (2c^2 - \lambda)\cdot \lambda \cdot (8(c^2 + \frac{1}{c^2}) + \lambda)
$$

Da cui 

$$
\operatorname{spec}(H_f(\frac{1}{c}, c, 0)) = \left\{2c^2, 0, -8(c^2 + \frac{1}{c^2})\right\}, c \in \mathbb{R}\backslash\{0\}
$$

Si ha che 

$$
2c^2 > 0, \; -8(c^2 + \frac{1}{c^2}) < 0, \; \forall c \in \mathbb{R}\backslash\{0\}
$$

L'hessiana è quindi indefinita, i punti della forma $(\frac{1}{c}, c, 0), \; c \in \mathbb{R}\backslash\{0\}$ sono quindi di sella

Consideriamo ora i punti della forma $\left\{
\begin{pmatrix}
    -\frac{1}{c} \\ c \\ 0
\end{pmatrix} : c \in \mathbb{R}\backslash\{0\}
\right\}$ si ha che 

$$
H_f(-\frac{1}{c}, c, 0) = 
\begin{pmatrix}
    -8c^2 & 8 & 0 \\
    8 & -8\frac{1}{c^2} & 0 \\ 
    0 & 0 & 2c^2 
\end{pmatrix}
$$

Analogamente a prima, determiniamo il polinomio caratteristico 

$$
p_\lambda = \det(H_f(-\frac{1}{c}, c, 0)  - \lambda I_3) = (2c^2 - \lambda) \cdot \det\begin{pmatrix}
-8c^2 - \lambda & 8 \\
8 & -8\frac{1}{c^2} - \lambda
\end{pmatrix} = (2c^2 - \lambda) \cdot [8c^2\lambda + 8\frac{1}{c^2}\lambda +
\lambda^2] 
$$

$$
= (2c^2 - \lambda)\cdot \lambda \cdot (8(c^2 + \frac{1}{c^2}) + \lambda)
$$

Da cui 

$$
\operatorname{spec}(H_f(\frac{1}{c}, c, 0)) = \left\{2c^2, 0, -8(c^2 + \frac{1}{c^2})\right\}, c \in \mathbb{R}\backslash\{0\}
$$

Si ha che 

$$
2c^2 > 0, \; -8(c^2 + \frac{1}{c^2}) < 0, \; \forall c \in \mathbb{R}\backslash\{0\}
$$

L'hessiana è quindi indefinita, i punti della forma $(-\frac{1}{c}, c, 0), \; c \in \mathbb{R}\backslash\{0\}$ sono quindi di sella

Consideriamo ora i punti della forma $\left\{
\begin{pmatrix}
    0 \\ d \\ 0
\end{pmatrix} : d \in \mathbb{R}\backslash\{0\}
\right\}$ si ha che 

Si ha che 

$$
H_f(0, d, 0) = 
\begin{pmatrix}
    4d^2 & 0 & 0 \\
    0 & 0 & 0 \\
    0 & 0 & 2d^2
\end{pmatrix}
$$

Da cui 

$$
\operatorname{spec}(H_f(0, d, 0)) = \left\{4d^2, 0, 2d^2\right\}
$$

$$
\lambda \geq 0 \; \forall \lambda \in \operatorname{spec}(H_f(0, d, 0))
$$

$H_f(0, d, 0)$ è semidefinita positiva, i punti della forma $(0, d, 0)$ sono quindi o di minimo relativo o di sella 

Sfruttiamo la continuità di $f$ per determinare la natura dei punti critici 

$$
f(x,y,z) + 1 = y^2z^2 + y^2x^2 (2-y^2x^2) 
$$

Nel punto $(0, d,  0)$ si ha 

$$
x = 0, \quad y = d, \quad z = 0 
$$

Consideriamo la funzione continua 

$$
h(x,y)= 2−y^2x^2
$$

In $(0,d)$ si ha

$$
h(0, d) = 2 > 0
$$

Per continuità, esiste un intorno di $(0, d)$ in cui 
$$
2-y^2x^2 > 0
$$

Quindi, in un intorno di $(0,d,0)$, si ha

$$
f(x,y,z)+1=y^2z^2+y^2x^2(2−y^2x^2) \geq 0
$$

Infatti $x^2y^2 \geq 0$ e $y^2x^2(2-y^2x^2) \geq 0$ perché entrambi i fattori sono non negativi.

allora 

$$
f(x,y, z) \geq -1 = f(0, d, 0)
$$

Da cui, si ha che i punti della forma $(0, d, 0), \; d \in \mathbb{R}\backslash\{0\}$ sono punti di minimo relativo

La funzione cresce e decresce in continuazione, verifichiamo la sua limitatezza:

Considerimo la direzione $(x,y,z) = (0, 1, t)$
$$
f(0,1,t) = t^2 -1 \to +\infty \text{ per } t \to +\infty
$$

$f$ è quindi superiormente illimitata non ammette quindi massimi assoluti

Consideriamo la direzione $(x,y,z) = (1, t, 0)$

$$
f(1, t, 0) = 2y^2 - y^4 - 1 \to -\infty \text{ per } t \to -\infty
$$

$f$ è quindi inferiormente illimitata non ammette quindi  minimi assoluti

### Esercizio 2

- Determinare gli eventuali massimi e minimi relativi e/o assoluti della funzione $f(x,y,z) = y^2z^2 + 2y^2x^2 + y^4x^4 + 1$

- Determinare gli eventuali massimi e minimi relativi e/o assoluti della funzione $g(x,y,z) = \frac{1}{f(x,y,z)}$

### Soluzione

$f \in C^k(\mathbb{R})$ poiché composizione di funzioni elementari

$f$ è superiormente illimitata poichè se considero la direzione $(x,y,z) = (t, 0, 0)$ si ha che 

$$
\lim_{t \to +\infty} t^4 + 2t^4 + t^8 + 1 = +\infty
$$ 

Consideriamo il punto $(0,0,0) \in \mathbb{R}^3$ si ha che 

$$
f(0,0,0) = 1 
$$

Consideriamo quindi  

$$
f(x,y,z) - 1 = y^2z^2 + 2y^2x^2 + y^4x^4 \in C^k(\mathbb{R}^3)
$$

Si ha che 
$$
y^2z^2 + 2y^2x^2 + y^4x^4 \geq 0, \; \forall x,y,z \in \mathbb{R} \implies f(x, y, z) \geq 1 = f(0,0,0) \; \forall x,y,z \in \mathbb{R}
$$

$f$ è quindi inferiormente limitata $(0,0,0) \in \text{Dom}(f) \implies$ $(0,0,0)$ è punto di minimo globale

Cerchiamo eventuali massimi e minimi relativi
