## 1. Limiti di funzioni di più variabili reali

Sia $(X, d)$ uno spazio metrico e sia $f: \text{dom}f \longrightarrow \mathbb{R}$, con $\text{dom}f \subseteq X$ 

$
\lim_{x \to x_0} f(x) = l \in \mathbb{R}
$ se e solo se per ogni intorno $V$ di $l$ esiste un intorno $U$ di $x_0$ tale che $x \in U \cap \text{dom}f$, $x \neq x_0$ implica $f(x) \in V$

se e solo se per ogni $\epsilon > 0$ esiste $\delta > 0$ tale che se $d(x, x_0) < \delta$, $x \in \text{dom}f$, $x \neq x_0$ implica $|f(x) - l| < \epsilon$ se e solo $\lim_{x \to x_0} |f(x) - l| = 0$

Se $X = \mathbb{R}^n$, in particolare $\mathbb{R}^2$ vofliamo studiare qualche tecnica per calcolare limiti di più funzioni di variabili reali

### Esempio:
Consideriamo
$$
\lim_{(x,y) \to (0,0)} \frac{\sin(xy)}{xy}
$$ 
Si ha che 
$$
\text{dom}f := \{(x,y) \in \mathbb{R}^2 : xy \neq 0\} = \{(x,y) \in \mathbb{R}^2 : x \neq 0 \text{ e } y \neq 0\}
$$
Ponendo $xy = t$ si ottiene
$$
\lim_{(x,y) \to (0,0)} \frac{\sin(xy)}{xy} \overset{(t)}{=} \lim_{t \to 0} \frac{\sin(t)}{t} = 1
$$
Utilizzando la definizione di limite devo dimostrare che $\forall \epsilon > 0, \; \exists \delta > 0 : ||(x,y)||_{\infty} < \delta$, $x,y \neq 0$ allora $|\frac{\sin(xy)}{xy} - 1| < \epsilon$
