# Spazi Vettoriali, Normati e Metrici

---

## 1. Spazi vettoriali

### Definizione 1.1 — Spazio vettoriale

Sia $V$ un insieme non vuoto e sia $\mathbb{F}$ un campo. Una struttura di **spazio vettoriale** su $V$ è data da due operazioni:

$$
+ : V \times V \to V,
\qquad
\cdot : \mathbb{F} \times V \to V,
$$

dette rispettivamente **somma** e **prodotto per scalare**, tali che:

1. $(V,+)$ è un gruppo abeliano:
   - **Commutatività:** $x+y=y+x$ per ogni $x,y\in V$;
   - **Associatività:** $(x+y)+z=x+(y+z)$ per ogni $x,y,z\in V$;
   - **Esistenza dell'elemento neutro:** esiste $0_V\in V$ tale che $x+0_V=x$ per ogni $x\in V$;
   - **Esistenza dell'inverso rispetto alla somma:** per ogni $x\in V$ esiste $-x\in V$ tale che $x+(-x)=0_V$.

2. Il prodotto per scalare soddisfa:
   - **Distributività rispetto alla somma:** $\lambda(x+y)=\lambda x+\lambda y$ per ogni $\lambda\in\mathbb{F}$, $x,y\in V$;
   - **Distributività rispetto alla somma di scalari:** $(\lambda+\mu)x=\lambda x+\mu x$ per ogni $\lambda,\mu\in\mathbb{F}$, $x\in V$;
   - **Associatività:** $(\lambda\mu)x=\lambda(\mu x)$ per ogni $\lambda,\mu\in\mathbb{F}$, $x\in V$;
   - **Esistenza dell'elemento neutro:** $1_{\mathbb{F}}x=x$ per ogni $x\in V$.

Gli elementi di $V$ sono detti **vettori**, gli elementi di $\mathbb{F}$ **scalari**.

### Esempio 1.2

Lo spazio $\mathbb{F}^n$ con le operazioni componente per componente è uno spazio vettoriale su $\mathbb{F}$.

### Esempio 1.3

Lo spazio $\mathbb{R}^n$ con le operazioni di somma per componente e prodotto per scalare è uno spazio vettoriale su $\mathbb{R}$.

### Esempio 1.4

Lo spazio $C^0([a,b],\mathbb{R})$ delle funzioni continue $f:[a,b]\to\mathbb{R}$, con somma puntuale e prodotto per scalare puntuale, è uno spazio vettoriale su $\mathbb{R}$.

## 2. Prodotto scalare e norma indotta

### Definizione 2.1 — Prodotto scalare

Sia $V$ uno spazio vettoriale su $\mathbb{F}$. Un **prodotto scalare** o **prodotto interno** su $V$ è una funzione

$$
\langle \cdot,\cdot\rangle : V\times V \to \mathbb{F}
$$

tale che, per ogni $x,y,z\in V$ e ogni $\lambda,\mu\in\mathbb{F}$:

1. **Simmetria hermitiana:**
   $$
   \langle x,y\rangle = \overline{\langle y,x\rangle}.
   $$

   In $\mathbb{R}^n$ coincide con la **simmetria:**
   $$
   \langle x,y\rangle = \langle y,x\rangle.
   $$

2. **Linearità nella prima variabile:**
   $$
   \langle \lambda x+\mu y,z\rangle
   =
   \lambda\langle x,z\rangle+\mu\langle y,z\rangle.
   $$

3. **Positività definita:**
   $$
   \langle x,x\rangle \ge 0,
   \qquad
   \langle x,x\rangle = 0 \iff x=0_V.
   $$

La coppia $(V,\langle\cdot,\cdot\rangle)$ è detta **spazio pre-Hilbertiano**.

> **Osservazione 2.2.**  
> Dalla simmetria hermitiana e dalla linearità nella prima variabile segue la **antilinearità nella seconda variabile**:
> $$
> \langle x,\lambda y+\mu z\rangle
> =
> \overline{\lambda}\langle x,y\rangle+\overline{\mu}\langle x,z\rangle.
> $$
> In $\mathbb{R}^n$, dalla simmetria e dalla linearità nella prima variabile segue la **bilinearità**:
> $$
> \langle x,\lambda y+\mu z\rangle
> =
> \lambda\langle x,y\rangle+\mu\langle x,z\rangle.
> $$

### Definizione 2.3 — Norma indotta da un prodotto scalare

Se $(V,\langle\cdot,\cdot\rangle)$ è uno spazio pre-Hilbertiano, si definisce

$$
\|x\| := \sqrt{\langle x,x\rangle},
\qquad x\in V.
$$

### Teorema 2.4 — Disuguaglianza di Cauchy–Schwarz

Sia $(V,\langle\cdot,\cdot\rangle)$ uno spazio pre-Hilbertiano. Allora, per ogni $x,y\in V$,

$$
|\langle x,y\rangle| \le \|x\|\,\|y\|.
$$

### Proposizione 2.5 — La norma indotta è una norma

Sia $(V,\langle\cdot,\cdot\rangle)$ uno spazio pre-Hilbertiano e sia

$$
\|x\|=\sqrt{\langle x,x\rangle}.
$$

Allora $\|\cdot\|$ è una norma su $V$.

### Esempio 2.6 — Prodotto scalare euclideo

Su $\mathbb{R}^n$,

$$
\langle x,y\rangle = \sum_{i=1}^n x_i y_i
$$

è un prodotto scalare. La norma indotta è $\|x\|_2 = \sqrt{\sum_{i=1}^n x_i^2}$.

### Esempio 2.7 — Prodotto scalare hermitiano

Su $\mathbb{C}^n$,

$$
\langle x,y\rangle = \sum_{i=1}^n x_i \overline{y_i}
$$

è un prodotto scalare. La norma indotta è $\|x\|_2 = \sqrt{\sum_{i=1}^n |x_i|^2}$.

### Esempio 2.8 — Prodotto scalare su $C^0([0,1])$

Su $C^0([0,1],\mathbb{C})$,

$$
\langle f,g\rangle = \int_0^1 f(t)\overline{g(t)}\,dt
$$

è un prodotto scalare. La norma indotta è $\|f\|_2 = \left(\int_0^1 |f(t)|^2\,dt\right)^{1/2}$.

## 3. Spazi normati

### Definizione 3.1 — Norma

Sia $X$ uno spazio vettoriale su $\mathbb{F}$. Una **norma** su $X$ è una funzione

$$
\|\cdot\| : X \to \mathbb{R}
$$

tale che, per ogni $x,y\in X$ e ogni $\lambda\in\mathbb{F}$:

1. $\|x\|\ge 0$ e $\|x\|=0 \iff x=0_X$;
2. $\|\lambda x\|=|\lambda|\,\|x\|$;
3. $\|x+y\|\le \|x\|+\|y\|$ (disuguaglianza triangolare).

La coppia $(X,\|\cdot\|)$ è detta **spazio normato**.

### Esempio 3.2 — Norma euclidea su $\mathbb{R}^n$

Consideriamo lo spazio $\mathbb{R}^n$ e verifichiamo che la funzione

$$
\|x\|_2 = \sqrt{\sum_{i=1}^n x_i^2}
$$

soddisfa gli assiomi di norma.

**Primo assioma** ($\|x\|\ge 0$ e $\|x\|=0 \iff x=0_X$).  
Sia $x = (x_1, \dots, x_n)^T \in \mathbb{R}^n$. Si ha

$$
\|x\|_2 = \sqrt{\sum_{i=1}^n x_i^2} = \sqrt{x_1^2 +\dots + x_n^2}.
$$

Poiché $x_i^2 \geq 0$ per ogni $i$, si ha $\|x\|_2 \geq 0$. Inoltre

$$
\|x\|_2 = 0
\iff
\sqrt{\sum_{i=1}^n x_i^2} = 0
\iff
\sum_{i=1}^n x_i^2 = 0
\iff
x_i^2 = 0 \ \forall i
\iff
x_i = 0 \ \forall i
\iff
x = \vec{0}.
$$

**Secondo assioma** ($\|\lambda x\|=|\lambda|\,\|x\|$).  
Sia $x = (x_1, \dots, x_n)^T \in \mathbb{R}^n$ e sia $\lambda \in \mathbb{R}$. Allora

$$
\|\lambda x\|_2
=
\sqrt{\sum^n_{i= 1} \lambda^2 x_i^2}
=
\sqrt{\lambda^2\sum^n_{i = 1} x_i^2}
=
|\lambda|\sqrt{\sum^n_{i = 1} x_i^2}
=
|\lambda|\cdot\|x\|_2.
$$

**Terzo assioma** ($\|x+y\|\le \|x\|+\|y\|$).  
Siano $x,y \in \mathbb{R}^n$. Usando la disuguaglianza di Cauchy–Schwarz applicata al prodotto scalare euclideo,

$$
\langle x,y\rangle = \sum_{i=1}^n x_i y_i,
$$

otteniamo

$$
\|x+y\|_2^2
=
\sum_{i=1}^n (x_i+y_i)^2
=
\|x\|_2^2 + 2\langle x,y\rangle + \|y\|_2^2
\le
\|x\|_2^2 + 2\|x\|_2\|y\|_2 + \|y\|_2^2
=
(\|x\|_2+\|y\|_2)^2.
$$

Prendendo la radice quadrata si ottiene $\|x+y\|_2 \le \|x\|_2+\|y\|_2$. Dunque $\|\cdot\|_2$ è una norma.

### Esempio 3.3 — Norma $\|\cdot\|_1$ su $\mathbb{R}^n$

Consideriamo la funzione

$$
\|x\|_1 = |x_1| + \dots + |x_n|.
$$

Verifichiamo i tre assiomi.

**Positività definita.**  
$\|x\|_1 = \sum_{i=1}^n |x_i| \ge 0$. Inoltre

$$
\|x\|_1 = 0
\iff
\sum_{i=1}^n |x_i| = 0
\iff
|x_i|=0 \ \forall i
\iff
x_i=0 \ \forall i
\iff
x=0.
$$

**Omogeneità.**  
Per $\lambda\in\mathbb{R}$,

$$
\|\lambda x\|_1 = \sum_{i=1}^n |\lambda x_i| = \sum_{i=1}^n |\lambda|\,|x_i| = |\lambda|\sum_{i=1}^n |x_i| = |\lambda|\,\|x\|_1.
$$

**Disuguaglianza triangolare.**  
Usando la disuguaglianza triangolare in $\mathbb{R}$, $|a+b|\le |a|+|b|$,

$$
\|x+y\|_1
=
\sum_{i=1}^n |x_i+y_i|
\le
\sum_{i=1}^n (|x_i|+|y_i|)
=
\sum_{i=1}^n |x_i| + \sum_{i=1}^n |y_i|
=
\|x\|_1+\|y\|_1.
$$

Dunque $\|\cdot\|_1$ è una norma.

### Esempio 3.4 — Norme $p$ su $\mathbb{F}^n$

Per $p\ge 1$,

$$
\|x\|_p = \left(\sum_{i=1}^n |x_i|^p\right)^{1/p}
$$

è una norma su $\mathbb{F}^n$ (la disuguaglianza triangolare è la disuguaglianza di Minkowski). Per $p=\infty$,

$$
\|x\|_\infty = \max_{1\le i\le n}|x_i|.
$$

### Esempio 3.5 — Norma del sup

Su $C^0([0,1])$,

$$
\|f\|_\infty = \sup_{t\in[0,1]} |f(t)|
$$

è una norma.

**Verifica.**

- **Positività:** $\|f\|_\infty \ge 0$ perché è l'estremo superiore di una funzione non negativa. Inoltre $\|f\|_\infty = 0$ se e solo se $|f(t)|=0$ per ogni $t$, cioè $f\equiv 0$, per la continuità di $f$.
- **Omogeneità:** $\|\lambda f\|_\infty = \sup_t |\lambda f(t)| = |\lambda|\sup_t |f(t)| = |\lambda|\|f\|_\infty$.
- **Triangolare:** $|f(t)+g(t)| \le |f(t)|+|g(t)| \le \|f\|_\infty+\|g\|_\infty$ per ogni $t$, quindi passando al sup, $\|f+g\|_\infty \le \|f\|_\infty+\|g\|_\infty$.

### Esempio 3.6 — Norma integrale su $C^0([a,b])$

Sia $I = [a,b]$ un intervallo chiuso e limitato, e sia $f \in C^0([a,b])$. Definiamo

$$
\|f\|_1 = \int^b_a |f(x)|\,dx.
$$

Verifichiamo i tre assiomi.

**Positività definita.**  
$\|f\|_1 \ge 0$ perché l'integranda è non negativa. Inoltre, se $f$ è continua e $f\neq 0$, allora esiste $x_0\in [a,b]$ con $|f(x_0)|>0$. Per continuità, esiste un intorno $U$ di $x_0$ su cui $|f(x)|>|f(x_0)|/2>0$. Allora

$$
\|f\|_1 = \int_a^b |f(x)|\,dx \ge \int_U |f(x)|\,dx > 0.
$$

Quindi $\|f\|_1 = 0 \iff f=0$.

**Omogeneità.**  
Per $\lambda\in\mathbb{R}$,

$$
\|\lambda f\|_1 = \int_a^b |\lambda f(x)|\,dx = |\lambda|\int_a^b |f(x)|\,dx = |\lambda|\,\|f\|_1.
$$

**Disuguaglianza triangolare.**  
Usando $|f(x)+g(x)| \le |f(x)|+|g(x)|$ e la monotonia dell'integrale,

$$
\|f+g\|_1 = \int_a^b |f(x)+g(x)|\,dx \le \int_a^b (|f(x)|+|g(x)|)\,dx = \|f\|_1+\|g\|_1.
$$

Dunque $\|\cdot\|_1$ è una norma.

### Proposizione 3.7 — Da norma a metrica

Sia $(X,\|\cdot\|)$ uno spazio normato. Ponendo

$$
d(x,y) := \|x-y\|,
\qquad x,y\in X,
$$

si ottiene una metrica su $X$.

> **Osservazione 3.8.**  
> Non ogni metrica su uno spazio vettoriale è indotta da una norma. Per esempio, la metrica discreta
> $$
> d(x,y)=
> \begin{cases}
> 0, & x=y,\\
> 1, & x\neq y
> \end{cases}
> $$
> non è indotta da alcuna norma, perché se fosse $d(x,y)=\|x-y\|$, allora per $x\neq 0$ e $\lambda=2$,
> $$
> d(0,2x)=1
> \quad\text{ma}\quad
> \|2x\|=2\|x\|=2d(0,x)=2,
> $$
> contraddizione.

## 4. Spazi metrici

### Definizione 4.1 — Metrica

Sia $X$ un insieme non vuoto. Una **metrica** o **distanza** su $X$ è una funzione

$$
d : X\times X \to \mathbb{R}
$$

tale che, per ogni $x,y,z\in X$:

1. $d(x,y)\ge 0$ e $d(x,y)=0 \iff x=y$;
2. $d(x,y)=d(y,x)$;
3. $d(x,y)\le d(x,z)+d(z,y)$.

La coppia $(X,d)$ è detta **spazio metrico**.

### Esempio 4.2 — Metrica euclidea

Su $\mathbb{R}^n$,

$$
d_2(x,y) = \|x-y\|_2 = \sqrt{\sum_{i=1}^n (x_i-y_i)^2}
$$

è la metrica euclidea.

### Esempio 4.3 — Metrica $p$-esima con $0<p<1$

Sia $0<p<1$ e sia $d:\mathbb{R}\times\mathbb{R}\to\mathbb{R}$ definita da

$$
d(x,y) = |x-y|^p.
$$

Allora $d$ è una metrica su $\mathbb{R}$.

**Dimostrazione.**

- **Positività:** $d(x,y)=|x-y|^p\ge 0$ e $d(x,y)=0 \iff |x-y|=0 \iff x=y$.
- **Simmetria:** $d(x,y)=|x-y|^p=|y-x|^p=d(y,x)$.
- **Disuguaglianza triangolare:** dobbiamo mostrare che
  $$
  |x-y|^p \le |x-z|^p + |z-y|^p.
  $$
  Siano $a = |x-z|$ e $b = |z-y|$. Allora $|x-y| \le a+b$, quindi basta dimostrare che per $a,b\ge 0$ e $0<p<1$ vale
  $$
  (a+b)^p \le a^p + b^p.
  $$
  Consideriamo la funzione $g(t) = (1+t)^p - 1 - t^p$ per $t\ge 0$. Allora $g(0)=0$ e
  $$
  g'(t) = p(1+t)^{p-1} - p t^{p-1} = p\left[(1+t)^{p-1} - t^{p-1}\right].
  $$
  Poiché $0<p<1$, si ha $p-1<0$, quindi $t\mapsto t^{p-1}$ è decrescente su $(0,\infty)$. Dunque $(1+t)^{p-1} < t^{p-1}$ per $t>0$, cioè $g'(t) < 0$. Quindi $g$ è decrescente e $g(t)\le g(0)=0$ per ogni $t\ge 0$. Pertanto $(1+t)^p \le 1 + t^p$. Se $a>0$, poniamo $t=b/a$ e moltiplichiamo per $a^p$: $(a+b)^p \le a^p + b^p$. Se $a=0$, la disuguaglianza è banale.

  Di conseguenza
  $$
  d(x,y) = |x-y|^p \le (|x-z|+|z-y|)^p \le |x-z|^p + |z-y|^p = d(x,z)+d(z,y).
  $$

$\square$

### Esempio 4.4 — Metrica discreta

Su un insieme non vuoto $X$,

$$
d(x,y)=
\begin{cases}
0, & x=y,\\
1, & x\neq y.
\end{cases}
$$

**Verifica.** La positività e la simmetria sono immediate. Per la disuguaglianza triangolare: se $x=y$ è banale; se $x\neq y$ allora $d(x,y)=1$ e almeno uno tra $d(x,z)$ e $d(z,y)$ è $1$ (altrimenti $x=z=y$), quindi $d(x,y)\le d(x,z)+d(z,y)$.

### Definizione 4.5 — Palla aperta

Sia $(X,d)$ uno spazio metrico. La **palla aperta** di centro $x_0\in X$ e raggio $r>0$ è

$$
\mathcal{B}_r(x_0) = \{x\in X : d(x,x_0)<r\}.
$$

### Esempio 4.7 — Palle in $C^0([0,1])$

$$
\mathcal{B}_r(f) = \left\{g \in C^0([0,1]) : \|f-g\|_\infty < r\right\}
= \left\{g \in C^0([0,1]) : \sup_{x \in [0,1]} |f(x) - g(x)| < r\right\}.
$$

La palla è l'insieme delle funzioni continue il cui grafico resta in una striscia verticale di semi-ampiezza $r$ attorno al grafico di $f$.

## 5. Topologia degli spazi metrici

### Definizione 5.1 — Intorno

Sia $(X,d)$ uno spazio metrico e sia $x_0\in X$. Un **intorno** di $x_0$ è un sottoinsieme $U\subseteq X$ tale che esiste $r>0$ con

$$
\mathcal{B}_r(x_0)\subseteq U.
$$

In altre parole un intorno di $x_0$ è un sottoinsieme $U$ di $X$ che contiene una palla (aperta) di centro $x_0$.

### Definizione 5.2 — Insieme aperto

Sia $(X,d)$ uno spazio metrico. Un insieme $A\subseteq X$ è **aperto** se è intorno di ogni suo punto, cioè se

$$
\forall x_0\in A\ \exists r>0 : \mathcal{B}_r(x_0)\subseteq A.
$$

### Definizione 5.3 — Insieme chiuso

Un insieme $C\subseteq X$ è **chiuso** se il suo complementare $C^C = X\setminus C$ è aperto.

> **Osservazione 5.4.**  
> Non è vero che un insieme che non è aperto è necessariamente chiuso, e viceversa. Per esempio, in $\mathbb{R}$ con la metrica euclidea, l'insieme $I = (0,1]$ non è aperto, perché non esiste una palla centrata in $1$ contenuta in $I$. Non è nemmeno chiuso, perché $I^C = (-\infty,0]\cup(1,+\infty)$ non è aperto: non esiste una palla centrata in $0$ contenuta in $I^C$. Quindi $I$ non è né aperto né chiuso.

> **Osservazione 5.5.**  
> Ogni palla aperta $\mathcal{B}_r(x)$ è un insieme aperto.

**Dimostrazione.**  
Sia $y\in\mathcal{B}_r(x)$, cioè $d(y,x)<r$. Posto $s = r-d(y,x)>0$, per ogni $z\in\mathcal{B}_s(y)$ vale

$$
d(z,x) \le d(z,y)+d(y,x) < s + d(y,x) = r,
$$

quindi $\mathcal{B}_s(y)\subseteq\mathcal{B}_r(x)$. Dunque $\mathcal{B}_r(x)$ è intorno di ogni suo punto. $\square$

> **Osservazione 5.6.**  
> In ogni spazio metrico, $X$ e $\emptyset$ sono sia aperti sia chiusi. Infatti $X$ è aperto perché contiene ogni palla; $\emptyset$ è aperto perché non ha punti. Quindi $\emptyset = X^C$ è chiuso e $X = \emptyset^C$ è chiuso.

### Teorema 5.7 — Unione e intersezione di aperti

Sia $(X,d)$ uno spazio metrico.

1. Se $\{A_i\}_{i\in I}$ è una famiglia qualsiasi di aperti, allora $\bigcup_{i\in I}A_i$ è aperto.
2. Se $A_1,\dots,A_k$ sono aperti, allora $\bigcap_{i=1}^k A_i$ è aperto.

**Dimostrazione.**

1. Sia $x_0\in\bigcup_{i\in I}A_i$. Allora esiste $j\in I$ tale che $x_0\in A_j$. Poiché $A_j$ è aperto, esiste $r>0$ con $\mathcal{B}_r(x_0)\subseteq A_j$. Ma $A_j\subseteq\bigcup_{i\in I}A_i$, quindi $\mathcal{B}_r(x_0)\subseteq\bigcup_{i\in I}A_i$. Dunque l'unione è aperta.

2. Sia $x_0\in\bigcap_{i=1}^k A_i$. Allora $x_0\in A_i$ per ogni $i$. Poiché ogni $A_i$ è aperto, per ogni $i$ esiste $r_i>0$ con $\mathcal{B}_{r_i}(x_0)\subseteq A_i$. Posto $r = \min\{r_1,\dots,r_k\}>0$, si ha $\mathcal{B}_r(x_0)\subseteq \mathcal{B}_{r_i}(x_0)\subseteq A_i$ per ogni $i$, quindi $\mathcal{B}_r(x_0)\subseteq \bigcap_{i=1}^k A_i$. $\square$

> **Osservazione 5.8.**  
> L'intersezione arbitraria di aperti non è necessariamente aperta. Per esempio, in $\mathbb{R}$,
> $$
> A_n = \left(-\frac{1}{n},\frac{1}{n}\right),
> \qquad n\ge 1.
> $$
> Allora $\bigcap_{n=1}^\infty A_n = \{0\}$, che non è aperto.

### Corollario 5.9 — Unione e intersezione di chiusi

Sia $(X,d)$ uno spazio metrico.

1. L'unione finita di chiusi è chiusa.
2. L'intersezione arbitraria di chiusi è chiusa.

**Dimostrazione.**

1. Siano $C_1,\dots,C_k$ chiusi. Allora
   $$
   \left(\bigcup_{i=1}^k C_i\right)^C = \bigcap_{i=1}^k C_i^C.
   $$
   Poiché ogni $C_i^C$ è aperto, per il Teorema 5.7 l'intersezione finita è aperta. Dunque l'unione è chiusa.

2. Sia $\{C_i\}_{i\in I}$ una famiglia di chiusi. Allora
   $$
   \left(\bigcap_{i\in I} C_i\right)^C = \bigcup_{i\in I} C_i^C.
   $$
   Ogni $C_i^C$ è aperto, quindi l'unione arbitraria è aperta per il Teorema 5.7. Dunque l'intersezione è chiusa. $\square$

> **Osservazione 5.10.**  
> L'unione arbitraria di chiusi non è necessariamente chiusa. Per esempio, in $\mathbb{R}$,
> $$
> C_n = \left[-1+\frac{1}{n},\,1-\frac{1}{n}\right],
> \qquad n\ge 1.
> $$
> Allora $\bigcup_{n=1}^\infty C_n = (-1,1)$, che è aperto e non chiuso.

### Definizione 5.11 — Insieme limitato

Sia $(X, d)$ uno spazio metrico. Un insieme $C \subseteq X$ è **limitato** se esistono $x \in X$ e $R > 0$ tali che $C \subseteq \mathcal{B}_R(x)$, cioè

$$
d(z,x) \leq R \qquad \forall z \in C.
$$

> **Osservazione 5.12.**  
> In uno spazio normato $(X, \|\cdot\|)$, il punto $x$ può essere preso come lo zero: $C \subseteq X$ è limitato se e solo se esiste $R > 0$ tale che $C \subseteq \mathcal{B}_R(0)$, cioè $\|z\| \leq R$ per ogni $z \in C$.

## 6. Punti interni, esterni, di accumulazione, isolati, chiusura e frontiera

### Definizione 6.1 — Punto interno

Sia $(X,d)$ uno spazio metrico e sia $A\subseteq X$. Un punto $x_0\in X$ è **interno** ad $A$ se esiste $r>0$ tale che

$$
\mathcal{B}_r(x_0)\subseteq A.
$$

L'insieme dei punti interni di $A$ è detto **interno** di $A$ e si indica con $\mathring{A}$.

> **Osservazione 6.2.**  
> $x_0$ è interno ad $A$ se e solo se esiste $r>0$ tale che
> $$
> \mathcal{B}_r(x_0)\cap A = \mathcal{B}_r(x_0).
> $$
> Infatti $B\cap A = B$ se e solo se $B\subseteq A$. In particolare, se $x_0$ è interno ad $A$, allora $x_0\in A$.

### Esempio 6.3

In $\mathbb{R}$ con la metrica euclidea, se $I=(0,1]$, allora $\mathring{I} = (0,1)$.

### Proposizione 6.4

$\mathring{A}$ è il più grande (nel senso dell'inclusione) aperto contenuto in $A$: se $B \subseteq A$ è aperto, allora $B \subseteq \mathring{A}$.

**Dimostrazione.**  
Se $x\in B$ e $B$ è aperto, allora esiste $r>0$ con $\mathcal{B}_r(x)\subseteq B\subseteq A$. Quindi $x\in\mathring{A}$. Dunque $B\subseteq\mathring{A}$.

Mostriamo che $\mathring{A}$ è aperto. Sia $x\in\mathring{A}$. Allora esiste $r>0$ con $\mathcal{B}_r(x)\subseteq A$. Per ogni $y\in\mathcal{B}_r(x)$, posto $s = r-d(x,y)>0$, si ha $\mathcal{B}_s(y)\subseteq\mathcal{B}_r(x)\subseteq A$, quindi $y\in\mathring{A}$. Dunque $\mathcal{B}_r(x)\subseteq\mathring{A}$. Questo prova che $\mathring{A}$ è aperto. $\square$

### Definizione 6.5 — Punto esterno

Sia $(X,d)$ uno spazio metrico e sia $A\subseteq X$. Un punto $x_0\in X$ è **esterno** ad $A$ se è interno ad $A^C$, cioè se esiste $r>0$ con

$$
\mathcal{B}_r(x_0)\subseteq A^C.
$$

> **Osservazione 6.6.**  
> - Se $x_0$ è interno ad $A$, allora $x_0\in A$. Non vale il viceversa: $x_0\in A$ non implica che $x_0$ sia interno ad $A$.
> - Se $x_0$ è esterno ad $A$, allora $x_0\in A^C$. Non vale il viceversa: $x_0\in A^C$ non implica che $x_0$ sia esterno ad $A$.
> - Un punto non può essere simultaneamente interno ed esterno ad $A$.

### Esempio 6.7

In $\mathbb{R}$, sia $A=(0,1]$. Il punto $2$ è esterno ad $A$, perché $\mathcal{B}_{1/2}(2)\subseteq A^C$. Il punto $0$ appartiene ad $A^C$ ma non è esterno ad $A$: ogni palla centrata in $0$ contiene punti di $A$.

### Definizione 6.8 — Punto di accumulazione

Sia $(X,d)$ uno spazio metrico e sia $A\subseteq X$. Un punto $x_0\in X$ è **di accumulazione** per $A$ se

$$
\forall r>0,\qquad
\left(\mathcal{B}_r(x_0)\setminus\{x_0\}\right)\cap A \neq \emptyset.
$$

Equivalentemente, ogni palla centrata in $x_0$ contiene almeno un punto di $A$ diverso da $x_0$. L'insieme dei punti di accumulazione di $A$ si indica con $A'$ ed è detto **insieme derivato**.

> **Osservazione 6.9.**  
> $x_0$ può essere di accumulazione senza appartenere ad $A$. Inoltre, in uno spazio metrico, se ogni palla centrata in $x_0$ contiene un punto di $A\setminus\{x_0\}$, allora ne contiene infiniti.

### Esempio 6.10

- In $\mathbb{R}$, sia $A=(0,1)$. Allora $0$ e $1$ sono di accumulazione per $A$, ma non appartengono ad $A$.
- In $\mathbb{R}$, sia $A=\left\{\frac{1}{n}:n\in\mathbb{N}\right\}\cup\{0\}$. Allora $0$ è l'unico punto di accumulazione.
- In $\mathbb{R}^2$, sia $A=\left\{\left(\frac{1}{n},0\right):n\in\mathbb{N}\right\}$. Allora $(0,0)$ è l'unico punto di accumulazione; il punto $(2,0)$ non è di accumulazione, perché $\mathcal{B}_{1/2}(2,0)\cap A=\emptyset$.

### Definizione 6.11 — Punto di chiusura

Sia $A\subseteq X$. Un punto $x_0\in X$ è **punto di chiusura** per $A$ se per ogni $r>0$,

$$
\mathcal{B}_r(x_0)\cap A\neq\emptyset.
$$

Cioè se ogni palla di centro $x_0$ interseca $A$. L'insieme dei punti di chiusura di $A$ è detto **chiusura** di $A$ e si indica con $\overline{A}$.

> **Osservazione 6.12.**  
> Ogni punto di accumulazione è un punto di chiusura, ma non viceversa: un punto isolato di $A$ è punto di chiusura ma non di accumulazione.

### Esempio 6.13

In $\mathbb{R}$, se $I=(0,1]$, allora $\overline{I}=[0,1]$.

### Proposizione 6.14

$\overline{A}$ è il più piccolo chiuso contenente $A$. In altre parole, se $C\supseteq A$ è chiuso, allora $\overline{A}\subseteq C$.

**Dimostrazione.**  
Sia $C$ chiuso con $A\subseteq C$. Se $x\notin C$, poiché $C^C$ è aperto, esiste $r>0$ con $\mathcal{B}_r(x)\subseteq C^C$. Allora $\mathcal{B}_r(x)\cap C=\emptyset$, quindi a maggior ragione $\mathcal{B}_r(x)\cap A=\emptyset$. Dunque $x\notin\overline{A}$. Quindi $\overline{A}\subseteq C$. $\square$

### Definizione 6.15 — Punto di frontiera

Sia $A\subseteq X$. Un punto $x_0\in X$ è **di frontiera** per $A$ se è punto di chiusura sia per $A$ sia per $A^C$, cioè se per ogni $r>0$,

$$
\mathcal{B}_r(x_0)\cap A\neq\emptyset
\qquad\text{e}\qquad
\mathcal{B}_r(x_0)\cap A^C\neq\emptyset.
$$

L'insieme dei punti di frontiera di $A$ è detto **frontiera** di $A$ e si indica con $\partial A$.

### Esempio 6.16

In $\mathbb{R}$, se $I=(0,1]$, allora $\partial I=\{0,1\}$.

### Definizione 6.17 — Punto isolato

Sia $A\subseteq X$. Un punto $x_0\in A$ è **isolato** in $A$ se esiste $r>0$ tale che

$$
\mathcal{B}_r(x_0)\cap A = \{x_0\}.
$$

### Esempio 6.18

- In $\mathbb{R}$, l'insieme $A=\mathbb{Z}$ ha tutti i punti isolati.
- In $\mathbb{R}^2$, l'insieme $A=\{(0,0),(1,0),(1,1)\}$ ha tutti i punti isolati.
- In $\mathbb{R}$, l'insieme $A=\left\{\frac{1}{n}:n\in\mathbb{N}\right\}$ ha tutti i punti isolati, mentre $0$ è di accumulazione ma non appartiene ad $A$.

### Osservazione 6.19 — Relazioni tra i tipi di punto

Sia $A\subseteq X$ e $x_0\in X$. Allora:

1. $x_0$ interno ad $A$ $\implies$ $x_0\in A$ e $x_0$ non è di frontiera per $A$.
2. $x_0$ esterno ad $A$ $\implies$ $x_0\in A^C$ e $x_0$ non è di frontiera per $A$.
3. $x_0$ isolato in $A$ $\implies$ $x_0\in A$ e $x_0$ non è di accumulazione per $A$.
4. $x_0$ di accumulazione per $A$ $\implies$ $x_0$ è punto di chiusura per $A$, ma $x_0$ può non appartenere ad $A$.
5. $x_0$ di frontiera per $A$ $\implies$ $x_0$ è punto di chiusura per $A$ e per $A^C$, ma $x_0$ può appartenere o no ad $A$.
6. Un punto è interno, esterno o di frontiera per $A$, e le tre categorie sono mutuamente esclusive e exhaustive in $X$.

### Osservazione 6.20 — Relazioni globali

Valgono:

$$
\mathring{A}\subseteq A\subseteq \overline{A},
$$

$$
A \text{ è aperto } \iff A=\mathring{A},
$$

$$
A \text{ è chiuso } \iff A=\overline{A},
$$

$$
\partial A = \overline{A}\cap \overline{A^C},
$$

$$
\overline{A} = A\cup \partial A = \mathring{A}\cup \partial A.
$$

### Esempio 6.21 — Classificazione dei punti in $\mathbb{R}^2$

Sia

$$
A = \{(x,y)\in\mathbb{R}^2 : x^2+y^2<1\}\cup\{(2,0)\}.
$$

Classificazione:

- $(0,0)$ è interno ad $A$.
- $(1,0)$ è di frontiera per $A$, non appartiene ad $A$.
- $(2,0)$ è isolato in $A$, non è di accumulazione, ed è di frontiera per $A$.
- $(3,0)$ è esterno ad $A$.

## 7. Successioni in spazi metrici

### Definizione 7.1 — Successione

Sia $(X,d)$ uno spazio metrico. Una **successione** in $X$ è una funzione

$$
x : \mathbb{N}\to X,
\qquad
k\mapsto x_k.
$$

Si indica con $\{x_k\}_{k\in\mathbb{N}}\subseteq X$.

### Esempio 7.2

Sia $X = \mathbb{R}^3$ e consideriamo

$$
(x_k, y_k, z_k) = \left(\frac{\sin k}{k}, 1+e^{-k}, k\right).
$$

Allora $\{(x_k, y_k, z_k)\}_{k \geq 1}$ è una successione di $\mathbb{R}^3$.

### Esempio 7.3

Sia $\{f_k\}_{k \in \mathbb{N}} \subseteq C^0([0,1])$ con $f_k(x) = x^k$. Allora $\{f_k\}_{k \in \mathbb{N}}$ è una successione nello spazio metrico $(C^0([0,1]),\|\cdot\|_\infty)$.

### Definizione 7.4 — Convergenza

Si dice che $\{x_k\}_{k\in\mathbb{N}}$ **converge** a $l\in X$, e si scrive

$$
\lim_{k\to+\infty}x_k=l,
$$

se

$$
\forall\varepsilon>0\ \exists N>0 : k>N \implies d(x_k,l)<\varepsilon.
$$

Equivalentemente, $\lim_{k\to+\infty}d(x_k,l)=0$.

### Osservazione 7.5

Sia $\{x_k\}_{k \in \mathbb{N}} \subseteq \mathbb{R}$. Allora

$$
\lim_{k \to +\infty} x_k = l \in \mathbb{R}
\iff
\forall \epsilon > 0 \ \exists N > 0 : k > N \implies |x_k - l| < \epsilon.
$$

Equivalentemente,

$$
\lim_{k \to +\infty} x_k = l
\iff
\lim_{k \to +\infty} |x_k - l| = 0,
$$

dove $|x_k - l|$ rappresenta la distanza su $\mathbb{R}$.

### Osservazione 7.6

Se $(X, \|\cdot\|)$ è uno spazio normato,

$$
\lim_{k \to +\infty} x_k = l \in X
\iff
\lim_{k \to +\infty} \|x_k - l\| = 0.
$$

### Teorema 7.7 — Unicità del limite

Sia $(X,d)$ uno spazio metrico e sia $\{x_k\}_{k\in\mathbb{N}}\subseteq X$. Se

$$
\lim_{k\to+\infty}x_k=l_1 \in X
\qquad\text{e}\qquad
\lim_{k\to+\infty}x_k=l_2 \in X,
$$

allora $l_1=l_2$.

**Dimostrazione.**  
Supponiamo $l_1\neq l_2$. Allora $d(l_1,l_2)>0$. Sia $\varepsilon = d(l_1,l_2)/3>0$. Poiché $x_k\to l_1$, esiste $N_1$ tale che per $k>N_1$, $d(x_k,l_1)<\varepsilon$. Poiché $x_k\to l_2$, esiste $N_2$ tale che per $k>N_2$, $d(x_k,l_2)<\varepsilon$. Per $k>\max\{N_1,N_2\}$, usando la disuguaglianza triangolare,

$$
d(l_1,l_2)
\le
d(l_1,x_k)+d(x_k,l_2)
<
2\varepsilon
=
\frac{2}{3}d(l_1,l_2),
$$

contraddizione. Quindi $l_1=l_2$. $\square$

### Proposizione 7.8 — Convergenza in $\mathbb{R}^n$

Sia $\{\vec{x}_k\}_{k\in\mathbb{N}}\subseteq\mathbb{R}^n$, con $\vec{x}_k=(x_{k1},\dots,x_{kn})$. Allora $\vec{x}_k\to \vec{l}=(l_1,\dots,l_n)$ se e solo se

$$
x_{kj}\to l_j \qquad \forall j=1,\dots,n.
$$

**Dimostrazione.**  
Usiamo le disuguaglianze

$$
\|\vec{x}\|_\infty \le \|\vec{x}\|_2 \le \|\vec{x}\|_1 \le n\|\vec{x}\|_\infty.
$$

Se $\vec{x}_k\to\vec{l}$, allora per ogni $j$,

$$
|x_{kj}-l_j|
\le
\|\vec{x}_k-\vec{l}\|_\infty
\le
\|\vec{x}_k-\vec{l}\|_2
\to 0.
$$

Viceversa, se $x_{kj}\to l_j$ per ogni $j$, allora

$$
\|\vec{x}_k-\vec{l}\|_\infty = \max_j |x_{kj}-l_j| \to 0,
$$

e poiché $\|\vec{x}_k-\vec{l}\|_2\le n\|\vec{x}_k-\vec{l}\|_\infty$, anche $\vec{x}_k\to\vec{l}$ in norma euclidea. $\square$

### Esempio 7.9

Consideriamo ancora $(x_k, y_k, z_k) = \left(\frac{\sin k}{k}, 1+e^{-k}, k\right)$ e studiamo la convergenza o la divergenza di $\{(x_k, y_k, z_k)\}_{k \geq 1}$.

Dalla proposizione precedente:

$$
\lim_{k \to +\infty} x_k = \lim_{k \to +\infty} \frac{\sin k}{k} = 0,
$$

$$
\lim_{k \to +\infty} y_k = \lim_{k \to +\infty} 1+e^{-k} = 1,
$$

$$
\lim_{k \to +\infty} z_k = \lim_{k \to +\infty} k = +\infty.
$$

Da cui la successione diverge:

$$
\lim_{k \to +\infty} (x_k, y_k, z_k) = +\infty,
$$

cioè

$$
\lim_{k \to +\infty} \|(x_k, y_k, z_k)\|_2 = +\infty,
$$

o equivalentemente

$$
\lim_{k \to +\infty} d((x_k, y_k, z_k), (0,0,0)) = +\infty.
$$

### Esempio 7.10

Sia $(x_k, y_k) = (k, 1+e^{-k}) \in \mathbb{R}^2$. Analogamente a prima,

$$
\lim_{k \to +\infty} x_k = \lim_{k \to +\infty} k = +\infty,
\qquad
\lim_{k \to +\infty} y_k = \lim_{k \to +\infty} 1+e^{-k} = 1.
$$

Da cui la successione diverge, cioè

$$
\lim_{k \to +\infty} \|(x_k, y_k)\|_2 = +\infty,
$$

o equivalentemente

$$
\lim_{k \to +\infty} d((x_k, y_k), (0,0)) = +\infty.
$$

### Esempio 7.11

Sia $(x_k, y_k, z_k) = \left(k^2\left(1-\cos \frac{1}{k}\right), e^{-k}, \arctan k\right)$. Studiamo la convergenza della successione $\{(x_k, y_k, z_k)\}_{k \geq 1}$.

$$
\lim_{k \to +\infty} x_k = \lim_{k \to +\infty} k^2\left(1-\cos \frac{1}{k}\right) = \frac{1}{2},
$$

$$
\lim_{k \to +\infty} y_k = \lim_{k \to +\infty} e^{-k} = 0,
$$

$$
\lim_{k \to +\infty} z_k = \lim_{k \to +\infty} \arctan k = \frac{\pi}{2}.
$$

La successione converge a $\left(\frac{1}{2}, 0, \frac{\pi}{2}\right)$.

### Teorema 7.12 — Caratterizzazione della chiusura tramite successioni

Sia $(X,d)$ uno spazio metrico e sia $A\subseteq X$. Allora $x\in\overline{A}$ se e solo se esiste una successione $\{x_k\}_{k\in\mathbb{N}}\subseteq A$ tale che $x_k\to x$.

**Dimostrazione.**  
Se $x\in\overline{A}$, per ogni $k\in\mathbb{N}$ la palla $\mathcal{B}_{1/k}(x)$ interseca $A$. Scegliamo $x_k\in \mathcal{B}_{1/k}(x)\cap A$. Allora $d(x_k,x)<1/k\to 0$, quindi $x_k\to x$.

Viceversa, se $\{x_k\}\subseteq A$ e $x_k\to x$, allora per ogni $r>0$ esiste $N$ tale che per $k>N$, $d(x_k,x)<r$. Dunque $x_k\in\mathcal{B}_r(x)\cap A$, quindi $\mathcal{B}_r(x)\cap A\neq\emptyset$. Pertanto $x\in\overline{A}$. $\square$

### Osservazione 7.13

Il teorema appena enunciato può essere riformulato nel seguente modo:

Sia $(X, d)$ uno spazio metrico e sia $A \subseteq X$. Allora $x \in \overline{A}$ se e solo se esiste una successione di elementi di $A$ che converge ad $x$.

### Corollario 7.14 — Caratterizzazione dei chiusi tramite successioni

Sia $(X,d)$ uno spazio metrico e sia $A\subseteq X$. Allora $A$ è chiuso se e solo se ogni successione convergente di elementi di $A$ ha limite in $A$.

**Dimostrazione.**  
Se $A$ è chiuso, allora $A=\overline{A}$. Se $\{x_k\}\subseteq A$ e $x_k\to x$, per il Teorema 7.12, $x\in\overline{A}=A$.

Viceversa, supponiamo che ogni successione convergente in $A$ abbia limite in $A$. Se $x\in\overline{A}$, per il Teorema 7.12 esiste $\{x_k\}\subseteq A$ con $x_k\to x$. Per ipotesi $x\in A$. Quindi $\overline{A}\subseteq A$, e poiché $A\subseteq\overline{A}$, si ha $A=\overline{A}$. Dunque $A$ è chiuso. $\square$

### Proposizione 7.15 — Le successioni convergenti sono limitate

Sia $(X,d)$ uno spazio metrico e sia $\{x_k\}_{k\in\mathbb{N}}\subseteq X$ una successione convergente a $l\in X$. Allora $\{x_k\}$ è limitata, cioè esistono $C>0$ e $x\in X$ tali che $d(x_k,x)<C$ per ogni $k$.

**Dimostrazione.**  
Poiché $x_k\to l$, per $\varepsilon=1$ esiste $N$ tale che per $k>N$, $d(x_k,l)<1$. Sia

$$
C = \max\{1, d(x_1,l),\dots,d(x_N,l)\}+1.
$$

Allora per ogni $k$, $d(x_k,l)<C$. Dunque la successione è contenuta nella palla $\mathcal{B}_C(l)$. $\square$

### Osservazione 7.16

La condizione "esistono $C>0$ e $x\in X$ tali che $d(x_k,x)<C$ per ogni $k$" può essere riformulata come $x_k \in \mathcal{B}_C(x)$ per ogni $k \in \mathbb{N}$.

## 8. Equivalenza di norme in $\mathbb{R}^n$

### Definizione 8.1 — Norme equivalenti

Sia $X$ uno spazio vettoriale. Due norme $\|\cdot\|_a$ e $\|\cdot\|_b$ su $X$ sono **equivalenti** se esistono costanti $c,C>0$ tali che

$$
c\|x\|_a \le \|x\|_b \le C\|x\|_a
\qquad \forall x\in X.
$$

### Proposizione 8.2 — Equivalenza delle norme $p$ in $\mathbb{R}^n$

Su $\mathbb{R}^n$, per ogni $x\in\mathbb{R}^n$,

$$
\|x\|_\infty \le \|x\|_2 \le \|x\|_1 \le n\|x\|_\infty.
$$

Di conseguenza, le norme $\|\cdot\|_1$, $\|\cdot\|_2$, $\|\cdot\|_\infty$ sono equivalenti.

**Dimostrazione.**  
Per ogni $i$, $|x_i|\le \|x\|_\infty$. Quindi

$$
\|x\|_2^2=\sum_{i=1}^n |x_i|^2 \le n\|x\|_\infty^2.
$$

D'altra parte $\|x\|_\infty \le \|x\|_2$ perché il massimo dei moduli è uno dei moduli. Inoltre,

$$
\|x\|_2^2=\sum_{i=1}^n |x_i|^2 \le \left(\sum_{i=1}^n |x_i|\right)^2 = \|x\|_1^2,
$$

quindi $\|x\|_2\le \|x\|_1$. Infine $\|x\|_1=\sum_{i=1}^n |x_i|\le n\|x\|_\infty$. $\square$

> **Osservazione 8.3.**  
> Poiché le norme sono equivalenti, un insieme è aperto rispetto a $\|\cdot\|_2$ se e solo se lo è rispetto a $\|\cdot\|_\infty$ se e solo se lo è rispetto a $\|\cdot\|_1$. Analogamente, una successione converge in una norma se e solo se converge nelle altre.

## 9. Limiti di funzioni tra spazi metrici

### 9.1 Punti di accumulazione e domini

Siano $(X,d_X)$ e $(Y,d_Y)$ spazi metrici, sia $f:\operatorname{dom}f\to Y$ con $\operatorname{dom}f\subseteq X$, e sia $x_0\in X$ un punto di accumulazione per $\operatorname{dom}f$.

### Definizione 9.1 — Limite di funzione

Si dice che $f$ ha **limite** $l\in Y$ per $x\to x_0$, e si scrive

$$
\lim_{x\to x_0} f(x) = l,
$$

se per ogni intorno $V$ di $l$ in $Y$ esiste un intorno $U$ di $x_0$ in $X$ tale che

$$
x\in U\cap \operatorname{dom}f,\ x\neq x_0
\implies
f(x)\in V.
$$

Equivalentemente, in termini di palle: per ogni $\varepsilon>0$ esiste $\delta>0$ tale che

$$
x\in \operatorname{dom}f,\quad
0<d_X(x,x_0)<\delta
\implies
d_Y(f(x),l)<\varepsilon.
$$

> **Osservazione 9.2.**  
> La condizione $x\neq x_0$ è essenziale: il valore $f(x_0)$ non influenza il limite, e $f$ può non essere definita in $x_0$. La richiesta che $x_0$ sia di accumulazione per $\operatorname{dom}f$ garantisce che esistano punti del dominio arbitrariamente vicini a $x_0$, rendendo la definizione non vuota.

### Esempio 9.3

Sia $f:\mathbb{R}\setminus\{0\}\to\mathbb{R}$, $f(x)=x\sin(1/x)$. Allora

$$
\lim_{x\to 0} x\sin(1/x)=0,
$$

perché $|x\sin(1/x)|\le |x|\to 0$.

### Esempio 9.4 — Limite che non esiste

Sia $f:\mathbb{R}\setminus\{0\}\to\mathbb{R}$, $f(x)=\sin(1/x)$. Il limite per $x\to 0$ non esiste: per ogni $l\in[-1,1]$ esiste una successione $x_k\to 0$ con $f(x_k)\to l$.

### Teorema 9.5 — Unicità del limite

Siano $(X,d_X)$ e $(Y,d_Y)$ spazi metrici, $f:\operatorname{dom}f\to Y$, e sia $x_0$ di accumulazione per $\operatorname{dom}f$. Se

$$
\lim_{x\to x_0} f(x)=l_1
\qquad\text{e}\qquad
\lim_{x\to x_0} f(x)=l_2,
$$

allora $l_1=l_2$.

**Dimostrazione.**  
Supponiamo $l_1\neq l_2$ e sia $\varepsilon = d_Y(l_1,l_2)/3>0$. Per definizione di limite, esistono $\delta_1,\delta_2>0$ tali che

$$
x\in \operatorname{dom}f,\ 0<d_X(x,x_0)<\delta_1 \implies d_Y(f(x),l_1)<\varepsilon,
$$

$$
x\in \operatorname{dom}f,\ 0<d_X(x,x_0)<\delta_2 \implies d_Y(f(x),l_2)<\varepsilon.
$$

Poiché $x_0$ è di accumulazione per $\operatorname{dom}f$, esiste $x\in\operatorname{dom}f$ con $0<d_X(x,x_0)<\min\{\delta_1,\delta_2\}$. Allora

$$
d_Y(l_1,l_2)
\le
d_Y(l_1,f(x))+d_Y(f(x),l_2)
<
2\varepsilon
=
\frac{2}{3}d_Y(l_1,l_2),
$$

contraddizione. Quindi $l_1=l_2$. $\square$

### Proposizione 9.6 — Caratterizzazione sequenziale del limite

Siano $(X,d_X)$ e $(Y,d_Y)$ spazi metrici, $f:\operatorname{dom}f\to Y$, e sia $x_0$ di accumulazione per $\operatorname{dom}f$. Allora

$$
\lim_{x\to x_0} f(x)=l
$$

se e solo se per ogni successione $\{x_k\}\subseteq \operatorname{dom}f\setminus\{x_0\}$ con $x_k\to x_0$ si ha $f(x_k)\to l$.

**Dimostrazione.**  
($\Rightarrow$) Fissiamo $\varepsilon>0$. Per definizione di limite esiste $\delta>0$ tale che

$$
x\in \operatorname{dom}f,\ 0<d_X(x,x_0)<\delta \implies d_Y(f(x),l)<\varepsilon.
$$

Poiché $x_k\to x_0$ e $x_k\neq x_0$, esiste $N$ tale che per $k>N$, $0<d_X(x_k,x_0)<\delta$. Dunque per $k>N$, $d_Y(f(x_k),l)<\varepsilon$, cioè $f(x_k)\to l$.

($\Leftarrow$) Supponiamo per assurdo che $f(x)\not\to l$. Allora esiste $\varepsilon>0$ tale che per ogni $\delta>0$ esiste $x\in\operatorname{dom}f$ con $0<d_X(x,x_0)<\delta$ e $d_Y(f(x),l)\ge\varepsilon$. Scegliendo $\delta=1/k$ otteniamo una successione $x_k\in\operatorname{dom}f\setminus\{x_0\}$ con $x_k\to x_0$ ma $f(x_k)\not\to l$, contraddizione. $\square$

### Proposizione 9.7 — Limite componente per componente

Siano $(X,d)$ uno spazio metrico, $\overline{f}:\operatorname{dom}\overline{f}\to\mathbb{R}^n$ con $\operatorname{dom}\overline{f}\subseteq X$, e sia $x_0$ di accumulazione per $\operatorname{dom}\overline{f}$. Scriviamo

$$
\overline{f}(x)=(f_1(x),\dots,f_n(x)).
$$

Allora

$$
\lim_{x\to x_0}\overline{f}(x)=\overline{l}=(l_1,\dots,l_n)
\iff
\lim_{x\to x_0} f_i(x)=l_i \quad \forall i=1,\dots,n.
$$

**Dimostrazione.**  
Usiamo le disuguaglianze

$$
\|\vec{y}\|_\infty \le \|\vec{y}\|_2 \le \|\vec{y}\|_1 \le n\|\vec{y}\|_\infty.
$$

Se $\overline{f}(x)\to\overline{l}$, allora per ogni $i$,

$$
|f_i(x)-l_i| \le \|\overline{f}(x)-\overline{l}\|_\infty \le \|\overline{f}(x)-\overline{l}\|_2 \to 0.
$$

Viceversa, se $f_i(x)\to l_i$ per ogni $i$, allora per ogni $\varepsilon>0$ esistono $\delta_1,\dots,\delta_n>0$ tali che

$$
0<d(x,x_0)<\delta_i \implies |f_i(x)-l_i|<\frac{\varepsilon}{n}.
$$

Posto $\delta=\min\{\delta_1,\dots,\delta_n\}>0$, per $0<d(x,x_0)<\delta$ si ha

$$
\|\overline{f}(x)-\overline{l}\|_1
=
\sum_{i=1}^n |f_i(x)-l_i|
<
n\cdot\frac{\varepsilon}{n}
=
\varepsilon.
$$

Poiché $\|\cdot\|_2\le\|\cdot\|_1$, anche $\|\overline{f}(x)-\overline{l}\|_2<\varepsilon$, quindi $\overline{f}(x)\to\overline{l}$. $\square$

### 9.2 Algebra dei limiti in $\mathbb{R}$

Siano $f,g:\operatorname{dom}\to\mathbb{R}$ con $x_0$ di accumulazione per $\operatorname{dom}$, e supponiamo che

$$
\lim_{x\to x_0} f(x)=l_1,
\qquad
\lim_{x\to x_0} g(x)=l_2.
$$

Allora:

1. $\displaystyle \lim_{x\to x_0} (f(x)+g(x)) = l_1+l_2$.
2. $\displaystyle \lim_{x\to x_0} (f(x)\cdot g(x)) = l_1\cdot l_2$.
3. Se $l_2\neq 0$, allora $\displaystyle \lim_{x\to x_0} \frac{f(x)}{g(x)} = \frac{l_1}{l_2}$, purché $g(x)\neq 0$ in un intorno di $x_0$.

**Idea della dimostrazione.**  
Le dimostrazioni si basano sulla disuguaglianza triangolare e sulla limitatezza locale delle funzioni convergenti. Per esempio, per il prodotto:

$$
\begin{aligned}
|f(x)g(x)-l_1 l_2|
&\le |f(x)-l_1|\,|g(x)| + |l_1|\,|g(x)-l_2| \\
&\le |f(x)-l_1|\,(|g(x)-l_2|+|l_2|) + |l_1|\,|g(x)-l_2|.
\end{aligned}
$$

Poiché $|g(x)-l_2|\to 0$ e $|f(x)-l_1|\to 0$, il membro di destra tende a $0$. $\square$

### Teorema 9.8 — Teorema del confronto

Siano $f,g:\operatorname{dom}\to\mathbb{R}$ con $x_0$ di accumulazione per $\operatorname{dom}$, e supponiamo che esista un intorno $U$ di $x_0$ tale che

$$
f(x)\le g(x)
\qquad \forall x\in U\cap\operatorname{dom},\ x\neq x_0.
$$

Se

$$
\lim_{x\to x_0} f(x)=l_1
\qquad\text{e}\qquad
\lim_{x\to x_0} g(x)=l_2,
$$

allora

$$
l_1\le l_2.
$$

**Dimostrazione.**  
Supponiamo per assurdo $l_1>l_2$. Sia $\varepsilon = (l_1-l_2)/3>0$. Per definizione di limite, esistono $\delta_1,\delta_2>0$ tali che

$$
x\in \operatorname{dom}f,\ 0<d_X(x,x_0)<\delta_1 \implies |f(x)-l_1|<\varepsilon,
$$

$$
x\in \operatorname{dom}f,\ 0<d_X(x,x_0)<\delta_2 \implies |g(x)-l_2|<\varepsilon.
$$

Poiché $x_0$ è di accumulazione per $\operatorname{dom}$ e $U$ è un intorno di $x_0$, esiste $x\in U\cap\operatorname{dom}$ con $0<d_X(x,x_0)<\min\{\delta_1,\delta_2\}$. Allora

$$
f(x)>l_1-\varepsilon
\qquad\text{e}\qquad
g(x)<l_2+\varepsilon.
$$

Ma $l_1-\varepsilon = l_1-(l_1-l_2)/3 = (2l_1+l_2)/3$ e $l_2+\varepsilon = l_2+(l_1-l_2)/3 = (l_1+2l_2)/3$. Poiché $l_1>l_2$, si ha

$$
\frac{2l_1+l_2}{3} > \frac{l_1+2l_2}{3},
$$

quindi $f(x)>g(x)$, in contraddizione con $f(x)\le g(x)$. Pertanto $l_1\le l_2$. $\square$

### Teorema 9.9 — Teorema del confronto (due carabinieri)

Siano $f,g,h:\operatorname{dom}\to\mathbb{R}$ con $x_0$ di accumulazione per $\operatorname{dom}$, e supponiamo che esista un intorno $U$ di $x_0$ tale che

$$
f(x)\le g(x)\le h(x)
\qquad \forall x\in U\cap\operatorname{dom},\ x\neq x_0.
$$

Se

$$
\lim_{x\to x_0} f(x)=\lim_{x\to x_0} h(x)=l,
$$

allora

$$
\lim_{x\to x_0} g(x)=l.
$$

**Dimostrazione.**  
Fissiamo $\varepsilon>0$. Per ipotesi esistono $\delta_1,\delta_2>0$ tali che

$$
0<d(x,x_0)<\delta_1 \implies |f(x)-l|<\varepsilon,
$$

$$
0<d(x,x_0)<\delta_2 \implies |h(x)-l|<\varepsilon.
$$

Sia $\delta=\min\{\delta_1,\delta_2\}$ e sia $x\in\operatorname{dom}$ con $0<d(x,x_0)<\delta$. Allora

$$
l-\varepsilon < f(x) \le g(x) \le h(x) < l+\varepsilon,
$$

quindi $|g(x)-l|<\varepsilon$. Dunque $g(x)\to l$. $\square$

## 10. Funzioni continue tra spazi metrici

Siano $(X, d_X)$ e $(Y, d_Y)$ spazi metrici e sia $f: \operatorname{dom}f \to Y$, con $\operatorname{dom}f \subseteq X$. Si dice che $f$ è **continua in** $x_0 \in \operatorname{dom}f$ se vale una delle due condizioni:

1. $x_0$ è un punto isolato per $\operatorname{dom}f$, ovvero esiste un intorno $U$ di $x_0$ tale che $U \cap \operatorname{dom}f = \{x_0\}$; in altre parole, esiste $r > 0$ tale che $\mathcal{B}_r(x_0) \cap \operatorname{dom}f = \{x_0\}$.

2. Se $x_0$ è un punto di accumulazione per $\operatorname{dom}f$,
   $$
   \lim_{x \to x_0} f(x) = f(x_0).
   $$
   In altre parole, $f$ è continua in $x_0$ se per ogni $\varepsilon > 0$ esiste $\delta > 0$ tale che
   $$
   x \in \mathcal{B}_{\delta}(x_0) \cap \operatorname{dom}f \implies f(x) \in \mathcal{B}_{\varepsilon}(f(x_0)),
   $$
   cioè se
   $$
   \forall \varepsilon > 0 \ \exists \delta > 0 : d_X(x,x_0) < \delta,\ x \in \operatorname{dom}f \implies d_Y(f(x), f(x_0)) < \varepsilon.
   $$

### Osservazione 10.1

Se $x_0$ è isolato per $\operatorname{dom}f$, ogni funzione è automaticamente continua in $x_0$, perché non ci sono altri punti del dominio su cui imporre la condizione. La continuità è quindi una nozione interessante solo nei punti di accumulazione.

### Esempio 10.2 — Funzione continua

La funzione $f:\mathbb{R}\to\mathbb{R}$, $f(x)=x^2$, è continua in ogni $x_0\in\mathbb{R}$, perché

$$
|f(x)-f(x_0)| = |x^2-x_0^2| = |x-x_0|\,|x+x_0|,
$$

e scegliendo $\delta$ in funzione di $\varepsilon$ e $x_0$ si ottiene la tesi.

### Esempio 10.3 — Funzione non continua

La funzione

$$
f(x)=
\begin{cases}
1, & x>0,\\
0, & x\le 0
\end{cases}
$$

non è continua in $x_0=0$: il limite per $x\to 0^+$ vale $1$, per $x\to 0^-$ vale $0$, quindi il limite non esiste e $f$ non è continua in $0$.

### Teorema 10.4 — Continuità e controimmagini di aperti

Siano $(X, d_X)$ e $(Y, d_Y)$ spazi metrici e sia $f: X \to Y$. Allora $f$ è continua (cioè è continua in ogni punto di $X$) se e solo se $f^{-1}(A)$ è un aperto di $X$ per ogni $A \subseteq Y$ aperto.

**Dimostrazione.**

($\Rightarrow$) Sia $A\subseteq Y$ aperto. Mostriamo che $f^{-1}(A)$ è aperto in $X$. Sia $x_0\in f^{-1}(A)$, cioè $f(x_0)\in A$. Poiché $A$ è aperto, esiste $\varepsilon>0$ con $\mathcal{B}_\varepsilon(f(x_0))\subseteq A$. Poiché $f$ è continua in $x_0$, esiste $\delta>0$ tale che

$$
d_X(x,x_0)<\delta \implies d_Y(f(x),f(x_0))<\varepsilon.
$$

Allora per ogni $x\in\mathcal{B}_\delta(x_0)$ si ha $f(x)\in\mathcal{B}_\varepsilon(f(x_0))\subseteq A$, quindi $x\in f^{-1}(A)$. Dunque $\mathcal{B}_\delta(x_0)\subseteq f^{-1}(A)$, e $f^{-1}(A)$ è aperto.

($\Leftarrow$) Sia $x_0\in X$ e $\varepsilon>0$. L'insieme $\mathcal{B}_\varepsilon(f(x_0))$ è aperto in $Y$. Per ipotesi, $f^{-1}(\mathcal{B}_\varepsilon(f(x_0)))$ è aperto in $X$ e contiene $x_0$. Quindi esiste $\delta>0$ con

$$
\mathcal{B}_\delta(x_0)\subseteq f^{-1}(\mathcal{B}_\varepsilon(f(x_0))).
$$

Allora per ogni $x\in X$ con $d_X(x,x_0)<\delta$ si ha $f(x)\in\mathcal{B}_\varepsilon(f(x_0))$, cioè $d_Y(f(x),f(x_0))<\varepsilon$. Dunque $f$ è continua in $x_0$. Poiché $x_0$ è arbitrario, $f$ è continua. $\square$

### Osservazione 10.5 — Continuità e controimmagini di chiusi

Se $f: X \to Y$ è continua e $C \subseteq Y$ è chiuso, allora $f^{-1}(C)$ è chiuso in $X$.

**Dimostrazione.**  
$C^C$ è aperto in $Y$. Per il Teorema 10.4, $f^{-1}(C^C)$ è aperto in $X$. Ora

$$
(f^{-1}(C))^C = \{x \in X : x \notin f^{-1}(C)\} = \{x \in X : f(x) \notin C\} = f^{-1}(C^C),
$$

che è aperto. Quindi $f^{-1}(C)$ è chiuso. $\square$

### Osservazione 10.6 — Algebra delle funzioni continue

Siano $f,g : X \to \mathbb{R}$ continue in $x_0 \in X$, con $(X, d)$ spazio metrico. Allora $f+g$, $f\cdot g$ e, se $g(x_0) \neq 0$, $\frac{f}{g}$ sono continue in $x_0$.

**Idea della dimostrazione.**  
Segue dalle proprietà dei limiti (algebra dei limiti in $\mathbb{R}$) applicate alla definizione di continuità. $\square$

### Proposizione 10.7 — Continuità componente per componente

Sia $(X, d)$ uno spazio metrico e sia $\overline{f}: \operatorname{dom}\overline{f} \to \mathbb{R}^n$, con $\overline{f}(x) = (f_1(x), \dots, f_n(x))$ e $\operatorname{dom}\overline{f} \subseteq X$. Allora $\overline{f}$ è continua in $x_0$ se e solo se tali sono $f_1, \dots, f_n$ con $f_i: \operatorname{dom}\overline{f} \to \mathbb{R}$, $i = 1,\dots, n$.

**Dimostrazione.**  
Segue dalla Proposizione 9.7 applicata alla definizione di continuità. Infatti, $\overline{f}$ è continua in $x_0$ se e solo se

$$
\lim_{x\to x_0} \overline{f}(x) = \overline{f}(x_0),
$$

e per la Proposizione 9.7 questo è equivalente a

$$
\lim_{x\to x_0} f_i(x) = f_i(x_0) \qquad \forall i=1,\dots,n,
$$

cioè alla continuità di ciascuna $f_i$ in $x_0$. $\square$

### Esempio 10.8

La funzione $\overline{f}:\mathbb{R}\to\mathbb{R}^2$, $\overline{f}(x)=(\cos x, \sin x)$, è continua perché lo sono le sue componenti $f_1(x)=\cos x$ e $f_2(x)=\sin x$. La sua immagine è la circonferenza unitaria.
