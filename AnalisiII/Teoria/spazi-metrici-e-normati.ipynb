# Spazi vettoriali, normati e metrici

---

## 1. Spazi vettoriali

### Definizione 1.1 — Spazio vettoriale

Sia $V$ un insieme non vuoto e sia $\mathbb{F}$ un campo (tipicamente $\mathbb{R}$ o $\mathbb{C}$). Una struttura di **spazio vettoriale** su $V$ è data da due operazioni

$$
+ : V \times V \to V, \qquad \cdot : \mathbb{F} \times V \to V,
$$

dette rispettivamente **somma** e **prodotto per scalare**, tali che:

1. $(V,+)$ è un gruppo abeliano:
   - **Commutatività:** $x+y=y+x$ per ogni $x,y\in V$;
   - **Associatività:** $(x+y)+z=x+(y+z)$ per ogni $x,y,z\in V$;
   - **Elemento neutro:** esiste $0_V\in V$ con $x+0_V=x$ per ogni $x\in V$;
   - **Inverso additivo:** per ogni $x\in V$ esiste $-x\in V$ con $x+(-x)=0_V$.

2. Il prodotto per scalare soddisfa:
   - **Distributività vettoriale:** $\lambda(x+y)=\lambda x+\lambda y$;
   - **Distributività scalare:** $(\lambda+\mu)x=\lambda x+\mu x$;
   - **Associatività mista:** $(\lambda\mu)x=\lambda(\mu x)$;
   - **Elemento neutro scalare:** $1_{\mathbb{F}}\,x=x$.

Gli elementi di $V$ sono **vettori**, gli elementi di $\mathbb{F}$ **scalari**.

> **Osservazione 1.1.1 — Proprietà elementari.**  
> Dalla definizione seguono:
> - $0_{\mathbb{F}}x=0_V$ per ogni $x\in V$;
> - $\lambda 0_V=0_V$ per ogni $\lambda\in\mathbb{F}$;
> - $(-1_{\mathbb{F}})x=-x$ per ogni $x\in V$.

**Dimostrazione.** Per la prima: $\lambda x=(\lambda+0_{\mathbb{F}})x=\lambda x+0_{\mathbb{F}}x$, quindi $0_{\mathbb{F}}x=0_V$. Le altre sono analoghe. $\square$

### Esempio 1.2
Lo spazio $\mathbb{F}^n$ con operazioni componente per componente è uno spazio vettoriale su $\mathbb{F}$.

### Esempio 1.3
$C^0([a,b],\mathbb{R})$ con somma e prodotto puntuali è uno spazio vettoriale su $\mathbb{R}$.

### Esempio 1.4
$\mathbb{R}^{m\times n}$ (matrici) e $\mathbb{R}[x]$ (polinomi) con le operazioni usuali sono spazi vettoriali.

## 2. Prodotto scalare e norma indotta

### Definizione 2.1 — Prodotto scalare

Sia $V$ uno spazio vettoriale su $\mathbb{F}$. Un **prodotto scalare** su $V$ è una funzione $\langle\cdot,\cdot\rangle:V\times V\to\mathbb{F}$ tale che, per ogni $x,y,z\in V$ e $\lambda,\mu\in\mathbb{F}$:

1. **Simmetria hermitiana:** $\langle x,y\rangle=\overline{\langle y,x\rangle}$;
2. **Linearità nella prima variabile:** $\langle\lambda x+\mu y,z\rangle=\lambda\langle x,z\rangle+\mu\langle y,z\rangle$;
3. **Positività definita:** $\langle x,x\rangle\ge 0$ e $\langle x,x\rangle=0\iff x=0_V$.

La coppia $(V,\langle\cdot,\cdot\rangle)$ è detta **spazio pre-Hilbertiano**.

> **Osservazione 2.2 — Antilinearità.**  
> Dalla simmetria hermitiana e dalla linearità nella prima variabile segue l'**antilinearità nella seconda**:
> $$
> \langle x,\lambda y+\mu z\rangle=\overline{\lambda}\langle x,y\rangle+\overline{\mu}\langle x,z\rangle.
> $$
> In $\mathbb{R}^n$ il prodotto scalare è **bilineare**.

### Definizione 2.3 — Norma indotta
$$
\|x\|:=\sqrt{\langle x,x\rangle},\qquad x\in V.
$$

### Teorema 2.4 — Disuguaglianza di Cauchy–Schwarz

Per ogni $x,y\in V$, $|\langle x,y\rangle|\le\|x\|\|y\|$.

**Dimostrazione.** Se $y=0$ è banale. Sia $y\neq 0$ e $t\in\mathbb{R}$. Allora
$$
0\le\|x+ty\|^2=\|x\|^2+2t\,\Re\langle x,y\rangle+t^2\|y\|^2.
$$
Il polinomio in $t$ è non negativo per ogni $t$, quindi il discriminante è $\le 0$:
$$
(2\Re\langle x,y\rangle)^2-4\|x\|^2\|y\|^2\le 0\implies|\Re\langle x,y\rangle|\le\|x\|\|y\|.
$$
Se $\mathbb{F}=\mathbb{C}$, si scelga $\lambda\in\mathbb{C}$, $|\lambda|=1$, con $\lambda\langle x,y\rangle=|\langle x,y\rangle|$. Allora
$$
|\langle x,y\rangle|=\langle\lambda x,y\rangle=\Re\langle\lambda x,y\rangle\le\|\lambda x\|\|y\|=\|x\|\|y\|.
$$
$\square$

### Proposizione 2.5 — La norma indotta è una norma

La funzione $\|x\|=\sqrt{\langle x,x\rangle}$ è una norma su $V$.

**Dimostrazione.** Positività e omogeneità sono immediate. Per la disuguaglianza triangolare:
$$
\|x+y\|^2=\|x\|^2+2\Re\langle x,y\rangle+\|y\|^2\le\|x\|^2+2\|x\|\|y\|+\|y\|^2=(\|x\|+\|y\|)^2.
$$
$\square$

### Esempi 2.6–2.9
- **Euclideo** su $\mathbb{R}^n$: $\langle x,y\rangle=\sum_i x_iy_i$, $\|x\|_2=\sqrt{\sum_i x_i^2}$.
- **Hermitiano** su $\mathbb{C}^n$: $\langle x,y\rangle=\sum_i x_i\overline{y_i}$.
- **Integrale** su $C^0([0,1])$: $\langle f,g\rangle=\int_0^1 f\overline{g}$, $\|f\|_2=\left(\int_0^1 |f|^2\right)^{1/2}$.
- **Pesato** su $\mathbb{R}^n$: fissata $A$ simmetrica definita positiva, $\langle x,y\rangle_A=x^TAy$.

## 3. Spazi normati

### Definizione 3.1 — Norma

Sia $X$ spazio vettoriale su $\mathbb{F}$. Una **norma** è $\|\cdot\|:X\to\mathbb{R}$ con:

1. $\|x\|\ge 0$ e $\|x\|=0\iff x=0_X$;
2. $\|\lambda x\|=|\lambda|\|x\|$;
3. $\|x+y\|\le\|x\|+\|y\|$.

La coppia $(X,\|\cdot\|)$ è uno **spazio normato**.

> **Osservazione 3.1.1 — Disuguaglianza triangolare inversa.**  
> $\bigl|\|x\|-\|y\|\bigr|\le\|x-y\|$.

**Dimostrazione.** $\|x\|=\|(x-y)+y\|\le\|x-y\|+\|y\|$, quindi $\|x\|-\|y\|\le\|x-y\|$. Scambiando $x$ e $y$ si ottiene la tesi. $\square$

### Esempi 3.2–3.6
- **Norma euclidea** $\|x\|_2=\sqrt{\sum_i x_i^2}$ su $\mathbb{R}^n$: verifica dei tre assiomi, con il terzo tramite Cauchy–Schwarz.
- **Norma 1** $\|x\|_1=\sum_i |x_i|$.
- **Norme $p$** $\|x\|_p=\left(\sum_i|x_i|^p\right)^{1/p}$ per $p\ge 1$; per $p=\infty$, $\|x\|_\infty=\max_i|x_i|$.
- **Norma del sup** $\|f\|_\infty=\sup_{t\in[0,1]}|f(t)|$ su $C^0([0,1])$.
- **Norma integrale** $\|f\|_1=\int_a^b|f(x)|dx$ su $C^0([a,b])$.

> **Osservazione 3.4.1 — $0<p<1$ non è norma.**  
> In $\mathbb{R}^2$, $\|(1,0)+(0,1)\|_{1/2}=4>2=\|(1,0)\|_{1/2}+\|(0,1)\|_{1/2}$.

### Proposizione 3.7 — Da norma a metrica

Ponendo $d(x,y):=\|x-y\|$, si ottiene una metrica su $X$.

**Dimostrazione.** Positività e simmetria sono immediate. Disuguaglianza triangolare:
$$
d(x,y)=\|(x-z)+(z-y)\|\le\|x-z\|+\|z-y\|=d(x,z)+d(z,y).
$$
$\square$

> **Osservazione 3.8 — Metriche non indotte da norme.**  
> La metrica discreta non è indotta da una norma: se fosse $d(0,2x)=\|2x\|=2\|x\|=2d(0,x)=2\neq 1$.

## 4. Spazi metrici

### Definizione 4.1 — Metrica

$d:X\times X\to\mathbb{R}$ è una **metrica** su $X$ se, per ogni $x,y,z\in X$:

1. $d(x,y)\ge 0$ e $d(x,y)=0\iff x=y$;
2. $d(x,y)=d(y,x)$;
3. $d(x,y)\le d(x,z)+d(z,y)$.

La coppia $(X,d)$ è uno **spazio metrico**.

### Esempio 4.2 — Metrica euclidea
$d_2(x,y)=\|x-y\|_2=\sqrt{\sum_i(x_i-y_i)^2}$ su $\mathbb{R}^n$.

### Esempio 4.3 — Metrica $p$-esima, $0<p<1$
$d(x,y)=|x-y|^p$ su $\mathbb{R}$ è una metrica.

**Dimostrazione.** Positività e simmetria sono immediate. Per la disuguaglianza triangolare occorre dimostrare che, per $a,b\ge 0$ e $0<p<1$,
$$
(a+b)^p\le a^p+b^p.
$$
Consideriamo $g(t)=(1+t)^p-1-t^p$ per $t\ge 0$. Allora $g(0)=0$ e
$$
g'(t)=p[(1+t)^{p-1}-t^{p-1}]<0
$$
perché $p-1<0$ implica che $t\mapsto t^{p-1}$ è decrescente, e $(1+t)^{p-1}<t^{p-1}$. Quindi $g$ è decrescente, $g(t)\le g(0)=0$, da cui $(1+t)^p\le 1+t^p$. Posto $t=b/a$ e moltiplicando per $a^p$ si ottiene la tesi. Da ciò segue
$$
d(x,y)=|x-y|^p\le(|x-z|+|z-y|)^p\le|x-z|^p+|z-y|^p=d(x,z)+d(z,y).
$$
$\square$

### Esempio 4.4 — Metrica discreta
$d(x,y)=0$ se $x=y$, $d(x,y)=1$ se $x\neq y$.

### Esempio 4.5 — Metrica uniforme
$d_\infty(f,g)=\sup_{t\in[a,b]}|f(t)-g(t)|$ su $C^0([a,b])$.

### Definizione 4.6 — Palla aperta
$$
\mathcal{B}_r(x_0)=\{x\in X:d(x,x_0)<r\}.
$$

### Esempio 4.7 — Palle in $\mathbb{R}^2$ con norme diverse

**Norma euclidea** $\|\cdot\|_2$: cerchio aperto di raggio $r$.
$$
\mathcal{B}_r(x_0,y_0)=\{(x,y):\sqrt{(x-x_0)^2+(y-y_0)^2}<r\}.
$$

**Norma del sup** $\|\cdot\|_\infty$: quadrato aperto di semi-lato $r$.
$$
\mathcal{B}_r(x_0,y_0)=\{(x,y):\max\{|x-x_0|,|y-y_0|\}<r\}.
$$

**Norma 1** $\|\cdot\|_1$: rombo aperto con vertici $(x_0\pm r,y_0)$, $(x_0,y_0\pm r)$.
$$
\mathcal{B}_r(x_0,y_0)=\{(x,y):|x-x_0|+|y-y_0|<r\}.
$$

### Esempio 4.8 — Palle in $C^0([0,1])$

**Norma del sup:** $g\in\mathcal{B}_r(f)$ se il grafico di $g$ sta in una striscia di semi-ampiezza $r$ attorno a quello di $f$.

**Norma integrale:** $\mathcal{B}_1(0)=\{g:\int_0^1|g|<1\}$ contiene funzioni con oscillazioni grandi e area piccola.

## 5. Topologia degli spazi metrici

### Definizione 5.1 — Intorno

$U\subseteq X$ è **intorno** di $x_0$ se esiste $r>0$ con $\mathcal{B}_r(x_0)\subseteq U$.

### Definizione 5.2 — Aperto

$A\subseteq X$ è **aperto** se è intorno di ogni suo punto.

### Definizione 5.3 — Chiuso

$C\subseteq X$ è **chiuso** se $C^C$ è aperto.

> **Osservazione 5.4 — Né aperti né chiusi.**  
> $I=(0,1]$ in $\mathbb{R}$ non è aperto (manca un intorno di $1$ contenuto in $I$) né chiuso ($I^C=(-\infty,0]\cup(1,\infty)$ non è aperto, perché non esiste un intorno di $0$ contenuto in $I^C$).

> **Osservazione 5.5 — Le palle aperte sono aperte.**  
> Sia $y\in\mathcal{B}_r(x)$ e $s=r-d(y,x)>0$. Per $z\in\mathcal{B}_s(y)$,
> $$
> d(z,x)\le d(z,y)+d(y,x)<s+d(y,x)=r.
> $$
> Quindi $\mathcal{B}_s(y)\subseteq\mathcal{B}_r(x)$.

> **Osservazione 5.6.**  
> In ogni spazio metrico $X$ e $\emptyset$ sono contemporaneamente aperti e chiusi.

### Teorema 5.7 — Unione e intersezione di aperti

1. Unione arbitraria di aperti è aperta.
2. Intersezione finita di aperti è aperta.

**Dimostrazione.** (1) Sia $x_0\in\bigcup_{i\in I}A_i$. Allora $x_0\in A_j$ per qualche $j$. Poiché $A_j$ è aperto, esiste $r>0$ con $\mathcal{B}_r(x_0)\subseteq A_j\subseteq\bigcup A_i$.
(2) Sia $x_0\in\bigcap_{i=1}^k A_i$. Per ogni $i$, esiste $r_i>0$ con $\mathcal{B}_{r_i}(x_0)\subseteq A_i$. Posto $r=\min\{r_1,\dots,r_k\}>0$, $\mathcal{B}_r(x_0)\subseteq\bigcap_i A_i$. $\square$

> **Osservazione 5.8 — Intersezione arbitraria di aperti.**  
> $A_n=(-1/n,1/n)$ in $\mathbb{R}$ dà $\bigcap_n A_n=\{0\}$, che non è aperto.

### Corollario 5.9 — Unione e intersezione di chiusi

1. Unione finita di chiusi è chiusa.
2. Intersezione arbitraria di chiusi è chiusa.

**Dimostrazione.** (1) $\left(\bigcup C_i\right)^C=\bigcap C_i^C$ è aperto.
(2) $\left(\bigcap C_i\right)^C=\bigcup C_i^C$ è aperto. $\square$

> **Osservazione 5.10.**  
> $C_n=[-1+1/n,1-1/n]$ ha $\bigcup_n C_n=(-1,1)$, che non è chiuso.

### Definizione 5.11 — Insieme limitato

$C\subseteq X$ è **limitato** se esistono $x\in X$, $R>0$ con $C\subseteq\mathcal{B}_R(x)$.

> **Osservazione 5.12.**  
> In uno spazio normato, $x$ può essere lo zero: $C$ è limitato se e solo se esiste $R>0$ con $\|z\|\le R$ per ogni $z\in C$.

### Caratterizzazione di limitatezza in $\mathbb{R}^n$

$A\subseteq\mathbb{R}^n$ è limitato se e solo se esistono $M_1,\dots,M_n>0$ con $-M_i\le x_i\le M_i$ per ogni $x\in A$, $i=1,\dots,n$.

## 6. Punti notevoli di un insieme

### Definizione 6.1 — Punto interno
$x_0$ è **interno** ad $A$ se esiste $r>0$ con $\mathcal{B}_r(x_0)\subseteq A$. L'insieme dei punti interni è $\mathring{A}$.

> **Osservazione 6.2.**  
> $x_0$ è interno ad $A$ se e solo se esiste $r>0$ con $\mathcal{B}_r(x_0)\cap A=\mathcal{B}_r(x_0)$.

### Esempio 6.3
$I=(0,1]$, $\mathring{I}=(0,1)$.

### Proposizione 6.4
$\mathring{A}$ è il più grande aperto contenuto in $A$.

**Dimostrazione.** Se $B\subseteq A$ è aperto e $x\in B$, allora esiste $r>0$ con $\mathcal{B}_r(x)\subseteq B\subseteq A$, quindi $x\in\mathring{A}$. Inoltre $\mathring{A}$ è aperto: se $x\in\mathring{A}$, esiste $r>0$ con $\mathcal{B}_r(x)\subseteq A$; per $y\in\mathcal{B}_r(x)$, posto $s=r-d(x,y)>0$, $\mathcal{B}_s(y)\subseteq\mathcal{B}_r(x)\subseteq A$, quindi $y\in\mathring{A}$. $\square$

### Definizione 6.5 — Punto esterno
$x_0$ è **esterno** ad $A$ se è interno ad $A^C$.

### Definizione 6.6 — Punto di accumulazione
$x_0\in X$ è **di accumulazione** per $A$ se
$$
\forall r>0,\quad(\mathcal{B}_r(x_0)\setminus\{x_0\})\cap A\neq\emptyset.
$$
L'insieme derivato è $A'$.

### Esempio 6.7
$A=\{1/n\}\cup\{0\}$ ha $0$ come unico punto di accumulazione.

### Definizione 6.8 — Punto di chiusura
$x_0$ è **punto di chiusura** per $A$ se $\forall r>0$, $\mathcal{B}_r(x_0)\cap A\neq\emptyset$. La **chiusura** è $\overline{A}$.

### Proposizione 6.9
$\overline{A}$ è il più piccolo chiuso contenente $A$.

**Dimostrazione.** Sia $C\supseteq A$ chiuso. Se $x\notin C$, esiste $r>0$ con $\mathcal{B}_r(x)\subseteq C^C$, quindi $\mathcal{B}_r(x)\cap C=\emptyset$, a maggior ragione $\mathcal{B}_r(x)\cap A=\emptyset$, da cui $x\notin\overline{A}$. Dunque $\overline{A}\subseteq C$. $\square$

### Definizione 6.10 — Punto di frontiera
$x_0$ è **di frontiera** per $A$ se per ogni $r>0$, $\mathcal{B}_r(x_0)\cap A\neq\emptyset$ e $\mathcal{B}_r(x_0)\cap A^C\neq\emptyset$. La **frontiera** è $\partial A$.

### Definizione 6.11 — Punto isolato
$x_0\in A$ è **isolato** in $A$ se esiste $r>0$ con $\mathcal{B}_r(x_0)\cap A=\{x_0\}$.

### Osservazione 6.12 — Relazioni
1. $x_0$ interno $\Rightarrow$ $x_0\in A$ e non di frontiera.
2. $x_0$ esterno $\Rightarrow$ $x_0\in A^C$ e non di frontiera.
3. $x_0$ isolato $\Rightarrow$ non di accumulazione.
4. $x_0$ di accumulazione $\Rightarrow$ di chiusura (ma può non appartenere ad $A$).
5. Ogni punto è interno, esterno o di frontiera (categorie esaustive e mutuamente esclusive).

### Osservazione 6.13 — Relazioni globali
$$
\mathring{A}\subseteq A\subseteq\overline{A},\quad
A\text{ aperto}\iff A=\mathring{A},\quad
A\text{ chiuso}\iff A=\overline{A},\quad
\partial A=\overline{A}\cap\overline{A^C},\quad
\overline{A}=A\cup\partial A.
$$

### Esempio 6.14
$A=\{(x,y):x^2+y^2<1\}\cup\{(2,0)\}$: $(0,0)$ interno, $(1,0)$ frontiera, $(2,0)$ isolato e di frontiera, $(3,0)$ esterno.

## 7. Successioni in spazi metrici

### Definizione 7.1 — Successione
$x:\mathbb{N}\to X$, $k\mapsto x_k$.

### Definizione 7.2 — Convergenza
$x_k\to l$ se
$$
\forall\varepsilon>0\ \exists N:k>N\implies d(x_k,l)<\varepsilon.
$$

### Teorema 7.3 — Unicità del limite
Se $x_k\to l_1$ e $x_k\to l_2$, allora $l_1=l_2$.

**Dimostrazione.** Sia $\varepsilon=d(l_1,l_2)/3$. Esistono $N_1,N_2$ con $d(x_k,l_1)<\varepsilon$ per $k>N_1$ e $d(x_k,l_2)<\varepsilon$ per $k>N_2$. Per $k>\max\{N_1,N_2\}$,
$$
d(l_1,l_2)\le d(l_1,x_k)+d(x_k,l_2)<2\varepsilon=\tfrac{2}{3}d(l_1,l_2),
$$
contraddizione. $\square$

### Proposizione 7.4 — Convergenza in $\mathbb{R}^n$
$x_k\to l$ in $\mathbb{R}^n$ se e solo se $x_{kj}\to l_j$ per ogni $j$.

**Dimostrazione.** $|x_{kj}-l_j|\le\|x_k-l\|_\infty\le\|x_k-l\|_2\to 0$. Viceversa, se le componenti convergono, $\|x_k-l\|_\infty=\max_j|x_{kj}-l_j|\to 0$. $\square$

### Teorema 7.5 — Caratterizzazione della chiusura
$x\in\overline{A}$ se e solo se esiste $\{x_k\}\subseteq A$ con $x_k\to x$.

**Dimostrazione.** Se $x\in\overline{A}$, per ogni $k$ si scelga $x_k\in\mathcal{B}_{1/k}(x)\cap A$. Allora $d(x_k,x)<1/k\to 0$.
Viceversa, se $x_k\to x$, per ogni $r>0$ esiste $N$ con $d(x_k,x)<r$ per $k>N$, quindi $x_k\in\mathcal{B}_r(x)\cap A$, da cui $x\in\overline{A}$. $\square$

### Corollario 7.6 — Caratterizzazione dei chiusi
$A$ è chiuso se e solo se ogni successione convergente di elementi di $A$ ha limite in $A$.

**Dimostrazione.** Se $A$ è chiuso, $A=\overline{A}$; se $x_k\to x$ con $x_k\in A$, allora $x\in\overline{A}=A$. Viceversa, se $x\in\overline{A}$, esiste $\{x_k\}\subseteq A$ con $x_k\to x$; per ipotesi $x\in A$, quindi $\overline{A}\subseteq A$, da cui $A=\overline{A}$. $\square$

### Proposizione 7.7 — Successioni convergenti sono limitate
Sia $x_k\to l$. Per $\varepsilon=1$ esiste $N$ con $d(x_k,l)<1$ per $k>N$. Posto $C=\max\{1,d(x_1,l),\dots,d(x_N,l)\}+1$, si ha $d(x_k,l)<C$ per ogni $k$. $\square$

## 8. Spazi metrici completi

### Definizione 8.1 — Successione di Cauchy
$\{x_k\}$ è **di Cauchy** se
$$
\forall\varepsilon>0\ \exists N:k,m>N\implies d(x_k,x_m)<\varepsilon.
$$

### Proposizione 8.2 — Ogni successione convergente è di Cauchy
Se $x_k\to l$, per $\varepsilon>0$ esiste $N$ con $d(x_k,l)<\varepsilon/2$ per $k>N$. Per $k,m>N$,
$$
d(x_k,x_m)\le d(x_k,l)+d(l,x_m)<\varepsilon.
$$
$\square$

> **Osservazione 8.3 — Il viceversa non vale.**  
> In $\mathbb{Q}$ con la metrica euclidea, la successione $x_k$ che approssima $\sqrt{2}$ è di Cauchy ma non converge in $\mathbb{Q}$.

### Definizione 8.4 — Spazio metrico completo
$(X,d)$ è **completo** se ogni successione di Cauchy converge in $X$.

### Esempio 8.5
$\mathbb{R}^n$ con la metrica euclidea è completo.

### Esempio 8.6
$C^0([a,b])$ con $\|f\|_\infty$ è completo (convergenza uniforme di funzioni continue è continua). Con $\|f\|_1$ non è completo.

### Definizione 8.7 — Spazio di Banach
Uno spazio normato completo è detto **spazio di Banach**.

### Definizione 8.8 — Spazio di Hilbert
Uno spazio pre-Hilbertiano completo è detto **spazio di Hilbert**.

### Teorema 8.9 — Sottospazi chiusi di spazi completi
Se $(X,d)$ è completo e $C\subseteq X$ è chiuso, allora $(C,d)$ è completo.

**Dimostrazione.** Sia $\{x_k\}\subseteq C$ di Cauchy in $C$. Poiché è di Cauchy in $X$ e $X$ è completo, $x_k\to x\in X$. Poiché $C$ è chiuso e $x_k\in C$, si ha $x\in C$. $\square$

## 9. Compattezza

### Definizione 9.1 — Ricoprimento aperto
Una famiglia $\{A_i\}_{i\in I}$ di aperti di $X$ è un **ricoprimento aperto** di $K\subseteq X$ se $K\subseteq\bigcup_{i\in I}A_i$.

### Definizione 9.2 — Compattezza per ricoprimenti
$K\subseteq X$ è **compatto** se ogni ricoprimento aperto di $K$ ammette un **sottoricoprimento finito**.

### Definizione 9.3 — Compattezza sequenziale
$K\subseteq X$ è **sequenzialmente compatto** se ogni successione in $K$ ammette una sottosuccessione convergente a un punto di $K$.

### Teorema 9.4 — Equivalenza
In uno spazio metrico $(X,d)$, $K$ è compatto se e solo se è sequenzialmente compatto.

**Idea della dimostrazione.** 
$(\Rightarrow)$ Se $K$ è compatto e $\{x_k\}\subseteq K$ non ha sottosuccessioni convergenti, per ogni $x\in K$ esiste $r_x>0$ con $\mathcal{B}_{r_x}(x)$ contenente solo finiti $x_k$. La famiglia $\{\mathcal{B}_{r_x}(x)\}_{x\in K}$ è un ricoprimento aperto; per compattezza ne esiste un sottoricoprimento finito, ma allora solo finiti $x_k$ apparterrebbero a $K$, contraddizione.
$(\Leftarrow)$ Dimostrazione tecnica che usa la completezza e la totale limitatezza. $\square$

### Teorema 9.5 — Heine–Borel
$K\subseteq\mathbb{R}^n$ è compatto se e solo se è chiuso e limitato.

**Dimostrazione.** 
$(\Rightarrow)$ Se $K$ è compatto, è chiuso (il complementare è aperto: per ogni $x\notin K$, per ogni $y\in K$ esistono palle disgiunte, e per compattezza se ne estrae un ricoprimento finito, da cui una palla centrata in $x$ disgiunta da $K$). È anche limitato: le palle $\mathcal{B}_k(0)$ per $k\in\mathbb{N}$ ricoprono $K$; per compattezza ne basta un numero finito, quindi $K\subseteq\mathcal{B}_N(0)$.
$(\Leftarrow)$ Se $K$ è chiuso e limitato, è contenuto in un compatto $[-M,M]^n$. Il prodotto di compatti è compatto (teorema di Tychonoff in dimensione finita), e un chiuso in un compatto è compatto. $\square$

### Teorema 9.6 — Bolzano–Weierstrass
Ogni successione limitata in $\mathbb{R}^n$ ammette una sottosuccessione convergente.

**Dimostrazione.** Sia $\{x_k\}$ limitata, contenuta in $[-M,M]^n$. Questo è compatto per Heine–Borel, quindi per il Teorema 9.4 esiste una sottosuccessione convergente. $\square$

### Teorema 9.7 — Weierstrass
Sia $K\subseteq\mathbb{R}^n$ compatto e $f:K\to\mathbb{R}$ continua. Allora $f$ ammette massimo e minimo su $K$.

**Dimostrazione.** $K$ compatto e $f$ continua implicano $f(K)$ compatto in $\mathbb{R}$; quindi $f(K)$ è chiuso e limitato, e quindi ammette massimo e minimo. $\square$

### Proprietà 9.8 — Compattezza e chiusura
- Un compatto è chiuso e limitato (in $\mathbb{R}^n$).
- Un chiuso contenuto in un compatto è compatto.
- L'unione finita di compatti è compatta.
- L'intersezione arbitraria di compatti è compatta.

### Esempio 9.9
- $[0,1]$, $[0,1]^n$, la sfera unitaria $S^{n-1}$ in $\mathbb{R}^n$ sono compatti.
- $(0,1]$, $\mathbb{R}$, $\mathbb{Q}$ non sono compatti.

## 10. Connessione

### Definizione 10.1 — Spazio connesso
Uno spazio metrico $(X,d)$ è **connesso** se non esistono due aperti $A,B$ non vuoti e disgiunti con $X=A\cup B$. Equivalentemente, gli unici sottoinsiemi di $X$ che sono contemporaneamente aperti e chiusi sono $\emptyset$ e $X$.

### Definizione 10.2 — Connessione per archi
$X$ è **connesso per archi** se, per ogni $x,y\in X$, esiste una funzione continua $\gamma:[0,1]\to X$ con $\gamma(0)=x$ e $\gamma(1)=y$.

### Proposizione 10.3
Connessione per archi $\implies$ connessione.

**Dimostrazione.** Sia $X$ connesso per archi e supponiamo $X=A\cup B$ con $A,B$ aperti non vuoti e disgiunti. Siano $a\in A$, $b\in B$ e $\gamma:[0,1]\to X$ continua con $\gamma(0)=a$, $\gamma(1)=b$. Allora $[0,1]=\gamma^{-1}(A)\cup\gamma^{-1}(B)$ con $\gamma^{-1}(A)$, $\gamma^{-1}(B)$ aperti non vuoti e disgiunti in $[0,1]$, che è connesso. Contraddizione. $\square$

### Teorema 10.4 — Immagine continua di connesso
Se $f:X\to Y$ è continua e $X$ è connesso, allora $f(X)$ è connesso.

**Dimostrazione.** Sia $f(X)=A\cup B$ con $A,B$ aperti disgiunti di $f(X)$. Allora $X=f^{-1}(A)\cup f^{-1}(B)$, con $f^{-1}(A)$, $f^{-1}(B)$ aperti disgiunti. Poiché $X$ è connesso, uno è vuoto, quindi uno tra $A,B$ è vuoto. $\square$

### Teorema 10.5 — Teorema del valore intermedio
Se $X$ è connesso, $f:X\to\mathbb{R}$ continua, $x,y\in X$ e $c$ è compreso tra $f(x)$ e $f(y)$, allora esiste $z\in X$ con $f(z)=c$.

**Dimostrazione.** $f(X)$ è connesso in $\mathbb{R}$, quindi è un intervallo. Poiché $f(x),f(y)\in f(X)$, ogni valore tra essi è in $f(X)$. $\square$

### Esempio 10.6
- $\mathbb{R}$ è connesso.
- $\mathbb{Q}$ non è connesso (è totalmente disconnesso).
- Intervalli di $\mathbb{R}$ sono connessi.

## 11. Continuità di funzioni tra spazi metrici

### Definizione 11.1 — Continuità in un punto
$f:X\to Y$ è **continua in** $x_0\in X$ se
$$
\forall\varepsilon>0\ \exists\delta>0:d_X(x,x_0)<\delta,\ x\in X\implies d_Y(f(x),f(x_0))<\varepsilon.
$$

### Definizione 11.2 — Continuità globale
$f$ è **continua** su $X$ se è continua in ogni punto di $X$.

### Teorema 11.3 — Caratterizzazioni equivalenti
Siano $(X,d_X)$, $(Y,d_Y)$ spazi metrici e $f:X\to Y$. Sono equivalenti:
1. $f$ è continua.
2. Per ogni aperto $A\subseteq Y$, $f^{-1}(A)$ è aperto in $X$.
3. Per ogni chiuso $C\subseteq Y$, $f^{-1}(C)$ è chiuso in $X$.
4. Per ogni successione $x_k\to x$ in $X$, $f(x_k)\to f(x)$ in $Y$.

**Dimostrazione.**
(1)$\Rightarrow$(2): Sia $A$ aperto e $x_0\in f^{-1}(A)$. Poiché $f(x_0)\in A$ ed $A$ è aperto, esiste $\varepsilon>0$ con $\mathcal{B}_\varepsilon(f(x_0))\subseteq A$. Per continuità, esiste $\delta>0$ con $f(\mathcal{B}_\delta(x_0))\subseteq\mathcal{B}_\varepsilon(f(x_0))\subseteq A$, quindi $\mathcal{B}_\delta(x_0)\subseteq f^{-1}(A)$.
(2)$\Rightarrow$(3): $C$ chiuso $\Rightarrow$ $C^C$ aperto $\Rightarrow$ $f^{-1}(C^C)=(f^{-1}(C))^C$ aperto $\Rightarrow$ $f^{-1}(C)$ chiuso.
(3)$\Rightarrow$(1): Fissati $x_0$ e $\varepsilon>0$, $\mathcal{B}_\varepsilon(f(x_0))$ è aperto, quindi $f^{-1}(\mathcal{B}_\varepsilon(f(x_0)))$ è aperto e contiene $x_0$; esiste $\delta>0$ con $\mathcal{B}_\delta(x_0)\subseteq f^{-1}(\mathcal{B}_\varepsilon(f(x_0)))$, cioè $f(\mathcal{B}_\delta(x_0))\subseteq\mathcal{B}_\varepsilon(f(x_0))$.
(1)$\Rightarrow$(4): Sia $x_k\to x$. Per $\varepsilon>0$, esiste $\delta>0$ con $d_X(x,x_0)<\delta\Rightarrow d_Y(f(x),f(x_0))<\varepsilon$. Poiché $x_k\to x$, esiste $N$ con $d_X(x_k,x)<\delta$ per $k>N$, quindi $d_Y(f(x_k),f(x))<\varepsilon$.
(4)$\Rightarrow$(1): Per assurdo, se $f$ non fosse continua in $x_0$, esisterebbe $\varepsilon>0$ tale che per ogni $k$ esiste $x_k$ con $d_X(x_k,x_0)<1/k$ e $d_Y(f(x_k),f(x_0))\ge\varepsilon$. Allora $x_k\to x_0$ ma $f(x_k)\not\to f(x_0)$. $\square$

### Teorema 11.4 — Algebra delle funzioni continue
Se $f,g:X\to\mathbb{R}$ sono continue in $x_0$, allora $f+g$, $f\cdot g$ e, se $g(x_0)\neq 0$, $f/g$ sono continue in $x_0$.

### Teorema 11.5 — Composizione
Se $f:X\to Y$ è continua in $x_0$ e $g:Y\to Z$ è continua in $f(x_0)$, allora $g\circ f$ è continua in $x_0$.

**Dimostrazione.** Fissato $\varepsilon>0$, esiste $\eta>0$ con $d_Y(y,f(x_0))<\eta\Rightarrow d_Z(g(y),g(f(x_0)))<\varepsilon$. Esiste $\delta>0$ con $d_X(x,x_0)<\delta\Rightarrow d_Y(f(x),f(x_0))<\eta$. Allora $d_Z(g(f(x)),g(f(x_0)))<\varepsilon$. $\square$

### Teorema 11.6 — Weierstrass
Sia $K\subseteq X$ compatto e $f:K\to\mathbb{R}$ continua. Allora $f$ ammette massimo e minimo su $K$.

**Dimostrazione.** $K$ compatto e $f$ continua implicano $f(K)$ compatto in $\mathbb{R}$, quindi chiuso e limitato, quindi con massimo e minimo. $\square$

### Teorema 11.7 — Immagine continua di compatto
Se $f:X\to Y$ è continua e $K\subseteq X$ è compatto, allora $f(K)$ è compatto.

**Dimostrazione.** Sia $\{A_i\}$ un ricoprimento aperto di $f(K)$. Allora $\{f^{-1}(A_i)\}$ è un ricoprimento aperto di $K$. Per compattezza, esiste un sottoricoprimento finito $f^{-1}(A_{i_1}),\dots,f^{-1}(A_{i_m})$ che ricopre $K$. Allora $A_{i_1},\dots,A_{i_m}$ ricoprono $f(K)$. $\square$

## 12. Continuità uniforme, Lipschitz, Hölder

### Definizione 12.1 — Continuità uniforme
$f:X\to Y$ è **uniformemente continua** se
$$
\forall\varepsilon>0\ \exists\delta>0:x,y\in X,\ d_X(x,y)<\delta\implies d_Y(f(x),f(y))<\varepsilon.
$$

> **Osservazione 12.2 — Differenza con continuità puntuale.**  
> Nella continuità puntuale, $\delta$ dipende da $\varepsilon$ **e da $x_0$**. In quella uniforme, $\delta$ dipende solo da $\varepsilon$.

### Esempio 12.3
$f(x)=x$ è uniformemente continua su $\mathbb{R}$ ($\delta=\varepsilon$).

### Esempio 12.4
$f(x)=x^2$ **non** è uniformemente continua su $\mathbb{R}$: per $\varepsilon=1$, comunque si scelga $\delta$, prendendo $x=1/\delta$, $y=1/\delta+\delta/2$ si ha $|f(x)-f(y)|>1$.

### Esempio 12.5
$f(x)=\sin x$ è uniformemente continua su $\mathbb{R}$: $|\sin x-\sin y|\le|x-y|$.

### Esempio 12.6
$f(x)=\sqrt{x}$ è uniformemente continua su $(0,1)$: $|\sqrt{x}-\sqrt{y}|\le\sqrt{|x-y|}$.

### Teorema 12.7 — Heine–Cantor
Se $f:K\subseteq X\to Y$ è continua su un compatto $K$, allora è uniformemente continua su $K$.

**Dimostrazione.** Supponiamo per assurdo che non lo sia. Esiste $\varepsilon>0$ e successioni $x_k,y_k\in K$ con $d_X(x_k,y_k)<1/k$ e $d_Y(f(x_k),f(y_k))\ge\varepsilon$. Per compattezza, esiste sottosuccessione $x_{k_j}\to x^*\in K$; allora anche $y_{k_j}\to x^*$. Per continuità, $f(x_{k_j})\to f(x^*)$ e $f(y_{k_j})\to f(x^*)$, quindi $d_Y(f(x_{k_j}),f(y_{k_j}))\to 0$, contraddizione. $\square$

### Definizione 12.8 — Lipschitzianità
$f:X\to Y$ è **Lipschitziana** di costante $L\ge 0$ se
$$
d_Y(f(x),f(y))\le L\,d_X(x,y)\quad\forall x,y\in X.
$$

### Definizione 12.9 — Hölderianità
$f$ è **Hölderiana** di esponente $\alpha\in(0,1]$ e costante $L$ se
$$
d_Y(f(x),f(y))\le L\,d_X(x,y)^\alpha.
$$

### Proposizione 12.10 — Gerarchia
$$
\text{Lipschitziana}\implies\text{Hölderiana}\implies\text{uniformemente continua}\implies\text{continua}.
$$
Nessuna implicazione è reversibile.

**Dimostrazione.** Lipschitziana $\Rightarrow$ Hölderiana con $\alpha=1$. Hölderiana $\Rightarrow$ uniformemente continua: scelto $\delta=(\varepsilon/L)^{1/\alpha}$, $d_X(x,y)<\delta\Rightarrow d_Y(f(x),f(y))\le L\delta^\alpha=\varepsilon$. Uniformemente continua $\Rightarrow$ continua ovvia. $\square$

### Esempi 12.11
- La norma $\|\cdot\|:\mathbb{R}^n\to\mathbb{R}$ è Lipschitziana di costante $1$.
- Proiezioni $\pi_i(x)=x_i$ sono Lipschitziane di costante $1$.
- Applicazioni lineari $T:\mathbb{R}^n\to\mathbb{R}^m$ sono Lipschitziane con $L=\|T\|_{\text{op}}$.
- $f(x)=\sqrt{|x|}$ è Hölderiana di esponente $1/2$ ma non Lipschitziana in $0$.
- $f(x)=1/\ln x$ su $(0,1/2]$ è uniformemente continua ma non Hölderiana.

### Teorema 12.12 — Funzioni $C^1$ su convessi
Sia $f:\Omega\subseteq\mathbb{R}^n\to\mathbb{R}^m$ di classe $C^1$ su un aperto convesso $\Omega$, con $\|J_f\|$ limitato su $\Omega$. Allora $f$ è Lipschitziana su $\Omega$.

**Dimostrazione.** Per il teorema del valor medio applicato a $g(t)=f(y+t(x-y))$,
$$
f(x)-f(y)=\int_0^1 J_f(y+t(x-y))(x-y)\,dt,
$$
quindi
$$
\|f(x)-f(y)\|\le\sup_\Omega\|J_f\|\cdot\|x-y\|.
$$
$\square$

## 13. Contrazioni e teorema del punto fisso

### Definizione 13.1 — Contrazione
$f:X\to X$ è una **contrazione** se esiste $0\le L<1$ con
$$
d(f(x),f(y))\le L\,d(x,y)\quad\forall x,y\in X.
$$

### Teorema 13.2 — Banach–Caccioppoli (punto fisso)
Sia $(X,d)$ completo e $f:X\to X$ una contrazione. Allora esiste un unico $x^*\in X$ con $f(x^*)=x^*$. Inoltre, per ogni $x_0\in X$, la successione $x_{k+1}=f(x_k)$ converge a $x^*$.

**Dimostrazione.** 
*Esistenza.* Sia $x_0\in X$ e $x_{k+1}=f(x_k)$. Allora
$$
d(x_{k+1},x_k)=d(f(x_k),f(x_{k-1}))\le L\,d(x_k,x_{k-1})\le\dots\le L^k d(x_1,x_0).
$$
Per $m>k$,
$$
d(x_m,x_k)\le\sum_{j=k}^{m-1}d(x_{j+1},x_j)\le d(x_1,x_0)\sum_{j=k}^{m-1}L^j\le\frac{L^k}{1-L}d(x_1,x_0)\to 0.
$$
Quindi $\{x_k\}$ è di Cauchy; per completezza, $x_k\to x^*\in X$. Per continuità di $f$ (è Lipschitziana), $f(x^*)=x^*$.

*Unicità.* Se $f(x^*)=x^*$ e $f(y^*)=y^*$, allora
$$
d(x^*,y^*)=d(f(x^*),f(y^*))\le L\,d(x^*,y^*)\implies d(x^*,y^*)=0.
$$
$\square$

### Esempio 13.3 — Applicazione a equazioni differenziali
Il teorema delle contrazioni è alla base del teorema di esistenza e unicità di Cauchy–Lipschitz per equazioni differenziali ordinarie.

## 14. Isometrie e omeomorfismi

### Definizione 14.1 — Isometria
$f:(X,d_X)\to(Y,d_Y)$ è un'**isometria** se
$$
d_Y(f(x),f(y))=d_X(x,y)\quad\forall x,y\in X.
$$

### Definizione 14.2 — Omeomorfismo
$f:X\to Y$ è un **omeomorfismo** se è biettiva, continua e con inversa continua. In tal caso $X$ e $Y$ sono **omeomorfi**.

### Osservazione 14.3
- Ogni isometria è un omeomorfismo (ma non viceversa).
- Un omeomorfismo preserva le proprietà topologiche (aperti, chiusi, compattezza, connessione), ma non necessariamente le proprietà metriche (distanze, limitatezza).

### Esempio 14.4
$f(x)=\arctan x$ è un omeomorfismo tra $\mathbb{R}$ e $(-\pi/2,\pi/2)$, ma non è un'isometria.

## 15. Equivalenza di norme in $\mathbb{R}^n$

### Definizione 15.1 — Norme equivalenti
Due norme $\|\cdot\|_a$, $\|\cdot\|_b$ su $X$ sono **equivalenti** se esistono $c,C>0$ con
$$
c\|x\|_a\le\|x\|_b\le C\|x\|_a\quad\forall x\in X.
$$

### Proposizione 15.2
Su $\mathbb{R}^n$,
$$
\|x\|_\infty\le\|x\|_2\le\|x\|_1\le n\|x\|_\infty.
$$

**Dimostrazione.** 
- $\|x\|_\infty\le\|x\|_2$: sia $j$ con $|x_j|=\|x\|_\infty$; allora $\|x\|_\infty^2=|x_j|^2\le\sum_i|x_i|^2=\|x\|_2^2$.
- $\|x\|_2\le\|x\|_1$: $\|x\|_2^2=\sum_i|x_i|^2\le(\sum_i|x_i|)^2=\|x\|_1^2$.
- $\|x\|_1\le n\|x\|_\infty$: $\|x\|_1=\sum_i|x_i|\le n\|x\|_\infty$. $\square$

### Teorema 15.3
Su $\mathbb{R}^n$, tutte le norme sono equivalenti.

**Idea.** Si fissa $\|\cdot\|_2$ e si mostra che ogni altra norma $\|\cdot\|$ è equivalente. La sfera unitaria $S=\{x:\|x\|_2=1\}$ è compatta. La funzione $\|\cdot\|:S\to\mathbb{R}$ è continua (per la disuguaglianza triangolare inversa), quindi ammette minimo $m>0$ e massimo $M>0$ su $S$. Per ogni $x\neq 0$, $\frac{x}{\|x\|_2}\in S$, quindi
$$
m\le\left\|\frac{x}{\|x\|_2}\right\|\le M\implies m\|x\|_2\le\|x\|\le M\|x\|_2.
$$
$\square$

> **Osservazione 15.4 — Topologia euclidea standard.**  
> Poiché tutte le norme su $\mathbb{R}^n$ sono equivalenti, inducono la stessa topologia (stessi aperti, chiusi, convergenza, continuità). Questa è detta **topologia euclidea standard**. La scelta della norma è irrilevante.

> **Osservazione 15.5 — Dimensione infinita.**  
> In dimensione infinita, l'equivalenza delle norme **fallisce**.

## 16. Limiti di funzioni tra spazi metrici

### Definizione 16.1 — Limite
Siano $(X,d_X)$, $(Y,d_Y)$ spazi metrici, $f:\operatorname{dom}f\subseteq X\to Y$, e $x_0$ di accumulazione per $\operatorname{dom}f$. Si dice che
$$
\lim_{x\to x_0}f(x)=l\in Y
$$
se per ogni $\varepsilon>0$ esiste $\delta>0$ con
$$
x\in\operatorname{dom}f,\ 0<d_X(x,x_0)<\delta\implies d_Y(f(x),l)<\varepsilon.
$$

> **Osservazione 16.2.**  
> La condizione $x\neq x_0$ è essenziale: il valore $f(x_0)$ non influenza il limite, e $f$ può non essere definita in $x_0$.

### Teorema 16.3 — Unicità del limite
Se $f\to l_1$ e $f\to l_2$, allora $l_1=l_2$.

**Dimostrazione.** Sia $\varepsilon=d_Y(l_1,l_2)/3$. Esistono $\delta_1,\delta_2$ con $d_Y(f(x),l_1)<\varepsilon$ per $0<d_X(x,x_0)<\delta_1$ e $d_Y(f(x),l_2)<\varepsilon$ per $0<d_X(x,x_0)<\delta_2$. Per $x$ con $0<d_X(x,x_0)<\min\{\delta_1,\delta_2\}$,
$$
d_Y(l_1,l_2)\le d_Y(l_1,f(x))+d_Y(f(x),l_2)<2\varepsilon=\tfrac{2}{3}d_Y(l_1,l_2),
$$
contraddizione. $\square$

### Proposizione 16.4 — Caratterizzazione sequenziale
$f\to l$ se e solo se per ogni successione $x_k\to x_0$ con $x_k\in\operatorname{dom}f\setminus\{x_0\}$, $f(x_k)\to l$.

**Dimostrazione.** 
$(\Rightarrow)$ Per $\varepsilon>0$, esiste $\delta>0$ con $0<d_X(x,x_0)<\delta\Rightarrow d_Y(f(x),l)<\varepsilon$. Poiché $x_k\to x_0$, esiste $N$ con $0<d_X(x_k,x_0)<\delta$ per $k>N$, quindi $d_Y(f(x_k),l)<\varepsilon$.
$(\Leftarrow)$ Per assurdo, se $f\not\to l$, esiste $\varepsilon>0$ tale che per ogni $\delta>0$ esiste $x$ con $0<d_X(x,x_0)<\delta$ e $d_Y(f(x),l)\ge\varepsilon$. Scegliendo $\delta=1/k$, si costruisce $x_k\to x_0$ con $f(x_k)\not\to l$. $\square$

### Teorema 16.5 — Algebra dei limiti
Se $f\to l_1$, $g\to l_2$, allora $f+g\to l_1+l_2$, $fg\to l_1l_2$, $f/g\to l_1/l_2$ (se $l_2\neq 0$).

### Teorema 16.6 — Confronto
Se $f(x)\le g(x)\le h(x)$ in un intorno di $x_0$ e $f,h\to l$, allora $g\to l$.

### Teorema 16.7 — Composizione
Se $f\to y_0$ per $x\to x_0$ e $g\to l$ per $y\to y_0$, con $f(x)\neq y_0$ in un intorno di $x_0$, allora $g\circ f\to l$.
