# Limiti e continuità di funzioni da $\mathbb{R}^n$ a $\mathbb{R}^m$

---

## 0. Preliminari: norme e topologia in $\mathbb{R}^n$

### 0.1 Norme su $\mathbb{R}^n$

Su $\mathbb{R}^n$ consideriamo le norme usuali:
$$
\|x\|_1 = \sum_{i=1}^n |x_i|, \qquad
\|x\|_2 = \left(\sum_{i=1}^n |x_i|^2\right)^{1/2}, \qquad
\|x\|_\infty = \max_{1\le i\le n} |x_i|.
$$

#### Teorema 0.1 — Equivalenza delle norme

Per ogni $x \in \mathbb{R}^n$,
$$
\|x\|_\infty \le \|x\|_2 \le \|x\|_1 \le n\|x\|_\infty.
$$

**Dimostrazione.**

*Prima disuguaglianza:* $\|x\|_\infty \le \|x\|_2$.  
Sia $j$ l'indice per cui $|x_j| = \max_i |x_i| = \|x\|_\infty$. Allora
$$
\|x\|_\infty^2 = |x_j|^2 \le \sum_{i=1}^n |x_i|^2 = \|x\|_2^2.
$$
Prendendo la radice quadrata (entrambi i membri sono non negativi) si ottiene $\|x\|_\infty \le \|x\|_2$.

*Seconda disuguaglianza:* $\|x\|_2 \le \|x\|_1$.  
Poiché $|x_i| \ge 0$ per ogni $i$, si ha
$$
\|x\|_2^2 = \sum_{i=1}^n |x_i|^2 \le \sum_{i=1}^n |x_i|^2 + 2\sum_{i<j} |x_i||x_j| = \left(\sum_{i=1}^n |x_i|\right)^2 = \|x\|_1^2.
$$
Prendendo la radice quadrata si ottiene $\|x\|_2 \le \|x\|_1$.

*Terza disuguaglianza:* $\|x\|_1 \le n\|x\|_\infty$.  
Per ogni $i$, $|x_i| \le \max_j |x_j| = \|x\|_\infty$. Quindi
$$
\|x\|_1 = \sum_{i=1}^n |x_i| \le \sum_{i=1}^n \|x\|_\infty = n\|x\|_\infty.
$$

$\square$

> **Osservazione 0.2 — Catena completa.**
> Le tre disuguaglianze si possono concatenare:
> $$
> \|x\|_\infty \le \|x\|_2 \le \|x\|_1 \le n\|x\|_\infty \le n\|x\|_2 \le n^2\|x\|_\infty.
> $$
> Da cui segue che $\frac{1}{n}\|x\|_1 \le \|x\|_\infty \le \|x\|_1$, cioè $\|x\|_\infty$ e $\|x\|_1$ sono equivalenti con costanti $1$ e $n$.

> **Osservazione 0.3 — Perché l'equivalenza è cruciale.**
> L'equivalenza delle norme significa che esiste $C \ge 1$ tale che
> $$
> \frac{1}{C}\|x\|_a \le \|x\|_b \le C\|x\|_a \qquad \forall x \in \mathbb{R}^n.
> $$
> Quindi:
> - se $\|x_k - x\|_a \to 0$, allora $\|x_k - x\|_b \to 0$;
> - un insieme è aperto rispetto a $\|\cdot\|_a$ se e solo se lo è rispetto a $\|\cdot\|_b$;
> - le palle di raggio $r$ rispetto a $\|\cdot\|_a$ contengono una palla di raggio $r/C$ rispetto a $\|\cdot\|_b$.
>
> In dimensione infinita questa proprietà **fallisce**: esistono norme non equivalenti che generano topologie diverse.

### 0.2 Topologia metrica indotta

La metrica indotta da una norma è $d(x,y) = \|x-y\|$. Gli **aperti** sono le unioni arbitrarie di palle
$$
\mathcal{B}_r(x_0) = \{x \in \mathbb{R}^n : \|x-x_0\| < r\}.
$$
$A \subseteq \mathbb{R}^n$ è **aperto** se
$$
\forall x_0 \in A\ \exists r > 0 : \mathcal{B}_r(x_0) \subseteq A.
$$
$C \subseteq \mathbb{R}^n$ è **chiuso** se $C^C$ è aperto. Un insieme $K$ è **compatto** se ogni ricoprimento aperto ammette un sottoricoprimento finito; in $\mathbb{R}^n$, per il teorema di Heine–Borel, $K$ è compatto se e solo se è chiuso e limitato.

### 0.3 Successioni e convergenza

Una successione $\{x_k\} \subseteq \mathbb{R}^n$ converge a $x \in \mathbb{R}^n$, e si scrive $x_k \to x$, se
$$
\forall \varepsilon > 0\ \exists N \in \mathbb{N} : k > N \implies \|x_k - x\| < \varepsilon.
$$

**Proposizione 0.4 — Convergenza componente per componente.**
$x_k \to x$ in $\mathbb{R}^n$ se e solo se $x_{k,i} \to x_i$ in $\mathbb{R}$ per ogni $i = 1, \dots, n$.

**Dimostrazione.**

$(\Rightarrow)$ Per ogni $i$,
$$
|x_{k,i} - x_i| \le \|x_k - x\|_\infty \le \|x_k - x\|_2 \to 0.
$$

$(\Leftarrow)$ Se $x_{k,i} \to x_i$ per ogni $i$, allora per ogni $\varepsilon > 0$ esiste $N_i$ tale che $k > N_i \implies |x_{k,i} - x_i| < \varepsilon$. Posto $N = \max_i N_i$, per $k > N$ si ha
$$
\|x_k - x\|_\infty = \max_i |x_{k,i} - x_i| < \varepsilon,
$$
quindi $x_k \to x$. $\square$

## 1. Limiti di funzioni

### 1.1 Punti di accumulazione

**Definizione 1.1.** Sia $(X,d)$ uno spazio metrico e $A \subseteq X$. Un punto $x_0 \in X$ è **di accumulazione** per $A$ se
$$
\forall r > 0,\quad \mathcal{B}_r(x_0) \cap (A \setminus \{x_0\}) \neq \emptyset.
$$
Equivalentemente, $x_0 \in \overline{A \setminus \{x_0\}}$.

> **Osservazione 1.2.** La condizione che $x_0$ sia di accumulazione per $\operatorname{dom}f$ è essenziale nella definizione di limite: se $x_0$ fosse isolato, l'insieme $\{x \in \operatorname{dom}f : 0 < \|x-x_0\| < \delta\}$ sarebbe vuoto per $\delta$ piccolo, e la condizione di limite sarebbe vacuamente vera per qualsiasi $l$.

**Esempio 1.3.** In $\mathbb{R}$, $0$ è di accumulazione per $(0,1)$ ma $0 \notin (0,1)$. In $\mathbb{R}^2$, $(0,0)$ è di accumulazione per $\{(1/n,0) : n \in \mathbb{N}\}$ e per ogni palla $\mathcal{B}_r((0,0))$ interseca l'insieme in infiniti punti.

**Esempio 1.4.** In $\mathbb{R}^2$, l'insieme $\mathbb{Z}^2$ non ha punti di accumulazione in $\mathbb{R}^2$: ogni punto è isolato. Ogni punto $(m,n) \in \mathbb{Z}^2$ è isolato con raggio $r = 1/2$.

### 1.2 Definizione di limite

**Definizione 1.5.** Siano $(X, d_X)$ e $(Y, d_Y)$ spazi metrici, $f : \operatorname{dom}f \subseteq X \to Y$, e $x_0$ di accumulazione per $\operatorname{dom}f$. Si dice che
$$
\lim_{x \to x_0} f(x) = l \in Y
$$
se per ogni intorno $V$ di $l$ in $Y$ esiste un intorno $U$ di $x_0$ in $X$ tale che
$$
x \in U \cap \operatorname{dom}f,\ x \neq x_0 \implies f(x) \in V.
$$

Equivalentemente, in termini di distanze:
$$
\forall \varepsilon > 0\ \exists \delta > 0 : x \in \operatorname{dom}f,\ 0 < d_X(x,x_0) < \delta \implies d_Y(f(x), l) < \varepsilon.
$$

Nel caso $X = \mathbb{R}^n$, $Y = \mathbb{R}^m$:
$$
\forall \varepsilon > 0\ \exists \delta > 0 : 0 < \|x - x_0\| < \delta \implies \|f(x) - l\| < \varepsilon.
$$

> **Osservazione 1.6.** Nella definizione:
> - la condizione $x \neq x_0$ è essenziale: il valore $f(x_0)$ non influenza il limite, e $f$ può non essere definita in $x_0$;
> - la scelta della norma è irrilevante per l'equivalenza delle norme in $\mathbb{R}^n$;
> - la definizione è **locale**: dipende solo dal comportamento di $f$ in un intorno di $x_0$.

### 1.3 Limite all'infinito

**Definizione 1.7.** Sia $f : \operatorname{dom}f \subseteq \mathbb{R}^n \to \mathbb{R}^m$. Si dice che
$$
\lim_{\|x\| \to +\infty} f(x) = l
$$
se
$$
\forall \varepsilon > 0\ \exists R > 0 : x \in \operatorname{dom}f,\ \|x\| > R \implies \|f(x) - l\| < \varepsilon.
$$

**Esempio 1.8.**
$$
\lim_{(x,y) \to +\infty} \frac{1}{x^2 + y^2} = 0.
$$

**Dimostrazione.** Fissato $\varepsilon > 0$, scegliamo $R = 1/\sqrt{\varepsilon}$. Se $\|(x,y)\|_2 = \sqrt{x^2+y^2} > R$, allora $x^2+y^2 > 1/\varepsilon$, quindi
$$
\left|\frac{1}{x^2+y^2} - 0\right| = \frac{1}{x^2+y^2} < \varepsilon.
$$
$\square$

**Esempio 1.9.**
$$
\lim_{(x,y) \to +\infty} \frac{x+y}{x^2+y^2} = 0.
$$

**Dimostrazione.** Per ogni $(x,y) \ne (0,0)$,
$$
\left|\frac{x+y}{x^2+y^2}\right| \le \frac{|x|+|y|}{x^2+y^2} \le \frac{\sqrt{2}\sqrt{x^2+y^2}}{x^2+y^2} = \frac{\sqrt{2}}{\sqrt{x^2+y^2}},
$$
dove la seconda disuguaglianza segue da $|x|+|y| \le \sqrt{2}\sqrt{x^2+y^2}$ (disuguaglianza di Cauchy–Schwarz). Fissato $\varepsilon > 0$, scegliamo $R = \sqrt{2}/\varepsilon$. Se $\|(x,y)\|_2 > R$, allora $\sqrt{2}/\sqrt{x^2+y^2} < \varepsilon$, da cui la tesi. $\square$

### 1.4 Limite e restrizioni

**Teorema 1.10 (Teorema delle restrizioni).** Sia $f : \operatorname{dom}f \subseteq \mathbb{R}^n \to \mathbb{R}$ e $x_0$ di accumulazione per $\operatorname{dom}f$. Allora
$$
\lim_{x \to x_0} f(x) = l
\iff
\lim_{\substack{x \to x_0 \\ x \in A}} f(x) = l \qquad \forall A \subseteq \operatorname{dom}f \text{ con } x_0 \in A'.
$$

**Dimostrazione.**

$(\Rightarrow)$ Sia $A \subseteq \operatorname{dom}f$ con $x_0$ di accumulazione per $A$. Fissato $\varepsilon > 0$, per definizione di limite esiste $\delta > 0$ tale che
$$
x \in \operatorname{dom}f,\ 0 < \|x - x_0\| < \delta \implies |f(x) - l| < \varepsilon.
$$
In particolare, se $x \in A \subseteq \operatorname{dom}f$ e $0 < \|x-x_0\| < \delta$, la condizione vale. Quindi $\lim_{x \to x_0} f|_A(x) = l$.

$(\Leftarrow)$ Prendendo $A = \operatorname{dom}f$ si ottiene la tesi. $\square$

**Corollario 1.11 (Criterio di non esistenza).** Se esistono $A, B \subseteq \operatorname{dom}f$ con $x_0$ di accumulazione per entrambi e
$$
\lim_{\substack{x \to x_0 \\ x \in A}} f(x) \neq \lim_{\substack{x \to x_0 \\ x \in B}} f(x),
$$
allora $\lim_{x \to x_0} f(x)$ non esiste.

**Dimostrazione.** Se il limite esistesse, per il Teorema 1.10 tutte le restrizioni avrebbero lo stesso limite, contraddizione. $\square$

### 1.5 Caratterizzazione sequenziale

**Teorema 1.12.** Sia $f : \operatorname{dom}f \subseteq \mathbb{R}^n \to \mathbb{R}^m$ e $x_0$ di accumulazione per $\operatorname{dom}f$. Allora
$$
\lim_{x \to x_0} f(x) = l
\iff
\forall \{x_k\} \subseteq \operatorname{dom}f \setminus \{x_0\},\ x_k \to x_0 \implies f(x_k) \to l.
$$

**Dimostrazione.**

$(\Rightarrow)$ Fissato $\varepsilon > 0$, per definizione di limite esiste $\delta > 0$ tale che
$$
x \in \operatorname{dom}f,\ 0 < \|x - x_0\| < \delta \implies \|f(x) - l\| < \varepsilon.
$$
Poiché $x_k \to x_0$ e $x_k \ne x_0$, esiste $N$ tale che per $k > N$, $0 < \|x_k - x_0\| < \delta$. Allora per $k > N$, $\|f(x_k) - l\| < \varepsilon$. Dunque $f(x_k) \to l$.

$(\Leftarrow)$ Supponiamo per assurdo che $f(x) \not\to l$. Allora esiste $\varepsilon > 0$ tale che per ogni $\delta > 0$ esiste $x \in \operatorname{dom}f$ con $0 < \|x-x_0\| < \delta$ e $\|f(x)-l\| \ge \varepsilon$. Scegliendo successivamente $\delta = 1/k$, costruiamo una successione $x_k \in \operatorname{dom}f \setminus \{x_0\}$ con $\|x_k - x_0\| < 1/k$ e $\|f(x_k) - l\| \ge \varepsilon$. Allora $x_k \to x_0$ ma $f(x_k) \not\to l$, contraddizione. $\square$

> **Osservazione 1.13.** La caratterizzazione sequenziale è spesso più **comoda** della definizione $\varepsilon$-$\delta$ perché permette di usare tutti i teoremi sulle successioni (ad esempio, la legge dei grandi numeri, il teorema del confronto per successioni, ecc.).

### 1.6 Esempi di limiti che non esistono

**Esempio 1.14.** $\displaystyle\lim_{(x,y) \to (0,0)} \frac{xy}{x^2+y^2}$ non esiste.

**Dimostrazione.** Lungo $y = mx$ con $m \ne 0$:
$$
f(x, mx) = \frac{m x^2}{x^2(1+m^2)} = \frac{m}{1+m^2}.
$$
Il valore dipende da $m$: per $m = 1$ vale $1/2$; per $m = 0$ vale $0$. Poiché due restrizioni danno limiti diversi, per il Corollario 1.11 il limite non esiste. $\square$

**Esempio 1.15.** $\displaystyle\lim_{(x,y) \to (0,0)} \frac{x^2 - y^2}{x^2+y^2}$ non esiste.

**Dimostrazione.** Lungo $y = 0$: $f(x,0) = 1 \to 1$. Lungo $x = 0$: $f(0,y) = -1 \to -1$. Poiché $1 \neq -1$, il limite non esiste. $\square$

**Esempio 1.16.** $\displaystyle f(x,y) = \frac{x^2 y}{x^4 + y^2}$ non ammette limite in $(0,0)$.

**Dimostrazione.** Lungo $y = mx$:
$$
f(x,mx) = \frac{m x^3}{x^4 + m^2 x^2} = \frac{m x}{x^2 + m^2} \to 0.
$$
Lungo $y = x^2$:
$$
f(x,x^2) = \frac{x^4}{x^4 + x^4} = \frac{1}{2} \neq 0.
$$
Poiché lungo le rette il limite è $0$ e lungo la parabola è $1/2$, il limite non esiste. $\square$

> **Osservazione 1.17.** L'Esempio 1.16 mostra che **la sola convergenza lungo tutte le rette non è sufficiente** per garantire l'esistenza del limite. Esistono funzioni che convergono a $0$ lungo ogni retta passante per l'origine ma non hanno limite. Questo è un fenomeno tipico di $\mathbb{R}^2$ che non ha controparte in $\mathbb{R}$.

**Esempio 1.18.** $\displaystyle\lim_{(x,y) \to (0,0)} \frac{x^3 y}{x^4 + y^4}$ non esiste.

**Dimostrazione.** In coordinate polari:
$$
f(\rho\cos\theta, \rho\sin\theta) = \frac{\rho^4 \cos^3\theta \sin\theta}{\rho^4(\cos^4\theta + \sin^4\theta)} = \frac{\cos^3\theta \sin\theta}{\cos^4\theta + \sin^4\theta}.
$$
Il valore dipende da $\theta$: per $\theta = 0$ vale $0$; per $\theta = \pi/4$ vale $\frac{(1/\sqrt{2})^4}{2(1/4)} = \frac{1/4}{1/2} = \frac{1}{2}$. Poiché due direzioni danno limiti diversi, il limite non esiste. $\square$

## 2. Teoremi sui limiti

Siano $f, g, h : X \subseteq \mathbb{R}^n \to \mathbb{R}$ e $x_0$ di accumulazione per $X$.

### 2.1 Unicità del limite

**Teorema 2.1.** Se $\lim_{x \to x_0} f(x) = l_1$ e $\lim_{x \to x_0} f(x) = l_2$, allora $l_1 = l_2$.

**Dimostrazione.** Supponiamo $l_1 \neq l_2$ e sia $\varepsilon = |l_1 - l_2|/3 > 0$. Esistono $\delta_1, \delta_2 > 0$ tali che
$$
0 < \|x - x_0\| < \delta_1 \implies |f(x) - l_1| < \varepsilon,
$$
$$
0 < \|x - x_0\| < \delta_2 \implies |f(x) - l_2| < \varepsilon.
$$
Poiché $x_0$ è di accumulazione per $X$, esiste $x \in X$ con $0 < \|x-x_0\| < \min\{\delta_1, \delta_2\}$. Allora
$$
|l_1 - l_2| = |l_1 - f(x) + f(x) - l_2| \le |l_1 - f(x)| + |f(x) - l_2| < 2\varepsilon = \frac{2}{3}|l_1 - l_2|,
$$
dove la prima disuguaglianza è la **disuguaglianza triangolare**. Da $|l_1-l_2| < \frac{2}{3}|l_1-l_2|$ segue $|l_1-l_2| = 0$, cioè $l_1 = l_2$, contraddizione. $\square$

> **Osservazione 2.2.** L'unicità del limite giustifica la scrittura $\lim_{x \to x_0} f(x) = l$: se il limite esiste, è unico e quindi la notazione è ben posta.

### 2.2 Permanenza del segno

**Teorema 2.3.** Se $\lim_{x \to x_0} f(x) = l$ e $l > 0$ (risp. $l < 0$), esiste un intorno $U$ di $x_0$ tale che $f(x) > 0$ (risp. $f(x) < 0$) per ogni $x \in (U \cap X) \setminus \{x_0\}$.

**Dimostrazione.** Supponiamo $l > 0$ e scegliamo $\varepsilon = l/2 > 0$. Per definizione di limite, esiste $\delta > 0$ tale che
$$
0 < \|x - x_0\| < \delta \implies |f(x) - l| < \frac{l}{2}.
$$
Da $|f(x) - l| < l/2$ segue
$$
f(x) > l - \frac{l}{2} = \frac{l}{2} > 0.
$$
Il caso $l < 0$ è simmetrico, applicando lo stesso argomento a $-f$ (il cui limite è $-l > 0$) oppure scegliendo $\varepsilon = -l/2$. $\square$

**Corollario 2.4.** Se $f(x) \le g(x)$ in un intorno di $x_0$ (escluso al più $x_0$), $\lim_{x \to x_0} f(x) = l_1$, $\lim_{x \to x_0} g(x) = l_2$, allora $l_1 \le l_2$.

**Dimostrazione.** Supponiamo per assurdo $l_1 > l_2$. Allora $l_1 - l_2 > 0$. Applicando il Teorema 2.3 a $f - g$ (il cui limite è $l_1 - l_2 > 0$), esiste un intorno $U$ di $x_0$ tale che
$$
f(x) - g(x) > 0 \qquad \forall x \in (U \cap X) \setminus \{x_0\},
$$
contraddicendo $f(x) \le g(x)$. $\square$

> **Osservazione 2.5.** Il Corollario 2.4 è **stretto**: dalle ipotesi segue solo $l_1 \le l_2$, non $l_1 < l_2$. Ad esempio, $f(x) = 0$ e $g(x) = |x|$ hanno $f \le g$ in un intorno di $0$, e $\lim f = \lim g = 0$.

### 2.3 Algebra dei limiti

**Teorema 2.6.** Se $\lim_{x \to x_0} f(x) = l_1$ e $\lim_{x \to x_0} g(x) = l_2$, allora:
1. $\lim_{x \to x_0} (f + g)(x) = l_1 + l_2$;
2. $\lim_{x \to x_0} (f \cdot g)(x) = l_1 l_2$;
3. se $l_2 \neq 0$, $\lim_{x \to x_0} (f/g)(x) = l_1/l_2$, con $g \neq 0$ in un intorno di $x_0$;
4. $\lim_{x \to x_0} (\lambda f)(x) = \lambda l_1$ per ogni $\lambda \in \mathbb{R}$.

**Dimostrazione.**

**(1) Somma.** Fissato $\varepsilon > 0$, esistono $\delta_1, \delta_2 > 0$ con
$$
0 < \|x-x_0\| < \delta_1 \implies |f(x) - l_1| < \varepsilon/2,
$$
$$
0 < \|x-x_0\| < \delta_2 \implies |g(x) - l_2| < \varepsilon/2.
$$
Per $\delta = \min\{\delta_1, \delta_2\}$, se $0 < \|x-x_0\| < \delta$,
$$
|f(x) + g(x) - (l_1 + l_2)| \le |f(x) - l_1| + |g(x) - l_2| < \varepsilon,
$$
dove la disuguaglianza è **triangolare**.

**(2) Prodotto.** Scriviamo
$$
f(x)g(x) - l_1 l_2 = f(x)(g(x) - l_2) + l_2(f(x) - l_1).
$$
Allora
$$
|f(x)g(x) - l_1 l_2| \le |f(x)| |g(x) - l_2| + |l_2| |f(x) - l_1|.
$$
Per il Teorema 3.8 (limitatezza locale), esiste $M > 0$ e un intorno $U$ di $x_0$ con $|f(x)| \le M$ per $x \in (U \cap X) \setminus \{x_0\}$. Fissato $\varepsilon > 0$, scegliamo $\delta_1, \delta_2 > 0$ tali che
$$
0 < \|x-x_0\| < \delta_1 \implies |g(x) - l_2| < \frac{\varepsilon}{2M},
$$
$$
0 < \|x-x_0\| < \delta_2 \implies |f(x) - l_1| < \frac{\varepsilon}{2(|l_2|+1)}.
$$
Per $\delta = \min\{\delta_1, \delta_2\}$ e $x$ nell'intorno $U$, si ha $|f(x)g(x) - l_1 l_2| < \varepsilon$.

**(3) Quoziente.** Per il Teorema 2.3 applicato a $g$ con $l_2 > 0$ (o simmetricamente $l_2 < 0$), esiste un intorno $U$ di $x_0$ con $g(x) \neq 0$ per $x \in (U \cap X) \setminus \{x_0\}$. In tale intorno,
$$
\frac{1}{g(x)} - \frac{1}{l_2} = \frac{l_2 - g(x)}{g(x) l_2}.
$$
Poiché $|g(x)| \to |l_2| > 0$, $|g(x)| \ge |l_2|/2$ in un intorno sufficientemente piccolo di $x_0$ (per il Teorema 2.3 applicato a $|g| - |l_2|/2$, che ha limite $|l_2|/2 > 0$). Allora
$$
\left|\frac{1}{g(x)} - \frac{1}{l_2}\right| \le \frac{|l_2 - g(x)|}{|g(x)||l_2|} \le \frac{2|l_2 - g(x)|}{|l_2|^2} \to 0.
$$
Quindi $1/g \to 1/l_2$. Applicando (2) a $f$ e $1/g$ si conclude.

**(4) Prodotto per scalare.** Segue da (2) con $g(x) \equiv \lambda$. $\square$

### 2.4 Teorema del confronto

**Teorema 2.7 (Due carabinieri).** Se $f(x) \le g(x) \le h(x)$ in un intorno di $x_0$ (escluso al più $x_0$) e $\lim_{x \to x_0} f(x) = \lim_{x \to x_0} h(x) = l$, allora $\lim_{x \to x_0} g(x) = l$.

**Dimostrazione.** Fissato $\varepsilon > 0$, esistono $\delta_1, \delta_2 > 0$ con
$$
0 < \|x-x_0\| < \delta_1 \implies |f(x) - l| < \varepsilon \iff l - \varepsilon < f(x) < l + \varepsilon,
$$
$$
0 < \|x-x_0\| < \delta_2 \implies |h(x) - l| < \varepsilon \iff l - \varepsilon < h(x) < l + \varepsilon.
$$
Sia $\delta = \min\{\delta_1, \delta_2\}$. Per $0 < \|x-x_0\| < \delta$, usando l'ipotesi $f(x) \le g(x) \le h(x)$,
$$
l - \varepsilon < f(x) \le g(x) \le h(x) < l + \varepsilon,
$$
quindi $|g(x) - l| < \varepsilon$. $\square$

### 2.5 Limite della funzione composta

**Teorema 2.8.** Siano $f : \operatorname{dom}f \subseteq X \to Y$ e $g : \operatorname{dom}g \subseteq Y \to Z$. Sia $x_0$ di accumulazione per $\operatorname{dom}f$, $\lim_{x \to x_0} f(x) = y_0$ con $y_0$ di accumulazione per $\operatorname{dom}g$, e $\lim_{y \to y_0} g(y) = l$. Se esiste un intorno $U$ di $x_0$ tale che $f(x) \neq y_0$ per ogni $x \in (U \cap \operatorname{dom}f) \setminus \{x_0\}$, allora
$$
\lim_{x \to x_0} (g \circ f)(x) = l.
$$

**Dimostrazione.** Fissato $\varepsilon > 0$, poiché $\lim_{y \to y_0} g(y) = l$, esiste $\eta > 0$ tale che
$$
y \in \operatorname{dom}g,\ 0 < d_Y(y, y_0) < \eta \implies d_Z(g(y), l) < \varepsilon.
$$
Poiché $\lim_{x \to x_0} f(x) = y_0$, esiste $\delta > 0$ tale che
$$
x \in \operatorname{dom}f,\ 0 < d_X(x, x_0) < \delta \implies d_Y(f(x), y_0) < \eta.
$$
Per l'ipotesi aggiuntiva, possiamo supporre (restringendo $\delta$) che $f(x) \neq y_0$ per ogni $x \in \operatorname{dom}f$ con $0 < d_X(x, x_0) < \delta$. Allora per tali $x$ si ha $0 < d_Y(f(x), y_0) < \eta$, quindi $d_Z(g(f(x)), l) < \varepsilon$. $\square$

> **Osservazione 2.9.** L'ipotesi $f(x) \neq y_0$ è **essenziale**. Consideriamo $f \equiv 0$ e $g(y) = \mathbf{1}_{\{y \neq 0\}}$. Allora $\lim_{x \to 0} f(x) = 0$, $\lim_{y \to 0} g(y) = 1$, ma $(g \circ f)(x) \equiv 0$, quindi $\lim_{x \to 0} (g \circ f)(x) = 0 \neq 1$.

### 2.6 Limitatezza locale

**Teorema 2.10.** Se $\lim_{x \to x_0} f(x) = l \in \mathbb{R}$, allora esiste un intorno $U$ di $x_0$ tale che $f$ è **limitata** su $(U \cap X) \setminus \{x_0\}$.

**Dimostrazione.** Scegliamo $\varepsilon = 1$. Esiste $\delta > 0$ tale che
$$
0 < \|x-x_0\| < \delta \implies |f(x) - l| < 1.
$$
Allora $|f(x)| \le |f(x) - l| + |l| < 1 + |l|$, quindi $f$ è limitata con $M = 1 + |l|$. $\square$

### 2.7 o-piccolo

**Definizione 2.11.** Siano $f, g : X \subseteq \mathbb{R}^n \to \mathbb{R}$ e $x_0$ di accumulazione per $X$. Si dice che $f(x) = o(g(x))$ per $x \to x_0$ se esiste $\omega : X \to \mathbb{R}$ con $\omega(x) \to 0$ per $x \to x_0$ tale che $f(x) = \omega(x) g(x)$ in un intorno di $x_0$ (escluso al più $x_0$). Equivalentemente, se $g(x) \neq 0$ in un intorno di $x_0$,
$$
\lim_{x \to x_0} \frac{f(x)}{g(x)} = 0.
$$

> **Osservazione 2.12.** La notazione $o(g(x))$ è particolarmente utile per esprimere il **resto** negli sviluppi di Taylor. Ad esempio, $f(x) = o(\|x-x_0\|)$ significa che $f$ tende a $0$ più rapidamente di $\|x-x_0\|$.

## 3. Tecniche di calcolo dei limiti in $\mathbb{R}^n$

### 3.1 Sostituzione diretta

Se $f$ è **continua** in $x_0$ (cioè $\lim_{x \to x_0} f(x) = f(x_0)$), il limite si calcola direttamente. Questo vale per polinomi, funzioni razionali (dove il denominatore non si annulla), funzioni trigonometriche, esponenziali, logaritmiche, e loro composizioni.

**Esempio 3.1.**
$$
\lim_{(x,y) \to (1,2)} \frac{x^2 + y^2}{x + y} = \frac{1 + 4}{1 + 2} = \frac{5}{3}.
$$

**Esempio 3.2.**
$$
\lim_{(x,y,z) \to (0,1,2)} e^x \cos(y) + z^2 = 1 \cdot \cos(1) + 4 = \cos(1) + 4.
$$

### 3.2 Cambio di variabile

Si introduce una nuova variabile $t = g(x,y)$ per ricondurre il limite a un limite notevole in una variabile. Il cambio di variabile è lecito solo se la trasformazione è **localmente biunivoca** e $t \to t_0$ quando $(x,y) \to (x_0, y_0)$.

**Esempio 3.3.**
$$
\lim_{(x,y) \to (0,0)} \frac{\sin(xy)}{xy} \overset{t = xy}{=} \lim_{t \to 0} \frac{\sin t}{t} = 1.
$$

**Esempio 3.4.**
$$
\lim_{(x,y) \to (0,0)} \frac{1 - \cos(xy)}{x^2 y^2} \overset{t = xy}{=} \lim_{t \to 0} \frac{1 - \cos t}{t^2} = \frac{1}{2}.
$$

**Esempio 3.5.**
$$
\lim_{(x,y) \to (0,0)} \frac{e^{x^2 + y^2} - 1}{x^2 + y^2} \overset{t = x^2+y^2}{=} \lim_{t \to 0} \frac{e^t - 1}{t} = 1.
$$

### 3.3 Maggiorazione con disuguaglianze

Si maggiora $|f(x,y)|$ con una funzione di $\|(x,y)\|$ che tende a $0$. Disuguaglianze fondamentali:
1. $|\sin t| \le |t|$ per ogni $t \in \mathbb{R}$;
2. $|1 - \cos t| \le t^2/2$ per ogni $t \in \mathbb{R}$;
3. $|\ln(1+t)| \le |t|$ per $|t|$ sufficientemente piccolo;
4. $|e^t - 1| \le 2|t|$ per $|t|$ sufficientemente piccolo;
5. $|a + b| \le |a| + |b|$;
6. $\sqrt{a^2 + b^2} \le |a| + |b|$.

**Esempio 3.6.**
$$
0 \le \left|\frac{x \sin(xy)}{x^2 + y^2}\right| \le |x| \frac{|xy|}{x^2+y^2} = \frac{x^2 |y|}{x^2+y^2} \le |y| \to 0.
$$
La prima disuguaglianza è $|\sin t| \le |t|$; la seconda è $\frac{x^2}{x^2+y^2} \le 1$.

**Esempio 3.7.**
$$
0 \le \left|\frac{x^2 y}{x^2+y^2}\right| \le \frac{|y|(x^2+y^2)}{x^2+y^2} = |y| \to 0.
$$

### 3.4 Sviluppi di Taylor

**Esempio 3.8.**
$$
\lim_{(x,y) \to (0,0)} \frac{\ln(1+xy) - \sin(xy)}{2x^2+y^2}.
$$

**Dimostrazione.** Con $t = xy$, usando gli sviluppi
$$
\ln(1+t) = t - \frac{t^2}{2} + o(t^2), \qquad \sin(t) = t - \frac{t^3}{6} + o(t^4),
$$
otteniamo
$$
\ln(1+xy) - \sin(xy) = -\frac{x^2 y^2}{2} + o(x^2 y^2).
$$
Allora
$$
\frac{\ln(1+xy) - \sin(xy)}{2x^2+y^2} = -\frac{1}{2} \cdot \frac{x^2 y^2}{2x^2+y^2} + o\left(\frac{x^2 y^2}{2x^2+y^2}\right).
$$
Ora $0 \le \frac{x^2 y^2}{2x^2+y^2} \le \frac{x^2 y^2}{2|x||y|} = \frac{|x||y|}{2} \to 0$, quindi il limite è $0$. $\square$

**Esempio 3.9.**
$$
\lim_{(x,y) \to (0,0)} \frac{e^{xy} - 1 - xy}{x^2+y^2} = 0.
$$

**Dimostrazione.** Con $t = xy$, $e^t - 1 - t = t^2/2 + o(t^2)$. Allora
$$
\frac{e^{xy} - 1 - xy}{x^2+y^2} = \frac{x^2 y^2/2 + o(x^2 y^2)}{x^2+y^2}.
$$
La tesi segue dalla maggiorazione $0 \le \frac{x^2 y^2}{x^2+y^2} \le \frac{x^2 y^2}{2|x||y|} = \frac{|x||y|}{2} \to 0$. $\square$

### 3.5 Coordinate polari

**Teorema 3.10.** Sia $f : X \subseteq \mathbb{R}^2 \to \mathbb{R}$ e $(x_0, y_0)$ di accumulazione per $X$. Sia
$$
f(x_0 + \rho\cos\theta, y_0 + \rho\sin\theta) = g(\rho) h(\theta).
$$
Se $\lim_{\rho \to 0^+} g(\rho) = 0$ e esiste $M > 0$ tale che $|h(\theta)| \le M$ per ogni $\theta$, allora
$$
\lim_{(x,y) \to (x_0, y_0)} f(x,y) = 0.
$$

**Dimostrazione.** Fissato $\varepsilon > 0$, poiché $g(\rho) \to 0$ per $\rho \to 0^+$, esiste $\delta > 0$ con
$$
0 < \rho < \delta \implies |g(\rho)| < \frac{\varepsilon}{M}.
$$
Sia $(x,y) \in X$ con $0 < \|(x,y)-(x_0,y_0)\|_2 < \delta$. Allora $\rho \in (0, \delta)$ e
$$
|f(x,y)| = |g(\rho)| |h(\theta)| < \frac{\varepsilon}{M} \cdot M = \varepsilon.
$$
$\square$

**Esempio 3.11.**
$$
f(x,y) = \frac{x^2 y}{x^2+y^2} \implies f(\rho\cos\theta, \rho\sin\theta) = \rho\cos^2\theta \sin\theta.
$$
Qui $g(\rho) = \rho \to 0$ e $|h(\theta)| \le 1$, quindi il limite è $0$.

**Esempio 3.12.**
$$
f(x,y) = \frac{x^3 + y^3}{x^2+y^2} \implies f(\rho\cos\theta, \rho\sin\theta) = \rho(\cos^3\theta + \sin^3\theta).
$$
Qui $|h(\theta)| \le 2$, quindi il limite è $0$.

**Controesempio 3.13.**
$$
f(x,y) = \frac{x^2 y}{x^4 + y^2} \implies f(\rho\cos\theta, \rho\sin\theta) = \frac{\rho\cos^2\theta \sin\theta}{\rho^2\cos^4\theta + \sin^2\theta}.
$$
Non si decompone come $g(\rho) h(\theta)$. Infatti, lungo $y = x^2$ il limite vale $1/2 \neq 0$, quindi il limite non esiste.

> **Osservazione 3.14.** La condizione di **limitatazza uniforme** in $\theta$ è essenziale. Il semplice fatto che, per ogni $\theta$ fissato, $f(\rho\cos\theta, \rho\sin\theta) \to 0$ non garantisce che il limite esista: occorre che la convergenza sia uniforme rispetto a $\theta$.

### 3.6 Limiti iterati

**Definizione 3.15.** Sia $f : \operatorname{dom}f \subseteq \mathbb{R}^2 \to \mathbb{R}$. I **limiti iterati** sono
$$
\lim_{x \to x_0}\left(\lim_{y \to y_0} f(x,y)\right), \qquad \lim_{y \to y_0}\left(\lim_{x \to x_0} f(x,y)\right).
$$

**Osservazione 3.16.** L'esistenza e l'uguaglianza dei limiti iterati **non garantiscono** l'esistenza del limite doppio. Ad esempio, per
$$
f(x,y) = \frac{x^2 - y^2}{x^2 + y^2},
$$
si ha
$$
\lim_{y \to 0}\left(\lim_{x \to 0} f(x,y)\right) = -1, \qquad \lim_{x \to 0}\left(\lim_{y \to 0} f(x,y)\right) = 1.
$$
Poiché i due limiti iterati sono diversi, il limite doppio non esiste.

**Esempio 3.17.** Esistono funzioni per cui i limiti iterati esistono, sono uguali, ma il limite doppio non esiste. Ad esempio,
$$
f(x,y) = \frac{xy}{x^2 + y^2}.
$$
I limiti iterati valgono entrambi $0$, ma il limite doppio non esiste (lungo $y = x$ vale $1/2$).

## 4. Continuità

### 4.1 Definizione di continuità

**Definizione 4.1.** Sia $f : X \subseteq \mathbb{R}^n \to \mathbb{R}^m$ e $x_0 \in X$. Si dice che $f$ è **continua** in $x_0$ se vale una delle due condizioni:

1. $x_0$ è un **punto isolato** di $X$ (cioè esiste $r > 0$ con $\mathcal{B}_r(x_0) \cap X = \{x_0\}$); oppure
2. $x_0$ è di accumulazione per $X$ e
   $$
   \lim_{x \to x_0} f(x) = f(x_0).
   $$

$f$ è **continua su** $X$ se è continua in ogni punto di $X$.

> **Osservazione 4.2.** La continuità è una proprietà **locale**: dipende solo dal comportamento di $f$ in un intorno di $x_0$. Inoltre, se $x_0$ è isolato, ogni funzione è automaticamente continua in $x_0$: la condizione di limite è vacuamente soddisfatta.

**Esempio 4.3.** $f(x,y) = x^2 + y^2$ è continua su $\mathbb{R}^2$ perché è un polinomio (funzione elementare).

**Esempio 4.4.** $f(x,y) = \frac{1}{x^2+y^2}$ è continua su $\mathbb{R}^2 \setminus \{(0,0)\}$.

**Esempio 4.5.** $f : \mathbb{Z}^2 \to \mathbb{R}$, $f(m,n) = m+n$, è continua su $\mathbb{Z}^2$ perché ogni punto è isolato.

### 4.2 Caratterizzazione con successioni

**Teorema 4.6.** $f : X \subseteq \mathbb{R}^n \to \mathbb{R}^m$ è continua in $x_0 \in X$ se e solo se per ogni successione $\{x_k\} \subseteq X$ con $x_k \to x_0$ si ha $f(x_k) \to f(x_0)$.

**Dimostrazione.** Segue dalla caratterizzazione sequenziale del limite (Teorema 1.12) e dalla definizione di continuità. Se $x_0$ è isolato, ogni successione $x_k \to x_0$ è definitivamente costante uguale a $x_0$, quindi $f(x_k) \to f(x_0)$. $\square$

> **Osservazione 4.7.** La caratterizzazione sequenziale è spesso utile per **dimostrare la non continuità**: basta trovare una successione $x_k \to x_0$ con $f(x_k) \not\to f(x_0)$.

**Esempio 4.8.** $f(x,y) = \frac{xy}{x^2+y^2}$ per $(x,y) \neq (0,0)$, $f(0,0) = 0$. Questa funzione non è continua in $(0,0)$: lungo $x_k = (1/k, 1/k)$, $f(x_k) = 1/2 \not\to 0 = f(0,0)$.

### 4.3 Continuità componente per componente

**Teorema 4.9.** $f = (f_1, \dots, f_m) : X \subseteq \mathbb{R}^n \to \mathbb{R}^m$ è continua in $x_0$ se e solo se ogni $f_i : X \to \mathbb{R}$ è continua in $x_0$.

**Dimostrazione.** Segue dal Teorema 2.1 (limiti componente per componente) applicato alla definizione di continuità. $\square$

### 4.4 Algebra delle funzioni continue

**Teorema 4.10.** Se $f, g : X \subseteq \mathbb{R}^n \to \mathbb{R}$ sono continue in $x_0$, allora:
1. $f + g$ è continua in $x_0$;
2. $f \cdot g$ è continua in $x_0$;
3. se $g(x_0) \neq 0$, $f/g$ è continua in $x_0$.

**Dimostrazione.** Segue dall'algebra dei limiti (Teorema 2.6) applicata alla definizione di continuità: $\lim_{x \to x_0}(f+g)(x) = l_1 + l_2 = f(x_0) + g(x_0) = (f+g)(x_0)$, e analogamente per prodotto e quoziente. $\square$

**Esempio 4.11.** $f(x,y) = \frac{x^2 + y^2}{xy + 1}$ è continua in $(0,0)$: numeratore e denominatore sono polinomi, e $xy + 1 \neq 0$ in $(0,0)$.

### 4.5 Continuità della composizione

**Teorema 4.12.** Siano $f : X \subseteq \mathbb{R}^n \to \mathbb{R}^m$ continua in $x_0$ con $f(X) \subseteq Y$, e $g : Y \subseteq \mathbb{R}^m \to \mathbb{R}^k$ continua in $y_0 = f(x_0) \in Y$. Allora $g \circ f : X \to \mathbb{R}^k$ è continua in $x_0$.

**Dimostrazione.** Fissato $\varepsilon > 0$, poiché $g$ è continua in $y_0$, esiste $\eta > 0$ tale che
$$
y \in Y,\ \|y - y_0\| < \eta \implies \|g(y) - g(y_0)\| < \varepsilon.
$$
Poiché $f$ è continua in $x_0$, esiste $\delta > 0$ tale che
$$
x \in X,\ \|x - x_0\| < \delta \implies \|f(x) - f(x_0)\| < \eta.
$$
Allora per $x \in X$ con $\|x-x_0\| < \delta$, si ha $f(x) \in Y$ e $\|f(x) - y_0\| < \eta$, quindi
$$
\|g(f(x)) - g(f(x_0))\| < \varepsilon.
$$
$\square$

**Esempio 4.13.** $f(x,y) = x^2 + y^2$ è continua su $\mathbb{R}^2$; $g(t) = \sin t$ è continua su $\mathbb{R}$. Allora $g \circ f = \sin(x^2+y^2)$ è continua su $\mathbb{R}^2$.

### 4.6 Continuità delle funzioni elementari

**Teorema 4.14.** Le funzioni elementari (polinomi, funzioni razionali, trigonometriche, esponenziali, logaritmiche, e loro composizioni) sono continue su tutto il loro dominio.

**Esempio 4.15.** $\lim_{(x,y) \to (1,2)} e^x \cos y = e \cos 2$.

### 4.7 Caratterizzazione tramite controimmagini di aperti

**Teorema 4.16.** $f : X \subseteq \mathbb{R}^n \to \mathbb{R}^m$ è continua su $X$ se e solo se $f^{-1}(A)$ è aperto in $X$ (cioè $f^{-1}(A) = U \cap X$ per qualche $U$ aperto in $\mathbb{R}^n$) per ogni $A \subseteq \mathbb{R}^m$ aperto.

**Dimostrazione.**

$(\Rightarrow)$ Sia $A \subseteq \mathbb{R}^m$ aperto e $x_0 \in f^{-1}(A)$. Allora $f(x_0) \in A$, e poiché $A$ è aperto esiste $\varepsilon > 0$ con $\mathcal{B}_\varepsilon(f(x_0)) \subseteq A$. Per la continuità di $f$ in $x_0$, esiste $\delta > 0$ tale che
$$
x \in X,\ \|x-x_0\| < \delta \implies \|f(x) - f(x_0)\| < \varepsilon.
$$
Allora per ogni $x \in \mathcal{B}_\delta(x_0) \cap X$, si ha $f(x) \in \mathcal{B}_\varepsilon(f(x_0)) \subseteq A$, quindi $x \in f^{-1}(A)$. Dunque $\mathcal{B}_\delta(x_0) \cap X \subseteq f^{-1}(A)$, cioè $f^{-1}(A)$ è aperto in $X$.

$(\Leftarrow)$ Fissiamo $x_0 \in X$ e $\varepsilon > 0$. L'insieme $\mathcal{B}_\varepsilon(f(x_0))$ è aperto in $\mathbb{R}^m$. Per ipotesi, $f^{-1}(\mathcal{B}_\varepsilon(f(x_0)))$ è aperto in $X$ e contiene $x_0$. Quindi esiste $\delta > 0$ con
$$
\mathcal{B}_\delta(x_0) \cap X \subseteq f^{-1}(\mathcal{B}_\varepsilon(f(x_0))).
$$
Allora per ogni $x \in X$ con $\|x-x_0\| < \delta$, $f(x) \in \mathcal{B}_\varepsilon(f(x_0))$, cioè $\|f(x) - f(x_0)\| < \varepsilon$. $\square$

**Corollario 4.17.** $f : X \to \mathbb{R}^m$ è continua su $X$ se e solo se $f^{-1}(C)$ è chiuso in $X$ per ogni $C \subseteq \mathbb{R}^m$ chiuso.

**Dimostrazione.** $C$ chiuso $\iff$ $C^C$ aperto. Per il Teorema 4.16, $f^{-1}(C^C)$ è aperto in $X$. Ora
$$
f^{-1}(C^C) = \{x \in X : f(x) \notin C\} = X \setminus f^{-1}(C) = (f^{-1}(C))^C.
$$
Quindi $f^{-1}(C)$ è chiuso in $X$. $\square$

> **Osservazione 4.18.** Il Teorema 4.16 è il **motivo** per cui la topologia è lo strumento naturale per studiare la continuità: la continuità dipende solo dalla topologia (non dalla metrica specifica), e la caratterizzazione tramite controimmagini di aperti è puramente topologica.

**Esempio 4.19.** $f(x,y) = x^2 + y^2$ è continua. Allora:
- $\{(x,y) : x^2+y^2 < 1\} = f^{-1}((-\infty, 1))$ è aperto;
- $\{(x,y) : x^2+y^2 \le 1\} = f^{-1}((-\infty, 1])$ è chiuso.

### 4.8 Teorema di Weierstrass

**Teorema 4.20.** Sia $K \subseteq \mathbb{R}^n$ **compatto** (chiuso e limitato) e $f : K \to \mathbb{R}$ **continua** su $K$. Allora esistono $x_m, x_M \in K$ tali che
$$
f(x_m) = \min_K f, \qquad f(x_M) = \max_K f.
$$

**Dimostrazione.** Poiché $K$ è compatto e $f$ è continua, l'immagine $f(K)$ è un sottoinsieme compatto di $\mathbb{R}$ (per il teorema di continuità su compatti). In $\mathbb{R}$, un insieme compatto è chiuso e limitato, quindi ammette massimo e minimo. Dunque esistono $x_m, x_M \in K$ con $f(x_m) = \min f(K)$ e $f(x_M) = \max f(K)$. $\square$

> **Osservazione 4.21.** Il teorema di Weierstrass è un risultato **esistenziale**: garantisce l'esistenza di massimo e minimo, ma non fornisce un metodo per calcolarli. È uno strumento fondamentale in ottimizzazione.

**Esempio 4.22.** $f(x,y) = x^2 + y^2$ su $K = \{(x,y) : x^2+y^2 \le 1\}$ ha minimo $0$ in $(0,0)$ e massimo $1$ sulla circonferenza $x^2+y^2 = 1$.

### 4.9 Insiemi di sotto/sopra livello

**Teorema 4.23.** Sia $f : \mathbb{R}^n \to \mathbb{R}$ continua. Allora per ogni $c \in \mathbb{R}$:
1. $\{f < c\}$ è aperto;
2. $\{f \le c\}$ è chiuso;
3. $\{f > c\}$ è aperto;
4. $\{f \ge c\}$ è chiuso.

**Dimostrazione.** 
(1) $\{f < c\} = f^{-1}((-\infty, c))$, e $(-\infty, c)$ è aperto in $\mathbb{R}$. Per il Teorema 4.16, la controimmagine è aperta.
(2) $\{f \le c\} = \{f < c\}^C$ è il complementare di un aperto, quindi è chiuso.
(3) $\{f > c\} = f^{-1}((c, +\infty))$, e $(c, +\infty)$ è aperto.
(4) $\{f \ge c\} = \{f > c\}^C$ è chiuso. $\square$

**Esempio 4.24.** $f(x,y) = x^2 + y^2$. $\{f < 1\}$ è il disco aperto; $\{f \le 1\}$ è il disco chiuso; $\{f > 1\}$ è l'esterno aperto; $\{f \ge 1\}$ è l'esterno chiuso.

## 5. Continuità uniforme

### 5.1 Definizione

**Definizione 5.1.** Sia $f : X \subseteq \mathbb{R}^n \to \mathbb{R}^m$. $f$ è **uniformemente continua** su $X$ se
$$
\forall \varepsilon > 0\ \exists \delta > 0 : x, y \in X,\ \|x-y\| < \delta \implies \|f(x) - f(y)\| < \varepsilon.
$$

> **Osservazione 5.2 — Differenza cruciale con la continuità puntuale.**  
> Nella continuità **puntuale**, $\delta$ dipende da $\varepsilon$ **e da $x_0$**: per ogni punto $x_0$ si trova un $\delta$ diverso. Nella continuità **uniforme**, $\delta$ dipende **solo da $\varepsilon$** ed è lo stesso per tutti i punti di $X$. Questo è un requisito **più forte**.

> **Osservazione 5.3 — Interpretazione.**  
> La continuità uniforme dice che la funzione "cambia poco" su tutto il dominio, in modo controllato e uniforme. Non ci sono punti dove la funzione è "più ripida" di altri.

**Esempio 5.4.** $f(x) = x$ su $\mathbb{R}$ è uniformemente continua: $\delta = \varepsilon$.

**Verifica.** $|f(x) - f(y)| = |x - y| < \delta = \varepsilon$. $\square$

**Esempio 5.5.** $f(x) = \sin x$ su $\mathbb{R}$ è uniformemente continua.

**Verifica.** Usando la formula di prostaferesi:
$$
|\sin x - \sin y| = \left|2 \cos\left(\frac{x+y}{2}\right) \sin\left(\frac{x-y}{2}\right)\right| \le 2 \cdot 1 \cdot \frac{|x-y|}{2} = |x - y|,
$$
dove abbiamo usato $|\cos t| \le 1$ e $|\sin t| \le |t|$. Quindi $\delta = \varepsilon$. $\square$

**Esempio 5.6.** $f(x) = \sqrt{x}$ su $(0,1)$ è uniformemente continua.

**Verifica.** Per $x, y \in (0,1)$,
$$
|\sqrt{x} - \sqrt{y}| = \frac{|x-y|}{\sqrt{x} + \sqrt{y}} \le \sqrt{|x-y|},
$$
dove l'ultima disuguaglianza segue da $\sqrt{x} + \sqrt{y} \ge \sqrt{|x-y|}$ (si verifica elevando al quadrato: $x + y + 2\sqrt{xy} \ge |x-y|$, che è sempre vera). Quindi scegliendo $\delta = \varepsilon^2$, se $|x-y| < \delta$, allora $|\sqrt{x} - \sqrt{y}| < \varepsilon$. $\square$

**Esempio 5.7.** $f(x) = x^2$ su $\mathbb{R}$ è continua ma **non** uniformemente continua.

**Dimostrazione.** Fissiamo $\varepsilon = 1$. Per ogni $\delta > 0$, scegliamo $x = 1/\delta$ e $y = 1/\delta + \delta/2$. Allora $|x - y| = \delta/2 < \delta$, ma
$$
|f(x) - f(y)| = \left|\frac{1}{\delta^2} - \left(\frac{1}{\delta} + \frac{\delta}{2}\right)^2\right| = \left|\frac{1}{\delta^2} - \frac{1}{\delta^2} - \frac{1}{\delta} \cdot \delta - \frac{\delta^2}{4}\right| = 1 + \frac{\delta^2}{4} > 1 = \varepsilon.
$$
Quindi non esiste $\delta$ che soddisfi la condizione per $\varepsilon = 1$. $\square$

**Esempio 5.8.** $f(x) = \frac{1}{x}$ su $(0,1)$ è continua ma non uniformemente continua.

**Dimostrazione.** Scegliamo $x_k = 1/k$, $y_k = 1/(k+1)$. Allora $|x_k - y_k| = \frac{1}{k(k+1)} \to 0$, ma
$$
|f(x_k) - f(y_k)| = |k - (k+1)| = 1.
$$
Quindi per $\varepsilon = 1/2$, non esiste $\delta$ che soddisfi la condizione. $\square$

### 5.2 Teorema di Heine–Cantor

**Teorema 5.9 (Heine–Cantor).** Se $f : K \subseteq \mathbb{R}^n \to \mathbb{R}^m$ è **continua** su un **compatto** $K$, allora $f$ è **uniformemente continua** su $K$.

**Dimostrazione.** Supponiamo per assurdo che $f$ non sia uniformemente continua. Allora esiste $\varepsilon > 0$ tale che per ogni $k \in \mathbb{N}$ esistono $x_k, y_k \in K$ con
$$
\|x_k - y_k\| < \frac{1}{k} \qquad \text{e} \qquad \|f(x_k) - f(y_k)\| \ge \varepsilon.
$$

Poiché $K$ è compatto, per il **teorema di Bolzano–Weierstrass** esiste una sottosuccessione $\{x_{k_j}\}$ convergente a $x^* \in K$. Poiché $\|x_{k_j} - y_{k_j}\| < 1/k_j \to 0$, anche $y_{k_j} \to x^*$.

Per la **continuità di $f$ in $x^*$**, si ha $f(x_{k_j}) \to f(x^*)$ e $f(y_{k_j}) \to f(x^*)$. Quindi
$$
\|f(x_{k_j}) - f(y_{k_j})\| \le \|f(x_{k_j}) - f(x^*)\| + \|f(x^*) - f(y_{k_j})\| \to 0,
$$
contraddicendo $\|f(x_{k_j}) - f(y_{k_j})\| \ge \varepsilon$. $\square$

> **Osservazione 5.10.** Il teorema di Heine–Cantor è **fondamentale**: garantisce che la continuità su un compatto è automaticamente uniforme. Questo è uno dei motivi per cui i compatti sono così importanti in analisi.

> **Osservazione 5.11 — Controesempio alla compattezza.** Su $X = (0,1)$ (non compatto), $f(x) = 1/x$ è continua ma non uniformemente continua (Esempio 5.8). Questo mostra che l'ipotesi di compattezza non può essere omessa.

## 6. Funzioni Lipschitziane e Hölderiane

### 6.1 Definizioni

**Definizione 6.1.** Sia $f : X \subseteq \mathbb{R}^n \to \mathbb{R}^m$. $f$ è **Lipschitziana** di costante $L \ge 0$ su $X$ se
$$
\|f(x) - f(y)\| \le L \|x - y\| \qquad \forall x, y \in X.
$$
La più piccola costante $L$ per cui vale la disuguaglianza è detta **costante di Lipschitz** di $f$ e si indica con $\operatorname{Lip}(f)$.

**Definizione 6.2.** $f$ è **Hölderiana** di esponente $\alpha \in (0,1]$ e costante $L \ge 0$ su $X$ se
$$
\|f(x) - f(y)\| \le L \|x - y\|^\alpha \qquad \forall x, y \in X.
$$
Per $\alpha = 1$ si ritrova la Lipschitzianità.

> **Osservazione 6.3.** La Lipschitzianità è una condizione **più forte** della continuità uniforme. Intuitivamente, una funzione Lipschitziana non può "cambiare troppo rapidamente": il suo grafico è contenuto in un cono di pendenza $L$.

### 6.2 Gerarchia delle continuità

**Proposizione 6.4.** Vale la catena di implicazioni
$$
\text{Lipschitziana} \implies \text{Hölderiana} \implies \text{uniformemente continua} \implies \text{continua}.
$$
Nessuna implicazione è reversibile.

**Dimostrazione.**

**(Lipschitziana $\Rightarrow$ Hölderiana).** Ovvia con $\alpha = 1$ e la stessa costante $L$.

**(Hölderiana $\Rightarrow$ uniformemente continua).** Fissato $\varepsilon > 0$, scegliamo $\delta = (\varepsilon/L)^{1/\alpha}$ (se $L = 0$, la funzione è costante e la tesi è banale). Se $\|x-y\| < \delta$, allora
$$
\|f(x) - f(y)\| \le L \|x-y\|^\alpha < L \delta^\alpha = \varepsilon.
$$

**(Uniformemente continua $\Rightarrow$ continua).** Immediata dalla definizione: la continuità uniforme è una condizione più forte (il $\delta$ uniforme funziona anche per ogni $x_0$). $\square$

> **Osservazione 6.5 — Le implicazioni inverse.** Le implicazioni inverse **non valgono**:
> - $f(x) = \sqrt{|x|}$ è Hölderiana di esponente $1/2$ ma non Lipschitziana in $0$;
> - $f(x) = 1/\ln x$ su $(0,1/2]$ è uniformemente continua ma non Hölderiana;
> - $f(x) = x^2$ su $\mathbb{R}$ è continua ma non uniformemente continua.

### 6.3 Esempi

**Esempio 6.6 — Norma.** La norma $\|\cdot\| : \mathbb{R}^n \to \mathbb{R}$ è Lipschitziana di costante $1$.

**Dimostrazione.** Per la **disuguaglianza triangolare inversa**:
$$
\bigl| \|x\| - \|y\| \bigr| \le \|x - y\|.
$$
$\square$

**Esempio 6.7 — Proiezioni.** La proiezione $\pi_i : \mathbb{R}^n \to \mathbb{R}$, $\pi_i(x) = x_i$, è Lipschitziana di costante $1$ rispetto a $\|\cdot\|_\infty$ (e di costante $1$ rispetto a $\|\cdot\|_2$, perché $|x_i - y_i| \le \|x-y\|_2$).

**Dimostrazione.** $|\pi_i(x) - \pi_i(y)| = |x_i - y_i| \le \|x - y\|_\infty$. $\square$

**Esempio 6.8 — Applicazioni lineari.** Se $T : \mathbb{R}^n \to \mathbb{R}^m$ è lineare, allora $T$ è Lipschitziana con costante $\|T\|_{\text{op}} = \sup_{\|x\| = 1} \|Tx\|$.

**Dimostrazione.** $\|Tx - Ty\| = \|T(x-y)\| \le \|T\|_{\text{op}} \|x-y\|$. $\square$

**Esempio 6.9 — Funzione Hölderiana non Lipschitziana.** $f(x) = \sqrt{|x|}$ su $\mathbb{R}$.

**Dimostrazione.** Per $x, y \ge 0$,
$$
|\sqrt{x} - \sqrt{y}| = \frac{|x-y|}{\sqrt{x} + \sqrt{y}}.
$$
Se $x, y \to 0$, il denominatore tende a $0$, quindi non esiste $L$ tale che $|\sqrt{x} - \sqrt{y}| \le L|x-y|$ per ogni $x, y \in [0,1]$. Tuttavia,
$$
|\sqrt{x} - \sqrt{y}| \le \sqrt{|x-y|},
$$
quindi $f$ è Hölderiana di esponente $1/2$. $\square$

**Esempio 6.10 — Uniformemente continua non Hölderiana.** $f(x) = 1/\ln x$ su $(0, 1/2]$.

**Cenni.** La funzione è continua su $(0,1/2]$ e ha limite finito per $x \to 0^+$ (vale $0$), quindi si estende a una funzione continua su $[0,1/2]$, che è compatto. Per Heine–Cantor, è uniformemente continua. Tuttavia, non è Hölderiana: per $x, y \to 0$, il rapporto $\frac{|f(x)-f(y)|}{|x-y|^\alpha}$ non è limitato per alcun $\alpha > 0$.
