# Limiti e continuità di funzioni da $\mathbb{R}^n$ a $\mathbb{R}^m$

---

## 1. Richiami: limiti in spazi metrici

### Definizione 1.1 — Limite di una funzione tra spazi metrici

Siano $(X, d_X)$ e $(Y, d_Y)$ spazi metrici, sia $f : \operatorname{dom}f \subseteq X \to Y$ e sia $x_0 \in X$ un **punto di accumulazione** per $\operatorname{dom}f$. Si dice che

$$
\lim_{x \to x_0} f(x) = l \in Y
$$

se per ogni intorno $V$ di $l$ in $Y$ esiste un intorno $U$ di $x_0$ in $X$ tale che

$$
x \in U \cap \operatorname{dom}f,\ x \neq x_0 \implies f(x) \in V.
$$

Equivalentemente, in termini di distanze: per ogni $\varepsilon > 0$ esiste $\delta > 0$ tale che

$$
x \in \operatorname{dom}f,\ 0 < d_X(x, x_0) < \delta \implies d_Y(f(x), l) < \varepsilon.
$$

Nel caso $Y = \mathbb{R}^m$, la condizione diventa

$$
\forall \varepsilon > 0\ \exists \delta > 0\ :\ 0 < \|x - x_0\| < \delta \implies \|f(x) - l\| < \varepsilon,
$$

dove $\|\cdot\|$ è una qualsiasi norma su $\mathbb{R}^n$ (per il dominio) e su $\mathbb{R}^m$ (per il codominio).

Nel caso $Y = \mathbb{R}$, la condizione diventa

$$
\forall \varepsilon > 0\ \exists \delta > 0\ :\ x \in \operatorname{dom}f,\ 0 < d_X(x, x_0) < \delta \implies |f(x) - l| < \varepsilon,
$$

che è equivalente a

$$
\lim_{x \to x_0} |f(x) - l| = 0.
$$

> **Osservazione 1.2.**  
> La condizione $x \neq x_0$ è essenziale: il valore $f(x_0)$ non influenza il limite, e $f$ può non essere definita in $x_0$. La richiesta che $x_0$ sia di accumulazione per $\operatorname{dom}f$ garantisce che esistano punti del dominio arbitrariamente vicini a $x_0$, rendendo la definizione non vuota.

## 2. Limiti componente per componente

### Teorema 2.1 — Riduzione a limiti scalari

Sia $f : X \subseteq \mathbb{R}^n \to \mathbb{R}^m$ una funzione vettoriale, scritta in componenti come

$$
f(x_1, \dots, x_n) = \bigl(f_1(x_1, \dots, x_n), \dots, f_m(x_1, \dots, x_n)\bigr),
$$

dove $f_i : X \subseteq \mathbb{R}^n \to \mathbb{R}$ per ogni $i = 1, \dots, m$. Allora, per ogni $x_0$ di accumulazione per $X$,

$$
\lim_{x \to x_0} f(x) = l = (l_1, \dots, l_m) \in \mathbb{R}^m
\iff
\lim_{x \to x_0} f_i(x) = l_i \qquad \forall i = 1, \dots, m.
$$

**Dimostrazione.**

$(\Rightarrow)$ Supponiamo che $\lim_{x \to x_0} f(x) = l$. Fissiamo $\varepsilon > 0$. Per definizione di limite, esiste $\delta > 0$ tale che

$$
0 < \|x - x_0\| < \delta \implies \|f(x) - l\|_\infty < \varepsilon.
$$

Osserviamo che, per ogni $i = 1, \dots, m$,

$$
|f_i(x) - l_i| \le \max_{1 \le j \le m} |f_j(x) - l_j| = \|f(x) - l\|_\infty.
$$

Quindi, per lo stesso $\delta$,

$$
0 < \|x - x_0\| < \delta \implies |f_i(x) - l_i| < \varepsilon.
$$

Dunque $\lim_{x \to x_0} f_i(x) = l_i$ per ogni $i$.

$(\Leftarrow)$ Supponiamo che $\lim_{x \to x_0} f_i(x) = l_i$ per ogni $i = 1, \dots, m$. Fissiamo $\varepsilon > 0$. Per ogni $i$ esiste $\delta_i > 0$ tale che

$$
0 < \|x - x_0\| < \delta_i \implies |f_i(x) - l_i| < \frac{\varepsilon}{m}.
$$

Sia $\delta = \min\{\delta_1, \dots, \delta_m\} > 0$. Per $0 < \|x - x_0\| < \delta$ si ha, per ogni $i$,

$$
|f_i(x) - l_i| < \frac{\varepsilon}{m}.
$$

Allora

$$
\|f(x) - l\|_1 = \sum_{i=1}^m |f_i(x) - l_i| < m \cdot \frac{\varepsilon}{m} = \varepsilon.
$$

Poiché $\|\cdot\|_2 \le \|\cdot\|_1$, anche $\|f(x) - l\|_2 < \varepsilon$. Quindi $\lim_{x \to x_0} f(x) = l$. $\square$

> **Osservazione 2.2.**  
> Il calcolo di limiti di funzioni da $\mathbb{R}^n$ a $\mathbb{R}^m$ si riduce quindi al calcolo di $m$ limiti da $\mathbb{R}^n$ a $\mathbb{R}$. Questo permette di applicare tutta la teoria scalare componente per componente.

### Esempio 2.3

Sia

$$
f(x,y) = \left(\frac{\sin(xy)}{xy}, \frac{x^2 - y^2}{x^2 + y^2}\right) : \mathbb{R}^2 \setminus \{(0,0)\} \to \mathbb{R}^2.
$$

Per il Teorema 2.1,

$$
\lim_{(x,y) \to (0,0)} f(x,y) = \left(\lim_{(x,y) \to (0,0)} \frac{\sin(xy)}{xy}, \lim_{(x,y) \to (0,0)} \frac{x^2 - y^2}{x^2 + y^2}\right).
$$

Il primo limite vale $1$ (Esempio 3.1); il secondo non esiste (Esempio 8.3). Dunque il limite di $f$ **non esiste**.

## 3. Teoremi sui limiti

In questa sezione enunciamo e dimostriamo i principali teoremi sui limiti di funzioni da $\mathbb{R}^n$ a $\mathbb{R}$ (o $\mathbb{R}^m$). Tutti i risultati si estendono al caso vettoriale applicando il Teorema 2.1 componente per componente.


Siano $f, g, h : X \subseteq \mathbb{R}^n \to \mathbb{R}$ e sia $x_0$ un punto di accumulazione per $X$.

---

### 3.1 Teorema di unicità del limite

#### Teorema 3.1 — Unicità del limite

Se

$$
\lim_{x \to x_0} f(x) = l_1 \in \mathbb{R}
\qquad\text{e}\qquad
\lim_{x \to x_0} f(x) = l_2 \in \mathbb{R},
$$

allora $l_1 = l_2$.

**Dimostrazione.**  
Supponiamo per assurdo $l_1 \neq l_2$. Sia $\varepsilon = \frac{|l_1 - l_2|}{3} > 0$. Per definizione di limite, esistono $\delta_1, \delta_2 > 0$ tali che

$$
0 < \|x - x_0\| < \delta_1 \implies |f(x) - l_1| < \varepsilon,
$$

$$
0 < \|x - x_0\| < \delta_2 \implies |f(x) - l_2| < \varepsilon.
$$

Poiché $x_0$ è di accumulazione per $X$, esiste $x \in X$ con $0 < \|x - x_0\| < \min\{\delta_1, \delta_2\}$. Allora, per la disuguaglianza triangolare,

$$
|l_1 - l_2|
\le
|l_1 - f(x)| + |f(x) - l_2|
<
2\varepsilon
=
\frac{2}{3}|l_1 - l_2|,
$$

contraddizione. Quindi $l_1 = l_2$. $\square$

> **Osservazione 3.2.**  
> L'unicità del limite giustifica la scrittura $\lim_{x \to x_0} f(x) = l$: se il limite esiste, è unico.

---

### 3.2 Teorema di permanenza del segno

#### Teorema 3.3 — Permanenza del segno

Sia $\lim_{x \to x_0} f(x) = l \in \mathbb{R}$.

1. Se $l > 0$, allora esiste un intorno $U$ di $x_0$ tale che
   $$
   f(x) > 0 \qquad \forall x \in (U \cap X) \setminus \{x_0\}.
   $$
2. Se $l < 0$, allora esiste un intorno $U$ di $x_0$ tale che
   $$
   f(x) < 0 \qquad \forall x \in (U \cap X) \setminus \{x_0\}.
   $$

**Dimostrazione.**  
Dimostriamo il caso $l > 0$. Sia $\varepsilon = \frac{l}{2} > 0$. Per definizione di limite, esiste $\delta > 0$ tale che

$$
0 < \|x - x_0\| < \delta \implies |f(x) - l| < \frac{l}{2}.
$$

In particolare, per $x \in X$ con $0 < \|x - x_0\| < \delta$,

$$
f(x) > l - \frac{l}{2} = \frac{l}{2} > 0.
$$

Il caso $l < 0$ è simmetrico, applicando la stessa argomentazione a $-f$ (oppure prendendo $\varepsilon = -l/2 > 0$). $\square$

> **Osservazione 3.4.**  
> Il teorema di permanenza del segno afferma che, se il limite è strettamente positivo (o negativo), allora la funzione è strettamente positiva (o negativa) in un intorno sufficientemente piccolo di $x_0$, escluso al più il punto $x_0$ stesso. **Il viceversa non vale**: una funzione può essere positiva in un intorno di $x_0$ senza che il limite sia positivo (potrebbe essere $0$).

#### Corollario 3.5 — Confronto tra limiti

Se

$$
\lim_{x \to x_0} f(x) = l_1,
\qquad
\lim_{x \to x_0} g(x) = l_2,
\qquad
f(x) \le g(x) \quad \forall x \in (U \cap X) \setminus \{x_0\}
$$

per un opportuno intorno $U$ di $x_0$, allora

$$
l_1 \le l_2.
$$

**Dimostrazione.**  
Supponiamo per assurdo $l_1 > l_2$. Per il teorema di permanenza del segno applicato a $f - g$ (il cui limite è $l_1 - l_2 > 0$), esiste un intorno $V$ di $x_0$ tale che

$$
f(x) - g(x) > 0 \qquad \forall x \in (V \cap X) \setminus \{x_0\},
$$

contraddicendo l'ipotesi $f(x) \le g(x)$. Quindi $l_1 \le l_2$. $\square$

---

### 3.3 Teorema sull'algebra dei limiti

#### Teorema 3.6 — Algebra dei limiti

Siano $f, g : X \subseteq \mathbb{R}^n \to \mathbb{R}$ e sia $x_0$ di accumulazione per $X$. Supponiamo che

$$
\lim_{x \to x_0} f(x) = l_1 \in \mathbb{R},
\qquad
\lim_{x \to x_0} g(x) = l_2 \in \mathbb{R}.
$$

Allora:

1. **Somma:**
   $$
   \lim_{x \to x_0} \bigl(f(x) + g(x)\bigr) = l_1 + l_2.
   $$
2. **Prodotto:**
   $$
   \lim_{x \to x_0} \bigl(f(x) \cdot g(x)\bigr) = l_1 \cdot l_2.
   $$
3. **Quoziente:** se $l_2 \neq 0$, allora esiste un intorno $U$ di $x_0$ tale che $g(x) \neq 0$ per ogni $x \in (U \cap X) \setminus \{x_0\}$, e
   $$
   \lim_{x \to x_0} \frac{f(x)}{g(x)} = \frac{l_1}{l_2}.
   $$
4. **Prodotto per scalare:** per ogni $\lambda \in \mathbb{R}$,
   $$
   \lim_{x \to x_0} \lambda f(x) = \lambda l_1.
   $$

**Dimostrazione.**

1. **Somma.** Fissiamo $\varepsilon > 0$. Esistono $\delta_1, \delta_2 > 0$ tali che

   $$
   0 < \|x - x_0\| < \delta_1 \implies |f(x) - l_1| < \frac{\varepsilon}{2},
   $$

   $$
   0 < \|x - x_0\| < \delta_2 \implies |g(x) - l_2| < \frac{\varepsilon}{2}.
   $$

   Sia $\delta = \min\{\delta_1, \delta_2\}$. Per $0 < \|x - x_0\| < \delta$,

   $$
   |f(x) + g(x) - (l_1 + l_2)| \le |f(x) - l_1| + |g(x) - l_2| < \varepsilon.
   $$

2. **Prodotto.** Scriviamo

   $$
   f(x) g(x) - l_1 l_2 = f(x)(g(x) - l_2) + l_2 (f(x) - l_1).
   $$

   Allora

   $$
   |f(x) g(x) - l_1 l_2| \le |f(x)|\,|g(x) - l_2| + |l_2|\,|f(x) - l_1|.
   $$

   Poiché $f$ è limitata in un intorno di $x_0$ (essendo convergente), esiste $M > 0$ tale che $|f(x)| \le M$ in un intorno di $x_0$. Fissato $\varepsilon > 0$, scegliamo $\delta_1, \delta_2$ tali che

   $$
   0 < \|x - x_0\| < \delta_1 \implies |g(x) - l_2| < \frac{\varepsilon}{2M},
   $$

   $$
   0 < \|x - x_0\| < \delta_2 \implies |f(x) - l_1| < \frac{\varepsilon}{2(|l_2| + 1)}.
   $$

   Per $\delta = \min\{\delta_1, \delta_2\}$ si ottiene la tesi.

3. **Quoziente.** Poiché $l_2 \neq 0$, per il teorema di permanenza del segno esiste un intorno $U$ di $x_0$ tale che $g(x) \neq 0$ per ogni $x \in (U \cap X) \setminus \{x_0\}$. In tale intorno,

   $$
   \frac{1}{g(x)} - \frac{1}{l_2} = \frac{l_2 - g(x)}{g(x) l_2}.
   $$

   Poiché $|g(x)| \to |l_2| > 0$, $g$ è limitata inferiormente in modulo in un intorno di $x_0$, quindi $\frac{1}{g(x)} \to \frac{1}{l_2}$. Applicando il punto (2) a $f$ e $\frac{1}{g}$, si conclude.

4. **Prodotto per scalare.** Segue dal punto (2) prendendo $g(x) \equiv \lambda$ costante. $\square$

> **Osservazione 3.7.**  
> L'algebra dei limiti permette di calcolare limiti complessi a partire da limiti noti, senza dover applicare ogni volta la definizione $\varepsilon$-$\delta$. Essa è lo strumento operativo principale nel calcolo dei limiti.

---

### 3.4 Teorema del confronto (due carabinieri)

#### Teorema 3.8 — Teorema del confronto

Siano $f, g, h : X \subseteq \mathbb{R}^n \to \mathbb{R}$ e sia $x_0$ di accumulazione per $X$. Supponiamo che esista un intorno $U$ di $x_0$ tale che

$$
f(x) \le g(x) \le h(x) \qquad \forall x \in (U \cap X) \setminus \{x_0\},
$$

e che

$$
\lim_{x \to x_0} f(x) = \lim_{x \to x_0} h(x) = l \in \mathbb{R}.
$$

Allora

$$
\lim_{x \to x_0} g(x) = l.
$$

**Dimostrazione.**

Per ipotesi $\lim_{x \to x_0} f(x) = l$, cioè per ogni $\varepsilon > 0$ esiste $\delta_\varepsilon^1 > 0$ tale che, se $0 < \|x - x_0\| < \delta_\varepsilon^1$, si ha

$$
|f(x) - l| < \varepsilon \iff l - \varepsilon < f(x) < l + \varepsilon.
$$

Analogamente, per ipotesi $\lim_{x \to x_0} h(x) = l$, cioè per ogni $\varepsilon > 0$ esiste $\delta_\varepsilon^2 > 0$ tale che, se $0 < \|x - x_0\| < \delta_\varepsilon^2$, si ha

$$
|h(x) - l| < \varepsilon \iff l - \varepsilon < h(x) < l + \varepsilon.
$$

Sia

$$
\delta_\varepsilon = \min\{\delta_\varepsilon^1, \delta_\varepsilon^2\} > 0,
$$

e sia $x \in X$ con $0 < \|x - x_0\| < \delta_\varepsilon$. Per la scelta di $\delta_\varepsilon$, valgono simultaneamente

$$
l - \varepsilon < f(x)
\qquad\text{e}\qquad
h(x) < l + \varepsilon.
$$

Combinando con l'ipotesi $f(x) \le g(x) \le h(x)$ (valida in un intorno sufficientemente piccolo di $x_0$, che possiamo assumere incluso nella palla di raggio $\delta_\varepsilon$), si ottiene

$$
l - \varepsilon < f(x) \le g(x) \le h(x) < l + \varepsilon.
$$

Dunque

$$
|g(x) - l| < \varepsilon.
$$

Quindi $\lim_{x \to x_0} g(x) = l$. $\square$

> **Osservazione 3.9.**  
> Il teorema del confronto è particolarmente utile per calcolare limiti di funzioni che non si riescono a manipolare direttamente, ma che possono essere "schiacciate" tra due funzioni più semplici con lo stesso limite. In $\mathbb{R}^n$, la tecnica tipica è maggiorare $|g(x)|$ con una funzione che tende a $0$.

---

### 3.5 Teorema di permanenza del segno (forma forte)

#### Teorema 3.10 — Permanenza del segno (forma forte)

Sia $\lim_{x \to x_0} f(x) = l \in \mathbb{R}$ con $l > 0$ (rispettivamente $l < 0$). Allora esiste un intorno $U$ di $x_0$ tale che

$$
f(x) > \frac{l}{2} > 0 \qquad \forall x \in (U \cap X) \setminus \{x_0\}
$$

(rispettivamente $f(x) < \frac{l}{2} < 0$).

**Dimostrazione.**  
Segue dalla dimostrazione del Teorema 3.3, scegliendo $\varepsilon = \frac{|l|}{2}$ e osservando che $|f(x) - l| < \frac{|l|}{2}$ implica $f(x) > l - \frac{|l|}{2}$. Per $l > 0$, $l - \frac{l}{2} = \frac{l}{2} > 0$. $\square$

## 4. Esempi di applicazione dei teoremi

### Esempio 4.1 — Applicazione dell'algebra dei limiti

$$
\lim_{(x,y) \to (1,2)} \frac{x^2 + y^2}{x + y} = \frac{1^2 + 2^2}{1 + 2} = \frac{5}{3}.
$$

### Esempio 4.2 — Applicazione del confronto

$$
\lim_{(x,y) \to (0,0)} \frac{x \sin(xy)}{x^2 + y^2} = 0
$$

grazie alla maggiorazione $\left| \frac{x \sin(xy)}{x^2 + y^2} \right| \le |y|$.

### Esempio 4.3 — Applicazione della permanenza del segno

$$
f(x,y) = x^2 + y^2 - 1
\implies
\lim_{(x,y) \to (2,0)} f(x,y) = 3 > 0.
$$

Per il Teorema 3.3, esiste un intorno di $(2,0)$ in cui $x^2 + y^2 > 1$.

## 5. Tecniche di calcolo dei limiti in $\mathbb{R}^n$

### 5.1 Norme equivalenti in $\mathbb{R}^n$

#### Teorema 5.1 — Equivalenza delle norme in $\mathbb{R}^n$

Su $\mathbb{R}^n$, per ogni $x \in \mathbb{R}^n$,

$$
\|x\|_\infty \le \|x\|_2 \le \|x\|_1 \le n \|x\|_\infty.
$$

Di conseguenza, le norme $\|\cdot\|_1$, $\|\cdot\|_2$ e $\|\cdot\|_\infty$ sono **equivalenti**: un insieme è aperto rispetto a una norma se e solo se lo è rispetto alle altre, e una successione converge in una norma se e solo se converge nelle altre.

**Conseguenza.** Nel calcolo dei limiti in $\mathbb{R}^n$, possiamo usare la norma che risulta più comoda, tipicamente $\|\cdot\|_\infty$ o $\|\cdot\|_1$.

### 5.2 Tecniche principali

1. **Sostituzione diretta** (continuità in $x_0$).
2. **Cambio di variabile** (riduzione a un limite noto scalare).
3. **Maggiorazione con disuguaglianze** (confronto con funzioni che tendono a $0$).
4. **Sviluppi di Taylor** (per funzioni regolari).
5. **Coordinate polari** (per limiti in $\mathbb{R}^2$, con verifica di uniformità in $\theta$).
6. **Restrizione a curve o successioni** (per dimostrare la **non esistenza** del limite).

## 6. Esempi di calcolo dei limiti

### Esempio 6.1 — Limite notevole con $\sin(xy)/(xy)$

Consideriamo

$$
\lim_{(x,y) \to (0,0)} \frac{\sin(xy)}{xy}.
$$

**Dominio.** $\operatorname{dom}f = \{(x,y) : x \neq 0, y \neq 0\}$; il punto $(0,0)$ è di accumulazione.

**Cambio di variabile.** Poniamo $t = xy$. Quando $(x,y) \to (0,0)$, si ha $t \to 0$. Allora

$$
\lim_{(x,y) \to (0,0)} \frac{\sin(xy)}{xy} = \lim_{t \to 0} \frac{\sin t}{t} = 1.
$$

**Verifica con $\varepsilon$-$\delta$.** Fissato $\varepsilon > 0$, esiste $\delta_0 > 0$ tale che $0 < |t| < \delta_0 \implies \left| \frac{\sin t}{t} - 1 \right| < \varepsilon$. Osservando che $|xy| \le \|(x,y)\|_\infty^2$ e scegliendo $\delta = \min\{1, \delta_0\}$, se $\|(x,y)\|_\infty < \delta$ allora $0 < |xy| < \delta_0$ e quindi $\left| \frac{\sin(xy)}{xy} - 1 \right| < \varepsilon$. $\square$

### Esempio 6.2 — Limite all'infinito con $\arctan$

$$
\lim_{(x,y) \to +\infty} \arctan(|x| + |y|).
$$

Sia $t = \|(x,y)\|_1 = |x| + |y|$. Quando $(x,y) \to +\infty$, $t \to +\infty$, quindi

$$
\lim_{(x,y) \to +\infty} \arctan(|x| + |y|) = \lim_{t \to +\infty} \arctan(t) = \frac{\pi}{2}.
$$

### Esempio 6.3 — Limite con restrizione a una curva

$$
\lim_{(x,y) \to (0,0)} \frac{x \sin(xy)}{x^2 + y^2}.
$$

**Candidato.** Restrizione $y = x$:

$$
f(x,x) = \frac{x \sin(x^2)}{2x^2} \to 0.
$$

**Verifica con maggiorazione.** Poiché $|\sin(t)| \le |t|$,

$$
0 \le \left| \frac{x \sin(xy)}{x^2 + y^2} \right| \le |x| \cdot \frac{|xy|}{x^2 + y^2} = \frac{x^2 |y|}{x^2 + y^2} \le |y| \to 0.
$$

Per il teorema del confronto, il limite è $0$. $\square$

### Esempio 6.4 — Limite con sviluppi di Taylor

$$
\lim_{(x,y) \to (0,0)} \frac{\ln(1 + xy) - \sin(xy)}{2x^2 + y^2}.
$$

Ponendo $t = xy$ e usando gli sviluppi $\ln(1+t) = t - t^2/2 + o(t^2)$ e $\sin t = t - t^3/6 + o(t^4)$:

$$
\ln(1 + xy) - \sin(xy) = -\frac{x^2 y^2}{2} + o(x^2 y^2).
$$

Allora

$$
\frac{\ln(1 + xy) - \sin(xy)}{2x^2 + y^2} = -\frac{1}{2} \cdot \frac{x^2 y^2}{2x^2 + y^2} + o\left(\frac{x^2 y^2}{2x^2 + y^2}\right).
$$

Poiché $0 \le \frac{x^2 y^2}{2x^2 + y^2} \le y^2 \to 0$, il limite è $0$. $\square$

### Esempio 6.5 — Limite che non esiste

$$
\lim_{(x,y) \to +\infty} \arctan(|x| - |y|).
$$

Lungo $y = 0$ con $x \to +\infty$: $\arctan(t) \to \pi/2$.  
Lungo $x = 0$ con $y \to +\infty$: $\arctan(-t) \to -\pi/2$.  

Limiti diversi, quindi il limite **non esiste**.

## 7. Teorema delle restrizioni

### Teorema 7.1 — Caratterizzazione del limite tramite restrizioni

Sia $f : \operatorname{dom}f \subseteq \mathbb{R}^n \to \mathbb{R}$ e sia $x_0 \in \mathbb{R}^n \cup \{\infty\}$ un punto di accumulazione per $\operatorname{dom}f$. Allora

$$
\lim_{x \to x_0} f(x) = l \in \mathbb{R} \cup \{\pm\infty\}
$$

se e solo se per ogni $A \subseteq \operatorname{dom}f$ per cui $x_0$ è punto di accumulazione per $A$ si ha

$$
\lim_{\substack{x \to x_0 \\ x \in A}} f(x) = l.
$$

Indicando con $f|_A : A \to \mathbb{R}$ la restrizione di $f$ ad $A$,

$$
\lim_{x \to x_0} f(x) = l \iff \lim_{x \to x_0} f|_A(x) = l \qquad \forall A \subseteq \operatorname{dom}f \text{ con } x_0 \in A'.
$$

**Dimostrazione.**

$(\Rightarrow)$ Fissiamo $A \subseteq \operatorname{dom}f$ con $x_0$ di accumulazione per $A$. Sia $\varepsilon > 0$. Per definizione di limite, esiste $\delta > 0$ tale che

$$
x \in \operatorname{dom}f,\ 0 < d(x, x_0) < \delta \implies |f(x) - l| < \varepsilon.
$$

In particolare, per $x \in A$ con $0 < d(x, x_0) < \delta$, la stessa disuguaglianza vale. Dunque $\lim_{x \to x_0} f|_A(x) = l$.

$(\Leftarrow)$ Supponiamo che $\lim_{x \to x_0} f|_A(x) = l$ per ogni $A \subseteq \operatorname{dom}f$ con $x_0$ di accumulazione per $A$. Prendendo $A = \operatorname{dom}f$, otteniamo $\lim_{x \to x_0} f(x) = l$. $\square$

### Corollario 7.2 — Criterio di non esistenza

Se esistono due sottoinsiemi $A, B \subseteq \operatorname{dom}f$ per cui $x_0$ è di accumulazione e

$$
\lim_{\substack{x \to x_0 \\ x \in A}} f(x) \neq \lim_{\substack{x \to x_0 \\ x \in B}} f(x),
$$

allora $\lim_{x \to x_0} f(x)$ non esiste.

**Dimostrazione.**  
Se il limite esistesse, per il Teorema 7.1 tutte le restrizioni dovrebbero avere lo stesso limite. Poiché due restrizioni hanno limiti diversi, il limite non può esistere. $\square$

### Esempio 7.3 — Non esistenza tramite rette

$$
\lim_{(x,y) \to (0,0)} \frac{xy}{x^2 + y^2}.
$$

Lungo $y = mx$:

$$
f(x, mx) = \frac{m x^2}{x^2(1 + m^2)} = \frac{m}{1 + m^2},
$$

che dipende da $m$. Il limite **non esiste**.

## 8. Limiti e coordinate polari

### 8.1 Cambio in coordinate polari

In $\mathbb{R}^2$, il passaggio a **coordinate polari** è

$$
x = x_0 + \rho \cos\theta, \qquad y = y_0 + \rho \sin\theta, \qquad \rho \geq 0,\ \theta \in [0, 2\pi).
$$

Il punto $(x_0, y_0)$ corrisponde a $\rho = 0$, indipendentemente da $\theta$.

### Teorema 8.1 — Calcolo dei limiti tramite coordinate polari

Sia $f : X \subseteq \mathbb{R}^2 \to \mathbb{R}$ e sia $(x_0, y_0)$ di accumulazione per $X$. Poniamo

$$
\begin{cases}
x = x_0 + \rho \cos\theta, \\
y = y_0 + \rho \sin\theta.
\end{cases}
$$

Se

$$
f(x_0 + \rho\cos\theta,\ y_0 + \rho\sin\theta) = g(\rho) \, h(\theta)
$$

e valgono:

1. $\displaystyle \lim_{\rho \to 0^+} g(\rho) = 0$;
2. esiste $M > 0$ tale che $|h(\theta)| \le M$ per ogni $\theta \in \mathbb{R}$,

allora

$$
\lim_{(x,y) \to (x_0, y_0)} f(x,y) = 0.
$$

**Dimostrazione.**  
Fissiamo $\varepsilon > 0$. Poiché $g(\rho) \to 0$ per $\rho \to 0^+$, esiste $\delta > 0$ tale che

$$
0 < \rho < \delta \implies |g(\rho)| < \frac{\varepsilon}{M}.
$$

Sia $(x,y) \in X$ con $0 < \|(x,y) - (x_0, y_0)\|_2 < \delta$. Allora $\rho = \|(x,y) - (x_0, y_0)\|_2 \in (0, \delta)$ e

$$
|f(x,y) - 0| = |g(\rho)| \, |h(\theta)| < \frac{\varepsilon}{M} \cdot M = \varepsilon.
$$

Dunque $\lim_{(x,y) \to (x_0, y_0)} f(x,y) = 0$. $\square$

> **Osservazione 8.2.**  
> Il teorema fornisce una **condizione sufficiente** per l'esistenza del limite, con valore $0$. Se la decomposizione $f = g(\rho) h(\theta)$ non è possibile, il teorema non si applica direttamente, ma si possono comunque usare stime di maggiorazione.

> **Osservazione 8.3 — Uniformità in $\theta$.**  
> La condizione di **limitatazza uniforme** di $h(\theta)$ è cruciale. Non basta che, per ogni $\theta$ fissato, $f(\rho\cos\theta, \rho\sin\theta) \to 0$: occorre che la convergenza sia **uniforme** rispetto a $\theta$. Se il limite dipende da $\theta$, il limite non esiste.

### Esempio 8.4 — Limite che non esiste (coordinate polari)

$$
\lim_{(x,y) \to (0,0)} \frac{x^2 y}{x^4 + y^2}.
$$

In coordinate polari:

$$
f(\rho\cos\theta, \rho\sin\theta) = \frac{\rho^3 \cos^2\theta \sin\theta}{\rho^4 \cos^4\theta + \rho^2 \sin^2\theta} = \frac{\rho \cos^2\theta \sin\theta}{\rho^2 \cos^4\theta + \sin^2\theta}.
$$

Il limite puntuale in $\theta$ è $0$ per $\sin\theta \neq 0$. Tuttavia, lungo la restrizione $y = x^2$:

$$
f(x, x^2) = \frac{x^4}{2x^4} = \frac{1}{2}.
$$

Il limite **non esiste**. Questo mostra che il passaggio a coordinate polari è ingannevole se non si verifica l'uniformità.

## 9. Limiti e successioni

### Teorema 9.1 — Caratterizzazione sequenziale del limite

Sia $f : \operatorname{dom}f \subseteq \mathbb{R}^n \to \mathbb{R}$ e sia $x_0$ di accumulazione per $\operatorname{dom}f$. Allora

$$
\lim_{x \to x_0} f(x) = l
\iff
\forall \{x_k\} \subseteq \operatorname{dom}f \setminus \{x_0\},\ x_k \to x_0 \implies f(x_k) \to l.
$$

### Corollario 9.2 — Criterio di non esistenza tramite successioni

Se esistono due successioni $\{x_k\}, \{y_k\} \subseteq \operatorname{dom}f \setminus \{x_0\}$ con $x_k \to x_0$, $y_k \to x_0$ e

$$
f(x_k) \to l_1, \qquad f(y_k) \to l_2, \qquad l_1 \neq l_2,
$$

allora $\lim_{x \to x_0} f(x)$ non esiste.

### Esempio 9.3

$$
\lim_{(x,y) \to (0,0)} \frac{x^2 - y^2}{x^2 + y^2}.
$$

Lungo $x_k = (1/k, 0)$: $f(x_k) \to 1$. Lungo $y_k = (0, 1/k)$: $f(y_k) \to -1$. Il limite **non esiste**.

## 10. Limiti iterati e limite doppio

### Definizione 10.1 — Limite iterato

Sia $f : \operatorname{dom}f \subseteq \mathbb{R}^2 \to \mathbb{R}$. I **limiti iterati** sono

$$
\lim_{x \to x_0} \left( \lim_{y \to y_0} f(x,y) \right)
\qquad \text{e} \qquad
\lim_{y \to y_0} \left( \lim_{x \to x_0} f(x,y) \right).
$$

### Osservazione 10.2 — Relazione con il limite doppio

- se il limite doppio esiste e i limiti iterati esistono, allora tutti coincidono;
- il limite doppio può esistere mentre i limiti iterati non esistono;
- i limiti iterati possono esistere e essere diversi tra loro, mentre il limite doppio non esiste.

### Esempio 10.3 — Limiti iterati diversi

$$
f(x,y) = \frac{x^2 - y^2}{x^2 + y^2}.
$$

$$
\lim_{y \to 0} \left( \lim_{x \to 0} f(x,y) \right) = -1,
\qquad
\lim_{x \to 0} \left( \lim_{y \to 0} f(x,y) \right) = 1.
$$

I due limiti iterati sono diversi, quindi il limite doppio **non esiste**.

> **Osservazione 10.4.**  
> L'esistenza e l'uguaglianza dei limiti iterati **non garantiscono** l'esistenza del limite doppio.

## 11. Continuità

### Definizione 11.1 — Continuità in un punto

Sia $f : X \subseteq \mathbb{R}^n \to \mathbb{R}$ e sia $x_0 \in X$. Si dice che $f$ è **continua** in $x_0$ se:

1. $x_0$ è un **punto isolato** di $X$; oppure
2. $x_0$ è di accumulazione per $X$ e
   $$
   \lim_{x \to x_0} f(x) = f(x_0).
   $$

### Definizione 11.2 — Continuità su un insieme

$f$ è **continua su** $X \subseteq \mathbb{R}^n$ se è continua in $x_0$ per ogni $x_0 \in X$.

### Osservazione 11.3

$f : X \subseteq \mathbb{R}^n \to \mathbb{R}^m$ è continua in $x_0$ se e solo se ogni componente $f_i$ è continua in $x_0$ per $i = 1, \dots, m$.

**Dimostrazione.**  
Segue dal Teorema 2.1: $f(x) \to f(x_0)$ se e solo se $f_i(x) \to f_i(x_0)$ per ogni $i$. $\square$

### Teorema 11.4 — Algebra delle funzioni continue

Siano $f, g : X \subseteq \mathbb{R}^n \to \mathbb{R}$ continue in $x_0$. Allora:

1. $f + g$ è continua in $x_0$;
2. $f \cdot g$ è continua in $x_0$;
3. se $g(x_0) \neq 0$, $\frac{f}{g}$ è continua in $x_0$.

**Dimostrazione.**  
Segue dal Teorema 3.6 (algebra dei limiti) applicato alla definizione di continuità. $\square$

### Teorema 11.5 — Continuità delle funzioni composte

Siano $f : X \subseteq \mathbb{R}^n \to \mathbb{R}^m$ continua in $x_0$, con $f(X) \subseteq Y$, e sia $g : Y \subseteq \mathbb{R}^m \to \mathbb{R}^k$ continua in $y_0 = f(x_0) \in Y$. Allora la funzione composta

$$
g \circ f : X \to \mathbb{R}^k, \qquad (g \circ f)(x) = g(f(x)),
$$

è continua in $x_0$.

**Dimostrazione.**  
Sia $\varepsilon > 0$. Poiché $g$ è continua in $y_0$, esiste $\eta > 0$ tale che

$$
y \in Y,\ \|y - y_0\| < \eta \implies \|g(y) - g(y_0)\| < \varepsilon.
$$

Poiché $f$ è continua in $x_0$, esiste $\delta > 0$ tale che

$$
x \in X,\ \|x - x_0\| < \delta \implies \|f(x) - f(x_0)\| < \eta.
$$

Allora, per $x \in X$ con $\|x - x_0\| < \delta$, si ha $f(x) \in Y$ e $\|f(x) - y_0\| < \eta$, quindi

$$
\|g(f(x)) - g(f(x_0))\| < \varepsilon.
$$

Dunque $g \circ f$ è continua in $x_0$. $\square$

### Teorema 11.6 — Continuità delle funzioni elementari

Le funzioni elementari (polinomi, funzioni razionali, trigonometriche, esponenziali, logaritmiche) sono continue su tutto il loro dominio.

**Conseguenza.** Se $f$ è una funzione elementare e $x_0 \in \operatorname{dom}f$, allora

$$
\lim_{x \to x_0} f(x) = f(x_0).
$$

### Teorema 11.7 — Weierstrass

Sia $f : X \subseteq \mathbb{R}^n \to \mathbb{R}$ con $X$ **chiuso e limitato** (cioè **compatto** in $\mathbb{R}^n$) e $f$ **continua** su $X$. Allora esistono $x_m, x_M \in X$ tali che

$$
f(x_m) \leq f(x) \leq f(x_M) \qquad \forall x \in X.
$$

In altre parole, $f$ assume su $X$ il suo **minimo** e il suo **massimo**.

**Idea della dimostrazione.**  
Poiché $X$ è compatto e $f$ è continua, l'immagine $f(X)$ è un sottoinsieme compatto di $\mathbb{R}$ (per il teorema di continuità su compatti). In $\mathbb{R}$, un insieme compatto è chiuso e limitato, quindi ammette massimo e minimo. Dunque esistono $x_m, x_M \in X$ con $f(x_m) = \min f(X)$ e $f(x_M) = \max f(X)$. $\square$

> **Osservazione 11.8.**  
> Il teorema di Weierstrass è un risultato **esistenziale**: garantisce l'esistenza di massimo e minimo, ma non fornisce un metodo per calcolarli. È uno strumento fondamentale in ottimizzazione.

### Teorema 11.9 — Insiemi di sotto/sopra livello di funzioni continue

Sia $f : \mathbb{R}^n \to \mathbb{R}$ continua su $\mathbb{R}^n$. Fissato $c \in \mathbb{R}$:

1. l'insieme $\{x \in \mathbb{R}^n : f(x) < c\}$ è **aperto**;
2. l'insieme $\{x \in \mathbb{R}^n : f(x) \le c\}$ è **chiuso**;
3. l'insieme $\{x \in \mathbb{R}^n : f(x) > c\}$ è **aperto**;
4. l'insieme $\{x \in \mathbb{R}^n : f(x) \ge c\}$ è **chiuso**.

**Dimostrazione.**  
Dimostriamo (1). L'insieme $\{f < c\}$ è la controimmagine di $(-\infty, c)$, che è aperto in $\mathbb{R}$:

$$
\{f < c\} = f^{-1}\bigl((-\infty, c)\bigr).
$$

Poiché $f$ è continua, la controimmagine di un aperto è aperta. Dunque $\{f < c\}$ è aperto.

Il punto (2) segue dal fatto che $\{f \le c\} = \{f < c\}^C$ è il complementare di un aperto, quindi è chiuso.

I punti (3) e (4) sono analoghi, usando gli aperti $(c, +\infty)$ e i loro complementari. $\square$

> **Osservazione 11.10.**  
> Il teorema è una conseguenza diretta della **caratterizzazione della continuità tramite controimmagini di aperti**: $f$ è continua su $\mathbb{R}^n$ se e solo se $f^{-1}(A)$ è aperto per ogni $A \subseteq \mathbb{R}$ aperto.
