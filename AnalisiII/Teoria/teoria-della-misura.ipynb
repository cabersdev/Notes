# Teoria della misura e integrale di Lebesgue

---

## 0. Perché serve una teoria della misura

### Osservazione 0.1

L'integrale di Riemann, pur sufficiente per molte applicazioni, ha limiti strutturali:

1. **classi di funzioni integrabili troppo ristrette**: la funzione di Dirichlet $\mathbf{1}_{\mathbb{Q}\cap[0,1]}$ non è integrabile secondo Riemann, ma è "piccola" (nulla q.o.);
2. **scarsa compatibilità con i limiti**: non vale il teorema di convergenza monotona né la convergenza dominata nella forma generale;
3. **difficoltà nel passaggio al limite sotto il segno di integrale**;
4. **estensione a insiemi "patologici"**: non è chiaro quali insiemi abbiano una nozione coerente di "lunghezza", "area", "volume".

La **teoria della misura** fornisce un quadro in cui:
- si definisce una nozione di **misura** $\mu$ su una famiglia di insiemi (la $\sigma$-algebra);
- si costruisce l'**integrale di Lebesgue** rispetto a $\mu$;
- valgono potenti **teoremi di convergenza** (Beppo Levi, Fatou, Lebesgue dominata);
- si dimostra il **teorema del cambiamento di variabili** nella sua forma più generale, che è alla base delle formule con coordinate polari, sferiche, cilindriche.

## 1. Spazi di misura

### Definizione 1.1 — $\sigma$-algebra

Sia $X$ un insieme non vuoto. Una famiglia $\mathcal{A}\subseteq\mathcal{P}(X)$ è una **$\sigma$-algebra** su $X$ se:

1. $X \in\mathcal{A}$;
2. **chiusura per complementare**: se $A\in\mathcal{A}$, allora $X\setminus A\in\mathcal{A}$;
3. **chiusura per unioni numerabili**: se $\{A_n\}_{n\in\mathbb{N}}\subseteq\mathcal{A}$, allora $\bigcup_{n}A_n\in\mathcal{A}$.

Gli elementi di $\mathcal{A}$ sono detti **insiemi misurabili**.

### Proposizione 1.2 — Proprietà immediate

Da 1, 2, 3 seguono:

1. $\varnothing\in\mathcal{A}$;
2. **chiusura per intersezioni numerabili**: se $\{A_n\}\subseteq\mathcal{A}$, allora $\bigcap_n A_n\in\mathcal{A}$;
3. **chiusura per differenze**: se $A,B\in\mathcal{A}$, allora $A\setminus B\in\mathcal{A}$;
4. **chiusura per unioni e intersezioni finite**.

**Dimostrazione.**  
$\varnothing = X\setminus X\in\mathcal{A}$. Se $A_n\in\mathcal{A}$, allora $X\setminus A_n\in\mathcal{A}$ e per la 3 $\bigcup_n(X\setminus A_n)\in\mathcal{A}$; il complementare di questa unione è $\bigcap_n A_n$, che è in $\mathcal{A}$ per la 2. Le altre proprietà seguono per induzione. $\square$

### Esempio 1.3 — $\sigma$-algebre fondamentali

- **Banale**: $\mathcal{A}=\{\varnothing,X\}$.
- **Discreta**: $\mathcal{A}=\mathcal{P}(X)$.
- **Boreliana**: se $X$ è uno spazio topologico, la **$\sigma$-algebra di Borel** $\mathcal{B}(X)$ è la più piccola $\sigma$-algebra contenente gli aperti (equivalentemente, i chiusi). In $\mathbb{R}^n$ è la $\sigma$-algebra generata dagli intervalli aperti (o dalle palle aperte).
- **Generata da una famiglia**: data $\mathcal{G}\subseteq\mathcal{P}(X)$, si definisce $\sigma(\mathcal{G})$ come l'intersezione di tutte le $\sigma$-algebre contenenti $\mathcal{G}$; è la più piccola $\sigma$-algebra contenente $\mathcal{G}$.

### Definizione 1.4 — Misura

Sia $(X,\mathcal{A})$ uno spazio misurabile. Una **misura** (positiva) è una funzione
$$
\mu:\mathcal{A}\to[0,+\infty]
$$
tale che:

1. $\mu(\varnothing)=0$;
2. **$\sigma$-additività**: se $\{A_n\}\subseteq\mathcal{A}$ sono a due a due disgiunti, allora
$$
\mu\left(\bigcup_{n=1}^\infty A_n\right)=\sum_{n=1}^\infty\mu(A_n).
$$

La terna $(X,\mathcal{A},\mu)$ è detta **spazio di misura**.

### Proposizione 1.5 — Proprietà elementari

Valgono:

1. **monotonia**: se $A\subseteq B$, allora $\mu(A)\le\mu(B)$;
2. **sottrattività**: se $A\subseteq B$ e $\mu(A)<\infty$, allora $\mu(B\setminus A)=\mu(B)-\mu(A)$;
3. **subadditività numerabile**: $\mu\left(\bigcup_n A_n\right)\le\sum_n\mu(A_n)$;
4. **continuità dal basso**: se $A_n\uparrow A$, allora $\mu(A_n)\to\mu(A)$;
5. **continuità dall'alto**: se $A_n\downarrow A$ e $\mu(A_1)<\infty$, allora $\mu(A_n)\to\mu(A)$.

**Dimostrazione.**  
1. $B=A\cup(B\setminus A)$ con i due insiemi disgiunti; per additività $\mu(B)=\mu(A)+\mu(B\setminus A)\ge\mu(A)$.
2. Segue da 1 applicato a $B=A\cup(B\setminus A)$.
3. Si pongono $B_1=A_1$ e $B_n=A_n\setminus\bigcup_{k<n}A_k$; allora $\bigcup_n B_n=\bigcup_n A_n$, i $B_n$ sono disgiunti e $B_n\subseteq A_n$, quindi $\mu(\bigcup A_n)=\sum\mu(B_n)\le\sum\mu(A_n)$.
4. Si pongono $B_1=A_1$ e $B_n=A_n\setminus A_{n-1}$ per $n\ge2$; i $B_n$ sono disgiunti e $\bigcup_n B_n=A$. Allora $\mu(A_n)=\sum_{k=1}^n\mu(B_k)\to\sum_k\mu(B_k)=\mu(A)$.
5. Applicare 4 a $A_1\setminus A_n\uparrow A_1\setminus A$. $\square$

### Definizione 1.6 — Misura finita, $\sigma$-finita, completa

- $\mu$ è **finita** se $\mu(X)<\infty$.
- $\mu$ è **$\sigma$-finita** se $X=\bigcup_n X_n$ con $\mu(X_n)<\infty$.
- $\mu$ è **completa** se, dato $A\in\mathcal{A}$ con $\mu(A)=0$ e $B\subseteq A$, allora $B\in\mathcal{A}$ (e quindi $\mu(B)=0$).

### Esempio 1.7 — Misure fondamentali

1. **Misura di conteggio** su $(X,\mathcal{P}(X))$: $\mu(A)=|A|$ se $A$ finito, $+\infty$ altrimenti.
2. **Delta di Dirac** in $x_0$: $\delta_{x_0}(A)=1$ se $x_0\in A$, $0$ altrimenti.
3. **Misura di Lebesgue** $\lambda^n$ su $(\mathbb{R}^n,\mathcal{B}(\mathbb{R}^n))$: unica misura (a meno di normalizzazione) invariante per traslazioni e tale che $\lambda^n([0,1]^n)=1$. Per $n=1$, $\lambda([a,b])=b-a$.
4. **Misura con densità**: data $f:X\to[0,\infty]$ misurabile, $\mu_f(A)=\int_A f\,d\mu$. È una misura su $\mathcal{A}$.
5. **Misure di probabilità**: $\mu(X)=1$.

> **Osservazione 1.8.**  
> La misura di Lebesgue su $\mathbb{R}^n$ è **$\sigma$-finita** ma non finita. È **completa** se si considera la $\sigma$-algebra di Lebesgue $\mathcal{L}^n$ (completamento di $\mathcal{B}(\mathbb{R}^n)$ rispetto a $\lambda^n$), che contiene anche i sottoinsiemi dei boreliani di misura nulla. La distinzione tra $\mathcal{B}$ e $\mathcal{L}$ è tecnica; in pratica si lavora con $\mathcal{L}^n$.

## 2. Funzioni misurabili

### Definizione 2.1 — Funzione misurabile

Siano $(X,\mathcal{A})$ e $(Y,\mathcal{B})$ spazi misurabili. Una funzione $f:X\to Y$ è **misurabile** se
$$
f^{-1}(B)\in\mathcal{A}\qquad\forall B\in\mathcal{B}.
$$

Nel caso $Y=\mathbb{R}$ con la $\sigma$-algebra di Borel, $f$ è misurabile se e solo se, per ogni $t\in\mathbb{R}$,
$$
\{x\in X : f(x)>t\}\in\mathcal{A}
$$
(equivalentemente con $<$; si può usare qualsiasi intervallo).

### Proposizione 2.2 — Chiusura per operazioni

Se $f,g:X\to\mathbb{R}$ sono misurabili, allora sono misurabili:

1. $f+g$, $f\cdot g$;
2. $\max(f,g)$, $\min(f,g)$;
3. $f^+=\max(f,0)$, $f^-=\max(-f,0)$;
4. $|f|$;
5. $\sup_n f_n$, $\inf_n f_n$, $\limsup_n f_n$, $\liminf_n f_n$ (se le $f_n$ sono misurabili).

**Dimostrazione (idea).**  
Gli insiemi del tipo $\{x:f(x)+g(x)>t\}$ si scrivono come unioni numerabili di insiemi misurabili del tipo $\{x:f(x)>r\}\cap\{x:g(x)>s\}$ con $r+s>t$, $r,s\in\mathbb{Q}$. Le altre proprietà seguono da scritture analoghe. $\square$

### Definizione 2.3 — Funzione semplice

Una funzione $s:X\to[0,\infty)$ è **semplice** se assume un numero finito di valori non negativi. Ogni funzione semplice si scrive in forma canonica
$$
s=\sum_{k=1}^n a_k\mathbf{1}_{A_k},
$$
con $a_k\ge0$ e $A_k\in\mathcal{A}$ a due a due disgiunti.

### Teorema 2.4 — Approssimazione con funzioni semplici

Per ogni $f:X\to[0,\infty]$ misurabile esiste una successione crescente di funzioni semplici $s_n$ tale che $s_n(x)\uparrow f(x)$ per ogni $x\in X$.

**Dimostrazione.**  
Si definisce, per $n\ge1$ e $k=0,\dots,n2^n-1$,
$$
A_{n,k}=\left\{x : \frac{k}{2^n}\le f(x)<\frac{k+1}{2^n}\right\},
\qquad
B_n=\{x : f(x)\ge n\},
$$
e
$$
s_n=\sum_{k=0}^{n2^n-1}\frac{k}{2^n}\mathbf{1}_{A_{n,k}}+n\mathbf{1}_{B_n}.
$$
Allora $s_n$ è semplice, $0\le s_n\le s_{n+1}\le f$, e per ogni $x$ con $f(x)<\infty$ si ha $s_n(x)\to f(x)$; se $f(x)=\infty$, $s_n(x)=n\to\infty$. $\square$

> **Osservazione 2.5.**  
> Il teorema 2.4 è il punto di partenza per la costruzione dell'integrale di Lebesgue: si definisce l'integrale per le funzioni semplici e poi si estende per monotonia alle funzioni non negative, e per differenza a quelle di segno qualsiasi.

## 3. Integrale di Lebesgue

### Definizione 3.1 — Integrale di una funzione semplice

Sia $s=\sum_{k=1}^n a_k\mathbf{1}_{A_k}$ una funzione semplice non negativa in forma canonica. Si definisce
$$
\int_X s\,d\mu=\sum_{k=1}^n a_k\mu(A_k)\in[0,+\infty].
$$
La definizione è ben posta: non dipende dalla rappresentazione scelta, perché la forma canonica è unica.

### Definizione 3.2 — Integrale di una funzione non negativa

Sia $f:X\to[0,\infty]$ misurabile. Si definisce
$$
\int_X f\,d\mu=\sup\left\{\int_X s\,d\mu : s\text{ semplice},\ 0\le s\le f\right\}\in[0,+\infty].
$$

> **Osservazione 3.3.**  
> L'integrale di Lebesgue è definito come estremo superiore degli integrali delle funzioni semplici dominate da $f$. Per il Teorema 2.4, questo estremo è raggiunto lungo una successione crescente di funzioni semplici, e quindi è il limite degli integrali.

### Definizione 3.4 — Funzioni integrabili

Sia $f:X\to\mathbb{R}$ misurabile. Si decompone $f=f^+-f^-$ con $f^\pm\ge0$ misurabili. Si dice che $f$ è **integrabile** (o **sommabile**) rispetto a $\mu$ se
$$
\int_X f^+\,d\mu<\infty
\qquad\text{e}\qquad
\int_X f^-\,d\mu<\infty.
$$
In tal caso si definisce
$$
\int_X f\,d\mu=\int_X f^+\,d\mu-\int_X f^-\,d\mu.
$$
Si denota con $L^1(X,\mu)$ l'insieme delle funzioni integrabili (a meno di uguaglianza q.o.).

> **Osservazione 3.5.**  
> L'integrale di Lebesgue estende quello di Riemann: se $f$ è Riemann-integrabile su $[a,b]$, allora è Lebesgue-integrabile e i due integrali coincidono. Ma l'integrale di Lebesgue è definito su una classe molto più ampia di funzioni e insiemi.

### Proposizione 3.6 — Proprietà elementari

Siano $f,g$ integrabili e $\alpha,\beta\in\mathbb{R}$. Allora:

1. **linearità**: $\int(\alpha f+\beta g)\,d\mu=\alpha\int f\,d\mu+\beta\int g\,d\mu$;
2. **monotonia**: se $f\le g$ q.o., allora $\int f\,d\mu\le\int g\,d\mu$;
3. **disuguaglianza triangolare**: $\left|\int f\,d\mu\right|\le\int|f|\,d\mu$;
4. **integrale su insiemi disgiunti**: se $A\cap B=\varnothing$, $\int_{A\cup B}f\,d\mu=\int_A f\,d\mu+\int_B f\,d\mu$;
5. **funzioni nulle q.o.**: se $f=0$ q.o., allora $\int f\,d\mu=0$; in particolare, funzioni uguali q.o. hanno lo stesso integrale.

**Dimostrazione.**  
La linearità segue dalla definizione per le funzioni semplici e dal passaggio al limite per monotonia. La monotonia segue dalla definizione di integrale come sup. La disuguaglianza triangolare segue da $-|f|\le f\le|f|$ e dalla monotonia. L'additività sugli insiemi segue dalla definizione con le funzioni indicatrici. Per le funzioni nulle q.o., l'insieme $\{x:f(x)\ne0\}$ ha misura nulla, e il sup nella definizione di integrale può essere calcolato su funzioni semplici nulle su tale insieme. $\square$

## 4. Teoremi di convergenza

### Teorema 4.1 — Convergenza monotona (Beppo Levi)

Sia $\{f_n\}$ una successione di funzioni misurabili non negative con $f_n\uparrow f$ q.o. Allora
$$
\lim_{n\to\infty}\int_X f_n\,d\mu=\int_X f\,d\mu.
$$

**Dimostrazione.**  
Poiché $f_n\le f$, si ha $\int f_n\le\int f$, quindi $\limsup_n\int f_n\le\int f$. Per la disuguaglianza opposta, sia $s$ una funzione semplice con $0\le s\le f$ e sia $\varepsilon>0$. Si considerano gli insiemi $A_n=\{x:f_n(x)\ge(1-\varepsilon)s(x)\}$, che sono crescenti e la cui unione è $X$ (a meno di un insieme di misura nulla). Allora
$$
\int f_n\ge\int_{A_n}f_n\ge(1-\varepsilon)\int_{A_n}s\,d\mu\xrightarrow{n\to\infty}(1-\varepsilon)\int s\,d\mu.
$$
Passando al limite per $\varepsilon\to0$ e poi al sup su $s$, si ottiene $\liminf_n\int f_n\ge\int f$. $\square$

### Teorema 4.2 — Lemma di Fatou

Sia $\{f_n\}$ una successione di funzioni misurabili non negative. Allora
$$
\int_X\liminf_{n\to\infty}f_n\,d\mu\le\liminf_{n\to\infty}\int_X f_n\,d\mu.
$$

**Dimostrazione.**  
Si pone $g_n=\inf_{k\ge n}f_k$. Allora $g_n$ è misurabile non negativa, $g_n\uparrow\liminf_n f_n$, e $g_n\le f_n$. Per Beppo Levi,
$$
\int\liminf f_n=\lim_n\int g_n\le\liminf_n\int f_n.
$$
$\square$

### Teorema 4.3 — Convergenza dominata (Lebesgue)

Sia $\{f_n\}$ una successione di funzioni misurabili con $f_n\to f$ q.o. e tale che esista $g\in L^1$ con $|f_n|\le g$ q.o. per ogni $n$. Allora $f\in L^1$ e
$$
\lim_{n\to\infty}\int_X f_n\,d\mu=\int_X f\,d\mu.
$$

**Dimostrazione.**  
Si applica Fatou a $g+f_n\ge0$ e a $g-f_n\ge0$:
$$
\int(g+f)\le\liminf_n\int(g+f_n)
\;\Longrightarrow\;
\int f\le\liminf_n\int f_n,
$$
$$
\int(g-f)\le\liminf_n\int(g-f_n)
\;\Longrightarrow\;
-\int f\le\liminf_n\left(-\int f_n\right)
\;\Longrightarrow\;
\int f\ge\limsup_n\int f_n.
$$
Combinando, $\lim_n\int f_n=\int f$. $\square$

> **Osservazione 4.4.**  
> I tre teoremi di convergenza sono il cuore dell'analisi reale. Beppo Levi è lo strumento per costruire l'integrale (passaggio da funzioni semplici a funzioni generali). Fatou è un lemma tecnico ma potente, usato per dimostrare la dominata. La dominata è il teorema più usato nelle applicazioni: garantisce il passaggio al limite sotto il segno di integrale quando c'è una funzione dominante integrabile.

> **Osservazione 4.5.**  
> Senza l'ipotesi di dominazione, il passaggio al limite può fallire: la successione $f_n=n\mathbf{1}_{(0,1/n)}$ converge a $0$ puntualmente, ma $\int f_n=1$ per ogni $n$. La dominata è quindi una condizione **sufficiente**, non necessaria.

## 5. Misure prodotto e teorema di Fubini–Tonelli

### Definizione 5.1 — Misura prodotto

Siano $(X,\mathcal{A},\mu)$ e $(Y,\mathcal{B},\nu)$ spazi di misura $\sigma$-finiti. Sia $\mathcal{A}\otimes\mathcal{B}$ la $\sigma$-algebra su $X\times Y$ generata dai rettangoli $A\times B$ con $A\in\mathcal{A}$, $B\in\mathcal{B}$. Esiste un'unica misura $\mu\otimes\nu$ su $\mathcal{A}\otimes\mathcal{B}$ tale che
$$
(\mu\otimes\nu)(A\times B)=\mu(A)\nu(B)
\qquad\forall A\in\mathcal{A},\ B\in\mathcal{B}.
$$
Tale misura è detta **misura prodotto**.

> **Esempio 5.2.**  
> La misura di Lebesgue su $\mathbb{R}^{m+n}$ è il prodotto delle misure di Lebesgue su $\mathbb{R}^m$ e $\mathbb{R}^n$:
> $$
> \lambda^{m+n}=\lambda^m\otimes\lambda^n.
> $$

### Teorema 5.3 — Tonelli

Sia $f:X\times Y\to[0,\infty]$ misurabile rispetto a $\mathcal{A}\otimes\mathcal{B}$. Allora:

1. per $\mu$-q.o. $x$, la funzione $y\mapsto f(x,y)$ è $\mathcal{B}$-misurabile;
2. per $\nu$-q.o. $y$, la funzione $x\mapsto f(x,y)$ è $\mathcal{A}$-misurabile;
3. vale
$$
\int_{X\times Y}f\,d(\mu\otimes\nu)=\int_X\left(\int_Y f(x,y)\,d\nu(y)\right)d\mu(x)=\int_Y\left(\int_X f(x,y)\,d\mu(x)\right)d\nu(y).
$$

> **Osservazione 5.4.**  
> Tonelli non richiede ipotesi di integrabilità: $f$ può assumere valore $+\infty$, e l'uguaglianza vale in $[0,+\infty]$. Questo lo rende lo strumento naturale per verificare se una funzione è integrabile: se uno dei due integrali iterati è finito, allora $f\in L^1$ e vale l'uguaglianza.

### Teorema 5.5 — Fubini

Sia $f:X\times Y\to\mathbb{R}$ integrabile rispetto a $\mu\otimes\nu$. Allora:

1. per $\mu$-q.o. $x$, $y\mapsto f(x,y)$ è integrabile rispetto a $\nu$;
2. per $\nu$-q.o. $y$, $x\mapsto f(x,y)$ è integrabile rispetto a $\mu$;
3. vale
$$
\int_{X\times Y}f\,d(\mu\otimes\nu)=\int_X\left(\int_Y f(x,y)\,d\nu(y)\right)d\mu(x)=\int_Y\left(\int_X f(x,y)\,d\mu(x)\right)d\nu(y).
$$

> **Osservazione 5.6.**  
> La differenza tra Tonelli e Fubini: Tonelli vale per funzioni non negative, senza ipotesi di integrabilità; Fubini vale per funzioni integrabili, con segno qualsiasi. In pratica, si usa Tonelli per verificare l'integrabilità, poi Fubini per scambiare l'ordine di integrazione.

> **Esempio 5.7 — L'ordine di integrazione conta.**  
> La funzione
> $$
> f(x,y)=\frac{x^2-y^2}{(x^2+y^2)^2}
> $$
> su $(0,1)^2$ è integrabile in modulo? No. Infatti $\int_0^1\int_0^1|f|\,dx\,dy=+\infty$, quindi Fubini non si applica. E in effetti i due integrali iterati danno risultati diversi: $\int_0^1\int_0^1 f\,dx\,dy=\pi/4$ e $\int_0^1\int_0^1 f\,dy\,dx=-\pi/4$. Questo mostra che senza integrabilità l'ordine di integrazione non può essere scambiato.

## 6. Misura di Lebesgue in $\mathbb{R}^n$

### Definizione 6.1 — Misura di Lebesgue

Esiste un'unica misura $\lambda^n$ su $\mathcal{B}(\mathbb{R}^n)$ (o sul suo completamento $\mathcal{L}^n$) tale che:

1. $\lambda^n([0,1]^n)=1$;
2. $\lambda^n$ è **invariante per traslazioni**: $\lambda^n(A+v)=\lambda^n(A)$ per ogni $A$ e ogni $v\in\mathbb{R}^n$;
3. $\lambda^n$ è **regolare**: per ogni $A\in\mathcal{L}^n$,
$$
\lambda^n(A)=\inf\{\lambda^n(U) : U\supseteq A,\ U\text{ aperto}\}=\sup\{\lambda^n(K) : K\subseteq A,\ K\text{ compatto}\}.
$$

La misura $\lambda^n$ è detta **misura di Lebesgue** $n$-dimensionale. La sua restrizione a $\mathcal{B}(\mathbb{R}^n)$ è $\sigma$-finita e non atomica.

> **Osservazione 6.2.**  
> La regolarità è la proprietà che rende la misura di Lebesgue "compatibile" con la topologia: ogni insieme misurabile è approssimabile dall'esterno con aperti e dall'interno con compatti. Questo è il ponte tra teoria della misura e topologia, e permette di usare argomenti di continuità e compattezza.

### Teorema 6.3 — Insiemi di misura nulla

Un insieme $N\subseteq\mathbb{R}^n$ ha misura di Lebesgue nulla se e solo se per ogni $\varepsilon>0$ esiste un ricoprimento numerabile di $N$ con palle $B(x_k,r_k)$ tale che
$$
\sum_k r_k^n<\varepsilon.
$$

> **Esempio 6.4.**  
> - I punti hanno misura nulla: $\lambda^n(\{x\})=0$.
> - Le rette in $\mathbb{R}^2$ hanno misura nulla; i piani in $\mathbb{R}^3$ hanno misura nulla.
> - L'insieme di Cantor ha misura nulla in $\mathbb{R}$, ma non è numerabile.
> - L'insieme $\mathbb{Q}\cap[0,1]$ ha misura nulla.
> - Esistono insiemi non misurabili (Vitali, 1905), la cui esistenza richiede l'assioma della scelta; la $\sigma$-algebra di Lebesgue $\mathcal{L}^n$ è strettamente contenuta in $\mathcal{P}(\mathbb{R}^n)$.

> **Osservazione 6.5.**  
> L'integrale di Riemann e quello di Lebesgue coincidono per le funzioni Riemann-integrabili. Ma l'integrale di Lebesgue è definito su una classe molto più ampia: la funzione $\mathbf{1}_{\mathbb{Q}\cap[0,1]}$ è Lebesgue-integrabile (con integrale $0$), mentre non è Riemann-integrabile. Questo è il primo esempio di come la teoria della misura estenda in modo naturale il calcolo integrale.

## 7. Cambiamento di variabili

### Teorema 7.1 — Cambiamento di variabili (forma generale)

Siano $U,V\subseteq\mathbb{R}^n$ aperti e $\phi:U\to V$ un **diffeomorfismo** di classe $C^1$ (cioè biiettivo, con $\phi$ e $\phi^{-1}$ di classe $C^1$). Allora per ogni funzione $f:V\to[0,\infty]$ misurabile (o integrabile),
$$
\int_V f(x)\,dx=\int_U f(\phi(u))\,|\det J_\phi(u)|\,du,
$$
dove $J_\phi(u)$ è la matrice Jacobiana di $\phi$ in $u$.

**Idea della dimostrazione.**  
Si dimostra prima per applicazioni lineari ($\phi(u)=Au$, con $|\det A|$ che scala il volume), poi per diffeomorfismi che sono l'identità su un intorno del bordo e lineari localmente, e infine per diffeomorfismi generali usando una partizione dell'unità. Il punto chiave è che $|\det J_\phi(u)|$ misura il fattore di scala locale del volume: un cubetto infinitesimo di lato $du$ in $U$ viene mappato in un parallelepipedo di volume $|\det J_\phi(u)|\,du$ in $V$. $\square$

> **Osservazione 7.2.**  
> Il teorema 7.1 è la forma rigorosa della formula usata negli appunti di calcolo integrale:
> $$
> \int_D f(x)\,dx=\int_{\phi^{-1}(D)}f(\phi(u))|\det J_\phi(u)|\,du.
> $$
> Le coordinate polari, sferiche, cilindriche ed ellittiche sono tutte applicazioni di questo teorema, con $\phi$ definita esplicitamente.

### Esempio 7.3 — Coordinate polari in $\mathbb{R}^2$

Sia
$$
\phi(\rho,\theta)=(\rho\cos\theta,\rho\sin\theta),
\qquad
\rho>0,\ 0<\theta<2\pi.
$$
La matrice Jacobiana è
$$
J_\phi(\rho,\theta)=
\begin{pmatrix}
\cos\theta & -\rho\sin\theta\\
\sin\theta & \rho\cos\theta
\end{pmatrix},
$$
e
$$
\det J_\phi(\rho,\theta)=\rho(\cos^2\theta+\sin^2\theta)=\rho.
$$
Quindi, per ogni $D\subseteq\mathbb{R}^2$ misurabile,
$$
\int_D f(x,y)\,dx\,dy=\int_{\phi^{-1}(D)}f(\rho\cos\theta,\rho\sin\theta)\,\rho\,d\rho\,d\theta.
$$
Il fattore $\rho$ è il **fattore di scala locale del volume** in coordinate polari. L'ipotesi $\rho>0$, $0<\theta<2\pi$ esclude l'origine e il semiasse positivo delle ascisse; ma questi insiemi hanno misura nulla, quindi l'integrale non cambia (si veda l'Esempio 7.4).

> **Esempio 7.4 — Insiemi di misura nulla e cambio di variabili.**  
> Nel cambio di variabili, il diffeomorfismo può essere definito solo su un aperto che esclude un insieme di misura nulla. Ad esempio, in coordinate polari, $\phi$ è definita su $(0,\infty)\times(0,2\pi)$, che esclude il semiasse positivo delle ascisse; ma tale semiasse ha misura nulla, quindi $\int_D=\int_{D\setminus\text{semiasse}}$. In generale, se $D_0$ ha misura nulla,
> $$
> \int_D f=\int_{D\setminus D_0}f.
> $$
> Questo permette di applicare il teorema 7.1 anche quando $\phi$ non è globalmente biiettivo.

### Esempio 7.5 — Coordinate sferiche in $\mathbb{R}^3$

Sia
$$
\phi(\rho,\theta,\varphi)=(\rho\sin\varphi\cos\theta,\ \rho\sin\varphi\sin\theta,\ \rho\cos\varphi),
$$
con $\rho>0$, $0<\theta<2\pi$, $0<\varphi<\pi$. La matrice Jacobiana ha determinante
$$
\det J_\phi=\rho^2\sin\varphi.
$$
Il fattore di scala locale del volume è quindi $\rho^2\sin\varphi$.

**Volume della sfera di raggio $R$.**  
La sfera $D=\{x^2+y^2+z^2\le R^2\}$ corrisponde, in coordinate sferiche, al dominio
$$
\phi^{-1}(D)=\{0<\rho<R,\ 0<\theta<2\pi,\ 0<\varphi<\pi\}.
$$
Allora
$$
\operatorname{vol}(D)=\int_D 1\,dx\,dy\,dz
=\int_0^{2\pi}\int_0^\pi\int_0^R\rho^2\sin\varphi\,d\rho\,d\varphi\,d\theta
=2\pi\cdot 2\cdot\frac{R^3}{3}
=\frac{4}{3}\pi R^3.
$$

### Esempio 7.6 — Coordinate cilindriche in $\mathbb{R}^3$

Sia
$$
\phi(\rho,\theta,z)=(\rho\cos\theta,\rho\sin\theta,z),
$$
con $\rho>0$, $0<\theta<2\pi$, $z\in\mathbb{R}$. Allora
$$
\det J_\phi=\rho.
$$
Il fattore di scala locale è $\rho$, come in coordinate polari (il fattore $z$ è invariante).

**Volume del cilindro.**  
Il cilindro $D=\{x^2+y^2\le R^2,\ 0\le z\le h\}$ corrisponde a
$$
\phi^{-1}(D)=\{0<\rho<R,\ 0<\theta<2\pi,\ 0<z<h\}.
$$
Allora
$$
\operatorname{vol}(D)=\int_0^{2\pi}\int_0^R\int_0^h\rho\,dz\,d\rho\,d\theta
=\pi R^2 h.
$$

> **Osservazione 7.7.**  
> I tre esempi (polari, sferiche, cilindriche) mostrano che il teorema del cambiamento di variabili unifica tutte le formule di trasformazione usate negli appunti di calcolo integrale. Il punto comune è il calcolo del determinante Jacobiano, che rappresenta il fattore di scala locale del volume. La scelta delle coordinate è guidata dalla geometria del dominio: coordinate polari per domini circolari in $\mathbb{R}^2$, sferiche per domini sferici, cilindriche per domini a simmetria cilindrica.

### Esempio 7.8 — Coordinate ellittiche in $\mathbb{R}^2$

Per il dominio ellittico
$$
D=\left\{(x,y):\frac{x^2}{a^2}+\frac{y^2}{b^2}\le 1\right\},
$$
si usano le coordinate
$$
\phi(\rho,\theta)=(a\rho\cos\theta,b\rho\sin\theta),
\qquad
\rho>0,\ 0<\theta<2\pi.
$$
La matrice Jacobiana è
$$
J_\phi=
\begin{pmatrix}
a\cos\theta & -a\rho\sin\theta\\
b\sin\theta & b\rho\cos\theta
\end{pmatrix},
$$
e
$$
\det J_\phi=ab\rho(\cos^2\theta+\sin^2\theta)=ab\rho.
$$
Il fattore di scala è $ab\rho$, e l'area dell'ellisse è
$$
\operatorname{area}(D)=\int_0^{2\pi}\int_0^1 ab\rho\,d\rho\,d\theta
=2\pi\cdot ab\cdot\frac12
=\pi ab.
$$
Questo generalizza l'area del cerchio ($a=b=R$), che vale $\pi R^2$.

## 8. Formule di riduzione

### Teorema 8.1 — Riduzione per sezioni (o per fette)

Sia $D\subseteq\mathbb{R}^n$ misurabile, e sia $f:D\to[0,\infty]$ misurabile. Si scriva $\mathbb{R}^n=\mathbb{R}^k\times\mathbb{R}^{n-k}$ e si denoti con $x=(u,v)$, $u\in\mathbb{R}^k$, $v\in\mathbb{R}^{n-k}$. Allora
$$
\int_D f(x)\,dx
=
\int_{\mathbb{R}^k}\left(\int_{D_u}f(u,v)\,dv\right)du,
$$
dove $D_u=\{v\in\mathbb{R}^{n-k}:(u,v)\in D\}$ è la **sezione** di $D$ in corrispondenza di $u$.

**Dimostrazione.**  
Si applica il teorema di Tonelli alla funzione $(u,v)\mapsto f(u,v)\mathbf{1}_D(u,v)$, che è misurabile e non negativa. L'integrale su $\mathbb{R}^n$ si scrive come integrale iterato, e l'integrale interno è esattamente l'integrale sulla sezione. $\square$

> **Osservazione 8.2.**  
> Il teorema 8.1 è la base delle formule "per fette" e "per fili" usate negli appunti di calcolo integrale:
> - **per fette** (o sezioni): si integra su $k$ coordinate e poi sulle rimanenti;
> - **per fili**: si integra lungo una direzione (ad esempio $z$) e poi sulle altre.
>
> La scelta dipende dalla geometria del dominio: se $D$ è definito da disuguaglianze del tipo $g_1(u)\le v\le g_2(u)$, la sezione $D_u$ è un intervallo e l'integrale interno è un integrale in una variabile.

### Esempio 8.3 — Integrale per fili

Sia $D=\{(x,y,z):x^2+y^2\le z\le 1\}$. Allora $D$ è compreso tra il paraboloide $z=x^2+y^2$ e il piano $z=1$. Fissato $(x,y)$ con $x^2+y^2\le 1$, la sezione verticale è
$$
D_{(x,y)}=\{z\in\mathbb{R}:x^2+y^2\le z\le 1\}=[x^2+y^2,1].
$$
Quindi
$$
\int_D f\,dV
=
\int_{x^2+y^2\le 1}\left(\int_{x^2+y^2}^1 f(x,y,z)\,dz\right)dx\,dy.
$$
Questo è il metodo "per fili": si integra lungo $z$ e poi sulle altre variabili.

### Esempio 8.4 — Integrale per fette

Sia $D=\{(x,y,z):x^2+y^2+z^2\le R^2,\ z\ge0\}$ (emisfero superiore). Fissato $z\in[0,R]$, la sezione orizzontale è
$$
D_z=\{(x,y):x^2+y^2\le R^2-z^2\},
$$
che è un cerchio di raggio $\sqrt{R^2-z^2}$. Allora
$$
\int_D f\,dV
=
\int_0^R\left(\int_{x^2+y^2\le R^2-z^2} f(x,y,z)\,dx\,dy\right)dz.
$$
Questo è il metodo "per fette": si integra su $z$ e poi sulle variabili $x,y$ nella sezione.

> **Osservazione 8.5.**  
> I metodi "per fili" e "per fette" non sono alternativi al cambiamento di variabili, ma complementari. In molti problemi conviene combinare: si sceglie la variabile lungo cui il dominio ha una struttura semplice (fette o fili), e per le sezioni si usano coordinate adatte (polari, ecc.). Questo è il modo in cui si risolvono, ad esempio, gli integrali tripli su domini a simmetria cilindrica o sferica.
