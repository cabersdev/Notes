# Massimi e Minimi su Insiemi Aperti

## Definizione - Forma quadratica associata ad $A$
Sia $A \in M^{n\times n}$ una matrice reale e simmetrica

Si dice **forma quadratica associata ad $A$** l'applicazione:

$$
\begin{aligned}
    \mathbb{R}^n &\longrightarrow \mathbb{R} \\
    x &\mapsto \langle Ax, x \rangle
\end{aligned}
$$

### Esempio

Sia $x = (x_1, \dots, x_n) \in \mathbb{R}^n$ e sia 

$$
A = 
\begin{pmatrix}
a_{11} & a_{12} & \dots & a_{1n} \\ 
a_{21} & a_{22} & \dots & a_{2n} \\ 
\vdots & \vdots & \ddots & \vdots \\
a_{n1} & a_{n2} & \dots & a_{nn} 
\end{pmatrix}
$$

Si ha che 

$$
Ax = 
\begin{pmatrix}
    \sum_{i = 1}^n a_{1i}x_i \\
    \sum_{i = 1}^n a_{2i}x_i \\
    \vdots \\
    \sum_{i = 1}^n a_{ni}x_i
\end{pmatrix}
$$

$$
\langle Ax, x \rangle = \sum_{j = 1}^n (a_{ji}x_ix_j) = a_{11}x_1^2 + a_{22}x_2^2 + \dots + a_{nn}x_n^2 + 2a_{12}x_1x_2 + \dots + 2 a_{n-1}x_{n-1}x_n
$$

**Definizione:**
$A \in M^{n\times n}$, reale e simmetrica
- **Definita positiva (Negativa):** se $\forall x \in \mathbb{R}^n$, $x \neq 0$ si ha $\langle Ax, x \rangle > 0$ ($< 0$)

- **Semidefinita positiva (Negativa):** se $\forall x \in \mathbb{R}^n$ si ha $\langle Ax, x\rangle \geq 0 (\leq 0)$

- **Indefinita:** Se $\exists x,y \in \mathbb{R}^n$, $x \neq 0$, $y \neq 0$ tale che $\langle Ax, x\rangle > 0$ e $\langle Ay, y \rangle < 0$

## Caratterizzazione delle forme quadratiche

1. **Caraterizzazione forme quadratiche definite positive (negative):**
$A \in M^{n \times n}$ reale e simmetrica $A$ è definita positiva (negative) se e solo se $\exists \alpha \in \mathbb{R}$, $\alpha > 0$ tale che $\langle Ax, x\rangle \geq \alpha ||x||^2 \; \forall x \in \mathbb{R}^n \; (\leq -\alpha ||x||^2)$

**Osservazione:** Se $A$ $n\times n$ è reale e simmetrica allora ha $n$ autovalori reali (contati con la molteplicità algebrica)

2. **Caratterizzazione della matrici reali e simmetriche con gli autovalori**
$A \in M^{n \times n}$ reale e simmetrica

- $A$ è definita positiva (negativa) se e solo se tutti gli autovalori sono strettamente positivi (negativi)

- $A$ è semidefinita positiva (negativa) se e solo tutti gli autovalori sono maggiori o uguali a $0$ (minori)

- $A$ è indefinit se e solo se $\exists \; \lambda_1, \lambda_2$ autovalori di $A$ tali che $\lambda_1 < 0$ e $\lambda_2 > 0$

### Definizione - Massimo e Minimo
Sia $f:\mathbb{R}^n \longrightarrow \mathbb{R}$ e sia $x_0 \in X$

- $x_0$ si dice di minimo (massimo) relativo, o locale per $f$ in $X$ se $\exists \mathcal{B}_r(x_0): \forall x \in \mathcal{B}_r(x_0) \cap X$ si ha $f(x) \geq f(x_0)$ ($\leq$)

- $x_0$ si dice di minimo (massimo) assoluto, o globale, per $f$ in $X$ se $\forall x \in X$ si ha $f(x) \geq f(x_0)$ ($\leq$)

- Se $f$ è derivabile in $x_0$, $x_0$ si dice punto critico per $f$ se $\nabla f(x_0) = 0$ 
Quindi $x_0$ è un punto critico per $f$ se $f$ è derivabile in $x_0$ se e solo se 
$$
\begin{cases}
\frac{\partial f}{\partial x_1}(x_0) = 0 \\ 
\vedots \\
\frac{\partial f}{\partial x_n}(x_0) = 0
\end{cases}
$$

- Se $f$ è derivabile in $x_0$ e $x_0$ è un punto critico di $f$ e $x_0$ non è né massimo né minimo relativo si dice che $x_0$ è un **punto di sella**

### Ricerca di massimi e minimi in Insiemi aperti
E' rilevante le caratteristiche topologiche nell'insieme nel quale si ricerca il massimo o minimo

### Teorema - Condizioni necessarie del $I$ ordine per i massimi e minimi locali (fermat)

Sia $f: X \subseteq \mathbb{R}^n \longrightarrow \mathbb{R}$, $X$ aperto $x_0 \in X$ se $f$ è derivabile in $x_0$ e $x_0$ è di minimo o di massimo relativo allora si ha che $\nabla f(x_0) = 0$ (cioè $x_0$ è un punto critico per $f$)

### Dimsotrazione in $n = 2$

Sia quindi 

$$
f(x,y): X \subseteq \mathbb{R}^2 \longrightarrow \mathbb{R}, \quad X \text{ aperto}
$$

$(x_0, y_0) \in X$ sia di minimo relativo per $f$

Per definizione di minimo relativo si ha quindi che $\exists \mathcal{B}_r(x_0): \forall x \in \mathcal{B}_r(x_0, y_0)$ si ha $f(x,y) \geq f(x_0, y_0)$ in particolare si ha che $f(x,y_0) \geq f(x_0, y_0) \; \forall (x,y_0) \in \mathcal{B}_r(x_0, y_0)$ (equivalentemente $\forall x \in (x_0 - r, x_0 + r)$)

Sia ora 

$$
g(x) = f(x, y_0)\; g: (x_0 - r, x_0 + r) \longrightarrow \mathbb{R}
$$

$g$ è derivabile in $x_0$ e $g'(x_0) = \frac{\partial f}{\partial x} (x_0, y_0)$ dove $x_0$ è un punto di minimo per $g$ 

infatti $g(x) = f(x, y_0) \geq f(x_0, y_0) = g(x_0)$ da cui per il teorema di Fermat $g'(x_0) = 0$ cioè $\frac{\partial f}{\partial x} (x_0, y_0) = 0$


Analogamente $f(x_0, y) \geq f(x_0, y_0)$ $\forall (x_0, y) \in \mathcal{B}_r(x_0, y_0)$ equivalentemente $\forall y \in (y_0- r, y_0 + r)$

$h(y) = f(x_0, y)$ $h(y) \geq h(y_0) \; \forall y \in (y_0 - r, y_0 + r)$ $h$ è derivabile in $y_0$ e $h'(y_0) = \frac{\partial f}{\partial y}(x_0, y_0)$ analogamente a prima applichiamo il teorema di fermat segue che 

$$
g'(y_0) = 0 = \frac{\partial f}{\partial y}(x_0, y_0)
$$

Si ha quindi che $\nabla_f (x_0)$ $\square$

Per cercare punti di massimo minimi relativi di una funzione derivabile bisogna inanzittutto risolvere il sistema $\nabla_f (x_0)$ cioè: 

$$
\begin{cases}
\frac{\partial f}{\partial x_1} (x_0) = 0 \\
\vdots \\
\frac{\partial f}{\partial x_n} (x_0) = 0
\end{cases}
$$

### Teorema - condizioni necessarie del $II$ ordine per massimi e i minimi locali

Sia $f: X \subseteq \mathbb{R}^n \longrightarrow \mathbb{R}$, $X$ aperto, $x_0 \in X$ e sia $f \in C^2(X)$

Se $x_0$ è di minimo locale per $f$ si ha che $H_f(x_0)$ è simidefinita positiva

Se $x_0$ è di massimo locale per $f$ si ha che $H_f(x_0)$ è semidefinita negativa

### Teorema - condizioni sufficienti del $II$ ordine per i massimi e i minimi locali

Sia $f: X \subseteq \mathbb{R}^n \longrightarrow \mathbb{R}$, $X$ aperto, $x_0 \in X$, con $f \in C^2(X)$. Supponiamo anche che $\nabla f(x_0) = 0$

1. Se $H_f(x_0)$ è definita positiva allora si ha che $x_0$ è di minimo

2. Se $H_f(x_0)$ è definita negativa allora si ha che $x_0$ è di massimo

3. Se $H_f(x_0)$ è indefinita allora si ha che $x_0$ è di sella

Se $\overline{x}$ è tale che $\nabla_f (\overline{x}) = 0$

se $H_f(x)$ è 
1. Definita positia $\overline{x}$ è mminimo relativo
2. Se $H_f(x_0)$ è definita negativa $\overline{x}$ è di massimo 
3. Se $H_f(x_0)$ è indefinita $\overline{x}$ è di sella 
4. se è semidifinita positiva $\overline{x}$ è non di massimo, o è minimo o è di sella
5. se è semidefinita negativi $\overline{x}$ è non di minimo, o è massimo o è di sella

### Esempio

Sia $f(x,y) = x^2 + y^4$

Cerchiamo punti di massimo e minimi

$$
\nabla_f(x,y) = 
\begin{pmatrix}
2x \\
4y^3
\end{pmatrix}
$$

Mettiamo a sistema 

$$
\begin{cases}
2x = 0\\
4y^3 = 0
\end{cases} \implies (0,0) \text{ è punto critico per } f
$$

$$
H_f(0,0) = 
\begin{pmatrix}
\frac{\partial f^2}{\partial x^2} & \frac{\partial f^2}{\partial x \partial y} \\
\frac{\partial f^2}{\partial x\partial y} & \frac{\partial f^2}{\partial y^2}
\end{pmatrix} = \begin{pmatrix} 2 & 0 \\ 
0 & 0\end{pmatrix}
$$

Da cui $\text{spec}(H_f(0,0)) = \{0, 2\}$

La matrice è semidefinita positiva non posso concludere nulla ma posso escludere che $(0,0)$ sia un punto di massimo
Tuttavia è facile dimostrare che il punto è di minimo

### Esempio

$f(x,y) = x^2 - y^4$ ha un unico punto critico in $(0,0)$ e 
$$
H_f(0,0) = 
\begin{pmatrix}
\frac{\partial f^2}{\partial x^2} & \frac{\partial f^2}{\partial x \partial y} \\
\frac{\partial f^2}{\partial x\partial y} & \frac{\partial f^2}{\partial y^2}
\end{pmatrix} = \begin{pmatrix} 2 & 0 \\ 
0 & 0\end{pmatrix}
$$

Ma $f(x,0) = x^2 \geq 0 = f(0,0)$ e $f(0,y) = -y^4 \leq 0$ si ha quindi che $(0,0)$ è un punto di sella
