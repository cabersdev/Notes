# Massimi e minimi vincolati

---

## 0. Impostazione del problema

Sia $X\subseteq\mathbb{R}^n$ aperto e sia
$$
f:X\subseteq\mathbb{R}^n\longrightarrow\mathbb{R}
$$
una funzione reale. Sia $A\subseteq X$ un insieme ammissibile definito da vincoli.

### Definizione 0.1 — Insieme ammissibile con vincoli di disuguaglianza

Siano $g_1,\dots,g_m\in C^1(\mathbb{R}^n)$. Si dice **insieme ammissibile** con vincoli di disuguaglianza l'insieme
$$
A=\{x\in\mathbb{R}^n : g_i(x)\le 0,\ i=1,\dots,m\}.
$$
Se i vincoli sono di uguaglianza, cioè
$$
M=\{x\in\mathbb{R}^n : h_j(x)=0,\ j=1,\dots,k\}
$$
con $h_j\in C^1(\mathbb{R}^n)$, si parla di **vincolo di uguaglianza** e $M$ è detto **varietà vincolare**.

> **Osservazione 0.2.** Un vincolo di uguaglianza $h_j(x)=0$ è equivalente alla coppia di disuguaglianze $h_j(x)\le 0$ e $-h_j(x)\le 0$. Quindi i due formalismi sono sostanzialmente equivalenti, ma il trattamento analitico è diverso: i vincoli di uguaglianza definiscono una varietà di dimensione $n-k$, mentre i vincoli di disuguaglianza definiscono un insieme con interno non vuoto.

### Esempi 0.3

$$
A_1=\{(x,y)\in\mathbb{R}^2 : x+y\ge 0,\ x^2+y^2\le 1,\ 2x-y\le 0\}.
$$
$$
A_2=\{(x,y)\in\mathbb{R}^2 : x-y+2\le 0,\ x\ge -1,\ 2x+y+2\ge 0,\ 3x+\tfrac{y}{2}+5\ge 0,\ y\le 2\}.
$$
$$
A_3=\{(x,y,z)\in\mathbb{R}^3 : x^2+y^2+z^2\le 1,\ x\ge 0,\ y\ge 0\}.
$$

Tutti questi insiemi sono della forma $\{x: g_i(x)\le 0\}$ con $g_i$ polinomi, quindi di classe $C^1$.

### 0.4 Il problema di ottimizzazione

Il problema che si affronta è
$$
\min_{x\in A} f(x) \qquad\text{oppure}\qquad \max_{x\in A} f(x),
$$
dove $A$ è definito da vincoli. Si parla di **massimi e minimi vincolati**.

> **Osservazione 0.5.**  
> Su un insieme aperto, gli estremi locali sono necessariamente punti critici (Teorema di Fermat). Su un insieme con frontiera, gli estremi possono trovarsi:
> - nell'**interno** $A^\circ$, dove vale il teorema di Fermat;
> - sulla **frontiera** $\partial A$, dove il gradiente di $f$ può essere non nullo e la condizione necessaria è diversa (moltiplicatori di Lagrange).
>
> Dunque il problema si decompone in due sotto-problemi: analisi interna e analisi al bordo.

## 1. Topologia degli insiemi definiti da vincoli

### 1.1 Interno e frontiera

Sia
$$
A=\{x\in\mathbb{R}^n : g_i(x)\le 0,\ i=1,\dots,m\},
\qquad g_i\in C^1(\mathbb{R}^n).
$$
Si indica con $A^\circ$ l'interno di $A$ e con $\partial A$ la sua frontiera.

- L'**interno** è
  $$
  A^\circ\supseteq\{x\in\mathbb{R}^n : g_i(x)<0\ \forall i\}.
  $$
  In generale vale solo l'inclusione, perché l'interno può contenere punti in cui alcuni vincoli sono attivi se essi non “bloccano” localmente l'insieme. Per esempio, se $g_1(x)=x^2$, l'insieme $\{x^2\le 0\}=\{0\}$ ha interno vuoto in $\mathbb{R}$; oppure in $\mathbb{R}^2$, il vincolo $(x-1)^2\le 0$ è attivo solo in $x=1$ e non genera un insieme con interno.
- La **frontiera** è
  $$
  \partial A
  =
  \overline{A}\setminus A^\circ.
  $$
  Un punto $x_0\in\partial A$ si dice **punto di frontiera**. In un punto di frontiera almeno un vincolo è attivo, cioè $g_i(x_0)=0$ per qualche $i$.

### 1.2 Decomposizione della frontiera

Sia $I(x)=\{i : g_i(x)=0\}$ l'insieme degli **indici attivi** in $x$. Allora
$$
\partial A\subseteq\bigcup_{\emptyset\neq J\subseteq\{1,\dots,m\}}\{x : g_i(x)=0\ \forall i\in J,\ g_j(x)<0\ \forall j\notin J\}.
$$
Ogni insieme della forma
$$
M_J=\{x : g_i(x)=0\ \forall i\in J,\ g_j(x)<0\ \forall j\notin J\}
$$
è un “pezzo” della frontiera di dimensione $n-|J|$ (se i vincoli sono regolari). In pratica, la frontiera si decompone in:

- **facce** di dimensione $n-1$, dove un solo vincolo è attivo;
- **spigoli** di dimensione $n-2$, dove due vincoli sono attivi;
- e così via, fino ai **vertici** di dimensione $0$, dove $n$ vincoli indipendenti sono attivi.

**Esempio 1.1.** Per
$$
A_3=\{(x,y,z)\in\mathbb{R}^3 : x^2+y^2+z^2\le 1,\ x\ge 0,\ y\ge 0\},
$$
la frontiera è unione di:

- la calotta sferica $\{x^2+y^2+z^2=1,\ x>0,\ y>0\}$;
- il quarto di disco $\{x^2+y^2+z^2<1,\ x=0,\ y>0\}$;
- il quarto di disco $\{x^2+y^2+z^2<1,\ x>0,\ y=0\}$;
- l'arco $\{x^2+y^2+z^2=1,\ x=0,\ y>0\}$;
- l'arco $\{x^2+y^2+z^2=1,\ x>0,\ y=0\}$;
- il segmento $\{x^2+y^2+z^2<1,\ x=0,\ y=0\}$;
- il punto $\{x^2+y^2+z^2=1,\ x=0,\ y=0\}$.

In generale, ogni pezzo della frontiera è un insieme definito da $|J|$ vincoli di uguaglianza e $m-|J|$ vincoli di disuguaglianza stretta.

> **Osservazione 1.2.**  
> I punti in cui $n$ vincoli indipendenti sono attivi (vertici) vanno trattati separatamente: lì la frontiera si riduce a un punto e basta valutare $f$.

## 2. Estremi locali e globali su insiemi vincolati

### Definizione 2.1 — Estremi locali e globali su $A$

Sia $f:X\subseteq\mathbb{R}^n\to\mathbb{R}$, $A\subseteq X$, $x_0\in A$.

- $x_0$ è **punto di minimo locale** per $f$ su $A$ se esiste $r>0$ tale che
  $$
  f(x)\ge f(x_0)\qquad \forall x\in\mathcal{B}_r(x_0)\cap A.
  $$
- $x_0$ è **punto di massimo locale** per $f$ su $A$ se esiste $r>0$ tale che
  $$
  f(x)\le f(x_0)\qquad \forall x\in\mathcal{B}_r(x_0)\cap A.
  $$
- $x_0$ è **punto di minimo globale** per $f$ su $A$ se
  $$
  f(x)\ge f(x_0)\qquad \forall x\in A.
  $$
- $x_0$ è **punto di massimo globale** per $f$ su $A$ se
  $$
  f(x)\le f(x_0)\qquad \forall x\in A.
  $$

Se le disuguaglianze valgono in senso stretto per $x\neq x_0$, si parla di estremo **stretto**.

### Definizione 2.2 — Punto critico vincolato

Sia $f:X\to\mathbb{R}$ differenziabile, $A\subseteq X$, $x_0\in A$. Si dice che $x_0$ è un **punto critico vincolato** per $f$ su $A$ se $x_0$ è un estremo locale di $f$ su $A$ oppure se soddisfa le condizioni necessarie del primo ordine (teorema di Fermat nell'interno, moltiplicatori di Lagrange sulla frontiera).

### 2.3 Esistenza degli estremi: teorema di Weierstrass

> **Teorema 2.3 (Weierstrass).**  
> Se $A\subseteq\mathbb{R}^n$ è **compatto** (chiuso e limitato) e $f:A\to\mathbb{R}$ è continua, allora $f$ ammette massimo e minimo su $A$.

**Dimostrazione.**  
$A$ è compatto e $f$ è continua, quindi $f(A)$ è compatto in $\mathbb{R}$, cioè chiuso e limitato. Dunque $\inf f(A)\in f(A)$ e $\sup f(A)\in f(A)$, cioè esistono $x_m,x_M\in A$ con $f(x_m)=\min_A f$ e $f(x_M)=\max_A f$. $\square$

> **Osservazione 2.4.**  
> Gli insiemi del tipo $A=\{x : g_i(x)\le 0\}$ con $g_i$ continue sono **chiusi** in $\mathbb{R}^n$. Se inoltre $A$ è limitato, allora è compatto e Weierstrass garantisce l'esistenza di massimo e minimo. In molti problemi applicativi questa verifica è il primo passo.

**Esempio 2.5.**  
$$
A=\{(x,y)\in\mathbb{R}^2 : x^2+y^2\le 1\}
$$
è chiuso e limitato, quindi compatto. Per ogni $f$ continua, massimo e minimo su $A$ esistono.

**Esempio 2.6.**  
$$
A=\{(x,y)\in\mathbb{R}^2 : x\ge -1,\ y\le -x+1,\ y\ge x-1\}
$$
è chiuso (intersezione di tre chiusi) e limitato, perché
$$
-2\le x-1\le y\le -x+1\le 2
\quad\Longrightarrow\quad
-2\le y\le 2,\ -1\le x\le 3.
$$
Quindi $A$ è compatto.

## 3. Analisi dell'interno: teorema di Fermat

### Teorema 3.1 — Fermat per estremi interni

Sia $f:X\subseteq\mathbb{R}^n\to\mathbb{R}$ differenziabile, $A\subseteq X$, e sia $x_0\in A^\circ$ un punto di massimo o minimo locale per $f$ su $A$. Allora
$$
\nabla f(x_0)=0.
$$

**Dimostrazione.**  
Poiché $x_0\in A^\circ$, esiste $r>0$ con $\mathcal{B}_r(x_0)\subseteq A$. In tale palla, $x_0$ è anche estremo locale di $f$ su $\mathcal{B}_r(x_0)$, che è un aperto. Per il teorema di Fermat su insiemi aperti, $\nabla f(x_0)=0$. $\square$

> **Osservazione 3.2.**  
> Il teorema di Fermat si applica solo ai punti **interni** di $A$. Sui punti di frontiera il gradiente può essere non nullo e la condizione necessaria è diversa.

### 3.3 Procedura per l'interno

Per cercare estremi locali interni si risolve il sistema
$$
\begin{cases}
\dfrac{\partial f}{\partial x_1}(x)=0,\\
\vdots\\
\dfrac{\partial f}{\partial x_n}(x)=0,
\end{cases}
$$
e si selezionano le soluzioni che appartengono ad $A^\circ$. Tali soluzioni sono i candidati estremi interni.

## 4. Analisi della frontiera: il problema generale

### 4.1 Perché serve una teoria specifica

Sulla frontiera, il gradiente di $f$ può essere non nullo, perché la direzione di massima crescita può essere bloccata dal vincolo. La condizione necessaria non è $\nabla f(x_0)=0$, ma una condizione più debole che coinvolge i gradienti dei vincoli attivi.

### 4.2 Decomposizione in pezzi

Si decompone $\partial A$ in pezzi $M_J$ corrispondenti ai sottoinsiemi $J$ di indici attivi:
$$
M_J=\{x\in A : g_i(x)=0\ \forall i\in J,\ g_j(x)<0\ \forall j\notin J\}.
$$
Su ciascun pezzo $M_J$, il problema diventa un problema di estremi **vincolati da uguaglianze**:
$$
\min_{x\in M_J} f(x)
\qquad\text{con vincoli } g_i(x)=0\ \forall i\in J.
$$
La dimensione di $M_J$ è $n-|J|$ se i vincoli sono indipendenti. Per $|J|=1$ si ha una “faccia” di dimensione $n-1$; per $|J|=n$ si ha un vertice (dimensione $0$), che si tratta semplicemente valutando $f$ nei punti corrispondenti.

### 4.3 Procedura operativa

1. **Interno.** Trovare i punti critici $\nabla f=0$ in $A^\circ$.
2. **Facce.** Per ogni sottoinsieme $J$ di indici attivi con $|J|<n$, risolvere il problema vincolato su $M_J$, usando il teorema delle funzioni implicite o i moltiplicatori di Lagrange.
3. **Vertici.** Per ogni sottoinsieme $J$ con $|J|=n$, se i vincoli sono indipendenti, la soluzione è univoca: il punto si trova risolvendo il sistema $\{g_i(x)=0\}_{i\in J}$ e si valuta $f$.
4. **Confronto.** Confrontare i valori di $f$ in tutti i candidati (punti critici interni, punti critici sulle facce, vertici) e determinare massimo e minimo globale.

> **Osservazione 4.4.**  
> Il punto delicato è garantire che la soluzione del problema vincolato esista e sia regolare. Il teorema delle funzioni implicite e la nozione di vincolo regolare sono gli strumenti che permettono di applicare i moltiplicatori di Lagrange.

## 5. Teorema delle funzioni implicite

### 5.1 Versione in $\mathbb{R}^2$

> **Teorema 5.1 (Dini, $\mathbb{R}^2$).**  
> Sia $F:U\subseteq\mathbb{R}^2\to\mathbb{R}$ di classe $C^1$ su un aperto $U$, e sia $(x_0,y_0)\in U$ con
> $$
> F(x_0,y_0)=0,
> \qquad
> \frac{\partial F}{\partial y}(x_0,y_0)\neq 0.
> $$
> Allora esistono un intorno $I$ di $x_0$ in $\mathbb{R}$ e un'unica funzione $\varphi:I\to\mathbb{R}$ di classe $C^1$ tale che
> $$
> \varphi(x_0)=y_0,
> \qquad
> F(x,\varphi(x))=0\quad\forall x\in I.
> $$
> Inoltre
> $$
> \varphi'(x)
> =
> -\frac{\dfrac{\partial F}{\partial x}(x,\varphi(x))}{\dfrac{\partial F}{\partial y}(x,\varphi(x))}
> \qquad\forall x\in I.
> $$

**Idea della dimostrazione.**  
Si applica il teorema delle contrazioni alla mappa $y\mapsto y-\frac{F(x,y)}{F_y(x_0,y_0)}$ in un intorno sufficientemente piccolo di $(x_0,y_0)$. La formula per $\varphi'$ segue derivando rispetto a $x$ l'identità $F(x,\varphi(x))=0$ e applicando la regola della catena:
$$
F_x(x,\varphi(x))+F_y(x,\varphi(x))\varphi'(x)=0. \qquad\square
$$

> **Osservazione 5.2.**  
> La condizione $\partial F/\partial y(x_0,y_0)\neq 0$ è essenziale. Se si annulla, il teorema non si applica: la curva di livello può avere un punto singolare (per esempio una cuspide o un nodo).

### 5.2 Versione in $\mathbb{R}^n$

> **Teorema 5.3 (Dini, $\mathbb{R}^n$).**  
> Sia $F:U\subseteq\mathbb{R}^n\to\mathbb{R}$ di classe $C^1$ su un aperto $U$, e sia $x_0\in U$ con
> $$
> F(x_0)=0,
> \qquad
> \nabla F(x_0)\neq 0.
> $$
> Allora, a meno di riordinare le variabili, esistono un intorno $V$ di $(x_{0,1},\dots,x_{0,n-1})$ in $\mathbb{R}^{n-1}$ e un'unica funzione $\varphi:V\to\mathbb{R}$ di classe $C^1$ tale che
> $$
> \varphi(x_{0,1},\dots,x_{0,n-1})=x_{0,n},
> \qquad
> F(x_1,\dots,x_{n-1},\varphi(x_1,\dots,x_{n-1}))=0.
> $$

In altre parole, l'ipersuperficie di livello $\{F=0\}$ è localmente il grafico di una funzione $C^1$ di $n-1$ variabili.

### 5.3 Versione per sistemi di vincoli

> **Teorema 5.4 (Dini per sistemi).**  
> Siano $h_1,\dots,h_k:U\subseteq\mathbb{R}^n\to\mathbb{R}$ di classe $C^1$ su un aperto $U$, e sia $x_0\in U$ con
> $$
> h_j(x_0)=0\quad\forall j=1,\dots,k,
> $$
> e tale che i gradienti $\nabla h_1(x_0),\dots,\nabla h_k(x_0)$ siano **linearmente indipendenti**. Allora esistono un intorno $V$ di un punto di $\mathbb{R}^{n-k}$ e un'unica funzione $\varphi:V\to\mathbb{R}^k$ di classe $C^1$ tale che, a meno di riordinare le variabili,
> $$
> M\cap U=\{x\in U : h_j(x)=0\ \forall j\}
> $$
> è localmente il grafico di $\varphi$. In particolare, $M$ è una **varietà di dimensione $n-k$** in un intorno di $x_0$.

> **Osservazione 5.5.**  
> L'indipendenza lineare dei gradienti $\nabla h_j(x_0)$ è la condizione di regolarità del vincolo. Se viene a mancare, la varietà $M$ può avere una singolarità in $x_0$ e il teorema non si applica.

## 6. Vincoli regolari

### Definizione 6.1 — Vincolo regolare

Sia $g:\mathbb{R}^n\to\mathbb{R}$ di classe $C^1$, e sia $x_0\in\mathbb{R}^n$ con $g(x_0)=0$. Si dice che il vincolo $g(x)=0$ è **regolare** in $x_0$ se
$$
\nabla g(x_0)\neq 0.
$$

Siano $g_1,\dots,g_k:\mathbb{R}^n\to\mathbb{R}$ di classe $C^1$ e sia $x_0$ con $g_i(x_0)=0$ per ogni $i$. Si dice che il sistema di vincoli è **regolare** in $x_0$ se i gradienti $\nabla g_1(x_0),\dots,\nabla g_k(x_0)$ sono **linearmente indipendenti**, cioè se la matrice Jacobiana
$$
J_g(x_0)=
\begin{pmatrix}
\dfrac{\partial g_1}{\partial x_1}(x_0)&\cdots&\dfrac{\partial g_1}{\partial x_n}(x_0)\\
\vdots&\ddots&\vdots\\
\dfrac{\partial g_k}{\partial x_1}(x_0)&\cdots&\dfrac{\partial g_k}{\partial x_n}(x_0)
\end{pmatrix}
\in M^{k\times n}(\mathbb{R})
$$
ha rango $k$.

> **Osservazione 6.2.**  
> Se il sistema di vincoli è regolare in $x_0$, allora per il Teorema 5.4 la varietà $M$ è localmente un grafico $C^1$ di dimensione $n-k$ e ammette spazio tangente ben definito in $x_0$.

### Definizione 6.3 — Spazio tangente e spazio normale

Sia $M=\{x : g_j(x)=0,\ j=1,\dots,k\}$ una varietà regolare in $x_0$. Lo **spazio tangente** a $M$ in $x_0$ è
$$
T_{x_0}M
=
\{v\in\mathbb{R}^n : \langle\nabla g_j(x_0),v\rangle=0\ \forall j=1,\dots,k\}.
$$
Lo **spazio normale** a $M$ in $x_0$ è
$$
N_{x_0}M
=
\operatorname{span}\{\nabla g_1(x_0),\dots,\nabla g_k(x_0)\}.
$$
Per il teorema di rappresentazione, $\mathbb{R}^n=T_{x_0}M\oplus N_{x_0}M$, e
$$
T_{x_0}M=(N_{x_0}M)^\perp.
$$

> **Osservazione 6.4.**  
> Per $k=1$ si ha $N_{x_0}M=\operatorname{span}\{\nabla g(x_0)\}$, e $T_{x_0}M$ è l'iperpiano ortogonale a $\nabla g(x_0)$. In $\mathbb{R}^2$, la curva di livello $g=0$ ha tangente ortogonale a $\nabla g(x_0)$.

### Esempio 6.5 — Punto non regolare

Sia $g(x,y)=x^3-y^2$ e $M=\{g=0\}$. In $(0,0)$ si ha $\nabla g(0,0)=(0,0)$, quindi il vincolo non è regolare. In effetti $M$ è una cuspide con vertice in $(0,0)$: non è localmente un grafico $C^1$ di una funzione, e la nozione di spazio tangente in $(0,0)$ non è ben definita.

> **Osservazione 6.6.**  
> In presenza di punti non regolari, i moltiplicatori di Lagrange possono non esistere, come mostra l'Esempio 9.4. In questi casi bisogna studiare il problema direttamente.

## 7. Condizioni necessarie del primo ordine per vincoli di uguaglianza

### Teorema 7.1 — Condizione necessaria per un solo vincolo

Sia $f:U\subseteq\mathbb{R}^n\to\mathbb{R}$ di classe $C^1$ su un aperto $U$, e sia $g:U\to\mathbb{R}$ di classe $C^1$. Sia
$$
M=\{x\in U : g(x)=0\}.
$$
Supponiamo che $x_0\in M$ sia un punto di massimo o minimo locale per $f$ su $M$ e che il vincolo sia regolare in $x_0$, cioè $\nabla g(x_0)\neq 0$. Allora esiste $\lambda\in\mathbb{R}$ tale che
$$
\nabla f(x_0)=\lambda\nabla g(x_0).
$$

**Dimostrazione.**  
Per il teorema delle funzioni implicite, esiste un intorno $V$ di $x_0$ in cui $M$ è il grafico di una funzione $C^1$ $\varphi$ di $n-1$ variabili. Sia $\Phi:W\to M\cap V$, $\Phi(u)=(u,\varphi(u))$, una parametrizzazione locale di $M$ con $\Phi(u_0)=x_0$. Consideriamo la funzione composta
$$
\tilde f=f\circ\Phi:W\subseteq\mathbb{R}^{n-1}\to\mathbb{R}.
$$
Poiché $x_0$ è estremo locale di $f$ su $M$, $u_0$ è estremo locale di $\tilde f$ su $W$. Per il teorema di Fermat in $\mathbb{R}^{n-1}$,
$$
\nabla\tilde f(u_0)=0.
$$
Per la regola della catena,
$$
\nabla\tilde f(u_0)=J_\Phi(u_0)^T\nabla f(x_0)=0,
$$
cioè
$$
\nabla f(x_0)\perp\operatorname{Im}J_\Phi(u_0).
$$
Ora $\operatorname{Im}J_\Phi(u_0)=T_{x_0}M$ è lo spazio tangente a $M$ in $x_0$, che coincide con $(\nabla g(x_0))^\perp$. Dunque $\nabla f(x_0)$ è ortogonale a $(\nabla g(x_0))^\perp$, cioè
$$
\nabla f(x_0)\in\operatorname{span}\{\nabla g(x_0)\}.
$$
Esiste quindi $\lambda\in\mathbb{R}$ con $\nabla f(x_0)=\lambda\nabla g(x_0)$. $\square$

### Teorema 7.2 — Condizioni necessarie per più vincoli

Siano $f,g_1,\dots,g_k:U\subseteq\mathbb{R}^n\to\mathbb{R}$ di classe $C^1$, e sia
$$
M=\{x\in U : g_j(x)=0,\ j=1,\dots,k\}.
$$
Sia $x_0\in M$ un punto di massimo o minimo locale per $f$ su $M$, e supponiamo che il sistema di vincoli sia regolare in $x_0$, cioè che $\nabla g_1(x_0),\dots,\nabla g_k(x_0)$ siano linearmente indipendenti. Allora esistono $\lambda_1,\dots,\lambda_k\in\mathbb{R}$ tali che
$$
\nabla f(x_0)=\sum_{j=1}^k\lambda_j\nabla g_j(x_0).
$$

**Dimostrazione.**  
Si applica il teorema di Dini per sistemi: localmente $M$ è il grafico di una funzione $\varphi:W\subseteq\mathbb{R}^{n-k}\to\mathbb{R}^k$, e si considera la composta $\tilde f=f\circ\Phi$ con $\Phi(u)=(u,\varphi(u))$. Il teorema di Fermat in $\mathbb{R}^{n-k}$ dà
$$
J_\Phi(u_0)^T\nabla f(x_0)=0.
$$
Poiché $\operatorname{Im}J_\Phi(u_0)=T_{x_0}M=(N_{x_0}M)^\perp$ e $N_{x_0}M=\operatorname{span}\{\nabla g_1(x_0),\dots,\nabla g_k(x_0)\}$, si ottiene
$$
\nabla f(x_0)\in N_{x_0}M,
$$
cioè la tesi. $\square$

> **Osservazione 7.3.**  
> Il teorema è solo **necessario**: un punto che soddisfa $\nabla f(x_0)=\sum_j\lambda_j\nabla g_j(x_0)$ non è necessariamente un estremo. Si parla di **punto critico vincolato**.

## 8. Interpretazione geometrica

### 8.1 Ortogonalità tra gradiente e vincolo

Sia $M=\{g=0\}$ una varietà regolare. In un punto critico vincolato $x_0$, il gradiente $\nabla f(x_0)$ è ortogonale a $T_{x_0}M$, cioè è **ortogonale al vincolo**.

**Conseguenza.**  
La direzione di massima crescita di $f$ è ortogonale alla varietà vincolare. Se $\nabla f(x_0)$ avesse una componente tangente non nulla, spostandosi lungo $M$ nella direzione di tale componente si aumenterebbe (o diminuirebbe) $f$, e $x_0$ non sarebbe un estremo.

### 8.2 Curve di livello tangenti

In $\mathbb{R}^2$, sia $M$ una curva di livello $g=0$. Le curve di livello $f=c$ sono ortogonali a $\nabla f$. Se $\nabla f(x_0)=\lambda\nabla g(x_0)$, le due curve di livello $f=c$ e $g=0$ hanno la **stessa tangente** in $x_0$, cioè sono **tangenti**. Dunque un punto critico vincolato corrisponde a una curva di livello di $f$ tangente al vincolo.

### 8.3 Significato del moltiplicatore

Il moltiplicatore $\lambda$ misura la **sensibilità** del valore ottimo rispetto al vincolo. Se si considera il problema perturbato
$$
\min_{g(x)=\varepsilon} f(x),
$$
con valore ottimo $V(\varepsilon)$, si ha
$$
V'(0)=-\lambda
$$
(sotto opportune ipotesi di regolarità). Dunque $\lambda$ rappresenta il **prezzo ombra** del vincolo: di quanto migliora il valore ottimo se si allenta il vincolo di un'unità.

## 9. Teorema dei moltiplicatori di Lagrange

### 9.1 Funzione Lagrangiana

Sia $f:U\subseteq\mathbb{R}^n\to\mathbb{R}$ di classe $C^1$, e siano $g_1,\dots,g_k:U\to\mathbb{R}$ di classe $C^1$. Si definisce **funzione Lagrangiana**
$$
\mathcal{L}:\mathbb{R}^n\times\mathbb{R}^k\to\mathbb{R},
\qquad
\mathcal{L}(x,\lambda)
=
f(x)-\sum_{j=1}^k\lambda_j g_j(x).
$$
Le condizioni del primo ordine per un estremo vincolato si scrivono
$$
\nabla_x\mathcal{L}(x,\lambda)=0,
\qquad
\nabla_\lambda\mathcal{L}(x,\lambda)=0.
$$
La prima equazione è
$$
\nabla f(x)-\sum_{j=1}^k\lambda_j\nabla g_j(x)=0,
$$
la seconda è il vincolo $g_j(x)=0$ per ogni $j$. Dunque i punti critici vincolati sono i **punti stazionari** della Lagrangiana.

### Teorema 9.2 — Moltiplicatori di Lagrange

Siano $f,g_1,\dots,g_k:U\subseteq\mathbb{R}^n\to\mathbb{R}$ di classe $C^1$ su un aperto $U$. Sia
$$
M=\{x\in U : g_j(x)=0,\ j=1,\dots,k\}.
$$
Sia $x_0\in M$ un punto di massimo o minimo locale per $f$ su $M$, e supponiamo che il sistema di vincoli sia regolare in $x_0$. Allora esistono $\lambda_1,\dots,\lambda_k\in\mathbb{R}$ tali che
$$
\nabla f(x_0)=\sum_{j=1}^k\lambda_j\nabla g_j(x_0).
$$

La dimostrazione è quella del Teorema 7.2. I numeri $\lambda_j$ sono detti **moltiplicatori di Lagrange**.

### 9.3 Procedura operativa

Per trovare i punti critici vincolati di $f$ su $M$:

1. Scrivere la Lagrangiana
   $$
   \mathcal{L}(x,\lambda)=f(x)-\sum_{j=1}^k\lambda_j g_j(x).
   $$
2. Risolvere il sistema
   $$
   \begin{cases}
   \dfrac{\partial\mathcal{L}}{\partial x_1}=0,\\
   \vdots\\
   \dfrac{\partial\mathcal{L}}{\partial x_n}=0,\\
   g_1(x)=0,\\
   \vdots\\
   g_k(x)=0.
   \end{cases}
   $$
   Le soluzioni sono i punti critici vincolati.
3. Verificare la regolarità del vincolo nei punti trovati. Se il vincolo non è regolare, il punto può essere un estremo anche se non soddisfa le condizioni di Lagrange (vedi Esempio 9.4).
4. Confrontare i valori di $f$ nei punti critici vincolati, insieme ai punti non regolari e agli eventuali punti di frontiera, per determinare massimo e minimo globali.

> **Osservazione 9.3.**  
> Il sistema di Lagrange ha $n+k$ equazioni in $n+k$ incognite (le $x_i$ e i $\lambda_j$). In generale è non lineare e va risolto caso per caso. Spesso è utile ricavare i $\lambda_j$ da alcune equazioni e sostituirli nelle altre.

### Esempio 9.4 — Punto non regolare

Sia $f(x,y)=x$ e $g(x,y)=x^3-y^2$. Si vuole minimizzare $f$ su $M=\{g=0\}$. Il punto $(0,0)\in M$ è l'unico punto di minimo di $f$ su $M$ (la cuspide ha l'ascissa minima $x=0$). Tuttavia
$$
\nabla f(0,0)=(1,0),
\qquad
\nabla g(0,0)=(0,0),
$$
e non esiste $\lambda$ con $\nabla f(0,0)=\lambda\nabla g(0,0)$. Il punto non è regolare e il teorema di Lagrange non si applica. Questo mostra l'importanza dell'ipotesi di regolarità.
