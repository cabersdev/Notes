# Problemi di Programmazione Lineare

---

## 1. Programmazione Lineare

### Definizione 1.1 — Problema di Programmazione Lineare

Un **problema di Programmazione Lineare (PL)** è un problema di ottimizzazione vincolata della forma

$$
\begin{aligned}
\max \; (\text{o} \min) \quad & c_1 x_1 + \dots + c_n x_n \\
\text{soggetto a} \quad & a_{11}x_1 + \dots + a_{1n}x_n \sim b_1 \\
& \vdots \\
& a_{m1}x_1 + \dots + a_{mn}x_n \sim b_m
\end{aligned}
$$

dove:

- $x_1,\dots,x_n$ sono le **variabili decisionali** (le incognite);
- $c_1,\dots,c_n \in \mathbb{R}$ sono i **coefficienti di costo**;
- $a_{ij} \in \mathbb{R}$ sono i **coefficienti tecnologici**;
- $b_1,\dots,b_m \in \mathbb{R}$ sono i **termini noti**;
- ciascun simbolo $\sim$ sta per uno degli operatori $\geq$, $\leq$ o $=$.

La funzione

$$
f(x) = c_1 x_1 + \dots + c_n x_n
$$

è detta **funzione obiettivo**, mentre il sistema $A x \sim b$ rappresenta i **vincoli** del problema.

> **Osservazione 1.2.**  
> Non sono ammesse disequazioni strette ($<$ o $>$). I vincoli sono sempre non stretti.

### Definizione 1.3 — Forma matriciale

In forma matriciale, un problema di PL si scrive

$$
\begin{aligned}
\max \quad & c^T x \\
\text{soggetto a} \quad & A x \sim b
\end{aligned}
$$

dove $A \in \mathbb{R}^{m \times n}$, $c \in \mathbb{R}^n$, $b \in \mathbb{R}^m$, $x \in \mathbb{R}^n$.

### Definizione 1.4 — Soluzione ammissibile

Un vettore $x \in \mathbb{R}^n$ è detto **soluzione ammissibile** del problema di PL se soddisfa tutti i vincoli. Il corrispondente valore $c^T x$ è detto **valore** o **costo** della soluzione ammissibile $x$.

### Definizione 1.5 — Regione ammissibile

L'insieme di tutte le soluzioni ammissibili di un programma lineare è detto **regione ammissibile** e si indica con

$$
\mathcal{F} = \{x \in \mathbb{R}^n : A x \sim b\}.
$$

### Definizione 1.6 — Soluzione ottima

Un vettore $\bar{x} \in \mathbb{R}^n$ è una **soluzione ottima** del programma lineare se:

1. $\bar{x}$ è ammissibile, cioè $\bar{x} \in \mathcal{F}$;
2. $c^T \bar{x} \geq c^T x$ per ogni $x \in \mathcal{F}$ (se il problema è di massimizzazione);
3. $c^T \bar{x} \leq c^T x$ per ogni $x \in \mathcal{F}$ (se il problema è di minimizzazione).

Il corrispondente valore $c^T \bar{x}$ è detto **valore ottimo** del problema.

## 2. Teorema fondamentale della Programmazione Lineare

### Teorema 2.1 — Fondamentale della programmazione lineare

Dato un programma lineare, vale una e una sola delle seguenti alternative:

1. Il problema ha almeno una soluzione ottima;
2. Il problema è **inammissibile**, cioè non ha soluzioni ammissibili ($\mathcal{F} = \emptyset$);
3. Il problema è **illimitato**, cioè per ogni $\alpha \in \mathbb{R}$ esiste una soluzione ammissibile $x$ tale che $c^T x \geq \alpha$ (o $c^T x \leq \alpha$ se il problema è di minimizzazione).

> **Osservazione 2.2.**  
> Il teorema afferma che in programmazione lineare non è possibile che la funzione obiettivo abbia un estremo superiore finito che non sia anche un massimo. In altre parole, non esistono problemi di PL con valore ottimo finito ma non raggiunto.

> **Osservazione 2.3.**  
> Alcuni testi enunciano un **Teorema fondamentale della Programmazione Lineare** più forte, in cui si precisa che se si verifica il caso (1), allora esiste una soluzione ottima con particolari proprietà geometriche: se la regione ammissibile è un poliedro, esiste una soluzione ottima in un vertice (punto estremo) del poliedro.

## 3. Problemi in forma standard

### Definizione 3.1 — Forma standard

Un programma lineare è detto **in forma standard** se è scritto come

$$
\begin{aligned}
\max \quad & c^T x \\
\text{soggetto a} \quad & A x = b \\
& x \geq 0
\end{aligned}
$$

dove $A \in \mathbb{R}^{m \times n}$, $c \in \mathbb{R}^n$, $b \in \mathbb{R}^m$, $x \in \mathbb{R}^n$.

> **Osservazione 3.2.**  
> La scrittura $A x = b$ è un modo compatto di indicare il sistema di $m$ equazioni lineari
> $$
> a_i^T x = b_i \qquad i = 1,\dots,m,
> $$
> cioè $a_{i1}x_1 + \dots + a_{in}x_n = b_i$ per $i=1,\dots,m$.

> **Osservazione 3.3.**  
> Nella forma standard tutti i vincoli (esclusa la non-negatività) sono **equazioni** e tutte le variabili sono **non-negative**.

## 4. Equivalenza tra programmi lineari

### Definizione 4.1 — Equivalenza

Due programmi lineari $P_1$ e $P_2$ sono detti **equivalenti** quando vale la seguente proprietà:

per ogni $\alpha \in \mathbb{R}$, $P_1$ ha una soluzione ammissibile $x$ di costo $\alpha$ se e solo se $P_2$ ha una soluzione ammissibile $y$ di costo $\alpha$.

Inoltre, si richiede che data una tale soluzione $x$, la corrispondente soluzione $y$ possa essere ricavata tramite una **regola esplicita** (e viceversa).

### Proposizione 4.2 — Conseguenze dell'equivalenza

Se $P_1$ e $P_2$ sono programmi lineari equivalenti, allora valgono le seguenti proprietà:

1. $P_1$ ha una soluzione ammissibile se e solo se $P_2$ ha una soluzione ammissibile;
2. per ogni $\alpha \in \mathbb{R}$, $P_1$ ha una soluzione ottima di costo $\alpha$ se e solo se $P_2$ ha una soluzione ottima di costo $\alpha$;
3. $P_1$ è illimitato se e solo se $P_2$ è illimitato.

**Dimostrazione.**

1. Segue direttamente dalla definizione di equivalenza applicata a un generico $\alpha$: se $P_1$ ha una soluzione ammissibile di un certo costo $\alpha$, allora $P_2$ ne ha una dello stesso costo, e viceversa.

2. Supponiamo che $P_1$ abbia una soluzione ottima di costo $\alpha$. Allora $P_1$ ha una soluzione ammissibile di costo $\alpha$, quindi per equivalenza $P_2$ ha una soluzione ammissibile di costo $\alpha$. Se per assurdo $P_2$ avesse una soluzione ammissibile di costo $\beta > \alpha$ (nel caso di massimizzazione), allora per equivalenza $P_1$ avrebbe una soluzione ammissibile di costo $\beta > \alpha$, contraddicendo l'ottimalità di $\alpha$ per $P_1$. Quindi $\alpha$ è il valore ottimo anche per $P_2$. Il viceversa è simmetrico.

3. Supponiamo che $P_1$ sia illimitato. Allora per ogni $\alpha \in \mathbb{R}$ esiste una soluzione ammissibile di $P_1$ con costo maggiore di $\alpha$. Per equivalenza, per ogni $\alpha$ esiste una soluzione ammissibile di $P_2$ con lo stesso costo, quindi $P_2$ è illimitato. Il viceversa è simmetrico. $\square$

## 5. Riduzione alla forma standard

### Proposizione 5.1 — Equivalenza con la forma standard

Ogni programma lineare è equivalente a un programma lineare in forma standard.

**Dimostrazione.**

Sia dato un generico problema di PL nella forma

$$
\begin{aligned}
\max \; (\text{o} \min) \quad & c^T x \\
\text{soggetto a} \quad & A x \sim b \\
& x \in \mathbb{R}^n
\end{aligned}
$$

Mostriamo come trasformarlo in un problema equivalente in forma standard, attraverso una sequenza di passi.

---

**Riduzione a un problema di massimizzazione.**

Se il problema è di minimizzazione, cioè vogliamo determinare $\min c^T x$, allora esso è equivalente al problema di massimizzazione

$$
\max \; -c^T x.
$$

Infatti, minimizzare $c^T x$ equivale a massimizzare $-c^T x$: se $\bar{x}$ minimizza $c^T x$, allora $\bar{x}$ massimizza $-c^T x$, e i valori ottimi sono opposti. Quindi possiamo assumere senza perdita di generalità che il problema sia di massimizzazione.

---

**Distinzione tra variabili non-negative e variabili libere.**

Diciamo che una variabile $x_j$ è **non-negativa** se il sistema dei vincoli contiene esplicitamente il vincolo $x_j \geq 0$. Altrimenti, la variabile è detta **libera**.

Nel passaggio alla forma standard tratteremo le variabili non-negative e quelle libere in modo diverso.

---

**Normalizzazione dei vincoli di disuguaglianza.**

Consideriamo una disuguaglianza della forma

$$
a_i^T x \geq b_i
$$

che non sia un vincolo di non-negatività su una variabile. Tale disuguaglianza è equivalente a

$$
-a_i^T x \leq -b_i.
$$

Quindi possiamo assumere che ciascun vincolo del problema (eccetto quelli che impongono la non-negatività di qualche variabile) sia del tipo

$$
a_i^T x \leq b_i
\qquad \text{oppure} \qquad
a_i^T x = b_i.
$$

---

**Introduzione delle variabili di scarto.**

Consideriamo un vincolo della forma

$$
a_i^T x \leq b_i.
$$

Aggiungiamo al problema una **variabile di scarto** (o **slack**) $s_i \geq 0$ e rimpiazziamo il vincolo precedente con

$$
a_i^T x + s_i = b_i, \qquad s_i \geq 0.
$$

Mostriamo che questa trasformazione produce un problema equivalente.

- Se $x$ soddisfa $a_i^T x \leq b_i$, allora ponendo
  $$
  s_i = b_i - a_i^T x \geq 0,
  $$
  le condizioni $a_i^T x + s_i = b_i$ e $s_i \geq 0$ risultano verificate.

- Viceversa, se $x$ e $s_i$ soddisfano $a_i^T x + s_i = b_i$ con $s_i \geq 0$, allora
  $$
  a_i^T x = b_i - s_i \leq b_i,
  $$
  quindi $x$ soddisfa il vincolo di partenza.

Pertanto, poiché non abbiamo modificato la funzione obiettivo, il problema così ottenuto è equivalente a quello di partenza.

Ripetendo questo procedimento per ogni vincolo di disuguaglianza, otteniamo un problema di massimizzazione in cui tutti i vincoli (eccetto quelli di non-negatività) sono **equazioni**.

---

**Eliminazione delle variabili libere.**

Resta da gestire il caso in cui alcune variabili $x_j$ siano libere, cioè non soggette al vincolo $x_j \geq 0$.

L'idea si basa sulla semplice osservazione che ogni numero reale può essere scritto come differenza di due numeri non-negativi. Introduciamo quindi due variabili non-negative $x_j'$ e $x_j''$ e imponiamo

$$
x_j = x_j' - x_j'', \qquad x_j', x_j'' \geq 0.
$$

Sostituiamo $x_j$ con $x_j' - x_j''$ nella funzione obiettivo e in tutti i vincoli del problema. Otteniamo così un programma lineare in cui $x_j$ non compare più e le nuove variabili introdotte sono non-negative.

Mostriamo che il problema ottenuto è equivalente a quello di partenza.

- **Da una soluzione del nuovo problema a una del vecchio.**  
  Data una soluzione ammissibile $(x', x'')$ del nuovo problema, definiamo
  $$
  x_j = x_j' - x_j''
  $$
  per ogni variabile libera $x_j$, e lasciamo invariate le altre variabili. Allora $x$ soddisfa tutti i vincoli del problema originario e ha lo stesso costo.

- **Da una soluzione del vecchio problema a una del nuovo.**  
  Data una soluzione ammissibile $x$ del problema originario, scegliamo valori non-negativi per $x_j'$ e $x_j''$ tali che $x_j = x_j' - x_j''$. Possiamo porre, ad esempio,
  $$
  x_j' = \max\{0, x_j\}, \qquad x_j'' = \max\{0, -x_j\}.
  $$
  Allora $x_j', x_j'' \geq 0$ e $x_j' - x_j'' = x_j$. Sostituendo nelle equazioni e nella funzione obiettivo, otteniamo una soluzione ammissibile del nuovo problema con lo stesso costo.

Poiché le due costruzioni sono inverse l'una dell'altra, i due problemi sono equivalenti.

---

**Conclusione.**

Combinando i passi precedenti, ogni programma lineare è equivalente a un problema di massimizzazione della forma

$$
\begin{aligned}
\max \quad & c^T x \\
\text{soggetto a} \quad & A x = b \\
& x \geq 0
\end{aligned}
$$

cioè a un problema in forma standard. $\square$

### Esempio 5.2 — Riduzione alla forma standard

Consideriamo il problema

$$
\begin{aligned}
\min \quad & 2x_1 - 3x_2 \\
\text{soggetto a} \quad & x_1 + x_2 \geq 4 \\
& x_1 - x_2 = 1 \\
& x_1 \geq 0
\end{aligned}
$$

dove $x_2$ è libera.

**Passo 1 — Massimizzazione.**

Il problema equivale a

$$
\begin{aligned}
\max \quad & -2x_1 + 3x_2 \\
\text{soggetto a} \quad & x_1 + x_2 \geq 4 \\
& x_1 - x_2 = 1 \\
& x_1 \geq 0
\end{aligned}
$$

**Passo 3 — Normalizzazione della disuguaglianza.**

Il vincolo $x_1 + x_2 \geq 4$ diventa $-x_1 - x_2 \leq -4$.

**Passo 4 — Variabile di scarto.**

Introduciamo $s_1 \geq 0$ e scriviamo

$$
-x_1 - x_2 + s_1 = -4.
$$

**Passo 5 — Variabile libera.**

Poniamo $x_2 = x_2' - x_2''$ con $x_2', x_2'' \geq 0$. Sostituendo:

$$
\begin{aligned}
\max \quad & -2x_1 + 3x_2' - 3x_2'' \\
\text{soggetto a} \quad & -x_1 - x_2' + x_2'' + s_1 = -4 \\
& x_1 - x_2' + x_2'' = 1 \\
& x_1, x_2', x_2'', s_1 \geq 0
\end{aligned}
$$

Questo problema è in forma standard.
