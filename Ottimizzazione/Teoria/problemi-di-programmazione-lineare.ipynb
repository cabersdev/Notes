# Programmazione Lineare

---

## 0. Introduzione

La Programmazione Lineare è la branca dell’ottimizzazione in cui la funzione obiettivo e i vincoli sono **lineari** rispetto alle variabili decisionali. Essa costituisce il modello fondamentale per una vasta classe di problemi applicativi: produzione, trasporti, allocazione di risorse, pianificazione finanziaria, reti, logistica.

Il motivo per cui la linearità è così importante è duplice:

- dal punto di vista **modellistico**, molti problemi reali ammettono una schematizzazione lineare ragionevole;
- dal punto di vista **algoritmico**, i problemi lineari hanno una struttura geometrica e combinatoria che consente di risolverli in modo efficiente e con garanzie teoriche forti (teorema fondamentale, dualità, simplesso, punti interni).

---

## 1. Definizioni fondamentali

### Definizione 1.1 — Problema di Programmazione Lineare

Un **problema di Programmazione Lineare** è un problema di ottimizzazione vincolata della forma
$$
\begin{aligned}
\max \; (\text{oppure } \min) \quad & c_1x_1+\dots+c_nx_n \\
\text{soggetto a} \quad & a_{11}x_1+\dots+a_{1n}x_n \sim b_1,\\
& \vdots\\
& a_{m1}x_1+\dots+a_{mn}x_n \sim b_m,
\end{aligned}
$$
dove:

- $x_1,\dots,x_n$ sono le **variabili decisionali**;
- $c_1,\dots,c_n\in\mathbb{R}$ sono i **coefficienti di costo**;
- $a_{ij}\in\mathbb{R}$ sono i **coefficienti tecnologici**;
- $b_1,\dots,b_m\in\mathbb{R}$ sono i **termini noti**;
- ciascun simbolo $\sim$ denota uno degli operatori $\le$, $\ge$ oppure $=$.

La funzione
$$
f(x)=c_1x_1+\dots+c_nx_n
$$
è detta **funzione obiettivo**, mentre il sistema $Ax\sim b$ rappresenta i **vincoli**.

> **Osservazione 1.2.**  
> Non sono ammesse disuguaglianze strette ($<$ o $>$). I vincoli sono sempre non stretti. Inoltre la funzione obiettivo e i vincoli sono lineari.

### Definizione 1.3 — Forma matriciale

In forma matriciale, un problema di PL si scrive
$$
\begin{aligned}
\max \quad & c^T x\\
\text{soggetto a} \quad & Ax\sim b,
\end{aligned}
$$
dove $A\in\mathbb{R}^{m\times n}$, $c\in\mathbb{R}^n$, $b\in\mathbb{R}^m$, $x\in\mathbb{R}^n$. La matrice $A$ è detta **matrice dei coefficienti tecnologici**.

### Definizione 1.4 — Soluzione ammissibile

Un vettore $x\in\mathbb{R}^n$ è detto **soluzione ammissibile** del problema di PL se soddisfa tutti i vincoli. Il corrispondente valore $c^Tx$ è detto **valore** o **costo** della soluzione ammissibile $x$.

### Definizione 1.5 — Regione ammissibile

L’insieme di tutte le soluzioni ammissibili di un programma lineare è detto **regione ammissibile** e si indica con
$$
\mathcal{F}=\{x\in\mathbb{R}^n : Ax\sim b\}.
$$
La regione ammissibile è intersezione di semispazi e iperpiani, quindi è un **poliedro** (eventualmente vuoto, limitato o illimitato).

### Definizione 1.6 — Soluzione ottima

Un vettore $\bar{x}\in\mathbb{R}^n$ è una **soluzione ottima** del programma lineare se:

1. $\bar{x}$ è ammissibile, cioè $\bar{x}\in\mathcal{F}$;
2. se il problema è di massimizzazione,
   $$
   c^T\bar{x}\ge c^Tx \qquad \forall x\in\mathcal{F};
   $$
3. se il problema è di minimizzazione,
   $$
   c^T\bar{x}\le c^Tx \qquad \forall x\in\mathcal{F}.
   $$

Il corrispondente valore $c^T\bar{x}$ è detto **valore ottimo** del problema.

> **Osservazione 1.7.**  
> Un problema può avere nessuna, una, infinite soluzioni ottime. Inoltre può essere **inammissibile** (nessuna soluzione ammissibile) oppure **illimitato** (funzione obiettivo non limitata superiormente o inferiormente sulla regione ammissibile). Il teorema fondamentale della Programmazione Lineare chiarisce che non ci sono altre possibilità.

## 2. Geometria della regione ammissibile

### Definizione 2.1 — Poliedro

Un **poliedro** in $\mathbb{R}^n$ è un insieme della forma
$$
P=\{x\in\mathbb{R}^n : Ax\le b\},
$$
con $A\in\mathbb{R}^{m\times n}$ e $b\in\mathbb{R}^m$. Un poliedro è un insieme **convesso**: se $x,y\in P$, allora per ogni $\lambda\in[0,1]$ si ha $\lambda x+(1-\lambda)y\in P$.

La regione ammissibile di un problema di PL è sempre un poliedro, eventualmente scritto con uguaglianze e disuguaglianze.

### Definizione 2.2 — Punto estremo (vertice)

Sia $P\subseteq\mathbb{R}^n$ un insieme convesso. Un punto $x\in P$ è detto **punto estremo** (o **vertice**) di $P$ se non può essere scritto come combinazione convessa stretta di due punti distinti di $P$: cioè se
$$
x=\lambda y+(1-\lambda)z,\qquad y,z\in P,\quad \lambda\in(0,1),
$$
implica $y=z=x$.

> **Osservazione 2.3.**  
> I punti estremi sono i “vertici” del poliedro. In dimensione $2$ sono i vertici del poligono; in dimensione $3$ sono i vertici del poliedro solido.

### Definizione 2.4 — Soluzione di base ammissibile

Sia il problema in forma standard
$$
\max\{c^Tx : Ax=b,\ x\ge 0\},
$$
con $A\in\mathbb{R}^{m\times n}$ di rango $m$ (righe linearmente indipendenti). Una **soluzione di base ammissibile** (BFS, basic feasible solution) è un vettore $x\in\mathbb{R}^n$ tale che:

- $Ax=b$ e $x\ge 0$;
- esistono $m$ indici $B\subseteq\{1,\dots,n\}$ tali che le colonne $A_B$ sono linearmente indipendenti e $x_j=0$ per ogni $j\notin B$.

In altri termini, si scelgono $m$ colonne indipendenti, si pongono a zero le altre variabili e si risolve il sistema quadrato $A_Bx_B=b$; se la soluzione $x_B$ è non negativa, si ottiene una BFS.

> **Teorema 2.5 — Corrispondenza tra vertici e BFS.**  
> Sia $P=\{x\in\mathbb{R}^n : Ax=b,\ x\ge 0\}$ con $A$ di rango $m$. Allora $x$ è un punto estremo di $P$ se e solo se $x$ è una soluzione di base ammissibile.

## 3. Teorema fondamentale della Programmazione Lineare

### Teorema 3.1 — Alternativa fondamentale

Dato un programma lineare, vale una e una sola delle seguenti alternative:

1. il problema ha almeno una soluzione ottima;
2. il problema è **inammissibile**, cioè $\mathcal{F}=\emptyset$;
3. il problema è **illimitato**, cioè per ogni $\alpha\in\mathbb{R}$ esiste $x\in\mathcal{F}$ con $c^Tx\ge\alpha$ (nel caso di massimizzazione) oppure $c^Tx\le\alpha$ (nel caso di minimizzazione).

> **Osservazione 3.2.**  
> Il teorema esclude il caso intermedio in cui la funzione obiettivo abbia un estremo superiore finito che non sia raggiunto. In PL, se il valore ottimo è finito, allora è assunto in almeno un punto ammissibile.

### Teorema 3.3 — Ottimo in un vertice

Sia
$$
P=\{x\in\mathbb{R}^n : Ax=b,\ x\ge 0\}
$$
un poliedro non vuoto, con $A$ di rango $m$, e sia $f(x)=c^Tx$. Allora:

1. se il problema di massimizzazione ha una soluzione ottima unica, questa è un punto estremo di $P$;
2. se il problema ha più di una soluzione ottima, almeno una di esse è un punto estremo di $P$.

> **Osservazione 3.4.**  
> In forma standard, il teorema precedente si può equivalentemente enunciare dicendo che, se esiste un ottimo, allora esiste una soluzione di base ammissibile ottima. Questo è il fondamento del metodo del simplesso, che cerca l’ottimo tra le BFS.

## 4. Forma standard

### Definizione 4.1 — Forma standard

Un programma lineare è detto **in forma standard** se è scritto come
$$
\begin{aligned}
\max \quad & c^Tx\\
\text{soggetto a} \quad & Ax=b,\\
& x\ge 0,
\end{aligned}
$$
dove $A\in\mathbb{R}^{m\times n}$, $c\in\mathbb{R}^n$, $b\in\mathbb{R}^m$, $x\in\mathbb{R}^n$.

> **Osservazione 4.2.**  
> La scrittura $Ax=b$ è un modo compatto di indicare il sistema di $m$ equazioni lineari
> $$
> a_i^Tx=b_i,\qquad i=1,\dots,m,
> $$
> cioè $a_{i1}x_1+\dots+a_{in}x_n=b_i$ per $i=1,\dots,m$.

> **Osservazione 4.3.**  
> Nella forma standard tutti i vincoli (esclusa la non-negatività) sono **equazioni** e tutte le variabili sono **non-negative**. La scelta di massimizzare è una convenzione: ogni problema di minimizzazione può essere ricondotto a uno di massimizzazione cambiando segno alla funzione obiettivo.

> **Osservazione 4.4.**  
> In forma standard si assume spesso che la matrice $A$ abbia rango $m$ (righe linearmente indipendenti). Se così non fosse, alcune equazioni sarebbero ridondanti o incompatibili; le righe ridondanti possono essere eliminate, mentre l’incompatibilità rende il problema inammissibile.

## 5. Equivalenza tra programmi lineari

### Definizione 5.1 — Equivalenza

Due programmi lineari $P_1$ e $P_2$ sono detti **equivalenti** quando vale la seguente proprietà:

per ogni $\alpha\in\mathbb{R}$, $P_1$ ha una soluzione ammissibile $x$ di costo $\alpha$ se e solo se $P_2$ ha una soluzione ammissibile $y$ di costo $\alpha$.

Inoltre si richiede che, data una tale soluzione $x$, la corrispondente soluzione $y$ possa essere ricavata tramite una **regola esplicita** (e viceversa).

> **Osservazione 5.2.**  
> L’equivalenza non richiede una bigezione tra le variabili, ma solo la corrispondenza dei valori ammissibili e degli ottimi. Questo è importante perché alcune trasformazioni, come lo splitting delle variabili libere, non sono iniettive.

### Proposizione 5.3 — Conseguenze dell’equivalenza

Se $P_1$ e $P_2$ sono programmi lineari equivalenti, allora:

1. $P_1$ ha una soluzione ammissibile se e solo se $P_2$ ha una soluzione ammissibile;
2. per ogni $\alpha\in\mathbb{R}$, $P_1$ ha una soluzione ottima di costo $\alpha$ se e solo se $P_2$ ha una soluzione ottima di costo $\alpha$;
3. $P_1$ è illimitato se e solo se $P_2$ è illimitato.

**Dimostrazione.**

1. Segue direttamente dalla definizione di equivalenza applicata a un generico $\alpha$: se $P_1$ ha una soluzione ammissibile di un certo costo $\alpha$, allora $P_2$ ne ha una dello stesso costo, e viceversa.

2. Supponiamo che $P_1$ abbia una soluzione ottima di costo $\alpha$. Allora $P_1$ ha una soluzione ammissibile di costo $\alpha$, quindi per equivalenza $P_2$ ha una soluzione ammissibile di costo $\alpha$. Se per assurdo $P_2$ avesse una soluzione ammissibile di costo $\beta>\alpha$ (nel caso di massimizzazione), allora per equivalenza $P_1$ avrebbe una soluzione ammissibile di costo $\beta>\alpha$, contraddicendo l’ottimalità di $\alpha$ per $P_1$. Quindi $\alpha$ è il valore ottimo anche per $P_2$. Il viceversa è simmetrico.

3. Supponiamo che $P_1$ sia illimitato. Allora per ogni $\alpha\in\mathbb{R}$ esiste una soluzione ammissibile di $P_1$ con costo maggiore di $\alpha$. Per equivalenza, per ogni $\alpha$ esiste una soluzione ammissibile di $P_2$ con lo stesso costo, quindi $P_2$ è illimitato. Il viceversa è simmetrico. $\square$

## 6. Riduzione alla forma standard

### Proposizione 6.1 — Equivalenza con la forma standard

Ogni programma lineare è equivalente a un programma lineare in forma standard.

**Dimostrazione.**

Sia dato un generico problema di PL nella forma
$$
\begin{aligned}
\max \; (\text{o} \min) \quad & c^Tx\\
\text{soggetto a} \quad & Ax\sim b,\\
& x\in\mathbb{R}^n.
\end{aligned}
$$
Mostriamo come trasformarlo in un problema equivalente in forma standard attraverso una sequenza di passi.

---

**Passo 1 — Riduzione a un problema di massimizzazione.**

Se il problema è di minimizzazione, cioè vogliamo determinare $\min c^Tx$, allora esso è equivalente al problema di massimizzazione
$$
\max\; -c^Tx.
$$
Infatti, minimizzare $c^Tx$ equivale a massimizzare $-c^Tx$: se $\bar{x}$ minimizza $c^Tx$, allora $\bar{x}$ massimizza $-c^Tx$, e i valori ottimi sono opposti. Quindi possiamo assumere senza perdita di generalità che il problema sia di massimizzazione.

---

**Passo 2 — Distinzione tra variabili non-negative e variabili libere.**

Diciamo che una variabile $x_j$ è **non-negativa** se il sistema dei vincoli contiene esplicitamente il vincolo $x_j\ge 0$. Altrimenti, la variabile è detta **libera**. Nel passaggio alla forma standard tratteremo le variabili non-negative e quelle libere in modo diverso.

---

**Passo 3 — Normalizzazione dei vincoli di disuguaglianza.**

Consideriamo una disuguaglianza della forma
$$
a_i^Tx\ge b_i
$$
che non sia un vincolo di non-negatività su una variabile. Tale disuguaglianza è equivalente a
$$
-a_i^Tx\le -b_i.
$$
Quindi possiamo assumere che ciascun vincolo del problema (eccetto quelli che impongono la non-negatività di qualche variabile) sia del tipo
$$
a_i^Tx\le b_i
\qquad\text{oppure}\qquad
a_i^Tx=b_i.
$$

---

**Passo 4 — Introduzione delle variabili di scarto.**

Consideriamo un vincolo della forma
$$
a_i^Tx\le b_i.
$$
Aggiungiamo al problema una **variabile di scarto** (o **slack**) $s_i\ge 0$ e rimpiazziamo il vincolo precedente con
$$
a_i^Tx+s_i=b_i,\qquad s_i\ge 0.
$$

Mostriamo che questa trasformazione produce un problema equivalente.

- Se $x$ soddisfa $a_i^Tx\le b_i$, allora ponendo
  $$
  s_i=b_i-a_i^Tx\ge 0,
  $$
  le condizioni $a_i^Tx+s_i=b_i$ e $s_i\ge 0$ risultano verificate.

- Viceversa, se $x$ e $s_i$ soddisfano $a_i^Tx+s_i=b_i$ con $s_i\ge 0$, allora
  $$
  a_i^Tx=b_i-s_i\le b_i,
  $$
  quindi $x$ soddisfa il vincolo di partenza.

Pertanto, poiché non abbiamo modificato la funzione obiettivo, il problema così ottenuto è equivalente a quello di partenza.

Se invece si parte da un vincolo $a_i^Tx\ge b_i$, dopo averlo moltiplicato per $-1$ si ottiene $-a_i^Tx\le -b_i$ e si procede come sopra. Equivalentemente, si può introdurre una **variabile di surplus** $s_i\ge 0$ e scrivere
$$
a_i^Tx-s_i=b_i,\qquad s_i\ge 0.
$$
Le due formulazioni sono equivalenti.

Ripetendo questo procedimento per ogni vincolo di disuguaglianza, otteniamo un problema di massimizzazione in cui tutti i vincoli (eccetto quelli di non-negatività) sono **equazioni**.

---

**Passo 5 — Eliminazione delle variabili libere.**

Resta da gestire il caso in cui alcune variabili $x_j$ siano libere, cioè non soggette al vincolo $x_j\ge 0$.

L’idea si basa sull’osservazione che ogni numero reale può essere scritto come differenza di due numeri non-negativi. Introduciamo quindi due variabili non-negative $x_j^+$ e $x_j^-$ e imponiamo
$$
x_j=x_j^+-x_j^-,\qquad x_j^+,x_j^-\ge 0.
$$
Sostituiamo $x_j$ con $x_j^+-x_j^-$ nella funzione obiettivo e in tutti i vincoli del problema. Otteniamo così un programma lineare in cui $x_j$ non compare più e le nuove variabili introdotte sono non-negative.

Mostriamo che il problema ottenuto è equivalente a quello di partenza.

- **Da una soluzione del nuovo problema a una del vecchio.**  
  Data una soluzione ammissibile $(x^+,x^-)$ del nuovo problema, definiamo
  $$
  x_j=x_j^+-x_j^-
  $$
  per ogni variabile libera $x_j$, e lasciamo invariate le altre variabili. Allora $x$ soddisfa tutti i vincoli del problema originario e ha lo stesso costo.

- **Da una soluzione del vecchio problema a una del nuovo.**  
  Data una soluzione ammissibile $x$ del problema originario, scegliamo valori non-negativi per $x_j^+$ e $x_j^-$ tali che $x_j=x_j^+-x_j^-$. Possiamo porre, ad esempio,
  $$
  x_j^+=\max\{0,x_j\},\qquad x_j^-=\max\{0,-x_j\}.
  $$
  Allora $x_j^+,x_j^-\ge 0$ e $x_j^+-x_j^-=x_j$. Sostituendo nelle equazioni e nella funzione obiettivo, otteniamo una soluzione ammissibile del nuovo problema con lo stesso costo.

Poiché le due costruzioni sono inverse l’una dell’altra, i due problemi sono equivalenti.

---

**Conclusione.**

Combinando i passi precedenti, ogni programma lineare è equivalente a un problema di massimizzazione della forma
$$
\begin{aligned}
\max \quad & c^Tx\\
\text{soggetto a} \quad & Ax=b,\\
& x\ge 0,
\end{aligned}
$$
cioè a un problema in forma standard. $\square$

## 7. Esempio completo di riduzione a forma standard

Consideriamo il programma lineare
$$
\begin{aligned}
\mathcal{P}: \quad & \max\; x_1+7x_2+5x_3\\
& -x_1+2x_2+x_3\ge 8,\\
& 2x_1-x_2+3x_3\le 9,\\
& x_1+3x_2=5,\\
& x_1\ge 0,\quad x_2\le 0,
\end{aligned}
$$
con $x_3$ libera.

### Passo 1 — Vincolo $\ge$

Il primo vincolo è equivalente a
$$
x_1-2x_2-x_3\le -8.
$$
Introduciamo una variabile di scarto $s_1\ge 0$:
$$
x_1-2x_2-x_3+s_1=-8.
$$

### Passo 2 — Vincolo $\le$

Il secondo vincolo diventa introducendo una variabile slack $s_2 \ge 0$
$$
2x_1-x_2+3x_3+s_2=9,\qquad s_2\ge 0.
$$

Il terzo vincolo è già un’uguaglianza:
$$
x_1+3x_2=5.
$$

### Passo 3 — Variabile $x_2\le 0$

Poniamo
$$
x_2=-x_2',\qquad x_2'\ge 0.
$$
Sostituendo:

- funzione obiettivo:
  $$
  x_1+7x_2+5x_3=x_1-7x_2'+5x_3;
  $$
- primo vincolo:
  $$
  x_1-2(-x_2')-x_3+s_1=x_1+2x_2'-x_3+s_1=-8;
  $$
- secondo vincolo:
  $$
  2x_1-(-x_2')+3x_3+s_2=2x_1+x_2'+3x_3+s_2=9;
  $$
- terzo vincolo:
  $$
  x_1+3(-x_2')=x_1-3x_2'=5.
  $$

Il problema diventa
$$
\begin{aligned}
\max \quad & x_1-7x_2'+5x_3\\
\text{soggetto a} \quad
& x_1+2x_2'-x_3+s_1=-8,\\
& 2x_1+x_2'+3x_3+s_2=9,\\
& x_1-3x_2'=5,\\
& x_1\ge 0,\quad x_2'\ge 0,\quad s_1\ge 0,\quad s_2\ge 0,
\end{aligned}
$$
con $x_3$ ancora libera.

### Passo 4 — Eliminazione della variabile libera $x_3$

#### Metodo A — Splitting

Poniamo
$$
x_3=x_3^+-x_3^-,\qquad x_3^+\ge 0,\quad x_3^-\ge 0.
$$
Sostituendo si ottiene il problema in forma standard:
$$
\begin{aligned}
\max \quad & x_1-7x_2'+5x_3^+-5x_3^-\\
\text{soggetto a} \quad
& x_1+2x_2'-x_3^++x_3^-+s_1=-8,\\
& 2x_1+x_2'+3x_3^+-3x_3^-+s_2=9,\\
& x_1-3x_2'=5,\\
& x_1,x_2',x_3^+,x_3^-,s_1,s_2\ge 0.
\end{aligned}
$$

Questa è una forma standard: massimizzazione, vincoli di uguaglianza, tutte le variabili non-negative. Il termine noto $-8$ nella prima equazione non è un problema: la forma standard non richiede $b\ge 0$. Se si desidera $b\ge 0$, si può moltiplicare la prima equazione per $-1$, ottenendo
$$
-x_1-2x_2'+x_3^+-x_3^--s_1=8.
$$

#### Metodo B — Sostituzione tramite un’uguaglianza

Dal primo vincolo
$$
x_1+2x_2'-x_3+s_1=-8
$$
si ricava
$$
x_3=x_1+2x_2'+s_1+8.
$$
Sostituendo nella funzione obiettivo:
$$
\begin{aligned}
x_1-7x_2'+5x_3
&=x_1-7x_2'+5(x_1+2x_2'+s_1+8)\\
&=6x_1+3x_2'+5s_1+40.
\end{aligned}
$$
Sostituendo nel secondo vincolo:
$$
\begin{aligned}
2x_1+x_2'+3(x_1+2x_2'+s_1+8)+s_2&=9\\
5x_1+7x_2'+3s_1+s_2+24&=9,
\end{aligned}
$$
cioè
$$
5x_1+7x_2'+3s_1+s_2=-15.
$$
Il terzo vincolo resta $x_1-3x_2'=5$. Otteniamo il problema equivalente
$$
\begin{aligned}
\max \quad & 6x_1+3x_2'+5s_1+40\\
\text{soggetto a} \quad
& 5x_1+7x_2'+3s_1+s_2=-15,\\
& x_1-3x_2'=5,\\
& x_1,x_2',s_1,s_2\ge 0.
\end{aligned}
$$
La variabile $x_3$ è definita da $x_3=x_1+2x_2'+s_1+8$.

> **Osservazione 7.1 — Inammissibilità dell’esempio.**  
> Dall’equazione $x_1-3x_2'=5$ si ha $x_1=5+3x_2'$. Sostituendo nella seconda equazione:
> $$
> 5(5+3x_2')+7x_2'+3s_1+s_2=-15
> \;\Longrightarrow\;
> 22x_2'+3s_1+s_2=-40.
> $$
> Poiché $x_2',s_1,s_2\ge 0$, il membro di sinistra è non negativo, mentre il membro di destra è negativo. Dunque il problema è **inammissibile**. La riduzione a forma standard è comunque corretta: essa mette in evidenza l’incompatibilità dei vincoli.
