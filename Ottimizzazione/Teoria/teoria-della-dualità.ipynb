# Teoria della dualità

> **Osservazione.** Questi appunti introducono la **teoria della dualità** per i problemi di Programmazione Lineare. L'idea centrale è che a ogni problema di PL (detto **primale**) è associato un secondo problema (detto **duale**), legato al primo da relazioni profonde: il valore ottimo del duale fornisce un **upper bound** (o lower bound, a seconda della forma) per il valore ottimo del primale, e sotto condizioni di regolarità i due valori coincidono. La dualità è lo strumento che permette di interpretare i moltiplicatori di Lagrange come **prezzi ombra**, di costruire algoritmi duali (simplesso duale, metodi di punto interno) e di derivare le **condizioni di ottimalità** (scarti complementari).

---

## 0. Motivazione

### Osservazione 0.1 — Perché serve il duale

Dato un problema di PL in forma standard
$$
\begin{aligned}
\max \quad & c^T x\\
\text{s.a} \quad & Ax = b,\\
& x \ge 0,
\end{aligned}
\qquad (P_s)
$$
siamo interessati a **delimitare superiormente** il valore ottimo $z^*$: trovare cioè vettori $u$ tali che $z^* \le u$. Se riusciamo a costruire un problema di ottimizzazione che produce il **miglior upper bound possibile**, allora quel problema è il duale di $(P_s)$.

L'idea è semplice: per ogni $y \in \mathbb{R}^m$, ogni $x$ ammissibile per $(P_s)$ soddisfa
$$
y^T (Ax) = y^T b,
$$
perché $Ax = b$. Se inoltre $A^T y \ge c$, allora, poiché $x \ge 0$,
$$
c^T x \le (A^T y)^T x = y^T A x = y^T b = b^T y.
$$
Dunque, per ogni $y$ con $A^T y \ge c$, il valore $b^T y$ è un upper bound per $z^*$. Il **miglior** upper bound ottenibile in questo modo è la soluzione del problema
$$
\begin{aligned}
\min \quad & b^T y\\
\text{s.a} \quad & A^T y \ge c.
\end{aligned}
\qquad (D_s)
$$
Questo è il **problema duale** di $(P_s)$.

> **Osservazione 0.2.**  
> Il problema duale ha una struttura speculare a quella del primale: le variabili duali $y$ sono in numero pari ai vincoli del primale ($m$), i vincoli duali sono in numero pari alle variabili del primale ($n$), e la matrice dei coefficienti è $A^T$. La direzione dell'ottimizzazione è opposta (min vs max) e il verso delle disuguaglianze è invertito ($\ge$ invece di $=$).

## 1. Dualità per problemi in forma standard

### Definizione 1.1 — Problema primale e duale in forma standard

Sia $(P_s)$ il problema in forma standard
$$
\begin{aligned}
\max \quad & c^T x\\
\text{s.a} \quad & Ax = b,\\
& x \ge 0,
\end{aligned}
$$
con $A \in \mathbb{R}^{m \times n}$, $b \in \mathbb{R}^m$, $c \in \mathbb{R}^n$. Il **problema duale** è
$$
\begin{aligned}
\min \quad & b^T y\\
\text{s.a} \quad & A^T y \ge c.
\end{aligned}
\qquad (D_s)
$$
Il problema $(P_s)$ è detto **primale**.

> **Osservazione 1.2.**  
> Nel duale non ci sono vincoli di non-negatività espliciti sulle variabili $y$: le variabili duali sono **libere**. Questo riflette il fatto che i vincoli primali sono uguaglianze. Se il primale avesse vincoli di disuguaglianza, il duale avrebbe vincoli di non-negatività.

### Teorema 1.3 — Dualità debole (forma standard)

Se $x$ è ammissibile per $(P_s)$ e $y$ è ammissibile per $(D_s)$, allora
$$
c^T x \le b^T y.
$$

**Dimostrazione.**  
Poiché $Ax = b$, per ogni $y \in \mathbb{R}^m$ si ha $y^T A x = y^T b$. Poiché $A^T y \ge c$ e $x \ge 0$,
$$
c^T x \le (A^T y)^T x = y^T A x = y^T b = b^T y.
$$
$\square$

> **Osservazione 1.4.**  
> Il teorema di dualità debole è **sempre vero**, senza ipotesi aggiuntive. Esso implica che ogni valore ammissibile del duale è un upper bound per ogni valore ammissibile del primale. In particolare, se il primale è illimitato, il duale è inammissibile; se il duale è illimitato, il primale è inammissibile.

### Corollario 1.5 — Criterio di ottimalità

Siano $x^*$ ammissibile per $(P_s)$ e $y^*$ ammissibile per $(D_s)$. Se $c^T x^* = b^T y^*$, allora $x^*$ e $y^*$ sono soluzioni ottime dei rispettivi problemi.

**Dimostrazione.**  
Se $x^*$ non fosse ottima, esisterebbe $\bar{x}$ ammissibile per $(P_s)$ con $c^T \bar{x} > c^T x^* = b^T y^*$, contraddicendo la dualità debole. Analogamente per $y^*$. $\square$

### Corollario 1.6 — Illimitatezza e inammissibilità

Se uno dei due problemi $(P_s)$ e $(D_s)$ è illimitato, allora l'altro è inammissibile.

**Dimostrazione.**  
Supponiamo che $(D_s)$ abbia una soluzione ammissibile $\bar{y}$. Allora, per la dualità debole, $c^T x \le b^T \bar{y}$ per ogni $x$ ammissibile per $(P_s)$, quindi $(P_s)$ non è illimitato. L'altra metà si dimostra in modo analogo. $\square$

> **Osservazione 1.7.**  
> I Corollari 1.5 e 1.6 mostrano che la dualità debole ha conseguenze operative: fornisce un **certificato di ottimalità** (se i valori coincidono, entrambe le soluzioni sono ottime) e un **certificato di illimitatezza** (se un problema è illimitato, l'altro non può avere soluzioni ammissibili).

### Teorema 1.8 — Dualità forte (forma standard)

Se $(P_s)$ ha una soluzione ottima $x^*$, allora $(D_s)$ ha una soluzione ottima $y^*$ e
$$
c^T x^* = b^T y^*.
$$

**Dimostrazione.**  
La dimostrazione si basa sul legame tra basi ottime del primale e soluzioni del duale. $\square$

> **Osservazione 1.9.**  
> Il teorema di dualità forte afferma che, se il primale ha ottimo, il **miglior upper bound** ottenibile è **esattamente** il valore ottimo del primale. Non c'è gap tra primale e duale: il valore ottimo del duale coincide con quello del primale.

## 2. Dualità, basi ottime e metodo del simplesso

### Definizione 2.1 — Base e soluzione di base

Sia $B \subseteq \{1,\dots,n\}$ una **base** di $A$ (cioè $|B|=m$ e $A_B$ invertibile). La **soluzione di base** relativa a $B$ è
$$
\bar{x}_B = A_B^{-1} b,
\qquad
\bar{x}_N = 0.
$$
La base è **ammissibile nel primale** se $\bar{x}_B \ge 0$.

### Definizione 2.2 — Soluzione duale associata a una base

Data una base $B$, si definisce
$$
\bar{y} = A_B^{-T} c_B,
$$
dove $A_B^{-T} = (A_B^{-1})^T = (A_B^T)^{-1}$.

> **Osservazione 2.3.**  
> La soluzione duale $\bar{y}$ è definita a partire dalla base $B$ e dai coefficienti $c_B$ delle variabili in base. Non dipende dai valori delle variabili fuori base. Si noti che
> $$
> c^T \bar{x} = c_B^T \bar{x}_B = c_B^T A_B^{-1} b = (A_B^{-T} c_B)^T b = \bar{y}^T b = b^T \bar{y}.
> $$
> Quindi, per ogni base $B$, il valore della soluzione di base primale coincide con il valore della soluzione duale associata.

### Proposizione 2.4 — Condizioni di ammissibilità duale

La soluzione duale $\bar{y} = A_B^{-T} c_B$ è ammissibile per $(D_s)$ se e solo se
$$
\bar{c}_N = c_N - A_N^T A_B^{-T} c_B \le 0,
$$
dove $\bar{c}_N$ è il vettore dei **costi ridotti** delle variabili fuori base.

**Dimostrazione.**  
$\bar{y}$ è ammissibile per $(D_s)$ se e solo se $A^T \bar{y} \ge c$. Partizionando $A$ in $[A_B \mid A_N]$ e $c$ in $(c_B, c_N)$,
$$
A_B^T \bar{y} \ge c_B,
\qquad
A_N^T \bar{y} \ge c_N.
$$
La prima disuguaglianza è sempre verificata con uguaglianza, perché
$$
A_B^T \bar{y} = A_B^T A_B^{-T} c_B = c_B.
$$
La seconda diventa
$$
A_N^T A_B^{-T} c_B \ge c_N
\quad\Longleftrightarrow\quad
c_N - A_N^T A_B^{-T} c_B \le 0
\quad\Longleftrightarrow\quad
\bar{c}_N \le 0.
$$
$\square$

### Definizione 2.5 — Base ottima

Una base $B$ è detta **ottima** se è ammissibile sia nel primale ($\bar{x}_B \ge 0$) sia nel duale ($\bar{c}_N \le 0$).

### Teorema 2.6 — Corrispondenza tra basi ottime e soluzioni ottime

Se $B$ è una base ottima, allora $\bar{x}$ è soluzione ottima di $(P_s)$ e $\bar{y}$ è soluzione ottima di $(D_s)$.

**Dimostrazione.**  
Per la Proposizione 2.4, $\bar{y}$ è ammissibile per $(D_s)$. Per ipotesi, $\bar{x}$ è ammissibile per $(P_s)$. Inoltre, come osservato, $c^T \bar{x} = b^T \bar{y}$. Per il Corollario 1.5, $\bar{x}$ e $\bar{y}$ sono ottime. $\square$

> **Osservazione 2.7.**  
> Il Teorema 2.6 è il ponte tra la teoria della dualità e il metodo del simplesso. Il simplesso, nella sua seconda fase, mantiene una base ammissibile nel primale ad ogni iterazione e termina quando la base diventa ammissibile anche nel duale, cioè quando $\bar{c}_N \le 0$. A quel punto, la base è ottima e la soluzione duale associata è ottima per $(D_s)$. Questo dimostra il Teorema di Dualità Forte (Teorema 1.8) in modo costruttivo.

### Osservazione 2.8 — Interpretazione dei costi ridotti

I costi ridotti $\bar{c}_j$ hanno un'interpretazione duale: essi misurano la **violazione** del vincolo duale $A_j^T y \ge c_j$ da parte della soluzione duale $\bar{y}$. Se $\bar{c}_j > 0$, il vincolo duale è violato e la variabile $x_j$ ha convenienza a entrare in base. Se $\bar{c}_j \le 0$, il vincolo è soddisfatto e la variabile non conviene.

## 3. Dualità per problemi in altre forme

### 3.1 Forma canonica

#### Definizione 3.1 — Forma canonica

Un problema di PL è in **forma canonica** se è scritto come
$$
\begin{aligned}
\max \quad & c^T x\\
\text{s.a} \quad & Ax \le b,\\
& x \ge 0.
\end{aligned}
\qquad (P_c)
$$

#### Teorema 3.2 — Duale della forma canonica

Il duale di $(P_c)$ è
$$
\begin{aligned}
\min \quad & b^T y\\
\text{s.a} \quad & A^T y \ge c,\\
& y \ge 0.
\end{aligned}
\qquad (D_c)
$$

**Dimostrazione (Metodo 1 — upper bound).**  
Per ogni $y \ge 0$ e ogni $x$ ammissibile per $(P_c)$, si ha
$$
y^T (Ax) \le y^T b
$$
perché $Ax \le b$ e $y \ge 0$. Se inoltre $A^T y \ge c$, allora, poiché $x \ge 0$,
$$
c^T x \le (A^T y)^T x = y^T A x \le y^T b = b^T y.
$$
Quindi $b^T y$ è un upper bound per $z^*$. Il miglior upper bound si ottiene minimizzando $b^T y$ sotto i vincoli $A^T y \ge c$ e $y \ge 0$. $\square$

**Dimostrazione (Metodo 2 — riduzione a forma standard).**  
Si introducano variabili di scarto $s \ge 0$ e si scriva $(P_c)$ come
$$
\begin{aligned}
\max \quad & c^T x + 0^T s\\
\text{s.a} \quad & Ax + I s = b,\\
& x \ge 0,\ s \ge 0.
\end{aligned}
$$
Applicando la Definizione 1.1, il duale è
$$
\begin{aligned}
\min \quad & b^T y\\
\text{s.a} \quad & A^T y \ge c,\\
& I^T y \ge 0 \quad\Longleftrightarrow\quad y \ge 0.
\end{aligned}
$$
$\square$

> **Osservazione 3.3.**  
> La differenza rispetto alla forma standard è che nel duale della forma canonica le variabili $y$ sono **non-negative**. Questo riflette il fatto che i vincoli primali sono disuguaglianze $\le$: per costruire un upper bound, è necessario che $y \ge 0$.

### 3.2 Forma generale

#### Definizione 3.4 — Forma generale

Un problema di PL è in **forma generale** se è scritto come
$$
\begin{aligned}
\max \quad & c^T x\\
\text{s.a} \quad & Ax \le b.
\end{aligned}
\qquad (P_g)
$$
Non ci sono vincoli di non-negatività espliciti su $x$.

#### Teorema 3.5 — Duale della forma generale

Il duale di $(P_g)$ è
$$
\begin{aligned}
\min \quad & b^T y\\
\text{s.a} \quad & A^T y = c,\\
& y \ge 0.
\end{aligned}
\qquad (D_g)
$$

**Dimostrazione.**  
Si trasformi $(P_g)$ in forma canonica sostituendo $x = x^+ - x^-$ con $x^+, x^- \ge 0$. Il problema diventa
$$
\begin{aligned}
\max \quad & c^T x^+ - c^T x^-\\
\text{s.a} \quad & A x^+ - A x^- \le b,\\
& x^+, x^- \ge 0.
\end{aligned}
$$
Il duale è
$$
\begin{aligned}
\min \quad & b^T y\\
\text{s.a} \quad & A^T y \ge c,\\
& -A^T y \ge -c,\\
& y \ge 0,
\end{aligned}
$$
che è equivalente a $A^T y = c$ e $y \ge 0$. $\square$

> **Osservazione 3.6.**  
> Nel duale della forma generale, i vincoli sono **uguaglianze** ($A^T y = c$). Questo riflette il fatto che le variabili primali sono **libere**. La dualità tra tipi di vincoli e tipi di variabili è sistematica e sarà formalizzata nella "ricetta" della Sezione 4.

### 3.3 Duale del duale

#### Teorema 3.7 — Duale del duale

Dato un programma lineare $(P)$, il duale del duale è equivalente a $(P)$.

**Dimostrazione (per la forma standard).**  
Il duale di $(P_s)$ è $(D_s)$:
$$
\min\{b^T y : A^T y \ge c\}.
$$
Scriviamo $(D_s)$ in forma standard:
$$
\max\{-b^T y : A^T y - s = c,\ s \ge 0\}.
$$
Il duale di questo problema ha variabili $x$ (per i vincoli di uguaglianza) e $w$ (per i vincoli $s \ge 0$):
$$
\min\{c^T x : A x = b,\ x \ge 0\},
$$
che è equivalente a $(P_s)$ (a meno del segno, che si gestisce cambiando direzione). $\square$

> **Osservazione 3.8.**  
> Il Teorema 3.7 implica che la dualità è **simmetrica**: ogni problema di PL è il duale del suo duale. Questo permette di applicare i risultati della dualità in modo bidirezionale: se un teorema vale per il primale, vale anche per il duale, e viceversa.

### 3.4 Tabella delle possibilità

Il Teorema di Dualità Forte e il Corollario 1.6 implicano che ogni coppia primale/duale soddisfa una delle alternative rappresentate nella tabella seguente:
$$
\begin{array}{c|ccc}
 & \text{Primale ottimo} & \text{Primale inamm.} & \text{Primale illim.}\\
\hline
\text{Duale ottimo} & \text{Possibile} & \text{No} & \text{No}\\
\text{Duale inamm.} & \text{No} & \text{Possibile} & \text{Possibile}\\
\text{Duale illim.} & \text{No} & \text{Possibile} & \text{No}
\end{array}
$$

> **Osservazione 3.9.**  
> In particolare, se entrambi i problemi sono ammissibili, allora entrambi hanno ottimo e i valori ottimi coincidono. Questo è il contenuto della versione rafforzata del Teorema di Dualità Forte (Teorema 3.10).

#### Teorema 3.10 — Dualità forte (versione rafforzata)

Dati due programmi lineari $(P)$ e $(D)$, l'uno il duale dell'altro, se uno dei due ha soluzione ottima, allora entrambi hanno soluzione ottima e i valori ottimi coincidono.

## 4. Dualità in generale: la ricetta

### Osservazione 4.1 — Perché una ricetta

Le sezioni precedenti mostrano come costruire il duale per forme specifiche (standard, canonica, generale). Esiste una **regola sistematica** che permette di scrivere il duale direttamente dalla forma del primale, senza doverlo prima trasformare in forma standard. La regola si basa sulla corrispondenza tra:

- tipo di vincolo primale → segno della variabile duale;
- tipo di variabile primale → tipo di vincolo duale.

### Definizione 4.2 — Forma generale con tipi misti

Consideriamo un problema di PL nella forma
$$
\begin{aligned}
\max \quad & c^T x\\
\text{s.a} \quad & a_i^T x \le b_i, && i = 1,\dots,h,\\
& a_i^T x \ge b_i, && i = h+1,\dots,k,\\
& a_i^T x = b_i, && i = k+1,\dots,m,\\
& x_j \ge 0, && j = 1,\dots,p,\\
& x_j \le 0, && j = p+1,\dots,q,\\
& x_j \text{ libera}, && j = q+1,\dots,n.
\end{aligned}
\qquad (P)
$$

### Teorema 4.3 — Il duale nella forma generale

Il duale di $(P)$ è
$$
\begin{aligned}
\min \quad & b^T y\\
\text{s.a} \quad & A_j^T y \ge c_j, && j = 1,\dots,p,\\
& A_j^T y \le c_j, && j = p+1,\dots,q,\\
& A_j^T y = c_j, && j = q+1,\dots,n,\\
& y_i \ge 0, && i = 1,\dots,h,\\
& y_i \le 0, && i = h+1,\dots,k,\\
& y_i \text{ libera}, && i = k+1,\dots,m.
\end{aligned}
\qquad (D)
$$

La corrispondenza è riassunta nella tabella seguente:
$$
\begin{array}{c|c}
\text{Primale (max)} & \text{Duale (min)}\\
\hline
a_i^T x \le b_i & y_i \ge 0\\
a_i^T x \ge b_i & y_i \le 0\\
a_i^T x = b_i & y_i \text{ libera}\\
x_j \ge 0 & A_j^T y \ge c_j\\
x_j \le 0 & A_j^T y \le c_j\\
x_j \text{ libera} & A_j^T y = c_j
\end{array}
$$
La tabella si legge in entrambi i sensi: da sinistra a destra per ottenere il duale di un problema di massimo, da destra a sinistra per ottenere il duale di un problema di minimo.

> **Osservazione 4.4.**  
> La regola è **sistematica**: ogni vincolo primale genera una variabile duale con segno determinato dal verso del vincolo; ogni variabile primale genera un vincolo duale con tipo determinato dal segno della variabile. Il duale ottenuto con questa ricetta coincide con quello ottenuto per riduzione a forma standard.

## 5. Scarti complementari

### Teorema 5.1 — Scarti complementari (forma standard)

Sia $(P_s)$ in forma standard e $(D_s)$ il suo duale. Siano $\bar{x}$ ammissibile per $(P_s)$ e $\bar{y}$ ammissibile per $(D_s)$. Allora $\bar{x}$ e $\bar{y}$ sono ottime per i rispettivi problemi se e solo se, per ogni $j = 1,\dots,n$,
$$
\bar{x}_j = 0
\qquad\text{oppure}\qquad
A_j^T \bar{y} = c_j
$$
(cioè $\bar{y}$ soddisfa ad uguaglianza il $j$-esimo vincolo duale).

**Dimostrazione.**  
Dall'ammissibilità di $\bar{x}$ e $\bar{y}$,
$$
c^T \bar{x} \le (A^T \bar{y})^T \bar{x} = \bar{y}^T A \bar{x} = \bar{y}^T b = b^T \bar{y}.
$$
Per la dualità forte, $\bar{x}$ e $\bar{y}$ sono ottime se e solo se $c^T \bar{x} = b^T \bar{y}$, cioè se e solo se nella catena precedente c'è uguaglianza ovunque. L'uguaglianza $c^T \bar{x} = (A^T \bar{y})^T \bar{x}$ equivale a
$$
(A^T \bar{y} - c)^T \bar{x} = 0,
$$
cioè
$$
\sum_{j=1}^n (A_j^T \bar{y} - c_j) \bar{x}_j = 0.
$$
Poiché $\bar{x}_j \ge 0$ e $A_j^T \bar{y} - c_j \ge 0$ per ogni $j$, la somma è nulla se e solo se ogni termine è nullo, cioè se e solo se per ogni $j$ almeno uno tra $\bar{x}_j$ e $A_j^T \bar{y} - c_j$ è zero. $\square$

> **Osservazione 5.2.**  
> Le condizioni di scarti complementari hanno un'interpretazione economica: se una variabile primale è positiva ($\bar{x}_j > 0$), allora il corrispondente vincolo duale è attivo ($A_j^T \bar{y} = c_j$); se una variabile primale è nulla ($\bar{x}_j = 0$), il vincolo duale può essere non attivo. Simmetricamente, se un vincolo primale è attivo, la corrispondente variabile duale può essere positiva; se non è attivo, la variabile duale è nulla. Questo è il contenuto del teorema nella forma generale.

### Teorema 5.3 — Scarti complementari (forma generale)

Dato un programma lineare nella forma generale (4.2) e il suo duale, siano $\bar{x}$ e $\bar{y}$ soluzioni ammissibili. Allora $\bar{x}$ e $\bar{y}$ sono ottime se e solo se valgono le condizioni:

- per ogni $j = 1,\dots,n$: $\bar{x}_j = 0$ oppure $\bar{y}$ soddisfa ad uguaglianza il $j$-esimo vincolo duale;
- per ogni $i = 1,\dots,m$: $\bar{y}_i = 0$ oppure $\bar{x}$ soddisfa ad uguaglianza l'$i$-esimo vincolo primale.

> **Osservazione 5.4.**  
> Se una variabile primale è libera, il corrispondente vincolo duale è un'uguaglianza, quindi la condizione di complementarietà è automaticamente soddisfatta (per l'ammissibilità). Simmetricamente, se una variabile duale è libera, il corrispondente vincolo primale è un'uguaglianza, e la condizione è automaticamente soddisfatta. Questo spiega perché nella forma standard (dove tutte le variabili primali sono non-negative e tutti i vincoli primali sono uguaglianze) le condizioni relative alle variabili duali sono omesse.

> **Osservazione 5.5.**  
> Durante il metodo del simplesso, ad ogni iterazione le soluzioni primale e duale associate alla base corrente soddisfano **sempre** le condizioni di scarti complementari. Le variabili primali positive sono quelle in base; per queste, il corrispondente vincolo duale è soddisfatto ad uguaglianza perché fa parte del sottosistema $A_B^T \bar{y} = c_B$. Tuttavia, le due soluzioni non sono ottime fintantoché non è garantita l'ammissibilità di entrambe: il simplesso termina quando la base diventa ammissibile anche nel duale.

## 6. Interpretazione economica della dualità

### 6.1 Mercato ombra

Consideriamo un problema di produzione: un'azienda ha $m$ risorse in quantità $b_i$ e può produrre $n$ tipi di prodotti con profitto unitario $c_j$ e coefficienti tecnologici $a_{ij}$ (quantità di risorsa $i$ necessaria per un'unità di prodotto $j$). Il problema primale è
$$
\begin{aligned}
\max \quad & c^T x\\
\text{s.a} \quad & Ax \le b,\\
& x \ge 0.
\end{aligned}
$$
Il duale è
$$
\begin{aligned}
\min \quad & b^T y\\
\text{s.a} \quad & A^T y \ge c,\\
& y \ge 0.
\end{aligned}
$$

**Interpretazione.** Immaginiamo un operatore che voglia **acquistare le risorse** dell'azienda. Sia $y_i$ il prezzo offerto per un'unità di risorsa $i$. Il costo totale per l'acquirente è $b^T y$, che è la funzione obiettivo del duale (da minimizzare). Affinché l'offerta sia **concorrenziale**, l'azienda deve guadagnare dalla vendita delle risorse almeno quanto guadagnerebbe producendo e vendendo i prodotti. Per ogni prodotto $j$, il ricavo dalla vendita delle risorse necessarie a produrlo è $A_j^T y = \sum_i a_{ij} y_i$; il profitto dalla produzione è $c_j$. La condizione di concorrenzialità è quindi
$$
A_j^T y \ge c_j
\qquad\forall j,
$$
che è esattamente il vincolo duale. Inoltre, i prezzi devono essere non negativi: $y \ge 0$.

Il duale è quindi il problema dell'acquirente: **minimizzare il costo totale di acquisto** soggetto al vincolo che l'offerta sia conveniente per l'azienda.

> **Osservazione 6.2.**  
> Il Teorema di Dualità Debole afferma che qualsiasi offerta concorrenziale dell'acquirente è almeno pari al valore ottimo del primale: l'azienda non può perdere accettando un'offerta concorrenziale. Il Teorema di Dualità Forte afferma che, all'ottimo, l'azienda è **indifferente** tra produrre e vendere le risorse: i due valori coincidono.

### 6.2 Prezzi ombra

Le variabili duali $y_i^*$ hanno un'interpretazione come **prezzi ombra** (shadow prices) delle risorse. Se la disponibilità della risorsa $i$ aumenta di un'unità (cioè $b_i$ diventa $b_i + 1$), il valore ottimo del primale aumenta di $y_i^*$ (per perturbazioni sufficientemente piccole, cioè finché la base ottima non cambia).

**Interpretazione.** $y_i^*$ è il **valore marginale** della risorsa $i$: quanto sarebbe disposta a pagare l'azienda per un'unità aggiuntiva di quella risorsa. Se $y_i^* = 0$, la risorsa non è un collo di bottiglia: un'unità in più non aumenta il profitto. Se $y_i^* > 0$, la risorsa è un collo di bottiglia e un'unità in più aumenta il profitto di $y_i^*$.

> **Osservazione 6.3.**  
> I prezzi ombra sono la base per l'analisi di sensibilità: permettono di valutare l'impatto di variazioni nelle disponibilità delle risorse senza risolvere il problema da capo. Sono anche lo strumento concettuale per interpretare i moltiplicatori di Lagrange nei problemi di ottimizzazione vincolata.
