# Grafi e complessità degli algoritmi

---

## 1. Grafi non orientati

### Definizione 1.1 — Grafo non orientato

Un **grafo non orientato** è una coppia ordinata $G=(V,E)$, dove $V$ è un insieme finito di **nodi** (o **vertici**) e $E$ è un insieme di **spigoli** (o **archi**), cioè coppie non ordinate di elementi distinti di $V$:
$$
E\subseteq\{\{u,v\} : u,v\in V,\ u\neq v\}.
$$
Per semplicità si scrive $uv$ per $\{u,v\}$; poiché l'ordine non conta, $uv=vu$.

> **Osservazione 1.2.**  
> La finitezza di $V$ ed $E$ è essenziale per gli algoritmi: un calcolatore può manipolare solo insiemi finiti. I grafi infiniti sono oggetti teorici e non rientrano nella trattazione algoritmica.

**Esempio 1.3.**
```mermaid
graph LR
    a((a))
    b((b))
    c((c))
    d((d))
    e((e))
    f((f))

    a --- b
    a --- c
    b --- d
    c --- d
    c --- e
    d --- e
    d --- f
    e --- f
```

Un grafo può essere rappresentato graficamente con $|V|$ punti nel piano e un segmento per ogni spigolo. La Figura mostra un grafo con $V=\{a,b,c,d,e,f\}$ e spigoli $\{ab,ac,bd,cd,ce,de,df,ef\}$.

### Definizione 1.4 — Grado e vicinato

Dato $v\in V$, si definisce
$$
\delta(v)=\{uv : uv\in E\}
$$
l'insieme degli spigoli incidenti in $v$. Il **grado** di $v$ è
$$
\deg(v)=|\delta(v)|.
$$
Due nodi $u,v$ sono **adiacenti** se $uv\in E$.

### Lemma 1.5 — Somma dei gradi

Per ogni grafo non orientato $G=(V,E)$,
$$
\sum_{v\in V}\deg(v)=2|E|.
$$

**Dimostrazione.**  
La somma $\sum_{v\in V}\deg(v)$ conta, per ogni nodo $v$, il numero di spigoli incidenti in $v$. Nell'effettuare questo conteggio, ogni spigolo $uv$ viene considerato due volte: una come elemento di $\delta(u)$ e una come elemento di $\delta(v)$. Quindi la somma è $2|E|$. $\square$

> **Corollario 1.6.**  
> In ogni grafo il numero di nodi di grado dispari è pari.

### Definizione 1.7 — Sottografo e sottografo indotto

Un grafo $G'=(V',E')$ è un **sottografo** di $G=(V,E)$ se $V'\subseteq V$ ed $E'\subseteq E$. È il **sottografo indotto** da $V'$ se
$$
E'=\{uv\in E : u,v\in V'\},
$$
e si indica con $G[V']$.

### Definizione 1.8 — Cammino

Un **cammino** in $G$ è una sequenza alternata di nodi e spigoli
$$
v_0,e_1,v_1,e_2,\dots,v_{k-1},e_k,v_k,
$$
dove $k\ge0$, i nodi $v_0,\dots,v_k$ sono distinti e $e_i=v_{i-1}v_i\in E$ per $i=1,\dots,k$. Si indica con $V(P)=\{v_0,\dots,v_k\}$ e $E(P)=\{e_1,\dots,e_k\}$. Gli **estremi** sono $v_0$ e $v_k$; la **lunghezza** è $|E(P)|=k$. Un cammino tra $u$ e $v$ ha estremi $u$ e $v$.

> **Osservazione 1.9.**  
> La condizione che i nodi siano distinti è essenziale: un cammino non può ripetere nodi. La sequenza di soli nodi $v_0,\dots,v_k$ determina univocamente il cammino, perché gli spigoli sono le coppie consecutive.

### Definizione 1.10 — Connessione

Due nodi $u,v\in V$ sono **connessi** se esiste un cammino in $G$ con estremi $u$ e $v$. La relazione "essere connesso a" è denotata con $\sim$.

### Lemma 1.11 — La connessione è una relazione di equivalenza

La relazione $\sim$ su $V$ è riflessiva, simmetrica e transitiva.

**Dimostrazione.**  
- **Riflessività:** il cammino di lunghezza $0$ con un solo nodo $v$ connette $v$ a sé stesso.
- **Simmetria:** se $P=v_0,\dots,v_k$ è un cammino da $u$ a $v$, allora la sequenza inversa $v_k,\dots,v_0$ è un cammino da $v$ a $u$.
- **Transitività:** siano $P$ un cammino da $u$ a $v$ e $Q$ un cammino da $v$ a $w$. Sia $z$ il primo nodo di $Q$ che si incontra percorrendo $P$ (esiste, perché $P$ e $Q$ hanno in comune almeno $v$). Allora la concatenazione di $P_{uz}$ (sottocammino di $P$ da $u$ a $z$) e $Q_{zw}$ (sottocammino di $Q$ da $z$ a $w$) è un cammino da $u$ a $w$, perché i nodi sono distinti per la scelta di $z$ e gli spigoli appartengono a $E(P)\cup E(Q)\subseteq E$. $\square$

### Definizione 1.12 — Componenti connesse

Le classi di equivalenza di $\sim$ inducono i **sottografi connessi massimali**, detti **componenti connesse** di $G$. Un grafo è **connesso** se ha una sola componente connessa.

### Definizione 1.13 — Ciclo

Un **ciclo** in $G$ è una sequenza alternata di nodi e spigoli
$$
v_0,e_1,v_1,e_2,\dots,v_{k-1},e_k,v_k,
$$
con $k\ge3$, $v_0,\dots,v_{k-1}$ distinti, $e_i=v_{i-1}v_i\in E$ e $v_k=v_0$. La **lunghezza** è $k$. Un grafo **aciclico** non contiene cicli.

### Definizione 1.14 — Albero

Un **albero** è un grafo connesso e aciclico. Un nodo di grado $1$ in un albero è detto **foglia**.

### Lemma 1.15 — Ogni albero con almeno due nodi ha almeno due foglie

Sia $G=(V,E)$ un albero con $|V|\ge2$. Allora $G$ ha almeno due foglie.

**Dimostrazione.**  
Si scelga un cammino $P$ massimale rispetto all'inclusione (cioè non contenuto in nessun altro cammino), ottenuto partendo da un cammino qualsiasi e aggiungendo spigoli finché possibile. Sia $P=v_0,\dots,v_k$. Poiché $G$ è connesso e ha almeno due nodi, $k\ge1$. Mostriamo che $v_0$ è una foglia. Sia $u$ un vicino di $v_0$. Se $u\notin V(P)$, allora $u,v_0,\dots,v_k$ sarebbe un cammino più lungo, contro la massimalità. Quindi $u\in V(P)$. Se $u\neq v_1$, allora $u=v_i$ per qualche $i\ge2$, e $v_0,v_1,\dots,v_i,v_0$ sarebbe un ciclo, contro l'aciclicità. Quindi $u=v_1$, cioè $\deg(v_0)=1$. Per simmetria, anche $v_k$ è una foglia. $\square$

### Teorema 1.16 — Caratterizzazioni degli alberi

Sia $G=(V,E)$ un grafo non orientato. Le seguenti affermazioni sono equivalenti:

1. $G$ è un albero;
2. per ogni $u,v\in V$, esiste esattamente un cammino tra $u$ e $v$;
3. $G$ è aciclico e $|E|=|V|-1$;
4. $G$ è connesso e $|E|=|V|-1$.

**Dimostrazione.**

**(1) $\Rightarrow$ (2).**  
$G$ è connesso, quindi esiste almeno un cammino tra ogni coppia. Supponiamo per assurdo che esistano due cammini distinti $P$ e $Q$ tra $u$ e $v$. Sia $w$ l'ultimo nodo comune a $P$ e $Q$ prima che i cammini si separino, e sia $z$ il primo nodo di $Q$ che si incontra percorrendo $P$ dopo $w$. Allora $P_{wz}$ e $Q_{wz}$ non hanno nodi interni in comune e formano un ciclo, contro l'aciclicità. Quindi il cammino è unico.

**(2) $\Rightarrow$ (1).**  
La connessione è ovvia. Se ci fosse un ciclo, ci sarebbero due cammini distinti tra due nodi del ciclo, contro l'unicità. Quindi $G$ è aciclico.

**(1) $\Rightarrow$ (3) e (1) $\Rightarrow$ (4).**  
Dimostriamo per induzione su $|V|$ che se $G$ è un albero allora $|E|=|V|-1$. Per $|V|=1$ è ovvio. Per $|V|\ge2$, per il Lemma 1.15 esiste una foglia $v$; sia $u$ il suo unico vicino. Il sottografo indotto $G'=G[V\setminus\{v\}]$ è ancora un albero (connesso e aciclico). Per ipotesi induttiva $|E'|=|V'|-1$. Poiché $|E|=|E'|+1$ e $|V|=|V'|+1$, si ha $|E|=|V|-1$. Inoltre $G$ è connesso per definizione, quindi valgono (3) e (4).

**(3) $\Rightarrow$ (1).**  
Sia $k$ il numero di componenti connesse $G_1,\dots,G_k$. Ogni $G_i$ è connesso e aciclico, quindi è un albero, e per l'implicazione già dimostrata $|E_i|=|V_i|-1$. Sommando,
$$
|E|=\sum_{i=1}^k|E_i|=\sum_{i=1}^k(|V_i|-1)=|V|-k.
$$
Per ipotesi $|E|=|V|-1$, quindi $k=1$, cioè $G$ è connesso. Dunque $G$ è un albero.

**(4) $\Rightarrow$ (1).**  
Dimostriamo per induzione su $|V|$. Per $|V|=1$ è ovvio. Per $|V|\ge2$, mostriamo che esiste un nodo di grado $1$: se tutti i nodi avessero grado $\ge2$, allora per il Lemma 1.5
$$
2|E|=\sum_{v\in V}\deg(v)\ge2|V|,
$$
cioè $|E|\ge|V|$, contro l'ipotesi $|E|=|V|-1$. Quindi esiste $v$ con $\deg(v)\le1$. Poiché $G$ è connesso, $\deg(v)=1$. Sia $u$ il suo unico vicino. Il sottografo $G'=G[V\setminus\{v\}]$ è connesso e ha $|E'|=|E|-1=|V|-2=|V'|-1$. Per induzione $G'$ è aciclico. Poiché $v$ ha grado $1$, nessun ciclo può contenere $v$; quindi $G$ è aciclico. Dunque $G$ è un albero. $\square$

> **Corollario 1.17.**  
> Le condizioni (3) e (4) possono essere indebolite:
> - (3') $G$ è aciclico e $|E|\ge|V|-1$;
> - (4') $G$ è connesso e $|E|\le|V|-1$.
>
> **Dimostrazione.**  
> Se $G$ è aciclico con $|E|\ge|V|-1$, allora ragionando come in (3) $\Rightarrow$ (1) si ha $|E|=|V|-k$ con $k$ numero di componenti. Poiché $|E|\ge|V|-1$, si ottiene $k\le1$, quindi $k=1$ e $|E|=|V|-1$. Analogamente per (4'). $\square$

## 2. Grafi orientati

### Definizione 2.1 — Grafo orientato

Un **grafo orientato** (o **digrafo**) è una coppia ordinata $D=(V,A)$, dove $V$ è un insieme finito di nodi e $A$ è un insieme di **archi** orientati:
$$
A\subseteq\{(u,v)\in V\times V : u\neq v\}.
$$
Per un arco $e=(u,v)$, $u$ è la **coda** e $v$ la **testa**; si dice che $e$ esce da $u$ ed entra in $v$.

### Definizione 2.2 — Intorni orientati

Per ogni $v\in V$,
$$
\delta^+(v)=\{(v,u)\in A\},\qquad
\delta^-(v)=\{(u,v)\in A\}.
$$
Vale la relazione
$$
|A|=\sum_{v\in V}|\delta^-(v)|=\sum_{v\in V}|\delta^+(v)|.
$$

### Definizione 2.3 — Cammino orientato

Un **cammino orientato** in $D$ è una sequenza
$$
v_0,e_1,v_1,\dots,e_k,v_k
$$
con $k\ge0$, nodi distinti, ed $e_i=(v_{i-1},v_i)\in A$. Si dice cammino **da $v_0$ a $v_k$**. La **lunghezza** è $k$.

### Lemma 2.4 — La raggiungibilità è riflessiva e transitiva

La relazione su $V$ definita da "$u$ è raggiungibile da $v$ se esiste un cammino orientato da $v$ a $u$" è riflessiva e transitiva.

**Dimostrazione.**  
Riflessività: cammino di lunghezza $0$. Transitività: concatenazione di cammini orientati. $\square$

> **Osservazione 2.5.**  
> A differenza del caso non orientato, la relazione non è simmetrica in generale. Un grafo orientato può avere $u$ raggiungibile da $v$ senza che $v$ sia raggiungibile da $u$.

### Definizione 2.6 — Ciclo orientato

Un **ciclo orientato** è una sequenza
$$
v_0,e_1,v_1,\dots,e_k,v_k
$$
con $k\ge2$, nodi $v_0,\dots,v_{k-1}$ distinti, $e_i=(v_{i-1},v_i)\in A$ e $v_k=v_0$. La **lunghezza** è $k$.

### Definizione 2.7 — Grafo sottostante

Il **grafo non orientato sottostante** a $D=(V,A)$ è $G=(V,E)$ con
$$
E=\{uv : (u,v)\in A\ \text{oppure}\ (v,u)\in A\}.
$$
Vale $|E|\le|A|$, con uguaglianza se e solo se $D$ non contiene coppie di archi opposti $(u,v)$ e $(v,u)$.

### Definizione 2.8 — Connessione e forte connessione

$D$ è **connesso** se il grafo sottostante è connesso. È **fortemente connesso** se per ogni $u,v\in V$ esistono cammini orientati da $u$ a $v$ e da $v$ a $u$.

> **Osservazione 2.9.**  
> La forte connessione implica la connessione, ma non viceversa. Un grafo orientato aciclico è connesso se il sottostante lo è, ma non è mai fortemente connesso (a meno di un solo nodo), perché l'aciclicità impedisce l'esistenza di cammini di andata e ritorno tra nodi distinti.

### Definizione 2.10 — Arborescenza

Dato $r\in V$, un grafo orientato $D=(V,A)$ è un'**arborescenza con radice $r$** se
$$
|A|=|V|-1
$$
e per ogni $v\in V$ esiste un cammino orientato da $r$ a $v$.

### Lemma 2.11 — Caratterizzazione delle arborescenze

Sia $D=(V,A)$ un grafo orientato e $r\in V$. Allora $D$ è un'arborescenza con radice $r$ se e solo se il grafo sottostante è un albero, $|\delta^-(r)|=0$ e $|\delta^-(v)|=1$ per ogni $v\neq r$.

**Dimostrazione.**

**( $\Leftarrow$ )**  
Supponiamo che il grafo sottostante $G$ sia un albero, $|\delta^-(r)|=0$ e $|\delta^-(v)|=1$ per $v\neq r$. Allora
$$
|A|=\sum_{v\in V}|\delta^-(v)|=|V|-1.
$$
Resta da mostrare che ogni $v$ è raggiungibile da $r$. Fissato $v_0\neq r$, poiché $|\delta^-(v_0)|=1$ esiste $v_1$ con $(v_1,v_0)\in A$. Se $v_1=r$, abbiamo il cammino. Altrimenti $|\delta^-(v_1)|=1$, esiste $v_2$ con $(v_2,v_1)\in A$, e così via. La sequenza $v_0,v_1,v_2,\dots$ è fatta di nodi distinti: se $v_i=v_j$ con $i>j$, si formerebbe un ciclo orientato, contro l'ipotesi che $G$ sia un albero. Poiché $V$ è finito, la sequenza termina in $r$. Il cammino inverso è un cammino orientato da $r$ a $v_0$.

**( $\Rightarrow$ )**  
Supponiamo che $D$ sia un'arborescenza con radice $r$. Allora per ogni $v$ esiste un cammino orientato da $r$ a $v$, quindi il grafo sottostante è connesso. Inoltre $|E|\le|A|=|V|-1$. Per il Corollario 1.17, $G$ è un albero. Resta da mostrare le condizioni sui gradi entranti. Poiché ogni $v\neq r$ è raggiungibile da $r$, deve essere $|\delta^-(v)|\ge1$. Allora
$$
|V|-1=|A|=\sum_{v\in V}|\delta^-(v)|=|\delta^-(r)|+\sum_{v\neq r}|\delta^-(v)|\ge|V|-1.
$$
La disuguaglianza è quindi un'uguaglianza, da cui $|\delta^-(r)|=0$ e $|\delta^-(v)|=1$ per $v\neq r$. $\square$

## 3. Complessità degli algoritmi

### Definizione 3.1 — Problema e istanza

Un **problema** è un quesito espresso in termini di parametri (dati). Un'**istanza** è un problema con i parametri fissati a valori specifici. Un problema è quindi l'insieme delle sue istanze.

**Esempio 3.2.**  
Il problema della Programmazione Lineare in forma standard: dati $m,n\in\mathbb{N}$, $A\in\mathbb{Q}^{m\times n}$, $b\in\mathbb{Q}^m$, $c\in\mathbb{Q}^n$, trovare $x\in\mathbb{Q}^n$ che risolva
$$
\max\{c^Tx : Ax=b,\ x\ge0\}.
$$
Un'istanza è una specifica quintupla $(m,n,A,b,c)$.

> **Osservazione 3.3.**  
> Si assume che i dati siano numeri razionali, perché un calcolatore può rappresentare solo numeri con espansione finita. I numeri irrazionali non hanno rappresentazione finita e non possono essere input di un algoritmo.

### Definizione 3.4 — Codifica e taglia

Un'istanza $I$ è codificata in una stringa binaria. La **taglia** $\operatorname{size}(I)$ è il numero di bit necessari a codificare tutti i dati dell'istanza. Per un intero $a$ servono $1+\lceil\log_2(|a|+1)\rceil$ bit; per un razionale $p/q$ servono $1+\lceil\log_2(|p|+1)\rceil+\lceil\log_2(|q|+1)\rceil$ bit.

### Definizione 3.5 — Tempo di esecuzione

Dati un problema $P$ e un algoritmo $A$ per $P$, si indica con $\operatorname{time}(A,I)$ il numero di **operazioni elementari** (addizione, sottrazione, prodotto, divisione, confronto) eseguite da $A$ sull'istanza $I$.

### Definizione 3.6 — Algoritmo polinomiale

$A$ è **polinomiale** per $P$ se esistono costanti $c,d>0$ tali che
$$
\operatorname{time}(A,I)\le c\cdot\operatorname{size}(I)^d
\qquad\forall I\in P.
$$
Si scrive $\operatorname{time}(A,I)=O(\operatorname{size}(I)^d)$.

> **Osservazione 3.7.**  
> L'interesse per gli algoritmi polinomiali è che i polinomi crescono molto più lentamente delle funzioni esponenziali. Un algoritmo esponenziale, come l'enumerazione di tutte le soluzioni ammissibili, diventa impraticabile già per dimensioni moderate. La Tabella 5.1 del materiale mostra che $1.1^n$ supera $n^{100}$ per $n$ sufficientemente grande.

### Esempio 3.8 — Taglia di un'istanza di PL

Per un'istanza di PL con $A\in\mathbb{Q}^{m\times n}$, $b\in\mathbb{Q}^m$, $c\in\mathbb{Q}^n$, sia $T$ il numero massimo di bit per una componente. Allora
$$
\operatorname{size}(I)\le\operatorname{size}(m)+\operatorname{size}(n)+(mn+m+n)T.
$$
Poiché $\operatorname{size}(m)\le m+1$ e $\operatorname{size}(n)\le n+1$, ogni polinomio in $\operatorname{size}(I)$ è maggiorato da un polinomio in $m,n,T$, e viceversa. Quindi la polinomialità di un algoritmo per la PL può essere verificata stimando il tempo come polinomio in $m,n,T$.

### Esempio 3.9 — Taglia di un'istanza di grafo

Per un grafo $G$ con $n$ nodi e $m$ spigoli, codificato con $n$ e la lista degli spigoli,
$$
\operatorname{size}(n)+2m\le\operatorname{size}(I)\le\operatorname{size}(n)+2m\operatorname{size}(n).
$$
Poiché $n/2\le m\le n(n-1)/2$, la polinomialità può essere verificata stimando il tempo come polinomio in $n$.

### Esempio 3.10 — Ciclo hamiltoniano

Il problema di decidere se un grafo contiene un ciclo hamiltoniano (un ciclo che attraversa tutti i nodi) non ha algoritmi polinomiali noti. L'algoritmo banale enumera tutte le $n!$ permutazioni dei nodi, con tempo $\Omega(n!)$, che è più che esponenziale. Per $n=30$, $30!\approx2.65\times10^{32}$, e anche a un picosecondo per permutazione servirebbero oltre $8400$ miliardi di anni. La congettura prevalente è che non esista un algoritmo polinomiale per questo problema.

## 4. Complessità della Programmazione Lineare

### Osservazione 4.1 — Iterazione del simplesso

Una singola iterazione del metodo del simplesso può essere eseguita in tempo polinomiale nella taglia dell'istanza.

**Dimostrazione.**  
Per scegliere la variabile entrante, si scorre il vettore dei costi ridotti fino a trovare la prima componente positiva (o si conclude che l'ottimo è raggiunto): al più $n$ confronti. Per scegliere la variabile uscente (o concludere che il problema è illimitato), si calcolano e si confrontano al più $m$ rapporti. Il pivot richiede il calcolo di al più $(m+1)(n+1)$ componenti del tableau, ciascuna con un numero costante di operazioni elementari. Quindi il numero di operazioni è polinomiale in $m,n$ e nella taglia dei dati. $\square$

> **Osservazione 4.2.**  
> Se il simplesso eseguisse un numero di iterazioni polinomiale nella taglia, allora sarebbe un algoritmo polinomiale. Tuttavia, per tutte le regole di pivot proposte sono stati costruiti esempi in cui il numero di iterazioni è esponenziale. Il primo esempio è di Klee e Minty (1972) per la regola di Bland; altri esempi sono stati trovati successivamente. Non è noto se esista una regola di pivot che garantisca la polinomialità; la questione è aperta.

> **Osservazione 4.3.**  
> Nonostante il simplesso non sia polinomiale nel caso peggiore, il suo comportamento nella pratica è molto buono, e i solver commerciali lo usano insieme ai metodi di punto interno.

### Osservazione 4.4 — Algoritmi polinomiali per la PL

Nel 1979 Khachiyan ha proposto il **metodo dell'ellissoide**, il primo algoritmo polinomiale per la PL. Tuttavia, non è efficiente in pratica. I **metodi di punto interno**, sviluppati negli anni '90, sono polinomiali sia in teoria sia efficienti in pratica. I solver commerciali combinano simplesso (primale o duale) e metodi di punto interno, scegliendo in base al problema.
