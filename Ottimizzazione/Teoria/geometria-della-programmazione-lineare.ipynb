## 1. Insiemi convessi e poliedri

### Definizione 1.1 — Semispazio chiuso

Un **semispazio chiuso** in $\mathbb{R}^n$ è un insieme della forma

$$
H = \{x \in \mathbb{R}^n : a^T x \leq \beta\},
$$

dove $a \in \mathbb{R}^n \setminus \{0\}$ e $\beta \in \mathbb{R}$.

Scritto per esteso, un semispazio chiuso è l'insieme dei punti di $\mathbb{R}^n$ che soddisfano una disequazione del tipo

$$
a_1 x_1 + \dots + a_n x_n \leq \beta.
$$

Il vettore $a$ è detto **vettore normale** al semispazio, e la retta (o piano) $a^T x = \beta$ è detta **frontiera** del semispazio.

### Definizione 1.2 — Poliedro

Un **poliedro** in $\mathbb{R}^n$ è l'intersezione di un numero finito di semispazi chiusi. In altre parole, un poliedro è un insieme che può essere scritto nella forma

$$
P = \{x \in \mathbb{R}^n : A x \leq b\},
$$

dove $A \in \mathbb{R}^{m \times n}$ e $b \in \mathbb{R}^m$.

> **Osservazione 1.3.**  
> La regione ammissibile di un qualunque problema di programmazione lineare è un poliedro. Eventuali vincoli scritti sotto forma di equazione possono essere sostituiti ciascuno con una coppia di disequazioni opposte:
> $$
> a_i^T x = b_i
> \iff
> \begin{cases}
> a_i^T x \leq b_i,\\
> -a_i^T x \leq -b_i.
> \end{cases}
> $$

### Definizione 1.4 — Insieme convesso

Un insieme $K \subseteq \mathbb{R}^n$ si dice **convesso** se per ogni coppia di punti $y, z \in K$ il segmento di retta che congiunge $y$ e $z$ è contenuto in $K$, cioè se

$$
\lambda y + (1 - \lambda) z \in K
\qquad
\forall y, z \in K,\ \forall \lambda \in [0,1].
$$

### Definizione 1.5 — Combinazione convessa

Un'espressione del tipo $\lambda y + (1 - \lambda) z$ con $\lambda \in [0,1]$ è detta **combinazione convessa** di $y$ e $z$.

Un punto $x \in \mathbb{R}^n$ è detto **combinazione convessa propria** dei punti $y, z \in \mathbb{R}^n$ se è combinazione convessa di $y$ e $z$ e inoltre $x \neq y$ e $x \neq z$. In tal caso si ha necessariamente

$$
x = \lambda y + (1 - \lambda) z
\qquad \text{con } \lambda \in (0,1).
$$

### Proposizione 1.6 — Proprietà degli insiemi convessi

Valgono le seguenti proprietà:

1. Ogni semispazio chiuso è convesso.
2. L'intersezione di insiemi convessi è un insieme convesso.
3. Ogni poliedro è convesso.

**Dimostrazione.**

**(i)** Sia $S$ il semispazio chiuso definito dalla disequazione $a^T x \leq \beta$. Scegliamo arbitrariamente $y, z \in S$ e $\lambda \in [0,1]$. Allora

$$
a^T(\lambda y + (1 - \lambda) z)
=
\lambda a^T y + (1 - \lambda) a^T z
\leq
\lambda \beta + (1 - \lambda) \beta
=
\beta,
$$

dove la disuguaglianza vale perché $y, z \in S$, $\lambda \geq 0$ e $1 - \lambda \geq 0$. Dunque $\lambda y + (1 - \lambda) z \in S$, da cui segue che $S$ è convesso.

**(ii)** Per semplicità di scrittura diamo la dimostrazione per il caso di due insiemi convessi; l'argomento per più insiemi convessi è identico (anche nel caso in cui questi siano in numero infinito). Siano $K_1$ e $K_2$ due insiemi convessi e definiamo $K = K_1 \cap K_2$. Scegliamo arbitrariamente $y, z \in K$ e $\lambda \in [0,1]$. Allora il punto $\lambda y + (1 - \lambda) z$ appartiene sia a $K_1$ sia a $K_2$ (in quanto questi due insiemi sono convessi) e dunque a $K$. Ne segue che $K$ è convesso.

**(iii)** Poiché un poliedro è l'intersezione di un numero finito di semispazi chiusi, il risultato segue dalle proprietà (i) e (ii). $\square$

## 2. Punti estremi, punti esposti, soluzioni di base, vertici

### Definizione 2.1 — Punto estremo

Dato un insieme convesso $K \subseteq \mathbb{R}^n$, un punto $\bar{x} \in K$ è un **punto estremo** di $K$ se $\bar{x}$ non è combinazione convessa propria di due punti di $K$, cioè se non esistono due punti $y, z \in K \setminus \{\bar{x}\}$ e $\lambda \in (0,1)$ tali che

$$
\bar{x} = \lambda y + (1 - \lambda) z.
$$

### Definizione 2.2 — Punto esposto

Dato un insieme convesso $K \subseteq \mathbb{R}^n$, un punto $\bar{x} \in K$ è un **punto esposto** di $K$ se esiste $c \in \mathbb{R}^n$ tale che

$$
c^T x < c^T \bar{x}
\qquad \forall x \in K \setminus \{\bar{x}\}.
$$

In altre parole, $\bar{x}$ è un punto esposto di $K$ se esiste una funzione obiettivo lineare per cui $\bar{x}$ è l'unico punto ottimale di $K$.

> **Osservazione 2.3.**  
> Ogni punto esposto di un insieme convesso $K$ è un punto estremo di $K$, ma il viceversa non è vero in generale. **coincidono nel caso in cui $K$ sia un poliedro**.

### Definizione 2.4 — Vincoli linearmente indipendenti e vincoli attivi

Consideriamo un generico sistema di equazioni e/o disequazioni lineari in $\mathbb{R}^n$:

$$
\begin{cases}
a_i^T x = b_i, & i = 1,\dots,k,\\
a_i^T x \leq b_i, & i = k+1,\dots,m.
\end{cases}
$$

I vincoli sono detti **linearmente indipendenti** se i vettori $\{a_i\}_{i=1,\dots,m}$ sono linearmente indipendenti.

Dati un punto $\bar{x} \in \mathbb{R}^n$ e un indice $i \in \{1,\dots,m\}$, l'$i$-esimo vincolo è **attivo** in $\bar{x}$ se

$$
a_i^T \bar{x} = b_i.
$$

### Definizione 2.5 — Soluzione di base

Un punto $\bar{x} \in \mathbb{R}^n$ è detto **soluzione di base** del sistema di vincoli se:

1. $a_i^T \bar{x} = b_i$ per $i = 1,\dots,k$;
2. esistono $n$ vincoli del sistema attivi in $\bar{x}$ che sono linearmente indipendenti.

> **Osservazione 2.6.**  
> Non è richiesto che una soluzione di base soddisfi anche tutti i vincoli di disuguaglianza. Quando questo avviene, la soluzione di base è detta **soluzione di base ammissibile**.

### Teorema 2.7 — Caratterizzazione dei vertici di un poliedro

Sia $P \subseteq \mathbb{R}^n$ il poliedro definito dal generico sistema lineare. Dato un punto $\bar{x} \in P$, le seguenti affermazioni sono equivalenti:

1. $\bar{x}$ è un punto estremo di $P$;
2. $\bar{x}$ è un punto esposto di $P$;
3. $\bar{x}$ è una soluzione di base ammissibile del sistema.

**Dimostrazione.**

**(i) $\Rightarrow$ (iii)** Supponiamo che $\bar{x}$ sia un punto estremo di $P$. Poiché $\bar{x} \in P$, $\bar{x}$ soddisfa tutti i vincoli del sistema. Per dimostrare che $\bar{x}$ è una soluzione di base ammissibile, basterà dimostrare che esistono $n$ vincoli del sistema attivi in $\bar{x}$ che sono linearmente indipendenti.

Scriviamo il sistema nella forma matriciale $A x \sim b$, dove $A \in \mathbb{R}^{m \times n}$, $b \in \mathbb{R}^m$ e la notazione "$\sim$" indica che su ciascuna riga del sistema appare il simbolo "$=$" oppure il simbolo "$\leq$". Sia $A'$ la sottomatrice di $A$ formata dalle righe che corrispondono ai vincoli del sistema che sono attivi in $\bar{x}$; sia $A''$ la sottomatrice di $A$ formata dalle righe rimanenti. Definiamo $b'$ e $b''$ come i sottovettori di $b$ corrispondenti ad $A'$ e $A''$ rispettivamente. Allora valgono le condizioni

$$
A' \bar{x} = b', \qquad A'' \bar{x} < b''.
$$

Dire che esistono $n$ vincoli del sistema attivi in $\bar{x}$ che sono linearmente indipendenti equivale a dire che $\operatorname{rk}(A') = n$. Supponiamo per contraddizione che $\operatorname{rk}(A') < n$. Allora esiste un vettore non nullo $v \in \mathbb{R}^n$ tale che $A' v = 0$. Scelto uno scalare $\varepsilon > 0$ e definiti i vettori $y = \bar{x} + \varepsilon v$ e $z = \bar{x} - \varepsilon v$, si ha

$$
A' y = A' \bar{x} + \varepsilon A' v = A' \bar{x} + 0 = b',
\qquad
A' z = A' \bar{x} - \varepsilon A' v = A' \bar{x} - 0 = b'.
$$

Inoltre, se $\varepsilon$ è scelto sufficientemente piccolo, si ha

$$
A'' y = A'' \bar{x} + \varepsilon A'' v \leq b'',
\qquad
A'' z = A'' \bar{x} - \varepsilon A'' v \leq b''.
$$

Dunque $y$ e $z$ soddisfano tutti i vincoli del sistema e sono quindi punti di $P$. Notiamo che

$$
\bar{x} = \tfrac{1}{2} y + \tfrac{1}{2} z, \qquad \bar{x} \neq y, \qquad \bar{x} \neq z.
$$

Allora $\bar{x}$ è combinazione convessa propria di $y$ e $z$, in contraddizione con il fatto che $\bar{x}$ è un punto estremo di $P$.

**(iii) $\Rightarrow$ (ii)** Supponiamo che $\bar{x}$ sia una soluzione di base ammissibile. Definiamo $A', A'', b', b''$ come sopra. Possiamo assumere senza perdita di generalità che, per un qualche intero $p \geq 0$, le righe di $A'$ siano le prime $p$ righe di $A$, cioè i vettori $a_1^T, \dots, a_p^T$, e che $b' = (b_1,\dots,b_p)^T$.

Scegliamo

$$
c = a_1 + \dots + a_p.
$$

Dimostreremo che $c^T x < c^T \bar{x}$ per ogni $x \in P \setminus \{\bar{x}\}$, da cui $\bar{x}$ è un punto esposto di $P$.

Per ogni $x \in P$ si ha

$$
c^T x = a_1^T x + \dots + a_p^T x \leq b_1 + \dots + b_p = a_1^T \bar{x} + \dots + a_p^T \bar{x} = c^T \bar{x}.
$$

Dunque $c^T x \leq c^T \bar{x}$ per ogni $x \in P$, e l'uguaglianza vale se e solo se $a_i^T x = b_i$ per $i = 1,\dots,p$, cioè se e solo se $A' x = b'$. Poiché $\bar{x}$ è una soluzione di base, la matrice $A'$ ha rango $n$, dunque il sistema $A' x = b'$ ha $\bar{x}$ come unica soluzione. Pertanto $c^T x < c^T \bar{x}$ per ogni punto $x \in P \setminus \{\bar{x}\}$.

**(ii) $\Rightarrow$ (i)** Se $\bar{x}$ è un punto esposto di $P$, esiste $c \in \mathbb{R}^n$ tale che $c^T x < c^T \bar{x}$ per ogni $x \in P \setminus \{\bar{x}\}$. Assumiamo per assurdo che $\bar{x}$ non sia un punto estremo di $P$. Allora esistono due punti $y, z \in P$ e uno scalare $\lambda \in (0,1)$ tali che $\bar{x} = \lambda y + (1 - \lambda) z$, con $\bar{x} \neq y$ e $\bar{x} \neq z$. Per la scelta di $c$, abbiamo $c^T y < c^T \bar{x}$ e $c^T z < c^T \bar{x}$. Ne segue che

$$
c^T \bar{x}
=
c^T(\lambda y + (1 - \lambda) z)
=
\lambda c^T y + (1 - \lambda) c^T z
<
\lambda c^T \bar{x} + (1 - \lambda) c^T \bar{x}
=
c^T \bar{x},
$$

dove la disuguaglianza vale perché $\lambda > 0$ e $1 - \lambda > 0$. Abbiamo dunque ottenuto la contraddizione $c^T \bar{x} < c^T \bar{x}$. $\square$

> **Osservazione 2.8.**  
> Per dimostrare l'implicazione (ii) $\Rightarrow$ (i) non abbiamo sfruttato il fatto che $P$ è un poliedro. Ne segue che **ogni punto esposto di un insieme convesso $K$ è un punto estremo di $K$**, mentre il viceversa vale solo per poliedri. Infatti, per mostrare l'implicazione inversa siamo passati attraverso il concetto di soluzione di base ammissibile, che ha senso solo per sistemi lineari e dunque solo per poliedri.

### Definizione 2.9 — Vertice

I punti estremi (cioè i punti esposti) di un poliedro sono anche detti **vertici**. Nel caso di poliedri, le nozioni di vertice, punto estremo e punto esposto coincidono.

> **Osservazione 2.10.**  
> Nel seguito vedremo che le soluzioni di base ammissibili giocano un ruolo fondamentale in programmazione lineare. Conviene sempre tenere presente l'interpretazione geometrica delle soluzioni di base ammissibili, cioè il fatto che esse corrispondono ai punti estremi (cioè ai punti esposti) della regione ammissibile.

## 3. Soluzioni di base per problemi in forma standard

Sia $P \subseteq \mathbb{R}^n$ un poliedro definito da un sistema lineare in forma standard

$$
\begin{cases}
A x = b,\\
x \geq 0,
\end{cases}
$$

con $A \in \mathbb{R}^{m \times n}$ e $b \in \mathbb{R}^m$. Nel seguito assumeremo $P \neq \emptyset$. In particolare, il sistema $A x = b$ ammette una soluzione, da cui segue che $\operatorname{rk}(A) = \operatorname{rk}(A \mid b)$.

Possiamo assumere senza perdita di generalità che le righe di $A$ siano linearmente indipendenti (cioè che $\operatorname{rk}(A) = m$): in caso contrario, se si avesse cioè $\operatorname{rk}(A) = m' < m$, potremmo tenere nel sistema $m'$ equazioni linearmente indipendenti e rimuovere le altre senza alterare l'insieme delle soluzioni ammissibili, riducendoci così ad avere un sistema $A' x = b'$ con $m'$ righe e $\operatorname{rk}(A') = m'$.

### Definizione 3.1 — Base di una matrice

Sia $A \in \mathbb{R}^{m \times n}$ una matrice di rango $m$. Un insieme $B \subseteq \{1,\dots,n\}$ è detto **base** di $A$ se $|B| = m$ e la matrice $A_B$ (sottomatrice di $A$ formata dalle colonne $A_j$ con $j \in B$) è invertibile.

Con riferimento a un sistema in forma standard, $B$ viene anche detta **base del sistema**.

### Proposizione 3.2 — Caratterizzazione delle soluzioni di base

Un punto $\bar{x} \in \mathbb{R}^n$ è una soluzione di base del sistema in forma standard se e solo se $A \bar{x} = b$ ed esiste una base $B$ di $A$ tale che $\bar{x}_j = 0$ per ogni $j \notin B$.

**Dimostrazione.**

**($\Rightarrow$)** Supponiamo che $\bar{x}$ sia una soluzione di base del sistema. Allora esistono $n$ vincoli attivi in $\bar{x}$ che sono linearmente indipendenti. Si noti che gli $m$ vincoli del sistema $A x = b$ sono necessariamente attivi in $\bar{x}$. Inoltre, poiché $\operatorname{rk}(A) = m$, tali vincoli sono linearmente indipendenti. Dunque esistono $n - m$ vincoli di non-negatività attivi in $\bar{x}$, diciamo le disuguaglianze $x_j \geq 0$ per $j \in N$, dove $N \subseteq \{1,\dots,n\}$ e $|N| = n - m$, tali che i vincoli

$$
\begin{cases}
a_i^T x = b_i, & i = 1,\dots,m,\\
x_j \geq 0, & j \in N
\end{cases}
$$

siano linearmente indipendenti. In altre parole, la matrice del sistema è invertibile. A meno di permutare le variabili, possiamo assumere $N = \{m+1,\dots,n\}$. Definiamo inoltre $B = \{1,\dots,m\}$. Allora la matrice del sistema è

$$
\begin{pmatrix}
A_B & A_N\\
0 & I
\end{pmatrix},
$$

dove $I$ denota la matrice identità $(n - m) \times (n - m)$. Poiché questa matrice è invertibile, anche $A_B$ è invertibile. Dunque $B$ è una base di $A$. Inoltre, siccome i vincoli $x_j \geq 0$ per $j \in N$ sono attivi in $\bar{x}$, si ha $\bar{x}_j = 0$ per $j \in N$ (cioè per $j \notin B$).

**($\Leftarrow$)** Viceversa, supponiamo che si abbia $A \bar{x} = b$ e che esista una base $B$ di $A$ tale che $\bar{x}_j = 0$ per ogni $j \notin B$. A meno di permutare le variabili, possiamo assumere $B = \{1,\dots,m\}$. Definiamo inoltre $N = \{m+1,\dots,n\}$. Poiché $B$ è una base di $A$, la sottomatrice $A_B$ è invertibile, dunque anche la matrice precedente è invertibile. Pertanto i vincoli sono linearmente indipendenti e sono attivi in $\bar{x}$, cioè $\bar{x}$ è una soluzione di base del sistema. $\square$


### Corollario 3.3 — Caratterizzazione tramite colonne linearmente indipendenti

Dati un sistema in forma standard e un punto $\bar{x} \in \mathbb{R}^n$, sia

$$
S = \{j \in \{1,\dots,n\} : \bar{x}_j \neq 0\}.
$$

Allora $\bar{x}$ è una soluzione di base del sistema se e solo se $A \bar{x} = b$ e i vettori $\{A_j\}_{j \in S}$ sono linearmente indipendenti.

**Dimostrazione.**

Se $\bar{x}$ è una soluzione di base del sistema, allora per la Proposizione 3.2 esiste una base $B$ di $A$ tale che $\{A_j\}_{j \in S} \subseteq \{A_j\}_{j \in B}$; inoltre, tali vettori sono linearmente indipendenti per definizione di base.

Viceversa, se i vettori $\{A_j\}_{j \in S}$ sono linearmente indipendenti, allora possiamo scegliere una base $B$ di $A$ tale che $S \subseteq B$. Dunque $\bar{x}_j = 0$ per ogni $j \notin B$. Per la Proposizione 3.2, $\bar{x}$ è una soluzione di base del sistema. $\square$

### Osservazione 3.4 — Unicità della soluzione di base relativa a una base

Data una base $B$ del sistema in forma standard, esiste un unico vettore $\bar{x}$ che soddisfa $A \bar{x} = b$ e $\bar{x}_j = 0$ per ogni $j \notin B$.

**Dimostrazione.**

A meno di permutare le variabili, possiamo assumere $B = \{1,\dots,m\}$. Sia inoltre $N = \{m+1,\dots,n\}$. Se $x \in \mathbb{R}^n$ è un vettore che soddisfa $A x = b$ e $x_j = 0$ per ogni $j \notin B$, allora

$$
b = A x = A_B x_B + A_N x_N = A_B x_B.
$$

Poiché $B$ è una base di $A$, la matrice $A_B$ è invertibile. Pertanto l'unica scelta possibile per $x_B$ è data da

$$
x_B = A_B^{-1} b.
$$

Ne segue che vi è un unico vettore $\bar{x}$ che soddisfa $A \bar{x} = b$ e $\bar{x}_j = 0$ per ogni $j \notin B$, precisamente il vettore le cui componenti sono definite da

$$
\bar{x}_B = A_B^{-1} b, \qquad \bar{x}_N = 0.
$$

$\square$

### Definizione 3.5 — Soluzione di base relativa a una base

Data una base $B$ del sistema in forma standard, la **soluzione di base relativa a $B$** è l'unico vettore $\bar{x}$ che soddisfa $A \bar{x} = b$ e $\bar{x}_j = 0$ per ogni $j \notin B$, cioè il vettore definito da

$$
\bar{x}_B = A_B^{-1} b, \qquad \bar{x}_N = 0.
$$

### Definizione 3.6 — Base ammissibile

Se $\bar{x}$ è una soluzione ammissibile del sistema in forma standard, cioè se

$$
\bar{x}_B = A_B^{-1} b \geq 0,
$$

allora $B$ è detta **base ammissibile** del sistema.

### Teorema 3.7 — Esistenza di una soluzione ottima di base

Se un programma lineare in forma standard

$$
\begin{aligned}
\max \quad & c^T x \\
\text{s.a} \quad & A x = b \\
& x \geq 0
\end{aligned}
$$

ha una soluzione ottima, allora ha una soluzione ottima che è una soluzione di base ammissibile.

**Dimostrazione.**

Tra tutte le soluzioni ottime del problema dato (ce n'è almeno una per ipotesi), ne scegliamo una che abbia il numero massimo di componenti nulle e la chiamiamo $\bar{x}$. Dimostreremo che $\bar{x}$ è una soluzione di base (ovviamente ammissibile).

Supponiamo che $\bar{x}$ non sia una soluzione di base e definiamo

$$
S = \{j : \bar{x}_j > 0\} = \{j : \bar{x}_j \neq 0\}.
$$

Per il Corollario 3.3, i vettori $\{A_j\}_{j \in S}$ sono linearmente dipendenti. Dunque possiamo ottenere il vettore nullo come combinazione lineare di tali vettori, dove i coefficienti della combinazione non sono tutti nulli: in altre parole, esiste $v \in \mathbb{R}^S$ tale che $v \neq 0$ e $A_S v = 0$. Possiamo assumere senza perdita di generalità che $v$ abbia almeno una componente positiva (in caso contrario basta rimpiazzare $v$ con $-v$). Estendiamo $v$ ad un vettore $\bar{v} \in \mathbb{R}^n$ come segue:

$$
\bar{v}_j =
\begin{cases}
v_j & \text{se } j \in S,\\
0 & \text{se } j \notin S.
\end{cases}
$$

Si noti che $A \bar{v} = 0$ e $\bar{v}$ ha almeno una componente positiva.

Per ogni $\varepsilon \geq 0$ si ha

$$
A(\bar{x} \pm \varepsilon \bar{v}) = A \bar{x} \pm \varepsilon A \bar{v} = b.
$$

Inoltre, poiché $\bar{v}_j = 0$ per ogni $j \notin S$, si ha $\bar{v}_j = 0$ per ogni $j$ tale che $\bar{x}_j = 0$. Allora, se $\varepsilon > 0$ è scelto sufficientemente piccolo, entrambi i punti $\bar{x} + \varepsilon \bar{v}$ e $\bar{x} - \varepsilon \bar{v}$ sono non-negativi e sono dunque soluzioni ammissibili. E poiché $\bar{x}$ è una soluzione ottima, si ha

$$
c^T \bar{x} \geq c^T(\bar{x} + \varepsilon \bar{v}) = c^T \bar{x} + \varepsilon c^T \bar{v},
\qquad
c^T \bar{x} \geq c^T(\bar{x} - \varepsilon \bar{v}) = c^T \bar{x} - \varepsilon c^T \bar{v},
$$

da cui segue che $c^T \bar{v} = 0$.

Consideriamo un punto della forma $\bar{x} - \varepsilon \bar{v}$ con $\varepsilon \geq 0$. Poiché $c^T \bar{v} = 0$, il punto $\bar{x} - \varepsilon \bar{v}$ è una soluzione ottima per ogni $\varepsilon \geq 0$ tale che il punto stesso sia ammissibile.

Ci chiediamo ora quali siano i valori $\varepsilon \geq 0$ per cui il punto $\bar{x} - \varepsilon \bar{v}$ è ammissibile. Questo avviene se e solo se $\bar{x}_j - \varepsilon \bar{v}_j \geq 0$ per ogni $j \in \{1,\dots,n\}$. Si noti che se $j$ è un indice tale che $\bar{v}_j \leq 0$, allora, poiché $\bar{x}_j \geq 0$, si avrà $\bar{x}_j - \varepsilon \bar{v}_j \geq 0$ per ogni $\varepsilon \geq 0$. Dunque il punto $\bar{x} - \varepsilon \bar{v}$ è ammissibile se e solo se

$$
\varepsilon \leq \frac{\bar{x}_j}{\bar{v}_j}
\qquad
\forall j \in \{1,\dots,n\} \text{ tale che } \bar{v}_j > 0.
$$

Sia $h$ un indice tale che $\bar{v}_h > 0$ per cui il valore $\bar{x}_h / \bar{v}_h$ sia il più piccolo possibile. Come assunto in precedenza, il vettore $\bar{v}$ ha almeno una componente positiva, dunque $h$ è ben definito. Poniamo ora

$$
\varepsilon = \frac{\bar{x}_h}{\bar{v}_h},
\qquad
x^* = \bar{x} - \varepsilon \bar{v}.
$$

Poiché $\varepsilon$ soddisfa la condizione precedente, il punto $x^*$ è ammissibile. Inoltre

$$
x^*_h = \bar{x}_h - \varepsilon \bar{v}_h = 0,
$$

e per ogni $j \notin S$ si ha $x^*_j = \bar{x}_j - \varepsilon \bar{v}_j = 0 - 0 = 0$. Dunque $x^*$ è una soluzione ottima con almeno una componente nulla in più rispetto a $\bar{x}$, il che contraddice la scelta di $\bar{x}$. $\square$

> **Osservazione 3.8.**  
> Il teorema precedente chiarisce perché la programmazione lineare sia considerata parte dell'ottimizzazione discreta nonostante la regione ammissibile di un programma lineare sia una regione continua: il fatto che, quando esiste una soluzione ottima, ce ne sia una che è una soluzione di base ammissibile permette di limitare la ricerca dell'ottimo alle sole soluzioni di base ammissibili, che sono in numero finito (come segue dalla definizione di base). Dunque la programmazione lineare può essere vista come un problema di ottimizzazione discreta. questa osservazione è alla base del **metodo del simplesso**.

### Corollario 3.9 — Esistenza di una soluzione di base ammissibile

Se un sistema in forma standard ha una soluzione ammissibile, allora ha una soluzione di base ammissibile.

**Dimostrazione.**

Un sistema in forma standard ha la forma $A x = b$, $x \geq 0$. Scegliendo $c = 0$ come vettore dei coefficienti della funzione obiettivo, ogni soluzione ammissibile del problema di massimizzazione è una soluzione ottima (e viceversa). Il Teorema 3.7 mostra che il programma lineare così ottenuto ha una soluzione ottima che è una soluzione di base ammissibile; nel nostro caso, possiamo parafrasare dicendo che il problema ha una soluzione ammissibile che è una soluzione di base ammissibile. $\square$

> **Osservazione 3.10.**  
> Mentre abbiamo visto che ogni sistema in forma standard che ammetta soluzione ha una soluzione di base ammissibile, questo non è sempre vero per sistemi lineari generali. Ad esempio, il sistema di vincoli
> $$
> \begin{cases}
> x_1 + x_2 \leq 1,\\
> x_1 + x_2 \geq 0
> \end{cases}
> $$
> non ha alcuna soluzione di base, in quanto non è possibile estrarre dal sistema due vincoli linearmente indipendenti. Tuttavia il sistema ammette soluzioni (ad esempio l'origine). Geometricamente, questo sistema definisce un poliedro non vuoto privo di punti estremi e dunque privo di soluzioni di base ammissibili.
