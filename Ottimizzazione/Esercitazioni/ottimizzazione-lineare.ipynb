# Esercitazioni su Ottimizzazione Lineare 

---

### Esercizio 1.0
Un panificio vende il suo prodotto sotto forma di panini e di panini farciti. I lavoratori hanno a disposizione $200$ chilogrammi di farina a settimana. Ogni panino richiede mezzo etto di farina. 

Un fornitore esterno consegna al panificio la quantità richiesta di prosciutto, che però non può superare $80$ chilogrammi per settimana. Per preparare un panino farcito si utilizza mezzo etto di prosciutto.

Tutti gli altri ingredienti sono disponibili in gran quantità, diciamo senza limitazioni.
Ci sono a disposizione cinque lavoratori e ciascuno lavora $40$ ore a settimana.
La produzione di ciascun panino richiede $2$ minuti di lavoro (di un solo lavoratore), l’eventuale farcitura ne richiede altri $4$.
La richiesta della clientela è elevata, per cui tutta la produzione sarà venduta.

Ogni panino venduto vuoto dà un profitto netto di $0.1$ euro, ogni panino farcito dà un profitto netto di $0.2$ euro. La gestione del panificio deve individuare il piano di produzione settimanale che massimizza il suo profitto totale, e quindi quanti panini produrre e quanti farcirne in ogni settimana.

1. Formulate un modello di ottimizzazione lineare per questo problema (trascurando le condizioni di interezza sulle variabili).
2. Disegnate la regione ammissibile e individuate una politica ottima con il metodo grafico

### Soluzione
Sia $x_1$ il numero di panini vuoti prodotti in una settimana e $x_2$ il numero di panini farciti prodotti in una settimana. Entrambe sono variabili reali non negative (rilassamento continuo).

**Vincoli.**
- **Farina:** ogni panino (vuoto o farcito) richiede $0.05$ kg di farina. Disponibilità $200$ kg/settimana:
  $$
  0.05x_1 + 0.05x_2 \le 200 \quad\Longleftrightarrow\quad x_1+x_2 \le 4000.
  $$
- **Prosciutto:** ogni panino farcito richiede $0.05$ kg di prosciutto. Disponibilità massima $80$ kg/settimana:
  $$
  0.05x_2 \le 80 \quad\Longleftrightarrow\quad x_2 \le 1600.
  $$
- **Lavoro:** $5$ lavoratori $\times 40$ h $=200$ h $=12000$ min. Un panino vuoto richiede $2$ min, uno farcito $2+4=6$ min:
  $$
  2x_1 + 6x_2 \le 12000 \quad\Longleftrightarrow\quad x_1+3x_2 \le 6000.
  $$
- **Non negatività:** $x_1\ge 0,\; x_2\ge 0$.

**Funzione obiettivo.**  
Profitto netto: $0.1$ euro per panino vuoto, $0.2$ euro per panino farcito:
$$
\max f(x_1,x_2)=0.1x_1+0.2x_2.
$$

**Modello PL.**
$$
\begin{aligned}
\max\quad & 0.1x_1+0.2x_2\\
\text{s.a}\quad
& x_1+x_2 \le 4000,\\
& x_2 \le 1600,\\
& x_1+3x_2 \le 6000,\\
& x_1\ge 0,\; x_2\ge 0.
\end{aligned}
$$

La regione ammissibile è un poligono convesso limitato; l’ottimo, se esiste, è su un vertice. I vertici sono:
$$
(0,0),\quad (4000,0),\quad (3000,1000),\quad (1200,1600),\quad (0,1600).
$$
Valutando $f$:
$$
\begin{aligned}
f(0,0)&=0,\\
f(4000,0)&=400,\\
f(3000,1000)&=500,\\
f(1200,1600)&=440,\\
f(0,1600)&=320.
\end{aligned}
$$
Il massimo è $500$ in $(3000,1000)$. Quindi la politica ottima è produrre $3000$ panini vuoti e $1000$ panini farciti a settimana.

> **Osservazione 1.1**  
> Si poteva anche definire $x_1$ come numero totale di panini prodotti e $x_2$ come numero di panini farciti. Allora i panini vuoti sono $x_1-x_2$. Il modello diventa:
> $$
> \begin{aligned}
> \max\quad & 0.1(x_1-x_2)+0.2x_2 = 0.1x_1+0.1x_2\\
> \text{s.a}\quad
> & 0.05x_1 \le 200 \;\Rightarrow\; x_1 \le 4000,\\
> & 0.05x_2 \le 80 \;\Rightarrow\; x_2 \le 1600,\\
> & 2(x_1-x_2)+6x_2 \le 12000 \;\Rightarrow\; x_1+2x_2 \le 6000,\\
> & x_2 \le x_1,\\
> & x_1\ge 0,\; x_2\ge 0.
> \end{aligned}
> $$
> I vertici di questa nuova regione sono:
> $$
> (0,0),\quad (4000,0),\quad (4000,1000),\quad (2800,1600),\quad (1600,1600).
> $$
> La funzione obiettivo $0.1x_1+0.1x_2$ vale:
> $$
> f(0,0)=0,\quad f(4000,0)=400,\quad f(4000,1000)=500,\quad f(2800,1600)=440,\quad f(1600,1600)=320.
> $$
> Il massimo è ancora $500$ in $(x_1,x_2)=(4000,1000)$, che corrisponde a $4000$ panini totali di cui $1000$ farciti, cioè $3000$ vuoti e $1000$ farciti.

**Conclusione.**  
La soluzione ottima è $x_1^*=3000$, $x_2^*=1000$, con profitto massimo $500$ euro/settimana.

### Esercizio 2.0

Una macelleria prepara la carne macinata mediante una combinazione di manzo e maiale. La carne di manzo che viene macinata contiene l’$80\%$ di polpa e il $20\%$ di grasso e costa $8$ euro al kg; la carne di maiale contiene il $68\%$ di polpa e il $32\%$ di grasso e costa $6$ euro al kg. Quale quantità di ciascun tipo di carne deve impiegare il macellaio per preparare un kg di carne macinata se vuole minimizzare il costo ed evitare che il contenuto grasso del macinato superi il $25\%$?

1. Formulare il problema con un modello di ottimizzazione lineare.
2. Risolvere il modello per via grafica.

### Soluzione:
Sia $x_1$ la qunatità in kili di carne di manzo e sia $x_2$ la quantità in kili di carne di maiale. Entrambe sono variabili reali non negative (rilassamento continuo).

**Vincoli:**
- **Percentuale di grasso:** 
    $$
    0.20x_1 + 0.32x_2 \leq 0.25 \iff 20x_1 + 32x_2 \leq 25
    $$
- **Non negatività:** $x_1 \geq 0, x_2 \geq 0$ 

**Funzione obbiettivo:**
$8$ euro al kilo per la carne di manzo e $6$ euro al kilo per la carne di maiale, da cui: 
$$
f(x_1, x_2) = 8x_1 + 6x_2
$$

**Modello di programmazione lineare:**
$$
\begin{aligned}
\max\quad & 8x_1 + 6x_2\\
\text{s.a}\quad
& 20x_1 + 32x_2 \leq 25,\\
& x_1\ge 0,\; x_2\ge 0.
\end{aligned}
$$
