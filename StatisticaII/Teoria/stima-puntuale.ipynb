# Stima Puntuale

> **Convenzione.** Scriviamo $p_Y(y;\theta)$ invece di $p_Y(y,\theta)$ per sottolineare che $y$ è la variabile e $\theta$ il parametro.

---

## 1. Modello statistico parametrico

### Definizione 1.1 — Modello statistico parametrico

Sia $\Theta \subseteq \mathbb{R}^p$ non vuoto. Una famiglia di densità

$$
\mathcal{P} = \{ p_Y(\cdot;\theta) : \theta \in \Theta \}
$$

rispetto a una misura dominante $\mu$ è detta **modello statistico parametrico**. L'insieme $\Theta$ è detto **spazio parametrico**.

L'obiettivo della **stima puntuale** è, dato un campione osservato, assegnare a $\theta$ un valore plausibile.

> **Osservazione 1.2 — Identificabilità.**  
> Un modello statistico parametrico si dice **identificabile** se l'applicazione $\theta \mapsto p_Y(\cdot;\theta)$ è iniettiva, cioè se
> $$
> \theta_1 \neq \theta_2 \implies p_Y(\cdot;\theta_1) \neq p_Y(\cdot;\theta_2).
> $$
> L'identificabilità è una condizione necessaria per poter stimare $\theta$ in modo univoco. Se il modello non è identificabile, esistono due valori distinti del parametro che generano la stessa distribuzione, e nessun dato può distinguerli.

### Esempio 1.3

- Se $Y \sim \operatorname{Bernoulli}(p)$, allora $\Theta = (0,1)$ e $p_Y(y;p) = p^y(1-p)^{1-y}$ per $y \in \{0,1\}$.
- Se $Y \sim \mathcal{N}(\mu, \sigma^2)$, allora $\Theta = \mathbb{R} \times (0,\infty)$ e
  $$
  p_Y(y; \mu, \sigma^2) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(y-\mu)^2}{2\sigma^2}\right).
  $$
- Se $Y \sim \operatorname{Poisson}(\lambda)$, allora $\Theta = (0,\infty)$ e $p_Y(y;\lambda) = e^{-\lambda} \lambda^y / y!$ per $y \in \mathbb{N}$.

### Controesempio 1.4 — Modello non identificabile

Sia $Y \sim \mathcal{N}(\mu, \sigma^2)$ e si consideri il parametro $\theta = \mu^2$. Il modello non è identificabile rispetto a $\theta$: i valori $\mu$ e $-\mu$ generano la stessa distribuzione ma $\mu^2 = (-\mu)^2$. In questo caso, il parametro $\mu^2$ è identificabile solo se si restringe lo spazio parametrico a $\mu \geq 0$.

## 2. Campione casuale semplice

### Definizione 2.1 — Campione casuale semplice

Siano $Y_1, \dots, Y_n$ variabili aleatorie definite su $(\Omega, \mathcal{F}, P_\theta)$. Esse formano un **campione casuale semplice** da $Y$ se sono indipendenti e identicamente distribuite con legge $P_{Y;\theta}$.

Equivalentemente, il vettore $\mathbf{Y} = (Y_1, \dots, Y_n)$ ha legge prodotto

$$
P_{\mathbf{Y};\theta} = P_{Y;\theta}^{\otimes n}.
$$

Se il modello è dominato, la densità congiunta è

$$
p_{\mathbf{Y}}(\mathbf{y};\theta)
=
\prod_{i=1}^n p_Y(y_i;\theta),
\qquad
\mathbf{y} = (y_1,\dots,y_n) \in \mathbb{R}^n.
$$

Prima dell'estrazione, $\mathbf{Y}$ è un vettore aleatorio. Dopo l'estrazione, osserviamo una realizzazione

$$
\mathbf{y} = (y_1,\dots,y_n) \in \mathbb{R}^n.
$$

> **Osservazione 2.2.**  
> L'ipotesi di indipendenza e identica distribuzione è cruciale: garantisce che ogni osservazione porti la stessa quantità di informazione su $\theta$ e che le osservazioni non siano ridondanti. Se le osservazioni fossero dipendenti, la densità congiunta non sarebbe il prodotto delle densità marginali e tutta la teoria dovrebbe essere rivista.

### Esempio 2.3

Se $Y \sim \operatorname{Bernoulli}(p)$ e $n = 5$, con osservazioni $\mathbf{y} = (1, 0, 1, 1, 0)$, la densità congiunta è

$$
p_{\mathbf{Y}}(\mathbf{y};p) = p^3(1-p)^2.
$$

## 3. Statistica campionaria

### Definizione 3.1 — Statistica campionaria

Dato un campione casuale $\mathbf{Y} = (Y_1,\dots,Y_n)$ relativo al modello statistico $\{ p_Y(\cdot;\theta) : \theta \in \Theta \}$, una **statistica campionaria** è una funzione misurabile

$$
T_n = g(Y_1,\dots,Y_n),
$$

dove

$$
g : (\mathbb{R}^n, \mathcal{B}(\mathbb{R}^n)) \to (\mathbb{R}^k, \mathcal{B}(\mathbb{R}^k))
$$

è una funzione misurabile che **non dipende da $\theta$**.

Poiché $\mathbf{Y}$ è un vettore aleatorio, anche $T_n$ è una variabile aleatoria, o più in generale un vettore aleatorio.

### Esempi 3.2

- **Media campionaria:**
  $$
  \overline{Y} = \frac{1}{n} \sum_{i=1}^n Y_i.
  $$
- **Varianza campionaria:**
  $$
  S^2 = \frac{1}{n-1} \sum_{i=1}^n (Y_i - \overline{Y})^2.
  $$
- **Semisomma tra minimo e massimo:**
  $$
  \overline{Y}_{\min, \max} = \frac{Y_{(1)} + Y_{(n)}}{2}.
  $$
- **Media troncata:**
  $$
  \overline{Y}_{\text{trim}} = \frac{1}{n-2}\sum_{i = 2}^{n-1}Y_{(i)}.
  $$
- **Massimo campionario:**
  $$
  M_n = \max\{Y_1,\dots,Y_n\} = Y_{(n)}.
  $$
- **Minimo campionario:**
  $$
  m_n = \min\{Y_1,\dots,Y_n\} = Y_{(1)}.
  $$
- **Somma delle osservazioni:**
  $$
  \sum_{i = 1}^n Y_i.
  $$

> **Osservazione 3.3.**  
> La condizione che $g$ non dipenda da $\theta$ è essenziale: se $g$ dipendesse da $\theta$, non sarebbe una statistica ma una funzione del parametro, e non potrebbe essere calcolata senza conoscere $\theta$.

> **Osservazione 3.4 — Statistiche di ordine.**  
> Le **statistiche d'ordine** $Y_{(1)} \leq Y_{(2)} \leq \dots \leq Y_{(n)}$ sono le osservazioni ordinate in modo crescente. Esse costituiscono una statistica campionaria vettoriale di dimensione $n$ e sono alla base di stimatori come il massimo, il minimo e la mediana campionaria.

## 4. Stimatore

### Definizione 4.1 — Stimatore

Sia $\{  p_Y(\cdot;\theta) : \theta \in \Theta \}$ un modello statistico.

Uno **stimatore** di $\theta$ è una statistica campionaria

$$
\hat{\theta}_n = g(Y_1,\dots,Y_n)
$$

tale che

$$
g : (\mathbb{R}^n, \mathcal{B}(\mathbb{R}^n)) \to (\Theta, \mathcal{B}(\Theta))
$$

e $g$ non dipende da $\theta$.

Uno stimatore è quindi una **regola** che, applicata a un campione, produce un valore in $\Theta$. Prima di osservare il campione, $\hat{\theta}_n$ è una variabile aleatoria.

> **Osservazione 4.2.**  
> Ogni stimatore è una statistica campionaria, ma non ogni statistica campionaria è uno stimatore. Per essere uno stimatore, una statistica deve avere codominio in $\Theta$ ed essere usata per stimare $\theta$.

### Esempio 4.3

Sia $Y \sim \mathcal{N}(\mu, \sigma^2)$ con $\theta = \mu$. Allora $\overline{Y}$ è uno stimatore di $\mu$, mentre $S^2$ è uno stimatore di $\sigma^2$ ma non di $\mu$.

---

## 5. Stima

### Definizione 5.1 — Stima

Sia $\mathbf{y} = (y_1,\dots,y_n)$ il campione effettivamente osservato. Il valore

$$
\hat{\theta}_{\text{oss}} = g(y_1,\dots,y_n)
$$

è detto **stima** di $\theta$.

Mentre $\hat{\theta}_n$ è una variabile aleatoria, $\hat{\theta}_{\text{oss}}$ è un elemento di $\Theta$, quindi nel caso scalare è un numero.

Spesso si usa lo stesso simbolo $\hat{\theta}$ sia per lo stimatore sia per la stima. Formalmente è però opportuno distinguere:

$$
\hat{\theta}_n = \text{stimatore}, \qquad
\hat{\theta}_{\text{oss}} = \text{stima}.
$$

### Esempio 5.2

Se $Y \sim \operatorname{Bernoulli}(p)$ e osserviamo $\mathbf{y} = (1,0,1,1,0)$, allora

$$
\hat{p}_{\text{oss}} = \overline{y} = \frac{3}{5} = 0.6.
$$

Mentre $\hat{p}_n = \overline{Y}$ è uno stimatore (variabile aleatoria), $0.6$ è la stima (numero).

## 6. Distribuzione campionaria

### Definizione 6.1 — Distribuzione campionaria

Sia $\hat{\theta}_n = g(Y_1,\dots,Y_n)$ uno stimatore. La sua **distribuzione campionaria** è la legge immagine

$$
P_{\hat{\theta}_n;\theta}
=
P_\theta \circ \hat{\theta}_n^{-1}
$$

su $(\Theta, \mathcal{B}(\Theta))$. Esplicitamente, per ogni $A \in \mathcal{B}(\Theta)$,

$$
P_{\hat{\theta}_n;\theta}(A)
=
P_\theta(\hat{\theta}_n \in A)
=
P_\theta\bigl(\{\omega \in \Omega : \hat{\theta}_n(\omega) \in A\}\bigr).
$$

Se $\hat{\theta}_n$ è assolutamente continua rispetto a una misura, indichiamo la sua densità con

$$
f_{\hat{\theta}_n}(t;\theta).
$$

La distribuzione campionaria dipende da:

- il parametro $\theta$;
- la numerosità campionaria $n$;
- la procedura di stima $g$.

### Osservazione 6.2 — Perché è centrale

Senza la distribuzione campionaria, una stima è solo un numero privo di misura dell'incertezza. La distribuzione campionaria permette di:

- confrontare stimatori alternativi;
- costruire intervalli di confidenza;
- verificare ipotesi statistiche;
- calcolare errore standard, distorsione, varianza ed errore quadratico medio.

### Esempio 6.3 — Distribuzione campionaria della media per variabili normali

Sia $Y \sim \mathcal{N}(\mu, \sigma^2)$ e $\hat{\mu}_n = \overline{Y}$. Allora

$$
\overline{Y} \sim \mathcal{N}\!\left(\mu, \frac{\sigma^2}{n}\right).
$$

La distribuzione campionaria della media è quindi ancora normale, con media $\mu$ e varianza $\sigma^2/n$. Al crescere di $n$, la distribuzione si concentra sempre di più attorno a $\mu$.

### Esempio 6.4 — Distribuzione campionaria della varianza

Sia $Y \sim \mathcal{N}(\mu, \sigma^2)$ e $\hat{\sigma}^2_n = S^2$. Allora

$$
\frac{(n-1)S^2}{\sigma^2} \sim \chi^2_{n-1}.
$$

La distribuzione campionaria di $S^2$ è quindi legata alla distribuzione chi-quadro.

### Esempio 6.5 — Distribuzione campionaria per la Bernoulli

Sia $Y \sim \operatorname{Bernoulli}(p)$ e $\hat{p}_n = \overline{Y}$. Allora

$$
n \hat{p}_n = \sum_{i=1}^n Y_i \sim \operatorname{Bin}(n,p).
$$

Pertanto, per $k = 0, 1, \dots, n$:

$$
P_\theta\left(\hat{p}_n = \frac{k}{n}\right) = \binom{n}{k} p^k (1-p)^{n-k}.
$$

## 7. Principio del campionamento ripetuto

Il **principio del campionamento ripetuto** consiste nel valutare la bontà di uno stimatore immaginando di ripetere infinite volte l'esperimento:

1. si estrae un campione $\mathbf{Y}$;
2. si calcola la stima $\hat{\theta}_n(\mathbf{Y})$;
3. si ripete il procedimento.

In questo modo si ottiene una popolazione di stime, la cui distribuzione è proprio la distribuzione campionaria di $\hat{\theta}_n$.

> **Osservazione:** In altre parole l'adeguatezza di uno stimatore $\hat{\theta}_n$ si valuta considerandolo come una variabile aleatoria e studiando le proprietà di tale variabile aleatoria mediante la sua distribuzione campionaria.

Formalmente, studiamo le proprietà della variabile aleatoria $\hat{\theta}_n$ attraverso la sua legge $P_{\hat{\theta}_n;\theta}$. Questo principio non è un teorema, ma un criterio metodologico che giustifica l'uso di bias, varianza, MSE e consistenza.

## 8. Proprietà degli stimatori

Sia $Y \sim p_Y(y;\theta)$ e sia $\hat{\theta}_n = g(Y_1,\dots,Y_n)$ uno stimatore di $\theta$. Supponiamo che i momenti coinvolti siano finiti, salvo diversa indicazione.

Sono molti ed eterogenei gli aspetti da considerare nel confrontare $\theta$ con $\hat{\theta}$ e non sempre tali aspetti sono compatibili tra loro:

- sufficienza;
- correttezza o non distorsione;
- efficienza;
- consistenza (coerenza).

---

### 8.1 Sufficienza

#### Definizione 8.1.1 — Statistica sufficiente

Sia $T_n = g(Y_1,\dots,Y_n)$ una statistica. Si dice che $T_n$ è **sufficiente** per $\theta$ se la distribuzione condizionata di $\mathbf{Y}$ dato $T_n = t$ non dipende da $\theta$.

In forma più precisa, per ogni $A \in \mathcal{B}(\mathbb{R}^n)$ esiste una versione della probabilità condizionata

$$
P_\theta(\mathbf{Y} \in A \mid T_n = t)
$$

che non dipende da $\theta$.

In un modello dominato, ciò equivale a richiedere che la densità condizionata

$$
p_{\mathbf{Y} \mid T_n}(\mathbf{y} \mid t;\theta)
$$

non dipenda da $\theta$.

> **Interpretazione:** Una statistica sufficiente racchiude ed esaurisce tutte le informazioni riguardanti $\theta$ contenute nel campione. Una volta noto $T_n$, il campione completo non aggiunge alcuna informazione ulteriore su $\theta$.

> **Osservazione:** la sufficienza si definisce in rapporto ad un parametro $\theta$ che caratterizza uno specifico modello statistico $p_{Y}$; senza la specificazione del modello statistico non è possibile parlare di sufficienza.

> **Osservazione:** In molti casi è difficile stabilire se la condizione di sufficienza sia verificata sulla base della definizione stessa. Per questo motivo si usa il criterio di fattorizzazione.

#### Esempio 8.1.2 — Verifica diretta della sufficienza per la Bernoulli

Sia $Y \sim \operatorname{Bernoulli}(p)$ e consideriamo $T_n = \sum_{i=1}^n Y_i$. Vogliamo verificare **direttamente dalla definizione** che $T_n$ è sufficiente per $p$.

**Calcolo della densità condizionata.** Sia $\mathbf{y} = (y_1,\dots,y_n) \in \{0,1\}^n$ e $t = \sum_i y_i$. Allora

$$
p_{\mathbf{Y} \mid T_n}(\mathbf{y} \mid t; p)
=
\frac{p_{\mathbf{Y}}(\mathbf{y};p)}{p_{T_n}(t;p)}.
$$

La densità congiunta è

$$
p_{\mathbf{Y}}(\mathbf{y};p)
=
p^{\sum_i y_i}(1-p)^{n - \sum_i y_i}
=
p^t (1-p)^{n-t}.
$$

La distribuzione di $T_n$ è binomiale:

$$
P(T_n = t) = \binom{n}{t} p^t (1-p)^{n-t}.
$$

Quindi

$$
p_{\mathbf{Y} \mid T_n}(\mathbf{y} \mid t; p)
=
\frac{p^t (1-p)^{n-t}}{\binom{n}{t} p^t (1-p)^{n-t}}
=
\frac{1}{\binom{n}{t}}.
$$

**Conclusione.** La densità condizionata non dipende da $p$: quindi $T_n = \sum_i Y_i$ è sufficiente per $p$. $\square$

> **Interpretazione.** Una volta noto il numero totale di successi $t$, la particolare sequenza di successi e insuccessi non porta alcuna informazione aggiuntiva su $p$. Tutte le sequenze con lo stesso numero di successi sono equiprobabili, indipendentemente da $p$.

#### Teorema 8.1.3 — Criterio di fattorizzazione di Fisher–Neyman

Sia $\mathbf{Y} = (Y_1,\dots,Y_n)$ un campione casuale con densità congiunta $p_{\mathbf{Y}}(\mathbf{y};\theta)$. Una statistica $T_n = g(\mathbf{Y})$ è sufficiente per $\theta$ se e solo se esistono funzioni misurabili $h : \mathbb{R}^n \to [0,\infty)$ e $k : \mathbb{R}^k \times \Theta \to [0,\infty)$ tali che

$$
p_{\mathbf{Y}}(\mathbf{y};\theta)
=
h(\mathbf{y}) \, k(T_n(\mathbf{y});\theta)
$$

per quasi ogni $\mathbf{y}$ e per ogni $\theta \in \Theta$, con $h$ indipendente da $\theta$.

**Dimostrazione.**

**($\Rightarrow$)** Supponiamo che $T_n$ sia sufficiente. Allora la densità condizionata $p_{\mathbf{Y}\mid T_n}(\mathbf{y} \mid t;\theta)$ non dipende da $\theta$; indichiamola con $h(\mathbf{y})$. Allora

$$
p_{\mathbf{Y}}(\mathbf{y};\theta)
=
p_{T_n}(T_n(\mathbf{y});\theta) \cdot h(\mathbf{y}),
$$

dove il secondo fattore è indipendente da $\theta$. Ponendo $k(t;\theta) = p_{T_n}(t;\theta)$, si ottiene la fattorizzazione.

**($\Leftarrow$)** Supponiamo che valga la fattorizzazione. La densità di $T_n$ è

$$
p_{T_n}(t;\theta)
=
k(t;\theta) \int_{\{\mathbf{y}: T_n(\mathbf{y})=t\}} h(\mathbf{y}) \, d\mu(\mathbf{y}).
$$

Pertanto

$$
p_{\mathbf{Y} \mid T_n}(\mathbf{y} \mid t;\theta)
=
\frac{p_{\mathbf{Y}}(\mathbf{y};\theta)}{p_{T_n}(t;\theta)}
=
\frac{h(\mathbf{y})}{\int_{\{\mathbf{y}: T_n(\mathbf{y})=t\}} h(\mathbf{y}) \, d\mu(\mathbf{y})},
$$

che non dipende da $\theta$. Quindi $T_n$ è sufficiente. $\square$

> **Osservazione 8.1.3.1.**  
> Il criterio di fattorizzazione è lo strumento **operativo** principale per verificare la sufficienza. In pratica, si cerca di scrivere la densità congiunta come prodotto di un fattore che dipende solo da $\mathbf{y}$ (e non da $\theta$) e di un fattore che dipende da $\mathbf{y}$ solo attraverso $T_n$ e da $\theta$.

#### Proposizione 8.1.4 — Invarianza per trasformazioni biettive

Se $T_n$ è sufficiente per $\theta$ e $u$ è una funzione biettiva misurabile con inversa misurabile, allora anche $u(T_n)$ è sufficiente per $\theta$.

**Dimostrazione.**  
Poiché $u$ è biettiva, $\sigma(T_n) = \sigma(u(T_n))$. La distribuzione condizionata di $\mathbf{Y}$ dato $u(T_n)$ coincide con quella dato $T_n$, quindi non dipende da $\theta$. $\square$

#### Esempio 8.1.5 — Sufficienza per la normale con varianza nota

Sia $Y \sim \mathcal{N}(\mu, \sigma^2)$ con $\sigma^2$ nota. Allora

$$
p_{\mathbf{Y}}(\mathbf{y};\mu)
=
(2\pi\sigma^2)^{-n/2} \exp\left(-\frac{1}{2\sigma^2}\sum_{i=1}^n (y_i - \mu)^2\right).
$$

Sviluppando il quadrato:

$$
\sum_{i=1}^n (y_i - \mu)^2 = \sum_{i=1}^n y_i^2 - 2\mu \sum_{i=1}^n y_i + n\mu^2.
$$

Ponendo

$$
T_n = \sum_{i=1}^n Y_i,
\qquad
h(\mathbf{y}) = (2\pi\sigma^2)^{-n/2} e^{-\sum y_i^2 / (2\sigma^2)},
$$

$$
k(t;\mu) = e^{-(t^2 - 2n\mu t + n^2\mu^2)/(2n\sigma^2)},
$$

si ottiene la fattorizzazione. Dunque $\sum_i Y_i$ (e quindi $\overline{Y}$) è sufficiente per $\mu$.

#### Esempio 8.1.6 — Sufficienza per la Poisson

Sia $Y \sim \operatorname{Poisson}(\lambda)$. Allora

$$
p_{\mathbf{Y}}(\mathbf{y};\lambda)
=
\prod_{i=1}^n \frac{e^{-\lambda} \lambda^{y_i}}{y_i!}
=
e^{-n\lambda} \lambda^{\sum_i y_i} \prod_{i=1}^n \frac{1}{y_i!}.
$$

Ponendo

$$
T_n = \sum_{i=1}^n Y_i,
\qquad
h(\mathbf{y}) = \prod_{i=1}^n \frac{1}{y_i!},
\qquad
k(t;\lambda) = e^{-n\lambda} \lambda^t,
$$

si ottiene la fattorizzazione. Dunque $\sum_i Y_i$ è sufficiente per $\lambda$.

#### Esempio 8.1.7 — Sufficienza per la normale con entrambi i parametri incogniti

Sia $Y \sim \mathcal{N}(\mu, \sigma^2)$ con $\mu$ e $\sigma^2$ entrambi incogniti. Allora

$$
p_{\mathbf{Y}}(\mathbf{y};\mu,\sigma^2)
=
(2\pi\sigma^2)^{-n/2} \exp\left(-\frac{1}{2\sigma^2}\sum_{i=1}^n (y_i - \mu)^2\right).
$$

Sviluppando:

$$
\sum_{i=1}^n (y_i - \mu)^2 = \sum_{i=1}^n y_i^2 - 2\mu \sum_{i=1}^n y_i + n\mu^2.
$$

Ponendo

$$
T_n = \left(\sum_{i=1}^n Y_i, \sum_{i=1}^n Y_i^2\right),
$$

si ottiene la fattorizzazione con $h(\mathbf{y}) = 1$ e

$$
k(t_1, t_2; \mu, \sigma^2) = (2\pi\sigma^2)^{-n/2} e^{-\frac{1}{2\sigma^2}(t_2 - 2\mu t_1 + n\mu^2)}.
$$

Dunque $\left(\sum_i Y_i, \sum_i Y_i^2\right)$ è sufficiente per $(\mu, \sigma^2)$.

### 8.2 Non distorsione

#### Definizione 8.2.1 — Stimatore non distorto

Uno stimatore $\hat{\theta}_n$ di $\theta$ si dice **non distorto** o **corretto** se

$$
\mathbb{E}_\theta[\hat{\theta}_n] = \theta
\iff
\mathbb{E}_\theta[\hat{\theta}_n] - \theta = 0
\iff
\mathbb{E}_\theta[\hat{\theta}_n - \theta] = 0
\iff
\mathcal{B}(\hat{\theta}_n) = 0
\qquad \forall \theta \in \Theta.
$$

La **distorsione**, ovvero l'errore sistematico di stima, è definita come

$$
\mathcal{B}_\theta(\hat{\theta}_n)
=
\mathbb{E}_\theta[\hat{\theta}_n - \theta]
=
\mathbb{E}_\theta[\hat{\theta}_n] - \theta.
$$

La quantità $(\hat{\theta}_n - \theta)$ è detta **errore di stima**.

Uno stimatore è non distorto se e solo se $\mathcal{B}_\theta(\hat{\theta}_n)=0$ per ogni $\theta$.

> **Interpretazione.** La non distorsione garantisce che, in media, lo stimatore non sovrastima né sottostima sistematicamente il parametro. Immaginando di ripetere infinite volte l'esperimento, la media delle stime coincide con il vero valore del parametro.

#### Esempio 8.2.2 — Media campionaria

Sia $Y$ con $\mathbb{E}[Y]=\mu$. Allora

$$
\mathbb{E}[\overline{Y}]
=
\mathbb{E}\left[\frac{1}{n}\sum_{i=1}^n Y_i\right]
=
\frac{1}{n}\sum_{i=1}^n \mathbb{E}[Y_i]
=
\mu.
$$

Quindi $\overline{Y}$ è non distorto per $\mu$.

#### Esempio 8.2.3 — Varianza campionaria

Sia $\operatorname{Var}(Y)=\sigma^2$. Allora

$$
\mathbb{E}[S^2] = \sigma^2,
$$

dove

$$
S^2 = \frac{1}{n-1}\sum_{i=1}^n (Y_i - \overline{Y})^2.
$$

Quindi $S^2$ è non distorto per $\sigma^2$.

**Verifica.**

$$
\begin{aligned}
\sum_{i=1}^n (Y_i - \overline{Y})^2
&=
\sum_{i=1}^n (Y_i - \mu + \mu - \overline{Y})^2 \\
&=
\sum_{i=1}^n (Y_i - \mu)^2 - n(\overline{Y} - \mu)^2.
\end{aligned}
$$

Prendendo il valore atteso:

$$
\mathbb{E}\left[\sum_{i=1}^n (Y_i - \overline{Y})^2\right]
=
n\sigma^2 - n \cdot \frac{\sigma^2}{n}
=
(n-1)\sigma^2.
$$

Dividendo per $n-1$ si ottiene $\mathbb{E}[S^2] = \sigma^2$.

#### Controesempio 8.2.4 — Stimatore costante

Sia $\hat{\theta}_n = c$ con $c \in \Theta$ fissato, da cui $\mathbb{E}(\hat{\theta}_n) = c$. Allora

$$
\mathcal{B}_\theta(\hat{\theta}_n) = c - \theta.
$$

Se $c \neq \theta$, lo stimatore è distorto. Inoltre non è consistente, perché non dipende da $n$.

#### Esempio 8.2.5 — Stimatore non distorto per la varianza con media nota

Sia $Y \sim p_Y(y;\theta)$ con $\mathbb{E}(Y) = \mu$ nota e $\operatorname{Var}(Y) = \sigma^2$. Lo stimatore

$$
\hat{\sigma}^2_\mu = \frac{1}{n}\sum_{i=1}^n (Y_i - \mu)^2
$$

è non distorto per $\sigma^2$:

$$
\mathbb{E}[\hat{\sigma}^2_\mu]
=
\frac{1}{n}\sum_{i=1}^n \mathbb{E}[(Y_i - \mu)^2]
=
\frac{1}{n}\cdot n \sigma^2
=
\sigma^2.
$$

### 8.3 Non distorsione asintotica

#### Definizione 8.3.1 — Stimatore asintoticamente non distorto

Uno stimatore $\hat{\theta}_n$ è **asintoticamente non distorto** se

$$
\lim_{n \to +\infty} \mathbb{E}_\theta[\hat{\theta}_n] = \theta
\iff
\lim_{n \to +\infty} \mathbb{E}(\hat{\theta}_n - \theta) = 0
\iff
\lim_{n\to+\infty}\mathcal{B}(\hat{\theta}_n) = 0
\qquad \forall \theta \in \Theta.
$$

> **Interpretazione.** La non distorsione asintotica garantisce che, in media, lo stimatore non sovrastima né sottostima sistematicamente il parametro al crescere di $n$. Si richiede in pratica che l'eventuale distorsione dello stimatore tenda a ridursi fino a sparire all'aumentare della dimensione campionaria.

#### Esempio 8.3.2 — Varianza campionaria con denominatore $n$

Sia

$$
\hat{\sigma}^2_n = \frac{1}{n}\sum_{i=1}^n (Y_i - \overline{Y})^2.
$$

Allora

$$
\mathbb{E}[\hat{\sigma}^2_n] = \frac{n-1}{n}\sigma^2,
$$

quindi

$$
\mathcal{B}_\theta(\hat{\sigma}^2_n) = -\frac{\sigma^2}{n} \to 0.
$$

Lo stimatore è asintoticamente non distorto, ma non non distorto per $n$ finito.

#### Esempio 8.3.3 — Massimo campionario per una uniforme

Sia $Y \sim U(0, \theta)$ e sia $Y_1, \dots, Y_n$ un campione i.i.d. tratto da $Y$. Consideriamo lo stimatore

$$
\hat{\theta}_n = \max_{1\le i\le n} Y_i = Y_{(n)}.
$$

Mostriamo che è distorto ma asintoticamente corretto.

**Distribuzione del massimo.** Si ha

$$
(Y_{(n)} \leq y) \iff (Y_1 \leq y) \cap \dots \cap (Y_n \leq y).
$$

Da cui, sfruttando l'indipendenza,

$$
F_{Y_{(n)}}(y) = P(Y_{(n)} \leq y) = \prod_{i=1}^n P(Y_i \leq y) = F_Y(y)^n.
$$

Poiché $Y \sim U(0,\theta)$, si ha $p_Y(y) = 1/\theta$ e $F_Y(y) = y/\theta$ per $0 < y < \theta$. Derivando:

$$
p_{Y_{(n)}}(y) = \frac{d}{dy}F_Y(y)^n = n F_Y(y)^{n-1} p_Y(y) = \frac{n y^{n-1}}{\theta^n}, \quad 0 < y < \theta.
$$

**Media del massimo.**

$$
\mathbb{E}(Y_{(n)}) = \int_0^\theta y \cdot \frac{n y^{n-1}}{\theta^n}\, dy = \frac{n}{\theta^n} \int_0^\theta y^n\, dy = \frac{n}{\theta^n}\cdot\frac{\theta^{n+1}}{n+1} = \frac{n}{n+1}\theta.
$$

Poiché $\mathbb{E}(Y_{(n)}) = \frac{n}{n+1}\theta \neq \theta$, lo stimatore è **distorto**:

$$
\mathcal{B}_\theta(\hat{\theta}_n) = \frac{n}{n+1}\theta - \theta = -\frac{\theta}{n+1}.
$$

Tuttavia

$$
\lim_{n\to+\infty} \mathbb{E}(Y_{(n)}) = \lim_{n\to+\infty}\frac{n}{n+1}\theta = \theta,
$$

quindi lo stimatore è **asintoticamente non distorto**.

> **Osservazione 8.3.4.**  
> Il massimo campionario è uno stimatore **distorto** di $\theta$ ma **asintoticamente non distorto**. Lo stimatore corretto per $\theta$ è
> $$
> \tilde{\theta}_n = \frac{n+1}{n} Y_{(n)},
> $$
> che verifica
> $$
> \mathbb{E}[\tilde{\theta}_n] = \frac{n+1}{n}\mathbb{E}[Y_{(n)}] = \frac{n+1}{n}\cdot\frac{n}{n+1}\theta = \theta.
> $$

### 8.4 Varianza ed errore quadratico medio

#### Definizione 8.4.1 — Varianza di uno stimatore

$$
\mathbb{V}(\hat{\theta}_n) = \operatorname{Var}_\theta(\hat{\theta}_n)
=
\mathbb{E}_\theta\left[
\left(\hat{\theta}_n - \mathbb{E}_\theta[\hat{\theta}_n]\right)^2
\right].
$$

> **Interpretazione:** La varianza misura la **dispersione** dello stimatore attorno alla sua media.

#### Definizione 8.4.2 — Errore quadratico medio

L'**errore quadratico medio** di $\hat{\theta}_n$ è

$$
\operatorname{MSE}_\theta(\hat{\theta}_n)
=
\mathbb{E}_\theta\left[
(\hat{\theta}_n - \theta)^2
\right],
$$

ovvero la media dei quadrati degli errori di stima $\hat{\theta}_n - \theta$.

#### Teorema 8.4.3 — Decomposizione bias–varianza

Se $\mathbb{E}_\theta[\hat{\theta}_n^2] < \infty$, allora

$$
\operatorname{MSE}_\theta(\hat{\theta}_n)
=
\operatorname{Var}_\theta(\hat{\theta}_n)
+
\mathcal{B}_\theta(\hat{\theta}_n)^2.
$$

**Dimostrazione.**  
Scriviamo

$$
\hat{\theta}_n - \theta
=
(\hat{\theta}_n - \mathbb{E}_\theta[\hat{\theta}_n])
+
(\mathbb{E}_\theta[\hat{\theta}_n] - \theta).
$$

Elevando al quadrato e prendendo il valore atteso:

$$
\begin{aligned}
\operatorname{MSE}_\theta(\hat{\theta}_n)
&=
\mathbb{E}_\theta\left[
(\hat{\theta}_n - \mathbb{E}_\theta[\hat{\theta}_n])^2
\right]
+
2(\mathbb{E}_\theta[\hat{\theta}_n]-\theta)
\mathbb{E}_\theta[\hat{\theta}_n - \mathbb{E}_\theta[\hat{\theta}_n]]
\\
&\quad +
(\mathbb{E}_\theta[\hat{\theta}_n]-\theta)^2.
\end{aligned}
$$

Il termine centrale è nullo perché

$$
\mathbb{E}_\theta[\hat{\theta}_n - \mathbb{E}_\theta[\hat{\theta}_n]] = 0.
$$

Quindi

$$
\operatorname{MSE}_\theta(\hat{\theta}_n)
=
\operatorname{Var}_\theta(\hat{\theta}_n)
+
\mathcal{B}_\theta(\hat{\theta}_n)^2.
$$

$\square$

#### Osservazione 8.4.4

Per stimatori non distorti,

$$
\operatorname{MSE}_\theta(\hat{\theta}_n)
=
\operatorname{Var}_\theta(\hat{\theta}_n).
$$

Poiché $\mathcal{B}_\theta(\hat{\theta}_n) = 0$. Quindi, tra due stimatori non distorti, è preferibile quello con varianza minore.

> **Osservazione 8.4.5.**  
> L'MSE è una misura di bontà **globale**: tiene conto sia della distorsione sia della varianza. Uno stimatore distorto ma con varianza molto piccola può avere MSE inferiore a uno stimatore non distorto ma con varianza grande.

#### Esempio 8.4.6 — Confronto tra due stimatori

Sia $Y_1, Y_2$ un campione da $Y \sim \mathcal{N}(\mu, \sigma^2)$ con $\sigma^2$ nota. Consideriamo due stimatori di $\mu$:

$$
\hat{\mu}_1 = \frac{Y_1 + Y_2}{2}, \qquad \hat{\mu}_2 = Y_1.
$$

Entrambi sono non distorti:

$$
\mathbb{E}[\hat{\mu}_1] = \mu, \qquad \mathbb{E}[\hat{\mu}_2] = \mu.
$$

Ma

$$
\operatorname{Var}(\hat{\mu}_1) = \frac{\sigma^2}{2}, \qquad \operatorname{Var}(\hat{\mu}_2) = \sigma^2.
$$

Dunque $\hat{\mu}_1$ ha MSE minore ed è preferibile.

#### Esempio 8.4.7 — Confronto tra stimatori distorti e non distorti

Sia $Y \sim \mathcal{N}(\mu, \sigma^2)$ con $\sigma^2$ nota e $\mu$ incognito. Consideriamo due stimatori:

$$
\hat{\mu}_1 = \overline{Y} \quad (\text{non distorto}), \qquad \hat{\mu}_2 = 0.9 \overline{Y} \quad (\text{distorto}).
$$

Allora

$$
\mathbb{E}[\hat{\mu}_1] = \mu, \qquad \mathcal{B}(\hat{\mu}_1) = 0, \qquad \operatorname{Var}(\hat{\mu}_1) = \frac{\sigma^2}{n}.
$$

$$
\mathbb{E}[\hat{\mu}_2] = 0.9\mu, \qquad \mathcal{B}(\hat{\mu}_2) = -0.1\mu, \qquad \operatorname{Var}(\hat{\mu}_2) = 0.81 \frac{\sigma^2}{n}.
$$

L'MSE di $\hat{\mu}_2$ è

$$
\operatorname{MSE}(\hat{\mu}_2) = 0.81 \frac{\sigma^2}{n} + 0.01\mu^2.
$$

Se $\mu$ è piccolo rispetto a $\sigma$, l'MSE di $\hat{\mu}_2$ può essere inferiore a quello di $\hat{\mu}_1$. Questo mostra che la non distorsione non è sempre l'unico criterio da considerare.

### 8.5 Efficienza relativa

#### Definizione 8.5.1 — Efficienza relativa

Siano $\hat{\theta}_{1,n}$ e $\hat{\theta}_{2,n}$ due stimatori di $\theta$. L'**efficienza relativa** di $\hat{\theta}_{1,n}$ rispetto a $\hat{\theta}_{2,n}$ è

$$
\operatorname{eff}(\hat{\theta}_{1,n}, \hat{\theta}_{2,n})
=
\frac{\operatorname{MSE}_\theta(\hat{\theta}_{2,n})}
{\operatorname{MSE}_\theta(\hat{\theta}_{1,n})}.
$$

Se

$$
\operatorname{eff}(\hat{\theta}_{1,n}, \hat{\theta}_{2,n}) > 1,
$$

allora $\hat{\theta}_{1,n}$ è più efficiente di $\hat{\theta}_{2,n}$.

> **Attenzione.** L'MSE può dipendere da $\theta$ e da altre quantità incognite. Pertanto il confronto uniforme in $\theta$ non è sempre possibile.

#### Esempio 8.5.2

Siano $\hat{\mu}_1 = \overline{Y}$ e $\hat{\mu}_2 = Y_1$ due stimatori di $\mu$. Allora

$$
\operatorname{eff}(\hat{\mu}_1, \hat{\mu}_2)
=
\frac{\operatorname{Var}(\hat{\mu}_2)}{\operatorname{Var}(\hat{\mu}_1)}
=
\frac{\sigma^2}{\sigma^2/n}
=
n.
$$

Quindi $\hat{\mu}_1$ è $n$ volte più efficiente di $\hat{\mu}_2$.

### 8.6 Disuguaglianza di Cramér–Rao

Sia $\mathbf{Y} = (Y_1,\dots,Y_n)$ un campione i.i.d. con densità $p_Y(y;\theta)$. Supponiamo valide le condizioni di regolarità:

1. $\Theta$ è un intervallo aperto;
2. il supporto di $p_Y(\cdot;\theta)$ non dipende da $\theta$;
3. $p_Y(y;\theta)$ è derivabile rispetto a $\theta$;
4. è possibile scambiare derivata e integrale;
5. l'informazione di Fisher è finita e positiva.

Definiamo lo **score** per una osservazione:

$$
U_1(\theta;Y)
=
\frac{\partial}{\partial \theta} \log p_Y(Y;\theta).
$$

L'**informazione di Fisher** di una osservazione è

$$
I_1(\theta)
=
\mathbb{E}_\theta\left[
U_1(\theta;Y)^2
\right]
=
-\mathbb{E}_\theta\left[
\frac{\partial^2}{\partial \theta^2} \log p_Y(Y;\theta)
\right].
$$

Per un campione di numerosità $n$, lo score totale è

$$
U_n(\theta;\mathbf{Y})
=
\sum_{i=1}^n \frac{\partial}{\partial \theta} \log p_Y(Y_i;\theta),
$$

e l'informazione di Fisher campionaria è

$$
I_n(\theta) = n I_1(\theta).
$$

> **Osservazione 8.6.0.**  
> L'uguaglianza
> $$
> I_1(\theta) = \mathbb{E}_\theta[U_1^2] = -\mathbb{E}_\theta\left[\frac{\partial^2}{\partial \theta^2} \log p_Y(Y;\theta)\right]
> $$
> segue dall'identità
> $$
> \frac{\partial^2}{\partial \theta^2} \log p_Y = \frac{1}{p_Y}\frac{\partial^2 p_Y}{\partial \theta^2} - \left(\frac{1}{p_Y}\frac{\partial p_Y}{\partial \theta}\right)^2
> $$
> e dal fatto che, sotto le condizioni di regolarità,
> $$
> \int \frac{\partial^2 p_Y}{\partial \theta^2}\, d\mu = \frac{\partial^2}{\partial \theta^2}\int p_Y\, d\mu = 0.
> $$

#### Teorema 8.6.1 — Disuguaglianza di Cramér–Rao

Sotto le condizioni di regolarità, se $\hat{\theta}_n$ è uno stimatore non distorto di $\theta$, allora

$$
\operatorname{Var}_\theta(\hat{\theta}_n)
\ge
\frac{1}{I_n(\theta)}
=
\frac{1}{n I_1(\theta)}.
$$

**Dimostrazione.**  
Poiché $\hat{\theta}_n$ è non distorto,

$$
\mathbb{E}_\theta[\hat{\theta}_n] = \theta.
$$

Derivando rispetto a $\theta$ e scambiando derivata e integrale si ottiene

$$
\int \hat{\theta}_n(\mathbf{y}) \frac{\partial}{\partial \theta} p_{\mathbf{Y}}(\mathbf{y};\theta) \, d\mu(\mathbf{y}) = 1.
$$

Esprimendo la derivata come $\frac{\partial}{\partial \theta} p_{\mathbf{Y}} = p_{\mathbf{Y}} \cdot \frac{\partial}{\partial \theta} \log p_{\mathbf{Y}}$, si ha

$$
\int \hat{\theta}_n(\mathbf{y}) \frac{\partial}{\partial \theta} \log p_{\mathbf{Y}}(\mathbf{y};\theta) \, p_{\mathbf{Y}}(\mathbf{y};\theta) \, d\mu(\mathbf{y}) = 1,
$$

cioè

$$
\mathbb{E}_\theta[\hat{\theta}_n \cdot U_n] = 1.
$$

Inoltre $\mathbb{E}_\theta[U_n] = 0$, quindi

$$
\operatorname{Cov}_\theta(\hat{\theta}_n, U_n) = \mathbb{E}_\theta[\hat{\theta}_n U_n] - \mathbb{E}_\theta[\hat{\theta}_n] \cdot \mathbb{E}_\theta[U_n] = 1.
$$

Per la disuguaglianza di Cauchy–Schwarz,

$$
\operatorname{Cov}_\theta(\hat{\theta}_n, U_n)^2
\le
\operatorname{Var}_\theta(\hat{\theta}_n) \cdot \operatorname{Var}_\theta(U_n).
$$

Dunque

$$
1 \le \operatorname{Var}_\theta(\hat{\theta}_n) \cdot I_n(\theta),
$$

da cui

$$
\operatorname{Var}_\theta(\hat{\theta}_n)
\ge
\frac{1}{I_n(\theta)}.
$$

$\square$

#### Definizione 8.6.2 — Stimatore efficiente

Uno stimatore non distorto $\hat{\theta}_n$ si dice **efficiente** se raggiunge il limite di Cramér–Rao:

$$
\operatorname{Var}_\theta(\hat{\theta}_n)
=
\frac{1}{I_n(\theta)}.
$$

In tal caso si definisce l'efficienza come

$$
\operatorname{eff}(\hat{\theta}_n)
=
\frac{1}
{\operatorname{Var}_\theta(\hat{\theta}_n) \, I_n(\theta)}
=
\frac{\text{limite di Cramér–Rao}}{\operatorname{Var}_\theta(\hat{\theta}_n)}.
$$

Se $\hat{\theta}_n$ è efficiente, $\operatorname{eff}(\hat{\theta}_n)=1$. In generale,

$$
0 \le \operatorname{eff}(\hat{\theta}_n) \le 1.
$$

> **Metodo operativo per verificare l'efficienza.**  
> Per stabilire se uno stimatore non distorto $\hat{\theta}_n$ è efficiente, si procede nel modo seguente:
>
> 1. Si verifica la **non distorsione**: $\mathbb{E}_\theta[\hat{\theta}_n] = \theta$.
> 2. Si calcola la **varianza** dello stimatore: $\mathbb{V}(\hat{\theta}_n)$.
> 3. Si calcola lo **score** $\frac{\partial}{\partial \theta}\log p_Y(y;\theta)$ e l'**informazione di Fisher** $I_n(\theta)$.
> 4. Si verifica se vale l'uguaglianza
>    $$
>    \mathbb{V}(\hat{\theta}_n) = \frac{1}{I_n(\theta)}.
>    $$
>    Se sì, lo stimatore è **efficiente** (raggiunge il limite di Cramér–Rao). Se no, lo stimatore non è efficiente e il rapporto
>    $$
>    \operatorname{eff}(\hat{\theta}_n) = \frac{1}{\mathbb{V}(\hat{\theta}_n) I_n(\theta)} \in [0,1]
>    $$
>    misura quanto lo stimatore si avvicina al limite inferiore.

#### Esempio 8.6.3 — Bernoulli

Sia $Y \sim \operatorname{Bernoulli}(p)$. Allora

$$
\log p_Y(y;p) = y \log p + (1-y)\log(1-p).
$$

Lo score è

$$
U_1(p;Y)
=
\frac{Y}{p} - \frac{1-Y}{1-p}
=
\frac{Y-p}{p(1-p)}.
$$

Quindi

$$
I_1(p)
=
\mathbb{E}_p\left[
\frac{(Y-p)^2}{p^2(1-p)^2}
\right]
=
\frac{p(1-p)}{p^2(1-p)^2}
=
\frac{1}{p(1-p)}.
$$

Per un campione di numerosità $n$,

$$
I_n(p) = \frac{n}{p(1-p)}.
$$

Lo stimatore $\hat{p}_n = \overline{Y}$ è non distorto e

$$
\operatorname{Var}_p(\hat{p}_n)
=
\frac{p(1-p)}{n}
=
\frac{1}{I_n(p)}.
$$

Quindi $\overline{Y}$ è efficiente per $p$.

#### Esempio 8.6.4 — Normale con media nota

Sia $Y \sim \mathcal{N}(\mu, \sigma^2)$ con $\mu$ nota e $\theta = \sigma^2$. Allora

$$
\log p_Y(y;\sigma^2) = -\frac{1}{2}\log(2\pi\sigma^2) - \frac{(y-\mu)^2}{2\sigma^2}.
$$

Derivando due volte rispetto a $\sigma^2$ e prendendo il valore atteso si ottiene

$$
I_1(\sigma^2) = \frac{1}{2\sigma^4}.
$$

Quindi

$$
I_n(\sigma^2) = \frac{n}{2\sigma^4}.
$$

Il limite di Cramér–Rao è $2\sigma^4/n$. Lo stimatore

$$
\hat{\sigma}^2_\mu = \frac{1}{n}\sum_{i=1}^n (Y_i - \mu)^2
$$

ha varianza $2\sigma^4/n$ e quindi è efficiente.

#### Esempio 8.6.5 — Poisson (metodo operativo)

Sia $Y \sim \operatorname{Poisson}(\lambda)$ e sia $Y_1, \dots, Y_n$ un campione i.i.d. tratto da $Y$. Vogliamo mostrare che lo stimatore media campionaria $\overline{Y}$ è efficiente.

**Passo 1 — Non distorsione.**

$$
\mathbb{E}(\overline{Y}) = \mathbb{E}(Y) = \lambda.
$$

Lo stimatore è quindi corretto.

**Passo 2 — Varianza.**

$$
\mathbb{V}(\overline{Y}) = \frac{\mathbb{V}(Y)}{n} = \frac{\lambda}{n}.
$$

**Passo 3 — Informazione di Fisher.**

Poiché

$$
p_Y(y;\lambda) = e^{-\lambda}\frac{\lambda^y}{y!},
$$

si ha

$$
\log p_Y(y;\lambda) = -\lambda + y \log \lambda - \log(y!),
$$

e

$$
\frac{\partial}{\partial \lambda}\log p_Y(y;\lambda) = -1 + \frac{y}{\lambda} = \frac{y - \lambda}{\lambda}.
$$

Il quadrato è

$$
\left(\frac{\partial}{\partial \lambda}\log p_Y(y;\lambda)\right)^2 = \frac{(y-\lambda)^2}{\lambda^2}.
$$

Prendendo il valore atteso:

$$
I_1(\lambda) = \mathbb{E}\left[\frac{(Y-\lambda)^2}{\lambda^2}\right] = \frac{\mathbb{V}(Y)}{\lambda^2} = \frac{\lambda}{\lambda^2} = \frac{1}{\lambda}.
$$

Pertanto

$$
I_n(\lambda) = n I_1(\lambda) = \frac{n}{\lambda}.
$$

**Passo 4 — Confronto con il limite di Cramér–Rao.**

$$
\frac{1}{I_n(\lambda)} = \frac{\lambda}{n} = \mathbb{V}(\overline{Y}).
$$

Poiché la varianza dello stimatore coincide con il limite inferiore di Cramér–Rao, $\overline{Y}$ è **efficiente** per $\lambda$.

> **Osservazione:** Se lo stimatore raggiunge il limite di Cramér–Rao, possiamo concludere che è lo stimatore con varianza minima tra tutti gli stimatori non distorti del parametro. In questo senso, l'efficienza è una proprietà di **ottimalità**.

### 8.7 Efficienza asintotica

#### Definizione 8.7.1 — Efficienza relativa asintotica

Siano $\hat{\theta}_{1,n}$ e $\hat{\theta}_{2,n}$ due stimatori di $\theta$. L'**efficienza relativa asintotica** di $\hat{\theta}_{1,n}$ rispetto a $\hat{\theta}_{2,n}$ è

$$
\operatorname{ARE}(\hat{\theta}_{1,n}, \hat{\theta}_{2,n})
=
\lim_{n \to +\infty}
\frac{\operatorname{MSE}_\theta(\hat{\theta}_{2,n})}
{\operatorname{MSE}_\theta(\hat{\theta}_{1,n})},
$$

se il limite esiste.

Se $\operatorname{ARE} > 1$, allora $\hat{\theta}_{1,n}$ è asintoticamente più efficiente di $\hat{\theta}_{2,n}$.

Per stimatori asintoticamente non distorti, il confronto si riduce spesso al confronto delle varianze asintotiche:

$$
\operatorname{ARE}(\hat{\theta}_{2,n}, \hat{\theta}_{1,n})
=
\lim_{n\to+\infty} \frac{\mathbb{V}(\hat{\theta}_{1,n})}{\mathbb{V}(\hat{\theta}_{2,n})}.
$$

> **Osservazione 8.7.2.**  
> Per gli stimatori di massima verosimiglianza, sotto condizioni di regolarità, vale
> $$
> \sqrt{n}(\hat{\theta}_{\text{MLE}} - \theta) \xrightarrow{D} \mathcal{N}\!\left(0, \frac{1}{I_1(\theta)}\right),
> $$
> e quindi la varianza asintotica è $1/(n I_1(\theta))$, che coincide con il limite di Cramér–Rao asintotico. Dunque gli stimatori di massima verosimiglianza sono **asintoticamente efficienti**.

### 8.8 Consistenza (coerenza)

#### Definizione 8.8.1 — Consistenza debole

Uno stimatore $\hat{\theta}_n$ è **consistente in senso debole** se

$$
\hat{\theta}_n \xrightarrow{P_\theta} \theta
\qquad \forall \theta \in \Theta,
$$

cioè per ogni $\varepsilon > 0$,

$$
P_\theta\left(
|\hat{\theta}_n - \theta| > \varepsilon
\right)
\to 0
\qquad (n \to +\infty).
$$

> **Interpretazione.** Al crescere della dimensione campionaria, lo stimatore si avvicina sempre di più al vero valore del parametro, con probabilità crescente.

> **Osservazione:** La consistenza è in senso debole in quanto la convergenza richiesta è in probabilità. Esiste tuttavia una **consistenza forte**, dove viene richiesta la convergenza quasi certa, e una **consistenza in media quadratica**, dove si richiede $\mathbb{E}_\theta[(\hat{\theta}_n - \theta)^2] \to 0$.

> **Osservazione:** La consistenza rappresenta la proprietà necessaria per uno stimatore, nel senso che un buon stimatore può non essere corretto o efficiente ma deve essere almeno consistente per essere considerato tale.

#### Teorema 8.8.2 — Condizione sufficiente tramite MSE

Se

$$
\lim_{n \to +\infty}
\operatorname{MSE}_\theta(\hat{\theta}_n) = 0
\qquad \forall \theta \in \Theta,
$$

allora $\hat{\theta}_n$ è consistente in senso debole.

**Dimostrazione.**  
Per la disuguaglianza di Markov, per ogni $\varepsilon > 0$,

$$
P_\theta\left(
|\hat{\theta}_n - \theta| > \varepsilon
\right)
\le
\frac{
\mathbb{E}_\theta[(\hat{\theta}_n - \theta)^2]
}{\varepsilon^2}
=
\frac{
\operatorname{MSE}_\theta(\hat{\theta}_n)
}{\varepsilon^2}.
$$

Se $\operatorname{MSE}_\theta(\hat{\theta}_n) \to 0$, allora il membro di destra tende a $0$. $\square$

> **Osservazione:** richiedere che l'MSE $\to 0$ per $n \to +\infty$ equivale a richiedere la convergenza in media quadratica.

#### Corollario 8.8.3

Se

$$
\lim_{n \to +\infty} \mathbb{E}_\theta[\hat{\theta}_n] = \theta
\qquad \text{e} \qquad
\lim_{n \to +\infty} \operatorname{Var}_\theta(\hat{\theta}_n) = 0,
$$

allora $\hat{\theta}_n$ è consistente in senso debole.

**Dimostrazione.**  
Per la decomposizione bias–varianza,

$$
\operatorname{MSE}_\theta(\hat{\theta}_n)
=
\operatorname{Var}_\theta(\hat{\theta}_n)
+
\mathcal{B}_\theta(\hat{\theta}_n)^2.
$$

Se entrambi i termini tendono a $0$, anche l'MSE tende a $0$. Si applica il Teorema 8.8.2. $\square$

> **Osservazione:** condizione necessaria della consistenza risulta quindi essere la non distorsione asintotica.

#### Controesempio 8.8.4 — Stimatore non consistente

Sia $Y_1,\dots,Y_n$ un campione i.i.d. con media $\mu$. Consideriamo

$$
\hat{\mu}_n = Y_1.
$$

Allora $\hat{\mu}_n$ è non distorto:

$$
\mathbb{E}[\hat{\mu}_n] = \mu,
$$

ma

$$
\operatorname{Var}(\hat{\mu}_n) = \sigma^2,
$$

che non tende a $0$. Quindi $\hat{\mu}_n$ non è consistente.

#### Teorema 8.8.5 — Legge dei grandi numeri

Sia $\{Y_n\}$ una successione di variabili aleatorie i.i.d. con $\mathbb{E}(Y_n) = \mu$ e $\operatorname{Var}(Y_n) = \sigma^2 < \infty$. Allora

$$
\overline{Y}_n = \frac{1}{n}\sum_{i=1}^n Y_i \xrightarrow{q.c.} \mu.
$$

In particolare, $\overline{Y}_n$ è consistente (forte) per $\mu$.

**Conseguenza.** La media campionaria è uno stimatore consistente per la media della popolazione, sotto l'ipotesi di varianza finita.

## 9. Esempio completo: Bernoulli

Sia $Y \sim \operatorname{Bernoulli}(p)$, con $p \in (0,1)$. Consideriamo un campione i.i.d. $Y_1, \dots, Y_n$.

### 9.1 Modello

$$
p_Y(y; p) = p^y(1-p)^{1-y}, \qquad y \in \{0, 1\}.
$$

Densità congiunta:

$$
p_{\mathbf{Y}}(\mathbf{y}; p)
=
p^{\sum_{i=1}^n y_i}(1-p)^{n - \sum_{i=1}^n y_i}.
$$

### 9.2 Stimatore

$$
\hat{p}_n = \overline{Y} = \frac{1}{n} \sum_{i=1}^n Y_i.
$$

### 9.3 Distribuzione campionaria

$$
n\hat{p}_n = \sum_{i=1}^n Y_i \sim \operatorname{Bin}(n,p).
$$

Quindi

$$
P\left(\hat{p}_n = \frac{k}{n}\right)
=
\binom{n}{k} p^k (1-p)^{n - k}, \qquad k = 0, 1, \dots, n.
$$

### 9.4 Non distorsione

$$
\mathbb{E}[\hat{p}_n] = p \implies \mathcal{B}_p(\hat{p}_n) = \mathbb{E}[\hat{p}_n] - p = 0.
$$

Lo stimatore è quindi non distorto.

### 9.5 Varianza e MSE

$$
\operatorname{Var}_p(\hat{p}_n) = \frac{p(1-p)}{n}.
$$

Poiché lo stimatore è non distorto,

$$
\operatorname{MSE}_p(\hat{p}_n) = \operatorname{Var}_p(\hat{p}_n) = \frac{p(1-p)}{n}.
$$

### 9.6 Efficienza

$$
I_1(p) = \frac{1}{p(1-p)}, \qquad I_n(p) = \frac{n}{p(1-p)}.
$$

Si ha

$$
\operatorname{Var}_p(\hat{p}_n) = \frac{p(1-p)}{n} = \frac{1}{I_n(p)}.
$$

Quindi $\hat{p}_n$ è **efficiente**.

### 9.7 Consistenza

Poiché

$$
\lim_{n \to +\infty} \mathbb{E}[\hat{p}_n] = p
\qquad \text{e} \qquad
\lim_{n \to +\infty} \operatorname{Var}_p(\hat{p}_n) = \lim_{n \to +\infty} \frac{p(1-p)}{n} = 0,
$$

per il Corollario 8.8.3 lo stimatore $\hat{p}_n$ è **consistente in senso debole**.

### 9.8 Sufficienza

Per il criterio di fattorizzazione, $T_n = \sum_{i=1}^n Y_i$ è sufficiente per $p$.

## 10. Stima della media campionaria

Volendo stimare la media $\mu$ di una variabile aleatoria, la strada più intuitiva consiste nel calcolare la media delle osservazioni campionarie.

### Definizione 10.1 — Media campionaria

Dato un campione $Y_1, \dots, Y_n$ da una variabile aleatoria $Y$, la **media campionaria** è

$$
\overline{Y} = \frac{\sum_{i=1}^n Y_i}{n}.
$$

### Proprietà 10.2

Siano $Y_1, \dots, Y_n$ i.i.d. con $\mathbb{E}(Y_i) = \mu$ e $\operatorname{Var}(Y_i) = \sigma^2$. Allora:

1. **Non distorsione:**
   $$
   \mathbb{E}(\overline{Y}) = \mu.
   $$
   Quindi $\overline{Y}$ è uno stimatore corretto per $\mu$.

2. **Varianza:**
   $$
   \operatorname{Var}(\overline{Y}) = \frac{\sigma^2}{n}.
   $$

3. **Consistenza:**
   $$
   \operatorname{MSE}(\overline{Y}) = \operatorname{Var}(\overline{Y}) + \mathcal{B}(\overline{Y})^2 = \frac{\sigma^2}{n} \xrightarrow{n \to +\infty} 0.
   $$
   Quindi $\overline{Y}$ è consistente per $\mu$.

4. **Distribuzione campionaria (caso normale):** se $Y \sim \mathcal{N}(\mu, \sigma^2)$, allora
   $$
   \overline{Y} \sim \mathcal{N}\!\left(\mu, \frac{\sigma^2}{n}\right),
   $$
   e $\overline{Y}$ è **efficiente** per $\mu$.

5. **Teorema del limite centrale:** se $Y$ ha media $\mu$ e varianza $\sigma^2$ finite, allora
   $$
   \frac{\overline{Y} - \mu}{\sigma/\sqrt{n}} \xrightarrow{d} \mathcal{N}(0,1).
   $$

## 11. Stima della varianza

Volendo stimare la varianza $\sigma^2$ di una variabile aleatoria, consideriamo dapprima uno stimatore che presuppone nota la media.

### Definizione 11.1 — Stimatore della varianza con media nota

$$
\hat{\sigma}^2_\mu = \frac{\sum_{i=1}^n (Y_i - \mu)^2}{n}.
$$

> **Proprietà.** Se $Y_i \sim \mathcal{N}(\mu, \sigma^2)$, allora $\hat{\sigma}^2_\mu$ è:
> - **corretto**: $\mathbb{E}[\hat{\sigma}^2_\mu] = \sigma^2$;
> - **consistente**;
> - con distribuzione campionaria
>   $$
>   \hat{\sigma}^2_\mu \sim \frac{\sigma^2}{n}\chi^2_n.
>   $$

Tuttavia, nella pratica $\mu$ non è nota ed è necessario sostituirla con una sua stima.

### Definizione 11.2 — Varianza campionaria

$$
\hat{\sigma}^2 = \frac{\sum_{i=1}^n (Y_i - \overline{Y})^2}{n}.
$$

> **Proprietà.**
> - La varianza campionaria è uno stimatore **distorto** per $\sigma^2$:
>   $$
>   \mathbb{E}(\hat{\sigma}^2) = \frac{n-1}{n}\sigma^2.
>   $$
> - È tuttavia **asintoticamente corretto**: $\mathbb{E}(\hat{\sigma}^2) \to \sigma^2$ per $n \to \infty$.

### Definizione 11.3 — Varianza campionaria corretta

$$
S^2 = \frac{\sum_{i=1}^n (Y_i - \overline{Y})^2}{n-1}.
$$

> **Proprietà.**
> - $S^2$ è uno stimatore **corretto** per $\sigma^2$: $\mathbb{E}[S^2] = \sigma^2$.
> - $S^2$ è **consistente** per $\sigma^2$.
> - Se $Y \sim \mathcal{N}(\mu, \sigma^2)$, allora
>   $$
>   S^2 \sim \frac{\sigma^2}{n-1}\chi^2_{n-1}.
>   $$

### Esempio 11.4

Sia $Y \sim \mathcal{N}(0, \sigma^2)$. Per $n = 5$ e campione $\mathbf{y} = (0.5, -0.3, 0.8, -0.1, 0.4)$:

$$
\overline{y} = \frac{0.5 - 0.3 + 0.8 - 0.1 + 0.4}{5} = 0.26.
$$

$$
\hat{\sigma}^2 = \frac{(0.5-0.26)^2 + (-0.3-0.26)^2 + (0.8-0.26)^2 + (-0.1-0.26)^2 + (0.4-0.26)^2}{5} \approx 0.1624.
$$

$$
S^2 = \frac{5}{4} \cdot 0.1624 \approx 0.203.
$$

La stima corretta $S^2$ è maggiore di $\hat{\sigma}^2$, come atteso.

## 12. Metodi per la costruzione di stimatori

Uno stimatore è una funzione

$$
\hat{\theta}_n = T_n = g(Y_1, \dots, Y_n)
$$

del campione, utilizzata per stimare un parametro $\theta$. Una volta definito uno stimatore è possibile caratterizzarne la qualità in base a proprietà come sufficienza, non distorsione, efficienza e consistenza.

Per la costruzione di $T$ esistono vari metodi:

- metodo dell'analogia tra stimatore e parametro;
- metodo dei momenti;
- metodo dei minimi quadrati;
- metodo della massima verosimiglianza.

---

### 12.1 Metodo dell'analogia

Il **metodo dell'analogia** è una classe di procedure euristiche che definisce uno stimatore in base all'analogia tra lo stimatore stesso e il parametro di interesse.

**Esempi.**

- Per stimare la media della popolazione si usa la media campionaria:
  $$
  \mu \longleftrightarrow \overline{Y}.
  $$
- Per stimare la proporzione della popolazione si usa la proporzione campionaria:
  $$
  p \longleftrightarrow \frac{\sum_{i=1}^n Y_i}{n}.
  $$
- Se $Y \sim \operatorname{Exp}(\lambda)$, essendo $\mathbb{E}(Y) = 1/\lambda$, una stima naturale di $\lambda$ è
  $$
  \hat{\lambda} = \frac{1}{\overline{Y}} = \frac{n}{\sum_{i=1}^n Y_i}.
  $$

> **Attenzione.** Talvolta l'analogo campionario non è la scelta migliore. Per esempio, per stimare la varianza della popolazione si può usare la varianza campionaria, ma tale stimatore è distorto. Inoltre, esistono spesso più analoghi campionari possibili.

#### Esempio 12.1.1 — Stima di $\lambda$ per una legge esponenziale

Sia $Y \sim \operatorname{Exp}(\lambda)$, con $\mathbb{E}(Y) = 1/\lambda$. Il metodo dell'analogia suggerisce

$$
\hat{\lambda} = \frac{1}{\overline{Y}}.
$$

Per il metodo dei momenti si ottiene lo stesso stimatore, che è distorto per $n$ finito:

$$
\mathbb{E}[\hat{\lambda}] = \frac{n\lambda}{n-1} > \lambda.
$$

---

### 12.2 Metodo dei momenti

Il **metodo dei momenti** conduce alla definizione di stimatori "naturali" dei parametri. Richiede due condizioni:

1. **Esistenza dei momenti** della variabile aleatoria di interesse in numero pari al numero di parametri da stimare.
   - Se $Y \sim \operatorname{Ga}(\alpha, \lambda)$, abbiamo garanzia che $\mathbb{E}(Y)$ ed $\mathbb{E}(Y^2)$ esistono finiti, quindi possiamo stimare sia $\alpha$ sia $\lambda$.
2. **Conoscenza delle relazioni** che legano i momenti ai parametri da stimare.
   - Se $Y \sim \operatorname{Ga}(\alpha, \lambda)$, sappiamo che
     $$
     \mathbb{E}(Y) = \frac{\alpha}{\lambda}, \qquad
     \operatorname{Var}(Y) = \mathbb{E}(Y^2) - \mathbb{E}(Y)^2 = \frac{\alpha}{\lambda^2}.
     $$

#### Definizione 12.2.1 — Stimatore del metodo dei momenti

Poiché i momenti di $Y \sim p_Y(y;\theta)$ con $\theta \in \mathbb{R}^m$ sono funzioni dei parametri $\mathbb{E}(Y^r) = g_r(\theta)$, il metodo dei momenti consiste nello stimare i momenti teorici con i corrispondenti **momenti campionari** e nel risolvere il sistema

$$
\frac{1}{n}\sum_{i=1}^n Y_i^r = g_r(\theta), \qquad r = 1, \dots, m.
$$

#### Esempio 12.2.2 — Distribuzione Gamma

Se $Y \sim \operatorname{Ga}(\alpha, \lambda)$, risolvendo il sistema

$$
\begin{cases}
\dfrac{1}{n}\sum_{i=1}^n Y_i = \dfrac{\alpha}{\lambda} \\[6pt]
\dfrac{1}{n}\sum_{i=1}^n Y_i^2 - \left(\dfrac{1}{n}\sum_{i=1}^n Y_i\right)^2 = \dfrac{\alpha}{\lambda^2}
\end{cases}
$$

si ottiene

$$
\hat{\lambda} = \frac{\frac{1}{n}\sum_i Y_i}{\frac{1}{n}\sum_i Y_i^2 - \left(\frac{1}{n}\sum_i Y_i\right)^2},
\qquad
\hat{\alpha} = \frac{\left(\frac{1}{n}\sum_i Y_i\right)^2}{\frac{1}{n}\sum_i Y_i^2 - \left(\frac{1}{n}\sum_i Y_i\right)^2}.
$$

#### Esempio 12.2.3 — Normale con entrambi i parametri incogniti

Sia $Y \sim \mathcal{N}(\mu, \sigma^2)$ con $\mu$ e $\sigma^2$ incogniti. Il sistema dei momenti è

$$
\begin{cases}
\overline{Y} = \mu \\[4pt]
\dfrac{1}{n}\sum_i Y_i^2 - \overline{Y}^2 = \sigma^2
\end{cases}
$$

da cui

$$
\hat{\mu} = \overline{Y}, \qquad \hat{\sigma}^2 = \frac{1}{n}\sum_{i=1}^n (Y_i - \overline{Y})^2.
$$

Lo stimatore di $\sigma^2$ è distorto ma asintoticamente non distorto.

#### Esempio 12.2.4 — Distribuzione uniforme

Sia $Y \sim U(0, \theta)$. Allora

$$
\mathbb{E}(Y) = \frac{\theta}{2}
\implies
\hat{\theta}_n = 2\overline{Y}
$$

per il metodo dei momenti. Questo stimatore è non distorto e consistente, ma non è il più efficiente (lo stimatore basato sul massimo campionario ha varianza minore).

#### Proprietà 12.2.5

Gli stimatori ottenuti con il metodo dei momenti sono funzioni dei momenti campionari e, in quanto tali, esprimibili come somme pesate di quantità aleatorie. Essi sono pertanto:

- **consistenti**;
- **asintoticamente non distorti**;
- **asintoticamente normali** (per il teorema del limite centrale).

Non è tuttavia garantita:

- la loro **correttezza finita**;
- la loro **efficienza**;
- la loro **coerenza** in termini di determinazione di una stima espressa nella stessa unità di misura di $Y$.

> **Osservazione:** il metodo dei momenti fornisce stimatori spesso non ottimali.

---

### 12.3 Metodo dei minimi quadrati

In molte situazioni è ragionevole assumere che il fenomeno di interesse $Y \sim p_Y$ sia esprimibile come

$$
Y_i = g_i(\theta) + \epsilon_i,
$$

dove $g_i(\theta)$ è una funzione nota e deterministica che dipende dal parametro $\theta$, e $\epsilon_i$ è una quantità stocastica che tipicamente esprime l'effetto di errori accidentali.

Si assume che

$$
\mathbb{E}(\epsilon_i) = 0, \qquad \operatorname{Var}(\epsilon_i) = \sigma^2, \qquad \operatorname{Cov}(\epsilon_i, \epsilon_j) = 0 \quad (i \neq j).
$$

#### Definizione 12.3.1 — Stimatore dei minimi quadrati

Il metodo dei minimi quadrati stima $\theta$ minimizzando la distanza

$$
\hat{\theta} = \arg\min_{\theta} \sum_{i=1}^n (y_i - g_i(\theta))^2.
$$

Se $g$ è una funzione regolare, $\hat{\theta}$ soddisfa

$$
\frac{d}{d\theta} \sum_{i=1}^n (y_i - g_i(\theta))^2 = 0,
\qquad
\frac{d^2}{d\theta^2} \sum_{i=1}^n (y_i - g_i(\theta))^2 > 0.
$$

> **Caso lineare.** Se $g_i(\theta) = \theta x_i$, il metodo produce stimatori **corretti ed efficienti**.

#### Esempio 12.3.2 — Regressione lineare semplice

Sia $Y_i = \theta x_i + \epsilon_i$. La funzione obiettivo è

$$
Q(\theta) = \sum_{i=1}^n (y_i - \theta x_i)^2.
$$

Derivando rispetto a $\theta$:

$$
Q'(\theta) = -2 \sum_{i=1}^n x_i (y_i - \theta x_i) = 0
\implies
\hat{\theta} = \frac{\sum_i x_i y_i}{\sum_i x_i^2}.
$$

> **Osservazione:** Possiamo adottare il metodo dei minimi quadrati quando possiamo esprimere $Y_i$ come $Y_i = g_i(\theta) + \epsilon_i$, la funzione $g_i$ è nota e deterministica, $\epsilon_i$ è una quantità stocastica solitamente avente $\mathbb{E}(\epsilon_i) = 0$, $\operatorname{Var}(\epsilon_i) = \sigma^2$ e $\operatorname{Cov}(\epsilon_i, \epsilon_j) = 0$ per $i \neq j$.

---

### 12.4 Metodo della massima verosimiglianza

Il **metodo della massima verosimiglianza** si basa su un principio elementare: tra tutti i possibili valori di $\theta$, viene selezionato quello che corrisponde alla massima probabilità di generare i dati osservati.

Se il campione $Y_1, \dots, Y_n$ deriva da un modello statistico $p_Y(y;\theta)$, si può calcolare la probabilità della specifica $n$-upla osservata, fissato $\theta = \theta_0$:

$$
p_{Y_1, \dots, Y_n}(y_1, \dots, y_n; \theta = \theta_0)
=
\prod_{i=1}^n p_Y(y_i; \theta_0).
$$

Se si calcola per $\theta = \theta_1 \neq \theta_0$, si ottiene

$$
p_{Y_1, \dots, Y_n}(y_1, \dots, y_n; \theta = \theta_1)
=
\prod_{i=1}^n p_Y(y_i; \theta_1)
\neq
p_{Y_1, \dots, Y_n}(y_1, \dots, y_n; \theta = \theta_0).
$$

#### Definizione 12.4.1 — Funzione di verosimiglianza

Si chiama **verosimiglianza** la funzione che si ottiene, fissato il campione osservato, al variare di tutti i possibili $\theta \in \Theta$:

$$
L(\theta; y_1, \dots, y_n)
=
p_{Y_1, \dots, Y_n}(y_1, \dots, y_n; \theta)
=
\prod_{i=1}^n p_Y(y_i; \theta).
$$

La **stima di massima verosimiglianza** (MLE) è il valore

$$
\hat{\theta}_{\text{MLE}} = \arg\max_{\theta \in \Theta} L(\theta; y_1, \dots, y_n),
$$

ovvero il parametro più plausibile alla luce dei dati.

> **Osservazione 12.4.2.**  
> Rispetto agli altri metodi di stima, la massima verosimiglianza richiede esplicitamente la specificazione del modello statistico sottostante i dati.

> **Osservazione 12.4.3 — Log-verosimiglianza.**  
> Poiché la funzione logaritmo è strettamente crescente, massimizzare $L(\theta;\mathbf{y})$ equivale a massimizzare la **log-verosimiglianza**
> $$
> \ell(\theta;\mathbf{y}) = \log L(\theta;\mathbf{y}) = \sum_{i=1}^n \log p_Y(y_i;\theta).
> $$
> In pratica, si lavora sempre con $\ell$ perché trasforma i prodotti in somme, semplificando il calcolo delle derivate.

#### Esempio 12.4.4 — MLE per la Bernoulli

Sia $Y \sim \operatorname{Bernoulli}(p)$. La log-verosimiglianza è

$$
\ell(p;\mathbf{y}) = \left(\sum_i y_i\right)\log p + \left(n - \sum_i y_i\right)\log(1-p).
$$

Derivando rispetto a $p$:

$$
\ell'(p) = \frac{\sum_i y_i}{p} - \frac{n - \sum_i y_i}{1-p} = 0
\implies
\hat{p}_{\text{MLE}} = \overline{y}.
$$

#### Esempio 12.4.5 — MLE per la Normale

Sia $Y \sim \mathcal{N}(\mu, \sigma^2)$ con entrambi i parametri incogniti. La log-verosimiglianza è

$$
\ell(\mu, \sigma^2) = -\frac{n}{2}\log(2\pi) - \frac{n}{2}\log\sigma^2 - \frac{1}{2\sigma^2}\sum_{i=1}^n (y_i - \mu)^2.
$$

Derivando rispetto a $\mu$ e $\sigma^2$ e ponendo le derivate a zero:

$$
\hat{\mu}_{\text{MLE}} = \overline{y},
\qquad
\hat{\sigma}^2_{\text{MLE}} = \frac{1}{n}\sum_{i=1}^n (y_i - \overline{y})^2.
$$

#### Proprietà 12.4.6 — Proprietà asintotiche degli stimatori di massima verosimiglianza

Sotto condizioni di regolarità, gli stimatori di massima verosimiglianza godono delle seguenti proprietà:

- **non distorsione asintotica**;
- **consistenza**;
- **efficienza asintotica**;
- **normalità asintotica**:
  $$
  \sqrt{n}(\hat{\theta}_{\text{MLE}} - \theta) \xrightarrow{d} \mathcal{N}\!\left(0, \frac{1}{I_1(\theta)}\right).
  $$

> **Osservazione 12.4.7 — Invarianza.**  
> Se $\hat{\theta}_{\text{MLE}}$ è la stima di massima verosimiglianza di $\theta$ e $g$ è una funzione misurabile, allora $g(\hat{\theta}_{\text{MLE}})$ è la stima di massima verosimiglianza di $g(\theta)$. Questa proprietà è nota come **invarianza della MLE**.
