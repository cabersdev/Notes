# Stima di massima verosimiglianza
---

## 0. Setup

### Osservazione 0.1

Sia $\mathbf{Y}=(Y_1,\dots,Y_n)$ un campione casuale con distribuzione congiunta $p_{\mathbf{Y}}(\mathbf{y};\theta)$, $\theta\in\Theta\subseteq\mathbb{R}^p$. La **funzione di verosimiglianza** basata sul campione osservato $\mathbf{y}_{\text{oss}}$ è
$$
L(\theta)=L(\theta;\mathbf{y}_{\text{oss}})=c(\mathbf{y}_{\text{oss}})\,p_{\mathbf{Y}}(\mathbf{y}_{\text{oss}};\theta),
$$
dove $c(\mathbf{y}_{\text{oss}})>0$ non dipende da $\theta$. La **log-verosimiglianza** è $\ell(\theta)=\log L(\theta)$.

La verosimiglianza fornisce una **misura di plausibilità** dei valori di $\theta$ alla luce dei dati. L'idea della stima di massima verosimiglianza è scegliere il valore di $\theta$ che rende massimamente plausibile il campione osservato.

---

## 1. Definizione di stima di massima verosimiglianza

### Definizione 1.1 — Stima di massima verosimiglianza

Sia $L(\theta)$ una funzione di verosimiglianza costruita sul modello statistico $\mathcal{F}=\{p_{\mathbf{Y}}(\cdot;\theta):\theta\in\Theta\}$ e sui dati osservati $\mathbf{y}_{\text{oss}}$. Un valore $\hat\theta\in\Theta$ è detto **stima di massima verosimiglianza** (SMV) di $\theta$ se
$$
L(\theta)\le L(\hat\theta)\qquad\forall\theta\in\Theta,
$$
cioè se
$$
L(\hat\theta)=\max_{\theta\in\Theta}L(\theta).
$$
Equivalentemente, poiché $\log$ è strettamente crescente,
$$
\ell(\hat\theta)=\max_{\theta\in\Theta}\ell(\theta).
$$

> **Osservazione 1.2.**  
> La SMV è un **valore** di $\theta$, non uno stimatore. Quando si vuole sottolineare la dipendenza dai dati, si scrive $\hat\theta(\mathbf{y}_{\text{oss}})$ e si parla di **stimatore di massima verosimiglianza**. Lo stimatore è una funzione misurabile del campione; la stima è il suo valore in corrispondenza di $\mathbf{y}_{\text{oss}}$.

> **Osservazione 1.3.**  
> L'uso di $L$ o $\ell$ è equivalente: massimizzare $L$ o $\ell$ produce lo stesso $\hat\theta$. Tuttavia $\ell$ è spesso più comoda perché trasforma prodotti in somme e rapporti in differenze.

### Osservazione 1.4 — Esistenza e unicità

La SMV può:

1. **esistere** e essere **unica**;
2. **esistere** ma **non essere unica** (più punti di massimo con lo stesso valore);
3. **non esistere** (il massimo non è raggiunto in $\Theta$);
4. esistere ma **non essere esprimibile in forma chiusa** (richiede metodi numerici).

La discussione che segue chiarisce in quali condizioni ciascuna situazione si verifica.

## 2. Problemi di stima regolari

### Definizione 2.1 — Problema di stima regolare

Un problema di stima è detto **regolare** se:

1. lo spazio parametrico $\Theta$ è un **intervallo aperto** contenuto in $\mathbb{R}^p$;
2. le funzioni di densità (o di probabilità) che specificano $\mathcal{F}$ hanno **tutte lo stesso supporto** (che quindi non dipende da $\theta$);
3. per gli elementi di $\mathcal{F}$ è possibile **scambiare due volte** il segno di integrale con quello di derivata rispetto a $\theta$.

### Osservazione 2.2

Le tre condizioni servono a garantire:

- **condizione 1**: $\Theta$ è aperto, quindi si può parlare di derivate in ogni punto e non ci sono "punti di frontiera" che creano asimmetrie;
- **condizione 2**: il supporto non dipende da $\theta$, quindi la verosimiglianza è una funzione derivabile del parametro (la dipendenza da $\theta$ è "liscia", non attraverso indicatori);
- **condizione 3**: si può derivare sotto il segno di integrale, il che permette di scambiare derivazione e integrazione nelle dimostrazioni delle proprietà asintotiche.

> **Osservazione 2.3.**  
> Nelle condizioni di regolarità, la verosimiglianza è una funzione **liscia** di $\theta$ e la stima può essere determinata con gli strumenti del calcolo differenziale. Inoltre, l'inferenza che ne deriva gode di **ottime proprietà teoriche** (normalità asintotica, efficienza, invarianza). Al di fuori delle condizioni di regolarità, queste proprietà possono fallire e l'analisi richiede procedure ad hoc.

## 3. Funzione punteggio e informazione osservata

### Definizione 3.1 — Funzione punteggio (score)

Sia $\ell(\theta)$ la log-verosimiglianza, $\theta\in\Theta\subseteq\mathbb{R}^p$. La **funzione punteggio** (o **score**) è la derivata prima di $\ell$:
$$
\ell^*(\theta)=
\begin{cases}
\dfrac{d\ell(\theta)}{d\theta}, & p=1,\\[2ex]
\nabla_\theta\ell(\theta)=\left(\dfrac{\partial\ell}{\partial\theta_1},\dots,\dfrac{\partial\ell}{\partial\theta_p}\right)^T, & p>1.
\end{cases}
$$

### Definizione 3.2 — Informazione osservata

La **matrice di informazione osservata** (o **informazione di Fisher osservata**) è l'opposto della derivata seconda di $\ell$:
$$
J(\theta)=
\begin{cases}
-\dfrac{d^2\ell(\theta)}{d\theta^2}, & p=1,\\[2ex]
-\nabla^2_\theta\ell(\theta)=-\left(\dfrac{\partial^2\ell}{\partial\theta_i\partial\theta_j}\right)_{i,j=1,\dots,p}, & p>1.
\end{cases}
$$

> **Osservazione 3.3.**  
> $J(\theta)$ è detta "informazione" perché misura quanto bene si riesce a discriminare tra il punto di massimo e i valori vicini. Se $J(\hat\theta)$ è grande, la log-verosimiglianza ha una curvatura pronunciata nel punto di massimo e la stima è "precisa"; se è piccola, la log-verosimiglianza è "piatta" e la stima è "incerta". Per il teorema di Schwarz, $J(\theta)$ è **simmetrica**.

### Proposizione 3.4 — Relazione con la derivata della verosimiglianza

Se $L>0$ e $\ell$ è derivabile, allora
$$
\ell^*(\theta)=\frac{L'(\theta)}{L(\theta)}.
$$

**Dimostrazione.**  
Per definizione $\ell=\log L$, quindi $\ell'=L'/L$. Per $p>1$ la relazione vale componente per componente. $\square$

> **Osservazione 3.5.**  
> La funzione punteggio è la derivata logaritmica della verosimiglianza. La condizione $\ell^*(\theta)=0$ è quindi equivalente a $L'(\theta)=0$ (quando $L>0$): i punti stazionari di $L$ e $\ell$ coincidono.

## 4. Determinazione della SMV in problemi regolari

### Teorema 4.1 — Condizioni per la SMV

Sia $\mathcal{F}=\{p_{\mathbf{Y}}(\cdot;\theta):\theta\in\Theta\}$ un modello regolare, e sia $\ell$ la log-verosimiglianza. Allora:

1. la SMV $\hat\theta$ è da ricercarsi tra i punti $\theta_0\in\Theta$ che soddisfano le **equazioni di verosimiglianza**
   $$
   \ell^*(\theta_0)=0;
   $$
2. se $\ell$ è **concava** su $\Theta$, ogni punto stazionario è un massimo globale e la SMV è unica (se esiste);
3. se $\ell$ è **localmente concava** nei punti stazionari, cioè se
   - per $p=1$: $J(\theta_0)>0$;
   - per $p>1$: $J(\theta_0)$ è definita positiva;
   
   e se $\ell$ tende a $-\infty$ (o a una costante) sulla frontiera di $\Theta$, allora $\hat\theta=\theta_0$.

**Dimostrazione.**  
1. Se $\hat\theta$ è un massimo interno e $\ell$ è differenziabile, allora $\nabla_\theta\ell(\hat\theta)=0$ per il teorema di Fermat. Poiché $\Theta$ è aperto, ogni massimo è interno.
2. Se $\ell$ è concava, ogni punto stazionario è un massimo globale per la definizione di concavità. Se esistono due punti stazionari distinti $\theta_1,\theta_2$ con lo stesso valore, $\ell$ è costante su tutto il segmento tra loro e la SMV non è unica; altrimenti è unica.
3. Se $\ell$ è localmente concava in $\theta_0$ e $J(\theta_0)$ è definita positiva, $\theta_0$ è un massimo locale stretto. Se $\ell\to-\infty$ sulla frontiera, il massimo globale esiste e coincide con il massimo locale. $\square$

> **Osservazione 4.2.**  
> In pratica, la determinazione della SMV si articola in tre passi:
> 1. calcolare la funzione punteggio $\ell^*$;
> 2. risolvere le equazioni di verosimiglianza $\ell^*(\theta)=0$;
> 3. verificare che i punti trovati siano massimi (tramite $J$ o concavità).
>
> Se $\ell$ è concava, il passo 3 è automatico.

### Osservazione 4.3 — Il ruolo della concavità

La concavità della log-verosimiglianza è una condizione **sufficiente** per l'unicità e la globalità del massimo. Molte famiglie esponenziali (normale, Poisson, Bernoulli, Gamma) hanno log-verosimiglianza concava, quindi il problema di stima è ben posto. In altri casi, la log-verosimiglianza può avere più massimi locali, e occorre un'analisi più fine.

## 5. Esempi

### 5.1 Caso $p=1$: modello esponenziale

Siano $Y_1,\dots,Y_n$ i.i.d. $\operatorname{Esp}(\lambda)$, $\lambda>0$, con densità $p_Y(y;\lambda)=\lambda e^{-\lambda y}\mathbf{1}_{(0,\infty)}(y)$. Si osservano i tempi di funzionamento (in migliaia di ore) di $n=5$ lampadine:
$$
\mathbf{y}_{\text{oss}}=(3.2,\,5.7,\,2.1,\,1.6,\,0.4).
$$

**Verosimiglianza.** Per l'indipendenza,
$$
L(\lambda)=\prod_{i=1}^n\lambda e^{-\lambda y_i}=\lambda^n e^{-\lambda\sum_i y_i}.
$$
Con i dati, $\sum_i y_i=13$, $n=5$:
$$
L(\lambda)=\lambda^5 e^{-13\lambda}.
$$

**Log-verosimiglianza.**
$$
\ell(\lambda)=5\log\lambda-13\lambda.
$$

**Funzione punteggio.**
$$
\ell^*(\lambda)=\frac{5}{\lambda}-13.
$$
Ponendo $\ell^*(\lambda)=0$:
$$
\frac{5}{\lambda}-13=0
\quad\Longleftrightarrow\quad
\hat\lambda=\frac{5}{13}.
$$

**Verifica del massimo.** La derivata seconda è
$$
\ell''(\lambda)=-\frac{5}{\lambda^2}<0\qquad\forall\lambda>0.
$$
Quindi $\ell$ è **concava** su $\Theta=(0,\infty)$ e l'unico punto stazionario è anche massimo globale. La SMV è
$$
\hat\lambda=\frac{5}{13}\approx0.385.
$$

> **Interpretazione.**  
> Il tasso di guasto stimato è $\hat\lambda=5/13$ guasti per migliaio di ore, cioè circa un guasto ogni $13/5=2.6$ migliaia di ore. Il valore corrisponde al reciproco della media campionaria $\bar y=13/5$: per il modello esponenziale, la SMV del tasso è l'inverso della media campionaria.

### 5.2 Caso $p>1$: modello normale

Siano $Y_1,\dots,Y_n$ i.i.d. $\mathcal{N}(\mu,\sigma^2)$, con $\theta=(\mu,\sigma^2)\in\Theta=\mathbb{R}\times(0,\infty)$. Si osservano le altezze di $n=10$ individui adulti di sesso maschile:
$$
\sum_i y_i=1770,\qquad\sum_i y_i^2=315636.
$$

**Verosimiglianza.**
$$
L(\mu,\sigma^2)=(2\pi\sigma^2)^{-n/2}\exp\left(-\frac{1}{2\sigma^2}\sum_i(y_i-\mu)^2\right).
$$

**Log-verosimiglianza.** Sviluppando il quadrato,
$$
\ell(\mu,\sigma^2)=-\frac n2\log(2\pi)-\frac n2\log\sigma^2-\frac{1}{2\sigma^2}\sum_i(y_i-\mu)^2.
$$
Omettendo la costante $-\frac n2\log(2\pi)$ (che non dipende da $\theta$):
$$
\ell(\mu,\sigma^2)\propto-\frac n2\log\sigma^2-\frac{1}{2\sigma^2}\sum_i(y_i-\mu)^2.
$$
Con $n=10$, $\sum_i y_i=1770$, $\sum_i y_i^2=315636$:
$$
\ell(\mu,\sigma^2)\propto-5\log\sigma^2-\frac{1}{2\sigma^2}\left(315636-2\mu\cdot1770+10\mu^2\right).
$$

**Funzione punteggio.**
$$
\frac{\partial\ell}{\partial\mu}=-\frac{10\mu}{\sigma^2}+\frac{1770}{\sigma^2},
\qquad
\frac{\partial\ell}{\partial\sigma^2}=-\frac{5}{\sigma^2}+\frac{315636-2\mu\cdot1770+10\mu^2}{2\sigma^4}.
$$
Ponendo le derivate a zero:
$$
\frac{\partial\ell}{\partial\mu}=0
\quad\Longleftrightarrow\quad
\hat\mu=\frac{1770}{10}=177.6.
$$
Sostituendo nella seconda equazione:
$$
-\frac{5}{\sigma^2}+\frac{315636-2\cdot177.6\cdot1770+10\cdot177.6^2}{2\sigma^4}=0.
$$
Si ottiene
$$
\hat\sigma^2=21.4.
$$

**Verifica del massimo.** La matrice Hessiana è
$$
\nabla^2\ell(\hat\mu,\hat\sigma^2)=
\begin{pmatrix}
-\dfrac{10}{\hat\sigma^2} & 0\\[2ex]
0 & -\dfrac{5}{\hat\sigma^4}
\end{pmatrix},
$$
che è **definita negativa**. Quindi $(\hat\mu,\hat\sigma^2)$ è un massimo locale stretto. Inoltre $\ell\to-\infty$ sulla frontiera di $\Theta$ (per $\sigma^2\to0$ e $\sigma^2\to\infty$), quindi il massimo è globale. La SMV è
$$
\hat\mu=177.6,\qquad\hat\sigma^2=21.4.
$$

> **Osservazione 5.1.**  
> Per la normale, la SMV della media coincide con la media campionaria e la SMV della varianza è la varianza campionaria **non corretta** (divisa per $n$, non per $n-1$). Lo stimatore corretto $S^2$ è una modificazione della SMV per renderla non distorta; la SMV è asintoticamente non distorta ma finita è distorta.

## 6. Eccezioni

Il procedimento appena enunciato si applica in una pluralità di situazioni, ma ci sono casi in cui le condizioni di regolarità falliscono o la SMV ha proprietà inattese. Le eccezioni principali sono:

1. problemi non regolari;
2. SMV non esistente finita o non appartenente a $\Theta$;
3. verosimiglianza con più punti di massimo;
4. SMV non esprimibile esplicitamente come funzione dei dati.

### 6.1 Problemi non regolari

#### 6.1.1 Spazio parametrico finito

Se $\Theta$ ha **cardinalità finita**, la SMV si determina per confronto diretto dei valori $L(\theta)$. La derivata non è applicabile perché $\Theta$ è discreto.

**Esempio 6.1.**  
Sia $\Theta=\{1,2,4,7,9\}$ e supponiamo che i dati producano la verosimiglianza
$$
L(1)=1,\quad L(2)=2,\quad L(4)=5,\quad L(7)=3,\quad L(9)=2.
$$
Allora $\hat\theta=4$ e $L(\hat\theta)=5$.

Se i dati producono
$$
L(1)=3,\quad L(2)=5,\quad L(4)=5,\quad L(7)=1,\quad L(9)=2,
$$
allora ci sono **due** massimi, $\hat\theta=2$ e $\hat\theta=4$, con $L(\hat\theta)=5$. La SMV non è unica.

> **Osservazione 6.2.**  
> Se $\Theta$ è finito, la SMV esiste sempre (il massimo di un insieme finito è raggiunto), ma può non essere unica. In questo caso la scelta tra i massimi richiede un criterio aggiuntivo (ad esempio, il più piccolo, il più grande, o un valore basato su informazioni esterne).

#### 6.1.2 Supporto dipendente dal parametro

Se il supporto della distribuzione dipende da $\theta$, la verosimiglianza può contenere **indicatori** che la rendono non derivabile. La teoria standard non si applica.

**Esempio 6.3 — Uniforme $U(0,\theta)$.**  
Siano $Y_1,\dots,Y_n$ i.i.d. $U(0,\theta)$, $\theta>0$. La densità è
$$
p_Y(y;\theta)=\frac1\theta\mathbf{1}_{(0,\theta)}(y).
$$
La verosimiglianza è
$$
L(\theta)=\frac{1}{\theta^n}\prod_{i=1}^n\mathbf{1}_{(0,\theta)}(y_i).
$$
Il prodotto degli indicatori vale $1$ se e solo se $\theta\ge y_i$ per ogni $i$, cioè se e solo se
$$
\theta\ge y^{(n)}=\max_i y_i.
$$
Pertanto
$$
L(\theta)=
\begin{cases}
\dfrac{1}{\theta^n}, & \theta\ge y^{(n)},\\[2ex]
0, & \theta<y^{(n)}.
\end{cases}
$$
Su $(0,y^{(n)})$ la verosimiglianza è nulla; su $[y^{(n)},\infty)$ è decrescente. Il massimo è in $\theta=y^{(n)}$. Quindi
$$
\hat\theta=y^{(n)}.
$$

> **Osservazione 6.4.**  
> La SMV dipende dai dati solo attraverso il massimo campionario $y^{(n)}$, che è una statistica sufficiente minimale per $\theta$. Il problema non è regolare perché il supporto dipende da $\theta$ e la verosimiglianza non è derivabile in $\theta=y^{(n)}$. L'analisi richiede un ragionamento ad hoc basato sulla forma della verosimiglianza.

### 6.2 SMV non esistente finita o non appartenente a $\Theta$

La SMV è definita come il massimo di $L$ **su** $\Theta$. Se il massimo di $L$ su un insieme più ampio non appartiene a $\Theta$, la SMV può non esistere.

**Esempio 6.5 — Normale con spazio parametrico ristretto.**  
Siano $Y_1,\dots,Y_n$ i.i.d. $\mathcal{N}(\theta,1)$, con $\theta$ il parametro di interesse. La log-verosimiglianza è
$$
\ell(\theta)\propto-\frac12\sum_i(y_i-\theta)^2.
$$
Massimizzare $\ell$ equivale a minimizzare $\sum_i(y_i-\theta)^2$, che è minimo in $\theta=\bar y$ (proprietà della media aritmetica).

- Se $\Theta=\mathbb{R}$: $\hat\theta=\bar y\in\mathbb{R}$, SMV esiste.
- Se $\Theta=\mathbb{R}^+\cup\{0\}$: se $\bar y\ge0$, $\hat\theta=\bar y$; se $\bar y<0$, $\hat\theta=0$. La SMV esiste ma non è unica se $\bar y=0$.
- Se $\Theta=\mathbb{R}^+$: se $\bar y>0$, $\hat\theta=\bar y$; se $\bar y\le0$, il massimo non è raggiunto in $\Theta$ (perché $\theta=0\notin\Theta$ e la funzione è decrescente su $\Theta$), quindi la SMV **non esiste**.

> **Osservazione 6.6.**  
> L'esempio mostra che la scelta di $\Theta$ è cruciale. Un problema di stima su $\Theta=\mathbb{R}$ può avere SMV, ma la stessa verosimiglianza ristretta a $\Theta=\mathbb{R}^+$ può non ammettere SMV. In pratica, la scelta di $\Theta$ riflette la conoscenza a priori sul parametro, e la non esistenza della SMV è un segnale che il modello o lo spazio parametrico vanno ripensati.

### 6.3 Verosimiglianza con più punti di massimo

Ci sono due casi distinti:

1. **Più massimi globali:** esistono $\theta_1\ne\theta_2$ con $L(\theta_1)=L(\theta_2)=\max_\theta L(\theta)$. In tal caso la SMV non è unica e si indica con $\hat\theta=(\theta_1,\theta_2)$.
2. **Un solo massimo globale ma più massimi locali:** la SMV esiste ed è unica, ma non basta individuare i punti stazionari e la concavità locale; occorre confrontare i valori di $\ell$ nei vari massimi locali.

**Esempio 6.7 — Cauchy.**  
Siano $Y_1,\dots,Y_n$ i.i.d. $\operatorname{Cauchy}(\theta)$, con densità
$$
p_Y(y;\theta)=\frac{1}{\pi(1+(y-\theta)^2)}.
$$
La distribuzione è unimodale e simmetrica rispetto a $\theta$ (che è la mediana), ma non ammette momenti finiti. La log-verosimiglianza è
$$
\ell(\theta)\propto-\sum_{i=1}^n\log(1+(y_i-\theta)^2).
$$
Per il campione
$$
(225,\,171,\,198,\,189,\,189,\,135,\,162,\,135,\,117,\,162),
$$
la log-verosimiglianza ammette **quattro** punti di massimo locale. Non è possibile determinare la SMV con i soli strumenti del calcolo differenziale; occorre confrontare i valori nei quattro massimi locali e scegliere il più grande. In questo caso, la molteplicità dei massimi locali suggerisce che la stima puntuale di $\theta$ potrebbe non essere la sintesi più appropriata, e che sia preferibile un approccio diverso (intervalli, stime robuste).

> **Osservazione 6.8.**  
> La Cauchy è un esempio importante perché mostra che la log-verosimiglianza può essere **multimodale** anche per un modello semplice e simmetrico. In questi casi, l'algoritmo di Newton–Raphson può convergere a un massimo locale che non è quello globale, a seconda del punto iniziale. La scelta del punto iniziale è cruciale.

### 6.4 SMV non esprimibile esplicitamente

Spesso la SMV esiste ma non è esprimibile in forma chiusa come funzione dei dati. In questi casi si usano metodi numerici.

**Esempio 6.9 — Weibull.**  
Sia $Y\sim W(\gamma,\lambda)$ con densità
$$
p_Y(y;\gamma,\lambda)=\frac{\gamma}{\lambda}y^{\gamma-1}e^{-(y/\lambda)^\gamma},\qquad\gamma>0,\ \lambda>0.
$$
Per il caso $\lambda=1$, la densità è
$$
p_Y(y;\gamma,1)=\gamma y^{\gamma-1}e^{-y^\gamma}.
$$
La log-verosimiglianza per un campione $y_1,\dots,y_n$ è
$$
\ell(\gamma)=n\log\gamma+(\gamma-1)\sum_i\log y_i-\sum_i y_i^\gamma.
$$
La funzione punteggio è
$$
\ell^*(\gamma)=\frac{n}{\gamma}+\sum_i\log y_i-\sum_i y_i^\gamma\log y_i.
$$
Si verifica che $\ell$ è concava, quindi la SMV è l'unico punto che annulla $\ell^*$. Tuttavia, l'equazione
$$
\frac{n}{\gamma}+\sum_i\log y_i-\sum_i y_i^\gamma\log y_i=0
$$
non è risolvibile esplicitamente in $\gamma$. La SMV si determina numericamente.

> **Osservazione 6.10.**  
> L'esempio mostra che la concavità della log-verosimiglianza garantisce l'unicità e la globalità del massimo, ma non la sua esprimibilità in forma chiusa. In questi casi, il calcolo numerico è l'unica strada.

#### 6.4.1 Metodo di Newton–Raphson

Il metodo di **Newton–Raphson** è uno dei più usati per trovare le radici di una funzione derivabile, e quindi per risolvere $\ell^*(\theta)=0$.

**Idea.** Si approssima la funzione $\ell^*$ con il suo sviluppo di Taylor al primo ordine intorno a un punto corrente $\theta^{(k)}$:
$$
\ell^*(\theta)\approx\ell^*(\theta^{(k)})+\ell''(\theta^{(k)})(\theta-\theta^{(k)}).
$$
Imponendo che l'approssimazione sia nulla:
$$
\theta^{(k+1)}=\theta^{(k)}-\frac{\ell^*(\theta^{(k)})}{\ell''(\theta^{(k)})}.
$$
In dimensione $p>1$, la formula diventa
$$
\theta^{(k+1)}=\theta^{(k)}-[\nabla^2\ell(\theta^{(k)})]^{-1}\nabla\ell(\theta^{(k)}).
$$
Sotto condizioni di regolarità, la successione converge al punto stazionario con velocità quadratica.

> **Osservazione 6.11.**  
> In pratica, si itera finché la differenza tra $\theta^{(k+1)}$ e $\theta^{(k)}$ è inferiore a una tolleranza prefissata, oppure finché $\|\ell^*(\theta^{(k)})\|$ è sufficientemente piccolo. La convergenza può dipendere dalla scelta del punto iniziale, soprattutto se la log-verosimiglianza è multimodale
