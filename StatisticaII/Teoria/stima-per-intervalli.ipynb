# Stima per intervalli

---

## 1. Motivazione

### 1.1 Limiti della stima puntuale

Sia $Y \sim p_Y(y;\theta)$ con $\theta \in \Theta$ incognito, e sia $\hat{\theta}_n = g(Y_1,\dots,Y_n)$ uno stimatore di $\theta$. Una **stima puntuale** $\hat{\theta}_{\text{oss}} = g(y_1,\dots,y_n)$ fornisce un singolo valore per $\theta$, ma presenta i seguenti limiti strutturali:

1. **Impossibilità di coincidenza esatta.** Se $\hat{\theta}_n$ è una v.c. assolutamente continua, allora
   $$
   P_\theta(\hat{\theta}_n = \theta) = 0.
   $$
   Se $\hat{\theta}_n$ è discreta, $P_\theta(\hat{\theta}_n = \theta)$ è in generale positivo ma tipicamente piccolo.

2. **Assenza di misura dell'incertezza.** Le proprietà di uno stimatore (non distorsione, efficienza, consistenza) descrivono il comportamento della **procedura** su tutti i campioni possibili, ma non forniscono un'informazione diretta sulla precisione della singola stima osservata.

Da queste considerazioni emerge la necessità di **affiancare alla stima una misura dell'incertezza**, oppure di sostituire la stima puntuale con un **insieme di valori plausibili** per $\theta$.

> **Osservazione 1.2 — Tre livelli di incertezza.**  
> Nella pratica statistica si distinguono tre livelli di quantificazione dell'incertezza:
> 1. **Errore standard**: un singolo numero che misura la dispersione tipica dello stimatore.
> 2. **Intervallo di confidenza**: un intervallo che, con probabilità prestabilita, contiene il vero valore del parametro.
> 3. **Distribuzione campionaria completa**: la legge dello stimatore, che contiene tutta l'informazione sull'incertezza.

## 2. Errore standard

### Definizione 2.1 — Errore standard

Sia $\hat{\theta}_n$ uno stimatore di $\theta$. Si chiama **errore standard** di $\hat{\theta}_n$ la quantità

$$
\operatorname{SE}(\hat{\theta}_n) = \sqrt{\widehat{\mathbb{V}}_\theta(\hat{\theta}_n)},
$$

dove $\widehat{\mathbb{V}}_\theta(\hat{\theta}_n)$ è una stima della varianza dello stimatore $\hat{\theta}_n$.

> **Osservazione 2.2.**  
> Se $\mathbb{V}_\theta(\hat{\theta}_n)$ è nota (perché dipende solo da quantità note), allora $\operatorname{SE}(\hat{\theta}_n) = \sqrt{\mathbb{V}_\theta(\hat{\theta}_n)}$. In caso contrario, si sostituiscono le quantità incognite con i loro stimatori consistenti.

> **Osservazione 2.2.1 — Errore standard vs deviazione standard.**  
> È importante non confondere l'**errore standard** con la **deviazione standard**:
> - la deviazione standard $\sigma$ misura la variabilità delle singole osservazioni $Y_i$;
> - l'errore standard $\operatorname{SE}(\hat{\theta}_n)$ misura la variabilità dello **stimatore**, cioè quanto la stima varia da campione a campione.
>
> La relazione tra le due è $\operatorname{SE}(\overline{Y}) = \sigma/\sqrt{n}$: l'errore standard decresce come $1/\sqrt{n}$, mentre la deviazione standard è una proprietà della popolazione e non dipende da $n$.

### Esempio 2.3 — Errore standard della media campionaria

Sia $Y_1,\dots,Y_n$ un campione i.i.d. con $\mathbb{E}(Y_i) = \mu$ e $\mathbb{V}(Y_i) = \sigma^2$. Lo stimatore $\hat{\mu}_n = \overline{Y}$ ha varianza

$$
\mathbb{V}(\overline{Y}) = \frac{\sigma^2}{n}.
$$

- Se $\sigma^2$ è nota, l'errore standard è
  $$
  \operatorname{SE}(\overline{Y}) = \frac{\sigma}{\sqrt{n}}.
  $$
- Se $\sigma^2$ è ignota, si stima con $S^2 = \frac{1}{n-1}\sum_i (Y_i - \overline{Y})^2$ e
  $$
  \operatorname{SE}(\overline{Y}) = \sqrt{\frac{S^2}{n}}.
  $$

L'errore standard quantifica la **dispersione tipica** della stima attorno al valore medio dello stimatore che, se lo stimatore è corretto, coincide con $\theta$.

### Esempio pratico 2.4 — Errore standard nel controllo di qualità

Riprendendo l'Esempio 1.3: $n = 25$, $s = 0.05$ mm. L'errore standard della media campionaria è

$$
\operatorname{SE}(\overline{Y}) = \sqrt{\frac{S^2}{n}} = \sqrt{\frac{0.05^2}{25}} = \frac{0.05}{5} = 0.01 \text{ mm}.
$$

La stima $\overline{y} = 10.02$ mm ha un'incertezza tipica di $0.01$ mm. In altre parole, la differenza $0.02$ mm rispetto al valore nominale è **circa 2 volte** l'errore standard.

### Esempio 2.5 — Errore standard per una proporzione

Sia $Y \sim \operatorname{Be}(\pi)$ e $\hat{\pi}_n = \overline{Y}$. Allora

$$
\mathbb{V}(\hat{\pi}_n) = \frac{\pi(1-\pi)}{n},
\qquad
\operatorname{SE}(\hat{\pi}_n) = \sqrt{\frac{\hat{\pi}_n(1-\hat{\pi}_n)}{n}}.
$$

**Esempio numerico.** Se in un sondaggio su $n = 1000$ elettori, $600$ si dichiarano favorevoli a una proposta, allora $\hat{\pi}_n = 0.6$ e

$$
\operatorname{SE}(\hat{\pi}_n) = \sqrt{\frac{0.6 \cdot 0.4}{1000}} \approx 0.0155.
$$

L'incertezza tipica sulla stima è di circa $1.5$ punti percentuali.

> **Osservazione 2.6.**  
> L'errore standard è una misura di incertezza **puntuale**: affianca alla stima un singolo numero. Per ottenere una misura più informativa si costruiscono **intervalli casuali** che, con una probabilità prestabilita, contengono il vero valore del parametro.

## 3. Intervallo casuale e intervallo di confidenza

Sia $Y \sim p_Y(y;\theta)$ una variabile aleatoria con funzione di probabilità o densità nota a meno del parametro $\theta \in \Theta$, e sia $\mathbf{Y} = (Y_1,\dots,Y_n)$ un campione casuale semplice da $Y$.

### Definizione 3.1 — Intervallo casuale

Un **intervallo casuale** per $\theta$ è una regione

$$
S(\mathbf{Y}) = S(Y_1,\dots,Y_n) \subseteq \Theta
$$

costruita a partire da una statistica $T_n = g(Y_1,\dots,Y_n)$. Poiché $\mathbf{Y}$ è un vettore aleatorio, $S(\mathbf{Y})$ è un insieme aleatorio.

> **Osservazione 3.1.1.**  
> Nota la distribuzione campionaria di $T_n$, è possibile calcolare la probabilità di $S$, cioè $P_\theta(\theta \in S(\mathbf{Y}))$.

### Esempio motivazionale 3.2 — Media normale con varianza nota

Sia $Y \sim \mathcal{N}(\mu, \sigma^2)$ con $\sigma^2$ nota. Si è interessati a fare inferenza sul parametro $\mu$. Fissato il livello di confidenza $1-\alpha$ (es. $95\%$), sulla base di un campione di osservazioni $Y_1,\dots, Y_n$ si vuole determinare il valore dei due estremi $a$ e $b$ in modo che risulti:

$$
P_r(a < \mu < b) = 1-\alpha.
$$

**Attenzione.** L'espressione $a < \mu < b$ **non è un evento**, perché non contiene nessuna quantità aleatoria: $\mu$ è un parametro fissato (incognito), $a$ e $b$ sono costanti. Non ha quindi senso calcolarne la probabilità.

Per risolvere il problema, l'evento $a < \mu < b$ viene rimpiazzato da un evento contenente una quantità stocastica:

$$
P\left(a_1 < \frac{\overline{Y} - \mu}{\sqrt{\frac{\sigma^2}{n}}} < b_1\right) = 1 - \alpha,
$$

dove $\overline{Y} = \frac{\sum_{i = 1}^n Y_i}{n}$ è la media campionaria.

Poiché $\overline{Y} \sim \mathcal{N}(\mu, \frac{\sigma^2}{n})$, la scrittura precedente equivale a

$$
P(a_1 < Z < b_1) = 1-\alpha,
$$

con $Z \sim \mathcal{N}(0,1)$.

Gli estremi sono ora di immediata determinazione, poiché sono i quantili rispettivamente di ordine $\alpha/2$ e $1-\alpha/2$ di una variabile aleatoria $\mathcal{N}(0,1)$.

> **Osservazione 3.2.1.**  
> Quanto maggiore è $1-\alpha$, tanto maggiore sarà l'ampiezza dell'intervallo.

Sviluppiamo i passaggi:

$$
\begin{aligned}
1-\alpha &= P\left(a_1 < \frac{\overline{Y} - \mu}{\sigma/\sqrt{n}} < b_1\right) \\
&= P\left(z_{\alpha/2} < \frac{\overline{Y} - \mu}{\sigma/\sqrt{n}} < z_{1-\alpha/2}\right) \\
&= P\left(-z_{1-\alpha/2} < \frac{\overline{Y} - \mu}{\sigma/\sqrt{n}} < z_{1-\alpha/2}\right) \\
&= P\left(\overline{Y} - \frac{\sigma}{\sqrt{n}} z_{1-\alpha/2} < \mu < \overline{Y} + \frac{\sigma}{\sqrt{n}} z_{1-\alpha/2}\right),
\end{aligned}
$$

dove si è usata la simmetria della normale standard: $z_{\alpha/2} = -z_{1-\alpha/2}$.

> **Osservazione 3.2.2 — Interpretazione del livello di confidenza.**  
> Il livello di confidenza $1-\alpha$ deve essere così interpretato: se si estraesse un elevato numero di campioni e si costruisse l'intervallo di confidenza con la medesima procedura, $1-\alpha$ sarebbe la proporzione di intervalli che contengono il valore $\mu$.
>
> Il termine "confidenza" è da intendersi come "fiducia": non sappiamo se il valore del parametro appartenga realmente all'intervallo determinato, ma poiché al variare dei campioni, il $100(1-\alpha)\%$ degli intervalli costruiti con la medesima procedura lo contiene, la nostra fiducia è che l'intervallo costruito sia uno di quei $100(1-\alpha)\%$.

### Definizione 3.3 — Intervallo di confidenza

Sia $\alpha \in (0,1)$. Un intervallo casuale $[L(\mathbf{Y}), U(\mathbf{Y})]$ è detto **intervallo di confidenza di livello $1-\alpha$** per $\theta$ (in simboli $\operatorname{IC}_{1-\alpha}(\theta)$) se

$$
P_\theta\bigl(L(\mathbf{Y}) \le \theta \le U(\mathbf{Y})\bigr) = 1 - \alpha \qquad \forall \theta \in \Theta.
$$

Gli estremi $L(\mathbf{Y})$ e $U(\mathbf{Y})$ sono detti **limiti di confidenza** (inferiore e superiore).

Il livello $1-\alpha$ è detto **livello di confidenza** e $\alpha$ **livello di significatività**.

> **Osservazione 3.3.1 — Interpretazione frequentista.**  
> Il livello $1-\alpha$ si interpreta nel senso del **principio del campionamento ripetuto**: se si estraesse un numero molto grande di campioni indipendenti e per ciascuno si costruisse l'intervallo $\operatorname{IC}_{1-\alpha}(\theta)$ con la medesima procedura, la frazione di intervalli che contengono il vero valore $\theta$ sarebbe approssimativamente $1-\alpha$.
>
> Formalmente, per la legge forte dei grandi numeri, se $I_k = \mathbf{1}\{L(\mathbf{Y}^{(k)}) \le \theta \le U(\mathbf{Y}^{(k)})\}$ per $k = 1, \dots, N$ campioni indipendenti, allora
> $$
> \frac{1}{N}\sum_{k=1}^N I_k \xrightarrow{q.c.} 1 - \alpha.
> $$

### Osservazione 3.4 — Interpretazione errata

Non è corretto affermare che "$\theta$ appartiene all'intervallo osservato $[L(\mathbf{y}), U(\mathbf{y})]$ con probabilità $1-\alpha$". Infatti:

- $\theta$ è un parametro **fissato ma incognito**, non una variabile aleatoria;
- l'evento $L(\mathbf{y}) \le \theta \le U(\mathbf{y})$, valutato sul campione osservato, è deterministico: o è vero o è falso.

Le quantità aleatorie sono gli **estremi** $L(\mathbf{Y})$ e $U(\mathbf{Y})$. L'affermazione corretta è:

$$
P_\theta\bigl(L(\mathbf{Y}) \le \theta \le U(\mathbf{Y})\bigr) = 1 - \alpha.
$$

### Definizione 3.5 — Ampiezza dell'intervallo

Si chiama **ampiezza** dell'intervallo di confidenza la quantità

$$
W(\mathbf{Y}) = U(\mathbf{Y}) - L(\mathbf{Y}).
$$

$W(\mathbf{Y})$ è anch'essa una variabile aleatoria. A parità di livello di confidenza, intervalli di ampiezza minore sono **preferibili** perché più informativi.


### Osservazione

In generale un modo ragionevole per costruire un intervallo di confidenza è comprendere il parametro nell'intervallo $[\hat{\theta} - \operatorname{SE}(\hat{\theta}), \hat{\theta} + \operatorname{SE}(\hat{\theta})]$

## 4. Quantità pivotali

### Definizione 4.1 — Quantità pivotale

Sia $\theta \in \Theta$ il parametro di interesse e $T_n = g(Y_1,\dots,Y_n)$ una statistica. Una funzione $Q(T_n, \theta)$ è detta **quantità pivotale** (o **pivot**) se:

1. $Q$ dipende dal campione solo attraverso $T_n$ e dal parametro $\theta$;
2. la distribuzione di probabilità di $Q(T_n,\theta)$ è **completamente nota** e **non dipende da $\theta$**.

> **Osservazione 4.1.** Completamente nota significa che non dipende dal parametro $\theta$ (ignoto)

> **Osservazione 4.2.**  
> Una quantità pivotale **non è una statistica**, perché dipende da $\theta$; tuttavia la sua legge è indipendente da $\theta$, il che permette di calcolare probabilità di eventi che coinvolgono $Q$ senza conoscere $\theta$.

### Esempio 4.3 — Pivot per la media normale con varianza nota

Se $Y \sim \mathcal{N}(\mu, \sigma^2)$ con $\sigma^2$ nota, la funzione

$$
Q(\overline{Y}, \mu) = \frac{\overline{Y} - \mu}{\sigma/\sqrt{n}}
$$

ha distribuzione $\mathcal{N}(0,1)$, che non dipende da $\mu$. Dunque $Q$ è una quantità pivotale per $\mu$.

### Esempio 4.4 — Pivot per la varianza normale con media nota

Se $Y \sim \mathcal{N}(\mu, \sigma^2)$ con $\mu$ nota, la funzione

$$
Q\!\left(\sum_{i=1}^n (Y_i-\mu)^2, \sigma^2\right) = \frac{\sum_{i=1}^n (Y_i-\mu)^2}{\sigma^2}
$$

ha distribuzione $\chi^2_n$, che non dipende da $\sigma^2$. Dunque è una quantità pivotale per $\sigma^2$.

### Esempio 4.5 — Pivot per la media normale con varianza ignota

Se $Y \sim \mathcal{N}(\mu, \sigma^2)$ con $\sigma^2$ ignota, la funzione

$$
Q(\overline{Y}, S^2, \mu) = \frac{\overline{Y} - \mu}{S/\sqrt{n}}
$$

ha distribuzione $t_{n-1}$ (si veda il Teorema 5.3 più sotto), che non dipende da $\mu$ né da $\sigma^2$. Dunque è una quantità pivotale per $\mu$.

### Procedura 4.6 — Costruzione di un IC tramite pivot

Sia $Q(T_n,\theta)$ una quantità pivotale per $\theta$, e supponiamo che $Q$ sia **strettamente monotona** in $\theta$ per ogni valore fissato di $T_n$. Fissato il livello $1-\alpha$:

1. Si determinano i quantili $q_{\alpha/2}$ e $q_{1-\alpha/2}$ di $Q$, cioè
   $$
   P_\theta\bigl(q_{\alpha/2} \le Q(T_n,\theta) \le q_{1-\alpha/2}\bigr) = 1 - \alpha.
   $$
2. Se $Q$ è **crescente** in $\theta$, allora
   $$
   1 - \alpha = P_\theta\bigl(Q^{-1}(q_{\alpha/2}, T_n) \le \theta \le Q^{-1}(q_{1-\alpha/2}, T_n)\bigr),
   $$
   e l'intervallo di confidenza è
   $$
   \operatorname{IC}_{1-\alpha}(\theta) = \bigl[Q^{-1}(q_{\alpha/2}, T_n),\ Q^{-1}(q_{1-\alpha/2}, T_n)\bigr].
   $$
3. Se $Q$ è **decrescente** in $\theta$, i quantili sono invertiti:
   $$
   \operatorname{IC}_{1-\alpha}(\theta) = \bigl[Q^{-1}(q_{1-\alpha/2}, T_n),\ Q^{-1}(q_{\alpha/2}, T_n)\bigr].
   $$

> **Osservazione 4.7.**  
> La monotonia di $Q$ in $\theta$ è essenziale: in assenza di monotonia, l'insieme $\{\theta : q_{\alpha/2} \le Q(T_n,\theta) \le q_{1-\alpha/2}\}$ può non essere un intervallo.

### Osservazione 4.8 — Ruolo dei risultati asintotici

Quando non è possibile determinare esattamente la distribuzione di una quantità pivotale, si sfruttano **approssimazioni asintotiche**. In particolare:

- il **teorema del limite centrale** garantisce che, per $n$ grande, statistiche come $\overline{Y}$ hanno distribuzione approssimativamente normale;
- per gli **stimatori di massima verosimiglianza**, sotto condizioni di regolarità, si ha
  $$
  \sqrt{n}(\hat{\theta}_{\text{MLE}} - \theta) \xrightarrow{D} \mathcal{N}\!\left(0, \frac{1}{I_1(\theta)}\right),
  $$
  e sostituendo $I_1(\theta)$ con una sua stima consistente $I_1(\hat{\theta}_{\text{MLE}})$ si ottiene una quantità **asintoticamente pivotale**.

## 5. Intervalli di confidenza per la media

### 5.1 Caso normale con varianza nota

Sia $Y \sim \mathcal{N}(\mu, \sigma^2)$ con $\sigma^2$ **nota**, e sia $Y_1,\dots,Y_n$ un campione i.i.d. tratto da $Y$.

**Quantità pivotale.** Poiché $\overline{Y} \sim \mathcal{N}(\mu, \sigma^2/n)$, la funzione

$$
Z = \frac{\overline{Y} - \mu}{\sigma/\sqrt{n}} \sim \mathcal{N}(0,1)
$$

è una quantità pivotale per $\mu$, ed è **crescente** in $\mu$ per ogni valore fissato di $\overline{Y}$.

**Costruzione.** Fissato $1-\alpha$, indicando con $z_{1-\alpha/2}$ il quantile di ordine $1-\alpha/2$ della normale standard:

$$
1 - \alpha = P\left(-z_{1-\alpha/2} \le \frac{\overline{Y} - \mu}{\sigma/\sqrt{n}} \le z_{1-\alpha/2}\right).
$$

Isolando $\mu$:

$$
1 - \alpha = P\left(\overline{Y} - z_{1-\alpha/2}\frac{\sigma}{\sqrt{n}} \le \mu \le \overline{Y} + z_{1-\alpha/2}\frac{\sigma}{\sqrt{n}}\right).
$$

**Conclusione.** L'intervallo di confidenza di livello $1-\alpha$ per $\mu$ è

$$
\operatorname{IC}_{1-\alpha}(\mu) = \left[\overline{Y} - z_{1-\alpha/2}\frac{\sigma}{\sqrt{n}},\ \overline{Y} + z_{1-\alpha/2}\frac{\sigma}{\sqrt{n}}\right].
$$

**Ampiezza.** $W = 2 z_{1-\alpha/2} \sigma/\sqrt{n}$, che decresce come $1/\sqrt{n}$ e cresce con $1-\alpha$.

### Esempio pratico 5.2 — Controllo di qualità (continuazione)

Riprendendo l'Esempio 1.3: $n = 25$, $\overline{y} = 10.02$ mm, $\sigma = 0.05$ mm (supponiamo nota da specifiche tecniche). Fissato $1-\alpha = 0.95$, si ha $z_{0.975} = 1.96$. L'intervallo è

$$
\operatorname{IC}_{0.95}(\mu) = \left[10.02 - 1.96 \cdot \frac{0.05}{5},\ 10.02 + 1.96 \cdot \frac{0.05}{5}\right] = [10.02 - 0.0196,\ 10.02 + 0.0196] = [10.0004,\ 10.0396].
$$

**Interpretazione.** L'intervallo **non contiene** il valore nominale $10$ mm, il che suggerisce che il processo potrebbe essere leggermente fuori taratura. La differenza è piccola ma statisticamente significativa al livello $5\%$.

### Esempio pratico 5.3 — Stima dell'altezza media

Uno studio vuole stimare l'altezza media $\mu$ di una popolazione adulta. Si assume $\sigma = 8$ cm (da studi precedenti). Con un campione di $n = 100$ individui, si osserva $\overline{y} = 172$ cm. L'intervallo al $95\%$ è

$$
\operatorname{IC}_{0.95}(\mu) = \left[172 - 1.96 \cdot \frac{8}{10},\ 172 + 1.96 \cdot \frac{8}{10}\right] = [172 - 1.568,\ 172 + 1.568] = [170.43,\ 173.57] \text{ cm}.
$$

### 5.4 Caso normale con varianza ignota

**Setup.** Sia $Y \sim \mathcal{N}(\mu, \sigma^2)$ con $\sigma^2$ **ignota**, e sia $Y_1,\dots,Y_n$ un campione i.i.d.

**Statistica usata.** Si stima $\sigma^2$ con

$$
S^2 = \frac{1}{n-1}\sum_{i=1}^n (Y_i - \overline{Y})^2.
$$

**Quantità pivotale.** Si dimostra che, se $Y \sim \mathcal{N}(\mu, \sigma^2)$, allora $\overline{Y}$ e $S^2$ sono **indipendenti** (Teorema di Fisher) e

$$
\frac{\overline{Y} - \mu}{\sigma/\sqrt{n}} \sim \mathcal{N}(0,1),
\qquad
\frac{(n-1)S^2}{\sigma^2} \sim \chi^2_{n-1}.
$$

Pertanto

$$
T = \frac{\overline{Y} - \mu}{S/\sqrt{n}} = \frac{\dfrac{\overline{Y} - \mu}{\sigma/\sqrt{n}}}{\sqrt{\dfrac{(n-1)S^2}{\sigma^2}\cdot\dfrac{1}{n-1}}} = \frac{Z}{\sqrt{\chi^2_{n-1}/(n-1)}} \sim t_{n-1}.
$$

$T$ è una quantità pivotale per $\mu$, ed è crescente in $\mu$.

**Costruzione.** Fissato $1-\alpha$ e indicando con $t_{n-1,1-\alpha/2}$ il quantile di ordine $1-\alpha/2$ della $t_{n-1}$:

$$
1 - \alpha = P\left(-t_{n-1,1-\alpha/2} \le \frac{\overline{Y} - \mu}{S/\sqrt{n}} \le t_{n-1,1-\alpha/2}\right).
$$

Isolando $\mu$:

$$
\operatorname{IC}_{1-\alpha}(\mu) = \left[\overline{Y} - t_{n-1,1-\alpha/2}\sqrt{\frac{S^2}{n}},\ \overline{Y} + t_{n-1,1-\alpha/2}\sqrt{\frac{S^2}{n}}\right].
$$

> **Osservazione 5.5.**  
> Poiché $t_{n-1,1-\alpha/2} > z_{1-\alpha/2}$ per ogni $n$ finito, l'intervallo con varianza ignota è più ampio di quello con varianza nota. Per $n \to \infty$, $t_{n-1,1-\alpha/2} \to z_{1-\alpha/2}$ e i due intervalli coincidono asintoticamente.

### Esempio pratico 5.6 — Test clinico

Un nuovo farmaco viene testato su $n = 20$ pazienti per ridurre la pressione sanguigna. Si osserva una riduzione media $\overline{y} = 8$ mmHg con deviazione standard campionaria $s = 5$ mmHg. Al livello $95\%$, $t_{19, 0.975} \approx 2.093$. L'intervallo è

$$
\operatorname{IC}_{0.95}(\mu) = \left[8 - 2.093 \cdot \frac{5}{\sqrt{20}},\ 8 + 2.093 \cdot \frac{5}{\sqrt{20}}\right] \approx [8 - 2.34,\ 8 + 2.34] = [5.66,\ 10.34] \text{ mmHg}.
$$

**Interpretazione.** L'intervallo non contiene $0$, quindi la riduzione media è statisticamente significativa al livello $5\%$.

### 5.7 Caso non normale con grandi campioni

Sia $Y$ una v.c. qualsiasi con media $\mu$ e varianza $\sigma^2$ finite, e sia $Y_1,\dots,Y_n$ un campione i.i.d. con $n$ grande.

**Quantità asintoticamente pivotale.** Per il teorema del limite centrale,

$$
\frac{\overline{Y} - \mu}{\sigma/\sqrt{n}} \xrightarrow{D} \mathcal{N}(0,1).
$$

Sostituendo $\sigma^2$ con lo stimatore consistente $S^2$ e applicando il teorema di Slutsky,

$$
\frac{\overline{Y} - \mu}{S/\sqrt{n}} \xrightarrow{D} \mathcal{N}(0,1).
$$

**Intervallo approssimato.** Per $n$ grande,

$$
\operatorname{IC}_{1-\alpha}(\mu) \approx \left[\overline{Y} - z_{1-\alpha/2}\sqrt{\frac{S^2}{n}},\ \overline{Y} + z_{1-\alpha/2}\sqrt{\frac{S^2}{n}}\right],
$$

dove il livello di confidenza è **approssimato** e l'approssimazione migliora al crescere di $n$.

### 5.8 Caso Bernoulli con grandi campioni

Sia $Y \sim \operatorname{Be}(\pi)$ e sia $Y_1,\dots,Y_n$ un campione i.i.d. con $n$ grande. Poiché $\mathbb{E}(Y) = \pi$, lo stimatore è $\hat{\pi}_n = \overline{Y}$, e la varianza è $\mathbb{V}(Y) = \pi(1-\pi)$ ignota.

**Quantità asintoticamente pivotale.** Per il teorema del limite centrale e il teorema di Slutsky,

$$
\frac{\hat{\pi}_n - \pi}{\sqrt{\hat{\pi}_n(1-\hat{\pi}_n)/n}} \xrightarrow{D} \mathcal{N}(0,1).
$$

**Intervallo approssimato.**

$$
\operatorname{IC}_{1-\alpha}(\pi) \approx \left[\hat{\pi}_n - z_{1-\alpha/2}\sqrt{\frac{\hat{\pi}_n(1-\hat{\pi}_n)}{n}},\ \hat{\pi}_n + z_{1-\alpha/2}\sqrt{\frac{\hat{\pi}_n(1-\hat{\pi}_n)}{n}}\right].
$$

### Esempio pratico 5.9 — Sondaggio elettorale

In un sondaggio su $n = 1500$ elettori, $540$ si dichiarano favorevoli a un candidato, quindi $\hat{\pi} = 0.36$. L'intervallo al $95\%$ è

$$
\operatorname{IC}_{0.95}(\pi) \approx \left[0.36 - 1.96\sqrt{\frac{0.36 \cdot 0.64}{1500}},\ 0.36 + 1.96\sqrt{\frac{0.36 \cdot 0.64}{1500}}\right] \approx [0.336,\ 0.384].
$$

**Interpretazione.** Il candidato ha un consenso stimato tra il $33.6\%$ e il $38.4\%$.

> **Osservazione 5.9.1 — Margine di errore.**  
> La quantità $z_{1-\alpha/2}\sqrt{\hat{\pi}(1-\hat{\pi})/n}$ è spesso chiamata **margine di errore**. Per $1-\alpha = 0.95$ e $\hat{\pi} \approx 0.5$, essa vale approssimativamente $1/\sqrt{n}$. Ad esempio, per $n = 1000$, il margine è circa $3.2\%$, in accordo con i valori tipicamente riportati nei sondaggi.

### 5.10 Differenza tra medie di due popolazioni normali con varianze note

Siano $X \sim \mathcal{N}(\mu_X, \sigma_X^2)$ e $Y \sim \mathcal{N}(\mu_Y, \sigma_Y^2)$ con $\sigma_X^2, \sigma_Y^2$ **note**, e siano $X_1,\dots,X_{n_X}$ e $Y_1,\dots,Y_{n_Y}$ due campioni indipendenti.

**Quantità pivotale.** Poiché $\overline{X}$ e $\overline{Y}$ sono indipendenti,

$$
\overline{X} - \overline{Y} \sim \mathcal{N}\!\left(\mu_X - \mu_Y,\ \frac{\sigma_X^2}{n_X} + \frac{\sigma_Y^2}{n_Y}\right).
$$

Pertanto

$$
Z = \frac{(\overline{X} - \overline{Y}) - (\mu_X - \mu_Y)}{\sqrt{\dfrac{\sigma_X^2}{n_X} + \dfrac{\sigma_Y^2}{n_Y}}} \sim \mathcal{N}(0,1).
$$

**Intervallo di confidenza.**

$$
\operatorname{IC}_{1-\alpha}(\mu_X - \mu_Y) = \left[(\overline{X} - \overline{Y}) \pm z_{1-\alpha/2}\sqrt{\frac{\sigma_X^2}{n_X} + \frac{\sigma_Y^2}{n_Y}}\right].
$$

>**Osservazione:** Se l'intervallo di confidenza della differenza tra due medie comprende lo $0 \; (0 \in IC_{\alpha}(\mu))$ è plausibile che le due medie siano uguali

>**Osservazione:** Se $X$ e $Y$ non sono indipendenti nel calcolo della varianza di $\overline{X} - \overline{Y}$ rientra la covarianza
### Esempio pratico 5.11 — Confronto tra due metodi di produzione

Due linee di produzione $A$ e $B$ producono lo stesso componente. Si vuole stimare la differenza tra i diametri medi. Si assumono $\sigma_A = \sigma_B = 0.1$ mm. Si estraggono $n_A = 50$ e $n_B = 60$ componenti, con $\overline{x}_A = 20.05$ mm e $\overline{x}_B = 20.00$ mm. L'intervallo al $95\%$ è

$$
\operatorname{IC}_{0.95}(\mu_A - \mu_B) = \left[0.05 \pm 1.96\sqrt{\frac{0.01}{50} + \frac{0.01}{60}}\right] \approx [0.05 \pm 0.0378] = [0.0122,\ 0.0878].
$$

**Interpretazione.** Poiché l'intervallo **non contiene** $0$, le due linee producono diametri medi significativamente diversi.

### 5.12 Differenza tra medie di due popolazioni normali con varianza comune ignota

Siano $X \sim \mathcal{N}(\mu_X, \sigma^2)$ e $Y \sim \mathcal{N}(\mu_Y, \sigma^2)$ con $\sigma^2$ **ignota e comune**, e siano $X_1,\dots,X_{n_X}$ e $Y_1,\dots,Y_{n_Y}$ due campioni indipendenti.

**Statistica usata.** Si stima $\sigma^2$ con la **varianza pooled**

$$
S_{XY}^2 = \frac{(n_X - 1)S_X^2 + (n_Y - 1)S_Y^2}{n_X + n_Y - 2}.
$$

>**Osservazione:** Pur essendo uguali andiamo a utilizzare entrambe le varianze, facendo una media pesata ma togliondo alla singola variabile il grado di libertà per ricondorci a una t di student 

**Quantità pivotale.**

$$
T = \frac{(\overline{X} - \overline{Y}) - (\mu_X - \mu_Y)}{\sqrt{S_{XY}^2\left(\dfrac{1}{n_X} + \dfrac{1}{n_Y}\right)}} \sim t_{n_X + n_Y - 2}.
$$

**Intervallo di confidenza.**

$$
\operatorname{IC}_{1-\alpha}(\mu_X - \mu_Y) = \left[(\overline{X} - \overline{Y}) \pm t_{n_X+n_Y-2,1-\alpha/2}\sqrt{S_{XY}^2\left(\frac{1}{n_X} + \frac{1}{n_Y}\right)}\right].
$$

### Esempio pratico 5.13 — Confronto tra due gruppi di pazienti

Uno studio confronta l'efficacia di due diete. Gruppo A: $n_A = 15$, $\overline{x}_A = 5.2$ kg persi, $s_A = 1.5$ kg. Gruppo B: $n_B = 15$, $\overline{x}_B = 4.1$ kg persi, $s_B = 1.8$ kg. Varianza pooled:

$$
S_{AB}^2 = \frac{14 \cdot 2.25 + 14 \cdot 3.24}{28} = \frac{31.5 + 45.36}{28} \approx 2.745.
$$

Con $t_{28,0.975} \approx 2.048$:

$$
\operatorname{IC}_{0.95}(\mu_A - \mu_B) \approx \left[1.1 \pm 2.048\sqrt{2.745 \cdot \frac{2}{15}}\right] \approx [1.1 \pm 1.24] = [-0.14,\ 2.34].
$$

**Interpretazione.** L'intervallo **contiene** $0$, quindi non si può concludere che le due diete abbiano efficacia diversa al livello $5\%$.

### 5.14 Differenza tra medie con grandi campioni

Siano $X$ e $Y$ due v.c. con medie $\mu_X,\mu_Y$ e varianze $\sigma_X^2,\sigma_Y^2$ finite, e siano $X_1,\dots,X_{n_X}$ e $Y_1,\dots,Y_{n_Y}$ due campioni indipendenti con $n_X, n_Y$ grandi.

**Quantità asintoticamente pivotale.** Per il teorema del limite centrale e Slutsky,

$$
\frac{(\overline{X} - \overline{Y}) - (\mu_X - \mu_Y)}{\sqrt{\dfrac{S_X^2}{n_X} + \dfrac{S_Y^2}{n_Y}}} \xrightarrow{D} \mathcal{N}(0,1).
$$

**Intervallo approssimato.**

$$
\operatorname{IC}_{1-\alpha}(\mu_X - \mu_Y) \approx \left[(\overline{X} - \overline{Y}) \pm z_{1-\alpha/2}\sqrt{\frac{S_X^2}{n_X} + \frac{S_Y^2}{n_Y}}\right].
$$

>**Osservazione:** L'intuito porta a pensare di poter utilizzare una t di student tuttavia tanto più sono differenti le varianze delle due popolazioni tanto più l'approssimazione risulta instabile, è quindi necessario utilizzare il teorema del limite centrale
> In generale l'approssimazione con una $t$ di student non vale se $\sigma^2_X \neq \sigma^2_Y$, vale se le posso assumere uguali

>**Osservazione:** Problema di Behremo-Fisher 

>**Osservazione:** Posso tuttavia correggere l'approssimazione con una $t$ di student diminuendo i gradi di libertà, con la correzzione di Welch che pone riduce il grado di libertà $n_X+n_Y - 2$

### 5.15 Differenza tra proporzioni con grandi campioni

Siano $X \sim \operatorname{Be}(\pi_X)$ e $Y \sim \operatorname{Be}(\pi_Y)$, e siano $X_1,\dots,X_{n_X}$ e $Y_1,\dots,Y_{n_Y}$ campioni indipendenti con $n_X, n_Y$ grandi. Gli stimatori sono $\hat{\pi}_X, \hat{\pi}_Y$.

**Intervallo approssimato.**

$$
\operatorname{IC}_{1-\alpha}(\pi_X - \pi_Y) \approx \left[(\hat{\pi}_X - \hat{\pi}_Y) \pm z_{1-\alpha/2}\sqrt{\frac{\hat{\pi}_X(1-\hat{\pi}_X)}{n_X} + \frac{\hat{\pi}_Y(1-\hat{\pi}_Y)}{n_Y}}\right].
$$

### Esempio pratico 5.16 — Confronto tra due trattamenti

Due trattamenti per una malattia vengono testati su due gruppi. Trattamento A: $n_A = 500$, $250$ guariti ($\hat{\pi}_A = 0.5$). Trattamento B: $n_B = 600$, $360$ guariti ($\hat{\pi}_B = 0.6$). L'intervallo al $95\%$ sulla differenza è

$$
\operatorname{IC}_{0.95}(\pi_B - \pi_A) \approx \left[0.1 \pm 1.96\sqrt{\frac{0.5 \cdot 0.5}{500} + \frac{0.6 \cdot 0.4}{600}}\right] \approx [0.1 \pm 0.061] = [0.039,\ 0.161].
$$

**Interpretazione.** L'intervallo non contiene $0$, quindi il trattamento B è significativamente più efficace di A.

## 6. Intervalli di confidenza per la varianza

### 6.1 Caso normale

Sia $Y \sim \mathcal{N}(\mu, \sigma^2)$ e sia $Y_1,\dots,Y_n$ un campione i.i.d.

**Quantità pivotale.** Per il teorema di Cochran,

$$
Q = \frac{(n-1)S^2}{\sigma^2} \sim \chi^2_{n-1}.
$$

$Q$ è **decrescente** in $\sigma^2$ (poiché $\sigma^2$ appare al denominatore) e la sua distribuzione non dipende da $\sigma^2$.

**Costruzione.** Fissato $1-\alpha$, indicando con $\chi^2_{n-1,\delta}$ il quantile di ordine $\delta$ di $\chi^2_{n-1}$:

$$
1 - \alpha = P\left(\chi^2_{n-1,\alpha/2} \le \frac{(n-1)S^2}{\sigma^2} \le \chi^2_{n-1,1-\alpha/2}\right).
$$

Isolando $\sigma^2$ (ricordando che $Q$ è decrescente in $\sigma^2$):

$$
1 - \alpha = P\left(\frac{(n-1)S^2}{\chi^2_{n-1,1-\alpha/2}} \le \sigma^2 \le \frac{(n-1)S^2}{\chi^2_{n-1,\alpha/2}}\right).
$$

**Conclusione.**

$$
\operatorname{IC}_{1-\alpha}(\sigma^2) = \left[\frac{(n-1)S^2}{\chi^2_{n-1,1-\alpha/2}},\ \frac{(n-1)S^2}{\chi^2_{n-1,\alpha/2}}\right].
$$

> **Osservazione 6.2.**  
> A differenza dell'IC per la media, l'IC per la varianza **non è simmetrico** attorno a $S^2$, poiché la distribuzione $\chi^2$ è asimmetrica. Inoltre, l'ampiezza dipende dai quantili $\chi^2_{n-1,\alpha/2}$ e $\chi^2_{n-1,1-\alpha/2}$, che si allontanano da $n-1$ in modo asimmetrico.

> **Osservazione 6.3 — IC per la deviazione standard.**  
> Prendendo la radice quadrata degli estremi si ottiene un IC per la deviazione standard $\sigma$:
> $$
> \operatorname{IC}_{1-\alpha}(\sigma) = \left[\sqrt{\frac{(n-1)S^2}{\chi^2_{n-1,1-\alpha/2}}},\ \sqrt{\frac{(n-1)S^2}{\chi^2_{n-1,\alpha/2}}}\right].
> $$

### Esempio pratico 6.4 — Controllo di qualità (continuazione)

Riprendendo l'Esempio 1.3: $n = 25$, $s = 0.05$ mm. Al livello $95\%$, $\chi^2_{24,0.025} \approx 12.40$ e $\chi^2_{24,0.975} \approx 39.36$. L'IC per $\sigma^2$ è

$$
\operatorname{IC}_{0.95}(\sigma^2) = \left[\frac{24 \cdot 0.0025}{39.36},\ \frac{24 \cdot 0.0025}{12.40}\right] \approx [0.00152,\ 0.00484].
$$

Prendendo la radice quadrata, l'IC per $\sigma$ è

$$
\operatorname{IC}_{0.95}(\sigma) \approx [0.0390,\ 0.0696] \text{ mm}.
$$

**Interpretazione.** La deviazione standard è compresa tra $0.039$ e $0.070$ mm con confidenza del $95\%$.

## 7. Determinazione della numerosità campionaria

### 7.1 Caso della media con varianza nota

Un problema pratico frequente è: **quanti campioni servono per ottenere un IC di ampiezza desiderata?** Fissiamo l'ampiezza massima $W_{\max} = 2d$ (dove $d$ è il semiampiezza o **margine di errore**). Dalla formula dell'IC per la media:

$$
d = z_{1-\alpha/2}\frac{\sigma}{\sqrt{n}}.
$$

Risolvendo rispetto a $n$:

$$
n = \left(\frac{z_{1-\alpha/2} \sigma}{d}\right)^2.
$$

> **Osservazione 7.1.**  
> Se $\sigma$ è ignota, si usa una stima preliminare $\hat{\sigma}$ (da uno studio pilota o da dati storici). Se anche questa non è disponibile, si procede in modo iterativo.

### Esempio pratico 7.2

Nel controllo qualità (Esempio 1.3), supponiamo di volere un margine di errore $d = 0.01$ mm con $1-\alpha = 0.95$ e $\sigma = 0.05$ mm. Allora

$$
n = \left(\frac{1.96 \cdot 0.05}{0.01}\right)^2 = (9.8)^2 \approx 96.
$$

Occorrono almeno $96$ misurazioni per ottenere il margine desiderato.

### 7.3 Caso della proporzione

Per una proporzione, il margine di errore è

$$
d = z_{1-\alpha/2}\sqrt{\frac{\pi(1-\pi)}{n}}.
$$

Risolvendo rispetto a $n$:

$$
n = \left(\frac{z_{1-\alpha/2}}{d}\right)^2 \pi(1-\pi).
$$

Poiché $\pi$ è incognita e $\pi(1-\pi) \leq 1/4$, si usa la stima più conservativa $n = (z_{1-\alpha/2}/(2d))^2$.

### Esempio pratico 7.4

Un sondaggista vuole un margine di errore $d = 0.03$ (3 punti percentuali) al $95\%$. Allora

$$
n = \left(\frac{1.96}{2 \cdot 0.03}\right)^2 = (32.67)^2 \approx 1067.
$$

Occorrono circa $1067$ intervistati. Questo spiega perché i sondaggi nazionali usano tipicamente campioni di circa $1000$ persone.

## 8. Proprietà generali degli intervalli di confidenza

### Proposizione 8.1 — Relazione tra livello e ampiezza

A parità di numerosità campionaria $n$, di statistica $T_n$ e di varianza dello stimatore, l'ampiezza $W$ dell'intervallo di confidenza è **crescente** in $1-\alpha$.

**Dimostrazione.**  
Nella costruzione tramite quantità pivotale $Q$ con quantili $q_{\alpha/2}$ e $q_{1-\alpha/2}$, l'ampiezza è $W = Q^{-1}(q_{1-\alpha/2},T_n) - Q^{-1}(q_{\alpha/2},T_n)$. Poiché al diminuire di $\alpha$ i quantili $q_{\alpha/2}$ e $q_{1-\alpha/2}$ si allontanano (nel senso che $q_{\alpha/2}$ decresce e $q_{1-\alpha/2}$ cresce), la differenza $q_{1-\alpha/2} - q_{\alpha/2}$ cresce. Se $Q^{-1}$ è monotona, anche $W$ cresce. $\square$

### Proposizione 8.2 — Relazione tra ampiezza e numerosità campionaria

A parità di livello $1-\alpha$, l'ampiezza di un intervallo di confidenza basato su uno stimatore consistente con varianza dell'ordine $O(1/n)$ decresce come $O(1/\sqrt{n})$.

**Dimostrazione.**  
Nei casi considerati, l'ampiezza è della forma $W = 2c_{1-\alpha/2}\sqrt{\sigma^2/n}$, dove $c_{1-\alpha/2}$ è un quantile noto ($z_{1-\alpha/2}$ o $t_{n-1,1-\alpha/2}$). Allora

$$
W = O\left(\frac{1}{\sqrt{n}}\right).
$$

$\square$

> **Osservazione 8.3 — Trade-off precisione/confidenza.**  
> Esiste un **trade-off** tra precisione e confidenza: aumentando il livello $1-\alpha$ si ottiene una maggiore fiducia di copertura, ma a scapito di un intervallo più ampio (meno informativo). Viceversa, per ottenere un intervallo più preciso, occorre aumentare la numerosità campionaria o ridurre il livello di confidenza.

### Proposizione 8.4 — Consistenza degli intervalli di confidenza

Sotto le ipotesi standard, l'ampiezza dell'intervallo di confidenza tende a $0$ per $n \to \infty$:

$$
W(\mathbf{Y}) \xrightarrow{P_\theta} 0.
$$

**Dimostrazione.**  
Nei casi considerati, $W = 2c_{1-\alpha/2}\sqrt{\hat{\sigma}^2/n}$. Poiché $\hat{\sigma}^2$ è uno stimatore consistente di $\sigma^2$ (limitato in probabilità) e $1/\sqrt{n} \to 0$, il prodotto tende a $0$ in probabilità. $\square$

### Osservazione 8.5 — Intervalli unilaterali

In alcune applicazioni si è interessati a **limiti unilaterali**: un intervallo di confidenza di livello $1-\alpha$ del tipo $[L(\mathbf{Y}), +\infty)$ o $(-\infty, U(\mathbf{Y})]$. In tal caso, si sostituiscono i quantili $z_{\alpha/2}$ e $z_{1-\alpha/2}$ con i quantili $z_\alpha$ e $z_{1-\alpha}$ rispettivamente, ottenendo intervalli più stretti sul lato di interesse.

**Esempio.** Per la media normale con varianza nota, un intervallo unilaterale destro di livello $1-\alpha$ è

$$
\operatorname{IC}_{1-\alpha}^{+}(\mu) = \left[\overline{Y} - z_{1-\alpha}\frac{\sigma}{\sqrt{n}},\ +\infty\right).
$$

### Osservazione 8.6 — Relazione tra IC e test di ipotesi

Esiste una **dualità** tra intervalli di confidenza e test di ipotesi: un valore $\theta_0$ appartiene all'$\operatorname{IC}_{1-\alpha}(\theta)$ se e solo se il test di ipotesi $H_0: \theta = \theta_0$ contro $H_1: \theta \neq \theta_0$ non rifiuta $H_0$ al livello $\alpha$. Questa dualità permette di costruire test a partire da IC e viceversa.
