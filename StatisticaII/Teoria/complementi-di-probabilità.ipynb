## 0. Esperimenti aleatori e probabilità

### Definizione 0.1 — Esperimento aleatorio e spazio campionario

Un **esperimento aleatorio** è un esperimento il cui esito non è noto a priori; l'esecuzione in condizioni identiche può produrre esiti diversi. L'insieme $\Omega$ di tutti gli esiti possibili si dice **spazio campionario**. Un **evento** è un sottoinsieme $A\subseteq\Omega$.

### Definizione 0.2 — Assiomi di Kolmogorov

Una probabilità è una funzione $P:\mathcal{F}\to[0,1]$ definita su una $\sigma$-algebra $\mathcal{F}$ di sottoinsiemi di $\Omega$, tale che:

1. $P(\Omega)=1$;
2. $P(A)\ge0$ per ogni $A\in\mathcal{F}$;
3. **(additività numerabile)** se $\{A_n\}$ è una successione di eventi a due a due disgiunti, allora
$$
P\left(\bigcup_{n=1}^\infty A_n\right)=\sum_{n=1}^\infty P(A_n).
$$

Dagli assiomi seguono:

- $P(\varnothing)=0$;
- $P(A^c)=1-P(A)$;
- **(additività finita)** $P(A\cup B)=P(A)+P(B)$ se $A\cap B=\varnothing$;
- **(inclusione-esclusione)** $P(A\cup B)=P(A)+P(B)-P(A\cap B)$;
- **(monotonia)** se $A\subseteq B$, allora $P(A)\le P(B)$;
- **(continuità dal basso e dall'alto)** se $A_n\uparrow A$, allora $P(A_n)\to P(A)$; se $A_n\downarrow A$, allora $P(A_n)\to P(A)$.

### Definizione 0.3 — Probabilità condizionata e indipendenza

Se $P(B)>0$, la **probabilità condizionata** di $A$ dato $B$ è
$$
P(A\mid B)=\frac{P(A\cap B)}{P(B)}.
$$
Due eventi $A,B$ sono **indipendenti** se $P(A\cap B)=P(A)P(B)$; equivalentemente, se $P(A\mid B)=P(A)$ quando $P(B)>0$.

**Proposizione 0.4 (formula della probabilità totale e di Bayes).**  
Sia $\{B_n\}$ una partizione numerabile di $\Omega$ con $P(B_n)>0$. Allora per ogni evento $A$,
$$
P(A)=\sum_n P(A\mid B_n)P(B_n),
\qquad
P(B_k\mid A)=\frac{P(A\mid B_k)P(B_k)}{\sum_n P(A\mid B_n)P(B_n)}.
$$

**Dimostrazione.**  
La prima formula segue dall'additività applicata alla partizione $A=\bigcup_n(A\cap B_n)$, con $P(A\cap B_n)=P(A\mid B_n)P(B_n)$. La seconda è la definizione di probabilità condizionata applicata al numeratore, con il denominatore riscritto tramite la formula della probabilità totale. $\square$

## 1. Variabili casuali unidimensionali

### Definizione 1.1 — Variabile casuale

Dato $(\Omega,\mathcal{F},P)$, una **variabile casuale** (v.c.) è una funzione misurabile $X:\Omega\to\mathbb{R}$. La **legge** di $X$ è la probabilità immagine $P_X$ su $\mathbb{R}$, definita da
$$
P_X(B)=P(X^{-1}(B)),\qquad B\in\mathcal{B}(\mathbb{R}).
$$

### Definizione 1.2 — Supporto e classificazione

Il **supporto** di $X$ è l'insieme dei valori che $X$ assume con probabilità positiva (caso discreto) o in ogni intorno dei quali la densità è positiva (caso continuo). Si distingue:

- $X$ **discreta** se il supporto è finito o numerabile;
- $X$ **continua** se esiste una densità $f_X\ge0$ con $\int_{\mathbb{R}}f_X=1$ e
$$
P_X(B)=\int_B f_X(x)\,dx.
$$

### Definizione 1.3 — Funzione di ripartizione

La **funzione di ripartizione** di $X$ è
$$
F_X(x)=P(X\le x).
$$
Valgono:

1. $F_X$ è non decrescente;
2. $\lim_{x\to-\infty}F_X(x)=0$, $\lim_{x\to+\infty}F_X(x)=1$;
3. $F_X$ è continua a destra;
4. $P(a<X\le b)=F_X(b)-F_X(a)$.

Se $X$ è continua, $F_X$ è assolutamente continua e $F_X'=f_X$ q.o. Se $X$ è discreta, $F_X$ è costante a tratti, con salti nei punti del supporto.

> **Osservazione 1.4.**  
> Nel caso continuo, $P(X=x)=0$ per ogni $x$: la probabilità è concentrata su intervalli, non su punti. La densità non è una probabilità: solo il suo integrale lo è. Di conseguenza $F_X$ è continua ma non necessariamente derivabile ovunque.

## 2. Momenti e disuguaglianze

### Definizione 2.1 — Valore atteso

Se $X$ è discreta con supporto $S$ e $\sum_{x\in S}|x|P(X=x)<\infty$, si definisce
$$
E[X]=\sum_{x\in S}xP(X=x).
$$
Se $X$ è continua con densità $f_X$ e $\int_{\mathbb{R}}|x|f_X(x)\,dx<\infty$,
$$
E[X]=\int_{\mathbb{R}}x f_X(x)\,dx.
$$

Per una trasformazione $g$, se $E[|g(X)|]<\infty$,
$$
E[g(X)]=\begin{cases}\sum_x g(x)P(X=x) & \text{discreta}\\[2pt]\int_{\mathbb{R}}g(x)f_X(x)\,dx & \text{continua}\end{cases}
$$

**Proposizione 2.2 (linearità).**  
Per $a,b\in\mathbb{R}$, $E[aX+b]=aE[X]+b$, purché i momenti esistano.

### Definizione 2.3 — Varianza

Se $E[X^2]<\infty$, si definisce
$$
\operatorname{Var}(X)=E[(X-E[X])^2]=E[X^2]-E[X]^2.
$$
Lo **scarto quadratico medio** è $\sigma_X=\sqrt{\operatorname{Var}(X)}$.

Per $a,b\in\mathbb{R}$,
$$
\operatorname{Var}(aX+b)=a^2\operatorname{Var}(X).
$$

### Proposizione 2.4 — Disuguaglianza di Markov

Se $X\ge0$ e $E[X]<\infty$, allora per ogni $a>0$,
$$
P(X\ge a)\le\frac{E[X]}{a}.
$$

**Dimostrazione.**  
Sia $X\ge0$. Allora
$$
E[X]=\int_0^\infty x\,dF_X(x)\ge\int_a^\infty x\,dF_X(x)\ge a\int_a^\infty dF_X(x)=aP(X\ge a).
$$
Da cui la tesi. $\square$

### Proposizione 2.5 — Disuguaglianza di Chebyshev

Se $\operatorname{Var}(X)<\infty$ e $\varepsilon>0$,
$$
P(|X-E[X]|\ge\varepsilon)\le\frac{\operatorname{Var}(X)}{\varepsilon^2}.
$$

**Dimostrazione.**  
Applicare Markov a $Y=(X-E[X])^2$, che è non negativa e ha media $\operatorname{Var}(X)$. $\square$

**Corollario 2.6.**  
Per ogni $k>0$,
$$
P(|X-E[X]|\ge k\sigma_X)\le\frac{1}{k^2}.
$$
In particolare, $P(|X-E[X]|<3\sigma_X)\ge 8/9$.

> **Osservazione 2.7.**  
> Chebyshev è una disuguaglianza **distribuzionale**: vale per ogni legge con varianza finita e non richiede ipotesi sulla forma della distribuzione. Per questo è alla base di risultati di convergenza (legge debole dei grandi numeri) e di controllo della coda, ma è in genere **non ottimale**: per la normale, $P(|X-\mu|<3\sigma)\approx0.9973$, molto più di $8/9$.

### Proposizione 2.8 — Disuguaglianza di Jensen

Se $g:\mathbb{R}\to\mathbb{R}$ è convessa e $E[|X|]<\infty$, $E[|g(X)|]<\infty$, allora
$$
g(E[X])\le E[g(X)].
$$
Se $g$ è concava, la disuguaglianza si inverte.

**Dimostrazione.**  
Sia $g$ convessa. Per ogni $x_0$ esiste una sottogradiente $m(x_0)$ tale che
$$
g(x)\ge g(x_0)+m(x_0)(x-x_0)\qquad\forall x.
$$
Scegliendo $x_0=E[X]$ e prendendo il valore atteso,
$$
E[g(X)]\ge g(E[X])+m(E[X])(E[X]-E[X])=g(E[X]).
$$
$\square$

**Corollario 2.9.**  
Se $X>0$ q.c., $E[\log X]\le\log E[X]$.

## 3. Standardizzazione e trasformazioni

### Definizione 3.1 — Standardizzazione

Data $X$ con media $\mu$ e varianza $\sigma^2>0$, si definisce
$$
Z=\frac{X-\mu}{\sigma}.
$$
Allora $E[Z]=0$, $\operatorname{Var}(Z)=1$.

### Teorema 3.2 — Densità di una trasformazione monotona

Sia $X$ continua con densità $f_X$ su supporto $S$, e sia $g:S\to S^*$ strettamente monotona e differenziabile con $g'\neq0$. Allora $Y=g(X)$ ha densità
$$
f_Y(y)=f_X(g^{-1}(y))\left|\frac{d}{dy}g^{-1}(y)\right|,\qquad y\in S^*.
$$

**Dimostrazione.**  
Supponiamo $g$ crescente. Per $y\in S^*$,
$$
F_Y(y)=P(Y\le y)=P(g(X)\le y)=P(X\le g^{-1}(y))=F_X(g^{-1}(y)).
$$
Derivando rispetto a $y$ e applicando la regola della catena,
$$
f_Y(y)=f_X(g^{-1}(y))\frac{d}{dy}g^{-1}(y).
$$
Il caso decrescente segue con un argomento simmetrico, che introduce il valore assoluto. $\square$

> **Osservazione 3.3.**  
> La formula è il cambio di variabile per densità. In dimensione $p>1$ diventa
> $$
> f_Y(y)=f_X(g^{-1}(y))\left|\det\frac{\partial g^{-1}}{\partial y}(y)\right|,
> $$
> dove $\left|\det(\cdot)\right|$ è il valore assoluto del determinante dello Jacobiano. È esattamente la formula che useremo per la normale multivariata.

## 4. Variabili casuali doppie

### Definizione 4.1 — Variabile casuale doppia

Una **v.c. doppia** $(X,Y)$ è una funzione misurabile da $\Omega$ a $\mathbb{R}^2$. La sua legge è determinata da:

- nel caso discreto, dalle probabilità congiunte $p_{ij}=P(X=x_i,Y=y_j)$;
- nel caso continuo, dalla densità congiunta $f_{XY}$ con
$$
P((X,Y)\in B)=\int_B f_{XY}(x,y)\,dx\,dy.
$$

### Definizione 4.2 — Marginali e condizionate

Le **marginali** si ottengono integrando (o sommando) l'altra variabile:
$$
f_X(x)=\int_{\mathbb{R}}f_{XY}(x,y)\,dy,\qquad f_Y(y)=\int_{\mathbb{R}}f_{XY}(x,y)\,dx.
$$

Le **condizionate** sono
$$
f_{Y\mid X}(y\mid x)=\frac{f_{XY}(x,y)}{f_X(x)}\quad (f_X(x)>0),
\qquad
f_{X\mid Y}(x\mid y)=\frac{f_{XY}(x,y)}{f_Y(y)}\quad (f_Y(y)>0).
$$

### Definizione 4.3 — Momenti doppi

Si definiscono
$$
\mu_X=E[X],\quad\mu_Y=E[Y],\quad\sigma_X^2=\operatorname{Var}(X),\quad\sigma_Y^2=\operatorname{Var}(Y),
$$
e
$$
\sigma_{XY}=\operatorname{Cov}(X,Y)=E[(X-\mu_X)(Y-\mu_Y)]=E[XY]-\mu_X\mu_Y.
$$
Il **coefficiente di correlazione** è
$$
\rho_{XY}=\frac{\sigma_{XY}}{\sigma_X\sigma_Y}\in[-1,1].
$$

**Proposizione 4.4.**  
$|\rho_{XY}|\le1$, con uguaglianza $|\rho_{XY}|=1$ se e solo se esistono $a,b\in\mathbb{R}$ con $a\neq0$ tali che $Y=aX+b$ q.c.

**Dimostrazione.**  
Segue dalla disuguaglianza di Cauchy–Schwarz applicata a $X-\mu_X$ e $Y-\mu_Y$. Il caso di uguaglianza corrisponde alla proporzionalità lineare q.c. $\square$

### Teorema 4.5 — Indipendenza

$X$ e $Y$ sono **indipendenti** se e solo se $f_{XY}(x,y)=f_X(x)f_Y(y)$ per (q.o.) $(x,y)$. Conseguenze:

1. $E[g(X)h(Y)]=E[g(X)]E[h(Y)]$ per ogni $g,h$ per cui i momenti esistono;
2. $\operatorname{Cov}(X,Y)=0$;
3. le condizionate coincidono con le marginali.

> **Osservazione 4.6.**  
> L'indipendenza implica l'incorrelazione, ma **non** viceversa. Esistono variabili con $\operatorname{Cov}(X,Y)=0$ che non sono indipendenti. Un esempio classico: $X\sim\mathcal{N}(0,1)$ e $Y=X^2$. Allora $\operatorname{Cov}(X,Y)=E[X^3]-E[X]E[X^2]=0$, ma $Y$ è funzione deterministica di $X$. Per la **normale bivariata**, invece, incorrelazione e indipendenza coincidono (come vedremo).

### Proposizione 4.7 — Combinazioni lineari

Per $X,Y$ con momenti finiti e $a,b,c\in\mathbb{R}$,
$$
E[aX+bY+c]=aE[X]+bE[Y]+c,
$$
$$
\operatorname{Var}(aX+bY+c)=a^2\operatorname{Var}(X)+b^2\operatorname{Var}(Y)+2ab\operatorname{Cov}(X,Y).
$$
Più in generale, per $T=\sum_i\beta_iY_i$,
$$
\operatorname{Var}(T)=\sum_i\beta_i^2\operatorname{Var}(Y_i)+\sum_{i\neq j}\beta_i\beta_j\operatorname{Cov}(Y_i,Y_j).
$$
Se le $Y_i$ sono incorrelate, il secondo termine svanisce.

## 5. Variabili casuali multiple

### Definizione 5.1 — Vettore casuale

Un **vettore casuale** $\mathbf{Y}=(Y_1,\dots,Y_p)^T$ è una funzione misurabile da $\Omega$ a $\mathbb{R}^p$. Il **vettore delle medie** è
$$
\boldsymbol\mu=E[\mathbf{Y}]=(E[Y_1],\dots,E[Y_p])^T,
$$
e la **matrice di varianza-covarianza** è
$$
\boldsymbol\Sigma=\operatorname{Var}(\mathbf{Y})=E[(\mathbf{Y}-\boldsymbol\mu)(\mathbf{Y}-\boldsymbol\mu)^T]\in\mathbb{R}^{p\times p}.
$$

**Proposizione 5.2.**  
$\boldsymbol\Sigma$ è simmetrica e semidefinita positiva. È definita positiva se e solo se le componenti di $\mathbf{Y}$ non sono q.c. legate da una relazione lineare non banale.

**Dimostrazione.**  
Simmetria: $\Sigma_{ij}=\operatorname{Cov}(Y_i,Y_j)=\operatorname{Cov}(Y_j,Y_i)=\Sigma_{ji}$. Semi-definita positività: per ogni $a\in\mathbb{R}^p$,
$$
a^T\boldsymbol\Sigma a=\operatorname{Var}(a^T\mathbf{Y})\ge0.
$$
Il caso di uguaglianza corrisponde a $\operatorname{Var}(a^T\mathbf{Y})=0$, cioè $a^T\mathbf{Y}$ costante q.c., cioè una relazione lineare q.c. tra le componenti. $\square$

### Proposizione 5.3 — Trasformazioni lineari

Sia $\mathbf{T}=A\mathbf{Y}+\mathbf{b}$ con $A\in\mathbb{R}^{q\times p}$, $\mathbf{b}\in\mathbb{R}^q$. Allora
$$
E[\mathbf{T}]=A\boldsymbol\mu+\mathbf{b},
\qquad
\operatorname{Var}(\mathbf{T})=A\boldsymbol\Sigma A^T.
$$

**Dimostrazione.**  
Per la linearità del valore atteso, $E[A\mathbf{Y}+\mathbf{b}]=AE[\mathbf{Y}]+\mathbf{b}$. Per la varianza,
$$
\operatorname{Var}(\mathbf{T})=E[(A\mathbf{Y}+\mathbf{b}-A\boldsymbol\mu-\mathbf{b})(A\mathbf{Y}+\mathbf{b}-A\boldsymbol\mu-\mathbf{b})^T]=E[A(\mathbf{Y}-\boldsymbol\mu)(\mathbf{Y}-\boldsymbol\mu)^TA^T]=A\boldsymbol\Sigma A^T.
$$
$\square$

## 6. Convergenza di successioni di variabili casuali

### Definizione 6.1 — Modi di convergenza

Sia $\{X_n\}$ una successione di v.c. e $X$ una v.c.:

1. **Convergenza in media quadratica** ($X_n\xrightarrow{L^2}X$):
$$
\lim_{n\to\infty}E[(X_n-X)^2]=0.
$$
2. **Convergenza quasi certa** ($X_n\xrightarrow{q.c.}X$):
$$
P\left(\lim_{n\to\infty}X_n=X\right)=1.
$$
3. **Convergenza in probabilità** ($X_n\xrightarrow{p}X$):
$$
\forall\varepsilon>0,\quad\lim_{n\to\infty}P(|X_n-X|>\varepsilon)=0.
$$
4. **Convergenza in distribuzione** ($X_n\xrightarrow{D}X$):
$$
\lim_{n\to\infty}F_{X_n}(t)=F_X(t)\quad\text{per ogni punto }t\text{ di continuità di }F_X.
$$

### Proposizione 6.2 — Gerarchia delle convergenze

Valgono le implicazioni
$$
L^2\implies p\implies D,
\qquad
q.c.\implies p\implies D.
$$
In generale, $p$ non implica $q.c.$ e $D$ non implica $p$.

**Dimostrazione (idea).**  
$L^2\Rightarrow p$ segue da Markov applicato a $(X_n-X)^2$: $P(|X_n-X|>\varepsilon)\le E[(X_n-X)^2]/\varepsilon^2$.  
$q.c.\Rightarrow p$ segue dal fatto che, se $X_n\to X$ q.c., allora per ogni $\varepsilon>0$ la successione degli eventi $\{|X_n-X|>\varepsilon\}$ si verifica solo un numero finito di volte q.c.  
$p\Rightarrow D$ segue dal fatto che la convergenza in probabilità implica quella delle funzioni di ripartizione nei punti di continuità.  
Le implicazioni inverse non valgono in generale; controesempi classici mostrano che $p$ può valere senza $q.c.$ (successioni "a blocchi") e $D$ può valere senza $p$ (se $X_n$ e $X$ hanno leggi uguali ma non sono vicine come variabili). $\square$

> **Osservazione 6.3.**  
> La convergenza in distribuzione è la più debole e la più usata: è quella del teorema del limite centrale. La convergenza in probabilità è quella della legge debole dei grandi numeri. La convergenza quasi certa è quella della legge forte.

## 7. Legge dei grandi numeri e teorema del limite centrale

### Teorema 7.1 — Legge debole dei grandi numeri

Sia $\{X_n\}$ una successione di v.c. i.i.d. con $E[X_1]=\mu$ e $\operatorname{Var}(X_1)=\sigma^2<\infty$. Allora
$$
\bar{X}_n=\frac{1}{n}\sum_{i=1}^nX_i\xrightarrow{p}\mu.
$$

**Dimostrazione.**  
$\operatorname{Var}(\bar X_n)=\sigma^2/n$. Per Chebyshev, per ogni $\varepsilon>0$,
$$
P(|\bar X_n-\mu|\ge\varepsilon)\le\frac{\sigma^2}{n\varepsilon^2}\to0.
$$
$\square$

### Teorema 7.2 — Legge forte dei grandi numeri

Nelle stesse ipotesi (in realtà basta $E[|X_1|]<\infty$),
$$
\bar X_n\xrightarrow{q.c.}\mu.
$$

**Idea della dimostrazione.**  
Si dimostra prima per variabili limitate, usando il lemma di Borel–Cantelli e la disuguaglianza di Chebyshev su una sottosuccessione; poi si estende per troncamento. $\square$

### Teorema 7.3 — Teorema del limite centrale (Lindeberg–Lévy)

Sia $\{X_n\}$ una successione di v.c. i.i.d. con $E[X_1]=\mu$ e $\operatorname{Var}(X_1)=\sigma^2\in(0,\infty)$. Allora, posto $S_n=\sum_{i=1}^nX_i$ e
$$
Z_n=\frac{S_n-n\mu}{\sigma\sqrt n},
$$
si ha
$$
Z_n\xrightarrow{D}\mathcal{N}(0,1).
$$

**Idea della dimostrazione.**  
Si usa la funzione caratteristica $\varphi_{Z_n}(t)=E[e^{itZ_n}]$, che per l'indipendenza è
$$
\varphi_{Z_n}(t)=\left[\varphi_{(X-\mu)/\sigma}\left(\frac{t}{\sqrt n}\right)\right]^n.
$$
Sviluppando $\varphi$ al secondo ordine in $0$, $\varphi(u)=1-u^2/2+o(u^2)$, si ottiene
$$
\varphi_{Z_n}(t)=\left(1-\frac{t^2}{2n}+o\left(\frac1n\right)\right)^n\to e^{-t^2/2},
$$
che è la funzione caratteristica di $\mathcal{N}(0,1)$. Per il teorema di continuità di Lévy, la convergenza delle funzioni caratteristiche implica la convergenza in distribuzione. $\square$

> **Osservazione 7.4.**  
> Il CLT è la ragione per cui la normale è ovunque in inferenza. Non richiede che le $X_i$ siano normali: la somma standardizzata è approssimativamente normale indipendentemente dalla forma della distribuzione, purché la varianza sia finita. L'approssimazione migliora con $n$ e con la simmetria della distribuzione; per distribuzioni molto asimmetriche servono $n$ più grandi.

## 8. Risultati di Slutsky e metodo delta

### Teorema 8.1 — Slutsky

Siano $\{X_n\}$ e $\{Y_n\}$ successioni di v.c. con $X_n\xrightarrow{D}X$ e $Y_n\xrightarrow{p}c$ (costante deterministica). Allora:

1. $X_n+Y_n\xrightarrow{D}X+c$;
2. $X_nY_n\xrightarrow{D}cX$;
3. se $c\neq0$, $X_n/Y_n\xrightarrow{D}X/c$;
4. se $g$ è continua in $c$, $g(Y_n)\xrightarrow{p}g(c)$.

> **Osservazione 8.2.**  
> Slutsky è lo strumento che permette di "sostituire" una quantità che converge in probabilità a una costante all'interno di una convergenza in distribuzione. È il motivo per cui nel CLT si può sostituire $\sigma$ con la sua stima campionaria $S_n$ e continuare a ottenere la stessa normale limite.

### Teorema 8.3 — Metodo delta

Sia $\{X_n\}$ una successione di v.c. con
$$
\sqrt n(X_n-\theta)\xrightarrow{D}\mathcal{N}(0,\sigma^2),
$$
e sia $g$ derivabile in $\theta$ con $g'(\theta)\neq0$. Allora
$$
\sqrt n(g(X_n)-g(\theta))\xrightarrow{D}\mathcal{N}(0,\sigma^2[g'(\theta)]^2).
$$

**Dimostrazione.**  
Per il teorema di Taylor al primo ordine,
$$
g(X_n)=g(\theta)+g'(\theta)(X_n-\theta)+R_n,
$$
dove $R_n=o_P(|X_n-\theta|)$. Moltiplicando per $\sqrt n$,
$$
\sqrt n(g(X_n)-g(\theta))=g'(\theta)\sqrt n(X_n-\theta)+\sqrt n R_n.
$$
Il primo termine converge a $\mathcal{N}(0,\sigma^2[g'(\theta)]^2)$ per Slutsky. Il secondo termine è $o_P(1)$ perché $\sqrt n R_n=o_P(\sqrt n|X_n-\theta|)=o_P(1)$ per l'ipotesi di convergenza in distribuzione. $\square$

## 9. Distribuzioni notevoli discrete

### 9.1 Bernoulli

**Definizione.** $X\sim\operatorname{Ber}(\pi)$, $\pi\in(0,1)$, se
$$
P(X=1)=\pi,\qquad P(X=0)=1-\pi.
$$

**Momenti.** $E[X]=\pi$, $\operatorname{Var}(X)=\pi(1-\pi)$.

**Applicazione.** Modella un esperimento dicotomico (successo/insuccesso): esito di una prova, risposta a un trattamento, presenza/assenza di un difetto.

### 9.2 Binomiale

**Definizione.** $Y\sim\operatorname{Bi}(n,\pi)$ se
$$
P(Y=y)=\binom{n}{y}\pi^y(1-\pi)^{n-y},\qquad y=0,1,\dots,n.
$$

**Caratterizzazione.** $Y=\sum_{i=1}^nX_i$, con $X_i\sim\operatorname{Ber}(\pi)$ i.i.d.

**Momenti.** $E[Y]=n\pi$, $\operatorname{Var}(Y)=n\pi(1-\pi)$.

**Proprietà di additività.** Se $Y_1\sim\operatorname{Bi}(n_1,\pi)$ e $Y_2\sim\operatorname{Bi}(n_2,\pi)$ indipendenti, allora $Y_1+Y_2\sim\operatorname{Bi}(n_1+n_2,\pi)$.

**Applicazione.** Conteggio di successi in $n$ prove indipendenti con la stessa probabilità: numero di pezzi difettosi in un lotto, numero di pazienti che rispondono a una terapia.

**Approssimazione di Poisson.** Se $n\to\infty$, $\pi\to0$ con $n\pi\to\lambda>0$, allora
$$
\operatorname{Bi}(n,\pi)\xrightarrow{D}\operatorname{Po}(\lambda).
$$
**Applicazione tipica.** Conteggio di eventi rari in un grande numero di prove (guasti, difetti rari, arrivi in un call center in un piccolo intervallo).

**Approssimazione normale.** Per $n$ grande, per il CLT,
$$
\frac{Y-n\pi}{\sqrt{n\pi(1-\pi)}}\xrightarrow{D}\mathcal{N}(0,1).
$$
**Regola pratica.** Buona se $n\pi\ge5$ e $n(1-\pi)\ge5$.

### 9.3 Binomiale negativa

**Definizione.** $Y\sim\operatorname{Bineg}(k,\pi)$ se conta il numero di **insuccessi** prima di ottenere $k$ successi:
$$
P(Y=y)=\binom{y+k-1}{y}\pi^k(1-\pi)^y,\qquad y=0,1,\dots
$$

**Momenti.** $E[Y]=k(1-\pi)/\pi$, $\operatorname{Var}(Y)=k(1-\pi)/\pi^2$.

**Proprietà di additività.** Somma di $k$ variabili geometriche indipendenti.

**Applicazione.** Modellazione di tempi di attesa "contati in insuccessi" prima di un numero fissato di successi; in pratica, è spesso usata come alternativa alla Poisson quando i dati sono **sovradispersi** (varianza maggiore della media).

### 9.4 Geometrica

**Definizione.** $Y\sim\operatorname{Ge}(\pi)$ è il caso $k=1$ della binomiale negativa: conta il numero di insuccessi prima del primo successo.
$$
P(Y=y)=\pi(1-\pi)^y,\qquad y=0,1,\dots
$$

**Momenti.** $E[Y]=(1-\pi)/\pi$, $\operatorname{Var}(Y)=(1-\pi)/\pi^2$.

**Proprietà di assenza di memoria.**
$$
P(Y>x+y\mid Y>x)=P(Y>y).
$$
**Dimostrazione.**  
$P(Y>x+y\mid Y>x)=P(Y>x+y)/P(Y>x)$. Ora $P(Y>t)=(1-\pi)^{t+1}$, quindi
$$
\frac{(1-\pi)^{x+y+1}}{(1-\pi)^{x+1}}=(1-\pi)^y=P(Y>y).
$$
$\square$

**Applicazione.** Tempi di attesa discreti tra eventi indipendenti (numero di lanci di un dado prima di un 6). L'assenza di memoria la rende l'analogo discreto dell'esponenziale.

### 9.5 Poisson

**Definizione.** $Y\sim\operatorname{Po}(\lambda)$ se
$$
P(Y=y)=e^{-\lambda}\frac{\lambda^y}{y!},\qquad y=0,1,\dots
$$

**Momenti.** $E[Y]=\lambda$, $\operatorname{Var}(Y)=\lambda$.

**Proprietà di additività.** Se $Y_1\sim\operatorname{Po}(\lambda_1)$ e $Y_2\sim\operatorname{Po}(\lambda_2)$ indipendenti, $Y_1+Y_2\sim\operatorname{Po}(\lambda_1+\lambda_2)$.

**Approssimazione normale.** Per $\lambda$ grande,
$$
\frac{Y-\lambda}{\sqrt\lambda}\xrightarrow{D}\mathcal{N}(0,1).
$$
Regola pratica: buona per $\lambda\ge10$.

**Applicazione.** Conteggio di eventi in un dominio fissato (tempo, spazio), con eventi indipendenti e tasso costante: numero di chiamate a un call center in un'ora, numero di guasti in un mese, numero di particelle emesse in un intervallo.

### 9.6 Correzione di continuità

**Osservazione 9.6.1 — Perché serve.**  
Quando si approssima una distribuzione **discreta** (Binomiale, Poisson) con una **continua** (Normale), si commette un errore sistematico: la discreta concentra massa su punti, la continua su intervalli. La correzione di continuità migliora l'approssimazione aggiustando gli estremi dell'intervallo di un fattore $1/2$.

**Regola.** Se $X$ è discreta e si vuole approssimare $P(X\le k)$, $P(X\ge k)$, $P(a\le X\le b)$, si usano le versioni continue con $\pm0.5$:

$$
P(X\le k)\approx P\left(Z\le\frac{k+0.5-n\pi}{\sqrt{n\pi(1-\pi)}}\right),
$$
$$
P(X\ge k)\approx P\left(Z\ge\frac{k-0.5-n\pi}{\sqrt{n\pi(1-\pi)}}\right),
$$
$$
P(a\le X\le b)\approx P\left(\frac{a-0.5-n\pi}{\sqrt{n\pi(1-\pi)}}\le Z\le\frac{b+0.5-n\pi}{\sqrt{n\pi(1-\pi)}}\right),
$$
dove $Z\sim\mathcal{N}(0,1)$.

**Esempio.** Sia $Y\sim\operatorname{Bi}(50,0.4)$, $E[Y]=20$, $\operatorname{Var}(Y)=12$, $\sigma\approx3.464$. Si vuole $P(Y\le 15)$.
Senza correzione: $P(Z\le(15-20)/3.464)=P(Z\le-1.443)\approx0.0745$.
Con correzione: $P(Z\le(15.5-20)/3.464)=P(Z\le-1.299)\approx0.0970$.
Il valore esatto è $P(Y\le15)\approx0.0955$, quindi la correzione riduce l'errore di un ordine di grandezza.

**Osservazione 9.6.2.**  
La correzione di continuità è utile quando $n$ non è grandissimo. Per $n$ molto grande, l'errore introdotto dal $0.5$ diventa trascurabile rispetto alla scala $\sqrt{n\pi(1-\pi)}$ e si può omettere.

## 10. Distribuzioni notevoli continue

### 10.1 Uniforme

**Definizione.** $X\sim U(a,b)$, $a<b$, con densità
$$
f_X(x)=\frac{1}{b-a}\mathbf{1}_{(a,b)}(x).
$$

**Momenti.** $E[X]=\frac{a+b}{2}$, $\operatorname{Var}(X)=\frac{(b-a)^2}{12}$.

**Applicazione.** Modella l'incertezza totale su un intervallo limitato: tempo di attesa a un semaforo con rosso di durata fissata, posizione di un ago fatto girare su una ruota. È anche la base per la generazione di numeri pseudocasuali.

### 10.2 Normale

**Definizione.** $X\sim\mathcal{N}(\mu,\sigma^2)$, $\mu\in\mathbb{R}$, $\sigma^2>0$, con densità
$$
f_X(x)=\frac{1}{\sqrt{2\pi\sigma^2}}\exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right).
$$

**Momenti.** $E[X]=\mu$, $\operatorname{Var}(X)=\sigma^2$.

**Chiusura per trasformazioni affini.** Se $X\sim\mathcal{N}(\mu,\sigma^2)$ e $Y=a+bX$, allora $Y\sim\mathcal{N}(a+b\mu,b^2\sigma^2)$.

**Standardizzazione.** Se $X\sim\mathcal{N}(\mu,\sigma^2)$, allora $Z=(X-\mu)/\sigma\sim\mathcal{N}(0,1)$.

**Combinazioni lineari.** Se $Y_i\sim\mathcal{N}(\mu_i,\sigma_i^2)$ e sono indipendenti, allora $\sum_i\beta_iY_i\sim\mathcal{N}\left(\sum_i\beta_i\mu_i,\sum_i\beta_i^2\sigma_i^2\right)$.

**Approssimazione del resto.** La normale è la distribuzione limite nel CLT; approssima la Binomiale e la Poisson per parametri grandi, con eventuale correzione di continuità.

**Applicazione.** Errori di misura, altezze, pesi, rumore additivo, e qualunque grandezza che sia somma di molti contributi indipendenti.

### 10.3 Gamma

**Definizione.** $X\sim\Gamma(\alpha,\lambda)$, $\alpha>0$, $\lambda>0$, con densità
$$
f_X(x)=\frac{\lambda^\alpha}{\Gamma(\alpha)}x^{\alpha-1}e^{-\lambda x}\mathbf{1}_{(0,\infty)}(x),
$$
dove $\Gamma(\alpha)=\int_0^\infty t^{\alpha-1}e^{-t}\,dt$.

**Momenti.** $E[X]=\alpha/\lambda$, $\operatorname{Var}(X)=\alpha/\lambda^2$.

**Proprietà.**
1. Cambio di scala: se $X\sim\Gamma(\alpha,\lambda)$, $b>0$, allora $bX\sim\Gamma(\alpha,\lambda/b)$.
2. **Additività:** se $X_1\sim\Gamma(\alpha_1,\lambda)$ e $X_2\sim\Gamma(\alpha_2,\lambda)$ indipendenti, $X_1+X_2\sim\Gamma(\alpha_1+\alpha_2,\lambda)$. **Richiede lo stesso parametro di scala $\lambda$.**
3. Per $\alpha\to\infty$, $\Gamma(\alpha,\lambda)$ è approssimabile con $\mathcal{N}(\alpha/\lambda,\alpha/\lambda^2)$.

**Applicazione.** Modella tempi di attesa e durate, in modo flessibile grazie ai due parametri. La sua importanza è che contiene come casi particolari la $\chi^2$ e l'esponenziale.

### 10.4 Esponenziale

**Definizione.** $X\sim\operatorname{Esp}(\lambda)$, $\lambda>0$, con densità
$$
f_X(x)=\lambda e^{-\lambda x}\mathbf{1}_{(0,\infty)}(x).
$$

**Momenti.** $E[X]=1/\lambda$, $\operatorname{Var}(X)=1/\lambda^2$.

**Proprietà.**
1. **Assenza di memoria:** $P(X>s+t\mid X>s)=P(X>t)$.
2. **Cambio di scala:** se $X\sim\operatorname{Esp}(\lambda)$ e $b>0$, $bX\sim\operatorname{Esp}(\lambda/b)$.
3. **Connessione con la Poisson:** se il numero di arrivi per unità di tempo è $\operatorname{Po}(\lambda)$, allora il tempo di attesa prima del primo arrivo è $\operatorname{Esp}(\lambda)$ e il tempo prima dell'$n$-esimo arrivo è $\Gamma(n,\lambda)$.

**Dimostrazione dell'assenza di memoria.**
$$
P(X>s+t\mid X>s)=\frac{P(X>s+t)}{P(X>s)}=\frac{e^{-\lambda(s+t)}}{e^{-\lambda s}}=e^{-\lambda t}=P(X>t).
$$
$\square$

**Applicazione.** Tempi di vita di componenti senza usura, tempi di decadimento, tempi di attesa tra eventi indipendenti con tasso costante.

## 11. Distribuzioni derivate dalla normale

Tutte le distribuzioni seguenti si ottengono a partire da normali standard indipendenti e sono centrali per l'inferenza.

### 11.1 Chi-quadro

**Definizione.** Siano $Z_1,\dots,Z_\nu$ i.i.d. $\mathcal{N}(0,1)$. Allora
$$
U=\sum_{i=1}^\nu Z_i^2\sim\chi^2_\nu.
$$
In termini della Gamma:
$$
\chi^2_\nu=\Gamma\left(\frac{\nu}{2},\frac12\right).
$$

**Momenti.** $E[U]=\nu$, $\operatorname{Var}(U)=2\nu$.

**Additività.** Se $U\sim\chi^2_\nu$ e $V\sim\chi^2_\zeta$ indipendenti, $U+V\sim\chi^2_{\nu+\zeta}$.

**Applicazione.** Test di ipotesi sulla varianza di popolazioni normali, test di bontà di adattamento, test di indipendenza, analisi della devianza in ANOVA.

### 11.2 t di Student

**Definizione.** Siano $Z\sim\mathcal{N}(0,1)$ e $U\sim\chi^2_\nu$ indipendenti. Allora
$$
T=\frac{Z}{\sqrt{U/\nu}}\sim t_\nu.
$$

**Momenti.** $E[T]=0$ per $\nu>1$; $\operatorname{Var}(T)=\nu/(\nu-2)$ per $\nu>2$.

**Proprietà.** Simmetrica, campanulare, con code più pesanti della normale. Per $\nu\to\infty$, $t_\nu\to\mathcal{N}(0,1)$.

**Applicazione.** Inferenza sulla media di una popolazione normale con varianza **non nota**: $\frac{\bar X-\mu}{S/\sqrt n}\sim t_{n-1}$. Test e intervalli di confidenza per la differenza tra medie, dati appaiati, coefficienti di regressione.

### 11.3 F di Fisher

**Definizione.** Siano $U\sim\chi^2_\nu$ e $V\sim\chi^2_\zeta$ indipendenti. Allora
$$
F=\frac{U/\nu}{V/\zeta}\sim F_{\nu,\zeta}.
$$

**Proprietà.**
1. $1/F\sim F_{\zeta,\nu}$.
2. Se $T\sim t_\nu$, allora $T^2\sim F_{1,\nu}$.

**Applicazione.** Confronto tra varianze, ANOVA, test di ipotesi su modelli lineari. Il test F è il rapporto tra due varianze campionarie indipendenti sotto normalità.

## 12. Variabile casuale multinomiale

### Definizione 12.1 — Multinomiale

Sia $\pi=(\pi_1,\dots,\pi_K)$ un vettore di probabilità ($\pi_j>0$, $\sum_j\pi_j=1$). Il vettore
$$
\mathbf{Y}=(Y_1,\dots,Y_K)^T\sim\operatorname{Bi}_K(n,\pi)
$$
conta il numero di volte in cui ciascuna delle $K$ modalità si presenta in $n$ prove indipendenti. La funzione di probabilità è
$$
P(\mathbf{Y}=\mathbf{y})=\frac{n!}{y_1!\cdots y_K!}\prod_{j=1}^K\pi_j^{y_j},
\qquad \sum_j y_j=n,\ y_j\in\mathbb{N}.
$$

**Costruzione.** Ogni prova produce un vettore $\mathbf{Z}_i=(Z_{i1},\dots,Z_{iK})^T$ con una sola componente uguale a 1 (modalità osservata). Allora $Z_{ij}\sim\operatorname{Ber}(\pi_j)$ e $\mathbf{Y}=\sum_{i=1}^n\mathbf{Z}_i$.

**Momenti.**
$$
E[\mathbf{Y}]=n\pi,
\qquad
\operatorname{Var}(\mathbf{Y})=n(\operatorname{diag}(\pi)-\pi\pi^T).
$$
La matrice di varianza-covarianza ha $\operatorname{Var}(Y_j)=n\pi_j(1-\pi_j)$ e $\operatorname{Cov}(Y_j,Y_h)=-n\pi_j\pi_h$ per $j\neq h$.

**Proprietà.**
1. **Marginali.** $Y_j\sim\operatorname{Bi}(n,\pi_j)$.
2. **Marginali congiunte.** Per ogni sottoinsieme di componenti, la loro distribuzione congiunta è multinomiale con le stesse probabilità ristrette e l'ultima componente che raccoglie le rimanenti.
3. **Condizionate.** Date le componenti diverse dalla $j$-esima, $Y_j$ è binomiale.
4. **Additività.** Se $\mathbf{Y}\sim\operatorname{Bi}_K(n,\pi)$ e $\mathbf{X}\sim\operatorname{Bi}_K(m,\pi)$ indipendenti, allora $\mathbf{Y}+\mathbf{X}\sim\operatorname{Bi}_K(n+m,\pi)$.
5. **Approssimazione normale.** Per $n\to\infty$,
$$
\frac{\mathbf{Y}-n\pi}{\sqrt n}\xrightarrow{D}\mathcal{N}_K(\mathbf{0},\Sigma),
$$
dove $\Sigma=\operatorname{diag}(\pi)-\pi\pi^T$, di rango $K-1$.

**Applicazione.** Tabelle di contingenza, preferenze su $K$ alternative, gruppi sanguigni, esiti di un sondaggio categorico.

> **Osservazione 12.2.**  
> La multinomiale è la generalizzazione naturale della binomiale e la base dei test chi-quadro. La dipendenza negativa tra le componenti ($\operatorname{Cov}<0$) riflette il vincolo $\sum_jY_j=n$: se una modalità ha più osservazioni del previsto, le altre devono averne meno.

## 13. Variabile casuale normale multivariata

### 13.1 Normale multivariata standard

**Definizione.** Siano $Z_1,\dots,Z_p$ i.i.d. $\mathcal{N}(0,1)$. Il vettore $\mathbf{Z}=(Z_1,\dots,Z_p)^T$ ha densità
$$
f_{\mathbf{Z}}(\mathbf{z})=(2\pi)^{-p/2}\exp\left(-\frac12\mathbf{z}^T\mathbf{z}\right),
$$
e si scrive $\mathbf{Z}\sim\mathcal{N}_p(\mathbf{0},I_p)$.

### 13.2 Normale multivariata generale

**Definizione.** Sia $\boldsymbol\mu\in\mathbb{R}^p$, $\boldsymbol\Sigma$ matrice $p\times p$ simmetrica definita positiva. Sia $A$ tale che $\boldsymbol\Sigma=AA^T$ (esiste per il teorema spettrale o per la fattorizzazione di Cholesky). Si definisce
$$
\mathbf{Y}=A\mathbf{Z}+\boldsymbol\mu,
$$
dove $\mathbf{Z}\sim\mathcal{N}_p(\mathbf{0},I_p)$.

Applicando il cambio di variabili in dimensione $p$ con $g^{-1}(\mathbf{y})=A^{-1}(\mathbf{y}-\boldsymbol\mu)$, lo Jacobiano ha determinante $|A|^{-1}=|\boldsymbol\Sigma|^{-1/2}$, e
$$
\mathbf{z}^T\mathbf{z}=(\mathbf{y}-\boldsymbol\mu)^T(A^{-1})^TA^{-1}(\mathbf{y}-\boldsymbol\mu)=(\mathbf{y}-\boldsymbol\mu)^T\boldsymbol\Sigma^{-1}(\mathbf{y}-\boldsymbol\mu).
$$
Quindi
$$
f_{\mathbf{Y}}(\mathbf{y})=\frac{1}{(2\pi)^{p/2}|\boldsymbol\Sigma|^{1/2}}\exp\left(-\frac12(\mathbf{y}-\boldsymbol\mu)^T\boldsymbol\Sigma^{-1}(\mathbf{y}-\boldsymbol\mu)\right).
$$
Si scrive $\mathbf{Y}\sim\mathcal{N}_p(\boldsymbol\mu,\boldsymbol\Sigma)$.

**Momenti.** $E[\mathbf{Y}]=\boldsymbol\mu$, $\operatorname{Var}(\mathbf{Y})=\boldsymbol\Sigma$.

### 13.3 Proprietà

1. **Trasformazioni lineari.** Se $\mathbf{Y}\sim\mathcal{N}_p(\boldsymbol\mu,\boldsymbol\Sigma)$ e $\mathbf{X}=B\mathbf{Y}+\mathbf{b}$, allora
$$
\mathbf{X}\sim\mathcal{N}_q(B\boldsymbol\mu+\mathbf{b},B\boldsymbol\Sigma B^T).
$$
2. **Marginali.** Ogni sotto-vettore di $\mathbf{Y}$ è normale multivariato con la media e la varianza corrispondenti.
3. **Condizionate.** Le distribuzioni condizionate sono normali multivariate. In particolare, partizionando $\mathbf{Y}=(\mathbf{Y}_1,\mathbf{Y}_2)$ con medie e varianze coerenti,
$$
\mathbf{Y}_1\mid\mathbf{Y}_2=\mathbf{y}_2\sim\mathcal{N}\left(\boldsymbol\mu_1+\boldsymbol\Sigma_{12}\boldsymbol\Sigma_{22}^{-1}(\mathbf{y}_2-\boldsymbol\mu_2),\,\boldsymbol\Sigma_{11}-\boldsymbol\Sigma_{12}\boldsymbol\Sigma_{22}^{-1}\boldsymbol\Sigma_{21}\right).
$$
4. **Incorrelazione implica indipendenza.** Per la normale multivariata, se $\operatorname{Cov}(Y_i,Y_j)=0$ per ogni coppia di componenti di due sotto-vettori, allora i sotto-vettori sono indipendenti. In generale, per altre distribuzioni, questo non vale.

### 13.4 Caso bivariato

Sia $\mathbf{Y}=(Y_1,Y_2)^T\sim\mathcal{N}_2(\boldsymbol\mu,\boldsymbol\Sigma)$, con
$$
\boldsymbol\Sigma=\begin{pmatrix}\sigma_1^2&\rho\sigma_1\sigma_2\\\rho\sigma_1\sigma_2&\sigma_2^2\end{pmatrix}.
$$
Allora $|\boldsymbol\Sigma|=\sigma_1^2\sigma_2^2(1-\rho^2)$ e
$$
f_{Y_1,Y_2}(y_1,y_2)=\frac{1}{2\pi\sigma_1\sigma_2\sqrt{1-\rho^2}}\exp\left\{-\frac{1}{2(1-\rho^2)}\left[\left(\frac{y_1-\mu_1}{\sigma_1}\right)^2-2\rho\frac{y_1-\mu_1}{\sigma_1}\frac{y_2-\mu_2}{\sigma_2}+\left(\frac{y_2-\mu_2}{\sigma_2}\right)^2\right]\right\}.
$$
Per $\rho=0$, la densità fattorizza e le componenti sono indipendenti.

### 13.5 Distanza di Mahalanobis

**Definizione.** Dato $\mathbf{Y}\sim\mathcal{N}_p(\boldsymbol\mu,\boldsymbol\Sigma)$, la **distanza di Mahalanobis** al quadrato di $\mathbf{y}$ da $\boldsymbol\mu$ è
$$
d_M(\mathbf{y},\boldsymbol\mu)^2=(\mathbf{y}-\boldsymbol\mu)^T\boldsymbol\Sigma^{-1}(\mathbf{y}-\boldsymbol\mu).
$$

**Proprietà.**
1. È una norma indotta da $\boldsymbol\Sigma^{-1}$: $d_M(\mathbf{y}_1,\mathbf{y}_2)^2=(\mathbf{y}_1-\mathbf{y}_2)^T\boldsymbol\Sigma^{-1}(\mathbf{y}_1-\mathbf{y}_2)$.
2. È invariante per trasformazioni lineari non singolari: se $\mathbf{X}=B\mathbf{Y}$, allora $d_M(\mathbf{x},B\boldsymbol\mu)^2=d_M(\mathbf{y},\boldsymbol\mu)^2$.
3. Per $\boldsymbol\Sigma=I_p$, si riduce alla distanza euclidea.

**Teorema 13.6.**  
Se $\mathbf{Y}\sim\mathcal{N}_p(\boldsymbol\mu,\boldsymbol\Sigma)$ con $\boldsymbol\Sigma$ definita positiva, allora
$$
(\mathbf{Y}-\boldsymbol\mu)^T\boldsymbol\Sigma^{-1}(\mathbf{Y}-\boldsymbol\mu)\sim\chi^2_p.
$$

**Dimostrazione.**  
Sia $A$ tale che $\boldsymbol\Sigma=AA^T$ e $\mathbf{Z}=A^{-1}(\mathbf{Y}-\boldsymbol\mu)\sim\mathcal{N}_p(\mathbf{0},I_p)$. Allora
$$
(\mathbf{Y}-\boldsymbol\mu)^T\boldsymbol\Sigma^{-1}(\mathbf{Y}-\boldsymbol\mu)=\mathbf{Z}^T\mathbf{Z}=\sum_{i=1}^pZ_i^2,
$$
che è somma di $p$ quadrati di normali standard indipendenti, quindi $\chi^2_p$. $\square$

> **Osservazione 13.7.**  
> Le curve di livello della densità normale multivariata sono ellissoidi centrati in $\boldsymbol\mu$, definiti da $d_M^2=\text{costante}$. Al variare del livello, si ottengono ellissoidi che racchiudono una frazione crescente di probabilità. Per $\boldsymbol\Sigma$ diagonale, gli assi dell'ellissoide sono paralleli agli assi coordinati; per $\boldsymbol\Sigma$ con correlazioni non nulle, sono ruotati. La forma e l'orientamento dell'ellissoide sono determinati dagli autovettori e dagli autovalori di $\boldsymbol\Sigma$.

## 14. Teorema del limite centrale multivariato

**Teorema 14.1.**  
Sia $\{\mathbf{Y}^{(i)}\}$ una successione di vettori casuali $p$-dimensionali i.i.d. con $E[\mathbf{Y}^{(i)}]=\boldsymbol\mu$ e $\operatorname{Var}(\mathbf{Y}^{(i)})=\boldsymbol\Sigma$ definita positiva. Allora
$$
\sqrt n(\bar{\mathbf{Y}}_n-\boldsymbol\mu)\xrightarrow{D}\mathcal{N}_p(\mathbf{0},\boldsymbol\Sigma),
$$
dove $\bar{\mathbf{Y}}_n=\frac1n\sum_{i=1}^n\mathbf{Y}^{(i)}$.

**Idea della dimostrazione.**  
Si dimostra che, per ogni $\mathbf{a}\in\mathbb{R}^p$ con $\mathbf{a}\neq\mathbf{0}$, la successione $\mathbf{a}^T\sqrt n(\bar{\mathbf{Y}}_n-\boldsymbol\mu)$ converge in distribuzione a $\mathcal{N}(0,\mathbf{a}^T\boldsymbol\Sigma\mathbf{a})$ per il CLT univariato applicato alle variabili $\mathbf{a}^T\mathbf{Y}^{(i)}$. Poiché la convergenza vale per ogni combinazione lineare, per il teorema di Cramér–Wold la convergenza in distribuzione vale per il vettore. $\square$

> **Osservazione 14.2.**  
> Il CLT multivariato è lo strumento che rende la normale multivariata la distribuzione limite della media campionaria di un vettore. Da esso deriva la distribuzione asintotica $\chi^2$ delle forme quadratiche $n(\bar{\mathbf{Y}}_n-\boldsymbol\mu)^T\boldsymbol\Sigma^{-1}(\bar{\mathbf{Y}}_n-\boldsymbol\mu)$, che è alla base dei test su vettori di medie.
