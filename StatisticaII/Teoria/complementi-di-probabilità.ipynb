# Complementi di Probabilità

---

## 0. Notazioni e richiami

In queste note, $(\Omega, \mathcal{F}, P)$ indica uno spazio di probabilità, dove:

- $\Omega$ è lo **spazio campionario**, cioè l'insieme di tutti i possibili esiti di un esperimento aleatorio;
- $\mathcal{F}$ è una **$\sigma$-algebra** su $\Omega$, cioè una famiglia di sottoinsiemi di $\Omega$ chiusa rispetto a complementare e unione numerabile;
- $P : \mathcal{F} \to [0,1]$ è una **misura di probabilità**, cioè una funzione $\sigma$-additiva con $P(\Omega)=1$.

Gli elementi di $\mathcal{F}$ sono detti **eventi**. Per una variabile aleatoria $X$, indichiamo con $F_X$ la sua funzione di ripartizione e con $p_X$ (o $f_X$) la sua funzione di probabilità (o densità).

# 1. Definizione di Spazio di Probabilità

## 1.1 Esperimenti aleatori e spazio campionario

### Definizione 1.1 — Esperimento aleatorio

Un **esperimento aleatorio** (o casuale) è un esperimento il cui risultato non è noto a priori. Singole esecuzioni dello stesso esperimento, condotte in identiche condizioni, possono dar luogo a risultati diversi.

### Definizione 1.2 — Evento elementare e spazio campionario

Il singolo risultato di un esperimento aleatorio si chiama **evento elementare**. Gli eventi elementari sono tra loro:

- **incompatibili**: il verificarsi dell'uno esclude il verificarsi degli altri;
- **esaustivi**: certamente uno di essi si verifica.

L'insieme di tutti i possibili eventi elementari è detto **spazio degli eventi** o **spazio campionario**, e si indica con $\Omega$.

### Definizione 1.3 — Evento aleatorio

Un **evento aleatorio** è un sottoinsieme dello spazio campionario, cioè un elemento di $\mathcal{F}$.

### Esempio 1.4 — Lancio di un dado

Si lanci un dado a sei facce. Gli eventi elementari sono

$$
E_1 = \text{“esce 1”},\quad E_2 = \text{“esce 2”},\quad \dots,\quad E_6 = \text{“esce 6”}.
$$

Lo spazio campionario è

$$
\Omega = \{E_1, E_2, E_3, E_4, E_5, E_6\}.
$$

L'evento aleatorio “esce un numero dispari” è il sottoinsieme

$$
A = \{E_1, E_3, E_5\}.
$$

## 1.2 Definizioni di probabilità

### Definizione 1.5 — Probabilità (definizione soggettivista)

La probabilità è una **misura del grado di fiducia** relativo al verificarsi di un evento.

### Definizione 1.6 — Probabilità (definizione classica)

La probabilità di un evento è il rapporto tra il numero di casi favorevoli al verificarsi dello stesso e il numero di casi possibili, purché tutti i casi siano equiprobabili:

$$
P(A) = \frac{|A|}{|\Omega|}.
$$

### Definizione 1.7 — Probabilità (definizione frequentista)

In una serie di esperimenti ripetuti un gran numero di volte nelle stesse condizioni, la probabilità di un evento è la **frequenza relativa** con cui esso si verifica:

$$
P(A) = \lim_{n \to \infty} \frac{n_A}{n},
$$

dove $n_A$ è il numero di volte in cui $A$ si verifica su $n$ prove.

### Definizione 1.8 — Probabilità (definizione assiomatica)

La probabilità è una funzione $P : \mathcal{F} \to [0,1]$ definita sullo spazio campionario che soddisfa i seguenti **assiomi di Kolmogorov**:

1. **Normalizzazione:** $P(\Omega) = 1$;
2. **Non negatività:** $P(A) \geq 0$ per ogni evento $A \in \mathcal{F}$;
3. **$\sigma$-additività:** se $\{A_i\}_{i \in \mathbb{N}}$ è una famiglia numerabile di eventi a due a due disgiunti, allora
   $$
   P\left(\bigcup_{i=1}^\infty A_i\right) = \sum_{i=1}^\infty P(A_i).
   $$

### Proprietà 1.9 — Conseguenze degli assiomi

Valgono le seguenti proprietà:

1. Se $A^C$ è l'evento complementare di $A$, allora
   $$
   P(A^C) = 1 - P(A).
   $$
2. Se $\emptyset$ è l'evento impossibile, allora $P(\emptyset) = 0$.
3. Per due eventi qualsiasi $A, B \in \mathcal{F}$,
   $$
   P(A \cup B) = P(A) + P(B) - P(A \cap B).
   $$
4. **Probabilità condizionata:** se $P(B) > 0$,
   $$
   P(A \mid B) = \frac{P(A \cap B)}{P(B)}.
   $$
5. **Formula della probabilità totale:** se $\{B_i\}_{i \in I}$ è una partizione di $\Omega$ con $P(B_i)>0$,
   $$
   P(A) = \sum_{i \in I} P(A \mid B_i) P(B_i).
   $$
6. **Formula di Bayes:** se $P(A) > 0$,
   $$
   P(B_i \mid A) = \frac{P(A \mid B_i) P(B_i)}{\sum_j P(A \mid B_j) P(B_j)}.
   $$

# 2. Variabili aleatorie unidimensionali

## 2.1 Definizione di variabile aleatoria

### Definizione 2.1 — Variabile aleatoria

Dato uno spazio di probabilità $(\Omega, \mathcal{F}, P)$, una **variabile aleatoria** (v.c.) è una funzione misurabile

$$
X : (\Omega, \mathcal{F}) \to (\mathbb{R}, \mathcal{B}(\mathbb{R}))
$$

che associa a ogni possibile risultato dell'esperimento un numero reale.

### Definizione 2.2 — Supporto

Il **supporto** di una v.c. $X$ è l'insieme di tutti i possibili valori che $X$ può assumere:

$$
S(X) = \{x \in \mathbb{R} : P(X = x) > 0 \text{ se discreta}\},
$$

oppure, nel caso continuo, la chiusura dell'insieme dei valori a densità positiva.

Si distingue tra:

- **v.c. discreta**: il supporto è un insieme finito o numerabile;
- **v.c. continua**: il supporto è un intervallo dell'asse reale.

### Esempio 2.3

- Il numero di incidenti su un tratto di strada in una giornata è una v.c. discreta con supporto $\{0,1,2,\dots\}$.
- Il supporto della v.c. “lancio di un dado” è $\{1,2,3,4,5,6\}$.
- Il tempo di attesa ha supporto continuo e non negativo.

## 2.2 Distribuzione di probabilità

### Definizione 2.4 — Distribuzione di probabilità (caso discreto)

Per una v.c. discreta $X$ con supporto $S$, la **distribuzione di probabilità** è la funzione

$$
p_X(x) = P(X = x), \qquad x \in S,
$$

che soddisfa:

$$
p_X(x) \geq 0, \qquad \sum_{x \in S} p_X(x) = 1.
$$

### Esempio 2.5 — Lancio di un dado

Sia $X$ l'esito del lancio di un dado. Allora $S = \{1,2,3,4,5,6\}$ e

$$
p_X(x) = \frac{1}{6}, \quad x \in S.
$$

Si ha $\sum_{x=1}^6 \frac{1}{6} = 1$.

### Definizione 2.6 — Distribuzione di probabilità (caso continuo)

Per una v.c. continua $X$ con supporto $S$ (un intervallo), la distribuzione è data da una **funzione di densità** $f_X : S \to \mathbb{R}^+$ tale che:

$$
f_X(x) \geq 0, \qquad \int_S f_X(x)\, dx = 1,
$$

e per ogni $[a,b] \subseteq S$,

$$
P(a \leq X \leq b) = \int_a^b f_X(x)\, dx.
$$

> **Osservazione 2.7.**  
> Per una v.c. continua, $P(X = x) = 0$ per ogni $x$, ma la probabilità di un intervallo è positiva.

### Esempio 2.8 — Ago rotante

Sia $X$ la posizione, in gradi, in cui si ferma un ago fatto girare su una ruota, con $S = [0,360]$. Allora $P(X = x) = 0$ per ogni $x$, ma

$$
P(X \in [0,90]) = \frac{1}{4}.
$$

## 2.3 Funzione di ripartizione

### Definizione 2.9 — Funzione di ripartizione

La **funzione di ripartizione** di una v.c. $X$ è

$$
F_X(x) = P(X \leq x) =
\begin{cases}
\displaystyle \sum_{u \leq x} p_X(u), & \text{se } X \text{ è discreta},\\[6pt]
\displaystyle \int_{-\infty}^x f_X(u)\, du, & \text{se } X \text{ è continua}.
\end{cases}
$$

### Proprietà 2.10

La funzione di ripartizione gode delle seguenti proprietà:

1. $F_X : \mathbb{R} \to [0,1]$;
2. $\displaystyle \lim_{x \to -\infty} F_X(x) = 0$ e $\displaystyle \lim_{x \to +\infty} F_X(x) = 1$;
3. $F_X$ è monotona non decrescente;
4. $P(a < X \leq b) = F_X(b) - F_X(a)$ per $b > a$;
5. $F_X$ è continua a destra (per v.c. discrete è una funzione a gradini).

### Definizione 2.11 — Quantile

Il **quantile di ordine $p$** (con $p \in (0,1)$) di una v.c. $X$ è il valore $q$ tale che

$$
F_X(q) \geq p.
$$

## 2.4 Momenti

### Definizione 2.12 — Media (valore atteso)

La **media** (o **valore atteso**) di una v.c. $X$ è

$$
\mathbb{E}(X) =
\begin{cases}
\displaystyle \sum_{x \in S} x \, p_X(x), & \text{se } X \text{ è discreta},\\[6pt]
\displaystyle \int_S x \, f_X(x)\, dx, & \text{se } X \text{ è continua}.
\end{cases}
$$

Non tutte le v.c. ammettono momento primo finito.

### Proprietà 2.13 — Trasformazioni

Se $Y = g(X)$, allora

$$
\mathbb{E}(Y) = \mathbb{E}[g(X)] =
\begin{cases}
\displaystyle \sum_{g(x) \in S} g(x) \, p_X(x), & \text{discreta},\\[6pt]
\displaystyle \int_S g(x) \, f_X(x)\, dx, & \text{continua}.
\end{cases}
$$

In particolare:

- Se $Y = aX + b$, allora $\mathbb{E}(Y) = a\mathbb{E}(X) + b$.
- Se $Y = a$ (v.c. degenere), allora $\mathbb{E}(Y) = a$.

### Teorema 2.14 — Disuguaglianza di Jensen

Se $\mathbb{E}(X) < \infty$ e $g(\cdot)$ è una funzione **convessa**, allora

$$
\mathbb{E}(g(X)) \geq g(\mathbb{E}(X)).
$$

Se $g$ è concava, la disuguaglianza si inverte.

### Esempio 2.15

Poiché il logaritmo è una funzione concava,

$$
\mathbb{E}(\log X) \leq \log(\mathbb{E}(X)).
$$

### Definizione 2.16 — Varianza

La **varianza** di una v.c. $X$ è

$$
\mathbb{V}(X) = \mathbb{E}\left[(X - \mu)^2\right] =
\begin{cases}
\displaystyle \sum_{x \in S} (x-\mu)^2 p_X(x), & \text{discreta},\\[6pt]
\displaystyle \int_S (x-\mu)^2 f_X(x)\, dx, & \text{continua},
\end{cases}
$$

dove $\mu = \mathbb{E}(X)$.

### Proprietà 2.17

Valgono:

1. $\mathbb{V}(X) = \mathbb{E}(X^2) - \mathbb{E}(X)^2$;
2. lo **scarto quadratico medio** è $\sigma_X = +\sqrt{\mathbb{V}(X)}$;
3. se $Y = aX + b$, allora $\mathbb{V}(Y) = a^2 \mathbb{V}(X)$;
4. se $Y = a$ (degenere), allora $\mathbb{V}(Y) = 0$.

### Teorema 2.18 — Disuguaglianza di Chebyshev

Per ogni $\varepsilon > 0$,

$$
P(|X - \mu| < \varepsilon) \geq 1 - \frac{\sigma^2}{\varepsilon^2}.
$$

**Dimostrazione.**  
Sia $A = \{\omega : |X(\omega) - \mu| \geq \varepsilon\}$. Allora

$$
\sigma^2 = \mathbb{E}[(X-\mu)^2] \geq \int_A (X-\mu)^2 \, dP \geq \varepsilon^2 P(A),
$$

da cui $P(A) \leq \sigma^2/\varepsilon^2$. Passando al complementare si ottiene la tesi. $\square$

### Definizione 2.19 — Standardizzazione

La **standardizzazione** di una v.c. $X$ con media $\mu$ e varianza $\sigma^2$ è la trasformazione

$$
Z = \frac{X - \mu}{\sigma},
$$

per cui $\mathbb{E}(Z) = 0$ e $\mathbb{V}(Z) = 1$.

## 2.5 Trasformazioni di variabili continue

### Teorema 2.20 — Densità di una trasformazione biunivoca

Se $X$ è una v.c. continua con densità $f_X : S \to \mathbb{R}^+$ e $g : S \to S^*$ è una funzione biunivoca con inversa derivabile, allora $Y = g(X)$ ha densità

$$
f_Y(y) = f_X(g^{-1}(y)) \left| \frac{d g^{-1}(y)}{dy} \right|, \qquad y \in S^*.
$$

**Idea della dimostrazione.**  
Si parte dalla funzione di ripartizione di $Y$:

$$
F_Y(y) = P(Y \leq y) = P(g(X) \leq y) = P(X \leq g^{-1}(y)) = F_X(g^{-1}(y)).
$$

Derivando rispetto a $y$ e applicando la regola della catena si ottiene la tesi. $\square$

# 3. Variabili aleatorie multidimensionali

## 3.1 Variabili casuali doppie

### Definizione 3.1 — Variabile casuale doppia

Una **variabile casuale doppia** è una coppia $(X, Y)$ di variabili aleatorie definite sullo stesso spazio di probabilità.

### Definizione 3.2 — Distribuzione congiunta

La **distribuzione congiunta** di $(X, Y)$ è data da:

- nel caso discreto: $p_{ij} = P(X = x_i \cap Y = y_j)$;
- nel caso continuo: una densità $f_{XY}(x,y)$ tale che
  $$
  P((X,Y) \in A) = \iint_A f_{XY}(x,y)\, dx\, dy.
  $$

### Definizione 3.3 — Distribuzioni marginali

Le **distribuzioni marginali** si ottengono integrando (o sommando) rispetto all'altra variabile:

$$
p_{i\cdot} = P(X = x_i) = \sum_j p_{ij},
\qquad
p_{\cdot j} = P(Y = y_j) = \sum_i p_{ij},
$$

oppure

$$
f_X(x) = \int_y f_{XY}(x,y)\, dy,
\qquad
f_Y(y) = \int_x f_{XY}(x,y)\, dx.
$$

### Definizione 3.4 — Covarianza e correlazione

La **covarianza** tra $X$ e $Y$ è

$$
\operatorname{Cov}(X,Y) = \mathbb{E}[(X-\mu_X)(Y-\mu_Y)] = \mathbb{E}(XY) - \mathbb{E}(X)\mathbb{E}(Y).
$$

Il **coefficiente di correlazione** è

$$
\rho = \operatorname{Cor}(X,Y) = \frac{\operatorname{Cov}(X,Y)}{\sigma_X \sigma_Y}.
$$

### Definizione 3.5 — Distribuzioni condizionate

Le **distribuzioni condizionate** sono:

$$
P(Y = y_j \mid X = x_i) = \frac{p_{ij}}{p_{i\cdot}},
\qquad
P(X = x_i \mid Y = y_j) = \frac{p_{ij}}{p_{\cdot j}},
$$

nel caso discreto, e

$$
f_{Y \mid X}(y \mid x) = \frac{f_{XY}(x,y)}{f_X(x)},
\qquad
f_{X \mid Y}(x \mid y) = \frac{f_{XY}(x,y)}{f_Y(y)},
$$

nel caso continuo.

### Definizione 3.6 — Indipendenza

Due variabili $X$ e $Y$ sono **indipendenti** se

$$
P(X = x_i \cap Y = y_j) = P(X = x_i) P(Y = y_j) \qquad \forall i,j,
$$

oppure, nel caso continuo, se

$$
f_{XY}(x,y) = f_X(x) f_Y(y).
$$

In tal caso:

- $P(Y = y_j \mid X = x_i) = P(Y = y_j)$;
- $\mathbb{E}(XY) = \mathbb{E}(X)\mathbb{E}(Y)$;
- $\operatorname{Cov}(X,Y) = 0$.

> **Osservazione 3.7.**  
> L'indipendenza implica l'incorrelazione, ma non viceversa. Esistono variabili con covarianza nulla che non sono indipendenti.

## 3.2 Trasformazioni lineari

### Proposizione 3.8 — Trasformazioni lineari e covarianza

Date due trasformazioni lineari

$$
X_\alpha = \alpha_0 + \alpha_1 X,
\qquad
Y_\beta = \beta_0 + \beta_1 Y,
$$

si ha

$$
\operatorname{Cov}(X_\alpha, Y_\beta) = \alpha_1 \beta_1 \operatorname{Cov}(X,Y).
$$

In particolare, se $Y = \alpha + \beta X$, allora

$$
\operatorname{Cov}(X,Y) = \beta \operatorname{Var}(X).
$$

### Proposizione 3.9 — Combinazioni lineari

Siano $X$ e $Y$ con medie $\mu_X, \mu_Y$ e varianze $\sigma_X^2, \sigma_Y^2$, e sia $\sigma_{XY} = \operatorname{Cov}(X,Y)$. Allora per $T = \alpha X + \beta Y$:

$$
\mathbb{E}(T) = \alpha \mu_X + \beta \mu_Y,
$$

$$
\mathbb{V}(T) = \alpha^2 \sigma_X^2 + \beta^2 \sigma_Y^2 + 2\alpha\beta \sigma_{XY}.
$$

Se $\operatorname{Cor}(X,Y) = 0$, allora

$$
\mathbb{V}(T) = \alpha^2 \sigma_X^2 + \beta^2 \sigma_Y^2.
$$

### Proposizione 3.10 — Combinazioni lineari di $n$ variabili

Date $Y_1,\dots,Y_n$ e $T = \sum_{i=1}^n \beta_i Y_i$, si ha

$$
\mathbb{E}(T) = \sum_{i=1}^n \beta_i \mathbb{E}(Y_i),
$$

$$
\mathbb{V}(T) = \sum_{i=1}^n \beta_i^2 \mathbb{V}(Y_i) + \sum_{i \neq j} \beta_i \beta_j \operatorname{Cov}(Y_i, Y_j).
$$

Se le $Y_i$ sono incorrelate, il secondo termine svanisce.

## 3.3 Variabili casuali multiple

### Definizione 3.11 — Variabile casuale multipla

Una **variabile casuale multipla** è un vettore

$$
\mathbf{Y} = (Y_1, \dots, Y_n)^T
$$

le cui componenti sono variabili aleatorie. La sua **media** è il vettore

$$
\boldsymbol{\mu} = \mathbb{E}(\mathbf{Y}) = (\mathbb{E}(Y_1), \dots, \mathbb{E}(Y_n))^T,
$$

e la sua **matrice di varianza** (o di dispersione) è

$$
\Sigma = \mathbb{V}(\mathbf{Y}) = \mathbb{E}[(\mathbf{Y} - \boldsymbol{\mu})(\mathbf{Y} - \boldsymbol{\mu})^T] =
\begin{pmatrix}
\sigma_1^2 & \sigma_{12} & \cdots & \sigma_{1n}\\
\sigma_{21} & \sigma_2^2 & \cdots & \sigma_{2n}\\
\vdots & \vdots & \ddots & \vdots\\
\sigma_{n1} & \sigma_{n2} & \cdots & \sigma_n^2
\end{pmatrix}.
$$

### Proprietà 3.12

La matrice $\Sigma$ è:

- **simmetrica**: $\sigma_{ij} = \sigma_{ji}$;
- **semidefinita positiva**: $\mathbf{a}^T \Sigma \mathbf{a} \geq 0$ per ogni $\mathbf{a} \in \mathbb{R}^n$.

La **matrice di correlazione** si ottiene dividendo $\sigma_{ij}$ per $\sigma_i \sigma_j$.

### Proposizione 3.13 — Trasformazioni lineari di vettori casuali

Sia $A$ una matrice $p \times n$ e $\mathbf{b} \in \mathbb{R}^p$. Se $\mathbf{Y}$ è una v.c. multipla con media $\boldsymbol{\mu}$ e matrice di varianza $\Sigma$, allora per $\mathbf{T} = A\mathbf{Y} + \mathbf{b}$:

$$
\mathbb{E}(\mathbf{T}) = A\boldsymbol{\mu} + \mathbf{b},
\qquad
\mathbb{V}(\mathbf{T}) = A \Sigma A^T.
$$

# 4. Successioni di variabili aleatorie

## 4.1 Tipi di convergenza

### Definizione 4.1 — Convergenza in media quadratica

Una successione $\{X_n\}$ converge **in media quadratica** a $X$ se

$$
\lim_{n \to \infty} \mathbb{E}[(X_n - X)^2] = 0.
$$

Si scrive $X_n \xrightarrow{L^2} X$.

### Definizione 4.2 — Convergenza quasi certa

Una successione $\{X_n\}$ converge **quasi certamente** a $X$ se

$$
P\left(\lim_{n \to \infty} X_n = X\right) = 1.
$$

Si scrive $X_n \xrightarrow{q.c.} X$.

### Definizione 4.3 — Convergenza in probabilità

Una successione $\{X_n\}$ converge **in probabilità** a $X$ se per ogni $\varepsilon > 0$,

$$
\lim_{n \to \infty} P(|X_n - X| > \varepsilon) = 0.
$$

Si scrive $X_n \xrightarrow{P} X$.

### Definizione 4.4 — Convergenza in distribuzione

Una successione $\{X_n\}$ converge **in distribuzione** a $X$ se

$$
\lim_{n \to \infty} F_n(t) = F(t)
$$

per ogni $t$ in cui $F$ è continua, dove $F_n$ e $F$ sono le funzioni di ripartizione di $X_n$ e $X$.

Si scrive $X_n \xrightarrow{D} X$.

### Proposizione 4.5 — Implicazioni tra convergenze

Valgono le seguenti implicazioni:

$$
\begin{aligned}
&X_n \xrightarrow{L^2} X \implies X_n \xrightarrow{P} X, \\
&X_n \xrightarrow{q.c.} X \implies X_n \xrightarrow{P} X, \\
&X_n \xrightarrow{P} X \implies X_n \xrightarrow{D} X.
\end{aligned}
$$

In generale, le implicazioni inverse non valgono.

## 4.2 Risultati fondamentali

### Teorema 4.6 — Legge forte dei grandi numeri

Sia $\{X_n\}$ una successione di v.c. indipendenti e identicamente distribuite (i.i.d.) con $\mathbb{E}(X_n) = \mu < \infty$. Allora

$$
\frac{1}{n} \sum_{i=1}^n X_i \xrightarrow{q.c.} \mu.
$$

### Teorema 4.7 — Teorema del limite centrale

Sia $\{X_n\}$ una successione di v.c. i.i.d. con media $\mu$ e varianza $\sigma^2$ finite, e sia

$$
S_n = \sum_{i=1}^n X_i.
$$

Allora la successione delle somme standardizzate

$$
Z_n = \frac{S_n - n\mu}{\sigma \sqrt{n}}
$$

converge in distribuzione alla normale standard:

$$
Z_n \xrightarrow{D} \mathcal{N}(0,1).
$$

### Teorema 4.8 — Risultati di Slutsky

Siano $\{X_n\}$ e $\{Y_n\}$ successioni di v.c. tali che $X_n \xrightarrow{D} X$ e $Y_n \xrightarrow{P} c$ (con $c$ costante deterministica). Allora:

1. Se $g$ è continua in $c$, allora $g(Y_n) \xrightarrow{P} g(c)$;
2. $X_n Y_n \xrightarrow{D} cX$;
3. $X_n + Y_n \xrightarrow{D} X + c$;
4. Se $c \neq 0$, $X_n / Y_n \xrightarrow{D} X/c$.

### Teorema 4.9 — Metodo Delta

Se $\{X_n\}$ è una successione di v.c. tale che

$$
\sqrt{n}(X_n - \theta) \xrightarrow{D} \mathcal{N}(0, \sigma^2)
$$

e $g$ è una funzione reale due volte derivabile con derivate continue e $g'(\theta) \neq 0$, allora

$$
\sqrt{n}(g(X_n) - g(\theta)) \xrightarrow{D} \mathcal{N}(0, \sigma^2 [g'(\theta)]^2).
$$

**Dimostrazione (idea).**  
Si sviluppa $g$ in serie di Taylor del primo ordine in $\theta$:

$$
g(X_n) \approx g(\theta) + g'(\theta)(X_n - \theta).
$$

Moltiplicando per $\sqrt{n}$ e applicando il teorema di Slutsky si ottiene la tesi. $\square$

# 5. Classi notevoli di variabili aleatorie unidimensionali

## 5.1 Distribuzione Binomiale

### Definizione 5.1 — Distribuzione Binomiale

Una v.c. $Y$ ha distribuzione **Binomiale** di parametri $n \in \{1,2,\dots\}$ e $\pi \in (0,1)$ se

$$
Y \sim \operatorname{Bi}(n,\pi),
\qquad
p_Y(y;\pi) = \binom{n}{y} \pi^y (1-\pi)^{n-y}, \quad y = 0,1,\dots,n.
$$

**Condizioni di applicazione:**

- il numero di prove $n$ è fissato a priori;
- le prove sono indipendenti;
- ciascuna prova ha due esiti (successo/insuccesso);
- la probabilità di successo $\pi$ è costante.

**Momenti:**

$$
S = \{0,1,\dots,n\},
\qquad
\mathbb{E}(Y) = n\pi,
\qquad
\mathbb{V}(Y) = n\pi(1-\pi).
$$

### Definizione 5.2 — Distribuzione di Bernoulli

Per $n=1$ si ottiene la distribuzione di **Bernoulli**:

$$
Y \sim \operatorname{Ber}(\pi),
\qquad
p_Y(y;\pi) = \pi^y (1-\pi)^{1-y}, \quad y \in \{0,1\}.
$$

### Proposizione 5.3 — Somma di Bernoulli

Se $Y_1,\dots,Y_n$ sono i.i.d. con $Y_i \sim \operatorname{Ber}(\pi)$, allora

$$
Y = \sum_{i=1}^n Y_i \sim \operatorname{Bi}(n,\pi).
$$

## 5.2 Distribuzione Binomiale Negativa

### Definizione 5.4 — Binomiale Negativa

Sia $Y$ la v.c. che conta il numero di **insuccessi** prima di conseguire $k$ successi in una sequenza di prove indipendenti con probabilità di successo $\pi$. Allora

$$
Y \sim \operatorname{Bineg}(k,\pi),
\qquad
p_Y(y;\pi) = \binom{y+k-1}{y} \pi^k (1-\pi)^y, \quad y \geq 0.
$$

**Momenti:**

$$
S = \{0,1,2,\dots\},
\qquad
\mathbb{E}(Y) = \frac{k(1-\pi)}{\pi},
\qquad
\mathbb{V}(Y) = \frac{k(1-\pi)}{\pi^2}.
$$

### Definizione 5.5 — Distribuzione Geometrica

Per $k=1$ si ottiene la distribuzione **Geometrica**:

$$
Y \sim \operatorname{Ge}(\pi),
\qquad
p_Y(y;\pi) = \pi (1-\pi)^y, \quad y \geq 0.
$$

**Momenti:**

$$
\mathbb{E}(Y) = \frac{1-\pi}{\pi},
\qquad
\mathbb{V}(Y) = \frac{1-\pi}{\pi^2}.
$$

### Proposizione 5.6 — Assenza di memoria

La distribuzione geometrica gode della proprietà di **assenza di memoria**:

$$
P(Y > x + y \mid Y > x) = P(Y > y).
$$

**Dimostrazione.**  
Poiché $P(Y > y) = (1-\pi)^y$, si ha

$$
P(Y > x + y \mid Y > x) = \frac{P(Y > x + y)}{P(Y > x)} = \frac{(1-\pi)^{x+y}}{(1-\pi)^x} = (1-\pi)^y = P(Y > y).
$$

$\square$

## 5.3 Distribuzione di Poisson

### Definizione 5.7 — Distribuzione di Poisson

Una v.c. $Y$ ha distribuzione di **Poisson** di parametro $\lambda > 0$ se

$$
Y \sim \operatorname{Po}(\lambda),
\qquad
p_Y(y;\lambda) = e^{-\lambda} \frac{\lambda^y}{y!}, \quad y \in \{0,1,2,\dots\}.
$$

**Momenti:**

$$
\mathbb{E}(Y) = \lambda,
\qquad
\mathbb{V}(Y) = \lambda.
$$

### Proposizione 5.8 — Proprietà additiva

Se $Y_i \sim \operatorname{Po}(\lambda_i)$ sono indipendenti e i domini sono disgiunti, allora

$$
Y = \sum_{i=1}^n Y_i \sim \operatorname{Po}\left(\sum_{i=1}^n \lambda_i\right).
$$

### Proposizione 5.9 — Limite della Binomiale

Se $Y_n \sim \operatorname{Bi}(n, \pi_n)$ con $n\pi_n \to \lambda$, allora

$$
Y_n \xrightarrow{D} \operatorname{Po}(\lambda).
$$

## 5.4 Distribuzione Uniforme

### Definizione 5.10 — Distribuzione Uniforme

Una v.c. continua $Y$ ha distribuzione **Uniforme** su $[a,b]$ se

$$
Y \sim U(a,b),
\qquad
f_Y(y) = \frac{1}{b-a}, \quad a \leq y \leq b.
$$

**Momenti:**

$$
\mathbb{E}(Y) = \frac{a+b}{2},
\qquad
\mathbb{V}(Y) = \frac{(b-a)^2}{12}.
$$

## 5.5 Distribuzione Normale

### Definizione 5.11 — Distribuzione Normale

Una v.c. continua $Y$ ha distribuzione **Normale** (o Gaussiana) di parametri $\mu \in \mathbb{R}$ e $\sigma^2 > 0$ se

$$
Y \sim \mathcal{N}(\mu, \sigma^2),
\qquad
p_Y(y;\mu,\sigma^2) = \frac{1}{\sqrt{2\pi\sigma^2}} e^{-\frac{1}{2}\left(\frac{y-\mu}{\sigma}\right)^2}.
$$

**Momenti:**

$$
S = \mathbb{R},
\qquad
\mathbb{E}(Y) = \mu,
\qquad
\mathbb{V}(Y) = \sigma^2.
$$

### Proprietà 5.12 — Chiusura per trasformazioni lineari

Se $Y \sim \mathcal{N}(\mu,\sigma^2)$ e $T = a + bY$, allora

$$
T \sim \mathcal{N}(a + b\mu, b^2\sigma^2).
$$

### Definizione 5.13 — Normale standard

Se $Y \sim \mathcal{N}(\mu,\sigma^2)$, allora

$$
Z = \frac{Y - \mu}{\sigma} \sim \mathcal{N}(0,1)
$$

è detta **Normale standard**.

### Proprietà 5.14 — Somma di normali indipendenti

Se $Y_i \sim \mathcal{N}(\mu_i, \sigma_i^2)$ sono indipendenti e $T = \sum_i \beta_i Y_i$, allora

$$
T \sim \mathcal{N}\left(\sum_i \beta_i \mu_i, \sum_i \beta_i^2 \sigma_i^2\right).
$$

### Osservazione 5.15 — Approssimazioni notevoli

Per il teorema del limite centrale:

- se $Y_n \sim \operatorname{Bi}(n,\pi)$, allora $Y_n \xrightarrow{D} \mathcal{N}(n\pi, n\pi(1-\pi))$;
- se $Y_n \sim \operatorname{Po}(n)$, allora $Y_n \xrightarrow{D} \mathcal{N}(n,n)$.

## 5.6 Distribuzione Gamma

### Definizione 5.16 — Distribuzione Gamma

Una v.c. continua $Y$ ha distribuzione **Gamma** di parametri $\alpha > 0$ (forma) e $\lambda > 0$ (scala) se

$$
Y \sim \operatorname{Ga}(\alpha,\lambda),
\qquad
p_Y(y;\alpha,\lambda) = \frac{\lambda^\alpha}{\Gamma(\alpha)} y^{\alpha-1} e^{-\lambda y}, \quad y > 0,
$$

dove $\Gamma$ è la funzione Gamma:

$$
\Gamma(\alpha) = \int_0^{+\infty} t^{\alpha-1} e^{-t}\, dt.
$$

**Momenti:**

$$
S = \mathbb{R}^+,
\qquad
\mathbb{E}(Y) = \frac{\alpha}{\lambda},
\qquad
\mathbb{V}(Y) = \frac{\alpha}{\lambda^2}.
$$

### Proprietà 5.17

Valgono:

1. **Scala:** se $Y \sim \operatorname{Ga}(\alpha,\lambda)$, allora $kY \sim \operatorname{Ga}(\alpha,\lambda/k)$.
2. **Additività:** se $Y_1 \sim \operatorname{Ga}(\alpha_1,\lambda)$ e $Y_2 \sim \operatorname{Ga}(\alpha_2,\lambda)$ sono indipendenti, allora
   $$
   Y_1 + Y_2 \sim \operatorname{Ga}(\alpha_1 + \alpha_2, \lambda).
   $$
3. Per $\alpha$ grande, $\operatorname{Ga}(\alpha,\lambda) \approx \mathcal{N}(\alpha/\lambda, \alpha/\lambda^2)$.

## 5.7 Distribuzione Esponenziale

### Definizione 5.18 — Distribuzione Esponenziale

Una v.c. continua $Y$ ha distribuzione **Esponenziale** di parametro $\lambda > 0$ se

$$
Y \sim \operatorname{Esp}(\lambda),
\qquad
p_Y(y;\lambda) = \lambda e^{-\lambda y}, \quad y > 0.
$$

**Momenti:**

$$
\mathbb{E}(Y) = \frac{1}{\lambda},
\qquad
\mathbb{V}(Y) = \frac{1}{\lambda^2}.
$$

### Osservazione 5.19

L'esponenziale è un caso particolare della Gamma:

$$
\operatorname{Esp}(\lambda) = \operatorname{Ga}(1,\lambda).
$$

### Proprietà 5.20 — Assenza di memoria

$$
P(Y > s + t \mid Y > s) = P(Y > t).
$$

### Proposizione 5.21 — Relazione con Poisson

Sia $Y \sim \operatorname{Po}(\lambda)$ il numero di arrivi per unità di tempo. Il tempo di attesa prima del primo arrivo è una v.c. esponenziale $X \sim \operatorname{Esp}(\lambda)$. In generale, il tempo di attesa prima dell'$n$-esimo arrivo ha distribuzione $\operatorname{Ga}(n,\lambda)$.

## 5.8 Distribuzione Chi-quadro

### Definizione 5.22 — Chi-quadro

Se $Z \sim \mathcal{N}(0,1)$, la v.c. $T = Z^2$ ha distribuzione **Chi-quadro** con 1 grado di libertà:

$$
T \sim \chi^2_1,
\qquad
T \sim \operatorname{Ga}\left(\tfrac{1}{2}, \tfrac{1}{2}\right).
$$

### Definizione 5.23 — Chi-quadro con $\nu$ gradi di libertà

Se $Y_1,\dots,Y_\nu$ sono i.i.d. $\mathcal{N}(0,1)$, allora

$$
U = \sum_{i=1}^\nu Y_i^2 \sim \chi^2_\nu,
\qquad
U \sim \operatorname{Ga}\left(\tfrac{\nu}{2}, \tfrac{1}{2}\right).
$$

## 5.9 Distribuzione $t$ di Student

### Definizione 5.24 — $t$ di Student

Siano $Z \sim \mathcal{N}(0,1)$ e $U \sim \chi^2_\nu$ indipendenti. La v.c.

$$
T = \frac{Z}{\sqrt{U/\nu}} \sim t_\nu
$$

ha distribuzione $t$ di Student con $\nu$ gradi di libertà.

**Momenti:**

$$
S = \mathbb{R},
\qquad
\mathbb{E}(T) = 0 \ (\nu > 1),
\qquad
\mathbb{V}(T) = \frac{\nu}{\nu-2} \ (\nu > 2).
$$

> **Osservazione 5.25.**  
> La distribuzione $t$ è simmetrica, unimodale, con code più pesanti della Normale. Per $\nu \to \infty$ tende alla $\mathcal{N}(0,1)$.

## 5.10 Distribuzione $F$ di Fisher

### Definizione 5.26 — $F$ di Fisher

Siano $U \sim \chi^2_\nu$ e $V \sim \chi^2_\zeta$ indipendenti. La v.c.

$$
F = \frac{U/\nu}{V/\zeta} \sim F_{\nu,\zeta}
$$

ha distribuzione $F$ di Fisher con $\nu, \zeta$ gradi di libertà.

**Relazioni notevoli:**

$$
\frac{1}{F_{\nu,\zeta}} \sim F_{\zeta,\nu},
\qquad
T^2 \sim F_{1,\nu}
\quad \text{se } T \sim t_\nu.
$$

# 6. Classi notevoli di variabili aleatorie multidimensionali

## 6.1 Variabile casuale multinomiale

### Definizione 6.1 — Multinomiale

Si consideri un esperimento con $K$ possibili esiti $E_1,\dots,E_K$ con probabilità $\pi_1,\dots,\pi_K$ (con $\sum_j \pi_j = 1$). Ripetendo l'esperimento $n$ volte in modo indipendente, il vettore

$$
\mathbf{Y} = (Y_1,\dots,Y_K)^T,
\qquad
Y_j = \text{numero di volte in cui si verifica } E_j,
$$

ha distribuzione **Multinomiale** di dimensione $K$, indice $n$ e parametri $\boldsymbol{\pi}$:

$$
\mathbf{Y} \sim \operatorname{Bi}_K(n, \boldsymbol{\pi}),
\qquad
P(\mathbf{Y} = \mathbf{y}) = \binom{n}{\mathbf{y}} \prod_{j=1}^K \pi_j^{y_j},
$$

dove $\sum_j y_j = n$ e

$$
\binom{n}{\mathbf{y}} = \frac{n!}{y_1! \cdots y_K!}.
$$

### Osservazione 6.2

Per $K=2$ la multinomiale si riduce alla binomiale.

### Proprietà 6.3

Valgono:

- ogni componente $Y_j \sim \operatorname{Bi}(n,\pi_j)$;
- somme di multinomiali con lo stesso $\boldsymbol{\pi}$ sono ancora multinomiali;
- condizionando una multinomiale alle restanti componenti si ottiene ancora una multinomiale.

### Momenti 6.4

Se $\mathbf{Y} \sim \operatorname{Bi}_K(n,\boldsymbol{\pi})$, allora

$$
\mathbb{E}(\mathbf{Y}) = (n\pi_1,\dots,n\pi_K)^T,
$$

$$
\mathbb{V}(\mathbf{Y}) =
\begin{pmatrix}
n\pi_1(1-\pi_1) & -n\pi_1\pi_2 & \cdots & -n\pi_1\pi_K\\
-n\pi_2\pi_1 & n\pi_2(1-\pi_2) & \cdots & -n\pi_2\pi_K\\
\vdots & \vdots & \ddots & \vdots\\
-n\pi_K\pi_1 & -n\pi_K\pi_2 & \cdots & n\pi_K(1-\pi_K)
\end{pmatrix}.
$$

## 6.2 Variabile casuale normale multipla

### Definizione 6.5 — Normale multipla standard

Siano $Z_1,\dots,Z_p$ i.i.d. $\mathcal{N}(0,1)$. Il vettore $\mathbf{Z} = (Z_1,\dots,Z_p)^T$ ha densità

$$
f_{\mathbf{Z}}(\mathbf{z}) = \frac{1}{(2\pi)^{p/2}} e^{-\frac{1}{2}\mathbf{z}^T \mathbf{z}},
\qquad
\mathbf{Z} \sim \mathcal{N}_p(\mathbf{0}, I_p).
$$

### Definizione 6.6 — Normale multipla

Sia $\mathbf{Z} \sim \mathcal{N}_p(\mathbf{0}, I_p)$, $A$ una matrice $p \times p$ invertibile e $\boldsymbol{\mu} \in \mathbb{R}^p$. Il vettore

$$
\mathbf{Y} = A\mathbf{Z} + \boldsymbol{\mu}
$$

ha distribuzione **Normale multipla** con media $\boldsymbol{\mu}$ e matrice di varianza $\Sigma = AA^T$:

$$
\mathbf{Y} \sim \mathcal{N}_p(\boldsymbol{\mu}, \Sigma),
$$

con densità

$$
f_{\mathbf{Y}}(\mathbf{y}) = \frac{1}{(2\pi)^{p/2} |\Sigma|^{1/2}} \exp\left[-\frac{1}{2}(\mathbf{y}-\boldsymbol{\mu})^T \Sigma^{-1} (\mathbf{y}-\boldsymbol{\mu})\right].
$$

### Proprietà 6.7

Valgono:

1. $\mathbb{E}(\mathbf{Y}) = \boldsymbol{\mu}$ e $\mathbb{V}(\mathbf{Y}) = \Sigma$;
2. se $\mathbf{X} = B\mathbf{Y} + \mathbf{b}$, allora $\mathbf{X} \sim \mathcal{N}_p(B\boldsymbol{\mu} + \mathbf{b}, B\Sigma B^T)$;
3. tutte le distribuzioni marginali e condizionate sono ancora normali multiple.

### Definizione 6.8 — Distanza di Mahalanobis

La **distanza di Mahalanobis** tra due vettori $\mathbf{x}, \mathbf{y} \in \mathbb{R}^p$ è

$$
d_M(\mathbf{x},\mathbf{y})^2 = (\mathbf{x}-\mathbf{y})^T \Sigma^{-1} (\mathbf{x}-\mathbf{y}).
$$

Essa gode delle proprietà di una metrica:

1. $d_M(\mathbf{x},\mathbf{y}) \geq 0$;
2. $d_M(\mathbf{x},\mathbf{y}) = 0 \iff \mathbf{x} = \mathbf{y}$;
3. $d_M(\mathbf{x},\mathbf{y}) = d_M(\mathbf{y},\mathbf{x})$;
4. $d_M(\mathbf{x},\mathbf{y}) \leq d_M(\mathbf{x},\mathbf{z}) + d_M(\mathbf{z},\mathbf{y})$.

### Osservazione 6.9

Se $\Sigma = I_p$, la distanza di Mahalanobis si riduce alla distanza euclidea.

### Teorema 6.10 — Forma quadratica e Chi-quadro

Se $\mathbf{Y} \sim \mathcal{N}_p(\boldsymbol{\mu}, \Sigma)$, allora

$$
(\mathbf{Y} - \boldsymbol{\mu})^T \Sigma^{-1} (\mathbf{Y} - \boldsymbol{\mu}) \sim \chi^2_p.
$$

### Caso particolare: Normale doppia

Se $\mathbf{Y} = (Y_1,Y_2)^T \sim \mathcal{N}_2(\boldsymbol{\mu},\Sigma)$ con

$$
\boldsymbol{\mu} = \begin{pmatrix}\mu_1\\\mu_2\end{pmatrix},
\qquad
\Sigma = \begin{pmatrix}\sigma_1^2 & \rho\sigma_1\sigma_2\\ \rho\sigma_1\sigma_2 & \sigma_2^2\end{pmatrix},
$$

allora la densità congiunta è

$$
f_{Y_1,Y_2}(y_1,y_2) = \frac{1}{2\pi\sigma_1\sigma_2\sqrt{1-\rho^2}} \exp\left\{-\frac{1}{2(1-\rho^2)}\left[\left(\frac{y_1-\mu_1}{\sigma_1}\right)^2 - 2\rho\frac{(y_1-\mu_1)(y_2-\mu_2)}{\sigma_1\sigma_2} + \left(\frac{y_2-\mu_2}{\sigma_2}\right)^2\right]\right\}.
$$

> **Osservazione.** Se $\rho = 0$, allora $Y_1$ e $Y_2$ sono indipendenti (caso eccezionale: per la normale, incorrelazione implica indipendenza).

### Teorema 6.11 — Teorema del limite centrale multivariato

Sia $\{\mathbf{Y}^{(n)}\}$ una successione di vettori casuali $p$-dimensionali i.i.d. con media $\boldsymbol{\mu}$ e matrice di varianza $\Sigma$ definita positiva. Allora

$$
\sqrt{n}(\overline{\mathbf{Y}}^{(n)} - \boldsymbol{\mu}) \xrightarrow{D} \mathcal{N}_p(\mathbf{0}, \Sigma),
$$

dove

$$
\overline{\mathbf{Y}}^{(n)} = \frac{1}{n} \sum_{i=1}^n \mathbf{Y}^{(i)}.
$$
