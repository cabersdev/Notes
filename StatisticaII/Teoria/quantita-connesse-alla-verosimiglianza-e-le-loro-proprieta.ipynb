# Quantità connesse alla verosimiglianza e loro proprietà

---

## 0. Impostazione

### Osservazione 0.1

Sia $\mathbf{Y}=(Y_1,\dots,Y_n)$ un campione i.i.d. con densità (o probabilità) $p_{\mathbf{Y}}(\mathbf{y};\theta)$, $\theta\in\Theta\subseteq\mathbb{R}^p$. La **log-verosimiglianza** è
$$
\ell(\theta)=\ell(\theta;\mathbf{y})=\sum_{i=1}^n\log p_{Y_i}(y_i;\theta),
$$
e la **SMV** è
$$
\hat\theta=\arg\max_{\theta\in\Theta}\ell(\theta).
$$
La log-verosimiglianza contiene tutta l'informazione su $\theta$ contenuta nei dati, ma i suoi **valori** possono essere estremamente piccoli (soprattutto per modelli continui) e il suo **andamento** vicino a $\hat\theta$ contiene l'informazione sulla precisione della stima. Le quantità che introduciamo formalizzano questo secondo aspetto.

### Esempio 0.2 — Perché la verosimiglianza non normalizzata è problematica

Sia $\mathbf{y}=(225, 171, 198, 189, 189, 135, 162, 135, 117, 162)$ un campione da una $\operatorname{Cauchy}(\theta)$, con densità
$$
p_Y(y;\theta)=\frac{1}{\pi(1+(y-\theta)^2)}.
$$
La verosimiglianza è
$$
L(\theta)=\prod_{i=1}^{10}\frac{1}{\pi(1+(y_i-\theta)^2)}.
$$
I valori sono dell'ordine di $10^{-24}$ e variano di molti ordini di grandezza al variare di $\theta$. Anche graficamente è difficile distinguere i massimi. E' necessario quindi **normalizzare** la funzione di verosimiglianza

---

## 1. Verosimiglianza normalizzata

### Definizione 1.1 — Verosimiglianza normalizzata

Sia $L(\theta)$ la verosimiglianza e $\hat\theta$ la SMV di $\theta$. Si definisce **verosimiglianza normalizzata** (o **verosimiglianza relativa**) la funzione
$$
L_N(\theta)=\frac{L(\theta)}{L(\hat\theta)},
\qquad\theta\in\Theta.
$$

### Proposizione 1.2 — Proprietà

Valgono:

1. $0\le L_N(\theta)\le 1$ per ogni $\theta\in\Theta$;
2. $L_N(\hat\theta)=1$;
3. $L_N$ è equivalente a $L$ ai fini dell'inferenza, poiché differisce da $L$ per la costante positiva $1/L(\hat\theta)$ che non dipende da $\theta$;
4. la log-verosimiglianza normalizzata è
$$
\ell_N(\theta)=\log L_N(\theta)=\ell(\theta)-\ell(\hat\theta)\le 0.
$$

**Dimostrazione.**  
Poiché $\hat\theta$ massimizza $L$, si ha $L(\theta)\le L(\hat\theta)$ per ogni $\theta$, quindi $L_N(\theta)\le 1$; la non-negatività segue da $L(\theta)\ge 0$. Inoltre $L_N(\hat\theta)=L(\hat\theta)/L(\hat\theta)=1$. La costante $1/L(\hat\theta)$ non dipende da $\theta$, quindi $L_N$ è una verosimiglianza equivalente. La formula per $\ell_N$ segue dal fatto che $\log$ è strettamente crescente. $\square$

> **Osservazione 1.3.**  
> La normalizzazione risolve **due problemi**:
> - rende i valori confrontabili (sempre in $[0,1]$), eliminando il divario di ordini di grandezza;
> - rende la **forma** della verosimiglianza indipendente dalla scala dei dati, perché il fattore $L(\hat\theta)$ "assorbe" la costante di normalizzazione della densità.
>
> Inoltre, $L_N(\theta)$ è una **verosimiglianza equivalente**: per il principio debole di verosimiglianza, due campioni con $L_N$ proporzionali conducono alle stesse conclusioni inferenziali.

### Esempio 1.4 — Cauchy

Per il campione dell'Esempio 0.2, la verosimiglianza normalizzata è
$$
L_N(\theta)=\prod_{i=1}^{10}\frac{1+(y_i-\hat\theta)^2}{1+(y_i-\theta)^2}.
$$
I valori sono ora nell'intervallo $[0,1]$, con massimo $1$ in $\hat\theta$. La struttura dei massimi locali, invisibile nella $L$ originaria, diventa leggibile.

## 2. Informazione osservata

### Definizione 2.1 — Informazione osservata (di Fisher)

Sia $\ell(\theta)$ la log-verosimiglianza, $\theta\in\Theta\subseteq\mathbb{R}^p$. L'**informazione osservata** è
$$
J(\theta)=
\begin{cases}
-\ell''(\theta)=-\dfrac{d^2\ell(\theta)}{d\theta^2}, & p=1,\\[2ex]
-\nabla^2_\theta\ell(\theta)=-\left(\dfrac{\partial^2\ell(\theta)}{\partial\theta_i\partial\theta_j}\right)_{i,j=1,\dots,p}, & p>1.
\end{cases}
$$
Per il teorema di Schwarz, $J(\theta)$ è **simmetrica**.

### Osservazione 2.2 — Interpretazione

$J(\theta)$ misura **quanto bene si riesce a discriminare** il punto di massimo dai valori vicini. Se $J(\hat\theta)$ è grande, la log-verosimiglianza ha curvatura pronunciata in $\hat\theta$ (la "collina" è ripida); se è piccola, la log-verosimiglianza è piatta e la stima è incerta.

### Proposizione 2.3 — Relazione con la derivata della verosimiglianza

Se $L>0$ e $\ell$ è derivabile due volte,
$$
\ell''(\theta)=\frac{L''(\theta)}{L(\theta)}-\left(\frac{L'(\theta)}{L(\theta)}\right)^2.
$$
In particolare, in un punto stazionario $\ell'(\theta_0)=0$, si ha
$$
J(\theta_0)=-\ell''(\theta_0)=-\frac{L''(\theta_0)}{L(\theta_0)}.
$$

**Dimostrazione.**  
Derivando $\ell=\log L$ due volte: $\ell'=L'/L$, $\ell''=L''/L-(L'/L)^2$. In $\theta_0$ con $\ell'(\theta_0)=0$, il secondo termine svanisce. $\square$

> **Osservazione 2.4.**  
> Nei punti stazionari, $J(\theta_0)>0$ se e solo se $L''(\theta_0)<0$, cioè se $\theta_0$ è un massimo locale. Quindi $J$ è il criterio naturale per verificare la concavità locale e classificare i punti stazionari.

## 3. Approssimazione parabolica di $L_N$

### Teorema 3.1 — Approssimazione parabolica

Sia $\ell$ due volte differenziabile in un intorno di $\hat\theta$. Allora, per $\theta$ vicino a $\hat\theta$,
$$
\ell(\theta)-\ell(\hat\theta)\approx-\frac12(\theta-\hat\theta)^2\,J(\hat\theta).
$$
Equivalentemente,
$$
\log L_N(\theta)\approx-\frac12(\theta-\hat\theta)^2\,J(\hat\theta).
$$
Nel caso $p>1$,
$$
\ell(\theta)-\ell(\hat\theta)\approx-\frac12(\theta-\hat\theta)^T J(\hat\theta)(\theta-\hat\theta).
$$

**Dimostrazione.**  
Si sviluppa $\ell$ in serie di Taylor al secondo ordine in $\hat\theta$:
$$
\ell(\theta)=\ell(\hat\theta)+(\theta-\hat\theta)\ell'(\hat\theta)+\frac12(\theta-\hat\theta)^2\ell''(\hat\theta)+o((\theta-\hat\theta)^2).
$$
Poiché $\hat\theta$ è punto stazionario, $\ell'(\hat\theta)=0$. Sostituendo $\ell''(\hat\theta)=-J(\hat\theta)$ e trascurando il resto:
$$
\ell(\theta)-\ell(\hat\theta)\approx-\frac12(\theta-\hat\theta)^2 J(\hat\theta).
$$
Nel caso $p>1$, la formula analoga con gradiente e Hessiana dà la tesi. $\square$

### Osservazione 3.2 — Interpretazione geometrica

La funzione
$$
\theta\mapsto-\frac12(\theta-\hat\theta)^2 J(\hat\theta)
$$
è una **parabola** con vertice in $\hat\theta$, concavità rivolta verso il basso (perché $J(\hat\theta)>0$) e curvatura proporzionale a $J(\hat\theta)$. Quindi:

- **$J(\hat\theta)$ grande**: parabola stretta e alta, la verosimiglianza cala rapidamente allontanandosi da $\hat\theta$;
- **$J(\hat\theta)$ piccola**: parabola larga e piatta, la verosimiglianza cala lentamente.

In altre parole, **più informazione osservata significa più precisione nella stima**.

### Esempio 3.3 — Bernoulli

Sia $Y_1,\dots,Y_n$ i.i.d. $\operatorname{Be}(\pi)$. La verosimiglianza è
$$
L(\pi)=\pi^{\sum_i y_i}(1-\pi)^{n-\sum_i y_i}.
$$
La log-verosimiglianza è
$$
\ell(\pi)=\left(\sum_i y_i\right)\log\pi+\left(n-\sum_i y_i\right)\log(1-\pi).
$$
**SMV.** Derivando:
$$
\ell'(\pi)=\frac{\sum_i y_i}{\pi}-\frac{n-\sum_i y_i}{1-\pi}.
$$
Ponendo $\ell'(\pi)=0$:
$$
\hat\pi=\frac{\sum_i y_i}{n}=\bar y.
$$
**Informazione osservata.** Derivando ancora:
$$
\ell''(\pi)=-\frac{\sum_i y_i}{\pi^2}-\frac{n-\sum_i y_i}{(1-\pi)^2},
$$
quindi
$$
J(\pi)=\frac{\sum_i y_i}{\pi^2}+\frac{n-\sum_i y_i}{(1-\pi)^2}.
$$
In $\hat\pi$:
$$
J(\hat\pi)=\frac{n\hat\pi}{\hat\pi^2}+\frac{n(1-\hat\pi)}{(1-\hat\pi)^2}=\frac{n}{\hat\pi}+\frac{n}{1-\hat\pi}=\frac{n}{\hat\pi(1-\hat\pi)}.
$$
**Approssimazione parabolica.** Per $\pi$ vicino a $\hat\pi$,
$$
\ell(\pi)-\ell(\hat\pi)\approx-\frac12(\pi-\hat\pi)^2\frac{n}{\hat\pi(1-\hat\pi)}.
$$
Quindi la log-verosimiglianza normalizzata è approssimata da una parabola con curvatura $n/[\hat\pi(1-\hat\pi)]$: **cresce linearmente in $n$**. Più il campione è grande, più la verosimiglianza è concentrata intorno alla SMV. Questo è il fondamento intuitivo della **consistenza** della SMV.

## 4. Informazione attesa

### Definizione 4.1 — Informazione attesa (di Fisher)

Poiché $J(\theta)=J(\theta;\mathbf{Y})$ dipende dal campione, è una **variabile casuale**. Si definisce **informazione attesa** (o **informazione di Fisher**) la quantità
$$
I(\theta)=I(\theta;\mathbf{Y})=E_\theta[J(\theta;\mathbf{Y})]=E_\theta[-\ell''(\theta;\mathbf{Y})].
$$
Nel caso $p>1$,
$$
I(\theta)=E_\theta[-\nabla^2_\theta\ell(\theta;\mathbf{Y})]=-E_\theta[\nabla^2_\theta\ell(\theta;\mathbf{Y})].
$$

### Proposizione 4.2 — Additività per campioni i.i.d.

Sia $\mathbf{Y}=(Y_1,\dots,Y_n)$ con $Y_i$ i.i.d. di densità $p_Y(y;\theta)$. Allora
$$
I(\theta)=n\,i(\theta),
$$
dove $i(\theta)=E_\theta[-\ell_1''(\theta;Y_1)]$ è l'informazione attesa di **una singola osservazione**.

**Dimostrazione.**  
Per l'indipendenza, la log-verosimiglianza è
$$
\ell(\theta;\mathbf{y})=\sum_{i=1}^n\ell_i(\theta;y_i),
\qquad
\ell_i(\theta;y_i)=\log p_Y(y_i;\theta).
$$
Derivando due volte e prendendo il valore atteso:
$$
I(\theta)=E_\theta\left[-\sum_{i=1}^n\ell_i''(\theta;Y_i)\right]=\sum_{i=1}^n E_\theta[-\ell_i''(\theta;Y_i)]=n\,i(\theta).
$$
$\square$

> **Osservazione 4.3.**  
> L'informazione attesa è **additiva**: ogni osservazione aggiunge un contributo $i(\theta)$ all'informazione totale. Questo è uno dei motivi per cui la SMV è efficiente: la precisione cresce come $\sqrt n$. Il legame quantitativo è dato dalla disuguaglianza di Cramér–Rao: $\operatorname{Var}(\hat\theta)\ge [n\,i(\theta)]^{-1}$ per ogni stimatore non distorto, e la SMV raggiunge asintoticamente questo bound.

### Esempio 4.4 — Bernoulli

Con $Y\sim\operatorname{Be}(\pi)$, la log-verosimiglianza individuale è
$$
\ell_1(\pi;y)=y\log\pi+(1-y)\log(1-\pi).
$$
La derivata seconda è
$$
\ell_1''(\pi;y)=-\frac{y}{\pi^2}-\frac{1-y}{(1-\pi)^2}.
$$
Prendendo il valore atteso (con $E[Y]=\pi$):
$$
i(\pi)=E\left[\frac{Y}{\pi^2}+\frac{1-Y}{(1-\pi)^2}\right]=\frac{\pi}{\pi^2}+\frac{1-\pi}{(1-\pi)^2}=\frac1\pi+\frac1{1-\pi}=\frac{1}{\pi(1-\pi)}.
$$
Quindi
$$
I(\pi)=n\,i(\pi)=\frac{n}{\pi(1-\pi)}.
$$
Coincide con $J(\hat\pi)$ nel caso bernoulliano, perché $J(\pi)$ non dipende da $y$ (la derivata seconda è deterministica). In generale, $I(\theta)\ne J(\theta;\mathbf{y})$: l'informazione attesa è la media di $J$, e solo per alcune famiglie (come la Bernoulli) $J$ è costante rispetto al campione.

## 5. Disuguaglianza di Wald

### Teorema 5.1 — Disuguaglianza di Wald

Sia $\mathbf{Y}$ un campione da $p_{\mathbf{Y}}(\cdot;\theta_0)$, con $\theta_0\in\Theta$. Allora, per ogni $\theta\in\Theta$,
$$
E_{\theta_0}[\ell(\theta_0;\mathbf{Y})]\ge E_{\theta_0}[\ell(\theta;\mathbf{Y})],
$$
con uguaglianza se e solo se $p_{\mathbf{Y}}(\cdot;\theta)=p_{\mathbf{Y}}(\cdot;\theta_0)$ q.o.

In altre parole, **la log-verosimiglianza assume in media il valore massimo nel vero valore del parametro**.

**Dimostrazione.**  
Sia $\mathbf{Y}$ continuo (il caso discreto è analogo, con somme al posto di integrali). Consideriamo il rapporto
$$
\frac{p_{\mathbf{Y}}(\mathbf{y};\theta)}{p_{\mathbf{Y}}(\mathbf{y};\theta_0)}.
$$
Il suo valore atteso sotto $\theta_0$ è
$$
E_{\theta_0}\left[\frac{p_{\mathbf{Y}}(\mathbf{Y};\theta)}{p_{\mathbf{Y}}(\mathbf{Y};\theta_0)}\right]=\int_{\mathcal{Y}}\frac{p_{\mathbf{Y}}(\mathbf{y};\theta)}{p_{\mathbf{Y}}(\mathbf{y};\theta_0)}\,p_{\mathbf{Y}}(\mathbf{y};\theta_0)\,d\mathbf{y}=\int_{\mathcal{Y}}p_{\mathbf{Y}}(\mathbf{y};\theta)\,d\mathbf{y}=1.
$$
Applicando la disuguaglianza di Jensen alla funzione concava $\log$:
$$
\log E_{\theta_0}\left[\frac{p_{\mathbf{Y}}(\mathbf{Y};\theta)}{p_{\mathbf{Y}}(\mathbf{Y};\theta_0)}\right]\ge E_{\theta_0}\left[\log\frac{p_{\mathbf{Y}}(\mathbf{Y};\theta)}{p_{\mathbf{Y}}(\mathbf{Y};\theta_0)}\right].
$$
Il membro di sinistra è $\log 1=0$, quindi
$$
0\ge E_{\theta_0}\left[\log p_{\mathbf{Y}}(\mathbf{Y};\theta)-\log p_{\mathbf{Y}}(\mathbf{Y};\theta_0)\right],
$$
cioè
$$
E_{\theta_0}[\ell(\theta;\mathbf{Y})]\le E_{\theta_0}[\ell(\theta_0;\mathbf{Y})].
$$
L'uguaglianza vale se e solo se $\log(p_\theta/p_{\theta_0})$ è costante q.o., cioè se $p_\theta=p_{\theta_0}$ q.o. $\square$

> **Osservazione 5.2.**  
> La disuguaglianza di Wald è un'affermazione **frequentista**: mediando su tutti i campioni possibili generati dal vero $\theta_0$, la log-verosimiglianza è massima in $\theta_0$. Questo è il fondamento teorico della **consistenza** della SMV: se $\hat\theta$ massimizza la log-verosimiglianza, e in media la log-verosimiglianza è massima in $\theta_0$, allora $\hat\theta$ deve convergere a $\theta_0$ (sotto opportune condizioni di uniformità e regolarità). La dimostrazione rigorosa richiede la legge dei grandi numeri uniforme e la teoria delle stime $M$, ma l'idea è contenuta nella disuguaglianza.

> **Osservazione 5.3.**  
> La quantità $E_{\theta_0}[\ell(\theta_0)-\ell(\theta)]$ è legata alla **divergenza di Kullback–Leibler**:
> $$
> D_{\mathrm{KL}}(p_{\theta_0}\|p_\theta)=E_{\theta_0}\left[\log\frac{p_{\theta_0}}{p_\theta}\right]\ge 0.
> $$
> La disuguaglianza di Wald afferma che $D_{\mathrm{KL}}(p_{\theta_0}\|p_\theta)\ge 0$ con uguaglianza se e solo se $\theta=\theta_0$. È lo stesso principio, riscritto in termini di verosimiglianza.

## 6. Identità di Bartlett

### Teorema 6.1 — Identità di Bartlett

Sotto le condizioni di regolarità del problema di stima, per ogni $\theta\in\Theta$,
$$
E_\theta[\ell'(\theta;\mathbf{Y})]=0,
$$
e
$$
\operatorname{Var}_\theta(\ell'(\theta;\mathbf{Y}))=E_\theta[(\ell'(\theta;\mathbf{Y}))^2]=E_\theta[-\ell''(\theta;\mathbf{Y})]=I(\theta).
$$

Nel caso $p>1$, le identità diventano
$$
E_\theta[\nabla_\theta\ell(\theta;\mathbf{Y})]=\mathbf{0},
\qquad
\operatorname{Var}_\theta(\nabla_\theta\ell(\theta;\mathbf{Y}))=E_\theta[\nabla_\theta\ell(\theta;\mathbf{Y})\nabla_\theta\ell(\theta;\mathbf{Y})^T]=I(\theta).
$$

### Dimostrazione (caso $p=1$)

**Parte 1: valore atteso nullo.**  
Per ogni $\theta\in\Theta$, la densità integra a $1$:
$$
\int_{\mathcal{Y}}p_Y(y;\theta)\,dy=1.
$$
Derivando rispetto a $\theta$ (lo scambio è lecito per le condizioni di regolarità):
$$
0=\frac{d}{d\theta}\int_{\mathcal{Y}}p_Y(y;\theta)\,dy=\int_{\mathcal{Y}}\frac{\partial}{\partial\theta}p_Y(y;\theta)\,dy.
$$
Riscrivendo $\frac{\partial}{\partial\theta}p_Y=\frac{\partial}{\partial\theta}\log p_Y\cdot p_Y$:
$$
0=\int_{\mathcal{Y}}\left(\frac{\partial}{\partial\theta}\log p_Y(y;\theta)\right)p_Y(y;\theta)\,dy=E_\theta[\ell'(\theta;\mathbf{Y})].
$$
Quindi $E_\theta[\ell'(\theta;\mathbf{Y})]=0$.

**Parte 2: varianza uguale all'informazione attesa.**  
Derivando ancora rispetto a $\theta$ l'identità $\int_{\mathcal{Y}}\ell'(y;\theta)p_Y(y;\theta)\,dy=0$:
$$
0=\frac{d}{d\theta}\int_{\mathcal{Y}}\ell'(y;\theta)p_Y(y;\theta)\,dy
=\int_{\mathcal{Y}}\left[\ell''(y;\theta)p_Y(y;\theta)+\ell'(y;\theta)\frac{\partial}{\partial\theta}p_Y(y;\theta)\right]dy.
$$
Sostituendo $\frac{\partial}{\partial\theta}p_Y=\ell'(y;\theta)p_Y(y;\theta)$:
$$
0=E_\theta[\ell''(\theta;\mathbf{Y})]+\int_{\mathcal{Y}}(\ell'(y;\theta))^2 p_Y(y;\theta)\,dy
=E_\theta[\ell''(\theta;\mathbf{Y})]+E_\theta[(\ell'(\theta;\mathbf{Y}))^2].
$$
Quindi
$$
E_\theta[(\ell'(\theta;\mathbf{Y}))^2]=-E_\theta[\ell''(\theta;\mathbf{Y})]=I(\theta).
$$
Poiché $E_\theta[\ell']=0$, si ha anche $\operatorname{Var}_\theta(\ell')=E_\theta[(\ell')^2]$, da cui la tesi. $\square$

> **Osservazione 6.2.**  
> Le identità di Bartlett hanno un significato profondo: la **funzione punteggio** $\ell'(\theta;\mathbf{Y})$ ha media nulla e varianza uguale all'informazione attesa. In altri termini, la variabilità del punteggio intorno allo zero è **esattamente** la curvatura attesa della log-verosimiglianza. Questo è il ponte tra la variabilità campionaria del punteggio e la precisione della stima.

### Corollario 6.3 — Interpretazione della disuguaglianza di Cramér–Rao

Combinando le identità di Bartlett con la disuguaglianza di Cauchy–Schwarz, si ottiene la **disuguaglianza di Cramér–Rao**: per ogni stimatore $T$ non distorto di $\theta$,
$$
\operatorname{Var}_\theta(T)\ge\frac{1}{I(\theta)}.
$$
La SMV raggiunge asintoticamente questo bound. Le identità di Bartlett sono il primo passo della dimostrazione.

**Idea.**  
Si applica Cauchy–Schwarz a $\ell'(\theta;\mathbf{Y})$ e $T(\mathbf{Y})$:
$$
\operatorname{Var}(T)\operatorname{Var}(\ell')\ge[\operatorname{Cov}(T,\ell')]^2.
$$
Ma $\operatorname{Cov}(T,\ell')=1$ per uno stimatore non distorto (derivando $E[T]=\theta$), e $\operatorname{Var}(\ell')=I(\theta)$ per Bartlett. Da cui la tesi. $\square$

> **Osservazione 6.4.**  
> La disuguaglianza di Cramér–Rao è il motivo per cui la SMV è considerata **ottimale** tra gli stimatori asintoticamente non distorti: la sua varianza asintotica è $1/I(\theta)$, il minimo possibile. L'informazione attesa è quindi la misura naturale della "quantità di informazione" che un campione porta sul parametro.
