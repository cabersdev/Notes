# La verosimiglianza statistica

---

## 0. Impostazione: l'inferenza come problema di riduzione

### Osservazione 0.1

Il fenomeno di interesse viene osservato attraverso un campione $\mathbf{y}=(y_1,\dots,y_n)$, che è una realizzazione di un vettore casuale $\mathbf{Y}=(Y_1,\dots,Y_n)$ con supporto $\mathcal{S}^n$. La distribuzione vera di $\mathbf{Y}$, che si indica con $p_0(\mathbf{y})$, è **ignota**. Lo scopo dell'inferenza è usare $\mathbf{y}$ per trarre conclusioni su $p_0$.

Poiché $p_0$ è ignota, si specifica un **insieme di alternative possibili** per $p_0$. Tale insieme è il **modello statistico** $\mathcal{F}$. L'inferenza sarà tanto più accurata quanto più $\mathcal{F}$ è ristretto e ben specificato: la natura del fenomeno, lo schema di campionamento e le informazioni pregresse delimitano $\mathcal{F}$.

### Definizione 0.2 — Spazio campionario

Lo **spazio campionario** $\mathcal{Y}$ è l'insieme di tutti i possibili valori del campione compatibili con il modello $\mathcal{F}$:
$$
\mathcal{Y}=\{\mathbf{y} : p(\mathbf{y};\theta)>0\ \text{per qualche }\theta\in\Theta\}.
$$
In generale $\mathcal{Y}\subseteq\mathcal{S}^n$, e coincide con $\mathcal{S}^n$ se il supporto non dipende da $\theta$.

> **Osservazione 0.3.**  
> La distinzione tra $\mathcal{S}$ (supporto della singola variabile) e $\mathcal{Y}$ (spazio campionario) è importante quando il supporto dipende dal parametro (es. $U(0,\theta)$): in quel caso $\mathcal{Y}$ è l'unione dei supporti al variare di $\theta\in\Theta$, e la verosimiglianza è definita su $\Theta$ ma nulla al di fuori di $\mathcal{Y}$.

## 1. Modelli statistici parametrici

### Definizione 1.1 — Modello statistico parametrico

Un **modello statistico parametrico** è una famiglia di distribuzioni di probabilità (o densità) indicizzata da un parametro $\theta$ appartenente a uno **spazio parametrico** $\Theta\subseteq\mathbb{R}^K$:
$$
\mathcal{F}=\{p_Y(\cdot;\theta) : \theta\in\Theta\} := \{p_{Y_1, \dots, Y_n} (y_1, \dots, y_n) = p_{Y_1, \dots, Y_n} (y_1, \dots, y_n | \theta), \theta \in \Theta, y_1, \dots, y_n \in \mathcal{S}^n \}
$$
Gli elementi di $\mathcal{F}$ sono tutti dello stesso tipo (tutti densità, o tutti funzioni di probabilità) e si distinguono tra loro unicamente per il valore di $\theta$.

> **Osservazione 1.2.**  
> Un modello **non parametrico** è un insieme $\mathcal{F}$ di distribuzioni che non è indicizzato da un parametro di dimensione finita. Esempi:
> - $\mathcal{F}_1$: tutte le densità derivabili su $\mathbb{R}$;
> - $\mathcal{F}_2=\mathcal{F}_1\cap\{\text{densità positive con logaritmo concavo}\}$;
> - $\mathcal{F}_3$: tutte le funzioni di probabilità con supporto $\mathbb{N}$.
>
> I modelli non parametrici sono più flessibili ma richiedono tecniche inferenziali diverse (kernel, serie ortogonali, metodi di verosimiglianza empirica).

### Definizione 1.4 - Correttamente specificato

Un modello parametrico è **correttamente specificato** se dato $\mathcal{F}$ e $p_0$

$$
p_0 \in \mathcal{F} \iff \exists \theta_0 : p_0(y) = p_Y(y; \theta_0)
$$

>**Osservazione 1.5:** Esiste una branchia della statistica che va a studiare modelli parametrici non correttamente specificati, (statistica robusta)

### Definizione 1.3 — Identificabilità

Un modello statistico parametrico è **identificabile** se
$$
\theta_1\neq\theta_2
\quad\Longrightarrow\quad
\exists B\subseteq\mathcal{Y} : P(\mathbf{Y}\in B;\theta_1)\neq P(\mathbf{Y}\in B;\theta_2).
$$
Equivalentemente, se $p_Y(\cdot;\theta_1)=p_Y(\cdot;\theta_2)$ q.o. implica $\theta_1=\theta_2$.

Se il modello non è identificabile si dice sovraparametrizzato, in questo caso si tende a restringere lo spazio parametrico, o eventualmente un campio di campo da $\mathbb{R}$ a $\mathbb{C}$

> **Esempio 1.4 — Modello sovraparametrizzato.**  
> Il modello
> $$
> \mathcal{F}=\{\mathcal{N}(0,\theta^2) : \theta\in\mathbb{R}\}
> $$
> non è identificabile, perché $\mathcal{N}(0,\theta^2)=\mathcal{N}(0,(-\theta)^2)$ per ogni $\theta$. Il vincolo $\theta\in\mathbb{R}^+\setminus\{0\}$ rende il modello identificabile. In generale, un modello è sovraparametrizzato quando valori distinti di $\theta$ producono la stessa distribuzione: in tal caso è impossibile distinguere la vera distribuzione $p_0$ sulla base dei dati.

> **Osservazione 1.5.**  
> L'identificabilità è una condizione **necessaria** per l'inferenza: se due valori del parametro producono la stessa legge, nessuna procedura inferenziale può distinguerli. È una proprietà del modello, non della procedura. Si assume nel seguito che il modello sia **correttamente specificato** (esiste $\theta_0\in\Theta$ con $p_Y(\cdot;\theta_0)=p_0$) e **identificabile**.

### Definizione 1.6 — Riparametrizzazione

Sia $h:\Theta\to\Psi$ una funzione **biunivoca**, con $\Psi=\{h(\theta) : \theta\in\Theta\}$. Allora il modello può essere riscritto equivalentemente come
$$
\mathcal{F}=\{p_Y(\cdot;\psi) : \psi\in\Psi\},
$$
dove $p_Y(\cdot;\psi)=p_Y(\cdot;h^{-1}(\psi))$. Le due parametrizzazioni sono equivalenti, e la scelta è di convenienza.

> **Esempio 1.7.**  
> Per $Y\sim\operatorname{Esp}(\lambda)$ con densità $p_Y(y;\lambda)=\lambda e^{-\lambda y}$, la parametrizzazione in termini del **tasso** $\lambda$ è equivalente a quella in termini della **media** $\mu=1/\lambda$, con densità $p_Y(y;\mu)=\frac{1}{\mu}e^{-y/\mu}$. Le due parametrizzazioni sono legate dalla funzione biunivoca $h(\lambda)=1/\lambda$. Le conclusioni inferenziali devono essere invarianti rispetto alla parametrizzazione.

> **Osservazione 1.8.**  
> L'invarianza per riparametrizzazione è una proprietà desiderabile: se un risultato inferenziale dipendesse dalla parametrizzazione scelta, non sarebbe una proprietà del fenomeno ma un artefatto della scrittura. La verosimiglianza, come vedremo, è invariante per riparametrizzazione, e questa è una delle sue proprietà fondamentali.

## 2. La funzione di verosimiglianza

### Definizione 2.1 — Verosimiglianza

Sia $\mathbf{y}_{\text{oss}}$ il campione osservato e sia $\mathcal{F}=\{p_Y(\cdot;\theta) : \theta\in\Theta\}$ il modello statistico. La **funzione di verosimiglianza** per $\theta$ basata su $\mathbf{y}_{\text{oss}}$ è la funzione
$$
L:\Theta\to[0,\infty),
\qquad
L(\theta)=L(\theta;\mathbf{y}_{\text{oss}})=c(\mathbf{y}_{\text{oss}})\cdot p_Y(\mathbf{y}_{\text{oss}};\theta),
$$
dove $c(\mathbf{y}_{\text{oss}})>0$ è una costante che può dipendere dai dati ma **non** da $\theta$.

> **Osservazione 2.2 — Verosimiglianza e densità.**  
> La funzione $p_Y(\mathbf{y};\theta)$ ha due letture:
> - come funzione di $\mathbf{y}$ a $\theta$ fissato, è la densità (o probabilità) del campione;
> - come funzione di $\theta$ a $\mathbf{y}$ fissato, è la verosimiglianza.
>
> La differenza è sostanziale: la densità integra (o somma) a $1$ su $\mathcal{Y}$; la verosimiglianza, come funzione di $\theta$, **non integra a $1$** su $\Theta$. Non è una probabilità sul parametro, ma una misura di **plausibilità relativa** dei valori di $\theta$ alla luce del campione.

### Esempio 2.3 — Verosimiglianza per un campione Poisson

Sia $\mathbf{y}_{\text{oss}}=(y_1,\dots,y_n)$ un campione i.i.d. da $Y\sim\operatorname{Po}(\lambda)$, con
$$
p_Y(y;\lambda)=e^{-\lambda}\frac{\lambda^y}{y!}.
$$
La probabilità congiunta è
$$
P(Y_1=y_1,\dots,Y_n=y_n;\lambda)
=\prod_{i=1}^n e^{-\lambda}\frac{\lambda^{y_i}}{y_i!}
=e^{-n\lambda}\frac{\lambda^{\sum_i y_i}}{\prod_i y_i!}.
$$
La verosimiglianza è quindi
$$
L(\lambda;\mathbf{y}_{\text{oss}})
=e^{-n\lambda}\frac{\lambda^{\sum_i y_i}}{\prod_i y_i!}
\propto e^{-n\lambda}\lambda^{\sum_i y_i}.
$$
La costante $\prod_i y_i!$ non dipende da $\lambda$ e può essere omessa.

**Esempio numerico.** Se $\mathbf{y}_{\text{oss}}=(5,6,3,4)$, allora $n=4$ e $\sum_i y_i=18$, e
$$
L(\lambda;\mathbf{y}_{\text{oss}})\propto e^{-4\lambda}\lambda^{18}.
$$

### Definizione 2.4 — Verosimiglianze equivalenti

Due funzioni di verosimiglianza sono **equivalenti** se differiscono per una costante moltiplicativa positiva che non dipende da $\theta$:
$$
L_1(\theta)\propto L_2(\theta)
\quad\Longleftrightarrow\quad
\exists c>0 : L_1(\theta)=c\,L_2(\theta)\ \forall\theta\in\Theta.
$$
La verosimiglianza è quindi una **classe di equivalenza** di funzioni, e ai fini dell'inferenza è sufficiente conoscerne una qualsiasi rappresentante.

> **Osservazione 2.5.**  
> Il rapporto tra due verosimiglianze valutate in punti diversi del parametro è invariante per la scelta della rappresentante:
> $$
> \frac{L(\theta_2)}{L(\theta_1)}
> =
> \frac{c(\mathbf{y}_{\text{oss}})\,p_Y(\mathbf{y}_{\text{oss}};\theta_2)}{c(\mathbf{y}_{\text{oss}})\,p_Y(\mathbf{y}_{\text{oss}};\theta_1)}
> =
> \frac{p_Y(\mathbf{y}_{\text{oss}};\theta_2)}{p_Y(\mathbf{y}_{\text{oss}};\theta_1)}.
> $$
> Questo è il senso in cui la verosimiglianza è definita "a meno di costanti": le conclusioni inferenziali dipendono solo dai rapporti, non dai valori assoluti.

**Esempio 2.6 — Confronto di plausibilità.**  
Con $\mathbf{y}_{\text{oss}}=(5,6,3,4)$, volendo confrontare $\lambda_1=4$ e $\lambda_2=3$,
$$
\frac{L(\lambda_2)}{L(\lambda_1)}
=
\frac{e^{-4\cdot3}3^{18}}{e^{-4\cdot4}4^{18}}
=
e^{4}\left(\frac{3}{4}\right)^{18}
\approx 0.31<1.
$$
Quindi $\lambda_1=4$ è più plausibile di $\lambda_2=3$ alla luce dei dati.

### Definizione 2.7 — Log-verosimiglianza

La **funzione di log-verosimiglianza** è
$$
\ell(\theta)=\ell(\theta;\mathbf{y}_{\text{oss}})=\log L(\theta),
$$
con la convenzione $\ell(\theta)=-\infty$ se $L(\theta)=0$.

> **Osservazione 2.8.**  
> Poiché $\log$ è strettamente crescente, $L(\theta_1)>L(\theta_2)\iff\ell(\theta_1)>\ell(\theta_2)$. Le due funzioni hanno gli stessi punti di massimo e lo stesso ordinamento. Due log-verosimiglianze sono equivalenti se differiscono per una costante **additiva** che non dipende da $\theta$. La log-verosimiglianza è spesso più comoda perché trasforma prodotti in somme e rapporti in differenze.

**Esempio 2.9 — Log-verosimiglianza Poisson.**  
Dall'Esempio 2.3,
$$
\ell(\lambda;\mathbf{y}_{\text{oss}})
=
-n\lambda+\left(\sum_i y_i\right)\log\lambda-\log\left(\prod_i y_i!\right)
\propto
-n\lambda+\left(\sum_i y_i\right)\log\lambda.
$$
Con $\mathbf{y}_{\text{oss}}=(5,6,3,4)$, $\ell(\lambda)\propto -4\lambda+18\log\lambda$.

### Proposizione 2.10 — Invarianza per riparametrizzazione

Sia $\psi=h(\theta)$ con $h$ biunivoca. Allora la verosimiglianza per $\psi$ è
$$
L_\psi(\psi;\mathbf{y}_{\text{oss}})=L_\theta(h^{-1}(\psi);\mathbf{y}_{\text{oss}}),
$$
e le due verosimiglianze hanno gli stessi punti di massimo (trasformati tramite $h$).

**Dimostrazione.**  
Per definizione, $L_\psi(\psi)=p_Y(\mathbf{y}_{\text{oss}};h^{-1}(\psi))$, mentre $L_\theta(\theta)=p_Y(\mathbf{y}_{\text{oss}};\theta)$. Posto $\theta=h^{-1}(\psi)$, si ha $L_\psi(\psi)=L_\theta(h^{-1}(\psi))$. Il massimo di $L_\psi$ su $\Psi$ si ottiene in $\hat\psi=h(\hat\theta)$, dove $\hat\theta$ massimizza $L_\theta$, perché $h$ è biunivoca. $\square$

> **Osservazione 2.11.**  
> L'invarianza per riparametrizzazione è una proprietà fondamentale: le conclusioni inferenziali (stime, intervalli, test) non dipendono dalla parametrizzazione scelta. Questo è uno dei motivi per cui la verosimiglianza è un principio inferenziale "naturale".

## 3. Il principio debole di verosimiglianza

### Principio 3.1 — Principio debole di verosimiglianza

Sia $\mathcal{F}=\{p_Y(\cdot;\theta) : \theta\in\Theta\}$ un modello statistico. Due punti $\mathbf{y}$ e $\mathbf{z}$ dello spazio campionario che producono verosimiglianze equivalenti,
$$
L(\theta;\mathbf{y})\propto L(\theta;\mathbf{z}),
$$
devono condurre alle **medesime conclusioni inferenziali** su $\theta$.

> **Osservazione 3.2.**  
> Il principio afferma che l'informazione sul parametro contenuta nel campione è **tutta** nella verosimiglianza. Se due campioni producono la stessa verosimiglianza (a meno di costanti), allora essi contengono la stessa informazione su $\theta$, anche se sono diversi come realizzazioni. Questo è il fondamento logico dell'inferenza di verosimiglianza.

### Esempio 3.3 — Due domini di Poisson

Si osservino due domini disgiunti $D_1$ e $D_2$, con $Y_1\sim\operatorname{Po}(\lambda)$ su $D_1$ e $Y_2\sim\operatorname{Po}(\lambda)$ su $D_2$, indipendenti. Si osservano $y_1=3$ e $y_2=4$.

**Opzione 1 — Usare solo la somma $S=Y_1+Y_2$.**  
Per la proprietà additiva della Poisson, $S\sim\operatorname{Po}(2\lambda)$, con $s_{\text{oss}}=7$. Allora
$$
L(\lambda;s_{\text{oss}})\propto e^{-2\lambda}(2\lambda)^7.
$$

**Opzione 2 — Usare il campione completo $(y_1,y_2)$.**  
Per l'indipendenza,
$$
L(\lambda;y_1,y_2)\propto e^{-\lambda}\lambda^{y_1}\cdot e^{-\lambda}\lambda^{y_2}=e^{-2\lambda}\lambda^{y_1+y_2}=e^{-2\lambda}\lambda^7.
$$
Le due verosimiglianze sono **equivalenti** (differiscono per la costante $2^7$, che non dipende da $\lambda$). Quindi, per il principio debole, le conclusioni inferenziali sono le stesse. La somma $S$ è una **statistica sufficiente** per $\lambda$: contiene tutta l'informazione su $\lambda$ contenuta nella coppia $(Y_1,Y_2)$.

> **Osservazione 3.4.**  
> L'esempio mostra che la verosimiglianza opera una **sintesi** dell'informazione: punti diversi dello spazio campionario che producono la stessa verosimiglianza (a meno di costanti) sono equivalenti ai fini dell'inferenza. La sintesi è operata da una statistica sufficiente, che formalizzeremo nella Sezione 4.

## 4. Statistiche sufficienti

### Definizione 4.1 — Statistica sufficiente

Sia $\mathcal{F}=\{p_Y(\cdot;\theta) : \theta\in\Theta\}$ un modello statistico. Una statistica $T(\mathbf{Y})$ è **sufficiente** per $\theta$ se la distribuzione condizionata di $\mathbf{Y}$ dato $T(\mathbf{Y})=t$ **non dipende da $\theta$**:
$$
p_{\mathbf{Y}\mid T}(\mathbf{y}\mid t;\theta)=p_{\mathbf{Y}\mid T}(\mathbf{y}\mid t)
\qquad\forall\theta\in\Theta.
$$

> **Osservazione 4.2.**  
> Il senso è che, una volta noto $T(\mathbf{y})$, il campione completo non aggiunge informazione su $\theta$: la variabilità residua è "puro rumore" rispetto al parametro. La statistica sufficiente **esaurisce** l'informazione su $\theta$ contenuta nel campione.

### Teorema 4.3 — Teorema di fattorizzazione di Neyman–Fisher

Sia $\mathcal{F}=\{p_Y(\cdot;\theta) : \theta\in\Theta\}$ un modello statistico. Una statistica $T(\cdot)$ è sufficiente per $\theta$ se e solo se esistono due funzioni $h(\cdot)$ e $g(\cdot;\cdot)$ tali che
$$
p_Y(\mathbf{y};\theta)=h(\mathbf{y})\,g(T(\mathbf{y});\theta)
\qquad\forall\mathbf{y}\in\mathcal{Y},\ \forall\theta\in\Theta,
$$
dove $h$ non dipende da $\theta$ e $g$ dipende dai dati solo attraverso $T(\mathbf{y})$.

**Dimostrazione (caso discreto).**  
Sia $T$ sufficiente. Allora
$$
p_{\mathbf{Y}\mid T}(\mathbf{y}\mid t;\theta)=\frac{p_Y(\mathbf{y};\theta)}{P(T(\mathbf{Y})=t;\theta)}
$$
non dipende da $\theta$. Posto $t=T(\mathbf{y})$, si ha
$$
p_Y(\mathbf{y};\theta)=p_{\mathbf{Y}\mid T}(\mathbf{y}\mid t)\cdot P(T(\mathbf{Y})=t;\theta).
$$
Basta porre $h(\mathbf{y})=p_{\mathbf{Y}\mid T}(\mathbf{y}\mid T(\mathbf{y}))$ e $g(t;\theta)=P(T(\mathbf{Y})=t;\theta)$.

Viceversa, se $p_Y(\mathbf{y};\theta)=h(\mathbf{y})g(T(\mathbf{y});\theta)$, allora
$$
P(T(\mathbf{Y})=t;\theta)=\sum_{\mathbf{y}:T(\mathbf{y})=t}h(\mathbf{y})g(t;\theta)=g(t;\theta)\sum_{\mathbf{y}:T(\mathbf{y})=t}h(\mathbf{y}),
$$
e quindi
$$
p_{\mathbf{Y}\mid T}(\mathbf{y}\mid t;\theta)
=
\frac{h(\mathbf{y})g(t;\theta)}{g(t;\theta)\sum_{\mathbf{y}'\,:\,T(\mathbf{y}')=t}h(\mathbf{y}')}
=
\frac{h(\mathbf{y})}{\sum_{\mathbf{y}'\,:\,T(\mathbf{y}')=t}h(\mathbf{y}')},
$$
che non dipende da $\theta$. Dunque $T$ è sufficiente. $\square$

> **Osservazione 4.4.**  
> Il teorema di fattorizzazione è lo strumento operativo per verificare la sufficienza: si scrive la densità (o probabilità) congiunta e si cerca di fattorizzarla in una parte che dipende dai dati solo attraverso $T$ e una parte che non dipende da $\theta$.

### Corollario 4.5 — Sufficienza e verosimiglianza

$T$ è sufficiente per $\theta$ se e solo se, per ogni coppia $\mathbf{y},\mathbf{z}\in\mathcal{Y}$ con $T(\mathbf{y})=T(\mathbf{z})$,
$$
L(\theta;\mathbf{y})\propto L(\theta;\mathbf{z}).
$$

**Dimostrazione.**  
Segue dal teorema di fattorizzazione: se $T(\mathbf{y})=T(\mathbf{z})=t$, allora
$$
p_Y(\mathbf{y};\theta)=h(\mathbf{y})g(t;\theta),
\qquad
p_Y(\mathbf{z};\theta)=h(\mathbf{z})g(t;\theta),
$$
e quindi $L(\theta;\mathbf{y})/L(\theta;\mathbf{z})=h(\mathbf{y})/h(\mathbf{z})$, che non dipende da $\theta$. Viceversa, se la verosimiglianza dipende dai dati solo attraverso $T$, allora la distribuzione condizionata dato $T$ non dipende da $\theta$, e $T$ è sufficiente. $\square$

> **Osservazione 4.6.**  
> Il corollario stabilisce il legame profondo tra sufficienza e verosimiglianza: la verosimiglianza è funzione dei dati **solo attraverso una statistica sufficiente**. Questo è il senso preciso del principio debole di verosimiglianza: tutta l'informazione su $\theta$ è contenuta nella verosimiglianza, e la verosimiglianza dipende dai dati solo attraverso la statistica sufficiente minimale.

### Esempi 4.7 — Statistiche sufficienti notevoli

1. **Bernoulli.** $Y_1,\dots,Y_n$ i.i.d. $\operatorname{Ber}(\pi)$. La verosimiglianza è
   $$
   L(\pi;\mathbf{y})\propto\pi^{\sum_i y_i}(1-\pi)^{n-\sum_i y_i}.
   $$
   Per il teorema di fattorizzazione, $T(\mathbf{y})=\sum_i y_i$ è sufficiente per $\pi$. Anche la proporzione campionaria $\bar y=T/n$ è sufficiente (è una funzione biunivoca di $T$).

2. **Poisson.** $Y_1,\dots,Y_n$ i.i.d. $\operatorname{Po}(\lambda)$. La verosimiglianza è
   $$
   L(\lambda;\mathbf{y})\propto e^{-n\lambda}\lambda^{\sum_i y_i}.
   $$
   $T(\mathbf{y})=\sum_i y_i$ è sufficiente per $\lambda$.

3. **Normale con $\sigma^2$ nota.** $Y_1,\dots,Y_n$ i.i.d. $\mathcal{N}(\mu,\sigma^2)$. La verosimiglianza è
   $$
   L(\mu;\mathbf{y})\propto\exp\left(-\frac{1}{2\sigma^2}\sum_i(y_i-\mu)^2\right)
   \propto\exp\left(-\frac{n}{2\sigma^2}(\bar y-\mu)^2\right),
   $$
   quindi $T(\mathbf{y})=\bar y$ è sufficiente per $\mu$.

4. **Normale con $\mu$ e $\sigma^2$ ignoti.** La verosimiglianza è
   $$
   L(\mu,\sigma^2;\mathbf{y})\propto(\sigma^2)^{-n/2}\exp\left(-\frac{1}{2\sigma^2}\sum_i(y_i-\bar y)^2-\frac{n}{2\sigma^2}(\bar y-\mu)^2\right).
   $$
   La coppia $T(\mathbf{y})=(\bar y,\sum_i(y_i-\bar y)^2)$ è sufficiente per $(\mu,\sigma^2)$.

5. **Uniforme.** $Y_1,\dots,Y_n$ i.i.d. $U(0,\theta)$. La verosimiglianza è
   $$
   L(\theta;\mathbf{y})=\theta^{-n}\mathbf{1}_{\{y_{(n)}\le\theta\}},
   $$
   dove $y_{(n)}=\max_i y_i$. Per il teorema di fattorizzazione, $T(\mathbf{y})=y_{(n)}$ è sufficiente per $\theta$.

## 5. Statistica sufficiente minimale

### Osservazione 5.1

Fissato un modello, esistono molte statistiche sufficienti. Banalmente, il campione stesso è sufficiente, così come lo è la statistica ordinata (campione ordinato) e la verosimiglianza stessa. Inoltre, ogni funzione biunivoca di una statistica sufficiente è sufficiente. Poiché l'obiettivo è ottenere la massima sintesi possibile senza perdita di informazione, si introduce il concetto di **statistica sufficiente minimale**.

### Definizione 5.2 — Statistica sufficiente minimale

Una statistica $T_n=T(\mathbf{y})$ è **sufficiente minimale** per $\theta$ se assume valori distinti solo in punti dello spazio campionario a cui corrispondono verosimiglianze **non equivalenti**:
$$
T(\mathbf{z}_1)=T(\mathbf{z}_2)
\quad\Longleftrightarrow\quad
L(\theta;\mathbf{z}_1)\propto L(\theta;\mathbf{z}_2).
$$

### Proposizione 5.3 — Unicità della statistica sufficiente minimale

Se esistono due statistiche sufficienti minimali $T_1$ e $T_2$, allora sono funzioni l'una dell'altra. In particolare, la statistica sufficiente minimale è unica a meno di trasformazioni biunivoche.

**Dimostrazione.**  
Siano $T_1$ e $T_2$ sufficienti minimali. Poiché $T_1$ è sufficiente, la verosimiglianza dipende dai dati solo attraverso $T_1$: se $T_1(\mathbf{y})=T_1(\mathbf{z})$, allora $L(\theta;\mathbf{y})\propto L(\theta;\mathbf{z})$. Poiché $T_2$ è minimale, la condizione $L(\theta;\mathbf{y})\propto L(\theta;\mathbf{z})$ implica $T_2(\mathbf{y})=T_2(\mathbf{z})$. Quindi $T_2$ è costante sulle fibre di $T_1$, cioè esiste una funzione $\phi$ con $T_2=\phi(T_1)$. Per simmetria, $T_1=\psi(T_2)$. Dunque $T_1$ e $T_2$ sono funzioni l'una dell'altra. $\square$

> **Osservazione 5.4.**  
> La statistica sufficiente minimale è la **massima riduzione** dei dati che non perde informazione su $\theta$. Ogni altra statistica sufficiente è una funzione della minimale, e quindi contiene più informazione (o la stessa, se biunivoca) ma non meno. In questo senso, la statistica sufficiente minimale è il "rappresentante canonico" dell'informazione contenuta nel campione.

### Esempio 5.5 — Statistica sufficiente minimale per la Bernoulli

Per $Y_1,\dots,Y_n$ i.i.d. $\operatorname{Ber}(\pi)$, la statistica sufficiente minimale è $T(\mathbf{y})=\sum_i y_i$. Infatti, la verosimiglianza dipende dai dati solo attraverso $\sum_i y_i$, e due campioni con la stessa somma hanno verosimiglianze equivalenti. La proporzione campionaria $\bar y=T/n$ è una trasformazione biunivoca di $T$, quindi anch'essa minimale.

### Esempio 5.6 — Statistica sufficiente minimale per la normale

Per $Y_1,\dots,Y_n$ i.i.d. $\mathcal{N}(\mu,\sigma^2)$ con entrambi i parametri ignoti, la statistica sufficiente minimale è
$$
T(\mathbf{y})=\left(\sum_i y_i,\sum_i y_i^2\right),
$$
oppure equivalentemente $(\bar y,S^2)$. Due campioni con la stessa media e la stessa varianza campionaria hanno verosimiglianze equivalenti, e non esistono riduzioni ulteriori che preservino l'informazione su $(\mu,\sigma^2)$.

### Esempio 5.7 — Statistica sufficiente minimale per l'uniforme

Per $Y_1,\dots,Y_n$ i.i.d. $U(0,\theta)$, la statistica sufficiente minimale è $T(\mathbf{y})=y_{(n)}=\max_i y_i$. La verosimiglianza è $\theta^{-n}\mathbf{1}_{\{y_{(n)}\le\theta\}}$, che dipende dai dati solo attraverso il massimo.
