# Introduzione all'inferenza di verosimiglianza

---

## 0. Motivazione dietro il paradigma

### Esempio 0.1 — Il disastro del Challenger

Il 28 gennaio 1986 lo Space Shuttle Challenger esplose 73 secondi dopo il decollo, causando la morte dei sette membri dell'equipaggio. L'incidente fu causato dal cedimento di una coppia di o-ring (guarnizioni in gomma) del Solid Rocket Booster destro. Gli o-ring sigillano le giunture del razzo e sono attivati dal calore al momento della partenza. A temperature eccezionalmente basse (31 °F, circa -0.5 °C) la gomma perde flessibilità e si rompe più facilmente.

Prima del lancio, i tecnici avevano a disposizione dati empirici sul danneggiamento degli o-ring in funzione della temperatura. La decisione di lanciare fu presa analizzando **solo i lanci in cui si era verificato un danneggiamento**, trascurando i lanci senza danneggiamento. Questo è un errore statistico grave: si è verificata una **distorsione da selezione** del campione.

La domanda che ci poniamo è:

> **Domanda.** Un uso corretto della statistica avrebbe potuto impedire il disastro?

### Osservazione 0.2 — Perché gli strumenti classici non bastano

Supponiamo di voler rispondere a due domande operative:

1. La temperatura $t$ influenza la **probabilità** di guasto degli o-ring?
2. Qual è la probabilità di guasto a $t=31$ °F?

Il modello naturale è
$$
Y_i\sim\operatorname{Ber}(\pi_i),
\qquad
Y_i=\begin{cases}1&\text{se il lancio }i\text{ ha avuto un guasto},\\0&\text{altrimenti},\end{cases}
$$
dove $\pi_i=P(Y_i=1)$ dipende dalla temperatura $t_i$ del lancio $i$.


- **Metodo dei momenti / analogia:** $\hat\pi=\sum_iY_i/n\approx 0.30$. La stima **non dipende da $t$**, quindi non risponde alla domanda 1.
- **Minimi quadrati:** si può porre $Y_i=\beta_0+\beta_1t_i + \epsilon_i$ con $\mathbb{E}(\epsilon_i)$ da cui $\mathbb{E}(Y_i) = \pi_i = \beta_0 + \beta_1 t_i$ e stimare $(\beta_0,\beta_1)$. Il problema è che $\pi_i$ deve stare in $(0,1)$, mentre la stima produce $\hat\pi=1.66>1$ per $t=31$ °F. Il modello lineare non rispetta il vincolo di probabilità.

Serve quindi una **trasformazione** che mappi $\mathbb{R}$ in $(0,1)$. La scelta canonica è la **funzione logistica**:
$$
\psi_i=\log\frac{\pi_i}{1-\pi_i},
\qquad
\psi_i=\gamma_0+\gamma_1t_i.
$$
Invertendo,
$$
\pi_i=\frac{e^{\gamma_0+\gamma_1t_i}}{1+e^{\gamma_0+\gamma_1t_i}}.
$$
Allora:

- $\pi_i$ è crescente in $t_i$ se $\gamma_1>0$;
- è decrescente se $\gamma_1<0$;
- è costante se $\gamma_1=0$ e non dipende da $t$.

La risposta alla domanda 1 diventa un **test di ipotesi** $H_0:\gamma_1=0$ contro $H_1:\gamma_1\neq0$. La risposta alla domanda 2 richiede la **stima** di $\gamma=(\gamma_0,\gamma_1)$ e la valutazione di $\hat\pi(t=31)$.

La funzione di probabilità del campione è
$$
P(Y_i=y_i)=\pi_i^{y_i}(1-\pi_i)^{1-y_i}
=
\frac{\exp\{(\gamma_0+\gamma_1t_i)y_i\}}{1+\exp\{\gamma_0+\gamma_1t_i\}}.
$$
Il problema è che **gli strumenti classici non permettono di stimare $\gamma$ in modo naturale**: né il metodo dei momenti né i minimi quadrati si adattano direttamente a questa parametrizzazione. Serve un principio inferenziale più generale.

### Osservazione 0.3 — L'idea della verosimiglianza

L'**inferenza di verosimiglianza** fornisce un principio unificante: si scrive la probabilità (o densità) congiunta del campione osservato come funzione del parametro, e si cercano i valori del parametro che rendono il campione osservato **più plausibile**. Da questa funzione si derivano:

- **stime** puntuali (massimizzazione);
- **intervalli di confidenza** (regioni in cui la verosimiglianza è alta);
- **test di ipotesi** (confronto tra la verosimiglianza massima vincolata e quella libera).

Il prezzo da pagare è che è necessario specificare un **modello statistico parametrico**: bisogna conoscere la distribuzione del campione condizionatamente al parametro.

## 1. Il modello statistico parametrico

### Definizione 1.1 — Modello statistico parametrico

Un **modello statistico parametrico** è una famiglia di distribuzioni di probabilità indicizzata da un parametro $\theta\in\Theta\subseteq\mathbb{R}^p$:
$$
\mathcal{P}=\{P_\theta : \theta\in\Theta\}.
$$
Se le osservazioni $Y_1,\dots,Y_n$ sono continue con densità $p(y;\theta)$ o discrete con funzione di probabilità $p(y;\theta)$, e se sono i.i.d., la densità (o probabilità) congiunta del campione $\mathbf{y}=(y_1,\dots,y_n)$ è
$$
p(\mathbf{y};\theta)=\prod_{i=1}^n p(y_i;\theta).
$$

> **Osservazione 1.2.**  
> La specificazione del modello è il **presupposto imprescindibile** dell'inferenza di verosimiglianza. Nella prima parte del corso, alcune procedure (metodo dei momenti, test asintotici basati sul CLT) funzionavano senza assumere una forma parametrica. Con la verosimiglianza, invece, la forma della distribuzione è parte integrante del modello: senza di essa non si può scrivere la funzione di verosimiglianza.

### Esempio 1.3 — Modelli parametrici comuni

- $Y_i\sim\mathcal{N}(\mu,\sigma^2)$, $\theta=(\mu,\sigma^2)$;
- $Y_i\sim\operatorname{Ber}(\pi)$, $\theta=\pi$;
- $Y_i\sim\operatorname{Po}(\lambda)$, $\theta=\lambda$;
- $Y_i\sim\operatorname{Exp}(\lambda)$, $\theta=\lambda$;
- $Y_i\sim\operatorname{Gamma}(\alpha,\lambda)$, $\theta=(\alpha,\lambda)$;
- modello di regressione logistica: $Y_i\sim\operatorname{Ber}(\pi_i)$, $\pi_i=g^{-1}(\gamma_0+\gamma_1x_i)$, $\theta=(\gamma_0,\gamma_1)$.

In tutti i casi, una volta fissato il modello, la densità congiunta è determinata e la verosimiglianza può essere scritta.

## 2. La funzione di verosimiglianza

### Definizione 2.1 — Funzione di verosimiglianza

Sia $\mathbf{y}=(y_1,\dots,y_n)$ un campione osservato da un modello parametrico con densità (o probabilità) congiunta $p(\mathbf{y};\theta)$. La **funzione di verosimiglianza** è
$$
L(\theta;\mathbf{y})=p(\mathbf{y};\theta)=\prod_{i=1}^n p(y_i;\theta),
\qquad\theta\in\Theta.
$$
La **log-verosimiglianza** è
$$
\ell(\theta;\mathbf{y})=\log L(\theta;\mathbf{y})=\sum_{i=1}^n\log p(y_i;\theta).
$$

> **Osservazione 2.2 — Verosimiglianza vs probabilità.**  
> La funzione $p(\mathbf{y};\theta)$ ha due letture:
> - come funzione di $\mathbf{y}$ a $\theta$ fissato, è la **densità (o probabilità) del campione**; è una distribuzione di probabilità, integra (o somma) a $1$ su $\mathbb{R}^n$;
> - come funzione di $\theta$ a $\mathbf{y}$ fissato, è la **verosimiglianza**; è una funzione da $\Theta$ a $[0,\infty)$, e in generale **non integra a $1$** su $\Theta$.
>
> Questa distinzione è cruciale: la verosimiglianza non è una probabilità sul parametro, non è una distribuzione a posteriori. È una misura di **plausibilità relativa** dei valori del parametro alla luce del campione osservato.

### Esempio 2.3 — Verosimiglianza per un campione bernoulliano

Sia $Y_1,\dots,Y_n$ i.i.d. $\operatorname{Ber}(\pi)$, con $\pi\in(0,1)$. La funzione di probabilità è
$$
P(Y_i=y_i)=\pi^{y_i}(1-\pi)^{1-y_i}.
$$
La verosimiglianza è
$$
L(\pi;\mathbf{y})=\prod_{i=1}^n\pi^{y_i}(1-\pi)^{1-y_i}
=\pi^{\sum_iy_i}(1-\pi)^{n-\sum_iy_i}.
$$
La log-verosimiglianza è
$$
\ell(\pi;\mathbf{y})=\left(\sum_iy_i\right)\log\pi+\left(n-\sum_iy_i\right)\log(1-\pi).
$$

### Esempio 2.4 — Il caso delle guarnizioni

Un'industria acquista guarnizioni in partite di $5000$ pezzi e vuole verificare che la frazione di difettosi non superi il $5\%$. Su un campione di $n=50$ guarnizioni si osservano $y=4$ difetti. Il modello è $Y_i\sim\operatorname{Ber}(\pi)$ con $\pi$ frazione di difettosi.

La verosimiglianza è
$$
L(\pi)=\pi^4(1-\pi)^{46}.
$$
Per $\pi=0.05$, $L(0.05)\approx0.9\times10^{-15}$; per $\pi=0.1$, $L(0.1)\approx0.8\times10^{-6}$. Il rapporto è
$$
\frac{L(0.1)}{L(0.05)}\approx888\times10^6.
$$
Quindi, se $\pi$ fosse $0.1$, osservare $4$ difetti su $50$ sarebbe quasi un miliardo di volte più probabile che se $\pi$ fosse $0.05$. Il campione osservato è **molto più plausibile** sotto $\pi=0.1$ che sotto $\pi=0.05$.

## 3. Lo stimatore di massima verosimiglianza

### Definizione 3.1 — MLE

Lo **stimatore di massima verosimiglianza** (MLE) di $\theta$ è
$$
\hat\theta=\arg\max_{\theta\in\Theta}L(\theta;\mathbf{y})=\arg\max_{\theta\in\Theta}\ell(\theta;\mathbf{y}).
$$
Se $\ell$ è differenziabile e il massimo è interno, $\hat\theta$ soddisfa l'**equazione di verosimiglianza**
$$
\frac{\partial\ell}{\partial\theta_j}(\hat\theta;\mathbf{y})=0,
\qquad j=1,\dots,p.
$$

> **Osservazione 3.2.**  
> Massimizzare $L$ o $\ell$ è equivalente, perché $\log$ è strettamente crescente. La log-verosimiglianza è di solito più comoda perché trasforma prodotti in somme, semplificando la derivazione.

### Esempio 3.3 — MLE per il modello bernoulliano

La log-verosimiglianza è
$$
\ell(\pi)=\left(\sum_iy_i\right)\log\pi+\left(n-\sum_iy_i\right)\log(1-\pi).
$$
Derivando rispetto a $\pi$:
$$
\ell'(\pi)=\frac{\sum_iy_i}{\pi}-\frac{n-\sum_iy_i}{1-\pi}.
$$
Ponendo $\ell'(\pi)=0$:
$$
\frac{\sum_iy_i}{\pi}=\frac{n-\sum_iy_i}{1-\pi}
\Longleftrightarrow
(1-\pi)\sum_iy_i=\pi\left(n-\sum_iy_i\right)
\Longleftrightarrow
\hat\pi=\frac{\sum_iy_i}{n}.
$$
La derivata seconda è
$$
\ell''(\pi)=-\frac{\sum_iy_i}{\pi^2}-\frac{n-\sum_iy_i}{(1-\pi)^2}<0,
$$
quindi $\hat\pi$ è un massimo.

> **Osservazione 3.4.**  
> Il MLE coincide con la **proporzione campionaria**, cioè con lo stimatore che già conoscevamo per ragioni intuitive. Il valore aggiunto dell'approccio di verosimiglianza non è il risultato in sé, ma il fatto che lo stesso principio si applica **a qualunque modello parametrico**, anche quando non esiste uno stimatore "naturale" ricavabile per analogia o minimi quadrati.

### Esempio 3.5 — MLE per la normale

Sia $Y_1,\dots,Y_n$ i.i.d. $\mathcal{N}(\mu,\sigma^2)$. La log-verosimiglianza è
$$
\ell(\mu,\sigma^2)=-\frac n2\log(2\pi)-\frac n2\log\sigma^2-\frac{1}{2\sigma^2}\sum_{i=1}^n(y_i-\mu)^2.
$$
Le derivate parziali danno
$$
\hat\mu=\bar y,
\qquad
\hat\sigma^2=\frac1n\sum_{i=1}^n(y_i-\bar y)^2.
$$
Si noti che $\hat\sigma^2$ è lo stimatore di massima verosimiglianza, distorto, e differisce dallo stimatore corretto $S^2=\frac{1}{n-1}\sum_i(y_i-\bar y)^2$ solo per il fattore $n/(n-1)$.

### Esempio 3.6 — MLE per il modello logistico

Nel modello logistico, la log-verosimiglianza è
$$
\ell(\gamma_0,\gamma_1)=\sum_{i=1}^n\left[y_i(\gamma_0+\gamma_1t_i)-\log\left(1+e^{\gamma_0+\gamma_1t_i}\right)\right].
$$
Le equazioni di verosimiglianza sono
$$
\begin{cases}
\sum_iy_i=\sum_i\dfrac{e^{\gamma_0+\gamma_1t_i}}{1+e^{\gamma_0+\gamma_1t_i}},\\[2ex]
\sum_iy_it_i=\sum_i t_i\dfrac{e^{\gamma_0+\gamma_1t_i}}{1+e^{\gamma_0+\gamma_1t_i}}.
\end{cases}
$$
Sono equazioni **non lineari** in $(\gamma_0,\gamma_1)$: non ammettono soluzione in forma chiusa. Si risolvono numericamente con un algoritmo di ottimizzazione (Newton–Raphson, Fisher scoring, o metodi quasi-Newton). Questo è uno dei motivi per cui il paradigma della verosimiglianza è potente ma richiede, in pratica, il supporto del calcolo numerico.

## 4. Verosimiglianza relativa e intervalli di confidenza

### Definizione 4.1 — Verosimiglianza relativa

Si definisce **verosimiglianza relativa** la funzione
$$
L^*(\theta)=\frac{L(\theta;\mathbf{y})}{L(\hat\theta;\mathbf{y})}
=\frac{L(\theta;\mathbf{y})}{\sup_{\vartheta\in\Theta}L(\vartheta;\mathbf{y})}.
$$
Poiché $\hat\theta$ massimizza $L$, si ha $0\le L^*(\theta)\le1$ per ogni $\theta\in\Theta$, con $L^*(\hat\theta)=1$.

### Definizione 4.2 — Regione di verosimiglianza

Per un livello $c\in(0,1)$, la **regione di verosimiglianza** di soglia $c$ è
$$
\Gamma_c=\{\theta\in\Theta : L^*(\theta)\ge c\}.
$$
L'idea è che i valori del parametro con verosimiglianza relativa alta sono quelli **plausibili** alla luce del campione; quelli con verosimiglianza relativa bassa sono incompatibili con i dati.

> **Osservazione 4.3.**  
> Le regioni $\Gamma_c$ sono annidate: se $c_1<c_2$, allora $\Gamma_{c_2}\subseteq\Gamma_{c_1}$. Sono inoltre invarianti rispetto a riparametrizzazioni: se $\phi=g(\theta)$ con $g$ biunivoca, allora $\Gamma_c$ per $\theta$ corrisponde a $g(\Gamma_c)$ per $\phi$.

### Esempio 4.4 — Regione di verosimiglianza per il modello bernoulliano

Riprendiamo il caso $n=50$, $\sum_iy_i=4$. La verosimiglianza relativa è
$$
L^*(\pi)=\frac{\pi^4(1-\pi)^{46}}{\hat\pi^4(1-\hat\pi)^{46}},
\qquad \hat\pi=0.08.
$$
Scegliendo $c=0.5$, si ottiene l'intervallo
$$
\Gamma_{0.5}\approx(0.043,0.133).
$$
Poiché $0.05\in\Gamma_{0.5}$, i dati non sono incompatibili con l'ipotesi $\pi=0.05$. Se invece si aumenta la numerosità (es. $n=100$ con $\sum_iy_i=8$, stessa proporzione $\hat\pi=0.08$), la verosimiglianza si "restringe" e $\Gamma_{0.5}\approx(0.048,0.128)$. Il valore $0.05$ è ancora appena incluso, ma la regione si è ristretta. Se $n$ cresce ancora, $0.05$ può uscire e i dati diventano incompatibili con l'ipotesi.

> **Osservazione 4.5.**  
> La scelta $c=0.5$ è arbitraria. In pratica si usa il legame con la teoria asintotica: sotto regolarità, la statistica
> $$
> -2\log L^*(\theta)=-2\log\frac{L(\theta)}{L(\hat\theta)}
> $$
> converge in distribuzione a una $\chi^2_p$ quando $\theta$ è il vero parametro. Questo permette di calibrare $c$ in modo che la regione abbia una copertura asintotica $1-\alpha$, oppure di costruire test con livello asintotico $\alpha$.

## 5. Verosimiglianza e test di ipotesi

### Definizione 5.1 — Test del rapporto di verosimiglianza

Sia $H_0:\theta\in\Theta_0$ contro $H_1:\theta\in\Theta\setminus\Theta_0$, con $\Theta_0$ sottoinsieme di $\Theta$. Si definisce **rapporto di verosimiglianza**
$$
\Lambda=\frac{\sup_{\theta\in\Theta_0}L(\theta;\mathbf{y})}{\sup_{\theta\in\Theta}L(\theta;\mathbf{y})}\in(0,1].
$$
Un valore di $\Lambda$ vicino a $1$ indica che il massimo vincolato è quasi uguale al massimo libero: l'ipotesi nulla non è contraddetta dai dati. Un valore di $\Lambda$ vicino a $0$ indica che l'ipotesi nulla impone una perdita di verosimiglianza consistente: i dati sono incompatibili con $H_0$.

La statistica test usuale è
$$
W=-2\log\Lambda=2\left[\ell(\hat\theta)-\ell(\hat\theta_0)\right],
$$
dove $\hat\theta$ è l'MLE libero e $\hat\theta_0$ è l'MLE vincolato a $\Theta_0$.

> **Teorema 5.2 (Wilks, asintotico).**  
> Sotto regolarità e sotto $H_0$, se $\Theta_0$ ha dimensione $p_0$ e $\Theta$ ha dimensione $p$, allora
> $$
> W\overset{D}{\longrightarrow}\chi^2_{p-p_0}.
> $$
> Di conseguenza, il test che rifiuta $H_0$ per $W>\chi^2_{p-p_0,1-\alpha}$ ha livello asintotico $\alpha$.

**Idea della dimostrazione.**  
Si sviluppa $\ell$ al secondo ordine intorno all'MLE e si usa la normalità asintotica di $\hat\theta$. Il risultato è una forma quadratica in una normale multidimensionale, la cui distribuzione è $\chi^2$ con gradi di libertà pari alla differenza di dimensioni. $\square$

### Esempio 5.3 — Test nel modello logistico

Nel modello logistico $\pi_i=g^{-1}(\gamma_0+\gamma_1t_i)$, per testare $H_0:\gamma_1=0$ contro $H_1:\gamma_1\neq0$, si calcola:

- $\ell(\hat\gamma_0,\hat\gamma_1)$ con l'MLE libero $(\hat\gamma_0,\hat\gamma_1)$;
- $\ell(\hat\gamma_0^{(0)})$ con l'MLE vincolato sotto $H_0$, cioè $\gamma_1=0$.

La statistica $W=2[\ell(\hat\gamma_0,\hat\gamma_1)-\ell(\hat\gamma_0^{(0)})]$ ha, sotto $H_0$, distribuzione asintotica $\chi^2_1$. Il p-value è $P(\chi^2_1>W_{\text{oss}})$.

Nel caso del Challenger, con i dati completi, si ottiene $\hat\gamma_0=15.043$, $\hat\gamma_1=-0.2322$, e $W$ corrisponde a un p-value $\alpha_{\text{oss}}\approx0.032$. Si rifiuta quindi $H_0$ al livello $5\%$: la temperatura ha un effetto significativo sulla probabilità di guasto. Per $t=31$ °F, $\hat\pi=0.999$: la probabilità stimata di guasto era praticamente $1$.

### Osservazione 5.4 — Dualità con le regioni di verosimiglianza

Le regioni di verosimiglianza e i test del rapporto di verosimiglianza sono duali: fissato $\alpha$, la regione
$$
\Gamma_{1-\alpha}=\{\theta : -2\log L^*(\theta)\le\chi^2_{p,1-\alpha}\}
$$
è un intervallo di confidenza asintotico di livello $1-\alpha$. Il test rifiuta $H_0:\theta=\theta_0$ se e solo se $\theta_0\notin\Gamma_{1-\alpha}$. Lo stesso legame che c'é tra intervalli di confidenza e test bilaterali.

## 6. Proprietà dello stimatore di massima verosimiglianza

### Teorema 6.1 — Invarianza per trasformazioni

Sia $\hat\theta$ l'MLE di $\theta$ e sia $\phi=g(\theta)$ con $g$ biunivoca. Allora l'MLE di $\phi$ è $\hat\phi=g(\hat\theta)$.

**Dimostrazione.**  
La verosimiglianza per $\phi$ è
$$
\tilde L(\phi;\mathbf{y})=L(g^{-1}(\phi);\mathbf{y}).
$$
Poiché $g$ è biunivoca, massimizzare $\tilde L$ su $\phi$ equivale a massimizzare $L$ su $\theta$ e poi applicare $g$. $\square$

**Conseguenza.** Nel modello bernoulliano, l'MLE di $\pi$ è $\hat\pi=\sum_iy_i/n$; l'MLE dell'odds $\pi/(1-\pi)$ è $\hat\pi/(1-\hat\pi)$; l'MLE della log-odds $\log(\pi/(1-\pi))$ è $\log(\hat\pi/(1-\hat\pi))$. Tutte le trasformazioni si ottengono applicando la funzione allo stimatore.

### Teorema 6.2 — Normalità asintotica

Sotto condizioni di regolarità, se $\theta$ è il vero parametro, allora
$$
\sqrt n(\hat\theta-\theta)\overset{D}{\longrightarrow}\mathcal{N}_p(\mathbf{0},\mathcal{I}(\theta)^{-1}),
$$
dove $\mathcal{I}(\theta)$ è la **matrice di informazione di Fisher**
$$
\mathcal{I}(\theta)=-E_\theta\left[\frac{\partial^2\ell}{\partial\theta\partial\theta^T}\right].
$$

> **Osservazione 6.3.**  
> La normalità asintotica dell'MLE permette di costruire intervalli di confidenza di Wald:
> $$
> \hat\theta_j\pm z_{1-\alpha/2}\sqrt{\frac{[\mathcal{I}(\hat\theta)^{-1}]_{jj}}{n}}.
> $$
> In alternativa, si usano gli intervalli basati sulla verosimiglianza (più accurati in campioni finiti), le cui soglie derivano dalla $\chi^2_{p,1-\alpha}$.

### Teorema 6.4 — Consistenza

Sotto regolarità, $\hat\theta\xrightarrow{p}\theta$ quando $n\to\infty$.

**Idea della dimostrazione.**  
Per la legge dei grandi numeri, $\frac1n\ell(\theta;\mathbf{Y})$ converge in probabilità a $E_\theta[\ell(\theta;\mathbf{Y})/n]$. Il massimo di questa funzione asintotica è in $\theta$ (per la disuguaglianza di Jensen applicata al logaritmo), e la convergenza è uniforme in un intorno di $\theta$. Quindi l'MLE converge in probabilità a $\theta$. $\square$

## 7. Perché serve un modello parametrico

### Osservazione 7.1 — Il compromesso

Il paradigma della verosimiglianza ha un **vantaggio** e un **costo**:

- **Vantaggio**: è un principio unico che produce stime, intervalli e test in modo coerente, applicabile a qualunque modello parametrico, anche quando non esistono stimatori "naturali" ricavabili per analogia.
- **Costo**: richiede di specificare un modello statistico parametrico, cioè di assumere una forma per la distribuzione dei dati. Se il modello è sbagliato, le conclusioni possono essere fuorvianti.

Alcune procedure (metodo dei momenti, test asintotici basati sul CLT) funzionavano senza assumere una forma parametrica. Con la verosimiglianza, la forma della distribuzione è parte integrante del modello.

### Osservazione 7.2 — Robustezza e modello sbagliato

Se il modello è **mal specificato**, l'MLE converge al valore del parametro che minimizza la **divergenza di Kullback–Leibler** tra il vero modello e il modello assunto, non necessariamente al vero parametro. Questo è il motivo per cui, in pratica, si usano modelli flessibili, diagnostica dei residui, e verifiche di specificazione.

> **Osservazione 7.3.**  
> Nel caso del Challenger, il modello logistico è una scelta naturale (probabilità legata a un predittore continuo, mappa logistica per garantire l'appartenenza a $(0,1)$, interpretazione dei coefficienti in termini di log-odds). L'analisi condotta con la verosimiglianza avrebbe identificato l'effetto della temperatura e avrebbe stimato una probabilità di guasto quasi $1$ a $31$ °F. La lezione è che la scelta del modello è parte integrante del problema inferenziale, e che **trascurare dati** (come i lanci senza danneggiamento) o **assumere modelli inadeguati** (come il lineare per $\pi_i$) può portare a conclusioni drammaticamente sbagliate.
