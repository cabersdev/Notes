# Proprietà dello stimatore di massima verosimiglianza

---

## 0. Impostazione

### Definizione 0.1 — Stimatore di massima verosimiglianza

Sia $\mathbf{Y}=(Y_1,\dots,Y_n)$ un campione casuale con distribuzione congiunta $p_{\mathbf{Y}}(\mathbf{y};\theta)$, $\theta\in\Theta\subseteq\mathbb{R}^p$. Per ogni realizzazione $\mathbf{y}$, sia
$$
\hat\theta(\mathbf{y})=\arg\max_{\theta\in\Theta}L(\theta;\mathbf{y})
$$
la stima di massima verosimiglianza. Se la funzione $\hat\theta(\cdot)$ esiste finita ed è unica, la variabile casuale
$$
\hat\theta(\mathbf{Y})=\hat\theta(Y_1,\dots,Y_n)
$$
è detta **stimatore di massima verosimiglianza** di $\theta$.

> **Osservazione 0.2.**  
> La distinzione tra **stima** e **stimatore** è cruciale:
> - la stima $\hat\theta(\mathbf{y})$ è un numero (o un vettore), calcolato sui dati osservati;
> - lo stimatore $\hat\theta(\mathbf{Y})$ è una variabile casuale, che dipende dal campione prima di osservarlo.
>
> Le proprietà dello stimatore (distorsione, varianza, consistenza, normalità asintotica) si valutano sulla distribuzione di $\hat\theta(\mathbf{Y})$, indotta dalla distribuzione del campione.

### Osservazione 0.3 — Perché studiare le proprietà

Le proprietà di uno stimatore si valutano secondo il **principio del campionamento ripetuto**: si immagina di estrarre infiniti campioni dalla popolazione e si studia il comportamento dello stimatore al variare del campione. Le proprietà di interesse sono:

1. **sufficienza**: lo stimatore dipende dai dati solo attraverso una statistica sufficiente?
2. **equivarianza**: se si riparametrizza il modello, lo stimatore si trasforma coerentemente?
3. **consistenza**: al crescere di $n$, lo stimatore converge al vero valore del parametro?
4. **distribuzione asintotica**: qual è la distribuzione limite dell'errore di stima? Serve per costruire intervalli di confidenza e test.
5. **efficienza**: la varianza asintotica è la più piccola possibile?

I risultati generali si ottengono sotto ipotesi di **regolarità** e per campioni i.i.d. con $n\to\infty$.

## 1. Sufficienza

### Proposizione 1.1 — La SMV è funzione di una statistica sufficiente

Sia $\mathbf{Y}$ un campione dal modello $\mathcal{F}=\{p_{\mathbf{Y}}(\cdot;\theta):\theta\in\Theta\}$ e sia $T(\mathbf{Y})$ una statistica sufficiente per $\theta$. Allora la SMV $\hat\theta(\mathbf{Y})$ è funzione di $T(\mathbf{Y})$, ed è quindi essa stessa una statistica sufficiente.

**Dimostrazione.**  
Per il teorema di fattorizzazione di Neyman–Fisher, se $T$ è sufficiente, allora
$$
L(\theta;\mathbf{y})=h(\mathbf{y})\,g(T(\mathbf{y});\theta),
$$
dove $h$ non dipende da $\theta$. La massimizzazione di $L$ rispetto a $\theta$ dipende da $\mathbf{y}$ solo attraverso $T(\mathbf{y})$, perché il fattore $h(\mathbf{y})$ è una costante positiva rispetto a $\theta$ e non influenza il punto di massimo. Quindi
$$
\hat\theta(\mathbf{y})=\arg\max_\theta g(T(\mathbf{y});\theta),
$$
cioè $\hat\theta$ è funzione di $T(\mathbf{y})$. Inoltre, essendo funzione di una statistica sufficiente, $\hat\theta(\mathbf{Y})$ è anch'esso sufficiente (perché ogni funzione di una statistica sufficiente è sufficiente). $\square$

> **Osservazione 1.2.**  
> La proposizione afferma che la SMV **non perde informazione**: essa sintetizza i dati in una statistica sufficiente, e quindi contiene tutta l'informazione su $\theta$ contenuta nel campione. In particolare, se $T$ è la statistica sufficiente minimale, la SMV è una funzione di $T$; non può essere "più fine" di $T$, ma può essere una sua trasformazione.

**Esempio 1.3 — Bernoulli.**  
Per $Y_1,\dots,Y_n$ i.i.d. $\operatorname{Be}(\pi)$, la statistica sufficiente minimale è $T(\mathbf{Y})=\sum_i Y_i$. La SMV è
$$
\hat\pi=\frac{\sum_i Y_i}{n}=\frac{T(\mathbf{Y})}{n},
$$
che è funzione di $T$. Questo è coerente con la Proposizione 1.1.

## 2. Equivarianza rispetto a riparametrizzazioni

### Definizione 2.1 — Riparametrizzazione

Sia $h:\Theta\to\Psi$ una funzione **biunivoca**, con $\Psi=\{h(\theta):\theta\in\Theta\}$. Il modello può essere riscritto equivalentemente come
$$
\mathcal{F}_\psi=\{p_{\mathbf{Y}}(\cdot;\psi):\psi\in\Psi\},
$$
dove $p_{\mathbf{Y}}(\cdot;\psi)=p_{\mathbf{Y}}(\cdot;h^{-1}(\psi))$. Le due parametrizzazioni sono equivalenti, e la scelta è di convenienza.

### Teorema 2.2 — Equivarianza della SMV

Sia $\hat\theta$ la SMV di $\theta$ sotto $\mathcal{F}_\theta$ e sia $\hat\psi$ la SMV di $\psi$ sotto $\mathcal{F}_\psi$. Allora
$$
\hat\psi=h(\hat\theta).
$$

**Dimostrazione.**  
Per costruzione, i due modelli sono equivalenti:
$$
p_{\mathbf{Y}}(\mathbf{y};\theta)=p_{\mathbf{Y}}(\mathbf{y};h^{-1}(\psi))=p_{\mathbf{Y}}(\mathbf{y};\psi)
\qquad\forall\mathbf{y}.
$$
Quindi, per ogni $\theta$ e $\psi=h(\theta)$,
$$
L_\theta(\theta;\mathbf{y})=L_\psi(\psi;\mathbf{y}).
$$
Poiché l'uguaglianza vale per ogni $\theta$, vale anche nei punti di massimo:
$$
L_\theta(\hat\theta;\mathbf{y})=\max_\theta L_\theta(\theta;\mathbf{y})=\max_\psi L_\psi(\psi;\mathbf{y})=L_\psi(\hat\psi;\mathbf{y}).
$$
Se $h$ è biunivoca, il massimo di $L_\psi$ si ottiene in $\psi=h(\hat\theta)$. Quindi $\hat\psi=h(\hat\theta)$. $\square$

> **Osservazione 2.3.**  
> L'equivarianza è una proprietà **forte**: se si cambia parametrizzazione, la SMV si trasforma in modo coerente. Non è necessario ricalcolare la SMV da zero: basta applicare $h$ alla SMV della parametrizzazione originale. Questo è uno dei motivi per cui la verosimiglianza è un principio inferenziale "naturale": le conclusioni non dipendono dalla scelta della parametrizzazione.

### Esempio 2.4 — Esponenziale

Sia $Y_1,\dots,Y_n$ i.i.d. $\operatorname{Esp}(\lambda)$, con densità $p_Y(y;\lambda)=\lambda e^{-\lambda y}$. La SMV del tasso $\lambda$ è
$$
\hat\lambda=\frac{1}{\bar y}.
$$
Se si riparametrizza in termini della media $\mu=1/\lambda$, la densità diventa $p_Y(y;\mu)=\frac1\mu e^{-y/\mu}$, e la SMV di $\mu$ è
$$
\hat\mu=\bar y.
$$
Coerentemente con l'equivarianza, $\hat\mu=h(\hat\lambda)=1/\hat\lambda=\bar y$.

### Proposizione 2.5 — Trasformazione delle quantità connesse

Sotto l'ipotesi che $h$ e $g=h^{-1}$ siano due volte derivabili, le quantità connesse alla verosimiglianza si trasformano come segue:

1. **funzione punteggio**:
$$
\ell^*_\psi(\psi)=\ell^*_\theta(g(\psi))\,g'(\psi);
$$
2. **derivata seconda**:
$$
\ell^{**}_\psi(\psi)=\ell^{**}_\theta(g(\psi))\,[g'(\psi)]^2+\ell^*_\theta(g(\psi))\,g''(\psi);
$$
3. **informazione osservata**:
$$
J_\psi(\psi)=J_\theta(g(\psi))\,[g'(\psi)]^2-\ell^*_\theta(g(\psi))\,g''(\psi);
$$
4. **informazione osservata in $\hat\psi$**: poiché $\ell^*_\theta(\hat\theta)=0$,
$$
J_\psi(\hat\psi)=J_\theta(\hat\theta)\,[g'(\hat\psi)]^2;
$$
5. **informazione attesa**:
$$
I_\psi(\psi)=I_\theta(g(\psi))\,[g'(\psi)]^2.
$$

**Dimostrazione.**  
Sono applicazioni della regola della catena. Per la funzione punteggio:
$$
\ell^*_\psi(\psi)=\frac{d}{d\psi}\ell(g(\psi))=\ell'_\theta(g(\psi))\,g'(\psi).
$$
Derivando ancora:
$$
\ell^{**}_\psi(\psi)=\frac{d}{d\psi}\left[\ell'_\theta(g(\psi))\,g'(\psi)\right]=\ell''_\theta(g(\psi))\,[g'(\psi)]^2+\ell'_\theta(g(\psi))\,g''(\psi).
$$
Le altre formule seguono per definizione di $J$ e $I$ e per il fatto che $\ell'_\theta(\hat\theta)=0$. $\square$

> **Osservazione 2.6.**  
> L'informazione attesa **non è invariante** per riparametrizzazione: si trasforma con il quadrato della derivata di $g$. Questo è coerente con il fatto che l'informazione è una **metrica** sullo spazio parametrico: cambiando coordinate, la metrica cambia secondo il fattore di scala $[g'(\psi)]^2$. L'informazione osservata $J(\hat\psi)$ si trasforma invece con $[g'(\hat\psi)]^2$, perché il termine con $\ell'$ svanisce nel punto di massimo.

**Esempio 2.7 — Esponenziale (continuazione).**  
Con $\mu=1/\lambda$, $g(\mu)=1/\mu$ e $g'(\mu)=-1/\mu^2$. Per $n=1$:
$$
\ell_\lambda(\lambda)=\log\lambda-\lambda y,\qquad \ell^*_\lambda(\lambda)=\frac1\lambda-y.
$$
Per $\mu$:
$$
\ell_\mu(\mu)=-\log\mu-\frac{y}{\mu},\qquad \ell^*_\mu(\mu)=-\frac1\mu+\frac{y}{\mu^2}.
$$
Verifica della formula di trasformazione:
$$
\ell^*_\mu(\mu)=\ell^*_\lambda(g(\mu))\,g'(\mu)=\left(\mu-y\right)\left(-\frac1{\mu^2}\right)=-\frac1\mu+\frac{y}{\mu^2}.
$$
Coerente. Inoltre $\hat\mu=h(\hat\lambda)=1/\hat\lambda=y$. Per l'informazione attesa:
$$
I_\lambda(\lambda)=\frac{1}{\lambda^2},\qquad I_\mu(\mu)=I_\lambda(g(\mu))\,[g'(\mu)]^2=\frac{1}{(1/\mu)^2}\cdot\frac{1}{\mu^4}=\frac{1}{\mu^2}.
$$

## 3. Consistenza

### Teorema 3.1 — Consistenza della SMV

Sia $\mathcal{F}_\theta=\{p_{\mathbf{Y}}(\cdot;\theta):\theta\in\Theta\}$ un modello con funzioni di densità (o probabilità) che hanno **tutte lo stesso supporto**. Allora lo stimatore di massima verosimiglianza $\hat\theta_n=\hat\theta(Y_1,\dots,Y_n)$ è **consistente** sia in senso debole che in senso forte:
$$
\hat\theta_n\xrightarrow{p}\theta_0,
\qquad
\hat\theta_n\xrightarrow{\text{q.c.}}\theta_0,
$$
dove $\theta_0$ è il vero valore del parametro.

**Idea della dimostrazione.**  
La dimostrazione si articola in tre passi:

1. **Massimalità della SMV.** Per costruzione,
$$
\ell(\hat\theta_n;\mathbf{y})\ge\ell(\theta;\mathbf{y})\qquad\forall\theta\in\Theta.
$$

2. **Disuguaglianza di Wald.** Per la disuguaglianza di Wald, per ogni $\theta\ne\theta_0$,
$$
E_{\theta_0}[\ell(\theta_0;\mathbf{Y})]>E_{\theta_0}[\ell(\theta;\mathbf{Y})].
$$
Quindi, in media, la log-verosimiglianza è massima in $\theta_0$.

3. **Legge dei grandi numeri.** Poiché $\ell(\theta;\mathbf{Y})=\sum_i\ell_i(\theta;Y_i)$ è una somma di variabili i.i.d., per la legge dei grandi numeri
$$
\frac{\ell(\theta;\mathbf{Y})}{n}\xrightarrow{\text{q.c.}}E_{\theta_0}[\ell_1(\theta;Y_1)].
$$
Combinando con la disuguaglianza di Wald, per ogni $\theta\ne\theta_0$,
$$
\frac{\ell(\theta;\mathbf{Y})-\ell(\theta_0;\mathbf{Y})}{n}\xrightarrow{\text{q.c.}}E_{\theta_0}[\ell_1(\theta;Y_1)-\ell_1(\theta_0;Y_1)]<0.
$$
Quindi, per $n$ grande, $\ell(\theta;\mathbf{Y})<\ell(\theta_0;\mathbf{Y})$ q.c. per ogni $\theta\ne\theta_0$ in un intorno. Poiché $\hat\theta_n$ massimizza $\ell$, deve essere vicino a $\theta_0$. Un argomento di compattezza (o di uniformità in $\theta$) conclude che $\hat\theta_n\to\theta_0$ q.c. e in probabilità. $\square$

> **Osservazione 3.2.**  
> La consistenza è una proprietà **asintotica**: garantisce che, al crescere della dimensione campionaria, la SMV si avvicina al vero valore del parametro. Non dice nulla sulla velocità di convergenza né sulla distribuzione dell'errore a $n$ finito: per questo servono i risultati sulla distribuzione asintotica (Sezione 4).

> **Osservazione 3.3.**  
> La consistenza **non richiede** che la verosimiglianza sia regolare né che la SMV sia unica. Ad esempio, per il modello $U(0,\theta)$, la SMV è $\hat\theta=Y_{(n)}$, che è consistente (converge a $\theta$ q.c.) anche se il problema non è regolare. La consistenza è una proprietà più robusta della normalità asintotica.

## 4. Distribuzione asintotica

### Teorema 4.1 — Normalità asintotica della SMV

Sotto le condizioni di regolarità del problema di stima, lo stimatore di massima verosimiglianza $\hat\theta_n$ ha distribuzione asintoticamente normale:
$$
\sqrt{n}(\hat\theta_n-\theta_0)\xrightarrow{D}\mathcal{N}\left(0,\,i(\theta_0)^{-1}\right),
$$
dove $i(\theta_0)$ è l'informazione attesa di **una singola osservazione**. Equivalentemente,
$$
\hat\theta_n\overset{\text{approx}}{\sim}\mathcal{N}\left(\theta_0,\,\frac{1}{n\,i(\theta_0)}\right)=\mathcal{N}\left(\theta_0,\,\frac{1}{I(\theta_0)}\right),
$$
dove $I(\theta_0)=n\,i(\theta_0)$ è l'informazione attesa totale.

**Dimostrazione (traccia).**

**Passo 1 — Sviluppo di Taylor della funzione punteggio.**  
Poiché $\hat\theta_n$ è consistente, è vicino a $\theta_0$ per $n$ grande. Sviluppando $\ell^*(\theta)$ in serie di Taylor al primo ordine intorno a $\theta_0$ e valutando in $\hat\theta_n$:
$$
0=\ell^*(\hat\theta_n)\approx\ell^*(\theta_0)+(\hat\theta_n-\theta_0)\,\ell^{**}(\theta_0).
$$
Quindi
$$
\hat\theta_n-\theta_0\approx-\frac{\ell^*(\theta_0)}{\ell^{**}(\theta_0)}=\frac{\ell^*(\theta_0)}{J(\theta_0)}.
$$

**Passo 2 — Comportamento del numeratore.**  
La funzione punteggio è $\ell^*(\theta_0;\mathbf{Y})=\sum_i\ell_i^*(\theta_0;Y_i)$. Per le identità di Bartlett, $E_{\theta_0}[\ell_i^*(\theta_0;Y_i)]=0$ e $\operatorname{Var}_{\theta_0}(\ell_i^*(\theta_0;Y_i))=i(\theta_0)$. Per il teorema del limite centrale,
$$
\frac{\ell^*(\theta_0;\mathbf{Y})}{\sqrt{n\,i(\theta_0)}}\xrightarrow{D}\mathcal{N}(0,1).
$$

**Passo 3 — Comportamento del denominatore.**  
L'informazione osservata è $J(\theta_0;\mathbf{Y})=\sum_i J_i(\theta_0;Y_i)$. Per la legge dei grandi numeri,
$$
\frac{J(\theta_0;\mathbf{Y})}{n}=\frac{1}{n}\sum_i J_i(\theta_0;Y_i)\xrightarrow{p}i(\theta_0),
$$
quindi $J(\theta_0;\mathbf{Y})\xrightarrow{p}I(\theta_0)=n\,i(\theta_0)$.

**Passo 4 — Slutsky.**  
Combinando i passi 2 e 3 con il teorema di Slutsky:
$$
\sqrt{n}(\hat\theta_n-\theta_0)\approx\frac{\ell^*(\theta_0;\mathbf{Y})/\sqrt{n}}{J(\theta_0;\mathbf{Y})/n}\xrightarrow{D}\frac{\mathcal{N}(0,i(\theta_0))}{i(\theta_0)}=\mathcal{N}\left(0,\frac{1}{i(\theta_0)}\right).
$$
$\square$

> **Osservazione 4.2.**  
> La normalità asintotica è il risultato **centrale** dell'inferenza di verosimiglianza: permette di costruire intervalli di confidenza e test di ipotesi approssimati per $n$ grande, senza conoscere la distribuzione esatta di $\hat\theta_n$. La varianza asintotica è $1/I(\theta_0)$, dove $I(\theta_0)$ è l'informazione attesa totale. Più informazione c'è, più la stima è precisa.

> **Osservazione 4.3.**  
> In pratica, $I(\theta_0)$ è ignota (dipende dal vero $\theta_0$). Si sostituisce con una stima consistente, ad esempio $I(\hat\theta_n)$ o $J(\hat\theta_n;\mathbf{Y})$. Per Slutsky, la normalità asintotica continua a valere. Questo è il fondamento degli **intervalli di confidenza di Wald**:
> $$
> \hat\theta_n\pm z_{1-\alpha/2}\sqrt{\frac{1}{I(\hat\theta_n)}}.
> $$

**Esempio 4.4 — Bernoulli.**  
Per $Y_1,\dots,Y_n$ i.i.d. $\operatorname{Be}(\pi)$, la SMV è $\hat\pi_n=\bar Y_n$. L'informazione attesa di una singola osservazione è
$$
i(\pi)=\frac{1}{\pi(1-\pi)},
$$
quindi $I(\pi)=n/[\pi(1-\pi)]$. Per il Teorema 4.1,
$$
\sqrt{n}(\hat\pi_n-\pi_0)\xrightarrow{D}\mathcal{N}(0,\pi_0(1-\pi_0)).
$$
Questo è il classico risultato del teorema del limite centrale per la proporzione campionaria, ottenuto qui come caso particolare della teoria asintotica della SMV.

## 5. Efficienza

### Definizione 5.1 — Efficienza assoluta (Cramér–Rao)

Sia $\hat\theta$ uno stimatore non distorto di $\theta$. Si dice che $\hat\theta$ è **efficiente** se raggiunge il limite inferiore di Cramér–Rao:
$$
\operatorname{Var}(\hat\theta)=\frac{1}{I(\theta)}.
$$
Il limite di Cramér–Rao afferma che, per ogni stimatore non distorto $T$ di $\theta$,
$$
\operatorname{Var}(T)\ge\frac{1}{I(\theta)}.
$$
La disuguaglianza si dimostra con Cauchy–Schwarz applicata a $\ell^*(\theta;\mathbf{Y})$ e $T(\mathbf{Y})$, usando le identità di Bartlett.

### Teorema 5.2 — Efficienza asintotica della SMV

Sotto le condizioni di regolarità, la SMV è **asintoticamente efficiente**:
$$
\operatorname{Var}(\hat\theta_n)\sim\frac{1}{I(\theta_0)}
\qquad(n\to\infty).
$$

**Dimostrazione.**  
Dal Teorema 4.1, $\sqrt{n}(\hat\theta_n-\theta_0)\xrightarrow{D}\mathcal{N}(0,i(\theta_0)^{-1})$. Quindi
$$
\operatorname{Var}(\hat\theta_n)\approx\frac{1}{n\,i(\theta_0)}=\frac{1}{I(\theta_0)},
$$
che è esattamente il limite inferiore di Cramér–Rao. Inoltre, la SMV è asintoticamente non distorta: $E[\hat\theta_n]\to\theta_0$. Quindi è asintoticamente efficiente. $\square$

> **Osservazione 5.3.**  
> L'efficienza asintotica è una proprietà **forte**: tra tutti gli stimatori asintoticamente normali e asintoticamente non distorti, la SMV ha la varianza asintotica più piccola possibile. Questo è il senso in cui la SMV è "ottimale": non si può fare meglio, asintoticamente, in termini di varianza.

### Corollario 5.4 — Proprietà BAN

Sotto le condizioni di regolarità, lo stimatore di massima verosimiglianza è **BAN** (Best Asymptotically Normal):
- **consistente**: $\hat\theta_n\to\theta_0$ in probabilità (e q.c.);
- **asintoticamente non distorto**: $E[\hat\theta_n]\to\theta_0$;
- **asintoticamente efficiente**: $\operatorname{Var}(\hat\theta_n)\sim 1/I(\theta_0)$;
- **asintoticamente normale**: $\sqrt{n}(\hat\theta_n-\theta_0)\xrightarrow{D}\mathcal{N}(0,i(\theta_0)^{-1})$.

In sintesi: è il **migliore** stimatore asintoticamente normale, in termini di efficienza.
