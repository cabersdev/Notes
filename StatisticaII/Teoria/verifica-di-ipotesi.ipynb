# Verifica di ipotesi

---

## 1. Concetti fondamentali

### Definizione 1.1 — Ipotesi statistica

Sia $Y\sim p_Y(y;\theta)$ con $\theta\in\Theta$. Una **ipotesi statistica** è un’affermazione o congettura riguardante $\theta$, cioè un sottoinsieme $\Theta_0\subseteq\Theta$.

### Definizione 1.2 — Ipotesi nulla e alternativa

- L’**ipotesi nulla** $H_0$ è un enunciato specifico su $\theta$, tipicamente del tipo $\theta=\theta_0$ oppure $\theta\le\theta_0$ oppure $\theta\ge\theta_0$.
- L’**ipotesi alternativa** $H_1$ è l’insieme delle alternative a $H_0$ non specificate da $H_0$.

Formalmente, se $H_0$ corrisponde a $\Theta_0\subseteq\Theta$, allora $H_1$ corrisponde a $\Theta_1=\Theta\setminus\Theta_0$.

> **Osservazione 1.3.**  
> $H_0$ è tipicamente l’ipotesi “conservativa”, quella che, se vera, non porta a scoperte rilevanti. $H_1$ è l’ipotesi che ha motivato l’esperimento.

### Definizione 1.4 — Ipotesi semplice e composta

- Un’ipotesi è **semplice** se specifica completamente la distribuzione $p_Y(y;\theta)$, cioè se $\Theta_0$ contiene un solo valore di $\theta$. Cioè $|\Theta_0| = 1$
- Un’ipotesi è **composta** se $\Theta_0$ contiene più di un valore di $\theta$. Cioè $|\Theta_0| > 1$

### Definizione 1.5 — Ipotesi unilaterale e bilaterale

- Un’ipotesi composta è **unilaterale** se riguarda valori di $\theta$ limitati in una direzione, ad esempio $\theta>\theta_0$ oppure $\theta<\theta_0$.
- È **bilaterale** se riguarda valori di $\theta$ da entrambe le direzioni, ad esempio $\theta\neq\theta_0$.

### Esempio

Un'azienda che produce lastre di metallo vuole verificare che il processo produttivo sia sotto controllo, ovvero che lo spessore medio $\mu$ delle lastre prodotte sia pari a quello nominale, pari a $14$ mm

Si ha che:

$$
H_0 : \mu = 14, \quad H_1 : \mu \neq 14
$$

### Esempio

In occasione di un referendum abrogativo, si vuole fare una previsione circa il raggiungimento del quorum (che si raggiunge quando la proporzione $\pi$ dei votanti super la metà degli aventi diritto)

$$
H_0: \pi \leq 0.5, \quad H_1: \pi > 0.5
$$

### Esempio
Siano $\mu_1$ e $\mu_2$ la pressione sistolica media degli individui ipertesi trattati e non trattati con un nuovo farmaco. Si vuole valutare se il nuovo farmaco è efficace nel ridurre la pressione sanguigna

$$
H_0: \mu_1 \geq \mu_2, \quad H_1: \mu_1 < \mu_2
$$

>**Osservazione:** In questo caso sono due le popolazioni di interesse

### Definizione 1.6 — Test statistico

Un **test statistico** è una regola, basata sull’osservazione del campione, che permette di decidere se rifiutare o non rifiutare $H_0$.

Formalmente, un test statistico è una funzione misurabile

$$
\phi:\mathbb{R}^n\to\{0,1\},
$$

dove:

- $\phi(\mathbf{y})=1$ significa “rifiutare $H_0$”;
- $\phi(\mathbf{y})=0$ significa “non rifiutare $H_0$”.

Equivalentemente, il test individua una **regione di rifiuto** (o regione critica)

$$
R=\{\mathbf{y}\in\mathbb{R}^n : \phi(\mathbf{y})=1\}\subseteq\mathbb{R}^n,
$$

e una **regione di non rifiuto (o regione di accettazione)** $R^C$ o $\overline{R}$. Formalmente:

$$
\overline{R} = R^C = \{\mathbf{y} \in \mathbb{R}^n : \phi(\mathbf{y}) = 0\}\subseteq\mathbb{R}^n
$$

### Definizione 1.7 — Errori di I e II tipo

Poiché la decisione di accettazione o rifiuto dell'ipotesi $H_0$ sarà presa sulla base di un campione potrà essere soggetta ad errore.

Quando si esegue un test, si possono verificare quattro situazioni.

- Se $H_0$ è vera e non si rifiuta $H_0$: decisione corretta.
- Se $H_0$ è vera e si rifiuta $H_0$: **errore di I tipo**, con probabilità $\alpha$.
- Se $H_1$ è vera e non si rifiuta $H_0$: **errore di II tipo**, con probabilità $\beta$.
- Se $H_1$ è vera e si rifiuta $H_0$: decisione corretta.

Formalmente:

$$
\alpha=P_\theta(\text{rifiutare }H_0\mid H_0)=P_\theta(\mathbf{Y}\in R\mid \theta\in\Theta_0),
$$

$$
\beta=P_\theta(\text{non rifiutare }H_0\mid H_1)=P_\theta(\mathbf{Y}\in R^C\mid \theta\in\Theta_1).
$$

$\alpha$ rappresenta quindi la probabilità di commettere un errore di $I$ tipo (ovvero rifiutare $H_0$ quando questa è vera), anche detto **Livello di significatività** del test

$\beta$ rappresenta la probabilità di commettere un errore di $II$ tipo (ovvero "accettare" $H_0$ quando questa è falsa)

### Definizione 1.8 — Livello di significatività

Il **livello di significatività** del test è

$$
\alpha=\sup_{\theta\in\Theta_0}P_\theta(\mathbf{Y}\in R).
$$

Si dice che il test ha livello $\alpha$ se la probabilità di errore di I tipo è al più $\alpha$ per ogni $\theta\in\Theta_0$.

> **Osservazione 1.9.**  
> Fissata la numerosità campionaria $n$, al decrescere di $\alpha$ aumenta $\beta$ e viceversa. Non è possibile minimizzare simultaneamente entrambi gli errori. La prassi è fissare $\alpha$ piccolo, tipicamente $0.05$ o $0.01$, e, tra i test di livello $\alpha$, scegliere quello con $\beta$ minimo, equivalentemente con potenza massima.

### Definizione 1.10 — p-value (livello di significatività osservato)

Sia $T_n$ una statistica test e sia $t_{\text{oss}}=T_n(\mathbf{y})$ il valore osservato. Il **p-value** è la probabilità, calcolata sotto $H_0$, che la statistica test assuma un valore “più estremo” di quello osservato:

$$
p=P_{H_0}(T_n\text{ è più estremo di }t_{\text{oss}}).
$$

Per un test unilaterale destro:

$$
p=P_{H_0}(T_n\ge t_{\text{oss}}).
$$

Per un test unilaterale sinistro:

$$
p=P_{H_0}(T_n\le t_{\text{oss}}).
$$

Per un test bilaterale, se la distribuzione di $T_n$ sotto $H_0$ è simmetrica:

$$
p=2\min\{P_{H_0}(T_n\le t_{\text{oss}}),\,P_{H_0}(T_n\ge t_{\text{oss}})\}.
$$

**Regola di decisione.**

- Se $p<\alpha$, si rifiuta $H_0$ al livello $\alpha$.
- Se $p\ge\alpha$, non si rifiuta $H_0$.

> **Osservazione 1.11.**  
> Il p-value è una misura dell’evidenza empirica contro $H_0$: più piccolo è $p$, più forte è l’evidenza contro $H_0$.

### Definizione 1.12 — Potenza del test

La **potenza** del test è

$$
\pi(\theta)=P_\theta(\mathbf{Y}\in R)=1-\beta(\theta),\qquad \theta\in\Theta_1.
$$

La **funzione di potenza** $\pi:\Theta\to[0,1]$ è definita su tutto $\Theta$ come

$$
\pi(\theta)=P_\theta(\mathbf{Y}\in R).
$$

Per $\theta\in\Theta_0$, $\pi(\theta)\le\alpha$. Per $\theta\in\Theta_1$, $\pi(\theta)$ misura la capacità del test di rilevare che $H_0$ è falsa.

### Teorema 1.13 — Dualità tra intervalli di confidenza e test

Sia $\operatorname{IC}_{1-\alpha}(\theta)$ un intervallo di confidenza di livello $1-\alpha$ per $\theta$. Allora, per ogni $\theta_0\in\Theta$, il test che rifiuta $H_0:\theta=\theta_0$ contro $H_1:\theta\neq\theta_0$ se e solo se $\theta_0\notin\operatorname{IC}_{1-\alpha}(\theta)$ ha livello $\alpha$.

**Dimostrazione.**  
Per definizione di intervallo di confidenza,

$$
P_\theta\bigl(\theta\in\operatorname{IC}_{1-\alpha}(\theta)\bigr)=1-\alpha.
$$

Allora

$$
\begin{aligned}
P_\theta\bigl(\theta_0\notin\operatorname{IC}_{1-\alpha}(\theta)\mid \theta=\theta_0\bigr)
&=
1-P_{\theta_0}\bigl(\theta_0\in\operatorname{IC}_{1-\alpha}(\theta)\bigr) \\
&=
1-(1-\alpha) \\
&=
\alpha.
\end{aligned}
$$

Quindi il test ha livello $\alpha$. $\square$

> **Osservazione 1.14.**  
> La dualità non è sempre valida per test unilaterali, ma è possibile stabilire corrispondenze analoghe con intervalli unilaterali.

## 2. Test per la media di una popolazione normale con varianza nota

Sia $Y\sim\mathcal{N}(\mu,\sigma^2)$ con $\sigma^2$ nota, e sia $Y_1,\dots,Y_n$ un campione i.i.d.

### 2.1 Test bilaterale

**Ipotesi:**

$$
H_0:\mu=\mu_0,\qquad H_1:\mu\neq\mu_0.
$$

**Statistica test.** Poiché $\overline{Y}\sim\mathcal{N}(\mu,\sigma^2/n)$, sotto $H_0$



$$
Z=\frac{\overline{Y}-\mu_0}{\sigma/\sqrt{n}}\sim\mathcal{N}(0,1).
$$

**Regione di rifiuto.** Per un test di livello $\alpha$:

$$
R=\left\{\mathbf{y} : |Z|>z_{1-\alpha/2}\right\}
$$

cioè

$$
R=
\left\{
\mathbf{y} :
\overline{y}<\mu_0-z_{1-\alpha/2}\frac{\sigma}{\sqrt{n}}
\ \text{oppure}\
\overline{y}>\mu_0+z_{1-\alpha/2}\frac{\sigma}{\sqrt{n}}
\right\}.
$$

**p-value.**

$$
p=P(|Z|>|z_{\text{oss}}|)=2(1-\Phi(|z_{\text{oss}}|)).
$$

**Dimostrazione della distribuzione.**  
Sotto $H_0$, $\overline{Y}\sim\mathcal{N}(\mu_0,\sigma^2/n)$. La standardizzazione

$$
Z=\frac{\overline{Y}-\mu_0}{\sigma/\sqrt{n}}
$$

ha media $0$ e varianza $1$, quindi $Z\sim\mathcal{N}(0,1)$. $\square$

### 2.2 Test unilaterale destro

**Ipotesi:**

$$
H_0:\mu\le\mu_0,\qquad H_1:\mu>\mu_0.
$$

**Statistica test.**

$$
Z=\frac{\overline{Y}-\mu_0}{\sigma/\sqrt{n}}.
$$

**Regione di rifiuto.**

$$
R=\left\{\mathbf{y} : Z>z_{1-\alpha}\right\}
=
\left\{\mathbf{y} : \overline{y}>\mu_0+z_{1-\alpha}\frac{\sigma}{\sqrt{n}}\right\}.
$$

**p-value.**

$$
p=P(Z>z_{\text{oss}})=1-\Phi(z_{\text{oss}}).
$$

> **Osservazione 2.1.**  
> Nel caso $H_0:\mu\le\mu_0$, il caso più sfavorevole è $\mu=\mu_0$, perché è il valore più vicino all’alternativa. Quindi la probabilità di errore di I tipo è massima in $\mu=\mu_0$ e vale $\alpha$.

### 2.3 Test unilaterale sinistro

**Ipotesi:**

$$
H_0:\mu\ge\mu_0,\qquad H_1:\mu<\mu_0.
$$

**Regione di rifiuto.**

$$
R=\left\{\mathbf{y} : Z<-z_{1-\alpha}\right\}
=
\left\{\mathbf{y} : \overline{y}<\mu_0-z_{1-\alpha}\frac{\sigma}{\sqrt{n}}\right\}.
$$

**p-value.**

$$
p=P(Z<z_{\text{oss}})=\Phi(z_{\text{oss}}).
$$

### Esempio 2.2 — Lastre di metallo

Sia $Y\sim\mathcal{N}(\mu,0.1)$, $n=5$, $\overline{y}=14.302$, $\mu_0=14$, $\alpha=0.05$.

Test bilaterale:

$$
z_{\text{oss}}=\frac{14.302-14}{\sqrt{0.1/5}}=\frac{0.302}{0.1414}\approx 2.135.
$$

Poiché $|z_{\text{oss}}|>z_{0.975}=1.96$, si rifiuta $H_0$. Il p-value è

$$
p=2(1-\Phi(2.135))\approx 2\cdot 0.0164=0.0328<\alpha.
$$

Test unilaterale destro, con $H_1:\mu>14$:

$$
p=P(Z>2.135)\approx 0.0164<\alpha,
$$

quindi si rifiuta $H_0$.

## 3. Test per la varianza di una popolazione normale

Sia $Y\sim\mathcal{N}(\mu,\sigma^2)$ con $\mu$ e $\sigma^2$ non note, e sia $Y_1,\dots,Y_n$ un campione i.i.d.

### 3.1 Test unilaterale destro

**Ipotesi:**

$$
H_0:\sigma^2\le\sigma_0^2,\qquad H_1:\sigma^2>\sigma_0^2.
$$

**Statistica test.** Sotto $H_0$,

$$
\chi^2=\frac{(n-1)S^2}{\sigma_0^2}\sim\chi^2_{n-1}.
$$

**Regione di rifiuto.**

$$
R=\left\{\mathbf{y} : \chi^2>\chi^2_{n-1,1-\alpha}\right\}
=
\left\{\mathbf{y} : S^2>\frac{\chi^2_{n-1,1-\alpha}\sigma_0^2}{n-1}\right\}.
$$

**p-value.**

$$
p=P(\chi^2_{n-1}>\chi^2_{\text{oss}}).
$$

**Dimostrazione della distribuzione.**  
Se $Y_1,\dots,Y_n$ sono i.i.d. $\mathcal{N}(\mu,\sigma^2)$, allora

$$
\frac{(n-1)S^2}{\sigma^2}\sim\chi^2_{n-1}.
$$

Sotto $H_0$, $\sigma^2=\sigma_0^2$, quindi la statistica test ha distribuzione $\chi^2_{n-1}$. $\square$

### 3.2 Test bilaterale

**Ipotesi:**

$$
H_0:\sigma^2=\sigma_0^2,\qquad H_1:\sigma^2\neq\sigma_0^2.
$$

**Regione di rifiuto.**

$$
R=
\left\{
\mathbf{y} :
\chi^2<\chi^2_{n-1,\alpha/2}
\ \text{oppure}\
\chi^2>\chi^2_{n-1,1-\alpha/2}
\right\}.
$$

**p-value.** Poiché la distribuzione $\chi^2$ è asimmetrica:

$$
p=2\min\{P(\chi^2_{n-1}<\chi^2_{\text{oss}}),\,P(\chi^2_{n-1}>\chi^2_{\text{oss}})\}.
$$

### Esempio 3.3 — Peso delle confezioni

Sia $Y\sim\mathcal{N}(\mu,\sigma^2)$, $n=5$, campione $98,103,96,100,109$. Si ha $\overline{y}=101.2$, $S^2=25.7$. Test:

$$
H_0:\sigma^2\le 9,\qquad H_1:\sigma^2>9.
$$

Statistica test:

$$
\chi^2_{\text{oss}}=\frac{(n-1)S^2}{\sigma_0^2}=\frac{4\cdot 25.7}{9}\approx 11.42.
$$

Per $\alpha=0.01$, $\chi^2_{4,0.99}=13.28$. Poiché $11.42<13.28$, non si rifiuta $H_0$. Il processo è sotto controllo.

## 4. Test per la media di una popolazione normale con varianza non nota

Sia $Y\sim\mathcal{N}(\mu,\sigma^2)$ con $\mu$ e $\sigma^2$ non note, e sia $Y_1,\dots,Y_n$ un campione i.i.d.

### 4.1 Test bilaterale

**Ipotesi:**

$$
H_0:\mu=\mu_0,\qquad H_1:\mu\neq\mu_0.
$$

**Statistica test.** Sotto $H_0$,

$$
T=\frac{\overline{Y}-\mu_0}{S/\sqrt{n}}\sim t_{n-1}.
$$

**Regione di rifiuto.**

$$
R=\left\{\mathbf{y} : |T|>t_{n-1,1-\alpha/2}\right\}.
$$

**p-value.**

$$
p=P(|t_{n-1}|>|t_{\text{oss}}|)=2P(t_{n-1}>|t_{\text{oss}}|).
$$

**Dimostrazione della distribuzione.**  
Se $Y_i\sim\mathcal{N}(\mu,\sigma^2)$ i.i.d., allora $\overline{Y}$ e $S^2$ sono indipendenti, e

$$
\frac{\overline{Y}-\mu}{\sigma/\sqrt{n}}\sim\mathcal{N}(0,1),
\qquad
\frac{(n-1)S^2}{\sigma^2}\sim\chi^2_{n-1}.
$$

Pertanto

$$
T=\frac{\overline{Y}-\mu}{S/\sqrt{n}}
=
\frac{
\dfrac{\overline{Y}-\mu}{\sigma/\sqrt{n}}
}{
\sqrt{
\dfrac{(n-1)S^2}{\sigma^2}\cdot\dfrac{1}{n-1}
}
}
\sim t_{n-1}.
$$

Sotto $H_0$, $\mu=\mu_0$, quindi la statistica test ha distribuzione $t_{n-1}$. $\square$

### 4.2 Test unilaterale

- Per $H_1:\mu>\mu_0$, si rifiuta se $T>t_{n-1,1-\alpha}$.
- Per $H_1:\mu<\mu_0$, si rifiuta se $T<-t_{n-1,1-\alpha}$.

### Esempio 4.3 — Sonnifero

Sia $Y\sim\mathcal{N}(\mu,\sigma^2)$, $n=10$, $\overline{y}=0.75$, $S^2=3.2$. Test:

$$
H_0:\mu\le 0,\qquad H_1:\mu>0.
$$

Statistica test:

$$
T_{\text{oss}}=\frac{0.75-0}{\sqrt{3.2/10}}=\frac{0.75}{0.5657}\approx 1.326.
$$

Per $\alpha=0.01$, $t_{9,0.99}=2.82$. Poiché $1.326<2.82$, non si rifiuta $H_0$. La sostanza non è efficace.

## 5. Test per la media con grandi campioni

Sia $Y_1,\dots,Y_n$ un campione i.i.d. da una popolazione con media $\mu$ e varianza $\sigma^2$ finite, con $n$ grande.

**Ipotesi:**

$$
H_0:\mu=\mu_0,\qquad H_1:\mu\neq\mu_0.
$$

**Statistica test.** Per il teorema del limite centrale,

$$
Z=\frac{\overline{Y}-\mu_0}{S/\sqrt{n}}\overset{D}{\longrightarrow}\mathcal{N}(0,1).
$$

**Regione di rifiuto.**

$$
R=\left\{\mathbf{y} : |Z|>z_{1-\alpha/2}\right\}.
$$

**p-value.**

$$
p\approx 2(1-\Phi(|z_{\text{oss}}|)).
$$

> **Osservazione 5.1.**  
> L’intervallo è approssimato: la copertura tende a $1-\alpha$ per $n\to\infty$. Per $n$ finito, l’approssimazione può essere inaccurate se la distribuzione è molto asimmetrica.

## 6. Test per il parametro di una Bernoulli con grandi campioni

Sia $Y\sim\operatorname{Be}(\pi)$ e $Y_1,\dots,Y_n$ un campione i.i.d. con $n$ grande. Sia $\hat{\pi}=\overline{Y}$.

**Ipotesi:**

$$
H_0:\pi=\pi_0,\qquad H_1:\pi\neq\pi_0.
$$

**Statistica test.** Sotto $H_0$, la varianza di $Y$ è $\pi_0(1-\pi_0)$, quindi

$$
Z=\frac{\hat{\pi}-\pi_0}{\sqrt{\pi_0(1-\pi_0)/n}}\overset{D}{\longrightarrow}\mathcal{N}(0,1).
$$

**Regione di rifiuto.**

$$
R=\left\{\mathbf{y} : |Z|>z_{1-\alpha/2}\right\}.
$$

**p-value.**

$$
p\approx 2(1-\Phi(|z_{\text{oss}}|)).
$$

### Esempio 6.1 — Sondaggio elettorale

$n=2500$, $\hat{\pi}=0.53$. Test:

$$
H_0:\pi\le 0.5,\qquad H_1:\pi>0.5.
$$

Statistica test:

$$
z_{\text{oss}}=\frac{0.53-0.5}{\sqrt{0.5\cdot 0.5/2500}}=\frac{0.03}{0.01}=3.
$$

p-value:

$$
p=P(Z>3)\approx 0.00135<\alpha.
$$

Si rifiuta $H_0$: il candidato ha buone probabilità di vincere.

## 7. Test per la differenza tra medie di due popolazioni normali

### 7.1 Varianze note

Siano $X\sim\mathcal{N}(\mu_X,\sigma_X^2)$ e $Y\sim\mathcal{N}(\mu_Y,\sigma_Y^2)$ con $\sigma_X^2,\sigma_Y^2$ note, e siano $X_1,\dots,X_{n_X}$ e $Y_1,\dots,Y_{n_Y}$ campioni indipendenti.

**Ipotesi:**

$$
H_0:\mu_X=\mu_Y,\qquad H_1:\mu_X\neq\mu_Y.
$$

**Statistica test.** Sotto $H_0$,

$$
Z=\frac{\overline{X}-\overline{Y}}{\sqrt{\dfrac{\sigma_X^2}{n_X}+\dfrac{\sigma_Y^2}{n_Y}}}\sim\mathcal{N}(0,1).
$$

**Regione di rifiuto.**

$$
|Z|>z_{1-\alpha/2}.
$$

### 7.2 Varianze uguali e non note

Siano $X\sim\mathcal{N}(\mu_X,\sigma^2)$ e $Y\sim\mathcal{N}(\mu_Y,\sigma^2)$ con $\sigma^2$ ignota e comune. Sia

$$
S_{XY}^2=\frac{(n_X-1)S_X^2+(n_Y-1)S_Y^2}{n_X+n_Y-2}
$$

la varianza pooled.

**Statistica test.** Sotto $H_0$,

$$
T=\frac{\overline{X}-\overline{Y}}{\sqrt{S_{XY}^2\left(\dfrac{1}{n_X}+\dfrac{1}{n_Y}\right)}}\sim t_{n_X+n_Y-2}.
$$

**Regione di rifiuto.**

$$
|T|>t_{n_X+n_Y-2,1-\alpha/2}.
$$

**Dimostrazione della distribuzione.**  
Se $X_i\sim\mathcal{N}(\mu_X,\sigma^2)$ e $Y_j\sim\mathcal{N}(\mu_Y,\sigma^2)$ sono indipendenti, allora

$$
\frac{(n_X-1)S_X^2}{\sigma^2}\sim\chi^2_{n_X-1},
\qquad
\frac{(n_Y-1)S_Y^2}{\sigma^2}\sim\chi^2_{n_Y-1},
$$

e sono indipendenti. Quindi

$$
\frac{(n_X+n_Y-2)S_{XY}^2}{\sigma^2}\sim\chi^2_{n_X+n_Y-2}.
$$

Inoltre $\overline{X}-\overline{Y}$ è indipendente da $S_{XY}^2$ e, sotto $H_0$,

$$
\frac{\overline{X}-\overline{Y}}{\sigma\sqrt{1/n_X+1/n_Y}}\sim\mathcal{N}(0,1).
$$

Pertanto il rapporto

$$
T=
\frac{
\dfrac{\overline{X}-\overline{Y}}{\sigma\sqrt{1/n_X+1/n_Y}}
}{
\sqrt{
\dfrac{(n_X+n_Y-2)S_{XY}^2}{\sigma^2}
\cdot
\dfrac{1}{n_X+n_Y-2}
}
}
\sim t_{n_X+n_Y-2}.
$$

$\square$

### 7.3 Varianze non note e diverse: correzione di Welch

Se $\sigma_X^2\neq\sigma_Y^2$ e non possono essere assunte uguali, la statistica

$$
T=\frac{\overline{X}-\overline{Y}}{\sqrt{\dfrac{S_X^2}{n_X}+\dfrac{S_Y^2}{n_Y}}}
$$

non ha distribuzione $t$ esatta. Si usa l’approssimazione di Welch:

$$
T\approx t_\nu,
$$

con

$$
\nu\approx
\frac{
\left(\dfrac{S_X^2}{n_X}+\dfrac{S_Y^2}{n_Y}\right)^2
}{
\dfrac{\left(S_X^2/n_X\right)^2}{n_X-1}
+
\dfrac{\left(S_Y^2/n_Y\right)^2}{n_Y-1}
}.
$$

La regione di rifiuto è

$$
|T|>t_{\nu,1-\alpha/2}.
$$

> **Osservazione 7.4 — Problema di Behrens–Fisher.**  
> Il problema di Behrens–Fisher consiste nel fatto che, quando le varianze sono ignote e diverse, non esiste una quantità pivotale esatta per $\mu_X-\mu_Y$. La distribuzione della statistica test dipende dal rapporto $\sigma_X^2/\sigma_Y^2$, che è un parametro di disturbo. La correzione di Welch–Satterthwaite fornisce un’approssimazione affidabile.

### 7.4 Grandi campioni

Se $n_X,n_Y$ sono grandi e le varianze sono ignote, si usa

$$
Z=\frac{\overline{X}-\overline{Y}}{\sqrt{\dfrac{S_X^2}{n_X}+\dfrac{S_Y^2}{n_Y}}}\overset{D}{\longrightarrow}\mathcal{N}(0,1).
$$

## 8. Test per la differenza tra proporzioni con grandi campioni

Siano $X\sim\operatorname{Be}(\pi_X)$ e $Y\sim\operatorname{Be}(\pi_Y)$, e siano $X_1,\dots,X_{n_X}$ e $Y_1,\dots,Y_{n_Y}$ campioni indipendenti con $n_X,n_Y$ grandi. Siano $\hat{\pi}_X=\overline{X}$ e $\hat{\pi}_Y=\overline{Y}$.

**Ipotesi:**

$$
H_0:\pi_X=\pi_Y,\qquad H_1:\pi_X\neq\pi_Y.
$$

Sotto $H_0$, $\pi_X=\pi_Y=\pi$. La varianza comune è stimata con

$$
\hat{\pi}=\frac{n_X\hat{\pi}_X+n_Y\hat{\pi}_Y}{n_X+n_Y}.
$$

**Statistica test.** Sotto $H_0$,

$$
Z=
\frac{\hat{\pi}_X-\hat{\pi}_Y}{
\sqrt{
\hat{\pi}(1-\hat{\pi})
\left(\dfrac{1}{n_X}+\dfrac{1}{n_Y}\right)
}
}
\overset{D}{\longrightarrow}\mathcal{N}(0,1).
$$

**Regione di rifiuto.**

$$
|Z|>z_{1-\alpha/2}.
$$

## 9. Test per dati appaiati

Siano $X_1,\dots,X_n$ e $Y_1,\dots,Y_n$ due campioni appaiati, con $X_i$ e $Y_i$ osservati sulla stessa unità. Sia

$$
\delta_i=X_i-Y_i,\qquad i=1,\dots,n.
$$

Supponiamo $\delta_i\sim\mathcal{N}(\mu_\delta,\sigma_\delta^2)$ i.i.d.

**Ipotesi:**

$$
H_0:\mu_\delta=0,\qquad H_1:\mu_\delta\neq 0.
$$

**Statistica test.** Sotto $H_0$,

$$
T=\frac{\overline{\delta}}{S_\delta/\sqrt{n}}\sim t_{n-1},
$$

dove

$$
\overline{\delta}=\frac{1}{n}\sum_{i=1}^n\delta_i,
\qquad
S_\delta^2=\frac{1}{n-1}\sum_{i=1}^n(\delta_i-\overline{\delta})^2.
$$

**Regione di rifiuto.**

$$
|T|>t_{n-1,1-\alpha/2}.
$$

> **Osservazione 9.1.**  
> Non è corretto trattare i dati appaiati come campioni indipendenti, perché $X_i$ e $Y_i$ sono correlati. La variabile $\delta_i$ elimina la dipendenza e permette di ricondursi a un test $t$ a una popolazione.

### Esempio 9.2 — Prova di memoria

$n=10$ soggetti, differenze $\delta_i$ con $\overline{\delta}=4.6$, $S_\delta^2=11.38$. Test:

$$
H_0:\mu_\delta\le 0,\qquad H_1:\mu_\delta>0.
$$

Statistica test:

$$
T_{\text{oss}}=\frac{4.6}{\sqrt{11.38/10}}=\frac{4.6}{1.0667}\approx 4.31.
$$

Per $\alpha=0.01$, $t_{9,0.99}=2.82$. Poiché $4.31>2.82$, si rifiuta $H_0$. La tecnica è efficace.

## 10. ANOVA a una via

Siano $Y_{ij}\sim\mathcal{N}(\mu_j,\sigma^2)$, $j=1,\dots,J$, $i=1,\dots,n_j$, con $\sigma^2$ comune e non nota. Sia $N=\sum_j n_j$.

**Ipotesi:**

$$
H_0:\mu_1=\dots=\mu_J,
\qquad
H_1:\text{almeno una media è diversa}.
$$

**Scomposizione della devianza.**

$$
\sum_{j=1}^J\sum_{i=1}^{n_j}(y_{ij}-\bar{y}_{..})^2
=
\underbrace{
\sum_{j=1}^J n_j(\bar{y}_{.j}-\bar{y}_{..})^2
}_{D_{\text{tra}}}
+
\underbrace{
\sum_{j=1}^J\sum_{i=1}^{n_j}(y_{ij}-\bar{y}_{.j})^2
}_{D_{\text{entro}}},
$$

dove

$$
\bar{y}_{.j}=\frac{1}{n_j}\sum_{i=1}^{n_j}y_{ij},
\qquad
\bar{y}_{..}=\frac{1}{N}\sum_{j=1}^J\sum_{i=1}^{n_j}y_{ij}.
$$

**Statistica test.** Sotto $H_0$,

$$
F=\frac{D_{\text{tra}}/(J-1)}{D_{\text{entro}}/(N-J)}\sim F_{J-1,N-J}.
$$

**Regione di rifiuto.**

$$
F>F_{J-1,N-J,1-\alpha}.
$$

**Dimostrazione della distribuzione.**  
Sotto $H_0$, tutte le medie sono uguali a $\mu$. Allora

$$
\frac{D_{\text{entro}}}{\sigma^2}\sim\chi^2_{N-J},
\qquad
\frac{D_{\text{tra}}}{\sigma^2}\sim\chi^2_{J-1},
$$

e le due quantità sono indipendenti. Pertanto il rapporto

$$
F=
\frac{
\dfrac{D_{\text{tra}}}{\sigma^2}\cdot\dfrac{1}{J-1}
}{
\dfrac{D_{\text{entro}}}{\sigma^2}\cdot\dfrac{1}{N-J}
}
\sim F_{J-1,N-J}.
$$

$\square$

### Esempio 10.1 — Fertilizzante

Tre gruppi, $n_1=4,n_2=4,n_3=3$, $N=11$. Medie: $14.5,17.5,21.0$, media generale $17.36$. Si ha:

$$
D_{\text{tra}}=72.52,
\qquad
D_{\text{entro}}=12.
$$

Statistica test:

$$
F_{\text{oss}}=\frac{72.52/2}{12/8}=\frac{36.26}{1.5}\approx 24.17.
$$

Per $\alpha=0.05$, $F_{2,8,0.95}=4.46$. Poiché $24.17>4.46$, si rifiuta $H_0$: almeno un gruppo ha media diversa.

## 11. Test non parametrici: test del chi-quadro

### 11.1 Test di indipendenza

Siano $A$ e $B$ due variabili categoriali con modalità $A_1,\dots,A_R$ e $B_1,\dots,B_C$. Si osserva un campione di $n$ unità e si costruisce la tabella di contingenza con frequenze osservate $f_{rc}$.

**Ipotesi:**

$$
H_0: A\text{ e }B\text{ sono indipendenti},
\qquad
H_1:\text{non sono indipendenti}.
$$

Sotto $H_0$, le frequenze attese sono

$$
f_{rc}^*=\frac{f_{r+}f_{+c}}{n}.
$$

**Statistica test.**

$$
X^2=\sum_{r=1}^R\sum_{c=1}^C\frac{(f_{rc}-f_{rc}^*)^2}{f_{rc}^*}.
$$

Sotto $H_0$, per $n$ grande,

$$
X^2\overset{D}{\longrightarrow}\chi^2_{(R-1)(C-1)}.
$$

**Regione di rifiuto.**

$$
X^2>\chi^2_{(R-1)(C-1),1-\alpha}.
$$

**Dimostrazione (idea).**  
La statistica $X^2$ è una misura di distanza tra le frequenze osservate e quelle attese sotto indipendenza. Sotto $H_0$, le frequenze congiunte sono stimate da $f_{r+}f_{+c}/n$, e la statistica $X^2$ converge in distribuzione a una $\chi^2$ con gradi di libertà $(R-1)(C-1)$ per il teorema del limite centrale applicato alle frequenze multinomiali. $\square$

### Esempio 11.2 — Telefonia mobile

Tabella $2\times 3$: attivazione (attivo/non attivo) e piano tariffario (A/B/C). $n=2659$. Frequenze attese calcolate come $f_{rc}^*=f_{r+}f_{+c}/n$. Statistica test:

$$
X^2_{\text{oss}}=9.69.
$$

Per $\alpha=0.05$, $\chi^2_{2,0.95}=5.99$. Poiché $9.69>5.99$, si rifiuta $H_0$: esiste associazione tra le due variabili.

### 11.3 Test di omogeneità

Siano $A$ e $B$ due variabili categoriali, ma si fissano a priori le marginali di riga o di colonna. Si vuole verificare se la distribuzione di $B$ è la stessa in ogni gruppo definito da $A$.

**Ipotesi:**

$$
H_0:\pi_{r_1}=\pi_{r_2}\ \forall r_1,r_2,
\qquad
H_1:\text{non tutte uguali}.
$$

La statistica test è la stessa del test di indipendenza:

$$
X^2=\sum_{r,c}\frac{(f_{rc}-f_{rc}^*)^2}{f_{rc}^*},
\qquad
f_{rc}^*=\frac{f_{r+}f_{+c}}{n}.
$$

Sotto $H_0$,

$$
X^2\approx\chi^2_{(R-1)(C-1)}.
$$

### 11.4 Test di bontà di adattamento

Si vuole verificare se una variabile $Y$ ha una certa distribuzione $p_0(y)$.

**Ipotesi:**

$$
H_0:p_Y(y)=p_0(y),
\qquad
H_1:p_Y(y)\neq p_0(y).
$$

Siano $l_1,\dots,l_K$ le modalità o classi e $f_1,\dots,f_K$ le frequenze osservate. Sotto $H_0$, le frequenze attese sono

$$
f_j^*=n\pi_j,
\qquad
\pi_j=P_{H_0}(Y=l_j).
$$

**Statistica test.**

$$
X^2=\sum_{j=1}^K\frac{(f_j-f_j^*)^2}{f_j^*}.
$$

Sotto $H_0$, per $n$ grande,

$$
X^2\overset{D}{\longrightarrow}\chi^2_{K-1}.
$$

**Regione di rifiuto.**

$$
X^2>\chi^2_{K-1,1-\alpha}.
$$

> **Osservazione 11.5.**  
> Se la distribuzione $p_0(y;\theta)$ dipende da un parametro $\theta$ ignoto, si stima $\theta$ e si applica il test con $\hat{\theta}$, ma i gradi di libertà si riducono del numero di parametri stimati: $K-1-p$.

### Esempio 11.6 — Coppie con tre figli

$n=1659$, $K=4$ modalità (0,1,2,3 figlie femmine). Sotto $H_0$, multinomiale con probabilità

$$
0.125,\quad 0.375,\quad 0.375,\quad 0.125,
$$

frequenze attese

$$
207.375,\quad 622.125,\quad 622.125,\quad 207.375.
$$

Statistica test:

$$
X^2_{\text{oss}}=13.32.
$$

Per $\alpha=0.01$, $\chi^2_{3,0.99}=11.34$. Poiché $13.32>11.34$, si rifiuta $H_0$: la distribuzione non è quella ipotizzata.
