# Verifica di ipotesi

---

## 0. Impostazione formale

Sia
$$
\mathbf{Y}=(Y_1,\dots,Y_n)
$$
un campione casuale con distribuzione congiunta $P_\theta$, $\theta\in\Theta\subseteq\mathbb{R}^p$. Lo spazio parametrico è $\Theta$, lo spazio campionario è $\mathcal{Y}$. Una **statistica test** è una funzione misurabile
$$
T_n:\mathcal{Y}\to\mathbb{R}.
$$
Un **test** è individuato da una regione di rifiuto $R\subseteq\mathcal{Y}$: si rifiuta $H_0$ se $\mathbf{Y}\in R$. Equivalentemente si usa la funzione indicatrice
$$
\varphi(\mathbf{y})=\mathbf{1}_{R}(\mathbf{y}),
$$
dove $\varphi=1$ significa "rifiuto $H_0$" e $\varphi=0$ significa "non rifiuto $H_0$".

> **Osservazione 0.1.**  
> Nella pratica, invece di definire direttamente $R\subseteq\mathcal{Y}$, si definisce una statistica test $T_n$ e si prende come regione di rifiuto un insieme del tipo $\{T_n>c\}$, $\{T_n<c\}$ oppure $\{|T_n|>c\}$. In questo modo il test si riduce al confronto tra un numero osservato $t_{\text{oss}}=T_n(\mathbf{y})$ e uno o più valori critici. Questo è il punto di partenza per l'interpretazione grafica: la distribuzione di $T_n$ sotto $H_0$ è una curva, e la regione di rifiuto è una coda (o due code) di quella curva.

## 1. Concetti fondamentali

### Definizione 1.1 — Ipotesi statistica

Sia $\mathbf{Y}\sim p_{\mathbf{Y}}(\mathbf{y};\theta)$ con $\theta\in\Theta$. Una **ipotesi statistica** è un'affermazione su $\theta$, cioè un sottoinsieme
$$
\Theta_0\subseteq\Theta.
$$
Dire che vale $\Theta_0$ significa affermare che il vero parametro appartiene a $\Theta_0$.

> **Osservazione 1.2.**  
> Una verifica di ipotesi rappresenta una **riduzione della dimensione** dello spazio parametrico: da $\Theta$, che può avere dimensione $p$, si passa a $\Theta_0$, che tipicamente ha dimensione minore (spesso $0$, quando $\Theta_0$ è un singolo punto).

### Definizione 1.3 — Ipotesi nulla e alternativa

- L'**ipotesi nulla** $H_0$ è l'ipotesi che si assume come riferimento; formalmente $H_0:\theta\in\Theta_0$.
- L'**ipotesi alternativa** $H_1$ è il complementare:
  $$
  H_1:\theta\in\Theta_1=\Theta\setminus\Theta_0.
  $$

> **Osservazione 1.4.**  
> $H_0$ è di solito l'ipotesi "conservativa": se vera, non c'è evidenza di un effetto o di una differenza. $H_1$ è l'ipotesi che motiva l'esperimento. Il test non "dimostra" $H_0$: se non si rifiuta $H_0$, si dice soltanto che i dati non forniscono evidenza sufficiente contro $H_0$.

### Definizione 1.5 — Ipotesi semplice e composta

- Un'ipotesi è **semplice** se specifica completamente la distribuzione, cioè se
  $$
  |\Theta_0|=1.
  $$
- Un'ipotesi è **composta** se
  $$
  |\Theta_0|>1.
  $$

> **Osservazione 1.6.**  
> Sia $Y\sim\mathcal{N}(\mu,\sigma^2)$ con $\sigma^2$ non nota. L'ipotesi $H_0:\mu=\mu_0$ non è semplice, perché non specifica $\sigma^2$: al variare di $\sigma^2$ si ottengono distribuzioni diverse. È quindi un'ipotesi composta con **parametro di disturbo** $\sigma^2$. In questi casi, la quantità pivotale serve proprio a eliminare il parametro di disturbo.

### Definizione 1.7 — Ipotesi unilaterale e bilaterale

Sia $\theta\in\mathbb{R}$.

- Un'ipotesi composta è **unilaterale** se $\Theta_0$ o $\Theta_1$ è contenuto in una semiretta, ad esempio
  $$
  H_0:\theta\le\theta_0,\quad H_1:\theta>\theta_0,
  $$
  oppure
  $$
  H_0:\theta\ge\theta_0,\quad H_1:\theta<\theta_0.
  $$
- Un'ipotesi è **bilaterale** se l'alternativa contiene valori da entrambe le direzioni, ad esempio
  $$
  H_0:\theta=\theta_0,\quad H_1:\theta\neq\theta_0.
  $$

>**Osservazione:** 
> In generale:
> $H_0:\theta\le\theta_0,\quad H_1:\theta>\theta_0$ induce un test unilaterale destro
> $H_0:\theta\ge\theta_0,\quad H_1:\theta<\theta_0$ induce un test unilaterale sinistro
> E' importante tutta via comprendere che la direzione del test dipende da come è orientata la statistica test, non solo dalla forma delle ipotesi
> Con la convenzione standard in cui si usa una statistica test crescente con il parametro ad esempio $T = \frac{\hat{\theta} - \theta_0}{\operatorname{SE}(\hat{\theta})}$ allora:
> $H_0:\theta\le\theta_0,\quad H_1:\theta>\theta_0$ induce un test unilaterale destro: si rifiuta $H_0$ per valori grandi di $T$, cioè nella coda destra: $R = \{T > c\}$ infatti, se $\theta > \theta_0$, ci si aspetta $\hat{\theta}> \theta_0$, quindi $T$ grande
> $H_0:\theta\ge\theta_0,\quad H_1:\theta<\theta_0$ induce un test unilaterale sinistro: si rifiuta $H_0$ per valori piccoli di $T$, cioè nella coda sinistra $R = \{T < c\}$ infatti, se $\theta < \theta_0$, ci si aspetta $\hat{\theta} < \theta_0$ quindi $T$ piccolo
> Se si definisce la statistica test con il segno opposto ad esempio $T' = \frac{\theta_0 - \hat{\theta}}{\operatorname{SE}(\hat{\theta})} = -T$ allora le code si scambierebbero. La direzione "destra" o "sinistra" è quindi relativa alla statistica test scelta.


**Esempio.** Un'azienda vuole verificare che lo spessore medio $\mu$ delle lastre sia pari a $14$ mm:
$$
H_0:\mu=14,\qquad H_1:\mu\neq 14.
$$

**Esempio.** Per un referendum si vuole verificare il raggiungimento del quorum:
$$
H_0:\pi\le 0.5,\qquad H_1:\pi>0.5.
$$

**Esempio.** Siano $\mu_1$ e $\mu_2$ la pressione sistolica media di pazienti trattati e non trattati con un nuovo farmaco:
$$
H_0:\mu_1\ge\mu_2,\qquad H_1:\mu_1<\mu_2.
$$
Qui sono coinvolte due popolazioni.

### Definizione 1.8 — Test statistico e statistica test

Un **test statistico** è una regola basata sul campione che porta a rifiutare o non rifiutare $H_0$. Formalmente è una coppia $(T_n,R)$ con $T_n$ statistica test e $R$ regione di rifiuto.

Una **statistica test** è una statistica $T_n(\mathbf{Y})$ usata per decidere. Spesso, sotto $H_0$, $T_n$ è **pivotale**, cioè ha distribuzione nota e indipendente da parametri ignoti.

> **Osservazione 1.9.**  
> La statistica test è una variabile aleatoria. Il valore osservato $t_{\text{oss}}=T_n(\mathbf{y})$ è un numero. Le decisioni si basano sul confronto tra $t_{\text{oss}}$ e valori critici della distribuzione di $T_n$ sotto $H_0$, tenendo conto della direzione del test.

### Definizione 1.10 — Errori di I e II tipo

Quando si decide essendo la decisione basata sull'estrazione di un campione, si possono commettere due tipi di errore:

- $H_0$ vera e non si rifiuta $H_0$: decisione corretta.
- $H_0$ vera e si rifiuta $H_0$: errore di I tipo, con probabilità $\alpha$.
- $H_1$ vera e non si rifiuta $H_0$: errore di II tipo, con probabilità $\beta$.
- $H_1$ vera e si rifiuta $H_0$: decisione corretta.

Formalmente, per $\theta\in\Theta_0$,
$$
\alpha(\theta)=P_\theta(\mathbf{Y}\in R)=P_\theta(\text{rifiutare }H_0)=P(\text{rifiutare }H_0\mid H_0),
$$
e per $\theta\in\Theta_1$,
$$
\beta(\theta)=P_\theta(\mathbf{Y}\in R^C)=P_\theta(\text{non rifiutare }H_0)=P(\text{non rifiutare }H_0\mid H_1).
$$

> **Osservazione.**  
> $\alpha(\theta)$ e $\beta(\theta)$ sono **funzioni di $\theta$**, non numeri fissi. Quando si dice "$\alpha=0.05$", si intende che il livello del test è $0.05$, cioè che la probabilità massima di errore di I tipo su tutto $\Theta_0$ è al più $0.05$.

> **Interpretazione.**  
> Si immagini di tracciare la densità di $T_n$ sotto $H_0$ e, separatamente, la densità di $T_n$ sotto un valore specifico $\theta_1\in\Theta_1$. Le due curve hanno forma simile (spesso normale o $t$) ma centri diversi: quella sotto $H_1$ è spostata nella direzione dell'alternativa. La regione di rifiuto è una coda (o due code) della distribuzione sotto $H_0$:
> - l'**area della coda sotto $H_0$** è $\alpha$, la probabilità di errore di I tipo;
> - l'**area della regione complementare sotto $H_1$** è $\beta$, la probabilità di errore di II tipo;
> - l'**area della regione di rifiuto sotto $H_1$** è $1-\beta$, la potenza.
> 
> Quando le due curve sono molto vicine, la coda di $H_0$ e la parte centrale di $H_1$ si sovrappongono molto: $\beta$ è grande e la potenza è piccola. Quando le due curve sono lontane, la coda di $H_0$ contiene quasi tutta la massa di $H_1$: $\beta$ è piccolo e la potenza è vicina a $1$.

### Definizione 1.11 — Livello di significatività

Il **livello di significatività** del test è
$$
\alpha=\sup_{\theta\in\Theta_0}P_\theta(\mathbf{Y}\in R).
$$
Un test ha livello $\alpha$ se
$$
P_\theta(\mathbf{Y}\in R)\le\alpha\qquad\forall\theta\in\Theta_0.
$$
Se vale l'uguaglianza in corrispondenza del valore più sfavorevole di $\theta$, $\alpha$ è anche detta **dimensione** del test.

> **Osservazione 1.12.**  
> Fissata la numerosità $n$, diminuire $\alpha$ rende più difficile rifiutare $H_0$, quindi aumenta $\beta$ in genere. Non si possono minimizzare simultaneamente i due errori. La prassi è fissare $\alpha$ piccolo, ad esempio $0.05$ o $0.01$, e tra i test di livello $\alpha$ scegliere quello con potenza massima.

> **Interpretazione del livello.**  
> Il livello $\alpha$ è l'area della coda (o delle due code) sotto la densità di $T_n$ valutata nel punto più sfavorevole di $\Theta_0$. Per un test unilaterale destro, $\alpha$ è l'area a destra del valore critico $c$. Per un test bilaterale, è la somma delle due aree esterne ai valori critici $c_1$ e $c_2$. Ridurre $\alpha$ significa spostare il valore critico più verso la coda, restringendo la regione di rifiuto.

### Definizione 1.13 — p-value

Sia $T_n$ una statistica test e $t_{\text{oss}}=T_n(\mathbf{y})$. Il **p-value** è la probabilità, calcolata sotto $H_0$, che $T_n$ assuma un valore "più estremo" di quello osservato, nella direzione indicata da $H_1$:
$$
p=P_{H_0}(T_n\text{ è più estremo di }t_{\text{oss}}).
$$

Equivalentemente, se la famiglia di regioni di rifiuto è annidata, cioè $R_\alpha$ cresce con $\alpha$, si può definire
$$
p(\mathbf{y})=\inf\{\alpha\in(0,1):\mathbf{y}\in R_\alpha\}.
$$
Questa definizione è la più generale e mette in luce che il p-value è il più piccolo livello di significatività al quale i dati osservati porterebbero a rifiutare $H_0$.

Per un test unilaterale destro, con statistica test $T_n$ grande sotto $H_1$:
$$
p=P_{H_0}(T_n\ge t_{\text{oss}}).
$$
Per un test unilaterale sinistro:
$$
p=P_{H_0}(T_n\le t_{\text{oss}}).
$$
Per un test bilaterale, se la distribuzione di $T_n$ sotto $H_0$ è simmetrica:
$$
p=2\min\{P_{H_0}(T_n\le t_{\text{oss}}),P_{H_0}(T_n\ge t_{\text{oss}})\}.
$$

**Regola di decisione.**
- Se $p<\alpha$, si rifiuta $H_0$ al livello $\alpha$.
- Se $p\ge\alpha$, non si rifiuta $H_0$ al livello $\alpha$.

> **Osservazione 1.14 — Che cosa non è il p-value.**  
> Il p-value **non** è la probabilità che $H_0$ sia vera. Non è $P(H_0\mid\mathbf{y})$. È una probabilità calcolata **assumendo $H_0$ vera** e riguarda la statistica test, non direttamente il parametro. Un p-value piccolo significa che, se $H_0$ fosse vera, sarebbe improbabile osservare un valore così estremo di $T_n$.

> **Interpretazione del p-value.**  
> Si tracci la densità di $T_n$ sotto $H_0$. Il valore osservato $t_{\text{oss}}$ è un punto sull'asse delle ascisse. Il p-value è l'area della coda che parte da $t_{\text{oss}}$ e va verso l'infinito nella direzione dell'alternativa (a destra per un test unilaterale destro, a sinistra per un test unilaterale sinistro, in entrambe le code per un test bilaterale simmetrico). Più $t_{\text{oss}}$ è vicino alla coda, più piccola è l'area, più piccolo è il p-value, più forte è l'evidenza contro $H_0$. Il p-value è quindi l'area "oltre" il valore osservato.

### Definizione 1.15 — Potenza del test

La **potenza** del test in $\theta\in\Theta_1$ è
$$
\pi(\theta)=P_\theta(\mathbf{Y}\in R)=1-\beta(\theta).
$$
La **funzione di potenza** è definita su tutto $\Theta$:
$$
\pi:\Theta\to[0,1],\qquad \pi(\theta)=P_\theta(\mathbf{Y}\in R).
$$
Per $\theta\in\Theta_0$, $\pi(\theta)\le\alpha$. Per $\theta\in\Theta_1$, $\pi(\theta)$ misura la capacità del test di rilevare che $H_0$ è falsa.

> **Osservazione 1.16 — Interpretazione della potenza.**  
> La potenza è la probabilità di rifiutare $H_0$ quando $H_1$ è vera. Un test con potenza $0.80$ in un certo $\theta\in\Theta_1$ rifiuta $H_0$ l'$80\%$ delle volte se il vero parametro è quel $\theta$. La potenza dipende da:
> - la distanza tra $\theta$ e $\Theta_0$;
> - la numerosità campionaria $n$;
> - il livello $\alpha$;
> - la variabilità dei dati.
>
> A parità di $n$, aumentare la potenza contro alternative lontane è facile; contro alternative vicine serve più informazione, cioè più dati.

> **Interpretazione grafica della potenza.**  
> Si traccino due densità di $T_n$: una sotto $H_0$ (centrata su un valore compatibile con $H_0$) e una sotto un'alternativa $\theta_1$ (centrata più a destra, se l'alternativa è $\theta_1>\theta_0$). La regione di rifiuto è la coda destra sotto $H_0$, delimitata dal valore critico $c$, con area $\alpha$. La potenza è **l'area sotto la densità di $H_1$ che cade a destra di $c$**. Se $\theta_1$ è vicino a $\theta_0$, le due curve si sovrappongono molto, la coda destra di $H_1$ è piccola e la potenza è bassa. Se $\theta_1$ è lontano, quasi tutta la massa di $H_1$ sta a destra di $c$ e la potenza è vicina a $1$. Aumentare $n$ restringe entrambe le curve e aumenta la separazione, aumentando la potenza.

### Teorema 1.17 — Dualità tra intervalli di confidenza e test

Sia $\operatorname{IC}_{1-\alpha}(\theta)$ un intervallo di confidenza di livello $1-\alpha$ per $\theta$. Allora, per ogni $\theta_0\in\Theta$, il test che rifiuta
$$
H_0:\theta=\theta_0
\qquad\text{contro}\qquad
H_1:\theta\neq\theta_0
$$
se e solo se
$$
\theta_0\notin\operatorname{IC}_{1-\alpha}(\theta)
$$
ha livello $\alpha$.

**Dimostrazione.**  
Per definizione di intervallo di confidenza,
$$
P_\theta\bigl(\theta\in\operatorname{IC}_{1-\alpha}(\theta)\bigr)=1-\alpha.
$$
Allora
$$
\begin{aligned}
P_\theta\bigl(\theta_0\notin\operatorname{IC}_{1-\alpha}(\theta)\mid \theta=\theta_0\bigr)
&=
1-P_{\theta_0}\bigl(\theta_0\in\operatorname{IC}_{1-\alpha}(\theta)\bigr) \\
&=
1-(1-\alpha) \\
&=
\alpha.
\end{aligned}
$$
Quindi il test ha livello $\alpha$. $\square$

> **Osservazione 1.18.**  
> La dualità è esatta per test bilaterali e intervalli bilaterali. Per test unilaterali si usano intervalli unilaterali, con la corrispondenza analoga.

> **Interpretazione della dualità.**  
> L'intervallo di confidenza è l'insieme dei valori $\theta_0$ che "non sarebbero rifiutati" dal test. Se si immagina di far variare $\theta_0$, l'intervallo di confidenza è l'insieme dei valori per cui il valore osservato $t_{\text{oss}}$ cade nella regione di accettazione. Rifiutare $H_0:\theta=\theta_0$ equivale a dire che $\theta_0$ è fuori dall'intervallo: il valore osservato è troppo estremo per essere compatibile con quel $\theta_0$.

## 2. Quantità pivotali e costruzione dei test

### Definizione 2.1 — Quantità pivotale

Sia $\mathbf{Y}\sim P_\theta$, $\theta\in\Theta$. Una **quantità pivotale** è una funzione
$$
Q(\mathbf{Y},\theta)
$$
tale che la sua distribuzione **non dipende da $\theta$**. In altri termini, per ogni $\theta\in\Theta$, $Q(\mathbf{Y},\theta)$ ha la stessa distribuzione.

> **Osservazione 2.2.**  
> Una quantità pivotale è lo strumento che permette di costruire sia intervalli di confidenza sia test di ipotesi. L'idea è che, se si conosce la distribuzione di $Q$ e questa non dipende da $\theta$, allora si possono calcolare probabilità del tipo $P(q_1\le Q\le q_2)$ senza conoscere $\theta$. Nei test, si fissa $\theta=\theta_0$ (sotto $H_0$) e si valuta $Q(\mathbf{Y},\theta_0)$: se il valore osservato è "estremo" rispetto alla distribuzione nota di $Q$, si rifiuta $H_0$.

### 2.3 Costruzione di un test da una quantità pivotale

Il procedimento standard è il seguente:

1. **Individuare una quantità pivotale** $Q(\mathbf{Y},\theta)$ per il parametro di interesse.
2. **Fissare $\theta=\theta_0$**, il valore specificato da $H_0$. La quantità $T_n=Q(\mathbf{Y},\theta_0)$ diventa una statistica test con distribuzione nota.
3. **Scegliere la direzione del test** in base a $H_1$:
   - se $H_1$ è del tipo $\theta>\theta_0$, si rifiuta per valori grandi di $T_n$;
   - se $H_1$ è del tipo $\theta<\theta_0$, si rifiuta per valori piccoli di $T_n$;
   - se $H_1$ è del tipo $\theta\neq\theta_0$, si rifiuta per valori di $T_n$ lontani da zero (in valore assoluto).
4. **Determinare il valore critico** in modo che l'area della coda (o delle due code) sotto la distribuzione di $T_n$ sotto $H_0$ sia $\alpha$.
5. **Confrontare $t_{\text{oss}}$ con il valore critico** e calcolare il p-value.

> **Interpretazione grafica.**  
> Il punto chiave è che, fissato $\theta=\theta_0$, la distribuzione di $T_n$ è una curva nota (normale, $t$, $\chi^2$, $F$). La regione di rifiuto è una coda di questa curva. La scelta della coda dipende dalla direzione dell'alternativa. Il valore critico è il punto sull'asse delle ascisse che lascia l'area $\alpha$ nella coda. Il valore osservato $t_{\text{oss}}$ cade o non cade in quella coda, e il p-value è l'area oltre $t_{\text{oss}}$.

## 3. Test per la media di una popolazione normale con varianza nota

Sia
$$
Y\sim\mathcal{N}(\mu,\sigma^2)
$$
con $\sigma^2$ nota, e sia $Y_1,\dots,Y_n$ un campione i.i.d.

### 3.1 La quantità pivotale

Poiché $\bar{Y}\sim\mathcal{N}(\mu,\sigma^2/n)$, la standardizzazione
$$
Z=\frac{\bar{Y}-\mu}{\sigma/\sqrt{n}}
$$
ha distribuzione $\mathcal{N}(0,1)$ per ogni $\mu$. La distribuzione di $Z$ **non dipende da $\mu$** (né da $\sigma$, che è nota): è quindi una quantità pivotale.

**Statistica test.** Fissato $\mu=\mu_0$ (il valore specificato da $H_0$), si ottiene
$$
Z_0=\frac{\bar{Y}-\mu_0}{\sigma/\sqrt{n}}\sim\mathcal{N}(0,1)
\qquad\text{sotto }H_0.
$$

> **Interpretazione grafica.**  
> Sotto $H_0$, $Z_0$ ha una densità a campana centrata in $0$. Valori di $Z_0$ vicini a $0$ sono compatibili con $H_0$. Valori grandi in valore assoluto sono nella coda (o nelle due code) e indicano che la media campionaria è lontana da $\mu_0$ più di quanto ci si aspetterebbe per puro effetto del caso.

### 3.2 Test bilaterale

**Ipotesi:**
$$
H_0:\mu=\mu_0,\qquad H_1:\mu\neq\mu_0.
$$

**Regione di rifiuto.** Per un test di livello $\alpha$, si rifiuta per valori di $|Z_0|$ grandi, cioè
$$
R=\left\{\mathbf{y}:|Z_0|>z_{1-\alpha/2}\right\},
$$
dove $z_{1-\alpha/2}$ è il quantile di ordine $1-\alpha/2$ della normale standard. Equivalentemente,
$$
R=
\left\{
\mathbf{y}:
\bar{y}<\mu_0-z_{1-\alpha/2}\frac{\sigma}{\sqrt{n}}
\ \text{oppure}\
\bar{y}>\mu_0+z_{1-\alpha/2}\frac{\sigma}{\sqrt{n}}
\right\}.
$$

**p-value.**
$$
p=P(|Z|>|z_{\text{oss}}|)=2(1-\Phi(|z_{\text{oss}}|)).
$$

> **Interpretazione grafica.**  
> La regione di rifiuto è l'unione di due code simmetriche: una a sinistra di $-z_{1-\alpha/2}$ e una a destra di $z_{1-\alpha/2}$. Ciascuna coda ha area $\alpha/2$, per un'area totale $\alpha$. Il p-value è due volte l'area della coda che parte da $|z_{\text{oss}}|$.

### 3.3 Test unilaterale destro

**Ipotesi:**
$$
H_0:\mu\le\mu_0,\qquad H_1:\mu>\mu_0.
$$

**Regione di rifiuto.**
$$
R=\left\{\mathbf{y}:Z_0>z_{1-\alpha}\right\}
=
\left\{\mathbf{y}:\bar{y}>\mu_0+z_{1-\alpha}\frac{\sigma}{\sqrt{n}}\right\}.
$$

**p-value.**
$$
p=P(Z>z_{\text{oss}})=1-\Phi(z_{\text{oss}}).
$$

**Livello del test.**  
Per $\mu\le\mu_0$,
$$
Z_0\sim\mathcal{N}\left(\frac{\mu-\mu_0}{\sigma/\sqrt{n}},1\right).
$$
Poiché $\mu-\mu_0\le0$,
$$
P_\mu(Z_0>z_{1-\alpha})
\le
P_{\mu_0}(Z_0>z_{1-\alpha})
=
\alpha.
$$
Quindi
$$
\sup_{\mu\le\mu_0}P_\mu(\mathbf{Y}\in R)=\alpha.
$$
Il caso più sfavorevole è $\mu=\mu_0$, il valore di $\Theta_0$ più vicino all'alternativa.

> **Interpretazione grafica.**  
> La regione di rifiuto è la **sola coda destra** della normale standard, con area $\alpha$. Per $\mu\le\mu_0$ la distribuzione di $Z_0$ è centrata in un valore $\le0$, quindi la coda destra ha area minore di $\alpha$. Il massimo si ottiene in $\mu=\mu_0$, quando la distribuzione è centrata in $0$ e la coda destra ha esattamente area $\alpha$. Questo è il senso grafico del "caso più sfavorevole".

### 3.4 Test unilaterale sinistro

**Ipotesi:**
$$
H_0:\mu\ge\mu_0,\qquad H_1:\mu<\mu_0.
$$

**Regione di rifiuto.**
$$
R=\left\{\mathbf{y}:Z_0<-z_{1-\alpha}\right\}
=
\left\{\mathbf{y}:\bar{y}<\mu_0-z_{1-\alpha}\frac{\sigma}{\sqrt{n}}\right\}.
$$

**p-value.**
$$
p=P(Z<z_{\text{oss}})=\Phi(z_{\text{oss}}).
$$

> **Interpretazione grafica.**  
> La regione di rifiuto è la **sola coda sinistra** della normale standard, con area $\alpha$. Per $\mu\ge\mu_0$ la distribuzione di $Z_0$ è centrata in un valore $\ge0$, quindi la coda sinistra ha area minore di $\alpha$. Il massimo si ottiene in $\mu=\mu_0$.

---

## 4. Test per la varianza di una popolazione normale

Sia $Y\sim\mathcal{N}(\mu,\sigma^2)$ con $\mu$ e $\sigma^2$ non note, e sia $Y_1,\dots,Y_n$ un campione i.i.d.

### 4.1 La quantità pivotale

La varianza campionaria corretta è
$$
S^2=\frac{1}{n-1}\sum_{i=1}^n(Y_i-\bar{Y})^2.
$$
La quantità
$$
Q=\frac{(n-1)S^2}{\sigma^2}
$$
ha distribuzione $\chi^2_{n-1}$, **indipendente da $\mu$ e da $\sigma^2$**: è quindi una quantità pivotale per $\sigma^2$.

> **Osservazione 4.2.**  
> Il fatto che $Q$ sia pivotale è un risultato esatto per popolazioni normali: la somma dei quadrati degli scarti dalla media campionaria, divisa per $\sigma^2$, è una somma di $n-1$ quadrati di normali standard indipendenti. Questo è il motivo per cui la distribuzione è $\chi^2_{n-1}$ e non dipende da $\mu$.

**Statistica test.** Fissato $\sigma^2=\sigma_0^2$, si ottiene
$$
W=\frac{(n-1)S^2}{\sigma_0^2}\sim\chi^2_{n-1}
\qquad\text{sotto }H_0.
$$

> **Interpretazione grafica.**  
> La distribuzione $\chi^2_{n-1}$ è asimmetrica a destra: ha una lunga coda destra e una coda sinistra più corta. Questo significa che la regione di rifiuto per un test unilaterale destro è una coda destra "sottile", mentre per un test unilaterale sinistro è una coda sinistra "corta". Per un test bilaterale, le due code hanno aree $\alpha/2$ ciascuna, ma i valori critici non sono simmetrici rispetto allo $0$: il valore critico di sinistra è piccolo (spesso vicino a $0$), quello di destra è grande.

### 4.2 Test unilaterale destro

**Ipotesi:**
$$
H_0:\sigma^2\le\sigma_0^2,\qquad H_1:\sigma^2>\sigma_0^2.
$$

**Regione di rifiuto.**
$$
R=\left\{\mathbf{y}:W>\chi^2_{n-1,1-\alpha}\right\}.
$$

**p-value.**
$$
p=P(\chi^2_{n-1}>W_{\text{oss}}).
$$

**Livello del test.**  
Se $\sigma^2\le\sigma_0^2$,
$$
W=\frac{\sigma^2}{\sigma_0^2}\cdot\frac{(n-1)S^2}{\sigma^2}.
$$
Poiché $\sigma^2/\sigma_0^2\le1$, per ogni $c>0$,
$$
P_{\sigma^2}(W>c)
=
P\left(\chi^2_{n-1}>\frac{\sigma_0^2}{\sigma^2}c\right)
\le
P(\chi^2_{n-1}>c).
$$
Quindi il sup su $H_0$ si ottiene per $\sigma^2=\sigma_0^2$ e vale $\alpha$.

> **Interpretazione grafica.**  
> La regione di rifiuto è la **coda destra** della $\chi^2_{n-1}$, con area $\alpha$. Il valore critico è $\chi^2_{n-1,1-\alpha}$. Per $\sigma^2<\sigma_0^2$, la statistica $W$ tende a essere più piccola, quindi la probabilità di cadere nella coda destra diminuisce. Il caso più sfavorevole è $\sigma^2=\sigma_0^2$.

### 4.3 Test bilaterale

**Ipotesi:**
$$
H_0:\sigma^2=\sigma_0^2,\qquad H_1:\sigma^2\neq\sigma_0^2.
$$

**Regione di rifiuto.**
$$
R=
\left\{
\mathbf{y}:
W<\chi^2_{n-1,\alpha/2}
\ \text{oppure}\
W>\chi^2_{n-1,1-\alpha/2}
\right\}.
$$

**p-value.** Poiché la distribuzione $\chi^2$ è asimmetrica, si usa
$$
p=2\min\{P(\chi^2_{n-1}<W_{\text{oss}}),P(\chi^2_{n-1}>W_{\text{oss}})\},
$$
eventualmente troncato a $1$.

> **Interpretazione grafica.**  
> La regione di rifiuto è l'unione di **due code asimmetriche**: una coda sinistra che parte da $\chi^2_{n-1,\alpha/2}$ e una coda destra che parte da $\chi^2_{n-1,1-\alpha/2}$. Le due aree sono entrambe $\alpha/2$, ma i valori critici non sono simmetrici rispetto alla media. Il p-value è due volte la più piccola delle due aree esterne.

## 5. Test per la media di una popolazione normale con varianza non nota

Sia $Y\sim\mathcal{N}(\mu,\sigma^2)$ con $\mu$ e $\sigma^2$ non note, e sia $Y_1,\dots,Y_n$ un campione i.i.d.

### 5.1 La quantità pivotale

La varianza $\sigma^2$ è un parametro di disturbo. Per eliminarlo, si studentizza la media campionaria usando $S^2$:
$$
T=\frac{\bar{Y}-\mu}{S/\sqrt{n}}.
$$
Se $Y_i\sim\mathcal{N}(\mu,\sigma^2)$ i.i.d., allora $\bar{Y}$ e $S^2$ sono indipendenti, e
$$
\frac{\bar{Y}-\mu}{\sigma/\sqrt{n}}\sim\mathcal{N}(0,1),
\qquad
\frac{(n-1)S^2}{\sigma^2}\sim\chi^2_{n-1}.
$$
Pertanto
$$
T=\frac{\bar{Y}-\mu}{S/\sqrt{n}}
=
\frac{
\dfrac{\bar{Y}-\mu}{\sigma/\sqrt{n}}
}{
\sqrt{
\dfrac{(n-1)S^2}{\sigma^2}\cdot\dfrac{1}{n-1}
}
}
\sim t_{n-1}.
$$
La distribuzione di $T$ **non dipende da $\mu$ né da $\sigma^2$**: è una quantità pivotale per $\mu$.

**Statistica test.** Fissato $\mu=\mu_0$, si ottiene
$$
T_0=\frac{\bar{Y}-\mu_0}{S/\sqrt{n}}\sim t_{n-1}
\qquad\text{sotto }H_0.
$$

> **Interpretazione grafica.**  
> La distribuzione $t_{n-1}$ è simile alla normale standard, ma con code più pesanti: è più "schiacciata" al centro e ha più massa nelle code. Questo riflette l'incertezza aggiuntiva dovuta alla stima di $\sigma^2$. Al crescere di $n$, la $t_{n-1}$ converge alla normale standard e le code si alleggeriscono.

### 5.2 Test bilaterale

**Ipotesi:**
$$
H_0:\mu=\mu_0,\qquad H_1:\mu\neq\mu_0.
$$

**Regione di rifiuto.**
$$
R=\left\{\mathbf{y}:|T_0|>t_{n-1,1-\alpha/2}\right\}.
$$

**p-value.**
$$
p=P(|t_{n-1}|>|t_{\text{oss}}|)=2P(t_{n-1}>|t_{\text{oss}}|).
$$

### 5.3 Test unilaterale

- Per $H_1:\mu>\mu_0$, si rifiuta se $T_0>t_{n-1,1-\alpha}$.
- Per $H_1:\mu<\mu_0$, si rifiuta se $T_0<-t_{n-1,1-\alpha}$.

> **Interpretazione grafica.**  
> Come per il test $Z$, la regione di rifiuto è una coda (o due code) della distribuzione $t_{n-1}$. La differenza rispetto al caso con varianza nota è che la curva è più "pesante" nelle code: a parità di $\alpha$, il valore critico è più grande in valore assoluto. Questo rende il test più conservativo, riflettendo l'incertezza su $\sigma^2$.

## 6. Test per la media con grandi campioni

Sia $Y_1,\dots,Y_n$ un campione i.i.d. da una popolazione con media $\mu$ e varianza $\sigma^2$ finite, con $n$ grande.

**Ipotesi:**
$$
H_0:\mu=\mu_0,\qquad H_1:\mu\neq\mu_0.
$$

### 6.1 La quantità pivotale asintotica

Per il teorema del limite centrale,
$$
\sqrt{n}(\bar{Y}-\mu)\overset{D}{\longrightarrow}\mathcal{N}(0,\sigma^2).
$$
Sostituendo $\sigma^2$ con lo stimatore consistente $S^2$, per il teorema di Slutsky,
$$
\sqrt{n}\frac{\bar{Y}-\mu}{S}\overset{D}{\longrightarrow}\mathcal{N}(0,1).
$$
Quindi
$$
Z=\frac{\bar{Y}-\mu}{S/\sqrt{n}}\overset{D}{\longrightarrow}\mathcal{N}(0,1)
$$
per ogni $\mu$. Questa è una quantità pivotale **asintotica**: la distribuzione limite non dipende da $\mu$ né da $\sigma^2$.

**Statistica test.** Sotto $H_0$,
$$
Z_0=\frac{\bar{Y}-\mu_0}{S/\sqrt{n}}\overset{D}{\longrightarrow}\mathcal{N}(0,1).
$$

**Regione di rifiuto.**
$$
R=\left\{\mathbf{y}:|Z_0|>z_{1-\alpha/2}\right\}.
$$

**p-value.**
$$
p\approx 2(1-\Phi(|z_{\text{oss}}|)).
$$

> **Osservazione 6.2.**  
> L'approssimazione è asintotica. Per $n$ finito può essere inaccurate se la distribuzione è molto asimmetrica o se ci sono outlier.

> **Interpretazione grafica.**  
> La regione di rifiuto è ancora una coda (o due code) della normale standard, ma la curva è solo un'approssimazione della vera distribuzione di $Z_0$. Al crescere di $n$, la vera distribuzione di $Z_0$ si avvicina alla normale standard per il teorema del limite centrale, quindi l'approssimazione diventa più accurata.

## 7. Test per il parametro di una Bernoulli con grandi campioni

Sia $Y\sim\operatorname{Be}(\pi)$ e $Y_1,\dots,Y_n$ un campione i.i.d. con $n$ grande. Sia $\hat{\pi}=\bar{Y}$.

**Ipotesi:**
$$
H_0:\pi=\pi_0,\qquad H_1:\pi\neq\pi_0.
$$

### 7.1 La quantità pivotale asintotica

Sotto $H_0$, la varianza di $Y$ è $\pi_0(1-\pi_0)$. Per il teorema del limite centrale,
$$
\sqrt{n}\frac{\hat{\pi}-\pi_0}{\sqrt{\pi_0(1-\pi_0)}}\overset{D}{\longrightarrow}\mathcal{N}(0,1).
$$
Quindi
$$
Z=\frac{\hat{\pi}-\pi_0}{\sqrt{\pi_0(1-\pi_0)/n}}\overset{D}{\longrightarrow}\mathcal{N}(0,1).
$$
Questa è una quantità pivotale asintotica: la varianza è calcolata sotto $H_0$, quindi non dipende da parametri ignoti.

**Regione di rifiuto.**
$$
R=\left\{\mathbf{y}:|Z|>z_{1-\alpha/2}\right\}.
$$

**p-value.**
$$
p\approx 2(1-\Phi(|z_{\text{oss}}|)).
$$

> **Interpretazione grafica.**  
> La regione di rifiuto è una coda (o due code) della normale standard. La differenza rispetto al test $Z$ per la media è che qui la varianza è determinata da $\pi_0$ sotto $H_0$, quindi non c'è incertezza aggiuntiva. L'approssimazione normale è affidabile quando $n\pi_0$ e $n(1-\pi_0)$ sono entrambi sufficientemente grandi (tipicamente $\ge5$).

## 8. Test per la differenza tra medie di due popolazioni normali

### 8.1 Varianze note

Siano $X\sim\mathcal{N}(\mu_X,\sigma_X^2)$ e $Y\sim\mathcal{N}(\mu_Y,\sigma_Y^2)$ con $\sigma_X^2,\sigma_Y^2$ note, e siano $X_1,\dots,X_{n_X}$ e $Y_1,\dots,Y_{n_Y}$ campioni indipendenti.

**Ipotesi:**
$$
H_0:\mu_X=\mu_Y,\qquad H_1:\mu_X\neq\mu_Y.
$$

**Quantità pivotale.** La differenza $\bar{X}-\bar{Y}$ è normale con media $\mu_X-\mu_Y$ e varianza $\sigma_X^2/n_X+\sigma_Y^2/n_Y$. Quindi
$$
Z=\frac{(\bar{X}-\bar{Y})-(\mu_X-\mu_Y)}{\sqrt{\dfrac{\sigma_X^2}{n_X}+\dfrac{\sigma_Y^2}{n_Y}}}\sim\mathcal{N}(0,1)
$$
per ogni $\mu_X,\mu_Y$. Sotto $H_0$, $\mu_X-\mu_Y=0$, quindi
$$
Z_0=\frac{\bar{X}-\bar{Y}}{\sqrt{\dfrac{\sigma_X^2}{n_X}+\dfrac{\sigma_Y^2}{n_Y}}}\sim\mathcal{N}(0,1).
$$

**Regione di rifiuto.**
$$
|Z_0|>z_{1-\alpha/2}.
$$

> **Interpretazione grafica.**  
> La regione di rifiuto è una coda (o due code) della normale standard, esattamente come nel test per una media. La differenza è che la varianza della statistica test combina le varianze delle due popolazioni, pesate con le numerosità campionarie.

### 8.2 Varianze uguali e non note

Siano $X\sim\mathcal{N}(\mu_X,\sigma^2)$ e $Y\sim\mathcal{N}(\mu_Y,\sigma^2)$ con $\sigma^2$ ignota e comune. Sia
$$
S_{XY}^2=\frac{(n_X-1)S_X^2+(n_Y-1)S_Y^2}{n_X+n_Y-2}
$$
la varianza pooled.

**Quantità pivotale.** Se $X_i\sim\mathcal{N}(\mu_X,\sigma^2)$ e $Y_j\sim\mathcal{N}(\mu_Y,\sigma^2)$ sono indipendenti, allora
$$
\frac{(n_X-1)S_X^2}{\sigma^2}\sim\chi^2_{n_X-1},
\qquad
\frac{(n_Y-1)S_Y^2}{\sigma^2}\sim\chi^2_{n_Y-1},
$$
e sono indipendenti. Quindi
$$
\frac{(n_X+n_Y-2)S_{XY}^2}{\sigma^2}\sim\chi^2_{n_X+n_Y-2}.
$$
Inoltre $\bar{X}-\bar{Y}$ è indipendente da $S_{XY}^2$ e
$$
\frac{(\bar{X}-\bar{Y})-(\mu_X-\mu_Y)}{\sigma\sqrt{1/n_X+1/n_Y}}\sim\mathcal{N}(0,1).
$$
Pertanto
$$
T=
\frac{
\dfrac{(\bar{X}-\bar{Y})-(\mu_X-\mu_Y)}{\sigma\sqrt{1/n_X+1/n_Y}}
}{
\sqrt{
\dfrac{(n_X+n_Y-2)S_{XY}^2}{\sigma^2}
\cdot
\dfrac{1}{n_X+n_Y-2}
}
}
\sim t_{n_X+n_Y-2}.
$$
Sotto $H_0$, $\mu_X-\mu_Y=0$, quindi
$$
T_0=\frac{\bar{X}-\bar{Y}}{\sqrt{S_{XY}^2\left(\dfrac{1}{n_X}+\dfrac{1}{n_Y}\right)}}\sim t_{n_X+n_Y-2}.
$$

**Regione di rifiuto.**
$$
|T_0|>t_{n_X+n_Y-2,1-\alpha/2}.
$$

> **Interpretazione grafica.**  
> La regione di rifiuto è una coda (o due code) della distribuzione $t_{n_X+n_Y-2}$. Rispetto al caso con varianze note, la curva è più pesante nelle code, riflettendo l'incertezza sulla varianza comune.

### 8.3 Varianze non note e diverse: correzione di Welch

Se $\sigma_X^2\neq\sigma_Y^2$ e non possono essere assunte uguali, la statistica
$$
T_0=\frac{\bar{X}-\bar{Y}}{\sqrt{\dfrac{S_X^2}{n_X}+\dfrac{S_Y^2}{n_Y}}}
$$
non ha distribuzione $t$ esatta. Si usa l'approssimazione di Welch:
$$
T_0\approx t_\nu,
$$
con
$$
\nu\approx
\frac{
\left(\dfrac{S_X^2}{n_X}+\dfrac{S_Y^2}{n_Y}\right)^2
}{
\dfrac{\left(S_X^2/n_X\right)^2}{n_X-1}
+
\dfrac{\left(S_Y^2/n_Y\right)^2}{n_Y-1}
}.
$$
La regione di rifiuto è
$$
|T_0|>t_{\nu,1-\alpha/2}.
$$

> **Osservazione 8.1 — Problema di Behrens–Fisher.**  
> Quando le varianze sono ignote e diverse, non esiste una quantità pivotale esatta per $\mu_X-\mu_Y$. La distribuzione della statistica test dipende dal rapporto $\sigma_X^2/\sigma_Y^2$, che è un parametro di disturbo. La correzione di Welch–Satterthwaite fornisce un'approssimazione affidabile.

### 8.4 Grandi campioni

Se $n_X,n_Y$ sono grandi e le varianze sono ignote, si usa
$$
Z_0=\frac{\bar{X}-\bar{Y}}{\sqrt{\dfrac{S_X^2}{n_X}+\dfrac{S_Y^2}{n_Y}}}\overset{D}{\longrightarrow}\mathcal{N}(0,1).
$$

## 9. Test per la differenza tra proporzioni con grandi campioni

Siano $X\sim\operatorname{Be}(\pi_X)$ e $Y\sim\operatorname{Be}(\pi_Y)$, e siano $X_1,\dots,X_{n_X}$ e $Y_1,\dots,Y_{n_Y}$ campioni indipendenti con $n_X,n_Y$ grandi. Siano $\hat{\pi}_X=\bar{X}$ e $\hat{\pi}_Y=\bar{Y}$.

**Ipotesi:**
$$
H_0:\pi_X=\pi_Y,\qquad H_1:\pi_X\neq\pi_Y.
$$

### 9.1 La quantità pivotale asintotica

Sotto $H_0$, $\pi_X=\pi_Y=\pi$. La varianza comune è stimata con
$$
\hat{\pi}=\frac{n_X\hat{\pi}_X+n_Y\hat{\pi}_Y}{n_X+n_Y}.
$$
Per il teorema del limite centrale e il teorema di Slutsky,
$$
Z_0=
\frac{\hat{\pi}_X-\hat{\pi}_Y}{
\sqrt{
\hat{\pi}(1-\hat{\pi})
\left(\dfrac{1}{n_X}+\dfrac{1}{n_Y}\right)
}
}
\overset{D}{\longrightarrow}\mathcal{N}(0,1).
$$
La stima $\hat{\pi}$ è la stima pooled, che è la scelta naturale per la varianza sotto l'ipotesi nulla.

**Regione di rifiuto.**
$$
|Z_0|>z_{1-\alpha/2}.
$$

> **Interpretazione grafica.**  
> La regione di rifiuto è una coda (o due code) della normale standard. La differenza rispetto al test per una proporzione è che qui la varianza è stimata con la pooled proportion, che combina le informazioni dei due campioni.

## 10. Test per dati appaiati

Siano $X_1,\dots,X_n$ e $Y_1,\dots,Y_n$ due campioni appaiati, con $X_i$ e $Y_i$ osservati sulla stessa unità. Sia
$$
\delta_i=X_i-Y_i,\qquad i=1,\dots,n.
$$
Supponiamo $\delta_i\sim\mathcal{N}(\mu_\delta,\sigma_\delta^2)$ i.i.d.

**Ipotesi:**
$$
H_0:\mu_\delta=0,\qquad H_1:\mu_\delta\neq 0.
$$

### 10.1 La quantità pivotale

Il problema si riduce a un test $t$ a una popolazione sulla variabile $\delta$. La statistica è
$$
T_0=\frac{\bar{\delta}}{S_\delta/\sqrt{n}},
$$
dove
$$
\bar{\delta}=\frac{1}{n}\sum_{i=1}^n\delta_i,
\qquad
S_\delta^2=\frac{1}{n-1}\sum_{i=1}^n(\delta_i-\bar{\delta})^2.
$$
Sotto $H_0$, $T_0\sim t_{n-1}$.

**Regione di rifiuto.**
$$
|T_0|>t_{n-1,1-\alpha/2}.
$$

> **Osservazione 10.1.**  
> Trattare dati appaiati come campioni indipendenti è scorretto, perché $X_i$ e $Y_i$ sono correlati. La differenza $\delta_i$ elimina la dipendenza e riconduce il problema a un test $t$ a una popolazione.

> **Interpretazione grafica.**  
> La regione di rifiuto è una coda (o due code) della distribuzione $t_{n-1}$. La differenza rispetto al test per una media è che qui la variabile di interesse è la differenza $\delta_i$, e la numerosità è il numero di coppie.

## 11. ANOVA a una via

Siano
$$
Y_{ij}\sim\mathcal{N}(\mu_j,\sigma^2),\qquad j=1,\dots,J,\quad i=1,\dots,n_j,
$$
con $\sigma^2$ comune e non nota. Sia $N=\sum_j n_j$.

**Ipotesi:**
$$
H_0:\mu_1=\dots=\mu_J,
\qquad
H_1:\text{almeno una media è diversa}.
$$

### 11.1 La quantità pivotale

La varianza pooled è
$$
S^2=\frac{D_{\text{entro}}}{N-J},
$$
dove $D_{\text{entro}}$ è la devianza entro i gruppi. Sotto $H_0$, tutte le medie sono uguali, e la devianza tra i gruppi $D_{\text{tra}}$ misura la variabilità dovuta alle differenze tra le medie campionarie. La statistica
$$
F=\frac{D_{\text{tra}}/(J-1)}{D_{\text{entro}}/(N-J)}
$$
confronta la variabilità tra gruppi con quella entro i gruppi. Sotto $H_0$,
$$
F\sim F_{J-1,N-J}.
$$
La distribuzione di $F$ **non dipende da $\mu$ né da $\sigma^2$**: è una quantità pivotale per l'ipotesi di uguaglianza delle medie.

### 11.2 Scomposizione della devianza

$$
\sum_{j=1}^J\sum_{i=1}^{n_j}(y_{ij}-\bar{y}_{..})^2
=
\underbrace{
\sum_{j=1}^J n_j(\bar{y}_{.j}-\bar{y}_{..})^2
}_{D_{\text{tra}}}
+
\underbrace{
\sum_{j=1}^J\sum_{i=1}^{n_j}(y_{ij}-\bar{y}_{.j})^2
}_{D_{\text{entro}}},
$$
dove
$$
\bar{y}_{.j}=\frac{1}{n_j}\sum_{i=1}^{n_j}y_{ij},
\qquad
\bar{y}_{..}=\frac{1}{N}\sum_{j=1}^J\sum_{i=1}^{n_j}y_{ij}.
$$

**Dimostrazione della scomposizione.**  
Si ha
$$
y_{ij}-\bar{y}_{..}
=
(y_{ij}-\bar{y}_{.j})+(\bar{y}_{.j}-\bar{y}_{..}).
$$
Elevando al quadrato e sommando,
$$
\sum_{j,i}(y_{ij}-\bar{y}_{..})^2
=
\sum_{j,i}(y_{ij}-\bar{y}_{.j})^2
+
\sum_{j,i}(\bar{y}_{.j}-\bar{y}_{..})^2
+
2\sum_{j,i}(y_{ij}-\bar{y}_{.j})(\bar{y}_{.j}-\bar{y}_{..}).
$$
Il termine misto è nullo perché
$$
\sum_{i=1}^{n_j}(y_{ij}-\bar{y}_{.j})=0
$$
per ogni $j$. Quindi la scomposizione vale. $\square$

### 11.3 Distribuzione sotto $H_0$

Sotto $H_0$, tutte le medie sono uguali a $\mu$. Allora
$$
\frac{D_{\text{entro}}}{\sigma^2}\sim\chi^2_{N-J},
\qquad
\frac{D_{\text{tra}}}{\sigma^2}\sim\chi^2_{J-1},
$$
e le due quantità sono indipendenti. Pertanto
$$
F=
\frac{
\dfrac{D_{\text{tra}}}{\sigma^2}\cdot\dfrac{1}{J-1}
}{
\dfrac{D_{\text{entro}}}{\sigma^2}\cdot\dfrac{1}{N-J}
}
\sim F_{J-1,N-J}.
$$

**Regione di rifiuto.**
$$
F>F_{J-1,N-J,1-\alpha}.
$$

> **Interpretazione grafica.**  
> La distribuzione $F_{J-1,N-J}$ è asimmetrica a destra, con una lunga coda destra. La regione di rifiuto è la **coda destra**: valori grandi di $F$ indicano che la variabilità tra i gruppi è molto più grande di quella entro i gruppi, il che è incompatibile con l'ipotesi che tutte le medie siano uguali. Il valore critico $F_{J-1,N-J,1-\alpha}$ è il punto che lascia l'area $\alpha$ nella coda destra.

## 12. Variabili multinomiali

Prima di introdurre i test del chi-quadro, formalizziamo il modello probabilistico che ne sta alla base.

### Definizione 12.1 — Prova multinomiale e distribuzione multinomiale

Siano $K\ge 2$ categorie possibili e sia
$$
\pi=(\pi_1,\dots,\pi_K),\qquad \pi_k>0,\qquad \sum_{k=1}^K\pi_k=1,
$$
un vettore di probabilità. Si considerano $n$ prove indipendenti, ciascuna delle quali produce uno e un solo esito tra le $K$ categorie, con probabilità $\pi_k$ di produrre l'esito $k$.

Sia
$$
N_k=\text{numero di prove che producono l'esito }k,
\qquad k=1,\dots,K.
$$
Il vettore aleatorio
$$
\mathbf{N}=(N_1,\dots,N_K)
$$
ha **distribuzione multinomiale** di parametri $n$ e $\pi$, e si scrive
$$
\mathbf{N}\sim\operatorname{Mult}(n,\pi).
$$

La funzione di probabilità congiunta è
$$
P(N_1=n_1,\dots,N_K=n_K)
=
\frac{n!}{n_1!\cdots n_K!}
\prod_{k=1}^K\pi_k^{n_k},
$$
per ogni $n_1,\dots,n_K\ge 0$ con $\sum_{k=1}^K n_k=n$, e vale $0$ altrimenti.

> **Osservazione 12.2.**  
> Per $K=2$ la distribuzione multinomiale si riduce alla **binomiale**: se $N_1\sim\operatorname{Bin}(n,\pi_1)$ e $N_2=n-N_1$, allora $(N_1,N_2)\sim\operatorname{Mult}(n,(\pi_1,1-\pi_1))$.

> **Osservazione 12.3.**  
> La condizione $\sum_{k=1}^K N_k=n$ è strutturale: il vettore $\mathbf{N}$ vive nel simplesso
> $$
> \left\{(n_1,\dots,n_K)\in\mathbb{Z}_{\ge 0}^K : \sum_{k=1}^K n_k=n\right\},
> $$
> che ha dimensione $K-1$. Questo spiega perché, nei test asintotici basati sulla multinomiale, i gradi di libertà sono $K-1$ e non $K$.

### Proposizione 12.4 — Momenti della multinomiale

Sia $\mathbf{N}\sim\operatorname{Mult}(n,\pi)$. Allora:

1. **Valore atteso:**
   $$
   E[N_k]=n\pi_k,\qquad k=1,\dots,K.
   $$
2. **Varianza:**
   $$
   \operatorname{Var}(N_k)=n\pi_k(1-\pi_k),\qquad k=1,\dots,K.
   $$
3. **Covarianza:**
   $$
   \operatorname{Cov}(N_i,N_j)=-n\pi_i\pi_j,\qquad i\neq j.
   $$

**Dimostrazione.**

Si osservi che
$$
N_k=\sum_{r=1}^n X_{r,k},
$$
dove $X_{r,k}=1$ se la $r$-esima prova produce l'esito $k$, e $X_{r,k}=0$ altrimenti. Per ogni $r$, il vettore $(X_{r,1},\dots,X_{r,K})$ ha distribuzione categoriale con
$$
P(X_{r,k}=1)=\pi_k.
$$
Inoltre le prove sono indipendenti, quindi i vettori $(X_{r,1},\dots,X_{r,K})$ sono indipendenti al variare di $r$.

**Punto 1.** Per linearità del valore atteso,
$$
E[N_k]=\sum_{r=1}^n E[X_{r,k}]=\sum_{r=1}^n\pi_k=n\pi_k.
$$

**Punto 2.** Poiché $X_{r,k}$ è una Bernoulli di parametro $\pi_k$, si ha $\operatorname{Var}(X_{r,k})=\pi_k(1-\pi_k)$. Per l'indipendenza tra le prove,
$$
\operatorname{Var}(N_k)=\sum_{r=1}^n\operatorname{Var}(X_{r,k})=n\pi_k(1-\pi_k).
$$

**Punto 3.** Per $i\neq j$, si ha $X_{r,i}X_{r,j}=0$ per ogni $r$, perché una singola prova non può produrre contemporaneamente due esiti distinti. Dunque
$$
E[X_{r,i}X_{r,j}]=0,
\qquad
E[X_{r,i}]E[X_{r,j}]=\pi_i\pi_j,
$$
e quindi
$$
\operatorname{Cov}(X_{r,i},X_{r,j})=-\pi_i\pi_j.
$$
Per l'indipendenza tra le prove,
$$
\operatorname{Cov}(N_i,N_j)
=
\sum_{r=1}^n\operatorname{Cov}(X_{r,i},X_{r,j})
=
-n\pi_i\pi_j.
$$
$\square$

> **Osservazione 12.5.**  
> Le covarianze sono negative: se una categoria ha più osservazioni del previsto, le altre devono necessariamente averne meno, perché il totale $n$ è fissato. Questo è il motivo per cui la matrice di covarianza di $\mathbf{N}$ è singolare (ha rango $K-1$), coerentemente con il fatto che $\mathbf{N}$ vive in dimensione $K-1$.

### Proposizione 12.6 — Distribuzioni marginali e condizionate

Sia $\mathbf{N}\sim\operatorname{Mult}(n,\pi)$.

1. **Marginali.** Per ogni $k$,
   $$
   N_k\sim\operatorname{Bin}(n,\pi_k).
   $$
2. **Marginali congiunte.** Per ogni sottoinsieme $S\subseteq\{1,\dots,K\}$, il vettore $(N_k)_{k\in S}$ ha distribuzione multinomiale di parametri $n$ e $(\pi_k)_{k\in S}$, con la convenzione che l'ultima componente rappresenta la somma delle categorie fuori da $S$.
3. **Condizionate.** Dato $N_k=m$, il vettore delle rimanenti componenti ha distribuzione multinomiale di parametri $m$ e probabilità rinormalizzate $\pi_j/(1-\pi_k)$, $j\neq k$.

**Idea della dimostrazione.**  
La marginalità segue raggruppando le categorie: $N_k$ conta i successi in $n$ prove bernoulliane di parametro $\pi_k$. La congiunta segue raggruppando le categorie in $S$ e nel complementare. La condizionata segue dalla definizione di probabilità condizionata e dal fatto che $N_k=m$ fissa il numero di prove disponibili per le altre categorie. $\square$

### Teorema 12.7 — Approssimazione normale della multinomiale

Sia $\mathbf{N}\sim\operatorname{Mult}(n,\pi)$. Allora
$$
\frac{\mathbf{N}-n\pi}{\sqrt{n}}\overset{D}{\longrightarrow}\mathcal{N}(\mathbf{0},\Sigma),
$$
dove
$$
\Sigma=\operatorname{diag}(\pi)-\pi\pi^T
$$
è una matrice di rango $K-1$, con $\Sigma_{kk}=\pi_k(1-\pi_k)$ e $\Sigma_{ij}=-\pi_i\pi_j$ per $i\neq j$.

**Dimostrazione.**  
Scriviamo
$$
\mathbf{N}=\sum_{r=1}^n\mathbf{X}_r,
$$
dove $\mathbf{X}_r=(X_{r,1},\dots,X_{r,K})$ sono vettori i.i.d. con $E[\mathbf{X}_r]=\pi$ e matrice di covarianza $\Sigma$. Per il teorema del limite centrale multidimensionale applicato alla successione $\{\mathbf{X}_r\}$,
$$
\sqrt{n}\left(\frac{1}{n}\sum_{r=1}^n\mathbf{X}_r-\pi\right)
\overset{D}{\longrightarrow}\mathcal{N}(\mathbf{0},\Sigma).
$$
Poiché $\sum_r\mathbf{X}_r=\mathbf{N}$, la tesi segue. La matrice $\Sigma$ ha rango $K-1$ perché $\Sigma\mathbf{1}=0$ (dove $\mathbf{1}$ è il vettore di tutti uno), coerentemente con il vincolo $\sum_k N_k=n$. $\square$

> **Osservazione 12.8.**  
> Il teorema 12.7 è la chiave per capire perché le statistiche del chi-quadro hanno distribuzione limite $\chi^2$ con $K-1$ gradi di libertà: la forma quadratica $\sum_k Z_k^2$, dove $Z_k=(N_k-n\pi_k)/\sqrt{n\pi_k}$, converge a una forma quadratica in una normale di rango $K-1$, che ha distribuzione $\chi^2_{K-1}$.

### 12.9 — La multinomiale come modello per le tabelle di contingenza

Sia $A$ una variabile categoriale con $R$ modalità e $B$ una variabile categoriale con $C$ modalità. Si osservano $n$ unità e si costruisce la tabella di contingenza $R\times C$ con frequenze $f_{rc}$.

Se le unità sono indipendenti e la coppia $(A,B)$ ha probabilità congiunte
$$
\pi_{rc}=P(A=r,\ B=c),\qquad r=1,\dots,R,\ c=1,\dots,C,
$$
allora il vettore delle frequenze
$$
\mathbf{f}=(f_{11},f_{12},\dots,f_{RC})
$$
ha distribuzione multinomiale con $K=RC$ categorie:
$$
\mathbf{f}\sim\operatorname{Mult}(n,(\pi_{11},\dots,\pi_{RC})).
$$
Infatti ogni unità cade in una e una sola cella della tabella, esattamente come una prova multinomiale cade in una delle $K$ categorie.

> **Osservazione 12.10.**  
> Il modello multinomiale è il punto di partenza del **test di indipendenza** e del **test di omogeneità**. Nel test di indipendenza, sotto $H_0$ si assume
> $$
> \pi_{rc}=\pi_{r+}\pi_{+c},
> $$
> dove $\pi_{r+}=\sum_c\pi_{rc}$ e $\pi_{+c}=\sum_r\pi_{rc}$ sono le probabilità marginali. Nel test di omogeneità, si assume invece che le probabilità condizionate di $B$ dato $A=r$ siano uguali per ogni $r$. In entrambi i casi, la statistica test confronta le frequenze osservate $f_{rc}$ con le frequenze attese sotto $H_0$, e la sua distribuzione asintotica è una $\chi^2$ con gradi di libertà determinati dalla dimensione del sottospazio dei vincoli.

### 12.11 — La multinomiale come modello per la bontà di adattamento

Sia $Y$ una variabile categoriale con $K$ modalità $l_1,\dots,l_K$. Si osserva un campione di $n$ unità e si contano le frequenze $f_1,\dots,f_K$ di ciascuna modalità. Allora
$$
\mathbf{f}=(f_1,\dots,f_K)\sim\operatorname{Mult}(n,(\pi_1,\dots,\pi_K)),
$$
dove $\pi_k=P(Y=l_k)$.

Se si vuole verificare che $Y$ abbia una certa distribuzione $p_0$, l'ipotesi nulla è
$$
H_0:\pi_k=p_0(l_k),\qquad k=1,\dots,K.
$$
La statistica del chi-quadro di Pearson confronta le frequenze osservate con quelle attese $f_k^*=np_0(l_k)$:
$$
X^2=\sum_{k=1}^K\frac{(f_k-f_k^*)^2}{f_k^*}.
$$
Sotto $H_0$, per $n$ grande,
$$
X^2\overset{D}{\longrightarrow}\chi^2_{K-1}.
$$
I gradi di libertà sono $K-1$ perché il vettore $\mathbf{f}$ ha $K$ componenti ma è soggetto al vincolo $\sum_k f_k=n$, che riduce la dimensione a $K-1$.

> **Osservazione 12.12.**  
> Se la distribuzione $p_0(y;\theta)$ dipende da un parametro $\theta$ ignoto, si stima $\theta$ con $\hat{\theta}$ e si usa
> $$
> X^2=\sum_{k=1}^K\frac{(f_k-np_0(l_k;\hat{\theta}))^2}{np_0(l_k;\hat{\theta})}.
> $$
> In questo caso i gradi di libertà si riducono di un'ulteriore unità per ogni parametro stimato: $K-1-p$, dove $p$ è il numero di parametri liberi di $\theta$.

## 13. Test del chi-quadro

### 13.1 Test di indipendenza

Siano $A$ e $B$ due variabili categoriali con modalità $A_1,\dots,A_R$ e $B_1,\dots,B_C$. Si osserva un campione di $n$ unità e si costruisce la tabella di contingenza con frequenze osservate $f_{rc}$.

**Ipotesi:**
$$
H_0:A\text{ e }B\text{ sono indipendenti},
\qquad
H_1:\text{non sono indipendenti}.
$$

**Derivazione della statistica test.** Sotto $H_0$, le probabilità congiunte sono
$$
\pi_{rc}=\pi_{r+}\pi_{+c}.
$$
Le probabilità marginali $\pi_{r+}$ e $\pi_{+c}$ sono ignote e vengono stimate con
$$
\hat{\pi}_{r+}=\frac{f_{r+}}{n},
\qquad
\hat{\pi}_{+c}=\frac{f_{+c}}{n}.
$$
Le frequenze attese sotto $H_0$ sono quindi
$$
f_{rc}^*=n\hat{\pi}_{r+}\hat{\pi}_{+c}=\frac{f_{r+}f_{+c}}{n}.
$$
La statistica del chi-quadro di Pearson misura la distanza tra frequenze osservate e attese:
$$
X^2=\sum_{r=1}^R\sum_{c=1}^C\frac{(f_{rc}-f_{rc}^*)^2}{f_{rc}^*}.
$$
Questa è la forma quadratica dei residui standardizzati. Sotto $H_0$, per $n$ grande, i residui $(f_{rc}-f_{rc}^*)/\sqrt{f_{rc}^*}$ sono asintoticamente normali con una matrice di covarianza di rango $(R-1)(C-1)$, e quindi
$$
X^2\overset{D}{\longrightarrow}\chi^2_{(R-1)(C-1)}.
$$
I gradi di libertà sono $(R-1)(C-1)$ e non $RC-1$ perché l'ipotesi di indipendenza impone $R-1+C-1=R+C-2$ vincoli aggiuntivi sui parametri.

**Regione di rifiuto.**
$$
X^2>\chi^2_{(R-1)(C-1),1-\alpha}.
$$

> **Interpretazione grafica.**  
> La distribuzione $\chi^2_{(R-1)(C-1)}$ è asimmetrica a destra. La regione di rifiuto è la **coda destra**: valori grandi di $X^2$ indicano che le frequenze osservate sono molto lontane da quelle attese sotto indipendenza, il che è incompatibile con $H_0$. Il p-value è l'area a destra di $X^2_{\text{oss}}$.

### 13.2 Test di omogeneità

Siano $A$ e $B$ due variabili categoriali, ma si fissano a priori le marginali di riga o di colonna. Si vuole verificare se la distribuzione di $B$ è la stessa in ogni gruppo definito da $A$.

**Ipotesi:**
$$
H_0:\pi_{r_1}=\pi_{r_2}\ \forall r_1,r_2,
\qquad
H_1:\text{non tutte uguali}.
$$

La statistica test è la stessa del test di indipendenza:
$$
X^2=\sum_{r,c}\frac{(f_{rc}-f_{rc}^*)^2}{f_{rc}^*},
\qquad
f_{rc}^*=\frac{f_{r+}f_{+c}}{n}.
$$
Sotto $H_0$,
$$
X^2\approx\chi^2_{(R-1)(C-1)}.
$$

> **Osservazione 13.1.**  
> La formula è identica al test di indipendenza, ma il disegno campionario è diverso: nell'omogeneità i totali di riga o colonna sono fissati, nell'indipendenza sono casuali. Asintoticamente la distribuzione è la stessa.

### 13.3 Test di bontà di adattamento

Si vuole verificare se una variabile $Y$ ha una certa distribuzione $p_0(y)$.

**Ipotesi:**
$$
H_0:p_Y(y)=p_0(y),
\qquad
H_1:p_Y(y)\neq p_0(y).
$$

Siano $l_1,\dots,l_K$ le modalità o classi e $f_1,\dots,f_K$ le frequenze osservate. Sotto $H_0$, le frequenze attese sono
$$
f_j^*=n\pi_j,
\qquad
\pi_j=P_{H_0}(Y=l_j).
$$

**Statistica test.**
$$
X^2=\sum_{j=1}^K\frac{(f_j-f_j^*)^2}{f_j^*}.
$$
Sotto $H_0$, per $n$ grande,
$$
X^2\overset{D}{\longrightarrow}\chi^2_{K-1}.
$$

**Regione di rifiuto.**
$$
X^2>\chi^2_{K-1,1-\alpha}.
$$

> **Osservazione 13.2.**  
> Se la distribuzione $p_0(y;\theta)$ dipende da un parametro $\theta$ ignoto, si stima $\theta$ con $\hat{\theta}$ e si applica il test con $\hat{\theta}$. I gradi di libertà si riducono del numero di parametri stimati: $K-1-p$.

### 13.4 Esempi

**Esempio 13.3 — Lancio di un dado.**  
Si lancia un dado $n=120$ volte. Le $K=6$ categorie sono le facce, con $\pi_k=1/6$. Le frequenze attese sono $f_k^*=20$. Se le frequenze osservate sono
$$
\mathbf{f}=(18,22,19,24,17,20),
$$
allora
$$
X^2=\frac{(18-20)^2}{20}+\frac{(22-20)^2}{20}+\dots+\frac{(20-20)^2}{20}
=\frac{4+4+1+16+9+0}{20}=1.7.
$$
Per $\alpha=0.05$, $\chi^2_{5,0.95}\approx 11.07$. Poiché $1.7<11.07$, non si rifiuta l'ipotesi che il dado sia equilibrato.

> **Interpretazione grafica.**  
> Il valore osservato $X^2_{\text{oss}}=1.7$ cade nella parte centrale della distribuzione $\chi^2_5$, molto lontano dalla coda destra. Il p-value è l'area a destra di $1.7$, che è grande (circa $0.89$). Quindi non c'è evidenza contro $H_0$.

**Esempio 13.4 — Tabella $2\times 2$.**  
Per una tabella $2\times 2$ con frequenze $f_{rc}$, la statistica
$$
X^2=\sum_{r=1}^2\sum_{c=1}^2\frac{(f_{rc}-f_{rc}^*)^2}{f_{rc}^*}
$$
converge a una $\chi^2$ con $(R-1)(C-1)=1$ grado di libertà. La riduzione da $K-1=3$ a $1$ riflette i vincoli aggiuntivi imposti dall'ipotesi di indipendenza.

> **Osservazione 13.5 — Regola pratica.**  
> L'approssimazione $\chi^2$ è affidabile quando le frequenze attese $f_k^*=n\pi_k$ sono sufficientemente grandi, tipicamente $f_k^*\ge 5$ per ogni $k$. Se alcune frequenze attese sono troppo piccole, si possono raggruppare categorie adiacenti oppure usare test esatti (ad esempio il test esatto di Fisher per tabelle $2\times 2$).
