# Intervalli di confidenza e test asintotici basati sulla verosimiglianza

---

## 0. Distribuzioni campionarie di riferimento

### Osservazione 0.1

Sotto le condizioni di regolarità, valgono i seguenti risultati asintotici:

1. **Normalità asintotica della SMV**:
$$
\sqrt{n}(\hat\theta_n-\theta_0)\xrightarrow{D}\mathcal{N}\bigl(0,\,i(\theta_0)^{-1}\bigr),
$$
dove $i(\theta_0)$ è l'informazione attesa di una singola osservazione. Equivalentemente,
$$
\hat\theta_n\overset{\text{approx}}{\sim}\mathcal{N}\bigl(\theta_0,\,I(\theta_0)^{-1}\bigr),
$$
con $I(\theta_0)=n\,i(\theta_0)$.

2. **Normalità asintotica della funzione punteggio**:
$$
\frac{\ell^*(\theta_0;\mathbf{Y})}{\sqrt{I(\theta_0)}}\xrightarrow{D}\mathcal{N}(0,1),
$$
dove $\ell^*(\theta)=\frac{\partial}{\partial\theta}\ell(\theta;\mathbf{Y})$ è la funzione punteggio.

Queste distribuzioni dipendono da $I(\theta_0)$, che è ignota. In pratica si sostituisce con uno stimatore consistente, tipicamente $J(\hat\theta_n)$ (informazione osservata valutata in $\hat\theta_n$) o $I(\hat\theta_n)$ (informazione attesa valutata in $\hat\theta_n$). Per il teorema di Slutsky, le conclusioni asintotiche restano valide.

## 1. Statistica di Wald

### Definizione 1.1 — Quantità pivotale di Wald

La **quantità pivotale di Wald** per il parametro scalare $\theta$ è
$$
r_e(\theta)=\sqrt{J(\hat\theta)}\,(\hat\theta-\theta),
$$
dove $\hat\theta$ è la SMV e $J(\hat\theta)=-\ell''(\hat\theta;\mathbf{Y})$ è l'informazione osservata valutata in $\hat\theta$. Sotto le condizioni di regolarità,
$$
r_e(\theta)\overset{\text{approx}}{\sim}\mathcal{N}(0,1).
$$
In modo equivalente si può usare
$$
r'_e(\theta)=\sqrt{I(\hat\theta)}\,(\hat\theta-\theta)\overset{\text{approx}}{\sim}\mathcal{N}(0,1),
$$
dove $I(\hat\theta)$ è l'informazione attesa valutata in $\hat\theta$.

### Costruzione dell'intervallo di confidenza

Fissato il livello $1-\alpha$, si ha
$$
1-\alpha\approx P\bigl(-z_{1-\alpha/2}\le r_e(\theta)\le z_{1-\alpha/2}\bigr).
$$
Risolvendo la disuguaglianza rispetto a $\theta$:
$$
1-\alpha\approx P\left(\hat\theta-\frac{z_{1-\alpha/2}}{\sqrt{J(\hat\theta)}}\le\theta\le\hat\theta+\frac{z_{1-\alpha/2}}{\sqrt{J(\hat\theta)}}\right).
$$
L'**intervallo di confidenza di Wald** di livello $1-\alpha$ è
$$
IC_{1-\alpha}(\theta)=\left[\hat\theta-\frac{z_{1-\alpha/2}}{\sqrt{J(\hat\theta)}},\ \hat\theta+\frac{z_{1-\alpha/2}}{\sqrt{J(\hat\theta)}}\right].
$$
Analogamente, usando $I(\hat\theta)$ al posto di $J(\hat\theta)$, si ottiene un intervallo asintoticamente equivalente.

### Test di ipotesi

Per verificare $H_0:\theta=\theta_0$ contro $H_1:\theta\ne\theta_0$, si usa la statistica test
$$
r_e(\theta_0)=\sqrt{J(\hat\theta)}\,(\hat\theta-\theta_0)\overset{H_0}{\sim}\mathcal{N}(0,1).
$$
La regione di rifiuto di livello $\alpha$ è
$$
|r_e(\theta_0)|>z_{1-\alpha/2}.
$$
Il p-value è
$$
\alpha_{\text{oss}}=2P\bigl(Z>|r_e(\theta_0)^{\text{oss}}|\bigr)=2\bigl(1-\Phi(|r_e(\theta_0)^{\text{oss}}|)\bigr).
$$
Per test unilaterali:
- $H_1:\theta>\theta_0$: si rifiuta se $r_e(\theta_0)>z_{1-\alpha}$, p-value $P(Z>r_e(\theta_0)^{\text{oss}})$;
- $H_1:\theta<\theta_0$: si rifiuta se $r_e(\theta_0)<-z_{1-\alpha}$, p-value $P(Z<r_e(\theta_0)^{\text{oss}})$.

### Forma quadratica $W_e(\theta)$

Poiché se $Z\sim\mathcal{N}(0,1)$ allora $Z^2\sim\chi^2_1$, si può equivalentemente usare la statistica
$$
W_e(\theta)=r_e(\theta)^2=J(\hat\theta)(\hat\theta-\theta)^2\overset{\text{approx}}{\sim}\chi^2_1.
$$
Per un test bilaterale, si rifiuta $H_0$ se $W_e(\theta_0)>\chi^2_{1,1-\alpha}$. L'intervallo di confidenza è
$$
IC_{1-\alpha}(\theta)=\{\theta:W_e(\theta)\le\chi^2_{1,1-\alpha}\}.
$$
La scelta tra $r_e$ e $W_e$ è solo di convenienza.

> **Osservazione 1.2 — Interpretazione.**  
> La statistica di Wald misura la distanza standardizzata tra la SMV e il valore ipotizzato $\theta_0$. Grandi valori indicano che la stima è lontana da $\theta_0$ più di quanto ci si aspetti per puro effetto del caso. L'intervallo di Wald è simmetrico intorno a $\hat\theta$ e ha ampiezza proporzionale a $1/\sqrt{J(\hat\theta)}$: più informazione c'è, più l'intervallo è stretto.

> **Osservazione 1.3 — Limiti.**  
> L'intervallo di Wald è basato sull'approssimazione parabolica della log-verosimiglianza. Quando tale approssimazione è scarsa (campioni piccoli, verosimiglianza asimmetrica), l'intervallo può essere inaccurato. In particolare:
> - può includere valori esterni allo spazio parametrico (es. proporzioni negative);
> - può non includere valori con verosimiglianza maggiore di quelli inclusi.

## 2. Statistica score di Rao

### Definizione 2.1 — Quantità pivotale score

La **quantità pivotale score** (o **di Rao**) è
$$
r_u(\theta)=\frac{\ell^*(\theta)}{\sqrt{I(\theta)}}\overset{\text{approx}}{\sim}\mathcal{N}(0,1),
$$
dove $\ell^*(\theta)$ è la funzione punteggio valutata in $\theta$ e $I(\theta)$ è l'informazione attesa. In pratica si sostituisce $I(\theta)$ con $I(\hat\theta)$ o $J(\hat\theta)$, ottenendo
$$
r_u(\theta)=\frac{\ell^*(\theta)}{\sqrt{I(\hat\theta)}}\overset{\text{approx}}{\sim}\mathcal{N}(0,1).
$$

### Intervallo di confidenza

L'intervallo di confidenza score è
$$
IC_{1-\alpha}(\theta)=\{\theta:|r_u(\theta)|\le z_{1-\alpha/2}\}
=\left\{\theta:\frac{|\ell^*(\theta)|}{\sqrt{I(\hat\theta)}}\le z_{1-\alpha/2}\right\}.
$$
A differenza dell'intervallo di Wald, gli estremi non sono esplicitabili in forma chiusa e devono essere determinati numericamente (risolvendo l'equazione $\ell^*(\theta)^2/I(\hat\theta)=\chi^2_{1,1-\alpha}$).

### Test di ipotesi

Per $H_0:\theta=\theta_0$, la statistica test è
$$
r_u(\theta_0)=\frac{\ell^*(\theta_0)}{\sqrt{I(\hat\theta)}}\overset{H_0}{\sim}\mathcal{N}(0,1),
$$
dove in pratica si usa $I(\theta_0)$ se nota, altrimenti $I(\hat\theta)$. La regione di rifiuto bilaterale è $|r_u(\theta_0)|>z_{1-\alpha/2}$. Il p-value è $2(1-\Phi(|r_u(\theta_0)^{\text{oss}}|))$. Per test unilaterali si procede come per Wald.

### Forma quadratica $W_u(\theta)$

Si può equivalentemente usare
$$
W_u(\theta)=r_u(\theta)^2=\frac{\ell^*(\theta)^2}{I(\hat\theta)}\overset{\text{approx}}{\sim}\chi^2_1,
$$
e l'intervallo di confidenza è $\{\theta:W_u(\theta)\le\chi^2_{1,1-\alpha}\}$.

> **Osservazione 2.2 — Equivalenza asintotica con Wald.**  
> Le statistiche $r_e$ e $r_u$ sono asintoticamente equivalenti. Infatti, sviluppando $\ell^*(\theta)$ in serie di Taylor intorno a $\hat\theta$ e usando $\ell^*(\hat\theta)=0$:
> $$
> \ell^*(\theta)\approx\ell^{**}(\hat\theta)(\theta-\hat\theta)=-J(\hat\theta)(\theta-\hat\theta).
> $$
> Quindi
> $$
> r_u(\theta)=\frac{\ell^*(\theta)}{\sqrt{I(\hat\theta)}}\approx\frac{-J(\hat\theta)(\theta-\hat\theta)}{\sqrt{I(\hat\theta)}}=-\sqrt{J(\hat\theta)}(\theta-\hat\theta)=-r_e(\theta),
> $$
> e $J(\hat\theta)/I(\hat\theta)\xrightarrow{p}1$. Dunque $r_u$ e $r_e$ hanno la stessa distribuzione limite.

> **Osservazione 2.3 — Vantaggio della score.**  
> La statistica score ha il vantaggio di non richiedere il calcolo della SMV $\hat\theta$ per essere valutata sotto $H_0$: si valuta $\ell^*(\theta_0)$ e $I(\theta_0)$ (o una loro stima) direttamente in $\theta_0$. Questo è utile quando la SMV è difficile da calcolare ma la funzione punteggio è semplice.

## 3. Rapporto di verosimiglianza

### Definizione 3.1 — Statistica del rapporto di verosimiglianza

La **statistica del rapporto di verosimiglianza** (log-likelihood ratio) è
$$
W(\theta)=2\bigl(\ell(\hat\theta)-\ell(\theta)\bigr)\overset{\text{approx}}{\sim}\chi^2_1.
$$
Essa misura di quanto la log-verosimiglianza massima supera il valore in $\theta$. Per costruzione $W(\theta)\ge0$ e $W(\hat\theta)=0$.

### Intervallo di confidenza

L'intervallo di confidenza basato sul rapporto di verosimiglianza è
$$
IC_{1-\alpha}(\theta)=\{\theta:W(\theta)\le\chi^2_{1,1-\alpha}\}
=\{\theta:2(\ell(\hat\theta)-\ell(\theta))\le\chi^2_{1,1-\alpha}\}.
$$
Equivalentemente, in termini di verosimiglianza normalizzata $L_N(\theta)=L(\theta)/L(\hat\theta)$:
$$
IC_{1-\alpha}(\theta)=\{\theta:\log L_N(\theta)\ge-\tfrac12\chi^2_{1,1-\alpha}\}.
$$
Gli estremi non sono esplicitabili in forma chiusa in generale e vanno determinati numericamente (risolvendo $W(\theta)=\chi^2_{1,1-\alpha}$).

### Test di ipotesi

Per $H_0:\theta=\theta_0$, la statistica test è
$$
W(\theta_0)=2\bigl(\ell(\hat\theta)-\ell(\theta_0)\bigr)\overset{H_0}{\sim}\chi^2_1.
$$
La regione di rifiuto bilaterale di livello $\alpha$ è
$$
W(\theta_0)>\chi^2_{1,1-\alpha}.
$$
Il p-value è
$$
\alpha_{\text{oss}}=P\bigl(\chi^2_1>W(\theta_0)^{\text{oss}}\bigr).
$$
Equivalentemente, si rifiuta $H_0$ se $\theta_0\notin IC_{1-\alpha}(\theta)$.

### Versione unilaterale: radice con segno

Poiché $W(\theta)$ è bilaterale, per test unilaterali si usa la **radice con segno**:
$$
r(\theta)=\operatorname{sgn}(\hat\theta-\theta)\sqrt{W(\theta)}\overset{\text{approx}}{\sim}\mathcal{N}(0,1),
$$
dove $\operatorname{sgn}(x)=1$ se $x>0$, $-1$ se $x<0$. Per $H_0:\theta=\theta_0$:
- $H_1:\theta>\theta_0$: si rifiuta se $r(\theta_0)>z_{1-\alpha}$;
- $H_1:\theta<\theta_0$: si rifiuta se $r(\theta_0)<-z_{1-\alpha}$.

Il p-value è rispettivamente $P(Z>r(\theta_0)^{\text{oss}})$ e $P(Z<r(\theta_0)^{\text{oss}})$.

> **Osservazione 3.2 — Vantaggi del rapporto di verosimiglianza.**  
> La statistica $W$ è basata direttamente sulla forma della log-verosimiglianza, non sulla sua approssimazione parabolica. Di conseguenza:
> - gli intervalli di confidenza non sono necessariamente simmetrici e possono adattarsi all'asimmetria della verosimiglianza;
> - la regione di confidenza è invariante per riparametrizzazione (come si vedrà nella Sezione 4);
> - è generalmente più accurata della statistica di Wald per campioni piccoli o verosimiglianze asimmetriche.
>
> Lo svantaggio è che richiede la soluzione numerica di un'equazione non lineare per gli estremi dell'intervallo.

## 4. Invarianza per riparametrizzazione

### Proposizione 4.1 — Comportamento sotto riparametrizzazione

Sia $\psi=h(\theta)$ una riparametrizzazione biunivoca, con inversa $g=h^{-1}$.

1. **Wald e score non sono invarianti.** Le quantità $r_e(\theta)$, $W_e(\theta)$, $r_u(\theta)$, $W_u(\theta)$ non si trasformano in modo coerente sotto riparametrizzazione. In particolare, gli intervalli di confidenza e le regioni di rifiuto cambiano se si cambia parametrizzazione, anche se il modello è lo stesso.
2. **Il rapporto di verosimiglianza è invariante.** La statistica $W(\theta)$ è invariante: $W(\psi)=W(g(\psi))=W(\theta)$. Di conseguenza, gli intervalli di confidenza e le regioni di rifiuto basati su $W$ sono invarianti per riparametrizzazione: se $\Theta(y)$ è l'intervallo per $\theta$, allora l'intervallo per $\psi$ è $\Psi(y)=h(\Theta(y))$.

**Dimostrazione (punto 2).**  
Per definizione,
$$
W(\psi)=2\bigl(\ell(\hat\psi)-\ell(\psi)\bigr).
$$
Poiché la verosimiglianza è invariante e $\hat\psi=h(\hat\theta)$, si ha $\ell(\hat\psi)=\ell(\hat\theta)$ e $\ell(\psi)=\ell(g(\psi))=\ell(\theta)$. Quindi
$$
W(\psi)=2\bigl(\ell(\hat\theta)-\ell(\theta)\bigr)=W(\theta).
$$
Pertanto l'insieme $\{\psi:W(\psi)\le c\}$ è l'immagine tramite $h$ dell'insieme $\{\theta:W(\theta)\le c\}$. $\square$

> **Osservazione 4.2 — Rilevanza pratica.**  
> L'invarianza è una proprietà desiderabile: le conclusioni inferenziali non devono dipendere dalla parametrizzazione scelta. Il rapporto di verosimiglianza la garantisce, mentre Wald e score no. Questo è uno dei motivi per cui il rapporto di verosimiglianza è spesso preferito, nonostante richieda calcoli numerici.

## 5. Confronto e considerazioni pratiche

### Osservazione 5.1 — Accuratezza

- **Wald** e **score** sono basati sull'approssimazione parabolica della log-verosimiglianza. Sono accurati quando la verosimiglianza è approssimativamente simmetrica e il campione è grande.
- **Rapporto di verosimiglianza** è basato sulla verosimiglianza reale (anche se la distribuzione limite deriva dall'approssimazione parabolica). È generalmente più accurato, specialmente per campioni piccoli o verosimiglianze asimmetriche.

### Osservazione 5.2 — Simmetria degli intervalli

- Gli intervalli di Wald sono simmetrici intorno a $\hat\theta$. Se la verosimiglianza è asimmetrica, l'intervallo può non riflettere la reale incertezza (può includere valori meno plausibili e escluderne di più plausibili).
- Gli intervalli basati sul rapporto di verosimiglianza non sono necessariamente simmetrici e si adattano alla forma della verosimiglianza.
- Gli intervalli score non sono simmetrici in generale, ma la loro costruzione richiede la soluzione numerica.

### Osservazione 5.3 — Calcolo

- **Wald**: richiede solo $\hat\theta$, $J(\hat\theta)$ (o $I(\hat\theta)$) e i quantili normali. È il più semplice da calcolare.
- **Score**: richiede $\ell^*(\theta_0)$ e $I(\theta_0)$ (o una stima). Non richiede $\hat\theta$ per il test, ma per l'intervallo sì (per la stima di $I$).
- **Rapporto di verosimiglianza**: richiede $\ell(\hat\theta)$ e $\ell(\theta)$, e la soluzione numerica di un'equazione per gli estremi dell'intervallo. È il più oneroso computazionalmente, ma il più accurato.

### Osservazione 5.4 — Multimodalità

Se la log-verosimiglianza è multimodale, l'insieme $\{\theta:W(\theta)\le c\}$ può essere **sconnesso** (unione di più intervalli). In tal caso si parla più propriamente di **regione di confidenza** che di intervallo. Gli intervalli di Wald, essendo simmetrici e centrati in $\hat\theta$, non riescono a catturare la struttura multimodale.

## 6. Esempi

### Esempio 6.1 — Bernoulli

Sia $Y_1,\dots,Y_n$ i.i.d. $\operatorname{Be}(\pi)$. La SMV è $\hat\pi=\bar y$. L'informazione osservata è
$$
J(\pi)=\frac{\sum_i y_i}{\pi^2}+\frac{n-\sum_i y_i}{(1-\pi)^2}.
$$
**Intervallo di Wald**:
$$
IC_{1-\alpha}(\pi)=\hat\pi\pm z_{1-\alpha/2}\sqrt{\frac{\hat\pi(1-\hat\pi)}{n}}.
$$
Per $n=40$, $\sum_i y_i=3$, $\hat\pi=0.075$, un IC al 95% è approssimativamente $(-0.007,\,0.157)$ (si noti che include valori negativi, fuori dallo spazio parametrico $[0,1]$).

**Intervallo basato sul rapporto di verosimiglianza**:
$$
IC_{1-\alpha}(\pi)=\left\{\pi:2\bigl(\ell(\hat\pi)-\ell(\pi)\bigr)\le\chi^2_{1,1-\alpha}\right\}.
$$
Con i dati dell'esempio, l'intervallo è $(0.016,\,0.169)$, che resta dentro $[0,1]$ e riflette l'asimmetria della verosimiglianza.

### Esempio 6.2 — Esponenziale

Sia $Y_1,\dots,Y_n$ i.i.d. $\operatorname{Esp}(\mu)$ con densità $p_Y(y;\mu)=\frac1\mu e^{-y/\mu}$. La SMV è $\hat\mu=\bar y$. L'informazione osservata è
$$
J(\mu)=\frac{n}{\mu^2}-\frac{2\sum_i y_i}{\mu^3}+\frac{\sum_i y_i^2}{\mu^4}\cdot\text{(termine)}.
$$
Per $n=10$, $\bar y=1$, un IC di Wald al 95% per $\mu$ è circa $(0.38,\,1.62)$. L'intervallo basato sul rapporto di verosimiglianza è leggermente diverso e non simmetrico.

### Esempio 6.3 — Confronto tra Wald e LR

Per la Bernoulli con $n=40$, $\sum y_i=3$:
- Wald: $(-0.007,\,0.157)$;
- LR: $(0.016,\,0.169)$.

L'intervallo di Wald include valori negativi, chiaramente impossibili. L'intervallo LR è più plausibile e riflette l'asimmetria della verosimiglianza.
