# Esercitazione sulla stima per intervalli

## Esercizio 0 — Quantità pivotale per una uniforme

Sia $Y\sim U(0,\theta)$, con $\theta>0$. Determinare una quantità pivotale per costruire un intervallo di confidenza per $\theta$.

### Soluzione

Sia
$$
Y_{(n)}=\max_{1\le i\le n}Y_i.
$$
La densità del massimo è
$$
p_{Y_{(n)}}(y;\theta)=\frac{n y^{n-1}}{\theta^n},
\qquad 0<y<\theta.
$$
Consideriamo
$$
X=\frac{Y_{(n)}}{\theta}.
$$
Allora $Y_{(n)}=\theta X$ e, per $0<x<1$,
$$
p_X(x)=p_{Y_{(n)}}(\theta x;\theta)\cdot \theta
=
\frac{n(\theta x)^{n-1}}{\theta^n}\cdot \theta
=
n x^{n-1}.
$$
Dunque
$$
X=\frac{Y_{(n)}}{\theta}\sim \operatorname{Beta}(n,1),
$$
con funzione di ripartizione
$$
F_X(x)=x^n,\qquad 0<x<1.
$$
Poiché $X$ dipende dal campione solo attraverso $Y_{(n)}$ e dal parametro $\theta$, ma la sua distribuzione non dipende da $\theta$, $X$ è una quantità pivotale.

Siano $q_{\alpha/2}$ e $q_{1-\alpha/2}$ i quantili di $X$:
$$
q_{\alpha/2}=\left(\frac{\alpha}{2}\right)^{1/n},
\qquad
q_{1-\alpha/2}=\left(1-\frac{\alpha}{2}\right)^{1/n}.
$$
Allora
$$
1-\alpha
=
P\left(q_{\alpha/2}<\frac{Y_{(n)}}{\theta}<q_{1-\alpha/2}\right).
$$
Isolando $\theta$:
$$
1-\alpha
=
P\left(
\frac{Y_{(n)}}{q_{1-\alpha/2}}
<
\theta
<
\frac{Y_{(n)}}{q_{\alpha/2}}
\right).
$$
Pertanto un intervallo di confidenza di livello $1-\alpha$ per $\theta$ è
$$
\operatorname{IC}_{1-\alpha}(\theta)
=
\left[
\frac{Y_{(n)}}{\left(1-\frac{\alpha}{2}\right)^{1/n}},
\frac{Y_{(n)}}{\left(\frac{\alpha}{2}\right)^{1/n}}
\right].
$$

> **Osservazione.**  
> Si può anche costruire un intervallo unilaterale esatto:
> $$
> P\left(Y_{(n)}\le \theta\le \frac{Y_{(n)}}{\alpha^{1/n}}\right)=1-\alpha.
> $$
> Infatti $Y_{(n)}\le \theta$ quasi certamente e
> $$
> P\left(\frac{Y_{(n)}}{\theta}\ge \alpha^{1/n}\right)
> =
> 1-\left(\alpha^{1/n}\right)^n
> =
> 1-\alpha.
> $$

## Esercizio 1 — Errore di stima della media

Una macchina per il caffè eroga quantità con media ignota $\mu$ e varianza nota $\sigma^2=1.8$. Si osservano $n=50$ erogazioni indipendenti e si stima $\mu$ con
$$
\overline{Y}=\frac{1}{n}\sum_{i=1}^n Y_i.
$$
Entro quali limiti, con probabilità approssimativamente $0.90$, è compreso l'errore di stima $\overline{Y}-\mu$?

### Soluzione

Per il teorema del limite centrale,
$$
\frac{\overline{Y}-\mu}{\sigma/\sqrt{n}}
\overset{D}{\longrightarrow}
\mathcal N(0,1).
$$
Cerchiamo $a>0$ tale che
$$
P(|\overline{Y}-\mu|<a)\approx 0.90.
$$
Equivalentemente,
$$
P\left(
-\frac{a}{\sigma/\sqrt n}
<
Z
<
\frac{a}{\sigma/\sqrt n}
\right)\approx 0.90,
\qquad Z\sim \mathcal N(0,1).
$$
Poiché $z_{0.95}\approx 1.64485$,
$$
\frac{a}{\sigma/\sqrt n}=z_{0.95}.
$$
Con $\sigma^2=1.8$ e $n=50$,
$$
\frac{\sigma}{\sqrt n}
=
\sqrt{\frac{1.8}{50}}
=
\sqrt{0.036}
\approx 0.1897367.
$$
Dunque
$$
a\approx 1.64485\cdot 0.1897367\approx 0.3121.
$$
Pertanto
$$
P\left(-0.3121<\overline{Y}-\mu<0.3121\right)\approx 0.90.
$$

## Esercizio 2 — Intervallo per la media e trasformazione monotona

Sia $Y\sim\mathcal N(\mu,9)$ e si osservi il campione
$$
1.2,\quad 3.4,\quad 0.6,\quad 5.6.
$$
1. Costruire un intervallo di confidenza di livello $0.95$ per $\mu$.
2. Posto
$$
\lambda=\tau(\mu)=\frac{e^\mu}{1+e^\mu},
$$
ottenere un intervallo di confidenza per $\lambda$.

### Soluzione

La media campionaria è
$$
\overline y=\frac{1.2+3.4+0.6+5.6}{4}=2.7.
$$
Poiché $\sigma^2=9$ è nota,
$$
\overline{Y}\sim\mathcal N\left(\mu,\frac{9}{n}\right).
$$
La quantità pivotale è
$$
Q(\overline{Y},\mu)
=
\frac{\overline{Y}-\mu}{\sqrt{9/n}}
\sim\mathcal N(0,1).
$$
Per $1-\alpha=0.95$ si ha $z_{0.975}=1.96$. Allora
$$
0.95
=
P\left(
-1.96<
\frac{\overline{Y}-\mu}{\sqrt{9/n}}
<
1.96
\right),
$$
da cui
$$
\operatorname{IC}_{0.95}(\mu)
=
\left[
\overline{Y}-1.96\sqrt{\frac{9}{n}},
\overline{Y}+1.96\sqrt{\frac{9}{n}}
\right].
$$
Con $n=4$ e $\overline y=2.7$,
$$
\operatorname{IC}_{0.95}(\mu)
=
\left[
2.7-1.96\cdot\frac{3}{2},
2.7+1.96\cdot\frac{3}{2}
\right]
=
(-0.24,\;5.64).
$$

Per la trasformazione, si ha
$$
\tau'(\mu)
=
\frac{e^\mu(1+e^\mu)-e^{2\mu}}{(1+e^\mu)^2}
=
\frac{e^\mu}{(1+e^\mu)^2}>0.
$$
Dunque $\tau$ è strettamente crescente. Allora
$$
\operatorname{IC}_{0.95}(\lambda)
=
\left[
\tau(-0.24),\tau(5.64)
\right]
\approx
(0.4403,\;0.9965).
$$

## Esercizio 3 — Intervallo esatto per la varianza

Si consideri un campione
$$
1.2,\quad 3.4,\quad 0.6,\quad 5.6
$$
da una popolazione $\mathcal N(\mu,\sigma^2)$. Costruire un intervallo di confidenza esatto di livello $0.95$ per $\sigma^2$.

### Soluzione

Si ha
$$
\overline y=2.7,
\qquad
S^2=\frac{1}{n-1}\sum_{i=1}^n (y_i-\overline y)^2.
$$
Le deviazioni dalla media sono
$$
-1.5,\quad 0.7,\quad -2.1,\quad 2.9,
$$
i cui quadrati sommano
$$
2.25+0.49+4.41+8.41=15.56.
$$
Quindi
$$
S^2=\frac{15.56}{3}=5.186667.
$$
La quantità pivotale è
$$
Q=\frac{(n-1)S^2}{\sigma^2}\sim\chi^2_{n-1}.
$$
Con $n=4$, $n-1=3$. Per $1-\alpha=0.95$,
$$
\chi^2_{3,0.025}\approx 0.2158,
\qquad
\chi^2_{3,0.975}\approx 9.3484.
$$
Allora
$$
0.95
=
P\left(
\chi^2_{3,0.025}
<
\frac{3S^2}{\sigma^2}
<
\chi^2_{3,0.975}
\right),
$$
da cui
$$
\operatorname{IC}_{0.95}(\sigma^2)
=
\left[
\frac{3S^2}{\chi^2_{3,0.975}},
\frac{3S^2}{\chi^2_{3,0.025}}
\right].
$$
Sostituendo,
$$
\operatorname{IC}_{0.95}(\sigma^2)
\approx
\left[
\frac{15.56}{9.3484},
\frac{15.56}{0.2158}
\right]
\approx
(1.664,\;72.10).
$$
Per la deviazione standard,
$$
\operatorname{IC}_{0.95}(\sigma)
\approx
(\sqrt{1.664},\sqrt{72.10})
\approx
(1.29,\;8.49).
$$

## Esercizio 4 — Pasticcini: intervallo, numerosità e dimezzamento

Si prelevano $n=55$ pasticcini. Il peso medio campionario è $\overline y=35$ g. Si sa che lo scarto quadratico medio del peso di tutti i pasticcini è $\sigma=4$ g.

1. Trovare l'intervallo di confidenza al $98\%$ per il peso medio.
2. Di quanto deve aumentare $n$ affinché l'ampiezza dell'intervallo si dimezzi?
3. Quanti pasticcini occorre ancora estrarre affinché
$$
P(|\overline{Y}-\mu|<1)\approx 0.96?
$$

### Soluzione

Poiché $\sigma=4$ è noto,
$$
\frac{\overline{Y}-\mu}{\sigma/\sqrt n}
\sim\mathcal N(0,1)
$$
in modo esatto se i pesi sono normali, approssimato per il teorema del limite centrale altrimenti.

Per $1-\alpha=0.98$, $z_{0.99}\approx 2.32635$. Dunque
$$
\operatorname{IC}_{0.98}(\mu)
=
\left[
\overline{Y}\pm 2.32635\frac{4}{\sqrt{55}}
\right].
$$
Con $\overline y=35$,
$$
2.32635\frac{4}{\sqrt{55}}
\approx
2.32635\cdot 0.53936
\approx
1.254.
$$
Pertanto
$$
\operatorname{IC}_{0.98}(\mu)
\approx
(33.746,\;36.254).
$$

L'ampiezza è
$$
W=2\cdot 2.32635\frac{4}{\sqrt{55}}
\approx 2.508.
$$
Per dimezzarla, cioè $W'=1.254$, occorre $n'$ tale che
$$
2\cdot 2.32635\frac{4}{\sqrt{n'}}
=
\frac{1}{2}\cdot 2\cdot 2.32635\frac{4}{\sqrt{55}}.
$$
Semplificando,
$$
\frac{1}{\sqrt{n'}}
=
\frac{1}{2\sqrt{55}},
$$
da cui
$$
n'=4\cdot 55=220.
$$
Quindi occorre passare da $55$ a $220$ unità: bisogna aggiungere $165$ pasticcini.

Infine, si vuole
$$
P(|\overline{Y}-\mu|<1)\approx 0.96.
$$
Poiché $z_{0.98}\approx 2.05375$,
$$
1=2.05375\frac{4}{\sqrt n}.
$$
Allora
$$
\sqrt n=2.05375\cdot 4=8.215,
\qquad
n\approx 67.49.
$$
Arrotondando per eccesso, $n=68$. Poiché sono già stati estratti $55$ pasticcini, occorre estrarne ancora
$$
68-55=13.
$$

> **Correzione importante.**  
> Nel testo si legge “scarto quadratico medio pari a $4$ g”, quindi $\sigma=4$ e $\sigma^2=16$. La soluzione originaria trattava invece $\sigma^2=4$, il che è incoerente con la frase “scarto quadratico medio”. Con $\sigma=4$ l'intervallo corretto è $(33.746,36.254)$, non $(34.372,35.628)$.

## Esercizio 5 — Proporzione di alberi malati

In un vivaio una proporzione incognita $p$ di alberi ha contratto una malattia.

1. Quanti alberi occorre controllare affinché l'intervallo di confidenza al $99\%$ per $p$ abbia ampiezza inferiore a $0.1$?
2. Si selezionano $100$ alberi e $40$ risultano malati. Calcolare l'intervallo di confidenza al $98\%$ per $p$.

### Soluzione

Sia $Y\sim\operatorname{Be}(p)$ e
$$
\hat p=\frac{1}{n}\sum_{i=1}^n Y_i.
$$
Per il teorema del limite centrale,
$$
\frac{\hat p-p}{\sqrt{\hat p(1-\hat p)/n}}
\overset{D}{\longrightarrow}
\mathcal N(0,1).
$$
L'ampiezza approssimata dell'intervallo è
$$
W
=
2z_{1-\alpha/2}
\sqrt{\frac{\hat p(1-\hat p)}{n}}.
$$
Per $1-\alpha=0.99$, $z_{0.995}\approx 2.57583$. Senza una stima preliminare di $p$, si usa lo scenario peggiore $p=1/2$, che massimizza $p(1-p)$. Allora
$$
W
=
2\cdot 2.57583
\sqrt{\frac{1}{4n}}
=
\frac{2.57583}{\sqrt n}.
$$
Imponendo $W<0.1$,
$$
\frac{2.57583}{\sqrt n}<0.1
\iff
\sqrt n>25.7583
\iff
n>663.49.
$$
Quindi $n\ge 664$.

Con $n=100$ e $\hat p=0.40$, per $1-\alpha=0.98$ si ha $z_{0.99}\approx 2.32635$. L'errore standard stimato è
$$
\sqrt{\frac{0.4\cdot 0.6}{100}}
=
\sqrt{0.0024}
\approx 0.04899.
$$
L'intervallo è
$$
\operatorname{IC}_{0.98}(p)
\approx
\left[
0.40\pm 2.32635\cdot 0.04899
\right]
\approx
(0.286,\;0.514).
$$

## Esercizio 6 — Pacchetti di caffè

Sia $Y$ il peso dei pacchetti di caffè. Da un lotto si estraggono $n=101$ pezzi e si ottiene
$$
\sum_i y_i=25149,
\qquad
\sum_i y_i^2=6267050.
$$

1. Supponendo $Y\sim\mathcal N(\mu,\sigma^2)$, costruire l'intervallo di confidenza al $90\%$ per $\mu$.
2. Come si modifica l'intervallo se non si assume la normalità?

### Soluzione

La media campionaria è
$$
\overline y=\frac{25149}{101}=249.
$$
La varianza campionaria corretta è
$$
S^2
=
\frac{1}{n-1}
\left(
\sum_i y_i^2-n\overline y^{\,2}
\right).
$$
Ora
$$
n\overline y^{\,2}=101\cdot 249^2=101\cdot 62001=6262101.
$$
Quindi
$$
S^2
=
\frac{6267050-6262101}{100}
=
\frac{4949}{100}
=
49.49.
$$

Se $Y$ è normale e $\sigma^2$ è ignota, la quantità pivotale è
$$
\frac{\overline{Y}-\mu}{S/\sqrt n}
\sim t_{n-1}.
$$
Per $1-\alpha=0.90$ e $n-1=100$,
$$
t_{100,0.95}\approx 1.660.
$$
Inoltre
$$
\frac{S}{\sqrt n}
=
\sqrt{\frac{49.49}{101}}
=
\sqrt{0.49}
=
0.7.
$$
Dunque
$$
\operatorname{IC}_{0.90}(\mu)
\approx
\left[
249\pm 1.660\cdot 0.7
\right]
=
(247.838,\;250.162).
$$

Se non si assume la normalità, per $n$ grande si usa il teorema del limite centrale:
$$
\frac{\overline{Y}-\mu}{S/\sqrt n}
\overset{D}{\longrightarrow}
\mathcal N(0,1).
$$
Per $1-\alpha=0.90$, $z_{0.95}\approx 1.64485$. Allora
$$
\operatorname{IC}_{0.90}(\mu)
\approx
\left[
249\pm 1.64485\cdot 0.7
\right]
\approx
(247.849,\;250.151).
$$

> **Correzione.**  
> Nel calcolo asintotico si deve usare $z_{0.95}=1.64485$, non $z_{0.05}$. Inoltre l'intervallo basato su $t$ è esatto solo sotto l'ipotesi di normalità; quello basato su $z$ è approssimato e richiede $n$ grande.
