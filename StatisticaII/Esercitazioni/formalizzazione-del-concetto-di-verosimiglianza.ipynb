# Esercitazione sulla formalizzazione del concetto di verosimiglianza

---

>**Osservazione:** devo correggere il fatto che il modello statistico non si specifica per la variabile aleatoria, ma si specifica per il campione

>**Osservazione:** rifinire la dimostrazione

### Esercizio 1

In un parcheggio, si desidera valutare proporzione $\pi$ di SUV in un giorno feriale. Il dipendente $A$ decide di osservare le prime $10$ auto che arrivano e registra il numero $x$ di SUV. Ottiene
$x = 3$. Il dipendente B osserva pure le prime $10$ auto, ma registra $y_i , i = 1,\dots, 10$, ponendo
$y_i = 1$ se l’i-esima auto è un SUV e zero altrimenti. Ottiene $y = (0, 0, 1, 1, 0, 0, 0, 0, 0, 1)$. Il dipendente C decide invece di continuare ad osservare l’arrivo delle auto finché non arrivano $3$ SUV e registra il numero $z$ di auto non-SUV che arrivano fino a quando non arriva il terzo SUV. Ottiene $z = 7$.

- Si specifichino modelli statistici adeguati per i dati $x, y, z$ ottenuti da A, B, C.
- Si ottengano le funzioni di verosimiglianza per $\pi$, $L_A(\pi)$, $L_B(\pi)$, $L_C(\pi)$ basate sui modelli specificati al punto (a) e sui dati $x = 3$, $y = (0, 0, 1, 1, 0, 0, 0, 0, 0, 1)$ e $z = 7$
- Si mostri che le tre funzioni di verosimiglianza sono equivalenti.
- Il dipendente A continua l’esperimento e osserva le $6$ auto successive registrando l’arrivo di $2$ SUV. Come può combinare l’esito di questa seconda osservazione con quello della prima per calcolare la funzione di verosimiglianza per $\pi$?

### Soluzione

Sia $x_{\text{oss}} = 3$, assumiamo $X \sim \operatorname{Bin}(n, \pi)$ da cui si ha che 

$$
\mathcal{F}_X = \{\operatorname{Bin}(n, \pi) : \pi \in [0,1]\}
$$ 

è il modello statistico parametrico per $x_{\text{oss}}$

Segue che 

$$
L_A(\pi) = p_X(3, 1, \pi) = \binom{10}{3} \pi^3(1-\pi)^7
$$

Sia $y = (0, 0, 1, 1, 0, 0, 0, 0, 0, 1)$, assumiamo $Y_i \sim \operatorname{Be}(\pi), i = 1,\dots, 10$ da cui si ha che 

$$
\mathcal{F}_Y = \{\operatorname{Be}(\pi): \pi \in [0,1]\}
$$

è il modello statistico parametrico per $y_{\text{oss}}$

Segue che 

$$
L_B(\pi) = \prod_{i = 1}^{10} p_Y(y, \pi) = \pi^{\sum_{i = 1}^{10} y_i}(1-\pi)^{10 - \sum_{i = 1}^n y_i} = \pi^3(1-\pi)^7 
$$

Sia $z_{\text{oss}} = 7$, assumiamo $Z \sim \operatorname{Binneg}(3, \pi)$ da cui si ha che 

$$
\mathcal{F}_Z = \{\operatorname{Binneg}(3, \pi) : \pi \in [0,1]\}
$$ 

è il modello statistico parametrico per $z_{\text{oss}}$

Segue che 

$$
L_C(\pi) = p_Z(z, 3, \pi) = \binom{y + 3 -1}{y}\pi^3(1-\pi)^y = \binom{9}{7}\pi^3(1-\pi)^7
$$

Dimostriamo l'equivalenza tra le funzioni di verosimiglianza

Devo dimostrare che $\exists c > 0 : L_A(\pi) = cL_B(\pi) \; \forall \pi \in [0,1]$

$$
L_A(\pi) = cL_B(\pi) \iff \binom{10}{3} \pi^3(1-\pi)^7 = \pi^3(1-\pi)^7 
$$

Fissando $\frac{1}{c} = \binom{10}{3} > 0$ si ottiene che 

$$
\binom{10}{3} \pi^3(1-\pi)^7 = \frac{1}{c} \pi^3(1-\pi)^7 \iff \pi^3(1-\pi)^7 = c \pi^3(1-\pi)^7 \iff L_A(\pi) = c L_B(\pi)
$$

Segue che:

$$
L_A(\pi) \propto L_B(\pi)
$$

Analogamente

$$
L_B(\pi) = L_C(\pi) \iff \pi^3(1-\pi)^7 = \binom{9}{7}\pi^3(1-\pi)^7
$$

fissando $c = \binom{9}{7} > 0$ si ottiene 

$$
\pi^3(1-\pi)^7 = \binom{9}{7}\pi^3(1-\pi)^7 \iff \pi^3(1-\pi)^7 = c \pi^3 (1-\pi)^7 \iff L_B(\pi) = cL_C(\pi)
$$

Segue che 

$$
L_B(\pi) \propto L_C(\pi)
$$

Essendo $\propto$ una relazione di equivalenza si ha inoltre che 

$$
L_A(\pi) \propto L_C(\pi)
$$

Le tre funzioni di verosimiglianza sono quindi equivalenti

Sia ${x_1}_{\text{oss}} = 3$ dove $x_1$ corrisponde all'osservazione $x$, sia ${x_2}_{\text{oss}} = 2$ assumiamo che $X_2 \sim \operatorname{Binom}(6, \pi)$ segue 

$$
\mathcal{F}_{X_2} = \{\operatorname{Binom}(6, \pi) : \pi \in [0,1]\}
$$

Segue che 

$$
L_{X_2}(\pi) = p_{X_2}(x_2, 6, \pi) = \binom{6}{2}\pi^2(1-\pi)^4 \propto \pi^2(1-\pi)^4 
$$

Possimao considerare quindi la coppia $({x_1}_{\text{oss}}, {x_2}_{\text{oss}})$

Sfruttando il fatto che $L_{X_1}(\pi) = L_A(\pi)$ e supponendo che $X_1 \perp X_2$, si ha che 

$$
L_{X_1, X_2}(\pi) = L_{X_1}(\pi)L_{X_2}(\pi) = \pi^3(1-\pi)^7 \pi^2(1-\pi)^4 = \pi^5(1-\pi)^{11}
$$

Analogamente per lo studio completo possiamo considerare la somma definendo:

$$
S = X_1 + X_2 \implies s_{\text{oss}} = 5
$$ 

Assumendo $S \sim X_1 + X_2 \sim \operatorname{Binom}(16, \pi)$

Si ha che il modella statistico parametrico per $s_{\text{oss}}$ è quindi 

$$
\mathcal{F}_{S} = \{\operatorname{Binom}(16, \pi): \pi \in [0,1]\}
$$

Da cui 

$$
L_S(\pi) = p_S(s, 16, \pi) = \binom{16}{5} \pi^5(1-\pi)^{11} \propto \pi^5(1-\pi)^{11}
$$

$L_{X_1, X_2}(\pi) \propto L_S(\pi)$ in quanto differiscono per una costante

### Esercizio 2

Sia $Y \sim \mathcal{N}(\mu, \sigma^2)$, con $\mu \in \mathbb{R}$ e $\sigma^2 > 0$ La distribuzione della variabile casuale $T = e^Y$ è detta lognormale con parametri $\mu$ e $\sigma^2$

- Si ottenga la distribuzione di $T$.
- Sia $y = (y_1, \dots, y_n)$ un vettore di realizzazioni indipendenti di $Y$ e sia $t = (t_1, \dots, t_n)$ il
vettore delle corrispondenti realizzazioni $T$, con $t_i = e^{y_i}, i = 1,\dots, n$ Si mostri che le funzioni di verosimiglianza per $\theta = (\mu, \sigma^2)$ basate su $y$ e su $t$ sono equivalenti.
- Si calcoli la funzione di log-verosimiglianza.
- Si generalizzi il risultato per qualunque trasformazione monotona e derivabile delle osservazioni $w = (w_1, \dots, w_n)$ con $w_i = g(y_i)$ e $g(\cdot)$ derivabile e monotona

### Soluzione

$$
T = e^Y \iff Y = \ln T, \quad \frac{\partial Y}{\partial T}(t) = \frac{1}{T}
$$

Il supporto di $T$ è quindi $]0, +\infty[$

Da cui 

$$
f_T(t) = f_Y(g^{-1}(t)) |\frac{\partial Y}{\partial T}(t)| \mathbf{1}_{]0,+\infty[}(t) = \frac{1}{\sqrt{2\pi\sigma^2}}e^{-(\frac{(\ln(t_i) - \mu)^2}{2\sigma^2})} \frac{1}{t}\mathbf{1}_{]0,+\infty[}(t) = \frac{1}{t\sqrt{2\pi\sigma^2}}e^{-(\frac{(\ln(t_i) - \mu)^2}{2\sigma^2})}\mathbf{1}_{]0,+\infty[}(t)
$$

Da cui:

$$
T \sim \operatorname{Lognormal}(\mu, \sigma^2)
$$

>**Osservazione:** $\mu$ e $\sigma^2$ sono media e varianza di $\ln T = Y$, non di $T$

Sia $y = (y_1, \dots, y_n)$ un campione di realizzazioni indipendenti di $Y$. La densità di una singola $Y_i$ è 

$$
f_Y(y_i; \mu; \sigma^2) = \frac{1}{\sqrt{2\pi\sigma^2}} e^{(-\frac{(y_i - \mu)^2}{2\sigma^2})}
$$

Per indipendenza la varosimiglianza basata su $y$ è 

$$
L_y(\mu, \sigma^2) = \prod^n_{i = 1} f_Y(y_i, \mu, \sigma^2) = (2\pi\sigma^2)^{-\frac{n}{2}}e^{-\frac{1}{2\sigma^2}\sum_{i = 1}^n (y_i - \mu)^2}
$$

Sia 

$$
t_i = e^{y_i}, \quad i = 1,\dots, n
$$

La densità di $T_i = e^{Y_i}$ è 

$$
f_T(t_i; \mu \sigma^2) = \frac{1}{t_i\sqrt{2\pi\sigma^2}}e^{-\frac{(\ln t_i - \mu)^2}{2\sigma^2}}
$$

Poiché $t_i = e^{y_i}$, si ha $\ln t_i = y_i$. Quindi 

$$
L_t(\mu, \sigma^2) = \prod^n_{i = 1} f_T(t_i, \mu, \sigma^2) = \prod_{i = 1}^n\left[ \frac{1}{t_i\sqrt{2\pi\sigma^2}} e^{(-\frac{(\ln t_i - \mu)^2}{2\sigma^2})}\right]
$$

Semplificando, 

$$
L_t(\mu, \sigma^2) = \left(\prod_{i = 1}^n \frac{1}{t_i}\right) (2\pi\sigma^2)^{-\frac{n}{2}}e^{(-\frac{1}{2\sigma^2})\sum_{i = 1}^n (\ln t_i - \mu)^2}
$$

Ma $\ln t_i = y_i$, dunque 

$$
L_t(\mu, \sigma^2) = \left(\prod^n_{i = 1} \frac{1}{t_i}\right) L_y(\mu, \sigma^2)
$$

Il fattore 

$$
\prod^n_{i = 1} \frac{1}{t_i}
$$

Dipende solo dai dati osservati $t$, non da $\theta = (\mu, \sigma^2)$. Quindi è una costante positiva rispetto a $\theta$. Pertanto

$$
L_t(\mu, \sigma^2) \propto L_y(\mu, \sigma^2)
$$

Dalla verosimiglianza basata su $y$ si ottiene:

$$
l_y(\mu, \sigma^2) = \log L_Y(\mu, \sigma^2) = -\frac{n}{2}\ln(2\pi) - \frac{n}{2}\ln(\sigma^2) -\frac{1}{2\sigma^2}\sum_{i = 1}^n(y_i - \mu)^2
$$

Equivalentemente, usando $t_i = e^{y_i}$, cioè $y_i = \ln t_i$

$$
l_t(\mu, \sigma^2) = -\frac{n}{2}\ln(2\pi) - \frac{n}{2}\ln(\sigma^2) - \sum_{i = 1}^n \ln t_i - \frac{1}{2\sigma^2}\sum_{i = 1}^n (\ln t_i - \mu)^2
$$

Le due log-verosimiglianze differiscono per la costante

$$
-\sum_{i = 1}^n \ln t
$$

che non dipende da $\mu$ e $\sigma^2$. Quindi massimizzare $l_y$​ o $l_y$ rispetto a $\theta$ porta agli stessi stimatori.

Generalizziamo quanto esposto

Sia 

$$
w_i = g(y_i), \quad i = 1,\dots, n
$$

con $g$ strettamente monotona e derivabile, con derivata non nulla. Sia $h = g^{-1}$ la funzione inversa, cioè 

$$
y_i = h(w_i)
$$

La densità della variabile trasformata $W = g(Y)$ è, per la formula di trasformazione 

$$
f_W(w, \theta) = f_Y(h(w); \theta)|h'(w)|
$$

Quindi la verosimiglianza basata su $w = (w_1, \dots, w_n)$ è 

$$
L_w(\theta) = \prod^n_{i = 1}f_W(w_i; \theta) = \prod^n_{i = 1}[f_Y(h(w_i); \theta)|h'(w_i)|]
$$

Poiché $h(w_i) = y_i$, si ha 

$$
L_w(\theta) = \left( \prod^n_{i = 1} |h'(w_i)| \right) \prod^n_{i = 1}f_Y(y_i; \theta)
$$

Ma 

$$
\prod^n_{i = 1} f_Y(y_i; \theta) = L_y(\theta)
$$

Pertanto 

$$
L_w(\theta) = (\prod^n_{i = 1}|h'(w_i)|)L_y(\theta)
$$

Il fattore jacobiano

$$
\prod^n_{i = 1}|h'(w_i)|
$$
Dipende solo dai dati osservati $w$, non da $\theta$. Quindi è una costante positiva rispetto a $\theta$, e si ha 

$$
L_w(\theta) \propto L_y(\theta)
$$

La log-verosimiglianza diventa

$$
l_w(\theta) = l_y(\theta) + \sum_{i = 1}^n \ln|h'(w_i)|
$$

Cioè differisce da $l_y(\theta)$ per una costante che non dipende da $\theta$

Nel caso della trasformazione logaritmica, $g(y) = e^y$, $h(w) = \ln w$, quindi 

$$
h'(w) = \frac{1}{w}
$$

e infatti il fattore jacobiano è 

$$
\prod_{i = 1}^n \frac{1}{t_i}
$$
