# Esercitazione sulla formalizzazione del concetto di verosimiglianza

---

### Esercizio 1

In un parcheggio, si desidera valutare proporzione $\pi$ di SUV in un giorno feriale. Il dipendente $A$ decide di osservare le prime $10$ auto che arrivano e registra il numero $x$ di SUV. Ottiene
$x = 3$. Il dipendente B osserva pure le prime $10$ auto, ma registra $y_i , i = 1,\dots, 10$, ponendo
$y_i = 1$ se l’i-esima auto è un SUV e zero altrimenti. Ottiene $y = (0, 0, 1, 1, 0, 0, 0, 0, 0, 1)$. Il dipendente C decide invece di continuare ad osservare l’arrivo delle auto finché non arrivano $3$ SUV e registra il numero $z$ di auto non-SUV che arrivano fino a quando non arriva il terzo SUV. Ottiene $z = 7$.

- Si specifichino modelli statistici adeguati per i dati $x, y, z$ ottenuti da A, B, C.
- Si ottengano le funzioni di verosimiglianza per $\pi$, $L_A(\pi)$, $L_B(\pi)$, $L_C(\pi)$ basate sui modelli specificati al punto (a) e sui dati $x = 3$, $y = (0, 0, 1, 1, 0, 0, 0, 0, 0, 1)$ e $z = 7$
- Si mostri che le tre funzioni di verosimiglianza sono equivalenti.
- Il dipendente A continua l’esperimento e osserva le $6$ auto successive registrando l’arrivo di $2$ SUV. Come può combinare l’esito di questa seconda osservazione con quello della prima per calcolare la funzione di verosimiglianza per $\pi$?

### Soluzione

Sia $x_{\text{oss}} = 3$, assumiamo $X \sim \operatorname{Bin}(10, \pi)$ da cui si ha che 

$$
\mathcal{F}_X = \{\operatorname{Bin}(10, \pi) : \pi \in [0,1]\}
$$ 

è il modello statistico parametrico per $x_{\text{oss}}$

Segue che 

$$
L_A(\pi) = p_X(3, 10, \pi) = \binom{10}{3} \pi^3(1-\pi)^7
$$

Sia $y = (0, 0, 1, 1, 0, 0, 0, 0, 0, 1)$, assumiamo $Y_i \sim \operatorname{Be}(\pi)$ da cui si ha che 

$$
\mathcal{F}_Y = \{\operatorname{Be}(\pi): \pi \in [0,1]\}
$$

è il modello statistico parametrico per $y_{\text{oss}}$

Segue che 

$$
L_B(\pi) = \prod_{i = 1}^{10} p_Y(y, \pi) = \pi^{\sum_{i = 1}^{10} y_i}(1-\pi)^{10 - \sum_{i = 1}^n y_i} = \pi^3(1-\pi)^7 
$$

Sia $z_{\text{oss}} = 7$, assumiamo $Z \sim \operatorname{Binneg}(3, \pi)$ da cui si ha che 

$$
\mathcal{F}_Z = \{\operatorname{Binneg}(3, \pi) : \pi \in [0,1]\}
$$ 

è il modello statistico parametrico per $z_{\text{oss}}$

Segue che 

$$
L_C(\pi) = p_Z(y, 3, \pi) = \binom{y + 3 -1}{y}\pi^3(1-\pi)^y = \binom{9}{7}\pi^3(1-\pi)^7
$$

Dimostriamo l'equivalenza tra le funzioni di verosimiglianza

Devo dimostrare che $\exists c > 0 : L_A(\pi) = cL_B(\pi) \; \forall \pi \in [0,1]$

$$
L_A(\pi) = L_B(\pi) \iff \binom{10}{3} \pi^3(1-\pi)^7 = \pi^3(1-\pi)^7 
$$

Fissando $c = \binom{10}{3} > 0$ si ottiene che 

$$
\binom{10}{3} \pi^3(1-\pi)^7 = \pi^3(1-\pi)^7 \iff c \pi^3(1-\pi)^7 = \pi^3(1-\pi)^7 \iff c L_A(\pi) = L_B(\pi)
$$

Segue che 

$$
L_A(\pi) \propto L_B(\pi)
$$

Analogamente

$$
L_B(\pi) = L_C(\pi) \iff \pi^3(1-\pi)^7 = \binom{9}{7}\pi^3(1-\pi)^7
$$

fissando $c = \binom{9}{7} > 0$ si ottiene 

$$
\pi^3(1-\pi)^7 = \binom{9}{7}\pi^3(1-\pi)^7 \iff \pi^3(1-\pi)^7 = c \pi^3 (1-\pi)^7
$$

Segue che 

$$
L_B(\pi) \propto L_C(\pi)
$$

Essendo $\propto$ una relazione di equivalenza si ha inoltre che 

$$
L_A(\pi) \propto L_C(\pi)
$$

Le tre funzioni di verosimiglianza sono quindi equivalenti

Sia ${x_1}_{\text{oss}} = 3$ dove $x_1$ corrisponde all'osservazione $x$, sia ${x_2}_{\text{oss}} = 2$ assumiamo che $X_2 \sim \operatorname{Binom}(6, \pi)$ segue 

$$
\mathcal{F}_{X_2} = \{\operatorname{Binom}(6, \pi) : \pi \in [0,1]\}
$$

Segue che 

$$
L_{X_2}(\pi) = p_{X_2}(x_2, 6, \pi) = \binom{6}{2}\pi^2(1-\pi)^4 \propto \pi^2(1-\pi)^4 
$$

Possimao considerare quindi la coppia $({x_1}_{\text{oss}}, {x_2}_{\text{oss}})$

Sfruttando il fatto che $L_{X_1}(\pi) = L_A(\pi)$ e supponendo che $X_1 \perp X_2$, si ha che 

$$
L_{X_1, X_2}(\pi) = L_{X_1}(\pi)L_{X_2}(\pi) = \pi^3(1-\pi)^7 \pi^2(1-\pi)^4 = \pi^5(1-\pi)^{11}
$$

Analogamente per lo studio completo possiamo considerare la somma definendo:

$$
S = X_1 + X_2 \implies s_{\text{oss}} = 5
$$ 

Assumendo $S \sim X_1 + X_2 \sim \operatorname{Binom}(16, \pi)$

Si ha che il modella statistico parametrico per $s_{\text{oss}}$ è quindi 

$$
\mathcal{F}_{S} = \{\operatorname{Binom}(16, \pi): \pi \in [0,1]\}
$$

Da cui 

$$
L_S(\pi) = p_S(s, 16, \pi) = \binom{16}{5} \pi^5(1-\pi)^{11} \propto \pi^5(1-\pi)^{11}
$$

$L_{X_1, X_2}(\pi) \propto L_S(\pi)$ in quanto differiscono per una costante
