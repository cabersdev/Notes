# Esercizi su verifica di ipotesi

## Esercizio 1 

Nella pubblicità di una dieta si dichiara che essa consente di perdere **almeno** $4.5$ kg in due mesi. Si osservano $n=13$ persone selezionate casualmente, con perdite (in kg):
$$
3.7,\ 4.3,\ 4.5,\ 5.5,\ 4.1,\ 6.5,\ 3.7,\ 5.7,\ 4.3,\ 3.7,\ 4.6,\ 4.3,\ 3.0.
$$
Si assume $Y\sim\mathcal N(\mu,\sigma^2)$ e, in un primo momento, $\sigma^2=1$ nota da studi precedenti.

### Soluzione

#### Dati campionari

Si ha
$$
n=13,\qquad
\sum_{i=1}^{13}y_i=57.9,\qquad
\sum_{i=1}^{13}y_i^2=268.75.
$$
La media campionaria è
$$
\bar y=\frac{57.9}{13}\approx 4.4538.
$$
La devianza campionaria è
$$
\sum_{i=1}^{13}(y_i-\bar y)^2
=
268.75-\frac{57.9^2}{13}
\approx 10.8723,
$$
da cui
$$
S^2=\frac{10.8723}{12}\approx 0.9060,
\qquad
S\approx 0.9518.
$$

#### (a) Test al livello $\alpha=0.05$ e regione di rifiuto

La pubblicità afferma $\mu\ge 4.5$. Per verificarne la veridicità, impostiamo il test unilaterale sinistro
$$
H_0:\mu\ge 4.5,
\qquad
H_1:\mu<4.5.
$$
Rifiutare $H_0$ significa avere evidenza contro la dichiarazione pubblicitaria.

Poiché $\sigma^2=1$ è nota, la statistica test è
$$
Z=\frac{\bar Y-4.5}{\sigma/\sqrt n}
=
\frac{\bar Y-4.5}{1/\sqrt{13}}.
$$
Sotto $H_0$, al valore di frontiera $\mu=4.5$, si ha $Z\sim\mathcal N(0,1)$.

Per $\alpha=0.05$,
$$
z_{0.95}\approx 1.64485.
$$
La regione di rifiuto (coda sinistra, coerente con $H_1:\mu<4.5$) è
$$
R
=
\left\{
Z<-z_{0.95}
\right\}
=
\left\{
Z<-1.64485
\right\}
=
\left\{
\bar y<4.5-\frac{1.64485}{\sqrt{13}}
\right\}
\approx
\left\{
\bar y<4.0438
\right\}.
$$

Il valore osservato della statistica è
$$
z_{\text{oss}}
=
\frac{4.4538-4.5}{1/\sqrt{13}}
\approx
-0.1664.
$$
Poiché
$$
-0.1664>-1.64485,
\qquad
4.4538>4.0438,
$$
il valore osservato **non cade nella regione di rifiuto**. Al livello $5\%$ **non si rifiuta $H_0$**: non c'è evidenza statistica sufficiente per smentire la pubblicità.

#### (b) Livello di significatività osservato (p-value)

Per un test unilaterale sinistro,
$$
p
=
P_{H_0}(Z\le z_{\text{oss}})
=
\Phi(-0.1664)
=
1-\Phi(0.1664)
\approx
1-0.5661
=
0.4339.
$$
Quindi
$$
p\approx 0.434>\alpha=0.05.
$$
La conclusione è la stessa: non si rifiuta $H_0$.

> **Interpretazione.**  
> Se ripetessimo l'esperimento molte volte e $H_0$ fosse vera, osserveremmo un valore della statistica test così piccolo o più piccolo di quello osservato circa il $43\%$ delle volte. Non è un evento raro: i dati sono del tutto compatibili con $\mu\ge 4.5$.

#### (c) Potenza per $\mu_1=4$, $\mu_1=3.5$, $\mu_1=3$

La potenza in $\mu$ è
$$
\pi(\mu)
=
P_\mu(\bar Y<4.0438)
=
P_\mu\left(
\frac{\bar Y-\mu}{1/\sqrt{13}}
<
\sqrt{13}(4.0438-\mu)
\right)
=
\Phi\left(\sqrt{13}(4.0438-\mu)\right).
$$

- Per $\mu=4$:
  $$
  \pi(4)
  =
  \Phi\left(\sqrt{13}\cdot 0.0438\right)
  =
  \Phi(0.1579)
  \approx 0.563.
  $$
- Per $\mu=3.5$:
  $$
  \pi(3.5)
  =
  \Phi\left(\sqrt{13}\cdot 0.5438\right)
  =
  \Phi(1.9608)
  \approx 0.975.
  $$
- Per $\mu=3$:
  $$
  \pi(3)
  =
  \Phi\left(\sqrt{13}\cdot 1.0438\right)
  =
  \Phi(3.7635)
  \approx 0.9999.
  $$

> **Interpretazione.**  
> Il test è poco potente contro l'alternativa $\mu=4$ (potenza $\approx 0.56$): se la vera perdita media fosse $4$ kg, il test rifiuterebbe $H_0$ solo il $56\%$ delle volte. È invece molto potente contro alternative lontane ($\mu=3.5$ e $\mu=3$). Questo è coerente con l'intuizione: distinguere $4$ da $4.5$ con sole $13$ osservazioni e $\sigma=1$ è difficile; distinguere $3$ da $4.5$ è facile.

#### (d) Test con varianza non nota

Se $\sigma^2$ è incognita, la statistica test diventa
$$
T=\frac{\bar Y-4.5}{S/\sqrt n}
=
\frac{\bar Y-4.5}{0.9518/\sqrt{13}}.
$$
Sotto $H_0$, al valore di frontiera $\mu=4.5$,
$$
T\sim t_{12}.
$$

Il valore osservato è
$$
t_{\text{oss}}
=
\frac{4.4538-4.5}{0.9518/\sqrt{13}}
\approx
\frac{-0.0462}{0.2640}
\approx
-0.1750.
$$

Per $\alpha=0.05$ e $12$ gradi di libertà,
$$
-t_{12,0.95}\approx -1.7823.
$$
La regione di rifiuto è
$$
R=\{T<-1.7823\}.
$$
Poiché
$$
-0.1750>-1.7823,
$$
**non si rifiuta $H_0$**.

Il p-value è
$$
p=P(t_{12}\le -0.1750)\approx 0.432.
$$
Anche in questo caso $p>\alpha=0.05$ e la conclusione non cambia.

> **Confronto tra (a) e (d).**  
> Le conclusioni coincidono, ma il test $t$ è meno potente del test $z$ a parità di dati, perché la stima di $\sigma^2$ introduce incertezza aggiuntiva. In questo caso l'effetto è modesto perché $n=13$ e la stima $S^2\approx 0.906$ è vicina al valore noto $\sigma^2=1$.

## Esercizio 2 

Si dispone di una sola osservazione $X_1$ da $X\sim\operatorname{Exp}(\lambda)$, con densità
$$
f_X(x;\lambda)=\lambda e^{-\lambda x},\qquad x>0.
$$
Per verificare
$$
H_0:\lambda=1,
\qquad
H_1:\lambda=2,
$$
si usa la regione di accettazione $\{x_1:x_1>0.2\}$, cioè si rifiuta $H_0$ se $X_1\le 0.2$. Determinare livello di significatività e potenza del test.

### Soluzione

La regione di rifiuto è
$$
R=\{x_1\le 0.2\}.
$$

#### Livello di significatività

$$
\alpha
=
P_{H_0}(X_1\in R)
=
P_{\lambda=1}(X_1\le 0.2)
=
\int_0^{0.2}e^{-x}\,dx
=
\left[-e^{-x}\right]_0^{0.2}
=
1-e^{-0.2}
\approx 0.1813.
$$

#### Potenza

La potenza è la probabilità di rifiutare $H_0$ sotto $H_1$:
$$
\pi(2)
=
P_{H_1}(X_1\in R)
=
P_{\lambda=2}(X_1\le 0.2)
=
\int_0^{0.2}2e^{-2x}\,dx
=
\left[-e^{-2x}\right]_0^{0.2}
=
1-e^{-0.4}
\approx 0.3297.
$$
Equivalentemente, l'errore di II tipo è
$$
\beta
=
P_{H_1}(X_1\notin R)
=
e^{-0.4}
\approx 0.6703,
$$
e la potenza è $1-\beta\approx 0.3297$.

> **Interpretazione.**  
> Il test ha livello $\alpha\approx 0.18$ e potenza $\approx 0.33$. Entrambi sono modesti perché si basa su **una sola osservazione**: la regione di rifiuto $\{x_1\le 0.2\}$ è scelta in modo che sotto $H_1$ il valore atteso $1/2=0.5$ sia più piccolo di quello sotto $H_0$ ($1$), ma con una sola osservazione la separazione tra le due distribuzioni è debole. La potenza è maggiore di $\alpha$, come deve essere per un test sensato, ma non di molto.

## Esercizio 3 

Un metodo standard produce misure con scarto quadratico medio $\sigma_0=1.6$. Un nuovo metodo viene testato su $n=9$ misure e produce uno scarto quadratico medio campionario $s_{\text{oss}}=0.99$. Assumendo normalità con media $\mu$ non nota, si vuole verificare al $10\%$ se il nuovo metodo è più preciso, cioè se riduce $\sigma$.

### Soluzione

#### Ipotesi e statistica test

Le ipotesi sono
$$
H_0:\sigma\ge 1.6
\quad\Longleftrightarrow\quad
H_0:\sigma^2\ge 2.56,
\qquad
H_1:\sigma<1.6
\quad\Longleftrightarrow\quad
H_1:\sigma^2<2.56.
$$
Si tratta di un test unilaterale sinistro sulla varianza.

La statistica test è
$$
T=\frac{(n-1)S^2}{\sigma_0^2}
=
\frac{8S^2}{2.56}.
$$
Sotto $H_0$, al valore di frontiera $\sigma^2=\sigma_0^2=2.56$,
$$
T\sim\chi^2_{8}.
$$

#### Regione di rifiuto e decisione

Per $\alpha=0.10$, la regione di rifiuto è nella coda sinistra:
$$
R=\left\{T<\chi^2_{8,0.10}\right\}.
$$
Dalle tavole,
$$
\chi^2_{8,0.10}\approx 3.490.
$$
Il valore osservato è
$$
t_{\text{oss}}
=
\frac{8\cdot 0.99^2}{1.6^2}
=
\frac{8\cdot 0.9801}{2.56}
\approx 3.063.
$$
Poiché
$$
3.063<3.490,
$$
si **rifiuta $H_0$** al livello $10\%$. I dati forniscono evidenza sufficiente per affermare che il nuovo metodo riduce lo scarto quadratico medio, cioè è più preciso.

#### Livello di significatività osservato

Il p-value è
$$
p=P_{H_0}(T\le t_{\text{oss}})=P(\chi^2_8\le 3.063).
$$
Dalle tavole,
$$
\chi^2_{8,0.05}\approx 2.733,
\qquad
\chi^2_{8,0.10}\approx 3.490.
$$
Poiché
$$
2.733<3.063<3.490,
$$
si ha
$$
0.05<p<0.10.
$$
Un'interpolazione lineare tra i due quantili fornisce
$$
p
\approx
0.05+\frac{3.063-2.733}{3.490-2.733}\cdot 0.05
\approx
0.05+0.0218
\approx
0.072.
$$
Quindi $p\approx 0.07$: l'evidenza contro $H_0$ è moderata, coerente con il rifiuto al $10\%$ ma non al $5\%$.

> **Osservazione sulla notazione.**  
> $\chi^2_{\nu,\alpha}$ denota il quantile di ordine $\alpha$, cioè il valore con area $\alpha$ **a sinistra**. In un test unilaterale sinistro la regione di rifiuto è $T<\chi^2_{\nu,\alpha}$. In un test unilaterale destro sarebbe $T>\chi^2_{\nu,1-\alpha}$.

## Esercizio 4 — Test sulla media con varianza nota e regione di rifiuto assegnata

Si verifica
$$
H_0:\mu=50,
\qquad
H_1:\mu>50,
$$
per $Y\sim\mathcal N(\mu,\sigma^2)$ con $\sigma^2$ nota, su un campione di $n=16$ unità. Al livello $\alpha=0.05$ la regione di rifiuto è risultata
$$
R=\{\bar Y>54\}.
$$

### Soluzione

#### (a) Tipo di test e natura delle ipotesi

L'alternativa è $\mu>50$, quindi il test è **unilaterale destro**: si rifiuta $H_0$ solo per valori grandi della media campionaria.

- $H_0:\mu=50$ fissa un unico valore del parametro: è un'ipotesi **semplice**.
- $H_1:\mu>50$ contiene infiniti valori possibili di $\mu$: è un'ipotesi **composta**.

#### (b) Varianza $\sigma^2$

Sotto $H_0$, $\bar Y\sim\mathcal N(50,\sigma^2/16)$. Per definizione di livello,
$$
\alpha
=
P_{H_0}(\bar Y>54)
=
P\left(Z>\frac{54-50}{\sigma/4}\right)
=
0.05,
$$
dove $Z\sim\mathcal N(0,1)$. Dunque
$$
\frac{54-50}{\sigma/4}=z_{0.95}\approx 1.64485.
$$
Risolvendo,
$$
\frac{16}{\sigma}=1.64485
\quad\Longrightarrow\quad
\sigma=\frac{16}{1.64485}\approx 9.727,
\qquad
\sigma^2\approx 94.61.
$$

#### (c) Livello di significatività osservato per $\bar y=51$

Il p-value è
$$
p
=
P_{H_0}(\bar Y\ge 51)
=
P\left(Z\ge\frac{51-50}{9.727/4}\right)
=
P(Z\ge 0.4112)
=
1-\Phi(0.4112)
\approx
0.3404.
$$
Quindi $p\approx 0.34>\alpha=0.05$: non si rifiuta $H_0$. Il valore osservato $\bar y=51$ è del tutto compatibile con $\mu=50$.

#### (d) Potenza quando $\mu$ raddoppia ($\mu=100$)

La potenza in $\mu=100$ è
$$
\pi(100)
=
P_{100}(\bar Y>54)
=
P\left(Z>\frac{54-100}{9.727/4}\right)
=
P(Z>-18.91)
\approx 1.
$$
La potenza è praticamente $1$: se la media fosse $100$, il test rifiuterebbe $H_0$ con probabilità essenzialmente certa. Il valore $54$ è enormemente lontano dalla media sotto $H_1$, quindi il test è estremamente potente contro questa alternativa.

#### (e) Numerosità per un intervallo di confidenza di ampiezza $\le 2$

Poiché $\sigma^2$ è nota, l'intervallo di confidenza di livello $1-\alpha$ per $\mu$ è
$$
\bar Y\pm z_{1-\alpha/2}\frac{\sigma}{\sqrt n}.
$$
L'ampiezza è
$$
W=2z_{1-\alpha/2}\frac{\sigma}{\sqrt n}.
$$
Per $1-\alpha=0.95$, $z_{0.975}\approx 1.96$. Imponendo $W\le 2$:
$$
2\cdot 1.96\cdot\frac{9.727}{\sqrt n}\le 2
\quad\Longleftrightarrow\quad
\frac{1.96\cdot 9.727}{\sqrt n}\le 1
\quad\Longleftrightarrow\quad
\sqrt n\ge 1.96\cdot 9.727\approx 19.065.
$$
Elevando al quadrato,
$$
n\ge 363.5.
$$
Arrotondando per eccesso,
$$
n=364.
$$

> **Osservazione.**  
> La numerosità richiesta è elevata perché $\sigma\approx 9.727$ è grande rispetto all'ampiezza target $2$. L'ampiezza dell'intervallo scala come $1/\sqrt n$: per dimezzarla occorre quadruplicare $n$. Questo è un esempio concreto del costo informativo della precisione.
