### Esercizio 1

Nella pubblicità di una dieta si dichiara che essa consente a chi la segue per due mesi di perdere almeno $4.5$ kg. Si sono rilevati i dati per $13$ persone, selezionate casualmente, che hanno seguito
la dieta per due mesi e si è rilevato che hanno perso $3.7, 4.3, 4.5, 5.5, 4.1, 6.5, 3.7, 5.7, 4.3, 3.7, 4.6, 4.3, 3.0$ kg. Nell’ipotesi che nella popolazione il peso perso si distribuisca come una gaussiana la cui media denotiamo con $\mu$, e la varianza, nota da studi precedenti, è pari a $1$, si vuole verificare se la pubblicità dice il vero.

- Verificare l’ipotesi ad un livello di significatività pari a $0.05$ e identificare la regione di rifiuto
- Si calcoli il livello di significatività osservato.
- Si determini la potenza del test per le seguenti ipotesi alternative: $\mu_1 = 4$, $\mu_1 = 3.5$, $\mu_1 = 3$.
- Si ripeta il test al punto a) nell’ipotesi che la varianza della distribuzione non sia nota

### Soluzione 

Sia $Y \sim \mathcal{N}(\mu, 1)$ e sia $Y_1, \dots, Y_{13}$ il campione i.i.d. tratto da $Y$

Definiamo le ipotesi:
$$
H_0 : \mu \geq 4.5, \quad H_1 \mu < 4.5
$$
La verifica è quindi un test unilaterale sinistro

Definiamo la statistica test:
$$
Z = \frac{\overline{Y} - 4.5}{\sqrt{\frac{1}{13}}} \overset{H_0}{\sim} \mathcal{N}(0,1)
$$

Costruiamo la regione di rifiuto sfruttando la simmetria della normale standard fissando $\alpha = 0.05$:

$$
0.05 = P(\frac{\overline{Y} - 4.5}{\sqrt{\frac{1}{13}}} \overset{H_0}{\sim} \mathcal{N}(0,1) < -z_{0.95}) = P(\overline{Y} < 4.5 - 1.645\sqrt{\frac{1}{13}})
$$

La regione di rifiuto è quindi:

$$
R = \{y : \overline{Y} <  4.043759\}
$$

Dal campione si ha che 

$$
\overline{Y}_{\text{oss}} = 4.453846 > 4.043759
$$

Non rifiuto $H_0$ al $5\%$ di significatività non c'è evidenza statistica sul fatto che la publicità stia dicendo il falso

$$
Z_{\text{oss}} = -0.1664106
$$

Da cui 

$$
p = P(Z < -0.1664106) = \phi(-0.17) = 1-\phi(0.17) = 0.4325 > 0.05 
$$

Il p-value conferma il rifiuto di $H_0$

Definiamo l'ipotesi

$$
H_0 : \mu = 4.5, \quad H_1 \mu = 4
$$

$$
1-\beta(4) = P(\overline{Y} < 4.043759 | H_1) = P(Z < \frac{4.043759 - 4}{\sqrt{\frac{1}{13}}}) = \phi(0.16) = 0.5636 
$$

Definiamo l'ipotesi

$$
H_0 : \mu = 4.5, \quad H_1 \mu = 3.5
$$

$$
1-\beta(3.5) = P(\overline{Y} < 4.043759 | H_1) = P(Z < \frac{4.043759 - 3.5}{\sqrt{\frac{1}{13}}}) = \phi(1.96) = 0.975 
$$

Definiamo l'ipotesi

$$
H_0 : \mu = 4.5, \quad H_1 \mu = 3
$$

$$
1-\beta(3) = P(\overline{Y} < 4.043759 | H_1) = P(Z < \frac{4.043759 - 3}{\sqrt{\frac{1}{13}}}) = \phi(3.76) = 0.99
$$

Sia $Y \sim \mathcal{N}(\mu, \sigma)$ e sia $Y_1, \dots, Y_{13}$ il campione i.i.d. tratto da $Y$

Definiamo le ipotesi:
$$
H_0 : \mu \geq 4.5, \quad H_1 \mu < 4.5
$$

Definiamo la statistica test:
$$
T = \frac{\overline{Y} - 4.5}{\sqrt{\frac{S^2}{13}}} \overset{H_0}{\sim} \mathcal{t}_{12}, 
$$

E costruiamo la regione di rifiuto fissando $\alpha = 0.05$ e sfruttando la simmetria delle $t$ di student

$$
0.05 = P(\frac{\overline{Y} - 4.5}{\sqrt{\frac{S^2}{13}}} < -z_{0.95}) = P(\overline{Y} < 4.5 - 1.782\sqrt{\frac{0.9060256}{13}}) 
$$

Da cui si ottiene la regione di rifiuto 

$$
R = \{y : \overline{Y} < 4.029558\}
$$

Da prima si ha che 

$$
\overline{Y}_{\text{oss}} = 4.453846 > 4.029558
$$

Non rifiuto $H_0$ al $5\%$ di significatività

### Esercizio 3

Un metodo comunemente usato per determinare il calore specifico del ferro è noto produca misure caratterizzate da uno scarto quadratico medio pari a $1.6$. Un nuovo metodo viene verificato su un campione $9$ volte e per le misure ottenute si calcola lo scarto quadratico medio campionario che risulta pari a $0.99$. Assumendo che le misure di calore specifico del ferro nella
popolazione abbiano distribuzione normale di media $\mu$ non nota, si può affermare, al $10\%$ di significatività, che il nuovo metodo sia più preciso, ovvero riduca lo scarto quadratico medio?
Facendo ricorso alle tavole dei quantili, è possibile fornire una misura relativamente precisa del livello di significatività osservato?

### Soluzione

Sia $Y \sim \mathcal{N}(\mu, \sigma^2)$ e sia $Y_1, \dots, Y_9$ il campione i.i.d. traddo da $Y$

Definiamo le ipotesi: 

$$
H_0 : \sigma^2 \geq 2.56, \quad H_1 : \sigma^2 < 2.56
$$

Si tratta di un test unilaterale sinistro

Definiamo la statistica test:

$$
X^2 = \frac{S^2(9-1)}{\sigma^2} \overset{H_0}{\sim} \chi^2_8
$$

$$
X^2_{\text{oss}} = 3.062812
$$

Da cui 

$$
p = P(X^2_8 < 3.062812) = 1- P(X^2_8 \geq 3.062812)
$$

Cerchiamo di limitare superiormente e inferiormente $p$

$$
1-P(X^2_8 \geq 2.73) \leq 1- P(X^2_8 \geq 3.062812) \leq 1-P(X^2_8 \geq 3.49) \iff 0.05 \leq p \leq 0.10
$$

Da cui prendendo il valore centrale 

$$
p \approx \frac{0.10+0.05}{2} = 0.075 < 0.10
$$

Possiamo rifiutare quindi $H_0$ al $10\%$ di significatività

### Esercizio 4

Si voglia sottoporre a verifica il seguente sistema di ipotesi relativo alla media $\mu$ di una variabile casuale $Y$ Gaussiana di varianza nota $\sigma^2$:

$$
H_0 : \mu = 50, \quad H_1 : \mu > 50
$$

La regione di rifiuto ottenuta avendo a disposizione un campione casuale semplice di $16$ unità con $\alpha = 0.05$ è risultata essere $\{\overline{Y} > 54\}$.

- Si stabilisca se il test è unilaterale o bilaterale e se le ipotesi sono semplici o composte.
- Quanto vale la varianza?
- Essendo risultata la media campionaria pari a $51$, calcolare il livello di significatività osservato.
- Determinare la potenza del test nell’ipotesi che la media $\mu$ raddoppi.
- Si determini la numerosità campionaria necessaria a costruire un intervallo di confidenza per $\mu$, a livello $0.95$, di ampiezza non superiore a $2$.

### Soluzione

Il test è unilaterale destro poiché $H_1$ specifica $\mu > 50$

L'ipotesi $H_0$ è semplice poiché $\sigma^2$ è nota e quindi specifica completamente la distribuzione, l'ipotesi $H_1$ è invece composta in quanto non va a specificare completamente la distribuzione

Poiché $\sigma^2$ è nota la statistica test che è stata utilizzata per ottenere la regione di rifiuto è 

$$
Z = \frac{\overline{Y} - 50}{\sqrt{\frac{\sigma^2}{16}}} \overset{H_0}{\sim} \mathcal{N}(0,1)
$$

Da cui 

$$
0.05 = P(\frac{\overline{Y} - 50}{\sqrt{\frac{\sigma^2}{16}}} > z_{0.95}) = P(\overline{Y} > 50 + 1.645\sqrt{\frac{\sigma^2}{16}})
$$

Da cui 

$$
50 + 1.645\sqrt{\frac{\sigma^2}{16}} = 54 \iff \sqrt{\frac{\sigma^2}{16}} = \frac{4}{1.645} \iff \sigma = \frac{16}{1.645} \iff \sigma^2 = (\frac{16}{1.645})^2 = 94.60371
$$

La statistica test osservata è:

$$
Z_{\text{oss}} = 0.41125
$$

Da cui

$$
p = P(Z > 0.41) = 0.6591
$$

Siano 

$$
H_0 : \mu = 50, \qquad H_1 : \mu = 100
$$

$$
1-\beta(100) = P(\overline{Y} > 54 | H_1) = P(Z > \frac{54-100}{\sqrt{\frac{94.60371}{16}}}) = P(Z > -18.9175) = 1-(1-\phi(18.9175)) \approx 1
$$

Definiamo la quantità pivotale:

$$
Q = \frac{\overline{Y} - \mu}{\sqrt{\frac{\sigma^2}{n}}}
$$

E costruiamo l'intervallo di livello $1-\alpha = 0.95$ sfruttando la simmetria della normale

$$
0.95 = P(-z_{0.975} < \frac{\overline{51} - \mu}{\sqrt{\frac{94.60371}{n}}} < z_{0.975}) \iff 0.95 = P(51 - 1.96\sqrt{\frac{94.60371}{n}} < \mu < 51 + 1.96\sqrt{\frac{94.60371}{n}})
$$

Da cui 

$$
IC_{0.95}(\mu) = [51 - 1.96\sqrt{\frac{94.60371}{n}}, 51 + 1.96\sqrt{\frac{94.60371}{n}}]
$$

Poniamo:

$$
W = 51 + 1.96\sqrt{\frac{94.60371}{n}} - (51 - 1.96\sqrt{\frac{94.60371}{n}}) = 2 \cdot(1.96\sqrt{\frac{94.60371}{n}}) \leq 2
$$

$$
\iff 1.96\sqrt{\frac{94.60371}{n}} \leq 1 \iff 1.96\sqrt{94.60371} \leq n \iff 19.06383 \leq n
$$

Da cui 

$$
n \geq 20, n \in \mathbb{N}
$$

### Esercizio 5

Henri Poincaré fu un illustre matematico e fisico francese vissuto a cavallo tra l’$800$ e il $900$. Come molti matematici, Poincaré era un tipo piuttosto abitudinario: faceva ricerca quattro ore al giorno, sempre nelle stesse fasce orarie (la mattina dalle dieci alle dodici e il pomeriggio dalle cinque alle sette), e si recava ogni giorno dal medesimo fornaio ad acquistare una pagnotta da un chilo. Un giorno, gli venne il sospetto che il panettiere lo ingannasse, non sul prezzo ma sul peso che lui temeva fosse inferiore al dovuto. Da perfetto matematico, paziente e meticoloso, Poincaré pesò accuratamente il pane per un anno.

- Supponendo che Poincaré si fosse riproposto di denunciare il fornaio se il peso medio delle pagnotte fosse stato inferiore ai $993.902$ gr., concedendosi una probabilità di errore pari a circa l’$1\%$, si determini una stima dello scarto quadratico medio delle pagnotte.
- Dagli scritti del matematico, solo parzialmente recuperati, si legge che la media del peso al quadrato delle pagnotte acquistate fosse pari a $905000$. Si dica se Poincaré alla fine denunciò il fornaio.
- Si dica rispetto a quale ipotesi alternativa, il test effettuato da Poincaré ha una probabilità di errore di secondo tipo pari alla probabilità di errore di primo tipo.

### Soluzione

Supponiamo che $Y:= \text{"peso della singola pagnotta"} \sim \mathcal{N}(\mu, \sigma^2)$

Definiamo le ipotesi:

$$
H_0 : \mu \geq 1000, \quad H_1 : \mu < 1000
$$

E' stata ottenuta la seguente regione di rifiuto al livello di significatività $1\%$

$$
R = \{y : \overline{Y} < 993.902\}
$$

La statistica test utilizzata è:

$$
T = \frac{\overline{Y} - 1000}{\sqrt{\frac{S^2}{365}}} \overset{H_0}{\sim} t_{364}
$$

Da cui sfruttando la simmetria della normale:

$$
0.01 = P(\frac{\overline{Y} - 1000}{\sqrt{\frac{S^2}{365}}} < -t_{364, 0.99}) = P(\overline{Y} < 1000 - 2.326 \cdot \sqrt{\frac{S^2}{365}})
$$

Da cui 

$$
1000 - 2.326 \cdot \sqrt{\frac{S^2}{365}} = 993.902 \iff 2.326 \cdot \sqrt{\frac{S^2}{365}} = 6.098 \iff S = \frac{6.098}{2.326}\sqrt{365} = 50.0869
$$

Per ipotesi 

$$
\frac{1}{365}\sum_{i = 1}^{365} y_i^2
$$

Per la disuguaglianza di Jensen si ha che: 

$$
\frac{1}{365}\sum_{i = 1}^{365} y_i^2 > (\frac{1}{365}\sum_{i = 1}^{365} y_i)^2
$$

Da cui 

$$
905000 > (\frac{1}{365}\sum_{i = 1}^{365} y_i)^2 \iff 951.3149 > \frac{1}{365}\sum_{i = 1}^{365} y_i = \overline{Y}
$$

Il che implica $\overline{Y} \in R$, Poincarré denuncio il fornaio

Siano 

$$
H_0: \mu 1000, \quad H_1 : \mu = \mu_1
$$

$$
1 - \beta = P(\overline{Y} < 993.902 | H_1) = P(T_{364} < \frac{993.902 - \mu_1}{\sqrt{\frac{50.0869}{365}}})
$$

$\beta = \alpha = 0.01 \iff 1 - \beta = 0.99$

Da cui 

$$
P(T_{364} < \frac{993.902 - \mu_1}{\sqrt{\frac{50.0869}{365}}}) = 0.99 \iff \frac{993.902 - \mu_1}{\sqrt{\frac{50.0869}{365}}} =  2.3366 \iff \mu_1 = 993.902 - 2.3366\sqrt{\frac{50.0869}{365}}
$$

$$
\iff \mu_1 = 993.0367
$$

### Esercizio 6

Secondo uno studio sulle attitudini sessuali e stili di vita, effettuato in Inghilterra e relativo al $1994$, la percentuale degli uomini inglesi di età compresa tra i $16$ e i $59$ anni che si sono sottoposti al test HIV in quell’anno è pari al $13.1\%$. Nel $2002$, sempre in Inghilterra, è stato selezionato un campione casuale di $124$ uomini di età compresa tra i $16$ e i $59$ anni, $21$ dei quali hanno dichiarato di essersi sottoposti al test HIV nel $2001$. Si vuole stabilire se la proporzione di uomini di età compresa tra i $16$ e i $59$ anni che si sono sottoposti al test HIV è aumentata dal $1994$ al $2001$.

- Rispondere concedendosi una probabilità di errore di primo tipo pari al $5%$, sapendo che nel $1994$ l’indagine fu svolta su un campione di $99$ soggetti
- Come cambia la risposta precedente se non è nota la dimensione del campione nella prima
indagine?
- Si calcoli la potenza del test (utilizzato al punto b) in corrispondenza dell’ipotesi che sia
pari al $20\%$ la proporzione di uomini inglesi (di età compresa tra i $16$ e i $59$ anni) che si sono sottoposti al test HIV nel $2001$.

### Soluzione

Definiamo le ipotesi: 

Sia $Y_{2001} \sim \operatorname{Be}(\pi_{2001})$ e analogamente sia $Y_{1994} \sim \operatorname{Be}(\pi_{1994})$

$$
H_0 : \pi_{2001} - \pi_{1994} \leq 0, \quad H_1 : \pi_{2001} - \pi_{1994} > 0
$$

Stimiamo la proporzione sotto $H_0$

$$
\hat{\pi} = \frac{124 \cdot 0.17 + 99 \cdot 0.131}{124 + 99} = 0.153
$$

Definiamo la statistica test 

$$
Z = \frac{\hat{\pi}_{2001} - \hat{\pi}_{1994}}{\sqrt{0.153(1 - 0.153) \cdot (\frac{1}{124} + \frac{1}{99})}} \underbrace{\overset{D}{\to}}_{H_0} \mathcal{N}(0,1)
$$

Costruiamo la regione di rifiuto fissato $\alpha = 0.05$ e sfruttando la simmetria della normale standard

$$
0.05 = P(\frac{\hat{\pi}_{2001} - \hat{\pi}_{1994}}{\sqrt{0.153(1 - 0.153) \cdot (\frac{1}{124} + \frac{1}{99})}} > z_{0.95}) = P(\frac{\hat{\pi}_{2001} - \hat{\pi}_{1994}}{\sqrt{0.153(1 - 0.153) \cdot (\frac{1}{124} + \frac{1}{99})}} > 1.645) 
$$

$$
= P(\hat{\pi}_{2001} - \hat{\pi}_{1994} > 1.645 \cdot \sqrt{0.153(1 - 0.153) \cdot (\frac{1}{124} + \frac{1}{99})}) = P(\hat{\pi}_{2001} - \hat{\pi}_{1994} > 0.07974653)
$$

$$
\hat{\pi}_{2001} - \hat{\pi}_{1994} = 0.17 - 0.131 = 0.039 < 0.07974653
$$

Non rifiuto $H_0$ al $5\%$ di significatività

Supponendo che non sia nota la numerosità del campione del $1994$ assumiamo $\hat{\pi}_{1994}$ come vero parametro della distribuzione 

Le ipotesi diventano quindi 

$$
H_0 \pi_{2001} \leq 0.131, \quad \pi_{2001} > 0.131
$$

Definiamo la statistica test 

$$
Z = \frac{\hat{\pi}_{2001} - 0.131}{\sqrt{\frac{0.131\cdot(1-0.131)}{124}}} \overset{H_0}{\to} \mathcal{N}(0,1)
$$

Analogamente a prima costruiamo la regione di rifiuto fissato $\alpha = 0.05$

$$
0.05 = P(\frac{\hat{\pi}_{2001} - 0.131}{\sqrt{\frac{0.131\cdot(1-0.131)}{124}}} > 1.645) = P(\hat{\pi}_{2001} > 0.131 + 1.645 \sqrt{\frac{0.131\cdot(1-0.131)}{124}})
$$

Da cui 

$$
R = \{y : \hat{\pi}_{2001} > 0.1808426\} \implies \hat{\pi}_{2001} \notin R 
$$

Non rifiuto $H_0$ al $5\%$ di significatività 

Sia 

$$
H_0: \mu = 0.131, \quad H_0 : \mu = 0.20
$$

Si ha che: 

$$
1-\beta(0.20) = P(\hat{\pi}_{2001} > 0.1808426 | H_1) = P(Z > \frac{0.18 - 0.20}{0.030}) = 1 - P(Z \leq \frac{0.18 - 0.20}{0.030})
$$

$$
= 1 -\phi(-0.6666667) = 1-(1- \phi(0.67)) = \phi (0.67) = 0.7486
$$

| 7 | 185 | 168 | -17 |
| 8 | 206 | 180 | -26 |
| 9 | 173 | 147 | -26 |
| 10 | 146 | 136 | -10 |
| 11 | 174 | 151 | -23 |
| 12 | 201 | 168 | -33 |

- Supponendo che la pressione sistolica abbia distribuzione Normale, valutare l’efficacia del farmaco nel ridurre la pressione sistolica media.
- Nella stessa ipotesi del punto precedente, calcolare l’intervallo di confidenza di livello $0.95$ per la variazione media dei livelli di pressione sistolica determinata dall’assunzione del farmaco
- Cosa cambierebbe nel caso non fosse possibile assumere la normalità del fenomeno di interesse?

### Soluzione

Notiamo che i dati sono appaiati sulla singola unità statistica sono rilevate due varibiali 

Siano quini $Y_p \sim \mathcal{N}(\mu_p, \sigma_p^2)$, $Y_d \sim \mathcal{N}(\mu_d, \sigma_d^2)$, definiamo 

$$
\Delta = Y_p - Y_d \sim \mathcal{N}(\mu_\delta, \sigma_\delta^2)
$$

Definiamo le ipotesi: 

$$
H_0: \mu_\delta \leq 0, \quad H_1: \mu_\delta > 0
$$

Definiamo la statistica test:

$$
T = \frac{\hat{\mu}_\delta}{\sqrt{\frac{S^2}{12}}} \overset{H_0}{\sim} t_{11}
$$

Dal campione si ha che:

$$
\hat{\mu}_\delta = 18.58333, \quad S^2 = 102.0833, \implies T_{\text{oss}} = 6.371429
$$

Da cui 

$$
p = P(t_{11} > 6.371429) \approx 0 
$$

Rifiuto $H_0$, c'è evidenza statistica per affermare che farmaco è efficace nel ridurre la pressione sistolica media.

Construiamo l'intervallo di confidenza al livello $0.95$ per la variabile $\Delta$ sfruttando la simmetria della $t$ di student

$$
0.95 = P(-t_{0.975} < \frac{\hat{\mu}_\delta - \mu_{\delta}}{\sqrt{\frac{S^2}{12}}} < t_{0.975}) = P(18.58333 - 2.201 \sqrt{\frac{102.0833}{12}} < \mu_\delta < 18.58333 + 2.201 \sqrt{\frac{102.0833}{12}})
$$

Da cui: 

$$
IC_{0.95}(\mu_{\delta}) = [12.16375, 25.00292]
$$

Non potendo assumere la normalità del fenomeno, possiamo sfruttare il teorema del limite centrale

In particolare definiamo la statistica test 

$$
Z = \frac{\hat{\mu}_\delta}{\sqrt{\frac{S^2}{12}}} \overset{H_0}{\longrightarrow} \mathcal{N}(0,1)
$$

Si ha che 

$$
Z_{\text{oss}} = 6.371429
$$

Da cui 

$$
p = P(Z > 6.371429) = 1-\phi(6.371429) \approx 0
$$

Che conferma il rifiuto di $H_0$

Tuttavia l'approssimazione non è ottimale

### Esercizio 8

I valori che seguono sono i livelli di colesterolo LDL (in mmol/l) rilevati su un campione di $12$ soggetti che hanno una dieta ricca di fiocchi di granoturco.

$$
4.61, 6.42, 5.40, 4.54, 3.98, 3.82, 5.01, 4.34, 3.80, 4.56, 5.35, 3.89
$$


Si assuma che la distribuzione del carattere in questione sia ben descritta da un elemento della famiglia parametrica Normale.

- Sapendo che il livello di colesterolo LDL non dovrebbe superare i $4$ mmol/l, stabilire se la dieta considerata sia adeguata o meno a contenere il colesterolo. Data la decisione presa, con quale probabilità al massimo questa potrà essere sbagliata?

- Si costruisca un intervallo di confidenza, di copertura 0.95, per il livello medio del colesterolo LDL per un soggetto che si sottopone alla dieta.

- Si verifichi l’ipotesi che la varianza del colesterolo sia pari a $0.5$ contro l’alternativa che sia maggiore.

### Soluzione

Definiamo le ipotesi 

$$
H_0: \mu \leq 4, \quad \mu > 4
$$

Definiamo la statistica test

$$
T = \frac{\hat{\mu} - 4}{\sqrt{\frac{S^2}{12}}} \overset{H_0}{\sim} t_{11}
$$

$$
\hat{\mu}_{\text{oss}} = 4.643333, \quad S^2_{\text{oss}} = 0.6245697 \implies T_{\text{oss}} = 2.819916
$$

Da cui:

$$
p = P(t_{11} > 2.819916) < 0.01
$$

Rifiuto $H_0$ al $1\%$ di significatività

Non avendo fissato $\alpha$ non è fissata la probabilità massima con la quale il rifiuto sarebbe sbagliato, tutta via dal calcolo del $p$ value è chiaro che tale probabilità è al massimo $1\%$


Costruiamo l'intervallo di confidenza sfruttando la simmetria delle $t$ di student

$$

### Esercizio 9

Un gruppo di ricercatori vuole valutare l’efficacia di tre diverse campagne di prevenzione per ridurre la pressione arteriosa media in una popolazione a rischio di ipertensione. Le tre campagne sono state condotte in tre distretti sanitari diversi, ciascuno con caratteristiche sociodemografiche simili. In ciascun distretto è stato selezionato un campione indipendente di persone adulte, e per ognuna si è misurata la pressione arteriosa sistolica (in mmHg) dopo sei mesi dall’inizio della campagna. E’ragionevole assumere che la pressione arteriosa sistolica possa essere descritta da una variabile casuale Normale e che le varianze nelle tre condizioni sperimentali siano omogenee. I risultati campionari sono riassunti di seguito. Nel distretto $1$ sono stati osservati $n_1 = 8$ soggetti, con una pressione media pari a $134.2$ mmHg. Nel distretto $2$ sono stati osservati $n_2 = 10$ soggetti, con media $130.1$ mmHg. Nel distretto $3$ sono stati osservati $n_3 = 12$ soggetti, con media $127.8$ mmHg. La devianza totale nel campione è risultata pari a $340.2$. I ricercatori vogliono capire se le differenze osservate nelle medie dei tre distretti sono compatibili con le normali fluttuazioni campionarie, oppure se indicano che almeno una delle campagne di prevenzione abbia avuto un’efficacia diversa dalle altre. Se si propendesse per quest’ultima ipotesi, con quale probabilità si commetterebbe un errore?

### Soluzione

Definiamo le ipotesi: 

$$
H_0: \mu_1 = \mu_2 = \mu_3, \quad H_1 : \text{non tutte uguali}
$$

$$
\hat{\mu}_{\cdot \cdot} = \frac{8 \cdot 134.2 + 10 \cdot 130.1 + 12 \cdot 127.8}{8 + 10+ 12} = 130.2733
$$

$$
D_{\text{tra}} = \sum_{j = 1}^3 n_j(\hat{\mu}_{\cdot j} - \hat{\mu}_{\cdot \cdot})^2 =
$$

$$
= 8 \cdot (134.2 - 130.2733)^2 + 10 \cdot (130.1 - 130.2733)^2 + 12 \cdot (127.8 - 130.2733)^2 = 197.0587
$$

Calcoliamo la devianza entro i gruppi sfruttando la relazione 

$$
D_{\text{totale}} = D_{\text{tra}} + D_{\text{entro}} \iff D_{\text{entro}} = 340.2 - 197.0587 = 143.1413
$$

Definiamo la statistica test

$$
F = \frac{\frac{D_{\text{tra}}}{\sigma^2 \cdot 2}}{\frac{D_{\text{entro}}}{\sigma^2\cdot 27}} \overset{H_0}{\sim} F_{2, 27}
$$

La statistica osservata è quindi: 

$$
F_{\text{oss}} = \frac{\frac{197.0587}{2}}{\frac{143.1413}{27}} = 18.585
$$

da cui 

$$
p = P(F_{2,27} > 18.585) \approx 0
$$

Rifiuto quindi $H_0$, se rifiuto $H_0$ la probabilità massima di commettere un errore è pressoché $0$
