### Esercizio 1

Le osservazioni che seguono si riferiscono alla concentrazione media annua (in microgr/metro $^3$) di biossido di zolfo in un campione casuale semplice di 53 città italiane:

$$
\begin{aligned}
    (6.7799, 6.0141, 2.3891, 0.9817, 3.3384, 5.3363, 1.2274, 3.7378, 4.1574, 7.5315, 6.8971, 5.6222, 3.2202, 3.7482, \\
    4.7032, 3.1935, 3.5361, 7.0275,4.3589, 3.0664, 8.8281, 3.4695, 3.4470, 7.9581, 3.7190, 3.3961, 9.0582, 5.1762, \\
    6.7417, 2.5557, 5.1762, 6.7417, 2.5557, 7.7905, 1.7145, 2.4070, 6.9218, 4.2567, 5.0736, 6.4474, 8.7040, 3.5854, \\
    2.6791, 3.4598, 5.7002, 5.3849, 5.2398, 7.2526, 4.9982, 3.9164, 7.9081, 2.6735, 2.4015)
\end{aligned}
$$

Si assuma che il campione sia tratto da una distribuzione esponenziale con media $\frac{1}{\lambda}$, $\operatorname{Exp}(\lambda)$.

- Si ottengano la funzione di verosimiglianza e di log-verosimiglianza per $\lambda$

- Si ottenga la stima di massima verosimiglianza di $\lambda$

- Si risponda ai due quesiti precedenti assumendo che i dati siano realizzazioni indipendenti di una variabile casuale con distribuzione $\operatorname{Ga}(\alpha, \lambda)$, con $\alpha = \alpha_0 \neq 1$ fissato

### Soluzione

Siano $Y_1, \dots, Y_{53}$ i.i.d da $Y \sim \operatorname{Exp}(\lambda)$

Determiniamo la funzione di verosimiglianza e log-verosimiglianza per $\lambda$

$$
l(\lambda) = \log L(\lambda) = \log \prod_{i = 1}^{53} \lambda e^{-\lambda y_i} = \sum_{i = 1}^{53}  \log(\lambda e^{-\lambda y_i}) = \sum_{i = 1}^{53} \log(\lambda) + \log(e^{-\lambda y_i})
$$

$$
= 53\log(\lambda) - \lambda \sum_{i = 1}^{53} y_i = l(\lambda) \implies L(\lambda) = e^{53\log(\lambda) - \lambda \sum_{i = 1}^{53} y_i} = \lambda^{53}e^{- \lambda \sum_{i = 1}^{53} y_i}
$$

Stimiamo 

$$
\hat{\lambda} = \arg\max_{\lambda \in \mathbb{R}^+} l(\lambda)
$$

$$
\frac{dl}{d\lambda} = 53 \cdot \frac{1}{\lambda} - \sum_{i = 1}^{53} y_i
$$

$$
\frac{dl}{d\lambda} = 0 \iff \frac{1}{\lambda} - \sum_{i = 1}^{53} y_i = 0 \iff \lambda = \frac{53}{\sum_{i = 1}^{53} y_i}
$$

$$
J(\lambda) = -\frac{d^2l}{d\lambda^2} = - \frac{-53}{\lambda^2} = \frac{53}{\lambda^2} > 0 \; \forall \lambda \implies J(\frac{53}{\sum_{i = 1}^{53} y_i}) > 0
$$

Da cui segue che 

$$
\hat{\lambda} = \frac{53}{\sum_{i = 1}^{53} y_i} \implies \hat{\lambda}_{\text{oss}} = 0.2085
$$

Siano $Y_1, \dots, Y_{53}$ i.i.d da $Y \sim \operatorname{Ga}(\alpha, \lambda)$

$$
p_Y(y, \alpha, \lambda) = \frac{\lambda^\alpha}{\Gamma(\alpha)}y^{\alpha - 1}e^{-\lambda y} \mathbb{1}_{]0,+\infty[}(y)
$$

Determiniamo la funzione di verosimiglianza e log-verosimiglianza per $\lambda$

$$
l(\lambda) = \log(\prod_{i = 1}^{53} \frac{\lambda^\alpha}{\Gamma(\alpha)}y_i^{\alpha - 1}e^{-\lambda y_i}) = \sum_{i = 1}^{53} \log(\frac{\lambda^\alpha}{\Gamma(\alpha)}y_i^{\alpha - 1}e^{-\lambda y_i})
$$

$$
= \sum_{i = 1}^{53} \alpha\log(\lambda) - \log(\Gamma(\alpha)) + (\alpha -1)\log(y_i) - \lambda y_i = 
$$

$$
= 53\alpha\log(\lambda) -53\log(\Gamma(\alpha)) + (\alpha -1)\sum_{i = 1}^{53}\log(y_i) -\lambda \sum_{i = 1}^{53} y_i \propto 53\alpha\log(\lambda)  -\lambda \sum_{i = 1}^{53} y_i := l(\lambda)
$$

Da cui 

$$
L(\lambda) = e^{53\alpha\log(\lambda) - \lambda \sum_{i = 1}^{53} y_i} := l(\lambda) = \lambda^{53\alpha} e^{-\lambda\sum_{i = 1}^{53} y_i}
$$
