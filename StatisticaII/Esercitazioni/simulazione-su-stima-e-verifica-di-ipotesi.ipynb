# Simulazione su Stima e Verifica di Ipotesi

---

Una catena di supermercati monitora il numero di reclami formali ricevuti al punto informazioni. In condizioni di normale gestione del punto vendita, è noto che il numero di reclami giornalieri segue una distribuzione di Poisson con parametro $\lambda = 1$. A seguito dell’introduzione di un nuovo sistema di casse automatiche, il management vuole verificare se il numero medio di reclami sia aumentato a causa di possibili difficoltà tecniche incontrate dai clienti. La direzione ha stabilito di avviare una procedura di revisione dei flussi operativi qualora il numero medio di reclami nei punti vendita delle province di Padova e Vicenza (dove si ritiene che il numero di punti vendita sia sufficientemente elevato) fosse maggiore di $1.245$. Dati i costi di revisione, si è stabilito un margine di errore nell’attivazione della procedura di revisione pari al $2.5\%$.

- Si mostri che il numero dei punti vendita nelle province di Padova e Vicenza è $64$.
- Sapendo che, alla luce dei dati raccolti, il più piccolo margine di errore che avrebbe comunque portato ad avviare la revisione è pari a $0.0162$, si dica quanti sono, in totale, i reclami registrati durante il monitoraggio.
- La distribuzione del numero di reclami sopraggiunti dai diversi punti vendita durante il monitoraggio è parzialmente riportata, a livello aggregato, nella tabella che segue. Si dica se l’ipotesi in merito alla distribuzione di Poisson per il numero di reclami sia sostenibile sulla base dei dati raccolti.

| Reclami giornalieri | 0 | 1 | 2 o più |
|---|---:|---:|---:|
| Frequenze Osservate | 25 | 30 | 17 |

- Alla luce di questa risposta, si commenti la validità del test discusso nei punti (a) e (b).

### Soluzione

Sia $Y$ il numero giornaliero di reclami in un punto vendita $Y \sim \operatorname{Poisson}(\mu)$. Si vuole verificare 

$$
H_0 : \mu \leq \mu_0 = 1, \quad H_1 : \mu > \mu_0 = 1
$$

La regola di decisione consiste nel rifiutare $H_0$ se $\overline{Y}_{\text{oss}} > 1.245$, con probabilità di errore di primo tipo $\alpha = 0.025$

La verifica di ipotesi in esame riguarda la media di una popolazione. Il testo esplicita che il numero di punti vendita è sufficientemente elevato e pertanto è ragionevole che il test effettuato usi la statistica

$$
Z = \frac{\overline{Y} - \mu_0}{\sqrt{\frac{\sigma^2}{n}}} \overset{H_0}{\longrightarrow} \mathcal{N}(0, 1)
$$

la cui distribuzione approssimativa, sotto $H_0$, è nomrmale standard in virtù del teorema del limite centrale. 

Si rifiuta $H_0$ se 

$$
z = \frac{\overline{Y} - \mu_0}{\sqrt{\frac{\sigma^2}{n}}} > z_{1-\alpha} \iff \overline{Y} > \mu_0 + z_{1-\alpha}\sqrt{\frac{\sigma^2}{n}}
$$

E' noto quindi che: 

$$
\mu_0 + z_{1-\alpha}\sqrt{\frac{\sigma^2}{n}} = 1.245
$$

Con $\mu_0 = 0$, $Y \sim \operatorname{Poisson}(\mu)$ e quindi $\sigma^2 = \mu = \mu_0$ sotto $H_0$ e $z_{1-\alpha} = z_{0.975} = 1.96$. Quindi 

$$
1+1.96\sqrt{\frac{1}{n}} = 1.245 \iff \sqrt{n} = \frac{1.96}{0.245} = 8 \implies n = 64
$$

Il più piccolo livello $\alpha$ a cui i dati portano a rifiutare $H_0$, è per definizione il livello di significatività osservato: si rifiuta se e solo se $z_{\text{oss}} > z_{1-\alpha}$, ossia se $\alpha > P(Z > z_{\text{oss}}) = \alpha_{\text{oss}}$

Pertanto: 

$$
\alpha_{\text{oss}} = P(Z > z_{\text{oss}}) = 0.0162 \implies z_{\text{oss}} = z_{0.9838} = 2.14
$$

Invertendo la statistica test si ha che 

$$
z_{\text{oss}} = \frac{\overline{Y} - \mu_0}{\sqrt{\frac{\mu_0}{n}}}
$$

con $z_{\text{oss}} = 2.14$, $\mu_0 = 1$, $n = 64$ si ottiene: 

$$
\overline{Y}_{\text{oss}} = \mu_0 + z_{\text{oss}}\sqrt{\frac{\mu_0}{n}} = 1 + 2.14\sqrt{\frac{1}{64}} = 1.2675
$$

da cui il numero totale di reclami $n \overline{Y}_{\text{oss}} = 64 \cdot 1.2675 \approx 81$ (il totale deve essere un numero intero l’approssimazione dipende da arrotondamenti)

La frequenza mancante si ottiene per differenza: $64 - 25 - 17 = 22$. Si verifica 

$$
H_0: Y \sim \operatorname{Poisson}(\mu), \quad H_1: \overline{H_0}
$$

Stimando $\mu$ con la stima ottenuta al punto precedente $\hat{\mu} = \overline{Y}_{\text{oss}} \approx 1.27$ (dove, in effetti, avevamo concluso che $\mu$ non fosse plausibilmente pari a $1$). Dalla funzione di probabilità $p_Y(y; \mu) = \mu^{y}e^{\mu}/y!$ si ottengono le frequenze attese $f_j^* = np_j$

| Reclami giornalieri | 0 | 1 | 2 o più |
|---|---:|---:|---:|
| fj (osservate) | 25 | 22 | 17 |
| pj = pY (· ; 1.27) | 0.281 | 0.357 | 0.362 |
| f 0 j = 64 pj | 17.98 | 22.85 | 23.17 |

Dove $P(Y \geq 2) = 1 - F_Y(1) = 1 - (0.281 + 0.357) = 0.362$ la statistica test è 

$$
\chi^2 = \sum_j \frac{(f_j - f_j^*)^2}{f_j^*} = 4.42
$$

Con $\chi^2 \overset{H_0}{\sim} \chi^2_{k-1-1} = \chi^2_1$ ($k = 3$ classi, un parametro stimato).

Poichè $\chi^2_{1, 0.95} = 3.84 < 4.42 < \chi^2_{1, 0.975} = 5.02$ si ha che: 

$$
0.025 < \alpha_{\text{oss}}= P(\chi^2_1 > 4.42) < 0.05
$$

Rifiutiamo $H_0$ al $5\%$ di significatività

Il risultato è borderline, ma i dati portano a ritenere poco sostenibile l’ipotesi di Poisson

Il test dei punti (a) e (b) standardizza $\overline{Y}$ con $Var(Y) = \mu_0$, cioè sfrutta l’uguaglianza media-varianza propria della Poisson. Se $Y$ non è Poisson, in generale $\sigma^{2} = \mu$ e lo standard error è errato: il TLC continua a garantire la normalità approssimata di $\overline{Y}$ , ma con varianza $\frac{\sigma^2}{n}$.
La decisione di avviare la revisione non è quindi affidabile, perché basata su una regione di rifiuto costruita sulla base di una varianza plausibilmente diversa. Inoltre, effettuando due test in sequenza non possiamo più fare affidamento sulla probabilità di commettere un errore di primo tipo, che dovrebbe essere aggiustata tenendo conto del fatto di aver svolto due test
