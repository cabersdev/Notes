# Esercitazioni su stima puntuale

## Esercizio 1

Sia $Y_1, \dots, Y_n$ un campione casuale semplice tratto da $Y \sim \mathcal{N}(\mu, \sigma^2)$. Si considerino i due stimatori di $\mu$

$$
T_1 = \overline{Y} = \frac{1}{n}\sum_{i=1}^n Y_i,
\qquad
T_2 = 2\overline{Y} - Y_1.
$$

Si stabilisca quale dei due stimatori è preferibile in termini di errore quadratico medio.

### Soluzione

**Studio di $T_1 = \overline{Y}$.**

Il valore atteso è

$$
\mathbb{E}_\mu[T_1]
=
\mathbb{E}_\mu[\overline{Y}]
=
\frac{1}{n}\sum_{i=1}^n \mathbb{E}_\mu[Y_i]
=
\frac{1}{n}\cdot n\mu
=
\mu.
$$

Quindi $T_1$ è **non distorto**: $\mathcal{B}_\mu(T_1) = 0$.

La varianza è

$$
\mathbb{V}_\mu(T_1)
=
\mathbb{V}(\overline{Y})
=
\frac{\sigma^2}{n}.
$$

Essendo non distorto,

$$
\operatorname{MSE}_\mu(T_1) = \mathbb{V}_\mu(T_1) = \frac{\sigma^2}{n}.
$$

**Studio di $T_2 = 2\overline{Y} - Y_1$.**

Il valore atteso è

$$
\mathbb{E}_\mu[T_2]
=
2\mathbb{E}_\mu[\overline{Y}] - \mathbb{E}_\mu[Y_1]
=
2\mu - \mu
=
\mu.
$$

Dunque anche $T_2$ è **non distorto**: $\mathcal{B}_\mu(T_2) = 0$, e quindi

$$
\operatorname{MSE}_\mu(T_2) = \mathbb{V}_\mu(T_2).
$$

Per la varianza si ha

$$
\mathbb{V}(T_2)
=
\mathbb{V}(2\overline{Y} - Y_1)
=
4\mathbb{V}(\overline{Y}) + \mathbb{V}(Y_1) - 4\operatorname{Cov}(\overline{Y}, Y_1).
$$

Calcoliamo i tre termini:

$$
\mathbb{V}(\overline{Y}) = \frac{\sigma^2}{n},
\qquad
\mathbb{V}(Y_1) = \sigma^2,
$$

$$
\operatorname{Cov}(\overline{Y}, Y_1)
=
\operatorname{Cov}\!\left(\frac{1}{n}\sum_{i=1}^n Y_i,\ Y_1\right)
=
\frac{1}{n}\operatorname{Cov}(Y_1, Y_1)
=
\frac{\sigma^2}{n}.
$$

Sostituendo:

$$
\mathbb{V}(T_2)
=
4\cdot\frac{\sigma^2}{n} + \sigma^2 - 4\cdot\frac{\sigma^2}{n}
=
\sigma^2.
$$

Pertanto

$$
\operatorname{MSE}_\mu(T_2) = \sigma^2.
$$

**Confronto.**

$$
\operatorname{MSE}_\mu(T_2) - \operatorname{MSE}_\mu(T_1)
=
\sigma^2 - \frac{\sigma^2}{n}
=
\sigma^2\left(1 - \frac{1}{n}\right)
\ge 0.
$$

- Se $n > 1$, allora $\operatorname{MSE}_\mu(T_1) < \operatorname{MSE}_\mu(T_2)$, quindi $T_1 = \overline{Y}$ è preferibile.
- Se $n = 1$, allora $T_1 = T_2 = Y_1$ e i due stimatori coincidono.

**Conclusione.** Per $n > 1$, lo stimatore preferibile in termini di MSE è $T_1 = \overline{Y}$, essendo non distorto e con varianza $\sigma^2/n$, strettamente minore di $\sigma^2$.

> **Osservazione 1.1.**  
> Entrambi gli stimatori sono non distorti: la differenza sta esclusivamente nella varianza. $T_2$ sfrutta in modo ridondante $Y_1$ (che compare sia in $\overline{Y}$ sia esplicitamente), introducendo correlazione tra le componenti e aumentando la varianza complessiva.

> **Osservazione 1.2.**  
> Per $n=1$, l'espressione $T_2 = 2Y_1 - Y_1 = Y_1$ coincide con $T_1$, il che rende il confronto banale.

## Esercizio 2

Sia $Y$ una variabile casuale con media $\mu$ e varianza $\sigma^2$, e si consideri il seguente stimatore di $\mu$ basato su un campione $(Y_1, Y_2, Y_3)$:

$$
T = \frac{1}{2}Y_1 + \frac{1}{4}Y_2 + aY_3.
$$

1. Si determini per quali valori di $a$ lo stimatore è non distorto e si calcoli la varianza in corrispondenza di tale valore.
2. Si produca una stima di $\mu$ se si osserva il campione $(Y_1, Y_2, Y_3) = (4, 2, 6)$.

### Soluzione

**Non distorsione.**

$$
\mathbb{E}_\mu[T]
=
\frac{1}{2}\mathbb{E}(Y_1) + \frac{1}{4}\mathbb{E}(Y_2) + a\mathbb{E}(Y_3)
=
\left(\frac{1}{2} + \frac{1}{4} + a\right)\mu.
$$

Lo stimatore è non distorto se e solo se

$$
\frac{1}{2} + \frac{1}{4} + a = 1
\iff
a = \frac{1}{4}.
$$

**Varianza.**

Fissato $a = \frac{1}{4}$ e sfruttando l'indipendenza di $Y_1, Y_2, Y_3$:

$$
\mathbb{V}_\mu(T)
=
\mathbb{V}\!\left(\frac{1}{2}Y_1 + \frac{1}{4}Y_2 + \frac{1}{4}Y_3\right)
=
\frac{1}{4}\mathbb{V}(Y_1) + \frac{1}{16}\mathbb{V}(Y_2) + \frac{1}{16}\mathbb{V}(Y_3).
$$

Poiché $\mathbb{V}(Y_i) = \sigma^2$ per ogni $i$:

$$
\mathbb{V}_\mu(T)
=
\left(\frac{1}{4} + \frac{1}{16} + \frac{1}{16}\right)\sigma^2
=
\frac{3}{8}\sigma^2.
$$

**Stima.**

Con il campione osservato $(4, 2, 6)$ e $a = \frac{1}{4}$:

$$
\hat{\mu}_{\text{oss}}
=
\frac{1}{2}\cdot 4 + \frac{1}{4}\cdot 2 + \frac{1}{4}\cdot 6
=
2 + 0.5 + 1.5
=
4.
$$

**Conclusione.** Lo stimatore è non distorto per $a = 1/4$, ha varianza $\frac{3}{8}\sigma^2$, e la stima sul campione è $\hat{\mu}_{\text{oss}} = 4$.

## Esercizio 3

Sia $Y \sim \mathcal{N}(\theta, \sigma^2)$ la variabile aleatoria che descrive l'altezza di un individuo. Ai fini della stima di $\theta$, si confronti lo stimatore media campionaria $\overline{Y}$ con lo stimatore banale $T = 175$.

### Soluzione

**Stimatore media campionaria.**

$$
\overline{Y} = \frac{1}{n}\sum_{i=1}^n Y_i.
$$

Essendo $\mathbb{E}(\overline{Y}) = \theta$, lo stimatore è non distorto, quindi

$$
\operatorname{MSE}(\overline{Y})
=
\mathbb{V}(\overline{Y})
=
\frac{\sigma^2}{n}.
$$

**Stimatore banale $T = 175$.**

$$
\mathbb{E}(T) = 175,
\qquad
\mathbb{V}(T) = 0.
$$

La distorsione è

$$
\mathcal{B}(T) = 175 - \theta,
$$

e quindi

$$
\operatorname{MSE}(T)
=
\mathbb{V}(T) + \mathcal{B}(T)^2
=
(175 - \theta)^2.
$$

**Confronto.**

$$
\operatorname{MSE}(\overline{Y}) < \operatorname{MSE}(T)
\iff
\frac{\sigma^2}{n} < (175 - \theta)^2.
$$

Prendendo la radice quadrata:

$$
|175 - \theta| > \frac{\sigma}{\sqrt{n}}.
$$

Dunque

$$
\theta < 175 - \frac{\sigma}{\sqrt{n}}
\qquad\text{oppure}\qquad
\theta > 175 + \frac{\sigma}{\sqrt{n}}.
$$

Equivalentemente:

$$
\operatorname{MSE}(\overline{Y}) < \operatorname{MSE}(T)
\iff
\theta \in \left(-\infty,\ 175 - \frac{\sigma}{\sqrt{n}}\right) \cup \left(175 + \frac{\sigma}{\sqrt{n}},\ +\infty\right).
$$

Viceversa,

$$
\operatorname{MSE}(T) < \operatorname{MSE}(\overline{Y})
\iff
\theta \in \left(175 - \frac{\sigma}{\sqrt{n}},\ 175 + \frac{\sigma}{\sqrt{n}}\right).
$$

**Conclusione.** Per $n > 1$, lo stimatore banale $T = 175$ è preferibile solo se il vero valore $\theta$ è sufficientemente vicino a $175$, cioè se la distanza $|\theta - 175|$ è inferiore a $\frac{\sigma}{\sqrt{n}}$. All'aumentare di $n$, la soglia $\frac{\sigma}{\sqrt{n}}$ si restringe e la media campionaria diventa conveniente anche per valori di $\theta$ molto vicini a $175$.

> **Osservazione 3.1.**  
> Questo esempio mostra che uno stimatore distorto può avere MSE inferiore a uno stimatore non distorto. In particolare, $T = 175$ ha varianza nulla, quindi tutta la sua imprecisione è dovuta alla distorsione. Quando la distorsione è piccola (cioè $\theta$ vicino a $175$), l'MSE di $T$ è minore di quello di $\overline{Y}$.

> **Osservazione 3.2.**  
> Al crescere di $n$, la media campionaria diventa sempre più efficiente perché la sua varianza tende a $0$. In altre parole, con abbastanza dati, è sempre preferibile usare la media campionaria, indipendentemente da $\theta$.

## Esercizio 4

Data una variabile casuale $Y$ con funzione di ripartizione $F_Y(y)$ e un campione casuale semplice $Y_1, \dots, Y_n$ tratto da $Y$, si stabilisca se la funzione di ripartizione empirica

$$
F_n(y) = \frac{\#\{i \in \{1,\dots,n\} : Y_i \le y\}}{n}
$$

è uno stimatore non distorto e consistente di $F_Y(y)$.

### Soluzione

Fissiamo $y \in \mathbb{R}$. Per ogni $i = 1, \dots, n$ definiamo

$$
T_i = \mathbf{1}_{\{Y_i \le y\}}
=
\begin{cases}
1, & Y_i \le y,\\
0, & Y_i > y.
\end{cases}
$$

Poiché le $Y_i$ sono i.i.d. con funzione di ripartizione $F_Y$, le $T_i$ sono i.i.d. con distribuzione di Bernoulli di parametro

$$
\pi = P(Y_i \le y) = F_Y(y).
$$

Allora

$$
F_n(y) = \frac{1}{n}\sum_{i=1}^n T_i,
\qquad
\sum_{i=1}^n T_i \sim \operatorname{Bin}(n,\pi).
$$

**Non distorsione.**

$$
\mathbb{E}[F_n(y)]
=
\mathbb{E}\!\left[\frac{1}{n}\sum_{i=1}^n T_i\right]
=
\frac{1}{n}\cdot n\pi
=
\pi
=
F_Y(y).
$$

Quindi $F_n(y)$ è **non distorto** per $F_Y(y)$: $\mathcal{B}(F_n(y)) = 0$.

**Varianza e MSE.**

Essendo non distorto,

$$
\operatorname{MSE}(F_n(y)) = \mathbb{V}(F_n(y)).
$$

Inoltre

$$
\mathbb{V}(F_n(y))
=
\mathbb{V}\!\left(\frac{1}{n}\sum_{i=1}^n T_i\right)
=
\frac{1}{n^2}\cdot n\pi(1-\pi)
=
\frac{\pi(1-\pi)}{n}
=
\frac{F_Y(y)\bigl(1-F_Y(y)\bigr)}{n}.
$$

**Consistenza.**

$$
\lim_{n\to+\infty} \operatorname{MSE}(F_n(y))
=
\lim_{n\to+\infty} \frac{F_Y(y)(1-F_Y(y))}{n}
=
0.
$$

Per il criterio sufficiente basato sull'MSE, $F_n(y)$ è **consistente in senso debole**:

$$
F_n(y) \xrightarrow{\mathbb{P}} F_Y(y).
$$

In realtà, per la legge forte dei grandi numeri, vale anche la convergenza **quasi certa**:

$$
F_n(y) \xrightarrow{q.c.} F_Y(y).
$$

**Conclusione.** $F_n(y)$ è uno stimatore non distorto e consistente di $F_Y(y)$.

> **Osservazione 4.1.**  
> La consistenza ottenuta è **puntuale**: per ogni $y$ fissato, $F_n(y) \to F_Y(y)$. Il **teorema di Glivenko–Cantelli** garantisce un risultato più forte, cioè la convergenza uniforme
> $$
> \sup_{y \in \mathbb{R}} |F_n(y) - F_Y(y)| \xrightarrow{q.c.} 0.
> $$
> Questo risultato non è richiesto dalla traccia, ma è utile tenerlo presente.

> **Osservazione 4.2.**  
> La varianza di $F_n(y)$ è massima quando $F_Y(y) = 1/2$, cioè quando il quantile $y$ è la mediana. In tal caso $\mathbb{V}(F_n(y)) = \frac{1}{4n}$.

> **Osservazione 4.3.**  
> La funzione di ripartizione empirica è uno stimatore non parametrico: non richiede di specificare un modello parametrico per $Y$. La sua distribuzione campionaria è completamente determinata da $F_Y(y)$ attraverso la distribuzione binomiale.

## Esercizio 5

Sia $Y$ una variabile casuale discreta con supporto $\mathcal{S} = \{-1, 0, 1\}$ e funzione di probabilità

$$
p_Y(y;\theta) = \left(\frac{\theta}{2}\right)^{|y|} (1-\theta)^{1-|y|},
\qquad \theta \in [0,1].
$$

Dato un campione casuale semplice $Y_1, \dots, Y_n$ tratto da $Y$:

1. si trovi lo stimatore di $\theta$ basato sul metodo dei momenti;
2. si stabilisca se è non distorto e consistente;
3. si produca una stima di $\theta$ in corrispondenza del campione $-1, -1, 0, 1, -1$.

### Soluzione

**Distribuzione e momenti.**

La funzione di probabilità assume i valori:

$$
p_Y(-1;\theta) = \frac{\theta}{2},
\qquad
p_Y(0;\theta) = 1-\theta,
\qquad
p_Y(1;\theta) = \frac{\theta}{2}.
$$

Il primo momento è

$$
\mathbb{E}(Y) = (-1)\cdot\frac{\theta}{2} + 0\cdot(1-\theta) + 1\cdot\frac{\theta}{2} = 0.
$$

Il secondo momento (grezzo) è

$$
\mathbb{E}(Y^2) = (-1)^2\cdot\frac{\theta}{2} + 0^2\cdot(1-\theta) + 1^2\cdot\frac{\theta}{2} = \theta.
$$

Ne segue

$$
\mathbb{V}(Y) = \mathbb{E}(Y^2) - \mathbb{E}(Y)^2 = \theta.
$$

**Metodo dei momenti.**

Il parametro è uno solo, $\theta$. Il primo momento $\mathbb{E}(Y) = 0$ **non dipende da $\theta$**, quindi non fornisce alcuna equazione utile per stimare $\theta$. L'unico momento che dipende da $\theta$ è il secondo momento grezzo $\mathbb{E}(Y^2) = \theta$.

Il metodo dei momenti impone l'uguaglianza tra il momento teorico e il corrispondente momento campionario:

$$
\frac{1}{n}\sum_{i=1}^n Y_i^2 = \theta.
$$

Da cui lo stimatore dei momenti

$$
\hat{\theta}_n = \frac{1}{n}\sum_{i=1}^n Y_i^2.
$$

> **Osservazione 5.1.**  
> Non è corretto impostare il sistema con la varianza campionaria, cioè con
> $$
> \frac{1}{n}\sum_{i=1}^n Y_i^2 - \left(\frac{1}{n}\sum_{i=1}^n Y_i\right)^2 = \theta.
> $$
> Questo condurrebbe allo stimatore
> $$
> \tilde{\theta}_n = \frac{1}{n}\sum_{i=1}^n Y_i^2 - \left(\frac{1}{n}\sum_{i=1}^n Y_i\right)^2,
> $$
> che è diverso da $\hat{\theta}_n$. Il metodo dei momenti utilizza i **momenti grezzi**, non i momenti centrali.

**Non distorsione.**

$$
\mathbb{E}_\theta[\hat{\theta}_n]
=
\mathbb{E}_\theta\!\left[\frac{1}{n}\sum_{i=1}^n Y_i^2\right]
=
\frac{1}{n}\cdot n\,\mathbb{E}_\theta[Y^2]
=
\theta.
$$

Quindi $\hat{\theta}_n$ è **non distorto**: $\mathcal{B}_\theta(\hat{\theta}_n) = 0$.

**Varianza e MSE.**

Poiché le $Y_i^2$ sono i.i.d. con

$$
Y_i^2 \sim \operatorname{Bernoulli}(\theta),
$$

si ha

$$
\mathbb{V}(Y_i^2) = \theta(1-\theta).
$$

Allora

$$
\mathbb{V}_\theta(\hat{\theta}_n)
=
\mathbb{V}\!\left(\frac{1}{n}\sum_{i=1}^n Y_i^2\right)
=
\frac{1}{n^2}\cdot n\,\theta(1-\theta)
=
\frac{\theta(1-\theta)}{n}.
$$

Essendo non distorto,

$$
\operatorname{MSE}_\theta(\hat{\theta}_n) = \frac{\theta(1-\theta)}{n}.
$$

**Consistenza.**

$$
\lim_{n\to+\infty} \operatorname{MSE}_\theta(\hat{\theta}_n)
=
\lim_{n\to+\infty} \frac{\theta(1-\theta)}{n}
=
0.
$$

Quindi $\hat{\theta}_n$ è **consistente in senso debole** (e, per la legge forte dei grandi numeri, anche quasi certamente).

**Stima sul campione.**

Campione: $-1, -1, 0, 1, -1$. I quadrati sono $1, 1, 0, 1, 1$, la cui somma è $4$. Dunque

$$
\hat{\theta}_{\text{oss}} = \frac{4}{5} = 0.8.
$$

**Conclusione.** Lo stimatore dei momenti è $\hat{\theta}_n = \frac{1}{n}\sum_{i=1}^n Y_i^2$, è non distorto, consistente, e la stima sul campione è $0.8$.

> **Osservazione 5.2.**  
> Lo stimatore $\hat{\theta}_n$ è la media campionaria delle variabili $Y_i^2$, che sono Bernoulli 
$(\theta)$.
>Questo collega l'esercizio al caso della media di una Bernoulli, di cui $\hat{\theta}_n$ eredita non distorsione, varianza $\theta(1-\theta)/n$ e consistenza.

> **Osservazione 5.3.**  
> Poiché $\mathbb{E}(Y) = 0$ per ogni $\theta$, la distribuzione è simmetrica attorno allo zero. Il parametro $\theta$ controlla la probabilità di osservare valori non nulli ($\pm 1$), mentre $1-\theta$ è la probabilità di osservare $0$. Lo stimatore $\hat{\theta}_n$ stima la proporzione di osservazioni non nulle nel campione.
