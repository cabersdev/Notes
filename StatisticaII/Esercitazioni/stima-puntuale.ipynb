**Esercizione 1:**

Sia $Y_1, \dots, Y_n$ un campione tratto da una variabile casuale $Y \sim \mathcal{N}(\mu, \sigma^2)$ e siano $T_1$ e $T_2$ due stimatori per $\mu$

$$
T_1 = \overline{Y} = \frac{1}{n}\sum^n_{i = 1}Y_i \quad T_2 = 2\overline{Y} - Y_1
$$

Si dica quale tra i due stimatori è preferibile in termini di errore quadratico medio

**Soluzione:**

Studio di $T_1 = \overline{Y}$

Abbiamo:
$$
\mathbb{E}_\mu(T_1) = \mathbb{E}_\mu(\overline{Y}) = \mathbb{E}_\mu\left( \frac{1}{n}\sum^n_{i = 1} Y_i \right) = \frac{1}{n} \sum^n_{i = 1} \mathbb{E}_\mu(Y_i) = \frac{1}{n} \cdot n\mu = \mu
$$

Quindi $T_1$ è non distorto:
$$
\mathcal{B}_\mu(T_1) = 0
$$ 
Inoltre: 
$$
\mathbb{V}_\mu(T_1) = \mathbb{V}(\overline{Y}) = \frac{\sigma^2}{n}
$$

Pertanto:
$$
\operatorname{MSE}_\mu(T_1) = \mathbb{V}_\mu(T_1) = \frac{\sigma^2}{n}
$$


Studio di $T_2 = 2\overline{Y} - Y_1$

Calcoliamo il valore atteso:
$$
\mathbb{E}_\mu(T_2) = \mathbb{E}_\mu(2\overline{Y} - Y_1) = 2\mathbb{E}_\mu(\overline{Y}) - \mathbb{E}_\mu(Y_1) = 2\mu - \mu = \mu
$$

Dunque anche $T_2$ è non distorto: 
$$
\mathcal{B}_\mu(T_2) = 0
$$

Di conseguenza:

$$
\operatorname{MSE}_\mu(T_2) = \mathbb{V}_\mu(T_2)
$$

Calcoliamo ora la varianza. Poiché 

$$
T_2 = 2\overline{Y} - Y_1
$$

Si ha

$$
\mathbb{V}(T_2) = \mathbb{V}(2\overline{Y} - Y_1) = 4\mathbb{V}(\overline{Y}) + \mathbb{V}(Y_1) - 4\operatorname{Cov}(\overline{Y}, Y_1)
$$

Ora 

$$
\mathbb{V}(\overline{Y}) = \frac{\sigma^2}{n}, \quad \mathbb{V}(Y_1) = \sigma^2
$$

e

$$
\operatorname{Cov}(\overline{Y}, Y_1) = \operatorname{Cov}\left(\frac{1}{n}\sum_{i = 1}^nY_i, Y_1\right) = \frac{1}{n}\mathbb{V}(Y_1) = \frac{\sigma^2}{n}
$$

Quindi

$$\mathbb{V}(T_2) = 4\frac{\sigma^2}{n} + \sigma^2 - 4\frac{\sigma^2}{n} = \sigma^2$$

Pertanto:
$$
\operatorname{MSE}_\mu(T_2) = \sigma^2
$$

Si ha quindi che:
$$
\operatorname{MSE}_\mu(T_1) = \frac{\sigma^2}{n}, \quad \operatorname{MSE}_\mu(T_2) = \sigma^2
$$

Dunque:

$$
\operatorname{MSE}_\mu(T_2) - \operatorname{MSE}_\mu(T_1) = \sigma^2 - \frac{\sigma^2}{n} = \sigma^2\left(1-\frac{1}{n}\right) \geq 0
$$

Quindi:
- se $n > 1$, allora 
$$
\operatorname{MSE}_\mu(T_1) < \operatorname{MSE}_\mu(T_2)
$$
per $T_1 = \overline{Y}$ è preferibile rispetto a $T_2 = 2\overline{Y} - Y_1$
- se $n = 1$, allora $T_1 = T_2$, quindi i due stimatori coincidono.

In conclusione, per $n > 1$, lo stimatore preferibile in termini di errore quadratico medio è 
$$
T_1 = \overline{Y}
$$
Poiché è non distorto e ha varianza $\frac{\sigma^2}{n}$, minore di quella di $T_2$ che è $\sigma^2$

**Esercizio 2:**

Data una variabile aleatoria $Y$ con media $\mu$ e varianza $\sigma^2$, si consideri il seguente stimatore della media: 

$$
T = \frac{1}{2}Y_1 + \frac{1}{4}Y_2 + aY_3
$$

- Si dica per quali valori di $a$ lo stimatore risulta corretto e calcolarne la varianza
- Determinare una stima di $\mu$ se si osserva il campione $(Y_1 = 4, Y_2 = 2, Y_3 = 6)$

**Soluzione:**

$$
\mathbb{E}_\mu(T) = \mathbb{E}(\frac{1}{2}Y_1 + \frac{1}{4}Y_2 + aY_3) = \frac{1}{2}\mathbb{E}(Y_1) + \frac{1}{4}\mathbb{E}(Y_2) + a\mathbb{E}(Y_3) = (\frac{1}{2}+\frac{1}{4}+a)\mu  
$$
Da cui $T$ è non distorto se e solo se 

$$
\frac{1}{2}+\frac{1}{4}+a = 1 \iff a = \frac{1}{4}
$$

Fissando quindi $a = \frac{1}{4}$ e sfruttando l'indipendenza di $Y_1, Y_2, Y_3$ si ha che: 

$$
\mathbb{V}_\mu(T) = \mathbb{V}(\frac{1}{2}Y_1 + \frac{1}{4}Y_2 + \frac{1}{4}Y_3) = \frac{1}{4}\mathbb{V}(Y_1) + \frac{1}{16}\mathbb{V}(Y_2) + \frac{1}{16}\mathbb{V}(Y_3) = (\frac{1}{4} + \frac{1}{16} + \frac{1}{16})\sigma^2 = \frac{3}{8}\sigma^2
$$

Da cui 

$$
T(Y_1 = 4, Y_2 = 2, Y_3 = 6) = 4
$$
